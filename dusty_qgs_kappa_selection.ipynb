# Dusty quenched galaxies, selected by the gas kinematics

**The question.** SIMBA's quenched galaxies split into a dusty and a dust-poor population, and the previous notebook
(`agn_states_dusty_vs_pure.ipynb`, Parts 22–36) ended with one clear answer to *what divides them*: the rotation support of
the gas, $\kappa_{\rm rot}$(gas), together with the stellar mass. This notebook is the ordered restatement of that result —
**one sample, one selection, and every part below cut on it.** The AGN dose, which the older notebook carried as a second
dividing axis, is not one here: it is measured, written to the CSVs, and never split on.

**The selection.** $\kappa_{\rm rot}$ is a ratio of sums over the member gas particles, and a quenched galaxy here has only
50–250 of them, so the obvious objection is that a gas-poor galaxy could be called *rotating* because a handful of particles
happened to line up. The answer is not to multiply $\kappa$ by some function of $N$ — that would make the gas mass part of
the morphology indicator, and the gas mass is most of the dust label the selection is meant to predict. It is to put the
particle count where it belongs, **as the uncertainty of the measurement**:

$$\kappa_{\rm cut}(N) \;=\; \tfrac{1}{3} \;+\; 2\,\sigma_\kappa(N), \qquad \sigma_\kappa(N) = \frac{0.385}{\sqrt{N}},$$

the isotropic expectation plus two standard deviations of its own sampling noise. A galaxy sampled by 50 particles must sit
$0.11$ above isotropy to be called rotating; one sampled by 500 only $0.034$. The **second axis** is the stellar mass,
$\log M_\star > 10.5$ (`ROT_MASS`), ANDed into that cut: the old notebook's Part 32 found the pair the strongest divider it
could build, and Part 2 here measures what it is worth box by box — a real second channel in m100 and m50, nothing at all in
m25, where the gas is resolved. Both cuts are the same numbers in every box, and `ROT_MASS = None` recovers the purely
kinematic rule everywhere.

Parts 0–4b are **pure reads** of the `paper_ism_prediction_boxes` products (Parts 3–4b also use a cached kinematic sidecar
walked out of the caesar catalogues, which Part 3 builds once per box). Parts 5–8 also read the m25 notebook's own tables,
the K-S track products and the reduced particle cutouts; each block gates on its files and prints what is missing, so the
notebook runs unchanged where those are absent. **Parts 9–13 are the 25 Mpc box only** — the one that resolves the gas, and
the only one with particle cutouts along the branch. They add four caches of their own (the branch index, Part 9; the particle
zone census, Part 10; the radial profiles, Part 12; the CGM phase census, Part 13), all written next to the figures and all
rebuilt by deleting the file.

**What the notebook concludes** (the numbers are printed by the cells themselves, and are the ones from the run behind this
text): the two classes are **identical before the truncation** — in the gas rotation, in the ISM and stellar surface densities
and in the size of the gas disc relative to the stars (Part 4b) — and separate only after it; the X-ray channel fires within one
snapshot of the truncation but **cannot be called its cause** — its own gate is a gas criterion satisfied at the same moment,
while its mass criterion was met ~1.4 Gyr earlier and did nothing; and in the channel-off box the same galaxies quench and
**keep their dust** (−0.55 dex against −2.7 to −3.7 dex). What separates the classes afterwards is how much of the dust comes
back: half of the rotating class recovers ≥ 0.5 dex from its post-truncation minimum, against a fifth of the rest.

Parts 9–11 press on the *when* and the *where* of that in m25. The X-ray channel is on in ~15 % of both classes a Gyr before
the truncation and in ~90 % of both within one snapshot of it — an occupancy that says the mode is switched by the truncation
rather than distributed differently between the classes. The three dust ratios fall together, which is the signature of grain
**destruction** and not of dilution. And at the truncation the two classes are indistinguishable in $\Sigma_{\rm H_2}$
(Part 11, AUC 0.53 raw / 0.51 at fixed $M_\star$) — consistent with Part 4b's finding that they are the same galaxies until
the truncation happens to them.

Part 12 asks that *where* question at the one scale the annuli of Parts 5 and 7 cannot reach: the inner kiloparsec. The gas
of these galaxies is measured face-on in its own plane at the four critical points, and the profiles are split by whether the
centre is **empty** — $\Sigma$ inside 1 kpc at least a factor 2 below the peak further out, with the particle count that
would have been there had it not been, so the emptiness is a measurement and not sparseness. The hole is then carried onto
the dust clock: a galaxy that quenches with a ring and one that quenches with a filled centre are two different objects, and
the question is whether the dust knows the difference.

Part 13 takes the *mechanism* that suggests itself and turns it into something falsifiable. The proposal is a chain: the
galaxies that end up dust-poor had already blown more of their ISM into the CGM, faster, **before** the X-ray channel fired;
the channel then heated that CGM, sputtered the grains in it and cut the cold inflow off, leaving an ISM with no way to grow
dust again — while the others kept a compact ISM, kept accreting, and kept $f_{\rm gas}$ above the gate that lets the channel
fire at all. Each link is a separate measurement on the clock of the **X-ray onset**, scored **before** and **after** it
separately, on the ISM/CGM decomposition of the particles: if the difference the first link needs before the onset is only
there after it, the causal order fails, and the scorecard says so link by link.

Its last link (H7) sharpens the first one into the question that decides it: not *how much* left the galaxy, but whether it
left **bound**. Gas pushed out below the local escape velocity is a fountain — it stays in the halo, cools, and comes back
with its dust; gas pushed out above it is gone, and with it the reservoir the dust would have re-formed from. Two galaxies
can displace the same mass and end on opposite sides of that. The escape velocity is the halo's, because the cutouts hold no
dark matter — $M_{200c}$ and $R_{200c}$ from the caesar catalogue, NFW, plus the baryons the cutout resolves — and because
in a $10^{13}\,M_\odot$ halo almost nothing is formally unbound, the part also measures the continuous version of the same
question: the **apocentre** the outflowing material will reach before it turns around.

```python
# ── Part 0 — configuration: the products read, the sample, the isotropic null of kappa_rot, the AGN states, the shared helpers ──
# This notebook is the ordered version of agn_states_dusty_vs_pure: one sample, ONE selection, and every part downstream cut on it.
# The selection itself is Part 2; everything here is the machinery it and the later parts share.
import os
import warnings
import numpy as np
import pandas as pd
import h5py
import matplotlib.pyplot as plt
from astropy.table import Table
from matplotlib.lines import Line2D
from scipy.stats import mannwhitneyu, spearmanr

IN  = os.path.join(os.getcwd(), "output", "box_resolution", "ism_prediction")    # the boxes notebook's products (this notebook only reads them)
OUT = os.path.join(os.getcwd(), "output", "box_resolution", "kappa_selection")
CACHE_FITS = os.path.join(IN, "ism_prediction_catalogue.fits")                   # its Part 1 cache: every galaxy with log M* > 9.5 per (box, anchor)
CLASS_CSV  = os.path.join(IN, "ism_prediction_agn_classes.csv")                  # its Part 6a: SFT / QT / tau_q per galaxy
HIST_FILE  = os.path.join(IN, "ism_prediction_histories_{box}.hdf5")             # its Part 6a: the main-branch chains (mstar, sfr, mgas, mbh, fedd, mdust, mh2, zgas)
os.makedirs(OUT, exist_ok=True)

BOXES = ["cis100", "cis50", "cis25"]        # the three resolutions, in this order everywhere
NOX_BOX = "cis50nox"                        # the counterfactual of Part 4: the same box with SIMBA's X-ray channel compiled out
BOX_LABEL = {"cis100": "m100n1024 (sim100)", "cis50": "m50n512 s50 (sim50)", "cis25": "m25n512 (sim25)",
             "cis50nox": "m50n512 s50nox [X-ray off]"}

# ── THE SAMPLE: the m25 quenched rule of paper_ism_prediction_boxes Part 0, verbatim, plus the mass / redshift / age cuts ──
MASS_FLOOR, PASSIVE_FACTOR, NGAS_MIN, NSTAR_MIN, DUST_TO_H2_MIN = 10.0, 0.2, 50, 500, 1e-15
MASS_MIN    = 10.25       # log M* strictly above this
Z_MAX       = 10           # anchors at or below this redshift (8 = every anchor of the ladder)
AGE_MIN     = 1.5           # Gyr, the mass-weighted stellar age at the anchor: at or above this
AGE_MAX     = None        # Gyr, an upper bound if wanted (None = none)
DUSTY_CUT   = -4.5         # log M_dust/M*: at or above = a dusty QG. THE LABEL, never the selection — see Part 2
P9_FLOOR    = -6.5        # zero dust is kept at this log (the end point of the removal, never dropped)
SSFR_FLOOR  = 1e-19
HE, R_PROJ_OVER_3D = 1.36, 1.
KEY = ["box", "snap", "gal_id"]

POPS = {"dusty": dict(color="#d7191c", lw=2.6, ls="-", label=rf"dusty QG  ($\log M_{{\rm dust}}/M_\star \geq {DUSTY_CUT:g}$)"),
        "pure":  dict(color="#2c7bb6", lw=2.2, ls="-", label=rf"dust-poor QG  ($\log M_{{\rm dust}}/M_\star < {DUSTY_CUT:g}$)")}

# ── THE SELECTION AXIS: kappa_rot(gas) measured against the isotropic null, with the particle count as its UNCERTAINTY ──
# A quenched galaxy in these boxes holds ~50-250 gas particles and kappa_rot = K_rot/K is a ratio of sums over them, so a
# gas-poor galaxy can be called "rotating" because a handful of particles happened to line up. The null is exact:
#   for an isotropic velocity field  E[kappa] = 1/3  (one of three components is azimuthal) and Var(v_phi^2/v^2) = 4/45 for a
#   uniform direction; the v^2 weighting of the denominator gives N_eff = N (E w)^2 / E w^2 = KAPPA_WEFF x N for a Maxwellian,
#   so  s_iso(N) = sqrt(4/45 / 0.6 N) = 0.385 / sqrt(N)  —  +-0.044 at N = 76, the median of the m100 sample.
# The particle count therefore enters as the ERROR BAR ON THE MEASUREMENT, not as a factor multiplying kappa: a galaxy with few
# particles must lie FARTHER above 1/3 before it is called rotating. That is the whole difference from a median split, and it is
# why the axis cannot smuggle the gas mass (and with it the dust) in through the back door — see Part 34 of the old notebook.
KAPPA_ISO  = 1.0 / 3.0                                  # the isotropic expectation
KAPPA_VARF, KAPPA_WEFF = 4.0 / 45.0, 0.6                # Var(v_phi^2/v^2) for a uniform direction, and N_eff / N under the v^2 weighting
KAPPA_C    = float(np.sqrt(KAPPA_VARF / KAPPA_WEFF))    # s_iso(N) = KAPPA_C / sqrt(N) = 0.385 / sqrt(N)
ROT_SIG    = 5.0                                        # THE CUT: rotating = kappa > 1/3 + ROT_SIG s_iso(N), i.e. 2 sigma above isotropy
# THE SECOND AXIS OF THE SELECTION. The old notebook's Part 32 scored every divider it could build and found the pair
# kappa_gas + log M* the strongest one, and Part 2 measures here what it is worth: out of fold it adds +0.02 (m100) / +0.05
# (m50) of AUC toward the dust label and roughly TRIPLES the purity of the selected group, and it is worth nothing at all in
# m25 (+0.00), where the gas is resolved well enough that the kinematics alone carry the information. The mass is therefore
# kept as a SECOND CUT and never mixed into kappa itself: the two axes stay separately readable, and setting this to None
# recovers the purely kinematic rule everywhere in the notebook.
ROT_MASS   = None                                       # log M* strictly above this, ANDed with the kinematic cut (None = off)
KAPPA_COL, NGAS_COL = "kappa_gas", "ngas"               # the catalogue columns the axis is built from (whole-galaxy kappa_rot of the gas)

sig_iso   = lambda n: KAPPA_C / np.sqrt(np.where(np.asarray(n, float) > 0, np.asarray(n, float), np.nan))
kappa_cut = lambda n, sig=None: KAPPA_ISO + (ROT_SIG if sig is None else sig) * sig_iso(n)
kappa_z   = lambda k, n: (np.asarray(k, float) - KAPPA_ISO) / sig_iso(n)     # the significance: how many s_iso above isotropy


def rot_rule(kz, log_mstar=None, sig=None, mcut="default"):
    """THE SELECTION, for any set of rows and in one place: the kinematic significance above ROT_SIG and, when the mass axis
    is on, log M* above ROT_MASS. Every part of the notebook that has to classify rows calls this — Part 2 for the sample,
    Parts 6-8 for the K-S products' own (larger) anchor set — so the rule cannot drift between them."""
    m = np.asarray(kz, float) > (ROT_SIG if sig is None else float(sig))
    mc = ROT_MASS if mcut == "default" else mcut
    if mc is not None:
        m = m & (np.asarray(log_mstar, float) > float(mc))
    return m

# THE TWO CLASSES everything below is split by. "rotating" is a statement about the MEASUREMENT (the gas is resolved to be
# rotating), not a morphological claim: kappa_rot ~ 0.5 is rotation-supported, 1/3 is isotropic, and the cut sits between them
# at a distance set by how well each galaxy is sampled. With the mass axis on, the second class is "everything else" — a
# galaxy can be in it for either reason, which is exactly what the selection means and is said in the labels.
_MTXT = "" if ROT_MASS is None else rf"  and  $\log M_\star > {ROT_MASS:g}$"
KCLS = {"rot": dict(color="#b2182b", lw=2.6, ls="-",  short="rotating" + ("" if ROT_MASS is None else ", massive"),
                    label=rf"rotating gas  ($\kappa_{{\rm gas}} > 1/3 + {ROT_SIG:g}\,s_{{\rm iso}}(N_{{\rm gas}})$)" + _MTXT),
        "iso": dict(color="#2166ac", lw=2.4, ls="-",
                    short="not resolved as rotating" if ROT_MASS is None else "the rest",
                    label=(rf"no resolved rotation  ($\kappa_{{\rm gas}} \leq 1/3 + {ROT_SIG:g}\,s_{{\rm iso}}$)"
                           if ROT_MASS is None else
                           rf"the rest  (no resolved rotation, or $\log M_\star \leq {ROT_MASS:g}$)"))}
KC = list(KCLS)                                          # ["rot", "iso"], the drawing order

# ── the AGN states: SIMBA's own feedback branches evaluated at every snapshot of the main branch ──
JET_LOGMBH, JET_FEDD = 7.5, 0.2       # SIMBA's jet criterion: M_BH above 10^7.5 AND f_Edd below 0.2 (the jet velocity then ramps as w_jet)
W_FULL        = 0.999                 # "full-velocity jet": w_jet = clip(log10(0.2 / f_Edd), 0, 1) at/above this, i.e. f_Edd <= 0.02
XRAY_FGAS_MAX = float(os.environ.get("XRAY_FGAS_MAX", "0.2"))   # SIMBA's X-ray condition, M_gas/M_* below this (Dave+19)
ETA_RAD       = 0.1                   # radiative efficiency behind Mdot_BH = f_Edd L_Edd / (eta c^2)
MDOT_PER_MBH  = 1.26e38 / (ETA_RAD * 8.98755e20) / 1.98892e33 * 3.15576e7        # Msun/yr of accretion per Msun of BH at f_Edd = 1 (= 1 / 4.5e7 yr)
FEDD_FLOOR, MDOT_FLOOR, FGAS_FLOOR = 1e-6, 1e-7, 1e-3

STATES = [("radiative", "#9467bd", r"radiative — accreting, no jets"),
          ("choked",    "#2ca02c", rf"X-ray choked — jets, $f_{{\rm gas}} \geq {XRAY_FGAS_MAX:g}$"),
          ("xray",      "#1f77b4", rf"X-ray on — full jets $+$ $f_{{\rm gas}} < {XRAY_FGAS_MAX:g}$"),
          ("dormant",   "0.55",    r"dormant — $\dot M_{\rm BH} = 0$")]
SKEYS  = [s for s, _, _ in STATES]
SCOL   = {s: c for s, c, _ in STATES}
SLAB   = {s: l for s, _, l in STATES}

# ── the clock and the estimator settings ──
TG     = np.arange(-3.0, 4.001, 0.25)      # THE CLOCK: t - t_SFT (and t - t_x1) [Gyr], the grid every track is interpolated onto
NMIN   = 5                                 # an epoch is drawn only where this many galaxies contribute
NBOOT  = int(os.environ.get("AGN_NBOOT", "500"))   # the bootstrap band; the runtime driver of Parts 3-5, drop it for a smoke test
PAPER_DPI = 300
plt.rcParams.update({"font.size": 11, "axes.titlesize": 11, "axes.labelsize": 11, "legend.fontsize": 9, "xtick.labelsize": 9.5,
                     "ytick.labelsize": 9.5, "axes.titleweight": "bold", "pdf.fonttype": 42})


def need(path, made_by):
    if not os.path.exists(path):
        raise FileNotFoundError(f"{path} is missing: {made_by}")
    return path


def paper_save(fig, stem):
    for ext in ("png", "pdf"):
        fig.savefig(os.path.join(OUT, f"{stem}.{ext}"), dpi=PAPER_DPI, bbox_inches="tight")
    print("figure ->", os.path.join(OUT, f"{stem}.{{png,pdf}}"))


def mw_auc(x, m1):
    """x over two populations (m1 = population 1): AUC = P(x_1 > x_0), the two-sided Mann-Whitney p, the two finite counts."""
    x = np.asarray(x, float); m1 = np.asarray(m1, bool)
    a, b = x[m1 & np.isfinite(x)], x[~m1 & np.isfinite(x)]
    if len(a) < 3 or len(b) < 3 or np.nanstd(np.concatenate([a, b])) == 0:
        return np.nan, np.nan, len(a), len(b)
    u = mannwhitneyu(a, b, alternative="two-sided")
    return float(u.statistic) / (len(a) * len(b)), float(u.pvalue), len(a), len(b)


def wilson(k, n, z=1.0):
    """The Wilson interval on a fraction (no statsmodels on the cluster kernel); z = 1 gives the 68 % band."""
    k, n = float(k), float(n)
    if n <= 0:
        return np.nan, np.nan, np.nan
    p = k / n; d = 1.0 + z * z / n
    c = (p + z * z / (2 * n)) / d
    h = z * np.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / d
    return p, max(0.0, c - h), min(1.0, c + h)


def med_band(A, nmin=None, seed=11, nboot=None):
    """A (n_gal, n_t) stack -> the median track, its bootstrap 16-84 band and the per-epoch finite count; epochs under nmin blanked."""
    A = np.asarray(A, float)
    nmin = NMIN if nmin is None else nmin
    nb = NBOOT if nboot is None else nboot
    if A.ndim != 2 or not len(A):
        return (np.full(len(TG), np.nan),) * 3 + (np.zeros(len(TG), int),)
    n_t = np.isfinite(A).sum(axis=0)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        cent = np.nanmedian(A, axis=0)
        rng = np.random.default_rng(seed)
        bs = np.array([np.nanmedian(A[rng.integers(0, len(A), len(A))], axis=0) for _ in range(nb)])
        lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
    ok = n_t >= nmin
    return np.where(ok, cent, np.nan), np.where(ok, lo, np.nan), np.where(ok, hi, np.nan), n_t


def to_grid(trel, y, grid=None):
    """One galaxy's track interpolated onto the clock; nothing is extrapolated past the sampled range."""
    g = TG if grid is None else grid
    ok = np.isfinite(trel) & np.isfinite(y)
    return np.interp(g, np.asarray(trel)[ok], np.asarray(y)[ok], left=np.nan, right=np.nan) if ok.sum() >= 3 else np.full(len(g), np.nan)


def agn_states(mstar, mgas, mbh, fedd):
    """The four exclusive AGN states of an (n_t, n_gal) history block, plus w_jet, f_gas, Mdot_BH and the diagnostic sub-masks.
    A cell where the galaxy does not exist is NaN everywhere; a cell whose BH columns are absent counts as dormant.
    NOTE f_gas = M_gas/M_* is SIMBA's OWN X-ray gate (Dave+19) and stays on the total gas — it is a feedback criterion, not an
    ISM diagnostic."""
    with np.errstate(all="ignore"):
        exist = np.isfinite(mstar) & (mstar > 0)
        fgas  = np.where(exist, mgas / np.where(exist, mstar, np.nan), np.nan)
        bh    = np.isfinite(mbh) & (mbh > 0)
        acc   = exist & bh & np.isfinite(fedd) & (fedd > 0)                       # the BH is accreting: Mdot_BH > 0
        big   = bh & (mbh > 10 ** JET_LOGMBH)
        wjet  = np.where(acc & big, np.clip(np.log10(JET_FEDD / np.clip(fedd, 1e-12, None)), 0.0, 1.0), 0.0)
        mdot  = np.where(acc, MDOT_PER_MBH * np.nan_to_num(fedd) * np.nan_to_num(mbh), 0.0)
        jet   = acc & big & (fedd < JET_FEDD)                                     # SIMBA's jet criterion
        xray  = jet & (wjet >= W_FULL) & (fgas < XRAY_FGAS_MAX)                   # full velocity AND gas-poor: the X-ray channel fires
        S = {"xray": xray, "choked": jet & ~xray, "radiative": acc & ~jet, "dormant": exist & ~acc}
        D = {"nobh": exist & ~bh, "big": big, "gate": exist & (fgas < XRAY_FGAS_MAX), "jet": jet}
    f = lambda m: np.where(exist, m.astype(float), np.nan)
    return ({k: f(v) for k, v in S.items()}, {k: f(v) for k, v in D.items()},
            dict(exist=exist, fgas=fgas, wjet=np.where(exist, wjet, np.nan), mdot=np.where(exist, mdot, np.nan)))

# ── the catalogue cache -> the quenched sample (the boxes notebook's Part 1 rule) -> the dust LABEL ──
CAT = Table.read(need(CACHE_FITS, "paper_ism_prediction_boxes Part 1")).to_pandas()
CAT["box"] = [(v.decode() if isinstance(v, bytes) else str(v)).strip() for v in CAT["box"]]
_ms, _sfr = CAT["mstar"].to_numpy(float), CAT["sfr"].to_numpy(float)
with np.errstate(divide="ignore", invalid="ignore"):
    _ssfr = np.where(_ms > 0, _sfr / _ms, np.nan)
    _lm   = np.log10(np.where(_ms > 0, _ms, np.nan))
_quenched = ((_lm > MASS_FLOOR) & (_ssfr < PASSIVE_FACTOR / (CAT["t_H_gyr"].to_numpy(float) * 1e9))
             & (CAT["ngas"].to_numpy(float) >= NGAS_MIN) & (CAT["nstar"].to_numpy(float) >= NSTAR_MIN)
             & (CAT["mdust"].to_numpy(float) >= DUST_TO_H2_MIN * CAT["mh2"].to_numpy(float)))
QM = CAT[_quenched].copy()
with np.errstate(divide="ignore", invalid="ignore"):
    ms = np.where(QM["mstar"].to_numpy(float) > 0, QM["mstar"].to_numpy(float), np.nan)
    QM["log_mstar"] = np.log10(ms)
    fd = QM["mdust"].to_numpy(float) / ms
    QM["lfd"]   = np.where(fd > 0, np.log10(np.where(fd > 0, fd, 1.0)), P9_FLOOR)       # zero dust at the floor, never dropped
    QM["lfgas"] = np.log10(np.where(QM["mgas"].to_numpy(float) > 0, QM["mgas"].to_numpy(float) / ms, np.nan))
    QM["lfh2"]  = np.log10(np.where(QM["mh2"].to_numpy(float) > 0, HE * QM["mh2"].to_numpy(float) / ms, np.nan))
    QM["lssfr"] = np.log10(np.clip(np.maximum(QM["sfr"].to_numpy(float), 0.0) / ms, SSFR_FLOOR, None))
    QM["re"]    = R_PROJ_OVER_3D * QM["r_half_star"].to_numpy(float)
    QM["lsig"]  = np.log10(np.where(QM["re"] > 0, 0.5 * ms / (np.pi * QM["re"] ** 2), np.nan))
    QM["lngas"] = np.log10(np.where(QM[NGAS_COL].to_numpy(float) > 0, QM[NGAS_COL].to_numpy(float), np.nan))
    mbh = QM["mbh"].to_numpy(float)
    QM["log_mbh"] = np.where(mbh > 0, np.log10(np.where(mbh > 0, mbh, 1.0)), np.nan)

_n_before = len(QM)
_cuts = [(f"log M* > {MASS_MIN:g}", QM["log_mstar"].to_numpy(float) > MASS_MIN)]
if Z_MAX is not None:
    _cuts.append((f"z <= {Z_MAX:g}", QM["z"].to_numpy(float) <= float(Z_MAX) + 1e-9))
if AGE_MIN is not None:
    _cuts.append((f"age >= {AGE_MIN:g} Gyr", QM["age"].to_numpy(float) >= float(AGE_MIN)))
if AGE_MAX is not None:
    _cuts.append((f"age <= {AGE_MAX:g} Gyr", QM["age"].to_numpy(float) <= float(AGE_MAX)))
_keep, _drop = np.ones(len(QM), bool), []
for _lab, _m in _cuts:                                    # each cut's own toll, in the order it is applied
    _drop.append((_lab, int((_keep & ~_m).sum())))
    _keep &= _m
QM = QM[_keep].reset_index(drop=True)
QM["pop"] = np.where(QM["lfd"].to_numpy(float) >= DUSTY_CUT, "dusty", "pure")

CLS = pd.read_csv(need(CLASS_CSV, "paper_ism_prediction_boxes Part 6a"))
CLS_COLS = [c for c in ["t_sft", "t_qt", "tau_q", "w_pre", "agn_class", "agn_class3", "agn_onset"] if c in CLS.columns]
QM = QM.merge(CLS[KEY + CLS_COLS], on=KEY, how="left")    # left join: a box / anchor without histories keeps NaN

print("THE SAMPLE (declared here, inherited by every part): the m25 quenched rule on the catalogue cache, then "
      + ", ".join(l for l, _ in _cuts) + f"  ->  {_n_before} -> {len(QM)} rows"
      + ("  (" + ", ".join(f"{l}: -{n}" for l, n in _drop if n) + ")" if any(n for _, n in _drop) else ""))
print("  per box: " + ", ".join(f"{b} {int((QM['box'] == b).sum())}" for b in BOXES if (QM['box'] == b).any())
      + f";  anchors z = {QM['z'].min():.2f}-{QM['z'].max():.2f}, age = {QM['age'].min():.2f}-{QM['age'].max():.2f} Gyr")
print(f"  the dust LABEL at log M_dust/M* = {DUSTY_CUT:g}: dusty {int((QM['pop'] == 'dusty').sum())}, "
      f"dust-poor {int((QM['pop'] == 'pure').sum())} — a label to be predicted, never a selector (Part 2 selects on the gas kinematics).")
print(f"the X-ray condition in force: f_gas < {XRAY_FGAS_MAX:g} (env XRAY_FGAS_MAX); full jet at w_jet >= {W_FULL:g}, "
      f"i.e. f_Edd <= {JET_FEDD * 10 ** -W_FULL:.3f}")
print(f"the isotropic null of kappa_rot: E = {KAPPA_ISO:.3f}, s_iso(N) = {KAPPA_C:.3f} / sqrt(N)  ->  the cut of Part 2 is "
      f"kappa_{KAPPA_COL.split('_')[-1]} > 1/3 + {ROT_SIG:g} s_iso({NGAS_COL})"
      + (f"  AND  log M* > {ROT_MASS:g} (the second axis; set ROT_MASS = None for the purely kinematic rule)"
         if ROT_MASS is not None else "  (kinematics only: ROT_MASS is off)"))

# ── which boxes have a histories cache, and what that cache carries ──
HAVE_MD, HAVE_MH2, HAVE_Z, BOXES_OK, CHAIN_COV = {}, {}, {}, [], {}
for _b in BOXES:
    _p = HIST_FILE.format(box=_b)
    if not os.path.exists(_p):
        print(f"{_b}: {_p} missing — paper_ism_prediction_boxes Part 6a builds it (cluster); skipped"); continue
    with h5py.File(_p, "r") as _f:
        _g0 = _f[next(iter(_f))]
        HAVE_MD[_b], HAVE_MH2[_b], HAVE_Z[_b] = "mdust" in _g0, "mh2" in _g0, "zgas" in _g0
        _sub, _nh = QM[QM["box"] == _b], 0                        # how much of THE SAMPLE the cache actually covers
        for _a, _Sa in _sub.groupby("snap"):
            _k = f"snap{int(_a):03d}"
            if _k in _f:
                _ids = set(int(_v) for _v in _f[_k]["gal_ids"][:])
                _nh += int(sum(int(_x) in _ids for _x in _Sa["gal_id"]))
        CHAIN_COV[_b] = (_nh, len(_sub))
    BOXES_OK.append(_b)
print("histories caches: " + ", ".join(f"{b} (mdust {'y' if HAVE_MD[b] else 'n'}, mh2 {'y' if HAVE_MH2[b] else 'n'}, "
                                       f"zgas {'y' if HAVE_Z[b] else 'n'})" for b in BOXES_OK))
print("the effective sample of every part that reads histories (3 onwards): "
      + ", ".join(f"{b} {CHAIN_COV[b][0]}/{CHAIN_COV[b][1]} ({100 * CHAIN_COV[b][0] / max(1, CHAIN_COV[b][1]):.0f} %)" for b in BOXES_OK))
```

## Part 1 — the census: one sample, and a dust label that is never a selector

**The sample is fixed once, in Part 0** — the m25 quenched rule of `paper_ism_prediction_boxes`, then $\log M_\star >$
`MASS_MIN`, $z \leq$ `Z_MAX` and age $\geq$ `AGE_MIN` — and every part of the notebook runs on exactly those galaxies. Part 0
prints each cut's own toll so the sample is auditable in one place.

The split at $\log M_{\rm dust}/M_\star =$ `DUSTY_CUT` is kept, but **only as a label**: it is what Part 2's kinematic
selection is scored against, and it is never used to select. That is the difference from the notebook this one replaces,
where the dust cut defined the populations and every later statement about the dust risked being circular.

This cell reports what the two dust populations look like in the quantities the label does *not* see — including
$\kappa_{\rm rot}$(gas) and $N_{\rm gas}$, the two numbers Part 2 is built from — and draws the two cuts themselves.

**On the age cut.** `AGE_MIN` is not neutral: it removes the young, recently truncated galaxies, and it is *not* calibrated on
the observed sample and cannot be — the ALMA-C11 ages are CIGALE `sfhdelayedbq` **onset** ages while `age` here is caesar's
**mass-weighted** age, which for a delayed-$\tau$ SFH sits roughly $2\tau$ lower. It is a cut on the simulation's own age and
is internally consistent; just do not read it as "the observed floor". Set `AGE_MIN = None` in Part 0 to drop it.

```python
# ── Part 1 — the census: what the cuts select, and what the dust label separates before anything is selected on it ──
Q1 = [("age", "age [Gyr]"), ("log_mstar", "log M*"), ("z", "redshift"), ("lfd", "log M_dust/M*"),
      ("lfh2", "log 1.36 M_H2/M*"), ("lssfr", "log sSFR"), ("lsig", "log Sigma_e"), ("kappa_gas", "kappa_rot gas"),
      ("kappa_star", "kappa_rot star"), ("lngas", "log N_gas"), ("log_mbh", "log M_BH"), ("tau_q", "tau_q [Gyr]")]

rows1 = []
print(f"the two dust populations per box (medians; AUC = P(dusty > dust-poor), '*' = Mann-Whitney p < 0.01)\n")
for b in BOXES:
    g = QM[QM["box"] == b]
    if not len(g):
        print(f"  {b}: absent from the catalogue cache\n"); continue
    mD, mP = g["pop"] == "dusty", g["pop"] == "pure"
    print(f"  {'=' * 104}\n  {BOX_LABEL[b]}: {len(g)} quenched galaxies over {g['snap'].nunique()} anchors — "
          f"dusty {int(mD.sum())} ({100 * mD.mean():.0f} %), dust-poor {int(mP.sum())}; with an SFT: dusty "
          f"{int(np.isfinite(g.loc[mD, 't_sft']).sum())}, dust-poor {int(np.isfinite(g.loc[mP, 't_sft']).sum())}; "
          f"satellites {100 * (1 - g['central'].mean()):.0f} %")
    print(f"  {'quantity':22s} {'dusty':>22s} {'dust-poor':>23s} {'AUC':>7s} {'N d/p':>12s}")
    for q, lab in Q1:
        if q not in g.columns:
            continue
        vd, vp = g.loc[mD, q].to_numpy(float), g.loc[mP, q].to_numpy(float)
        auc, pv, n1, n0 = mw_auc(np.r_[vd, vp], np.r_[np.ones(len(vd), bool), np.zeros(len(vp), bool)])
        f3 = lambda v: (np.nan, np.nan, np.nan) if not np.isfinite(v).any() else tuple(np.nanpercentile(v, [50, 16, 84]))
        md, lo_d, hi_d = f3(vd); mp, lo_p, hi_p = f3(vp)
        print(f"  {lab:22s} {md:7.2f} [{lo_d:6.2f},{hi_d:6.2f}] {mp:8.2f} [{lo_p:6.2f},{hi_p:6.2f}] {auc:7.2f}{'*' if np.isfinite(pv) and pv < 0.01 else ' '} {n1:5d}/{n0:5d}")
        rows1.append(dict(box=b, quantity=q, med_dusty=md, lo_dusty=lo_d, hi_dusty=hi_d, med_pure=mp, lo_pure=lo_p, hi_pure=hi_p,
                          auc=auc, p=pv, n_dusty=n1, n_pure=n0))
    print()

C1 = pd.DataFrame(rows1)
C1.to_csv(os.path.join(OUT, "kappa_census.csv"), index=False)
print(f"census -> {os.path.join(OUT, 'kappa_census.csv')}")

# the two cuts drawn: the dust axis the LABEL is made on, and the mass axis the sample is cut on (before the cut, so it is visible)
QALL = CAT[_quenched].copy()
with np.errstate(divide="ignore", invalid="ignore"):
    _msa = np.where(QALL["mstar"].to_numpy(float) > 0, QALL["mstar"].to_numpy(float), np.nan)
    QALL["log_mstar"] = np.log10(_msa)
    _fda = QALL["mdust"].to_numpy(float) / _msa
    QALL["lfd"] = np.where(_fda > 0, np.log10(np.where(_fda > 0, _fda, 1.0)), P9_FLOOR)

fig, axs = plt.subplots(2, len(BOXES), figsize=(4.2 * len(BOXES), 5.6), squeeze=False)
for ci, b in enumerate(BOXES):
    g, ga = QM[QM["box"] == b], QALL[QALL["box"] == b]
    ax = axs[0, ci]
    ax.hist(g["lfd"].to_numpy(float), bins=np.arange(P9_FLOOR - 0.25, -1.749, 0.25), color="0.75", edgecolor="0.35", lw=0.6)
    ax.axvline(DUSTY_CUT, color="k", lw=1.6, ls="--")
    for pn, m in (("dusty", g["pop"] == "dusty"), ("pure", g["pop"] == "pure")):
        ax.hist(g.loc[m, "lfd"].to_numpy(float), bins=np.arange(P9_FLOOR - 0.25, -1.749, 0.25), color=POPS[pn]["color"], alpha=0.55, lw=0)
    ax.set_title(BOX_LABEL[b], fontsize=10)
    ax.set_xlabel(r"$\log M_{\rm dust}/M_\star$"); ax.set_ylabel("galaxies" if ci == 0 else "")
    ax.text(0.03, 0.94, f"dusty {100 * (g['pop'] == 'dusty').mean():.0f} %\nof {len(g)}", transform=ax.transAxes, va="top", fontsize=8.5)
    ax.grid(alpha=0.25, lw=0.5)
    ax = axs[1, ci]
    bins_m = np.arange(9.9, 12.61, 0.1)
    ax.hist(ga["log_mstar"].to_numpy(float), bins=bins_m, color="0.85", edgecolor="0.6", lw=0.5)
    for pn, m in (("dusty", g["pop"] == "dusty"), ("pure", g["pop"] == "pure")):
        ax.hist(g.loc[m, "log_mstar"].to_numpy(float), bins=bins_m, color=POPS[pn]["color"], alpha=0.6, lw=0)
    ax.axvline(MASS_MIN, color="k", lw=1.6, ls="--")
    ax.set_xlabel(r"$\log M_\star$ [$M_\odot$]"); ax.set_ylabel("galaxies" if ci == 0 else "")
    ax.grid(alpha=0.25, lw=0.5)
_h1 = [Line2D([], [], color=POPS[p]["color"], lw=6, alpha=0.6, label=POPS[p]["label"]) for p in POPS]
_h1 += [Line2D([], [], color="0.8", lw=6, label="quenched sample before the cut"), Line2D([], [], color="k", lw=1.6, ls="--", label="the cut")]
fig.legend(handles=_h1, loc="lower center", ncol=4, frameon=False, fontsize=8.5, bbox_to_anchor=(0.5, -0.03))
fig.suptitle(rf"the sample: quenched, $\log M_\star > {MASS_MIN:g}$, age $\geq {AGE_MIN:g}$ Gyr — and the dust LABEL at "
             rf"$\log M_{{\rm dust}}/M_\star = {DUSTY_CUT:g}$", y=0.99, fontsize=11)
fig.tight_layout(rect=(0, 0.05, 1, 0.965))
paper_save(fig, "kappa_sample")
plt.show()
```

## Part 2 — the selection: $\kappa_{\rm rot}$(gas) against isotropy, with $N_{\rm gas}$ as the uncertainty, and the stellar mass

**The rule**, both axes, applied identically in every box:

$$\text{selected} \iff \kappa_{\rm gas} > \tfrac13 + \texttt{P2\_SIG}\,\cdot\,s_{\rm iso}(N_{\rm gas})
\quad\textbf{and}\quad \log M_\star > \texttt{ROT\_MASS},
\qquad s_{\rm iso}(N) = \sqrt{\frac{4/45}{0.6\,N}} = \frac{0.385}{\sqrt N}.$$

The first axis is what the rest of this section is about; the second is `ROT_MASS = 10.5` in Part 0, and *why* it is there is
the last section below. Setting `ROT_MASS = None` re-cuts the whole notebook on the kinematics alone.

**Where the null comes from.** For an isotropic velocity field one of the three velocity components is azimuthal, so
$\mathbb{E}[\kappa_{\rm rot}] = 1/3$; the variance of $f = v_\phi^2/v^2$ for a uniformly distributed direction is $4/45$; and
the $v^2$ weighting of the denominator in the Sales+12 estimator gives an effective sample size
$N_{\rm eff} = N\,(\mathbb{E}w)^2/\mathbb{E}w^2 = 0.6\,N$ for a Maxwellian. Hence $s_{\rm iso}(N) = 0.385/\sqrt N$: $\pm0.044$
at $N = 76$, the median of the m100 sample. The constants are `KAPPA_VARF` and `KAPPA_WEFF` in Part 0.

**Why the count enters here and nowhere else.** The alternative — shrinking $\kappa$ towards $1/3$ by a factor $N/(N+N_0)$, or
any other weighting of the value itself — makes the *gas mass* part of the morphology indicator. In this sample the gas mass
is very nearly the dust label (that is what the whole notebook is about), so such an axis would score well for a reason that
has nothing to do with rotation. Putting $N$ in the threshold instead leaves the measured quantity untouched and only asks a
poorly sampled galaxy to be further from isotropy before it is believed.

**How to read the figure.**

* **(a) the plane.** $\kappa$ against $N_{\rm gas}$, with $1/3$, the $\pm s_{\rm iso}(N)$ band and the cut curve. The dusty
  galaxies (the label, red) should sit above the curve if the selection is worth anything — and they should not be piled at
  low $N$, which is what a shot-noise artefact would look like.
* **(b) the null.** The significance $z = (\kappa - 1/3)/s_{\rm iso}(N)$ against the unit normal an isotropic population would
  produce. If $\kappa$ were sampling noise the histogram and the green curve would coincide. They do not: $\mathrm{sd}(z)$ is
  6–23 and a third of the sample is above $2\sigma$ where the null allows 2.3 %. **The spread of the axis is intrinsic.**
* **(c) the outcome.** The dusty fraction in equal-count bins of $z$, with Wilson 68 % intervals. Monotonic, and the rise
  happens at the cut, not somewhere else.
* **(d) the cut on trial.** Purity and completeness scanned over the threshold. The point is that nothing special happens at
  $2\sigma$ — the numbers move slowly and smoothly — so the choice is a physical one, not a tuned one.

### The second axis: what the stellar mass adds, and where it does not

The old notebook's Part 32 found $\kappa_{\rm gas}$ **and** $\log M_\star$ the strongest *pair*, and its Part 35 explained
why: the two are orthogonal ($\rho \approx 0$) and act independently on the same term, the share of the metals locked in
grains. The second figure (`kappa_selection_mass`) is that pair — **both cuts drawn**, the shaded quadrant the selected class,
the box median as a dotted line for scale — with the dusty share of each quadrant written in it and the out-of-fold score of
$z$ alone against $z + \log M_\star$ in the title. That is the honest measure of what the mass channel is worth, and it is
**not the same in the three boxes**:

| | m100 | m50 | m25 |
|---|---|---|---|
| out of fold, $z$ | 0.83 | 0.85 | 0.89 |
| out of fold, $z + \log M_\star$ | 0.85 | 0.90 | 0.89 |
| quadrants, dusty share (low/low → high/high) | 1 / 3 / 7 / **28 %** | 0 / 5 / 10 / **26 %** | 8 / 0 / 31 / **38 %** |

So in the coarse boxes the mass axis is a real second channel — it triples the purity of the selected group, at two thirds of
the completeness — while in **m25, the box that resolves the gas, it adds nothing** ($\mathrm{AUC}(\log M_\star) = 0.53$).
That is exactly what Part 35 of the old notebook predicted: the physical axis there is $\Sigma_{\rm gas}$, the grain-growth
density, for which $M_\star$ is the coarse boxes' unresolved stand-in.

**The notebook takes it anyway, and says what it costs.** With `ROT_MASS = 10.5` the selected group goes from 557 to 373
galaxies in m100 (purity 18 → 24 %, completeness 82 → 71 %) and from 64 to 44 in m50 (19 → 27 %, completeness unchanged at
80 %); in m25 it goes from 29 to 19 and buys nothing (35 → 32 % purity for half the completeness). The cell prints exactly
that trade, box by box, next to the rule written without the mass axis, so the price is never hidden. Because the second class
is now *everything else*, a galaxy can be in it for either reason — that is what the label "the rest" means throughout.

Downstream this is one rule in one place: `rot_rule()` in Part 0, called by Part 2 for the sample and by Parts 6–8 for the
K-S products' own (larger) anchor set, so the division cannot drift between them.

**The other knobs.** `P2_SIG` re-cuts the entire notebook. `KAPPA_COL` / `NGAS_COL` in Part 0 choose the tracer.

**What is *not* claimed.** "Rotating" here is a statement about the measurement — the gas is resolved to rotate more than an
isotropic field would — not a morphological classification. $\kappa_{\rm rot} \approx 0.5$ is rotation-supported and $1/3$ is
isotropic; the cut sits between them, at a distance set by how well each galaxy is sampled. The cell also reports the share
above $\kappa = 0.5$ outright, which is the classical statement, for comparison.

```python
# ── Part 2 — THE SELECTION: kappa_rot(gas) against isotropy, with N_gas as the uncertainty, AND the stellar mass ──
# The rule, once, for every box:   SELECTED <=> kappa_gas > 1/3 + P2_SIG x s_iso(N_gas)  AND  log M* > P2_MASS,   s_iso = 0.385/sqrt(N)
# The particle count enters ONLY as the width of the null: a galaxy sampled by 50 particles has to sit 0.11 above isotropy to be
# called rotating, one sampled by 500 only 0.034. Nothing multiplies kappa by a function of N, so the gas mass never becomes part
# of the morphology indicator — which matters here because the gas mass is most of the dust label this selection is tested against.
#   (a) THE PLANE      kappa against N_gas with the isotropic 1/3, its +-1 s_iso(N) band and the cut curve; the dust label as colour.
#   (b) THE NULL       the significance z = (kappa - 1/3)/s_iso(N) against the unit normal the isotropic null predicts. If kappa were
#                      shot noise the two would coincide; the width of the observed distribution is the answer.
#   (c) THE OUTCOME    the dusty fraction against z, equal-count bins, Wilson 68 % — is the axis monotonic in what it predicts?
#   (d) THE CUT ON TRIAL   the same two numbers (purity, completeness) scanned over the threshold, so the choice of 2 sigma can be
#                      seen not to be fine-tuned.
#   FIG 2  THE SECOND AXIS drawn: the z x log M* plane with BOTH cuts on it, and the dusty share of each quadrant written in it.
# PRINTED: the null check, the rule's confusion against the dust label, the same rule written four other ways (the median split,
#          a fixed cut at the same level, kappa > 0.5, and the kinematic cut with the mass axis dropped) — including how many
#          galaxies change side — and what the mass axis is worth out of fold, which is why it is in the rule at all.
P2_SIG    = ROT_SIG        # the cut, in units of s_iso(N). Part 0's ROT_SIG; change it here to re-cut the whole notebook
P2_MASS   = ROT_MASS       # THE SECOND AXIS, ANDed into the rule: log M* strictly above this. Part 0's ROT_MASS (10.5, the old
                           # notebook's Part 32 pair); None here re-cuts the whole notebook on the kinematics alone
P2_NBIN   = 6              # equal-count bins of z in panel (c)
P2_SCAN   = np.arange(0.0, 5.01, 0.25)     # the thresholds panel (d) scans
P2_ZLIM   = (-25, 25)      # the drawing range of the significance axis (values outside are piled into the end bins)
P2_HIROT  = 0.5            # "rotation-supported" in the classical sense — reported alongside the significance cut
P2_NBINZ  = 40
P2_MASSFIG = True          # the second figure: the z x log M* plane and what the mass axis adds to the selection
P2_CVK     = 5             # folds of the out-of-fold score behind that figure

_kv, _nv = QM[KAPPA_COL].to_numpy(float), QM[NGAS_COL].to_numpy(float)
QM["s_iso"] = sig_iso(_nv)
QM["kz"]    = kappa_z(_kv, _nv)
QM["k_cut"] = kappa_cut(_nv, P2_SIG)
_rot = rot_rule(QM["kz"], QM["log_mstar"], sig=P2_SIG, mcut=P2_MASS)     # Part 0's rule, in one place for the whole notebook
QM["rot"]  = _rot
QM["kcls"] = np.where(_rot, "rot", "iso")
QM["kappa_z_cat"] = QM["kz"]        # the same number under the name Parts 6-8 look for when they take this as their geometry axis
KROT = {(b_, int(s_), int(g_)): bool(r_) for b_, s_, g_, r_ in zip(QM["box"], QM["snap"], QM["gal_id"], QM["rot"])}
KCLS_OF = {k: ("rot" if v else "iso") for k, v in KROT.items()}      # THE lookup every later part joins on


def kcls_of(box, snaps, gal_ids):
    """The class of a set of rows, by (box, snap, gal_id); None where the galaxy is not in Part 0's sample."""
    return np.array([KCLS_OF.get((box, int(s), int(g))) for s, g in zip(np.asarray(snaps), np.asarray(gal_ids))], dtype=object)


def _cv2(X, y, k=P2_CVK, seed=0):
    """The out-of-fold AUC of a hand-rolled logistic fit (no sklearn on the cluster kernel): what a rule built on these columns
    scores on galaxies it has not seen. Columns are rank-transformed, so a single galaxy at z = 130 cannot drive the fit."""
    X = np.asarray(X, float); y = np.asarray(y, float)
    ok = np.isfinite(X).all(1) & np.isfinite(y)
    X, y = X[ok], y[ok]
    if len(y) < 40 or y.sum() < 5 or (1 - y).sum() < 5:
        return np.nan
    X = np.column_stack([np.argsort(np.argsort(c)) / max(len(c) - 1, 1) for c in X.T])
    X = np.column_stack([np.ones(len(X)), (X - X.mean(0)) / np.where(X.std(0) > 0, X.std(0), 1)])
    rng = np.random.default_rng(seed)                       # STRATIFIED folds: with a dusty fraction of 7 % an unstratified
    _p, _n = rng.permutation(np.flatnonzero(y > 0.5)), rng.permutation(np.flatnonzero(y <= 0.5))   # fold can hold no positive
    folds = [np.concatenate([a_, b_]) for a_, b_ in zip(np.array_split(_p, k), np.array_split(_n, k))]
    idx = np.concatenate(folds)
    p = np.full(len(X), np.nan)
    for f in folds:
        tr = np.setdiff1d(idx, f); w = np.zeros(X.shape[1])
        for _ in range(60):
            mu = 1.0 / (1.0 + np.exp(-np.clip(X[tr] @ w, -30, 30))); W = np.clip(mu * (1 - mu), 1e-6, None)
            H = X[tr].T @ (X[tr] * W[:, None]) + 1e-3 * np.eye(X.shape[1])
            w = w + np.linalg.solve(H, X[tr].T @ (y[tr] - mu))
        p[f] = X[f] @ w
    return mw_auc(p, y.astype(bool))[0]


from math import erf, sqrt as _sqrt
_null_above = float(0.5 * (1.0 - erf(P2_SIG / _sqrt(2.0))))  # the share of an isotropic population the cut would still call rotating

print("=" * 122)
print(f"Part 2 — THE SELECTION: selected <=> {KAPPA_COL} > 1/3 + {P2_SIG:g} s_iso({NGAS_COL}), s_iso(N) = {KAPPA_C:.3f}/sqrt(N)"
      + (f"  AND  log M* > {P2_MASS:g}" if P2_MASS is not None else "   (the mass axis is off)"))
print(f"          an isotropic population would pass that cut {100 * _null_above:.1f} % of the time")
print("=" * 122)
S2, ROWS2 = {}, []
for b in BOXES:
    g = QM[QM["box"] == b]
    if len(g) < 10:
        print(f"  {b}: {len(g)} rows — skipped"); continue
    k, n, z = g[KAPPA_COL].to_numpy(float), g[NGAS_COL].to_numpy(float), g["kz"].to_numpy(float)
    y, rot = (g["pop"] == "dusty").to_numpy(bool), g["rot"].to_numpy(bool)
    s = g["s_iso"].to_numpy(float)
    S2[b] = dict(g=g, k=k, n=n, z=z, y=y, rot=rot, s=s)
    base = float(np.mean(y))
    prec = float(np.mean(y[rot])) if rot.any() else np.nan
    rec  = float(np.sum(y & rot) / max(int(y.sum()), 1))
    print(f"\n  {'=' * 118}\n  {BOX_LABEL[b]}: {len(g)} quenched galaxies, N_gas median {np.median(n):.0f} "
          f"[16-84 {np.percentile(n, 16):.0f}-{np.percentile(n, 84):.0f}]  ->  s_iso median {np.median(s):.3f}, "
          f"cut at kappa = {np.median(kappa_cut(n, P2_SIG)):.3f} (median)")
    print(f"    (a) THE NULL   sd(kappa) = {np.nanstd(k):.3f} against the {np.mean(s):.3f} the shot noise allows "
          f"({np.nanstd(k) / np.mean(s):.1f}x) — the spread of the axis is intrinsic, not sampling;  "
          f"sd(z) = {np.nanstd(z):.1f} against 1")
    print(f"                   above the cut: {100 * np.mean(rot):.1f} % of the sample against the {100 * _null_above:.1f} % an "
          f"isotropic population would give;  {100 * np.mean(k > P2_HIROT):.1f} % are above kappa = {P2_HIROT:g} outright")
    print(f"    (b) THE CLASSES  {KCLS['rot']['short']} {int(rot.sum())} ({100 * np.mean(rot):.0f} %), "
          f"{KCLS['iso']['short']} {int((~rot).sum())}")
    print(f"    (c) AGAINST THE DUST LABEL   base rate {100 * base:.1f} % dusty;  dusty | selected {100 * prec:.1f} %, "
          f"dusty | not {100 * np.mean(y[~rot]):.1f} %  ->  enrichment x{prec / max(base, 1e-9):.1f}, completeness "
          f"{100 * rec:.0f} % of all dusty galaxies")
    _lm2 = g["log_mstar"].to_numpy(float)
    _a = {"significance z": z, f"{KAPPA_COL} (raw)": k, "log N_gas (the control)": g["lngas"].to_numpy(float),
          "log M* (the other axis)": _lm2}
    print("    (d) AUC toward the dust label: " + ",  ".join(f"{lab} {mw_auc(v, y)[0]:.3f}" for lab, v in _a.items()))
    _oz, _om, _ozm = _cv2(z[:, None], y), _cv2(_lm2[:, None], y), _cv2(np.column_stack([z, _lm2]), y)
    print(f"        out of fold ({P2_CVK}-fold, ranks): z {_oz:.3f},  log M* {_om:.3f},  z + log M* {_ozm:.3f}"
          f"   -> THE MASS CHANNEL IS WORTH {_ozm - _oz:+.3f} HERE")
    S2[b].update(oof_z=_oz, oof_m=_om, oof_zm=_ozm)
    ROWS2.append(dict(box=b, rule="out-of-fold z", n_sel=len(z), precision=_oz, recall=np.nan, base=base, n=len(g)))
    ROWS2.append(dict(box=b, rule="out-of-fold z + log M*", n_sel=len(z), precision=_ozm, recall=np.nan, base=base, n=len(g)))
    # the same rule written other ways
    _kc, _fix = float(np.nanmedian(k)), float(np.nanmedian(kappa_cut(n, P2_SIG)))
    _alt2 = [(f"kappa > {_kc:.2f} (the median split)", k > _kc),
             (f"kappa > {_fix:.2f} (a FIXED cut at the same median level)", k > _fix),
             (f"kappa > {P2_HIROT:g} (rotation-supported)", k > P2_HIROT)]
    if P2_MASS is not None:                     # what the rule would be with the mass axis dropped: the cost of the second cut
        _alt2.append((f"kappa > 1/3 + {P2_SIG:g} s_iso(N)  ALONE (no mass axis)", z > P2_SIG))
    _alt2.append((f"kappa > 1/3 + {P2_SIG:g} s_iso(N)"
                  + (f" AND log M* > {P2_MASS:g}" if P2_MASS is not None else "") + "   <- THE RULE", rot))
    for _lab, _m in _alt2:
        print(f"        {_lab:52s} N {int(_m.sum()):5d}   dusty | in {100 * np.mean(y[_m]) if _m.any() else np.nan:5.1f} %   "
              f"completeness {100 * np.sum(y & _m) / max(int(y.sum()), 1):5.1f} %   "
              + ("" if _m is rot else f"changes side for {int((_m != rot).sum())} galaxies"))
        ROWS2.append(dict(box=b, rule=_lab, n_sel=int(_m.sum()), precision=float(np.mean(y[_m])) if _m.any() else np.nan,
                          recall=float(np.sum(y & _m) / max(int(y.sum()), 1)), base=base, n=len(g)))
    if P2_MASS is not None:                     # THE TRADE the second cut makes, box by box: purity bought with completeness
        _k_only = z > P2_SIG
        _pk = float(np.mean(y[_k_only])) if _k_only.any() else np.nan
        print(f"        THE SECOND CUT, what it costs and buys here: log M* > {P2_MASS:g} takes the selected group from "
              f"{int(_k_only.sum())} to {int(rot.sum())} galaxies, purity {100 * _pk:.1f} -> {100 * prec:.1f} %, completeness "
              f"{100 * np.sum(y & _k_only) / max(int(y.sum()), 1):.0f} -> {100 * rec:.0f} %"
              + ("   (in the box that RESOLVES the gas it buys nothing — see the out-of-fold line above)" if b == "cis25" else ""))

R2 = pd.DataFrame(ROWS2)
R2.to_csv(os.path.join(OUT, "kappa_selection_rules.csv"), index=False)
QM[KEY + ["z", "log_mstar", "age", "lfd", "pop", KAPPA_COL, NGAS_COL, "s_iso", "kz", "k_cut", "rot", "kcls"]].to_csv(
    os.path.join(OUT, "kappa_selection_galaxies.csv"), index=False)
print(f"\n  the rules -> {os.path.join(OUT, 'kappa_selection_rules.csv')};  the per-galaxy classes -> "
      f"{os.path.join(OUT, 'kappa_selection_galaxies.csv')}")

# ── the figure ──
_BX2 = [b for b in BOXES if b in S2]
if _BX2:
    fig, AX = plt.subplots(4, len(_BX2), figsize=(4.9 * len(_BX2), 15.2), squeeze=False)
    for j, b in enumerate(_BX2):
        d = S2[b]; k, n, z, y, rot = d["k"], d["n"], d["z"], d["y"], d["rot"]
        # (a) the plane
        ax = AX[0, j]
        _ng = np.logspace(np.log10(max(np.nanmin(n), 10)), np.log10(np.nanmax(n) * 1.05), 200)
        ax.fill_between(_ng, KAPPA_ISO - sig_iso(_ng), KAPPA_ISO + sig_iso(_ng), color="0.55", alpha=0.22, lw=0,
                        label=r"isotropic $1/3 \pm s_{\rm iso}(N)$")
        ax.axhline(KAPPA_ISO, color="0.3", lw=1.2, ls=":")
        ax.plot(_ng, kappa_cut(_ng, P2_SIG), "k--", lw=1.8, zorder=6, label=rf"$1/3 + {P2_SIG:g}\,s_{{\rm iso}}(N)$ — THE CUT")
        ax.scatter(n[~y], k[~y], s=7, c=POPS["pure"]["color"], alpha=0.35, lw=0, zorder=3)
        ax.scatter(n[y], k[y], s=22, c=POPS["dusty"]["color"], alpha=0.9, lw=0.4, edgecolors="k", zorder=5)
        ax.set_xscale("log"); ax.set_xlabel(r"$N_{\rm gas}$  (member gas particles)")
        ax.set_ylim(0, 1.02)
        if j == 0:
            ax.set_ylabel(r"$\kappa_{\rm rot}$(gas)")
        ax.set_title(f"(a) {BOX_LABEL[b]}\n{int(rot.sum())} of {len(k)} above the cut", fontsize=10)
        ax.legend(frameon=False, fontsize=8, loc="lower right"); ax.grid(alpha=0.2, lw=0.5)
        # (b) the null
        ax = AX[1, j]
        _zc = np.clip(z, *P2_ZLIM)
        _bins = np.linspace(P2_ZLIM[0], P2_ZLIM[1], P2_NBINZ + 1)
        ax.hist(_zc, bins=_bins, color="0.8", edgecolor="0.4", lw=0.6, label=f"all ({len(z)})")
        ax.hist(_zc[y], bins=_bins, color=POPS["dusty"]["color"], alpha=0.7, lw=0, label=f"dusty ({int(y.sum())})")
        _w = _bins[1] - _bins[0]
        _xg = np.linspace(P2_ZLIM[0], P2_ZLIM[1], 400)
        _null = len(z) * _w * np.exp(-0.5 * _xg ** 2) / np.sqrt(2 * np.pi)
        ax.plot(_xg[_null >= 0.5], _null[_null >= 0.5], "-", color="#2ca02c", lw=2.2, label="the isotropic null, N(0,1)")
        ax.axvline(P2_SIG, color="k", lw=1.8, ls="--")
        ax.set_yscale("log"); ax.set_ylim(0.5, 4.0 * max(np.histogram(_zc, bins=_bins)[0].max(), 1))
        ax.set_xlabel(r"$z = (\kappa_{\rm gas} - 1/3)\,/\,s_{\rm iso}(N_{\rm gas})$")
        if j == 0:
            ax.set_ylabel("galaxies")
        ax.set_title(f"(b) the axis is not shot noise\nsd$(z)$ = {np.nanstd(z):.1f}, not 1", fontsize=10)
        ax.legend(frameon=False, fontsize=8, loc="upper right"); ax.grid(alpha=0.2, lw=0.5)
        # (c) the outcome
        ax = AX[2, j]
        # the bins are equal-count, so the x axis is the bin ITSELF (its median z as the tick): a single galaxy at z = 100
        # would otherwise squash the five bins that carry the trend into the left edge
        _q = np.quantile(z[np.isfinite(z)], np.linspace(0, 1, P2_NBIN + 1)); _q[0] -= 1e-9
        _xs, _ps, _lo, _hi, _ns, _zm = [], [], [], [], [], []
        for i in range(P2_NBIN):
            m = (z > _q[i]) & (z <= _q[i + 1])
            if m.sum() < 5:
                continue
            p, lo, hi = wilson(int(y[m].sum()), int(m.sum()))
            _xs.append(len(_xs)); _ps.append(p); _lo.append(lo); _hi.append(hi); _ns.append(int(m.sum()))
            _zm.append(float(np.median(z[m])))
            ROWS2.append(dict(box=b, rule="fraction bin", n_sel=int(m.sum()), precision=p, recall=float(np.median(z[m])),
                              base=float(np.mean(y)), n=len(z)))
        ax.errorbar(_xs, _ps, yerr=[np.array(_ps) - np.array(_lo), np.array(_hi) - np.array(_ps)], fmt="-",
                    color="#4d004b", lw=2.0, capsize=3, zorder=3)
        for x_, p_, n_, z_ in zip(_xs, _ps, _ns, _zm):
            ax.plot([x_], [p_], "o", ms=8, color="#4d004b", mfc="#4d004b" if z_ > P2_SIG else "w", mew=1.6, zorder=4)
            ax.annotate(str(n_), (x_, p_), textcoords="offset points", xytext=(0, 9), ha="center", fontsize=7.5, color="0.35")
        ax.axhline(float(np.mean(y)), color="0.45", lw=1.2, ls=":", label="the box's base rate")
        ax.set_xticks(_xs); ax.set_xticklabels([f"{v:.0f}" for v in _zm])
        ax.set_xlabel(r"$z$ at the bin centre"); ax.set_xlim(-0.5, len(_xs) - 0.5)
        ax.set_ylim(0, max(_hi) * 1.22)
        if j == 0:
            ax.set_ylabel("dusty fraction")
        ax.set_title("(c) what the axis predicts\nequal-count bins, filled $=$ above the cut", fontsize=10)
        ax.legend(frameon=False, fontsize=8, loc="upper left"); ax.grid(alpha=0.2, lw=0.5)
        # (d) the cut on trial
        ax = AX[3, j]
        _pr, _re, _fr = [], [], []
        for s_ in P2_SCAN:
            m = z > s_
            _pr.append(float(np.mean(y[m])) if m.sum() >= 5 else np.nan)
            _re.append(float(np.sum(y & m) / max(int(y.sum()), 1)))
            _fr.append(float(np.mean(m)))
        ax.plot(P2_SCAN, _pr, "-", color="#b2182b", lw=2.4, label="dusty fraction of the selected (purity)")
        ax.plot(P2_SCAN, _re, "-", color="#2166ac", lw=2.4, label="share of all dusty selected (completeness)")
        ax.plot(P2_SCAN, _fr, "--", color="0.45", lw=1.6, label="share of the sample selected")
        ax.axvline(P2_SIG, color="k", lw=1.8, ls="--")
        ax.axhline(float(np.mean(y)), color="0.45", lw=1.0, ls=":")
        ax.set_xlabel(r"the threshold, in units of $s_{\rm iso}(N)$"); ax.set_ylim(0, 1.02)
        if j == 0:
            ax.set_ylabel("fraction")
            ax.legend(frameon=False, fontsize=8, loc="center right")
        ax.set_title("(d) the cut on trial", fontsize=10); ax.grid(alpha=0.2, lw=0.5)
    fig.suptitle(r"THE SELECTION: $\kappa_{\rm rot}$(gas) measured against isotropy, with $N_{\rm gas}$ as the uncertainty — "
                 r"the dust label is only ever the OUTCOME", y=0.997, fontsize=12)
    fig.tight_layout(rect=(0, 0, 1, 0.985))
    paper_save(fig, "kappa_selection")
    plt.show()

# ── THE SECOND AXIS: what the stellar mass adds, drawn rather than argued ──
# The old notebook's Part 32 found kappa_gas AND log M* the strongest pair, and Part 35 explained why: the two are orthogonal
# (rho ~ 0) and act on the same term (the share of the metals locked in grains) independently. This figure is that pair, with
# the selection's own cut on one axis and the box's median mass on the other, so the four quadrants can be read off directly.
# In m25 — the box that RESOLVES the gas — the mass axis adds nothing (Part 35: there the physical axis is Sigma_gas, for which
# M* is the coarse boxes' stand-in). It is in the rule because of what it does in the two COARSE boxes, and the m25 column of
# this figure is the honest picture of its cost there. The horizontal line is THE CUT (P2_MASS), the box median beside it.
if P2_MASSFIG and _BX2:
    fig, AX = plt.subplots(1, len(_BX2), figsize=(4.9 * len(_BX2), 4.5), squeeze=False)
    for j, b in enumerate(_BX2):
        d = S2[b]; z, y = np.clip(d["z"], *P2_ZLIM), d["y"]
        lm = d["g"]["log_mstar"].to_numpy(float)
        _med = float(np.nanmedian(lm))
        mc = _med if P2_MASS is None else float(P2_MASS)        # the quadrants are cut where THE RULE cuts
        ax = AX[0, j]
        ax.scatter(z[~y], lm[~y], s=7, c=POPS["pure"]["color"], alpha=0.30, lw=0)
        ax.scatter(z[y], lm[y], s=26, c=POPS["dusty"]["color"], alpha=0.9, lw=0.4, edgecolors="k")
        ax.axvline(P2_SIG, color="k", lw=1.8, ls="--"); ax.axhline(mc, color="k", lw=1.8, ls="--")
        if P2_MASS is not None:
            ax.axhline(_med, color="0.45", lw=1.0, ls=":")      # the box median, for scale: the cut is not a per-box quantile
        ax.add_patch(plt.Rectangle((P2_SIG, mc), 1e3, 1e3, transform=ax.transData, facecolor=KCLS["rot"]["color"],
                                   alpha=0.07, lw=0, zorder=0))
        _q = [(z <= P2_SIG, lm <= mc, 0.02, 0.03, "left", "bottom"), (z > P2_SIG, lm <= mc, 0.98, 0.03, "right", "bottom"),
              (z <= P2_SIG, lm > mc, 0.02, 0.97, "left", "top"), (z > P2_SIG, lm > mc, 0.98, 0.97, "right", "top")]
        for _mz, _mm, _x, _yq, _ha, _va in _q:
            m = _mz & _mm
            if m.sum() < 3:
                continue
            ax.text(_x, _yq, f"{100 * np.mean(y[m]):.0f} % dusty\n({int(m.sum())})", transform=ax.transAxes, ha=_ha, va=_va,
                    fontsize=9, fontweight="bold",
                    bbox=dict(boxstyle="round,pad=0.25", fc="w", ec="0.6", alpha=0.85))
            ROWS2.append(dict(box=b, rule=f"quadrant z{'>' if _ha == 'right' else '<='}{P2_SIG:g}, "
                                          f"log M*{'>' if _va == 'top' else '<='}{mc:.2f}",
                              n_sel=int(m.sum()), precision=float(np.mean(y[m])), recall=np.nan, base=float(np.mean(y)), n=len(y)))
        ax.set_xlabel(r"$z = (\kappa_{\rm gas} - 1/3)\,/\,s_{\rm iso}(N_{\rm gas})$")
        if j == 0:
            ax.set_ylabel(r"$\log M_\star$  [$M_\odot$]")
        ax.set_title(f"{BOX_LABEL[b]}\nout of fold: $z$ {d.get('oof_z', np.nan):.3f}, $+\,\log M_\star$ "
                     f"{d.get('oof_zm', np.nan):.3f}", fontsize=10)
        ax.set_ylim(np.nanpercentile(lm, 0.2) - 0.05, np.nanpercentile(lm, 99.8) + 0.05)
        ax.grid(alpha=0.2, lw=0.5)
    fig.suptitle((r"THE RULE, both axes: $z > %g$ (vertical) and $\log M_\star > %g$ (horizontal) — the shaded quadrant is the "
                  r"selected class, and the four shares are what the second axis is worth" % (P2_SIG, P2_MASS))
                 if P2_MASS is not None else
                 (r"the second axis: the selection ($z >$ %g, vertical) against the stellar mass (the box median, horizontal) — "
                  r"the quadrant shares are what the mass channel would be worth" % P2_SIG), y=1.005, fontsize=11.5)
    fig.tight_layout(rect=(0, 0, 1, 0.97))
    paper_save(fig, "kappa_selection_mass")
    plt.show()
    R2 = pd.DataFrame(ROWS2)
    R2.to_csv(os.path.join(OUT, "kappa_selection_rules.csv"), index=False)
```

## Part 3 — the branch pass, and the dust clock of the two classes

One pass over the main-branch chains of every box builds everything the rest of the notebook needs: the per-galaxy **tracks**
interpolated onto two clocks — $t - t_{\rm SFT}$ (the star-formation truncation, from `paper_ism_prediction_boxes` Part 6a)
and $t - t_{x1}$ (the first firing of SIMBA's X-ray channel on that branch) — and the per-galaxy **scalars**: the AGN clocks,
the dose, and the post-truncation dust budget. The channel-off box `cis50nox` is carried through the same pass so that Part 4
has its control.

**The figure is the notebook's central one.** On the left, $\log M_{\rm dust}/M_\star$ on the truncation clock, one median
track per class with its bootstrap band, the individual chains behind it and the per-epoch source counts underneath. On the
right, on the same axis, **the same quantity at the anchor** — what a telescope would measure — as a distribution per class,
with the dust label's cut drawn across both panels.

**How to read it.** The two classes lie on top of each other for the whole 3 Gyr *before* the truncation: the rotating
galaxies are not simply the ones that were dustier to begin with. They separate afterwards, and the separation survives to
the anchor, where it is a ~1 dex offset in the median and a factor 9 in the dusty fraction (Part 2).

**Two conventions worth knowing.** (i) The dust ratio along the branch is **floored** at $10^{\texttt{P9\_FLOOR}}$, exactly as
Part 0 floors the anchor value: a step with no dust left is the end point of the removal, not a missing measurement. (ii) The
classes are defined **at the anchor** and the tracks are walked backwards from there, so a track is the history of the
galaxies that *ended* in that class, not the path of a fixed population — the same caveat that applies to every anchor-defined
track in this repository.

`P3_QTY` redraws the figure for any of the tracked quantities (the dust-to-metal ratio, $f_{\rm gas}$, $f_{\rm H_2}$, sSFR).

```python
# ── Part 3 — the branch pass, and THE DUST CLOCK of the two classes: log M_dust/M* against t - t_SFT, box by box ──
# ONE pass over the main-branch chains of every box, which is all any later part needs:
#   (i)  the TRACKS of each galaxy, interpolated onto the clock twice — on t - t_SFT (the star-formation truncation) and on
#        t - t_x1 (the first firing of SIMBA's X-ray channel on that branch) — for log M_dust/M*, the dust-to-metal ratio,
#        log f_gas, log 1.36 M_H2/M*, log sSFR, log Z_gas, log M* and the X-ray occupancy itself;
#   (ii) the per-galaxy SCALARS: the AGN clocks (t_x1, t_gate, t_big, t_jet, t_quiet), the dose (E_x, f_x, f_x_post) and the
#        post-truncation dust budget (the pre-SFT level, the minimum after it, the last value: the NET loss and the REBOUND).
# The dust-to-metal ratio is the bounded metal-share-in-grains form M_dust/(M_dust + Z M_gas): it needs no assumption about
# whether caesar's zgas already carries the grain metals, and it cannot exceed 1.
# THE FIGURE is the one the classes are for: the clock on the left, and on the right, for each box, the distribution of the
# SAME quantity AT THE ANCHOR — what a telescope would measure — on the same axis, with the dust label's cut drawn across both.
import time
P3_BOXES  = None            # None = every box with a histories cache; the channel-off box is added automatically for Part 4
P3_QTY    = "lfd"           # the quantity drawn: "lfd" | "ldtm" | "lfh2" | "lfgas" | "lssfr"
P3_SPAG   = True            # the individual galaxy tracks behind the medians
P3_NMIN   = NMIN            # an epoch is drawn only where this many galaxies contribute
P3_PRE    = (-1.0, -0.1)    # the window (relative to t_SFT) the "before" level of the dust budget is the median of
P3_POSTMIN = 1.0            # a galaxy needs this much branch after t_SFT for its net / rebound to be counted [Gyr]
P3_HBINS  = np.arange(P9_FLOOR - 0.125, -1.24, 0.25)     # the anchor histogram of the right-hand column
P3_TRK    = ["lfd", "ldtm", "lz", "lfgas", "lfh2", "lssfr", "lms", "xr", "fgas", "kappa_gas", "comp", "lr_gas", "lr_star",
             "lsig_star", "lsig_gas", "lsig_h2"]
# THE SURFACE DENSITIES (Part 4b) are built here because they need the branch masses AND the branch radii together:
#   Sigma_X(t) = 0.5 M_X(t) / (pi R_50,X(t)^2),  the half-mass convention Part 0 uses for the catalogue's own lsig — half the
# mass inside the half-mass radius. The stars use R_50(*), the gas and the H2 both use R_50(gas): caesar carries no H2 radius,
# so Sigma_H2 is the H2 mass spread over the COLD-GAS half-mass radius and is a surface density of H2 only to the extent that
# the H2 follows the gas. It is drawn as an ISM density, never as a resolved measurement — Parts 5 and 7 do that from particles.
# THE KINEMATICS ALONG THE BRANCH are not in the boxes notebook's histories (they carry masses only), so they come from a
# CACHED SIDECAR walked out of the caesar catalogues themselves: the same main branch (tree_data/progen_galaxy_star, most
# massive progenitor, the histories' own snapshot ladder), one file open per snapshot, kappa_rot(gas), kappa_rot(*), the two
# half-mass radii, M_HI and N_gas. It is the SAME product Part 36 of agn_states_dusty_vs_pure builds, and the default path is
# that notebook's, so an existing build is reused and only the missing boxes are walked. Delete a file to rebuild it.
P3_KIN_FILE = os.path.join(os.getcwd(), "output", "box_resolution", "agn_states", "agn_states_kin_histories_{box}.hdf5")
P3_KIN_BUILD = True         # walk the catalogues where the sidecar is missing (False = draw only what is already cached)
P3_CATDIR  = globals().get("P29_CATDIR", {})    # {"cis50": "/path/to/Groups"} overrides of the config's catalog_dir
P3_KIN_COLS = {"kappa_gas": "galaxy_data/dicts/rotation.gas_kappa_rot", "kappa_star": "galaxy_data/dicts/rotation.stellar_kappa_rot",
               "r_gas": "galaxy_data/dicts/radii.gas_half_mass", "r_star": "galaxy_data/dicts/radii.stellar_half_mass",
               "mhi": "galaxy_data/dicts/masses.HI", "ngas": "galaxy_data/ngas"}
P3_LAB    = {"kappa_gas": r"$\kappa_{\rm rot}$(gas)", "comp": r"$-\log\,R_{50}({\rm gas})/R_{50}(\star)$",
             "lr_gas": r"$\log R_{50}({\rm gas})$  [kpc]", "lfd": r"$\log M_{\rm dust}/M_\star$", "ldtm": r"$\log M_{\rm dust}/(M_{\rm dust} + Z M_{\rm gas})$",
             "lz": r"$\log Z_{\rm gas}$", "lfgas": r"$\log M_{\rm gas}/M_\star$", "lfh2": r"$\log 1.36\,M_{\rm H_2}/M_\star$",
             "lssfr": r"$\log$ sSFR [yr$^{-1}$]", "lms": r"$\log M_\star$", "xr": "X-ray channel on",
             "fgas": r"$M_{\rm gas}/M_\star$", "lr_star": r"$\log R_{50}(\star)$  [kpc]",
             "lsig_star": r"$\log \Sigma_\star$  [$M_\odot$ kpc$^{-2}$]",
             "lsig_gas": r"$\log \Sigma_{\rm gas}$  [$M_\odot$ kpc$^{-2}$]",
             "lsig_h2": r"$\log \Sigma_{\rm H_2}$  [$M_\odot$ kpc$^{-2}$]"}

_P3B = [b for b in (BOXES_OK if P3_BOXES is None else P3_BOXES)]
if os.path.exists(HIST_FILE.format(box=NOX_BOX)) and NOX_BOX not in _P3B:
    _P3B.append(NOX_BOX)                      # the counterfactual box: carried through the same pass, drawn only in Part 4


def _catcfg3(box):
    """(catalog_dir, file_format) of a box: P3_CATDIR wins, then ~/.simbanator/config.json, then the first CANDIDATE that
    exists on disk — the share's s50 catalogues moved out of .../SIMBA_50/s50/Groups into .../SIMBA_50/Groups while the config
    kept the old path, so the data_dir's and the grandparent's Groups are tried too."""
    import json as _json3
    try:
        cfg = _json3.load(open(os.path.expanduser("~/.simbanator/config.json"))); ent = (cfg.get("simulations", cfg)).get(box, {})
    except Exception:
        ent = {}
    _stem = {"cis100": "m100n1024", "cis50": "m50n512", "cis50nox": "m50n512", "cis25": "m25n512"}.get(box, box)
    fmt = ent.get("file_format", "%s_{snap:03d}.hdf5" % _stem)
    if box in P3_CATDIR:
        return P3_CATDIR[box], fmt
    _d = ent.get("catalog_dir")
    _cand = [_d, os.path.join(ent.get("data_dir", ""), "Groups") if ent.get("data_dir") else None,
             os.path.join(os.path.dirname(os.path.dirname(_d)), "Groups") if _d else None, os.path.dirname(_d) if _d else None]
    for _c in _cand:
        if _c and os.path.isdir(_c):
            return _c, fmt
    return _d, fmt


def _read3(box, snap):
    """ONE read of one catalogue: the kinematic columns, the scale factor and the most massive progenitor pointer
    (tree_data/progen_galaxy_star, first column), from the catalogue where caesar wrote it and from the progen_links sidecar
    otherwise (the m25 case). Anything missing comes back as None and lands as NaN in the branch."""
    cdir, fmt = _catcfg3(box); fn = fmt.format(snap=int(snap))
    _pg = lambda f: ((lambda a: (a[:, 0] if a.ndim == 2 else a).astype(np.int64))(f["tree_data/progen_galaxy_star"][:])
                     if "tree_data/progen_galaxy_star" in f else None)
    with h5py.File(os.path.join(cdir or "", fn), "r") as f:
        sf = float(f["simulation_attributes"].attrs["scale_factor"])
        cols = {k: (f[d][:].astype(float) if d in f else None) for k, d in P3_KIN_COLS.items()}
        prog = _pg(f)
    if prog is None:
        _q = os.path.join(os.getcwd(), "output", box, "progen_links", fn)
        if os.path.exists(_q):
            with h5py.File(_q, "r") as f:
                prog = _pg(f)
    for k in ("r_gas", "r_star"):                       # caesar radii are comoving kpc
        if cols.get(k) is not None:
            cols[k] = cols[k] * sf
    return cols, prog


def _kinbuild3(box):
    """The kinematic sidecar of one box: one group per anchor, arrays (n_snap, n_gal) built by walking the SAME main branch the
    histories were built from. One catalogue read per snapshot; anchors already in the file are left alone."""
    hp, kp = HIST_FILE.format(box=box), P3_KIN_FILE.format(box=box)
    if not os.path.exists(hp):
        return
    with h5py.File(hp, "r") as f:
        groups = {int(k[4:]): (f[k]["snaps"][:].astype(int), f[k]["gal_ids"][:].astype(np.int64)) for k in f if k.startswith("snap")}
    done = []
    if os.path.exists(kp):
        with h5py.File(kp, "r") as f:
            done = [int(k[4:]) for k in f if k.startswith("snap")]
    todo = {a: v for a, v in groups.items() if a not in done}
    if not todo:
        return
    cdir, _ = _catcfg3(box)
    if not cdir or not os.path.isdir(cdir):
        print(f"  [{box}] no catalogue directory ({cdir}) — the kinematic branch cannot be built (set P3_CATDIR); the kappa "
              "and size tracks stay empty for this box"); return
    os.makedirs(os.path.dirname(kp), exist_ok=True)
    smax, smin = max(todo), min(sn.min() for sn, _ in todo.values())
    print(f"  [{box}] building the kinematic branch of anchors {sorted(todo)}: {smax - smin + 1} catalogues, one read each, "
          f"from {cdir}", flush=True)
    st = {a: dict(idx=ids.astype(float), rec={k: [] for k in P3_KIN_COLS}, snaps=[], ids=ids) for a, (sn, ids) in todo.items()}
    t0, miss = time.time(), []
    for sn_ in range(smax, smin - 1, -1):
        act = [a for a, (sn, _) in todo.items() if a >= sn_ >= sn.min()]
        if not act:
            continue
        try:
            cols, prog = _read3(box, sn_)
        except Exception as e:
            miss.append(f"snap {sn_}: {type(e).__name__}: {e}"); cols, prog = {k: None for k in P3_KIN_COLS}, None
        n = len(cols["kappa_gas"]) if cols.get("kappa_gas") is not None else 0
        for a in act:
            S = st[a]; idx = S["idx"]; ok = np.isfinite(idx) & (idx >= 0) & (idx < n); ii = idx[ok].astype(int)
            for k in P3_KIN_COLS:
                arr = np.full(len(idx), np.nan)
                if cols.get(k) is not None:
                    arr[ok] = cols[k][ii]
                S["rec"][k].append(arr)
            S["snaps"].append(sn_)
            nxt = np.full(len(idx), np.nan)
            if prog is not None:
                q = prog[ii]; nxt[ok] = np.where(q >= 0, q, np.nan)
            S["idx"] = nxt
        if (smax - sn_) % 20 == 19:
            print(f"     snap {sn_:3d}, {time.time() - t0:5.0f} s", flush=True)
    for m in miss[:5]:
        print(f"     catalogue not read, {m}")
    with h5py.File(kp, "a") as f:
        for a, S in st.items():
            g = f.require_group(f"snap{a:03d}")
            for k in list(g.keys()):
                del g[k]
            g["snaps"] = np.array(S["snaps"], np.int32); g["gal_ids"] = S["ids"]
            for k in P3_KIN_COLS:
                g[k] = np.vstack(S["rec"][k])
    _cells = [np.vstack(S["rec"]["kappa_gas"]) for S in st.values()]      # one block per anchor: the anchors have different widths
    _fin, _tot = sum(int(np.isfinite(a).sum()) for a in _cells), sum(int(a.size) for a in _cells)
    print(f"  [{box}] kinematic branch -> {kp}  ({time.time() - t0:.0f} s, {100 * _fin / max(_tot, 1):.0f} % of the cells filled "
          "— a cell is empty where the branch has already ended, exactly as in the histories)")


def _kinread3(box):
    """{anchor: {col: (n_snap, n_gal)}} of the sidecar, keyed the same way as the histories."""
    kp = P3_KIN_FILE.format(box=box)
    if not os.path.exists(kp):
        return {}
    out = {}
    with h5py.File(kp, "r") as f:
        for k in f:
            g = f[k]; out[int(k[4:])] = {c: g[c][:] for c in g.keys()}
    return out


def _pass3(box):
    """One box's chains -> (the per-galaxy frame, the gridded tracks on both clocks). Only Part 0's sample enters."""
    sub = QM[QM["box"] == box]
    want = {(int(s), int(i)): r for (s, i), r in zip(zip(sub["snap"], sub["gal_id"]), sub.to_dict("records"))}
    rows, T = [], {c: {"sft": [], "x1": []} for c in P3_TRK}
    if P3_KIN_BUILD:
        _kinbuild3(box)
    KIN = _kinread3(box)
    with h5py.File(HIST_FILE.format(box=box), "r") as f:
        for gname in f:
            a = int(gname.replace("snap", ""))
            if not any(k[0] == a for k in want):
                continue
            gh = f[gname]
            t = gh["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
            if len(t) < 3:
                continue
            ms, sfr, mg, mbh, fedd = (gh[c][:][o] for c in ("mstar", "sfr", "mgas", "mbh", "fedd"))
            md = gh["mdust"][:][o] if "mdust" in gh else np.full_like(ms, np.nan)
            mh = gh["mh2"][:][o] if "mh2" in gh else np.full_like(ms, np.nan)
            zg = gh["zgas"][:][o] if "zgas" in gh else np.full_like(ms, np.nan)
            S, D, X = agn_states(ms, mg, mbh, fedd)
            dtw = np.gradient(t)
            # the sidecar is matched on the snapshot numbers and the galaxy ids THEMSELVES, never on the array order: a partial
            # build simply leaves those cells NaN instead of misaligning a branch
            _sh = np.asarray(gh["snaps"][:]).astype(int)[o] if "snaps" in gh else None
            _gid_h = np.asarray(gh["gal_ids"][:]).astype(np.int64)
            _K, kin = KIN.get(a), {}
            if _K is not None and _sh is not None:
                _rs = {int(v): i for i, v in enumerate(np.asarray(_K["snaps"]).astype(int))}
                _ri = {int(v): j for j, v in enumerate(np.asarray(_K["gal_ids"]).astype(np.int64))}
                _row = np.array([_rs.get(int(v), -1) for v in _sh]); _col = np.array([_ri.get(int(v), -1) for v in _gid_h])
                _mr, _mc = _row >= 0, _col >= 0
                for _cc in ("kappa_gas", "r_gas", "r_star", "ngas"):
                    if _cc in _K and _mr.any() and _mc.any():
                        _B = np.full((len(_sh), len(_gid_h)), np.nan)
                        _B[np.ix_(_mr, _mc)] = np.asarray(_K[_cc], float)[np.ix_(_row[_mr], _col[_mc])]
                        kin[_cc] = _B
            _nan2 = np.full_like(ms, np.nan)
            _kg, _rg, _rs2 = kin.get("kappa_gas", _nan2), kin.get("r_gas", _nan2), kin.get("r_star", _nan2)
            xray = np.nan_to_num(S["xray"]) > 0.5
            gate = np.nan_to_num(D["gate"]) > 0.5
            big  = np.nan_to_num(D["big"]) > 0.5
            jet  = np.nan_to_num(D["jet"]) > 0.5
            for j, gid in enumerate(gh["gal_ids"][:]):
                key = (a, int(gid))
                if key not in want:
                    continue
                r, ex = want[key], X["exist"][:, j]
                if ex.sum() < 3:
                    continue
                ka = int(np.flatnonzero(ex)[-1]); t_anc = float(t[ka])
                xj = xray[:, j] & ex
                first = lambda m: float(t[m][0]) if m.any() else np.nan
                t_x1, t_sft = first(xj), float(r.get("t_sft", np.nan))
                with np.errstate(all="ignore"):
                    den = np.where(ex, ms[:, j], np.nan)
                    met = zg[:, j] * mg[:, j]                                  # the metal mass in the gas phase
                    # the dust ratios are FLOORED at 10^P9_FLOOR, exactly as Part 0 floors the anchor value: a branch step with
                    # no dust left is the end point of the removal, not a missing measurement, and must not run off to -inf
                    v = {"lfd":   np.log10(np.where(ex, np.clip(md[:, j] / den, 10 ** P9_FLOOR, None), np.nan)),
                         "ldtm":  np.log10(np.where((met > 0) & ex, np.clip(md[:, j] / (md[:, j] + met), 10 ** P9_FLOOR, None), np.nan)),
                         "lz":    np.log10(np.where((zg[:, j] > 0) & ex, zg[:, j], np.nan)),
                         "lfgas": np.log10(np.where((mg[:, j] > 0) & ex, mg[:, j] / den, np.nan)),
                         "lfh2":  np.log10(np.where((mh[:, j] > 0) & ex, HE * mh[:, j] / den, np.nan)),
                         "lssfr": np.log10(np.where(ex, np.clip(np.maximum(sfr[:, j], 0.0) / den, SSFR_FLOOR, None), np.nan)),
                         "lms":   np.log10(np.where(ex, ms[:, j], np.nan)),
                         "xr":    np.where(ex, xj.astype(float), np.nan),
                         "fgas":  np.where(ex, X["fgas"][:, j], np.nan),
                         # the kinematics and the SIZE of the cold gas, from the sidecar (NaN where it was not built)
                         "kappa_gas": np.where(ex, _kg[:, j], np.nan),
                         "comp":  -np.log10(np.where((_rg[:, j] > 0) & (_rs2[:, j] > 0) & ex, _rg[:, j] / _rs2[:, j], np.nan)),
                         "lr_gas": np.log10(np.where((_rg[:, j] > 0) & ex, _rg[:, j], np.nan)),
                         "lr_star": np.log10(np.where((_rs2[:, j] > 0) & ex, _rs2[:, j], np.nan)),
                         # the surface densities: half the mass inside the half-mass radius, the same convention as Part 0's lsig
                         "lsig_star": np.log10(np.where((_rs2[:, j] > 0) & ex, 0.5 * ms[:, j] / (np.pi * _rs2[:, j] ** 2), np.nan)),
                         "lsig_gas": np.log10(np.where((_rg[:, j] > 0) & (mg[:, j] > 0) & ex,
                                                       0.5 * mg[:, j] / (np.pi * _rg[:, j] ** 2), np.nan)),
                         "lsig_h2": np.log10(np.where((_rg[:, j] > 0) & (mh[:, j] > 0) & ex,
                                                      0.5 * HE * mh[:, j] / (np.pi * _rg[:, j] ** 2), np.nan))}
                for c in P3_TRK:
                    T[c]["sft"].append(to_grid(t - t_sft, v[c]) if np.isfinite(t_sft) else np.full(len(TG), np.nan))
                    T[c]["x1"].append(to_grid(t - t_x1, v[c]) if np.isfinite(t_x1) else np.full(len(TG), np.nan))
                rec = dict(box=box, snap=a, gal_id=int(gid), z=float(r["z"]), pop=r["pop"], lfd=float(r["lfd"]),
                           lfgas=float(r["lfgas"]), lfh2=float(r["lfh2"]), lssfr=float(r["lssfr"]), age=float(r["age"]),
                           log_mstar=float(r["log_mstar"]), log_mbh=float(r["log_mbh"]), kappa=float(r[KAPPA_COL]),
                           ngas=float(r[NGAS_COL]), kz=float(r["kz"]), rot=bool(r["rot"]), kcls=str(r["kcls"]),
                           central=float(r["central"]) if "central" in r else np.nan,
                           t_anchor=t_anc, t_sft=t_sft, t_qt=float(r.get("t_qt", np.nan)), t_span=t_anc - float(t[ex][0]),
                           t_x1=t_x1, t_gate=first(gate[:, j] & ex), t_big=first(big[:, j] & ex), t_jet=first(jet[:, j] & ex),
                           t_quiet=t_anc - float(t[xj][-1]) if xj.any() else np.nan,
                           e_x=float(np.sum(dtw[xj])), n_x=int(xj.sum()), f_x=float(np.sum(dtw[xj]) / np.sum(dtw[ex])),
                           zgas_anchor=float(zg[ka, j]) if np.isfinite(zg[ka, j]) else np.nan,
                           lz_anchor=float(np.log10(zg[ka, j])) if np.isfinite(zg[ka, j]) and zg[ka, j] > 0 else np.nan)
                post = ex & (t >= t_sft) if np.isfinite(t_sft) else np.zeros(len(t), bool)
                rec["f_x_post"] = float(np.sum(dtw[xj & post]) / np.sum(dtw[post])) if post.any() else np.nan
                rec["t_post"] = float(t_anc - t_sft) if np.isfinite(t_sft) else np.nan
                for c in ("lfd", "ldtm"):                              # the post-truncation dust budget, on the native sampling
                    pre = ex & np.isfinite(t_sft) & (t - t_sft >= P3_PRE[0]) & (t - t_sft <= P3_PRE[1])
                    y = v[c]
                    b0 = float(np.nanmedian(y[pre])) if pre.any() and np.isfinite(y[pre]).any() else np.nan
                    yp = y[post] if post.any() else np.array([])
                    yp = yp[np.isfinite(yp)]
                    rec[f"{c}_pre"] = b0
                    rec[f"{c}_min"] = float(np.min(yp)) if len(yp) else np.nan
                    rec[f"{c}_end"] = float(yp[-1]) if len(yp) else np.nan
                    rec[f"{c}_net"] = rec[f"{c}_end"] - b0
                    rec[f"{c}_reb"] = rec[f"{c}_end"] - rec[f"{c}_min"] if len(yp) else np.nan
                rows.append(rec)
    F = pd.DataFrame(rows)
    T = {c: {k: np.asarray(v, float).reshape(len(F), len(TG)) for k, v in d.items()} for c, d in T.items()}
    return F, T


H, _t0 = {}, None
for _b in _P3B:
    _t0 = time.time()
    _F, _T = _pass3(_b)
    if not len(_F):
        print(f"  [{_b}] no chain of Part 0's sample is in the histories cache — skipped"); continue
    H[_b] = dict(F=_F, T=_T)
    _nx = int(np.isfinite(_F["t_x1"]).sum())
    _nk = int(np.isfinite(_T["kappa_gas"]["sft"]).any(axis=1).sum())
    print(f"  [{_b}] {len(_F)} chains ({time.time() - _t0:.0f} s): {int(_F['rot'].sum())} rotating / {int((~_F['rot']).sum())} not, "
          f"{int(np.isfinite(_F['t_sft']).sum())} with an SFT, {_nx} with a first X-ray firing"
          + (f", {len(_F) - _nx} never fire the channel" if len(_F) - _nx else "")
          + f";  {_nk} of them ({100 * _nk / max(len(_F), 1):.0f} %) carry the kinematic branch")
FR3 = pd.concat([H[b]["F"] for b in H], ignore_index=True) if H else pd.DataFrame()
if len(FR3):
    FR3.to_csv(os.path.join(OUT, "kappa_branch_galaxies.csv"), index=False)
    print(f"  the per-galaxy branch record -> {os.path.join(OUT, 'kappa_branch_galaxies.csv')}")

# ── the figure: the clock, and the anchor distribution beside it ──
_BX3 = [b for b in BOXES if b in H]
if _BX3:
    from matplotlib.gridspec import GridSpec
    _rows3 = []
    fig = plt.figure(figsize=(11.4, 3.5 * len(_BX3)))
    gs = GridSpec(len(_BX3), 2, width_ratios=[3.1, 1.0], wspace=0.04, hspace=0.32, figure=fig)
    for i, b in enumerate(_BX3):
        F, A = H[b]["F"], H[b]["T"][P3_QTY]["sft"]
        ax, axh = fig.add_subplot(gs[i, 0]), None
        axb = ax.twinx(); axb.set_zorder(ax.get_zorder() - 1); ax.patch.set_visible(False); axb.set_yticks([])
        _top = 1.0
        for c in KC:
            m = (F["kcls"] == c).to_numpy(bool)
            if m.sum() < 2:
                continue
            if P3_SPAG:
                _al = float(np.clip(25.0 / max(int(m.sum()), 1), 0.03, 0.30))     # a thousand chains at alpha 0.1 is a solid block
                for row in A[m]:
                    ax.plot(TG, row, color=KCLS[c]["color"], lw=0.4, alpha=_al, zorder=2)
            cen, lo, hi, n_t = med_band(A[m], nmin=P3_NMIN)
            _top = max(_top, float(np.nanmax(n_t)) if len(n_t) else 1.0)
            axb.bar(TG, n_t, width=0.9 * (TG[1] - TG[0]), color=KCLS[c]["color"], alpha=0.10, lw=0, zorder=1)
            ax.fill_between(TG, lo, hi, color=KCLS[c]["color"], alpha=0.20, lw=0, zorder=3)
            ax.plot(TG, cen, KCLS[c]["ls"], color=KCLS[c]["color"], lw=KCLS[c]["lw"], zorder=5,
                    label=f"{KCLS[c]['short']}  ({int(m.sum())})")
            for x_, m_, l_, h_, nn in zip(TG, cen, lo, hi, n_t):
                _rows3.append(dict(box=b, block="clock", quantity=P3_QTY, kcls=c, x=float(x_), n_gal=int(nn),
                                   med=float(m_), lo=float(l_), hi=float(h_)))
        axb.set_ylim(0, _top * 3.0)
        ax.axvline(0, color="k", lw=1.4, ls="--", alpha=0.9, zorder=1)
        ax.annotate("SFT", xy=(0, 1.0), xycoords=("data", "axes fraction"), xytext=(3, -12), textcoords="offset points",
                    fontsize=8.5, color="0.3")
        if P3_QTY == "lfd":
            ax.axhline(DUSTY_CUT, color="0.35", lw=1.1, ls=":", zorder=1)
        ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(P3_LAB[P3_QTY]); ax.grid(alpha=0.22, lw=0.5)
        with warnings.catch_warnings():                       # the floor is a real value, so the axis stops there
            warnings.simplefilter("ignore", RuntimeWarning)
            _y0, _y1 = np.nanpercentile(A, 0.5), np.nanpercentile(A, 99.8)
        ax.set_ylim(max(_y0 - 0.3, P9_FLOOR - 0.35), _y1 + 0.35)
        ax.set_title(f"{BOX_LABEL[b]} — the clock of the two classes", fontsize=10.5, loc="left")
        if i == len(_BX3) - 1:
            ax.set_xlabel(r"$t - t_{\rm SFT}$   [Gyr]")
        ax.legend(frameon=False, fontsize=8.6, loc="lower left")
        # the same quantity AT THE ANCHOR, on the same axis
        axh = fig.add_subplot(gs[i, 1], sharey=ax)
        _anc = {"lfd": F["lfd"], "lfgas": F["lfgas"], "lfh2": F["lfh2"], "lssfr": F["lssfr"]}.get(P3_QTY)
        if _anc is None:                                    # a quantity with no catalogue counterpart: take the last branch value
            _anc = pd.Series(np.array([r[np.isfinite(r)][-1] if np.isfinite(r).any() else np.nan for r in A]))
        _anc = _anc.to_numpy(float)
        for c in KC:
            m = (F["kcls"] == c).to_numpy(bool)
            axh.hist(np.clip(_anc[m], P3_HBINS[0], P3_HBINS[-1]), bins=P3_HBINS, orientation="horizontal",
                     color=KCLS[c]["color"], alpha=0.55, lw=0)
            axh.hist(np.clip(_anc[m], P3_HBINS[0], P3_HBINS[-1]), bins=P3_HBINS, orientation="horizontal",
                     histtype="step", color=KCLS[c]["color"], lw=1.6)
            _md = float(np.nanmedian(_anc[m]))
            axh.axhline(_md, color=KCLS[c]["color"], lw=2.0, ls="--", alpha=0.85)
            _rows3.append(dict(box=b, block="anchor", quantity=P3_QTY, kcls=c, x=np.nan, n_gal=int(m.sum()),
                               med=_md, lo=np.nan, hi=np.nan))
        if P3_QTY == "lfd":
            axh.axhline(DUSTY_CUT, color="0.35", lw=1.1, ls=":")
        _auc, _pv, _n1, _n0 = mw_auc(_anc, (F["kcls"] == "rot").to_numpy(bool))
        axh.set_title(f"at the anchor\nAUC {_auc:.2f}, gap "
                      f"{np.nanmedian(_anc[(F['kcls'] == 'rot').to_numpy(bool)]) - np.nanmedian(_anc[(F['kcls'] == 'iso').to_numpy(bool)]):+.2f} dex",
                      fontsize=9.5)
        plt.setp(axh.get_yticklabels(), visible=False)
        axh.set_xlabel("galaxies" if i == len(_BX3) - 1 else "")
        axh.grid(alpha=0.22, lw=0.5)
    fig.suptitle(rf"the dust on the truncation clock: {P3_LAB[P3_QTY]} for the two KINEMATIC classes — identical before the "
                 rf"truncation, separated after it", y=0.995, fontsize=12)
    fig.tight_layout(rect=(0, 0, 1, 0.98))
    paper_save(fig, f"kappa_clock_{P3_QTY}")
    plt.show()
    C3 = pd.DataFrame(_rows3)
    C3.to_csv(os.path.join(OUT, "kappa_clock.csv"), index=False)
    print(f"  the tracks -> {os.path.join(OUT, 'kappa_clock.csv')}")

    # ── what the figure says, in numbers ──
    print(f"\n  {P3_QTY} on the t - t_SFT clock (median of each class) and at the anchor:")
    print(f"    {'box':8s} {'class':10s} " + " ".join(f"{x:+5.1f}" for x in (-2, -1, 0, 1, 2, 3)) + "   anchor   AUC(rot)")
    for b in _BX3:
        F, A = H[b]["F"], H[b]["T"][P3_QTY]["sft"]
        _anc = F[P3_QTY].to_numpy(float) if P3_QTY in F.columns else np.full(len(F), np.nan)
        for c in KC:
            m = (F["kcls"] == c).to_numpy(bool)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", RuntimeWarning)
                _v = [np.nanmedian(A[m][:, int(np.argmin(np.abs(TG - x)))]) for x in (-2, -1, 0, 1, 2, 3)]
            print(f"    {b:8s} {KCLS[c]['short'][:10]:10s} " + " ".join(f"{x:+5.2f}" for x in _v)
                  + f"   {np.nanmedian(_anc[m]):+6.2f}" + (f"   {mw_auc(_anc, (F['kcls'] == 'rot').to_numpy(bool))[0]:.3f}" if c == "rot" else ""))
```

## Part 4 — the channel, the truncation, and where the classes part

Three questions, one row of the figure each, one column per box, **with the channel-off box last as the control**.

**(1) The two clocks.** When SIMBA's X-ray channel first fires on the branch, relative to the truncation — together with the
two other clocks the firing needs: `t_gate`, the first time $f_{\rm gas} <$ `XRAY_FGAS_MAX` (the channel's own gas criterion),
and `t_big`, the first time $M_{\rm BH} > 10^{7.5}$ (its mass criterion).

> **The honest reading, and it is the reason this row draws three clocks and not one.** The firing and the truncation are the
> same event to within a snapshot — median $+0.07$ Gyr, a bit more than half of them inside $\pm0.25$ Gyr. That is a
> coincidence, not a demonstration of cause, because the channel's own gate is a *gas* criterion: 90 % of the galaxies fire
> within one snapshot of the moment $f_{\rm gas}$ drops below the threshold, and the gas dropping below the threshold is
> nearly the same event as the truncation. Meanwhile the mass criterion was met ~1.4 Gyr earlier and nothing happened. So the
> clock says: **the channel waits for the gas to run low**, and the truncation cannot be attributed to it from timing alone.

**(2) What changes at the firing.** The median track of $\log$ sSFR, $\log f_{\rm gas}$, $\log M_{\rm dust}/M_\star$, the
**compactness of the cold gas** $-\log R_{50}({\rm gas})/R_{50}(\star)$ and $\kappa_{\rm rot}$(gas) itself on the $t - t_{x1}$
clock, each galaxy referred to **its own** value at the firing. The sSFR and the gas are already 1.2 dex down when the channel
fires; the dust is flat, and then it collapses by 2 dex. **The channel dates the dust destruction, and only the dust
destruction.** $\kappa_{\rm rot}$ is not a logarithm, so it is drawn against its own right-hand axis, dashed; it and the size
come from Part 3's kinematic sidecar, and a box without one simply shows the three ISM curves and says so. Reading those two
curves is the point of the row: the quantity the notebook *selects* on is itself changing at the firing, which is why the
classes cannot be read as a pre-existing property (Part 5 (A) makes the same point at the critical points).

**(3) The counterfactual.** `s50nox` is the same box with the X-ray heating compiled out and the jets left in, so its `t_x1`
is the *would-be* firing — the moment the same criteria would have been met. Its galaxies still quench on the same clocks, and
their dust does not move: a post-truncation net change of −0.55 dex against −2.7 to −3.7 dex in the fiducial boxes. That is
the closest thing to an experiment this notebook has, and it is what turns the coincidence of row (1) into a statement about
the dust.

**Where the classes part** is printed rather than drawn (the figure is two rows). The per-galaxy post-truncation budget: the
**net** (the end value minus the pre-truncation level) and the **rebound** (the end value minus the minimum reached after the
truncation). A class that merely loses less has a
shallower net and no rebound; a class whose dust comes back has both. Both are true here — the rotating class loses ~1 dex
less *and* about half of it recovers ≥ 0.5 dex from its own minimum, against a fifth of the rest — so the separation is not
only slower destruction, it is **re-growth**. One caveat is printed with the numbers: galaxies that reach the zero-dust floor
and then have dust again have their rebound measured *from* that floor, so its size depends on `P9_FLOOR`; the share and the
floor-free version are printed alongside.

```python
# ── Part 4 — the channel, the truncation, and where the two classes actually part company ──
# Three questions, one row of the figure each, one column per box (the channel-OFF box last, as the counterfactual):
#   (1) THE TWO CLOCKS      t_x1 - t_SFT: when SIMBA's X-ray channel first fires on the branch, against the star-formation
#       truncation. Drawn with the two OTHER clocks the firing needs — t_gate (the first time f_gas < XRAY_FGAS_MAX, the
#       channel's own gas criterion) and t_big (the first time M_BH > 10^7.5, its mass criterion). The ordering of those three is
#       what decides whether the coincidence can be read as causation, and it is printed as well as drawn.
#   (2) WHAT CHANGES AT THE FIRING   the median track of log sSFR, log f_gas and log M_dust/M* on the t - t_x1 clock, each
#       galaxy referred to ITS OWN value at the firing, so the three sit on one axis. The sSFR and the gas are already falling
#       when the channel fires; the dust is flat until that moment and then collapses. The channel DATES the dust destruction.
#       The gas KINEMATICS and the gas SIZE are on the same clock, from Part 3's sidecar: kappa_rot(gas) on a right-hand axis
#       (it is not a log) and the compactness -log R50(gas)/R50(*) with the rest. They answer the question the selection raises
#       — is the rotation the galaxy is selected on something the channel destroys, and does the cold gas shrink when it fires?
#   WHERE THE CLASSES PART is no longer a panel: the post-truncation budget per galaxy — the NET (end minus the pre-SFT level)
#       and the REBOUND (end minus the minimum reached after the truncation) — is PRINTED below with its AUCs and written to
#       the CSV. A class that merely loses less has a shallower net and no rebound; a class whose dust comes back has both.
# THE COUNTERFACTUAL. s50nox is the same box with the X-ray heating compiled out and the jets left in, so its "t_x1" is the
# WOULD-BE firing: the moment the same criteria would have been met. Its galaxies still quench, and the comparison of its dust
# track with the fiducial one is the closest thing to an experiment this notebook has.
P4_BOXES  = None            # None = the boxes of Part 3, with the channel-off box last
P4_DT     = 0.25            # "the same event" tolerance [Gyr] — the snapshot spacing of the coarse boxes
# the quantities of row (2): "L" = the left axis, in dex; "R" = a right-hand axis, for kappa_rot, which is not a log
P4_CLOCKQ = [("lssfr", "#7b3294", r"$\Delta \log$ sSFR", "L"),
             ("lfgas", "#1b9e77", r"$\Delta \log M_{\rm gas}/M_\star$", "L"),
             ("lfd",   "#d7191c", r"$\Delta \log M_{\rm dust}/M_\star$", "L"),
             ("comp",  "#e6ab02", r"$\Delta$ compactness $-\log R_{50}({\rm gas})/R_{50}(\star)$", "L"),
             ("kappa_gas", "k",   r"$\Delta \kappa_{\rm rot}$(gas)   (right axis)", "R")]
P4_KLIM   = None            # the right-hand axis of row (2): (lo, hi) for Delta kappa_rot, or None = symmetric, from the data
P4_BUDGET = "lfd"           # the budget of row 3: "lfd" | "ldtm"
P4_REB    = 0.5             # a REAL rebound: this many dex recovered from the post-truncation minimum
P4_TBINS  = np.arange(-3.0, 3.01, 0.25)

_BX4 = [b for b in (list(H) if P4_BOXES is None else P4_BOXES) if b in H]
_BX4 = [b for b in _BX4 if b != NOX_BOX] + [b for b in _BX4 if b == NOX_BOX]      # the counterfactual last
if not _BX4:
    print("Part 4: Part 3 built no branch records — run it first.")
else:
    R4 = []
    print("=" * 122)
    print("Part 4 — the channel and the truncation.  'the same event' means |Delta t| <= %.2f Gyr (about one snapshot)" % P4_DT)
    print("=" * 122)
    for b in _BX4:
        F = H[b]["F"]
        d = {k: (F[f"t_{k}"] - F["t_sft"]).to_numpy(float) for k in ("x1", "gate", "big", "jet")}
        _f = lambda v: v[np.isfinite(v)]
        print(f"\n  {BOX_LABEL[b]}{'   [the channel is NOT in this run: t_x1 is the WOULD-BE firing]' if b == NOX_BOX else ''}")
        for k, lab in (("x1", "the X-ray channel first fires"), ("gate", f"f_gas first drops below {XRAY_FGAS_MAX:g} (its gas gate)"),
                       ("big", "M_BH first exceeds 10^7.5 (its mass gate)"), ("jet", "the jet criterion is first met")):
            v = _f(d[k])
            if not len(v):
                continue
            print(f"    {lab:48s} t - t_SFT = {np.median(v):+6.2f} Gyr  [16-84 {np.percentile(v, 16):+.2f}, {np.percentile(v, 84):+.2f}]"
                  f"   before the SFT {100 * np.mean(v < 0):3.0f} %,  within {P4_DT:g} Gyr of it {100 * np.mean(np.abs(v) <= P4_DT):3.0f} %")
            R4.append(dict(box=b, block="clocks", label=k, med=float(np.median(v)), lo=float(np.percentile(v, 16)),
                           hi=float(np.percentile(v, 84)), f_before=float(np.mean(v < 0)), f_same=float(np.mean(np.abs(v) <= P4_DT)), n=len(v)))
        _lag = _f((F["t_x1"] - F["t_gate"]).to_numpy(float))
        _wait = _f((F["t_x1"] - F["t_big"]).to_numpy(float))
        if len(_lag):
            print(f"    -> the channel fires {np.median(_lag):+.2f} Gyr after its gas gate opens ({100 * np.mean(_lag <= P4_DT):.0f} % within "
                  f"{P4_DT:g} Gyr) and {np.median(_wait):+.2f} Gyr after its BH is massive enough ({100 * np.mean(_wait <= P4_DT):.0f} %):")
            print(f"       the BINDING CRITERION IS THE GAS, not the black hole — which is why the firing cannot be read off this "
                  f"coincidence as the CAUSE of the truncation (see the markdown).")
            R4.append(dict(box=b, block="lag", label="t_x1 - t_gate", med=float(np.median(_lag)), lo=np.nan, hi=np.nan,
                           f_before=np.nan, f_same=float(np.mean(_lag <= P4_DT)), n=len(_lag)))
        # the budget
        ok = (F["t_post"].to_numpy(float) >= P3_POSTMIN) & np.isfinite(F[f"{P4_BUDGET}_net"].to_numpy(float))
        rot = (F["kcls"] == "rot").to_numpy(bool)
        print(f"    the post-truncation dust budget ({int(ok.sum())} galaxies with >= {P3_POSTMIN:g} Gyr of branch after the SFT):")
        for c in KC:
            m = ok & (rot if c == "rot" else ~rot)
            if m.sum() < 3:
                continue
            _net, _reb = F.loc[m, f"{P4_BUDGET}_net"].to_numpy(float), F.loc[m, f"{P4_BUDGET}_reb"].to_numpy(float)
            _fl = F.loc[m, f"{P4_BUDGET}_min"].to_numpy(float) <= P9_FLOOR + 0.01      # lost ALL of it, then had some again
            print(f"      {KCLS[c]['short']:26s} N {int(m.sum()):5d}   net {np.nanmedian(_net):+6.2f} dex   rebound {np.nanmedian(_reb):+5.2f} dex   "
                  f"with a real rebound (>= {P4_REB:g} dex) {100 * np.nanmean(_reb >= P4_REB):4.0f} %")
            print(f"      {'':26s} {100 * np.mean(_fl):4.0f} % of them hit the zero-dust floor first, so their rebound is measured from "
                  f"log = {P9_FLOOR:g} and its SIZE is set by that floor; without them: net {np.nanmedian(_net[~_fl]):+.2f}, "
                  f"rebound {np.nanmedian(_reb[~_fl]):+.2f}, share {100 * np.nanmean(_reb[~_fl] >= P4_REB):.0f} %")
            R4.append(dict(box=b, block="budget", label=c, med=float(np.nanmedian(_net)), lo=float(np.nanmedian(_reb)),
                           hi=float(np.nanmean(_reb >= P4_REB)), f_before=np.nan, f_same=np.nan, n=int(m.sum())))
        for _q, _nm in ((f"{P4_BUDGET}_net", "net"), (f"{P4_BUDGET}_reb", "rebound")):
            _a, _p, _n1, _n0 = mw_auc(np.where(ok, F[_q].to_numpy(float), np.nan), rot)
            print(f"      AUC(rotating) on the {_nm:8s} {_a:.3f}" + (f"  (p = {_p:.1e})" if np.isfinite(_p) else ""))
            R4.append(dict(box=b, block="auc", label=_nm, med=_a, lo=_p, hi=np.nan, f_before=np.nan, f_same=np.nan, n=int(ok.sum())))

    # ── the figure ──
    fig, AX = plt.subplots(2, len(_BX4), figsize=(4.55 * len(_BX4), 8.6), squeeze=False)
    for j, b in enumerate(_BX4):
        F = H[b]["F"]; _nox = b == NOX_BOX
        # (1) the clocks
        ax = AX[0, j]
        for k, col, lab, ls in (("x1", "#1f77b4", r"first X-ray firing $t_{x1}$", "-"),
                                ("gate", "#2ca02c", rf"gas gate $f_{{\rm gas}} < {XRAY_FGAS_MAX:g}$", "--"),
                                ("big", "#7b3294", r"$M_{\rm BH} > 10^{7.5}$", ":")):
            v = (F[f"t_{k}"] - F["t_sft"]).to_numpy(float)
            v = np.clip(v[np.isfinite(v)], P4_TBINS[0], P4_TBINS[-1])
            if not len(v):
                continue
            ax.hist(v, bins=P4_TBINS, histtype="step", color=col, lw=2.2, ls=ls, label=f"{lab}   ({np.median(v):+.2f})")
        ax.axvspan(-P4_DT, P4_DT, color="0.75", alpha=0.35, lw=0, zorder=0)
        ax.axvline(0, color="k", lw=1.6, ls="--")
        ax.set_xlabel(r"$t - t_{\rm SFT}$   [Gyr]"); ax.set_xlim(P4_TBINS[0], P4_TBINS[-1])
        if j == 0:
            ax.set_ylabel("galaxies")
        _vx = (F["t_x1"] - F["t_sft"]).to_numpy(float); _vx = _vx[np.isfinite(_vx)]
        ax.set_title(f"(1) {BOX_LABEL[b]}\n" + (r"WOULD-BE firing (channel off)" if _nox else
                     f"{100 * np.mean(np.abs(_vx) <= P4_DT):.0f} % fire within {P4_DT:g} Gyr of the truncation"), fontsize=9.5,
                     color="#7f3b08" if _nox else "k")
        ax.legend(frameon=False, fontsize=7.8, loc="upper left"); ax.grid(alpha=0.22, lw=0.5)
        # (2) what changes at the firing
        ax = AX[1, j]
        _i0 = int(np.argmin(np.abs(TG)))
        _axR, _hnd, _rmax = None, [], 0.0
        for q, col, lab, side in P4_CLOCKQ:
            A = H[b]["T"][q]["x1"]
            ref = A[:, _i0]
            D = A - ref[:, None]
            D = D[np.isfinite(ref)]
            if not len(D) or not np.isfinite(D).any():
                continue
            cen, lo, hi, n_t = med_band(D)
            if side == "R":                                   # kappa is not a log: its own axis, dashed, so it cannot be misread
                if _axR is None:
                    _axR = ax.twinx()
                    _axR.set_ylabel(r"$\Delta \kappa_{\rm rot}$(gas)", fontsize=9, color=col)
                    _axR.tick_params(labelsize=8.5, colors=col)
                _axR.fill_between(TG, lo, hi, color=col, alpha=0.14, lw=0)
                _axR.plot(TG, cen, "--", color=col, lw=2.4)
                _rmax = max(_rmax, float(np.nanmax(np.abs(np.r_[cen, lo, hi]))) if np.isfinite(cen).any() else 0.0)
                _hnd.append(Line2D([], [], color=col, lw=2.4, ls="--", label=lab))
            else:
                ax.fill_between(TG, lo, hi, color=col, alpha=0.20, lw=0)
                ax.plot(TG, cen, "-", color=col, lw=2.6)
                _hnd.append(Line2D([], [], color=col, lw=2.6, label=lab))
        if _axR is not None:
            _axR.set_ylim(*(P4_KLIM if P4_KLIM is not None else (-1.25 * _rmax, 1.25 * _rmax)))
        else:
            ax.text(0.985, 0.03, "no kinematic sidecar for this box\n(Part 3 prints what to build)", transform=ax.transAxes,
                    fontsize=7.2, color="0.45", va="bottom", ha="right")
        ax.axvline(0, color="k", lw=1.6, ls="--")
        ax.axhline(0, color="0.5", lw=1.0, ls=":")
        ax.annotate("the channel fires" if not _nox else "it would have fired", xy=(0, 1.0), xycoords=("data", "axes fraction"),
                    xytext=(4, -12), textcoords="offset points", fontsize=8, color="0.3")
        ax.set_xlabel(r"$t - t_{x1}$   [Gyr]"); ax.set_xlim(TG[0], TG[-1]); ax.set_ylim(-4.2, 1.2)
        if j == 0:
            ax.set_ylabel("change since the firing [dex]")
        ax.set_title("(2) what changes at the firing" + ("\n— with the channel off, nothing does" if _nox else ""), fontsize=9.5,
                     color="#7f3b08" if _nox else "k")
        ax.legend(handles=_hnd, frameon=False, fontsize=7.6, loc="lower left"); ax.grid(alpha=0.22, lw=0.5)
        ax.set_zorder(2); ax.patch.set_visible(False)
    fig.suptitle("the X-ray channel fires AT the truncation, and it is the dust that it decides — the channel-off box is the control",
                 y=0.997, fontsize=12)
    fig.tight_layout(rect=(0, 0, 1, 0.985))
    paper_save(fig, "kappa_cause")
    plt.show()
    pd.DataFrame(R4).to_csv(os.path.join(OUT, "kappa_cause.csv"), index=False)
    print(f"\n  -> {os.path.join(OUT, 'kappa_cause.csv')}")
```

## Part 4b — where the two classes come from: the gas rotation, the ISM and the stellar surface densities

Part 4's second row shows the cold gas of the selected class getting **more compact** through the X-ray firing, and that
raises the question this part answers: is the selected class simply a *denser kind of galaxy* — something it already was — or
is the density made at the same moment as everything else?

Five quantities, all from Part 3's single branch pass (masses from the histories, radii and $\kappa_{\rm rot}$ from the
kinematic sidecar), on both sides of the same clock:

$$\Sigma_X \;=\; \frac{0.5\,M_X}{\pi R_{50,X}^2}\quad [M_\odot\,{\rm kpc}^{-2}]$$

— half the mass inside the half-mass radius, the convention Part 0 already uses for the catalogue's own $\Sigma_\star$. The
stars use $R_{50}(\star)$; **the gas and the H$_2$ both use $R_{50}$(gas)**, because caesar carries no H$_2$ radius, so
$\Sigma_{\rm H_2}$ is the H$_2$ mass spread over the cold-gas half-mass radius and is a surface density of H$_2$ only to the
extent that the H$_2$ follows the gas. It is drawn as an ISM density, never as a resolved measurement — Parts 5 and 7 do that
from the particles themselves.

* **Figure 1 — the clocks.** One row per box, one column per quantity, the two class medians on $t - t_{\rm SFT}$ with
  bootstrap bands. *Read the left half of every panel first.* Curves already apart at $t - t_{\rm SFT} = -2$ Gyr mean the
  property is **inherited**; curves lying on top of each other and parting at 0 mean it is **made** at the truncation.
* **Figure 2 — the distributions at the truncation**, class by class, with an **arrow** from each class's median there to its
  median at the anchor (what a telescope would find), the shift in dex written on it. It compresses Figure 1 into one number
  per class and shows what the classes look like at the epoch the selection is *about*.
* **Figure 3 — when the difference appears.** The gap between the class medians against the clock, all five quantities
  together with a joint-bootstrap band. A flat non-zero line is an inherited difference; a line that leaves zero at
  $t_{\rm SFT}$ is one made there. $\kappa_{\rm rot}$ is not a logarithm, so it has its own right-hand axis, dashed.

The cell also prints the table the figures come from — each class's median at $-2$, $-1$, $0$, $+1$ Gyr and at the anchor,
their difference and the AUC between the classes — and ends with a per-box verdict splitting the five quantities into
*already apart 2 Gyr before the truncation* and *apart only at or after it*.

**A box with no kinematic sidecar has no radii along its branch** and is skipped with a message (Part 3 builds
`agn_states_kin_histories_<box>.hdf5` on its next run, one catalogue read per snapshot). Nothing in this part is circular:
$\kappa_{\rm rot}$(gas) at the anchor *is* the selection axis and is expected to separate; the surface densities are not, and
neither is anything at $-2$ Gyr.

```python
# ── Part 4b — WHERE THE TWO CLASSES COME FROM: the gas rotation, the ISM and the stellar surface densities ──
# Part 4 row (2) shows the cold gas of the selected class becoming MORE COMPACT through the X-ray firing, which raises the
# obvious question: is the selected class simply a denser kind of galaxy — something it always was — or is the density made at
# the same moment as everything else? This part answers it with the five quantities the question is about, each of them on
# BOTH sides of the same clock:
#   Sigma_X = 0.5 M_X / (pi R_50,X^2)   [M_sun kpc^-2]   — half the mass inside the half-mass radius (Part 0's own convention)
#   FIG 1  THE CLOCKS      one row per box, one column per quantity: the median of each class on t - t_SFT with its bootstrap
#                          band. This is the "did they evolve differently" panel — read the LEFT half of every panel first: if
#                          the two curves are already apart at t - t_SFT = -2 Gyr the property is inherited, if they lie on
#                          top of each other and part at 0 it is made by the truncation.
#   FIG 2  THE DISTRIBUTIONS at the truncation, class by class, with an ARROW from each class's median at SFT to its median
#                          AT THE ANCHOR — the same galaxies as a telescope would find them. The arrow is the evolution the
#                          clocks draw, compressed into one number per class.
#   FIG 3  WHEN THE DIFFERENCE APPEARS   the difference of the two class medians against the clock, all quantities together,
#                          with the band of a joint bootstrap: a flat non-zero line is an inherited difference, a line that
#                          leaves zero at the truncation is a made one.
# PURE READS: everything comes from Part 3's single branch pass (the masses from the boxes notebook's histories, the radii and
# kappa_rot from the kinematic sidecar it walks) and from the catalogue at the anchor. A box whose sidecar has not been built
# has no radii along the branch and is skipped WITH A MESSAGE — the catalogue anchor values are still printed for it.
P4B_BOXES = None            # None = every box Part 3 built (the channel-off box included when it carries a sidecar)
P4B_QTY   = ["kappa_gas", "lsig_h2", "lsig_gas", "lsig_star", "comp"]   # the columns, in this order
P4B_DEX   = {"kappa_gas": False, "lsig_h2": True, "lsig_gas": True, "lsig_star": True, "comp": True}   # is it a log quantity
P4B_QCOL  = {"kappa_gas": "#000000", "lsig_h2": "#1b9e77", "lsig_gas": "#7570b3", "lsig_star": "#e6ab02", "comp": "#d95f02"}
P4B_EPOCH = [-2.0, 0.0, 2.0]           # the clock epochs the table reports (the anchor is added to them); the FIRST of them
                                       # is the "was it already there" test of the verdict line, so keep it well before 0
P4B_HB    = 20              # bins of the Fig 2 histograms
P4B_NBOOT = 300             # the bootstrap of Fig 3's difference band
P4B_NMIN  = NMIN            # an epoch is drawn only where this many galaxies of the class contribute
P4B_ARROW = True            # Fig 2: the SFT -> anchor median arrow
P4B_LAB   = {q: P3_LAB.get(q, q) for q in P4B_QTY}
P4B_CTRL  = ["lms", "lr_star", "lr_gas"]   # PRINTED under the table, not drawn: the mass and the two radii every Sigma is
                                           # built from. With the mass axis on (ROT_MASS), log M* IS half the selection, so
                                           # log Sigma_* is not an independent quantity — this is where that is read off.
P4B_SHORT = {"kappa_gas": "kappa_rot(gas)", "lsig_h2": "log Sigma_H2", "lsig_gas": "log Sigma_gas",
             "lsig_star": "log Sigma_*", "comp": "-log R50(gas)/R50(*)", "lms": "log M*  <- the rule's axis",
             "lr_star": "log R50(*)", "lr_gas": "log R50(gas)"}

_i0_4b = int(np.argmin(np.abs(TG)))            # the grid index of the truncation itself
if "H" not in globals() or not H:
    print("Part 4b: needs Part 3 — run it first (it builds the branch tracks this part draws).")
else:
    _BX4b = [b for b in ((BOXES + [NOX_BOX]) if P4B_BOXES is None else P4B_BOXES) if b in H]

    def _anc4b(b):
        """The anchor values of the same five quantities, straight from the catalogue rows of Part 0's sample, aligned to
        Part 3's frame. The catalogue's radii are physical kpc and identical to the sidecar's at the anchor snapshot (checked),
        so the arrow of Fig 2 compares like with like."""
        F = H[b]["F"]
        g = QM[QM["box"] == b].drop_duplicates(subset=["snap", "gal_id"]).copy()
        g = g.set_index([g["snap"].astype(int), g["gal_id"].astype(int)])
        key = list(zip(F["snap"].astype(int), F["gal_id"].astype(int)))
        take = lambda c: (g[c].reindex(key).to_numpy(float) if c in g.columns else np.full(len(key), np.nan))
        ms, mg, mh = take("mstar"), take("mgas"), take("mh2")
        rs, rg = take("r_half_star"), take("r_half_gas")
        with np.errstate(all="ignore"):
            out = {"kappa_gas": take(KAPPA_COL),
                   "lsig_star": np.log10(np.where((rs > 0) & (ms > 0), 0.5 * ms / (np.pi * rs ** 2), np.nan)),
                   "lsig_gas":  np.log10(np.where((rg > 0) & (mg > 0), 0.5 * mg / (np.pi * rg ** 2), np.nan)),
                   "lsig_h2":   np.log10(np.where((rg > 0) & (mh > 0), 0.5 * HE * mh / (np.pi * rg ** 2), np.nan)),
                   "comp":      -np.log10(np.where((rg > 0) & (rs > 0), rg / rs, np.nan)),
                   "lms":       np.log10(np.where(ms > 0, ms, np.nan)),          # the controls, so their anchor column fills too
                   "lr_star":   np.log10(np.where(rs > 0, rs, np.nan)),
                   "lr_gas":    np.log10(np.where(rg > 0, rg, np.nan))}
        return out

    def _dband4b(A, B, seed=13):
        """The difference of the two class medians on the clock, with the band of a JOINT bootstrap (both classes resampled)."""
        A, B = np.asarray(A, float), np.asarray(B, float)
        if not len(A) or not len(B):
            return (np.full(len(TG), np.nan),) * 3
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", RuntimeWarning)
            d = np.nanmedian(A, axis=0) - np.nanmedian(B, axis=0)
            rng = np.random.default_rng(seed)
            bs = np.array([np.nanmedian(A[rng.integers(0, len(A), len(A))], axis=0)
                           - np.nanmedian(B[rng.integers(0, len(B), len(B))], axis=0) for _ in range(P4B_NBOOT)])
            lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
        ok = (np.isfinite(A).sum(axis=0) >= P4B_NMIN) & (np.isfinite(B).sum(axis=0) >= P4B_NMIN)
        return np.where(ok, d, np.nan), np.where(ok, lo, np.nan), np.where(ok, hi, np.nan)

    # ── the per-box blocks: the tracks, the values at the truncation, the values at the anchor ──
    S4b, ROWS4b, _skip4b = {}, [], []
    for b in _BX4b:
        F, T = H[b]["F"], H[b]["T"]
        if not any(np.isfinite(T[q]["sft"]).any() for q in P4B_QTY if q != "kappa_gas"):
            _skip4b.append(b); continue
        m = {c: (F["kcls"] == c).to_numpy(bool) for c in KC}
        S4b[b] = dict(F=F, T=T, m=m, anc=_anc4b(b),
                      sft={q: T[q]["sft"][:, _i0_4b] for q in P4B_QTY})
    print("=" * 130)
    print("Part 4b — the gas rotation, the ISM and the stellar surface densities of the two classes: before, at, and after the "
          "truncation")
    print("=" * 130)
    for b in _skip4b:
        print(f"  [{b}] the kinematic sidecar was not built, so the branch has no radii — the clocks are skipped for this box "
              f"(delete nothing; Part 3 builds {os.path.basename(P3_KIN_FILE.format(box=b))} on its next run)")
    for b in S4b:
        d = S4b[b]; F, T, m, anc, sft = d["F"], d["T"], d["m"], d["anc"], d["sft"]
        _nk = int(np.isfinite(T["lsig_gas"]["sft"]).any(axis=1).sum())
        print(f"\n  {'=' * 126}\n  {BOX_LABEL.get(b, b)}: {int(m['rot'].sum())} {KCLS['rot']['short']} / "
              f"{int(m['iso'].sum())} {KCLS['iso']['short']};  {_nk} of {len(F)} chains carry the kinematic branch")
        _W4b = 24
        print(f"    {'quantity':26s} " + " ".join(f"{('t - t_SFT ' + f'{e:+.0f} Gyr') if e else 'AT THE TRUNCATION':^{_W4b}s}"
                                                 for e in P4B_EPOCH) + f" {'AT THE ANCHOR':^{_W4b}s}")
        print(f"    {'':26s} " + " ".join(f"{'sel / rest    D   AUC':^{_W4b}s}" for _ in list(P4B_EPOCH) + ["anchor"]))
        def _row4b(q):
            """One line of the table: the two class medians, their difference and the AUC at every epoch, the anchor last."""
            _cells = []
            for e in list(P4B_EPOCH) + ["anchor"]:
                v = anc.get(q) if e == "anchor" else T[q]["sft"][:, int(np.argmin(np.abs(TG - e)))]
                if v is None:
                    _cells.append(f"{'--':^{_W4b}s}"); continue
                a_, b_ = v[m["rot"]], v[m["iso"]]
                na, nb = int(np.isfinite(a_).sum()), int(np.isfinite(b_).sum())
                if na < P4B_NMIN or nb < P4B_NMIN:
                    _cells.append(f"{'--':^{_W4b}s}"); continue
                ma, mb = float(np.nanmedian(a_)), float(np.nanmedian(b_))
                au = mw_auc(v, m["rot"])[0]
                _cells.append(f"{ma:6.2f}/{mb:6.2f}{ma - mb:+6.2f}{au:6.2f}")
                ROWS4b.append(dict(box=b, block="epoch", quantity=q, epoch=("anchor" if e == "anchor" else float(e)),
                                   med_rot=ma, med_iso=mb, delta=ma - mb, auc=au, n_rot=na, n_iso=nb))
            print(f"    {P4B_SHORT.get(q, q):26s} " + " ".join(_cells))
        for q in P4B_QTY:
            _row4b(q)
        if P4B_CTRL:
            print(f"    {'-- controls, not drawn: the mass and the radii every Sigma is built from --':26s}")
            for q in P4B_CTRL:
                _row4b(q)
        # THE VERDICT, per box: inherited (already apart 2 Gyr before) against made (apart only at or after the truncation)
        _inh, _made = [], []
        for q in P4B_QTY:
            _a2 = next((r["auc"] for r in ROWS4b if r["box"] == b and r["quantity"] == q and r["epoch"] == P4B_EPOCH[0]), np.nan)
            _aa = next((r["auc"] for r in ROWS4b if r["box"] == b and r["quantity"] == q and r["epoch"] == "anchor"), np.nan)
            if np.isfinite(_a2) and abs(_a2 - 0.5) >= 0.10:
                _inh.append(f"{P4B_SHORT[q]} ({_a2:.2f})")
            elif np.isfinite(_aa) and abs(_aa - 0.5) >= 0.10:
                _made.append(f"{P4B_SHORT[q]} ({_a2:.2f} -> {_aa:.2f})")
        print(f"    ALREADY APART {P4B_EPOCH[0]:+.0f} Gyr before the truncation (inherited): "
              + (", ".join(_inh) if _inh else "none"))
        print(f"    APART ONLY AT OR AFTER IT (made):                       " + (", ".join(_made) if _made else "none"))
        if ROT_MASS is not None:
            print(f"      (log M* is the rule's own second axis, so log Sigma_* is half selection by construction — what makes "
                  f"it a statement is its {P4B_EPOCH[0]:+.0f} Gyr column, where the SAME galaxies are not yet apart)")

    # ── FIGURE 1: the clocks ──
    _B4b = [b for b in _BX4b if b in S4b]
    if _B4b:
        fig, AX = plt.subplots(len(_B4b), len(P4B_QTY), figsize=(3.5 * len(P4B_QTY), 2.95 * len(_B4b)),
                               squeeze=False, sharex=True, sharey="col")
        for i, b in enumerate(_B4b):
            d = S4b[b]
            for j, q in enumerate(P4B_QTY):
                ax = AX[i, j]
                for c in KC:
                    A = d["T"][q]["sft"][d["m"][c]]
                    if not len(A) or not np.isfinite(A).any():
                        continue
                    cen, lo, hi, n_t = med_band(A, nmin=P4B_NMIN, nboot=P4B_NBOOT)
                    ax.fill_between(TG, lo, hi, color=KCLS[c]["color"], alpha=0.20, lw=0, zorder=3)
                    ax.plot(TG, cen, KCLS[c]["ls"], color=KCLS[c]["color"], lw=KCLS[c]["lw"], zorder=5,
                            label=f"{KCLS[c]['short']}  ({int(d['m'][c].sum())})")
                    for x_, y_, l_, h_, nn in zip(TG, cen, lo, hi, n_t):
                        ROWS4b.append(dict(box=b, block="clock", quantity=q, kcls=c, x=float(x_), n_gal=int(nn),
                                           med=float(y_), lo=float(l_), hi=float(h_)))
                ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9, zorder=1)
                if q == "kappa_gas":
                    ax.axhline(KAPPA_ISO, color="0.4", lw=1.0, ls=":", zorder=1)
                if q == "comp":
                    ax.axhline(0.0, color="0.4", lw=1.0, ls=":", zorder=1)
                ax.grid(alpha=0.22, lw=0.5); ax.set_xlim(TG[0], TG[-1])
                if i == 0:
                    ax.set_title(P4B_LAB[q], fontsize=10)
                if j == 0:
                    ax.set_ylabel(BOX_LABEL.get(b, b), fontsize=9.5)
                if i == len(_B4b) - 1:
                    ax.set_xlabel(r"$t - t_{\rm SFT}$   [Gyr]")
                if i == 0 and j == 0:
                    ax.legend(frameon=False, fontsize=8.2, loc="best")
        fig.suptitle("Part 4b — the two classes on the truncation clock: read the left half of every panel. Curves that are "
                     "already apart before $t_{\\rm SFT}$ are an INHERITED difference, curves that part at 0 are a MADE one.",
                     fontsize=11.5, y=0.999)
        fig.tight_layout(rect=(0, 0, 1, 0.975))
        paper_save(fig, "kappa_sigma_clocks")
        plt.show()

        # ── FIGURE 2: the distributions at the truncation, with the arrow to the anchor ──
        fig, AX = plt.subplots(len(_B4b), len(P4B_QTY), figsize=(3.5 * len(P4B_QTY), 2.75 * len(_B4b)),
                               squeeze=False, sharex="col")
        _bins4b = {}
        for q in P4B_QTY:                                    # ONE binning per quantity, over every box and both epochs
            _all = np.concatenate([np.concatenate([S4b[b]["sft"][q], S4b[b]["anc"][q]]) for b in _B4b])
            _all = _all[np.isfinite(_all)]
            _bins4b[q] = (np.linspace(*np.percentile(_all, [1, 99]), P4B_HB + 1) if len(_all) > 10
                          else np.linspace(0, 1, P4B_HB + 1))
        for i, b in enumerate(_B4b):
            d = S4b[b]
            for j, q in enumerate(P4B_QTY):
                ax = AX[i, j]; bins = _bins4b[q]
                _hmax = 0.0
                for c in KC:
                    v = d["sft"][q][d["m"][c]]; v = v[np.isfinite(v)]
                    if len(v) < 3:
                        continue
                    h, _ = np.histogram(np.clip(v, bins[0], bins[-1]), bins=bins, density=True)
                    _hmax = max(_hmax, float(h.max()))
                    ax.stairs(h, bins, fill=True, color=KCLS[c]["color"], alpha=0.28, lw=0)
                    ax.stairs(h, bins, color=KCLS[c]["color"], lw=1.8)
                    ROWS4b.append(dict(box=b, block="at_sft", quantity=q, kcls=c, n_gal=len(v),
                                       med=float(np.median(v)), lo=float(np.percentile(v, 16)), hi=float(np.percentile(v, 84))))
                for k, c in enumerate(KC):                   # the arrow: the class median from SFT to the anchor
                    v0, v1 = d["sft"][q][d["m"][c]], d["anc"][q][d["m"][c]]
                    v0, v1 = v0[np.isfinite(v0)], v1[np.isfinite(v1)]
                    if len(v0) < 3 or len(v1) < 3:
                        continue
                    m0, m1 = float(np.median(v0)), float(np.median(v1))
                    ROWS4b.append(dict(box=b, block="at_anchor", quantity=q, kcls=c, n_gal=len(v1), med=m1,
                                       lo=float(np.percentile(v1, 16)), hi=float(np.percentile(v1, 84))))
                    if not P4B_ARROW:
                        continue
                    _yy = (1.0 + 0.11 * k) * _hmax * 1.02
                    ax.annotate("", xy=(np.clip(m1, bins[0], bins[-1]), _yy), xytext=(np.clip(m0, bins[0], bins[-1]), _yy),
                                arrowprops=dict(arrowstyle="-|>", color=KCLS[c]["color"], lw=1.9, shrinkA=0, shrinkB=0))
                    ax.plot([np.clip(m0, bins[0], bins[-1])], [_yy], "o", ms=4.5, color=KCLS[c]["color"])
                    ax.text(np.clip(m1, bins[0], bins[-1]), _yy, f"  {m1 - m0:+.2f}", color=KCLS[c]["color"], fontsize=7.6,
                            va="center", ha="left" if m1 >= m0 else "right")
                ax.set_ylim(0, max(_hmax * 1.42, 1e-3)); ax.set_yticks([])
                ax.grid(alpha=0.20, lw=0.5, axis="x")
                if i == 0:
                    ax.set_title(P4B_LAB[q], fontsize=10)
                if j == 0:
                    ax.set_ylabel(BOX_LABEL.get(b, b), fontsize=9.5)
                if i == len(_B4b) - 1:
                    ax.set_xlabel(P4B_LAB[q], fontsize=9)
        fig.legend(handles=[Line2D([], [], color=KCLS[c]["color"], lw=2.4,
                                   label=f"{KCLS[c]['short']}  —  at the truncation, then the arrow to the anchor")
                            for c in KC], loc="upper left", bbox_to_anchor=(0.008, 0.975), frameon=False, fontsize=9)
        fig.suptitle("Part 4b — the same five quantities AT THE TRUNCATION, class by class; the arrow runs from each class's "
                     "median there to its median AT THE ANCHOR, i.e. what a telescope would find (the shift in dex is written "
                     "on it).", fontsize=11.5, y=0.999)
        fig.tight_layout(rect=(0, 0, 1, 0.93))
        paper_save(fig, "kappa_sigma_atsft")
        plt.show()

        # ── FIGURE 3: when the difference appears ──
        fig, AX = plt.subplots(1, len(_B4b), figsize=(max(4.6 * len(_B4b), 11.0), 4.4), squeeze=False, sharey=True)
        for i, b in enumerate(_B4b):
            d = S4b[b]; ax = AX[0, i]
            axk, _kspan = ax.twinx(), []
            for q in P4B_QTY:
                A, B = d["T"][q]["sft"][d["m"]["rot"]], d["T"][q]["sft"][d["m"]["iso"]]
                cen, lo, hi = _dband4b(A, B)
                _ax = ax if P4B_DEX[q] else axk
                _ls = "-" if P4B_DEX[q] else (0, (4, 1.6))
                _ax.fill_between(TG, lo, hi, color=P4B_QCOL[q], alpha=0.14, lw=0)
                _ax.plot(TG, cen, ls=_ls, color=P4B_QCOL[q], lw=2.2,
                         label=P4B_LAB[q] + ("" if P4B_DEX[q] else "   (right axis)"))
                if not P4B_DEX[q]:
                    _kspan.append(np.concatenate([lo, hi, cen]))
                for x_, y_, l_, h_ in zip(TG, cen, lo, hi):
                    ROWS4b.append(dict(box=b, block="delta", quantity=q, x=float(x_), med=float(y_), lo=float(l_), hi=float(h_)))
            ax.axhline(0, color="k", lw=1.0); ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9)
            ax.annotate("SFT", xy=(0, 1.0), xycoords=("data", "axes fraction"), xytext=(3, -12), textcoords="offset points",
                        fontsize=8.5, color="0.3")
            with warnings.catch_warnings():           # the kappa axis is scaled to the kappa curve, symmetric about zero
                warnings.simplefilter("ignore", RuntimeWarning)
                _kl = float(np.nanmax(np.abs(np.concatenate(_kspan)))) if _kspan else np.nan
            axk.set_ylim(*((-1.6 * _kl, 1.6 * _kl) if np.isfinite(_kl) and _kl > 0 else (-0.1, 0.1)))
            axk.set_ylabel(r"$\Delta\,\kappa_{\rm rot}$(gas)", color=P4B_QCOL["kappa_gas"], fontsize=9.5)
            axk.tick_params(axis="y", colors=P4B_QCOL["kappa_gas"], labelsize=8.5)
            ax.set_xlim(TG[0], TG[-1]); ax.grid(alpha=0.22, lw=0.5)
            ax.set_xlabel(r"$t - t_{\rm SFT}$   [Gyr]")
            if i == 0:
                ax.set_ylabel(r"$\Delta$ (selected $-$ the rest)   [dex]")
            ax.set_title(BOX_LABEL.get(b, b), fontsize=10)
        fig.legend(handles=[Line2D([], [], color=P4B_QCOL[q], lw=2.2, ls="-" if P4B_DEX[q] else (0, (4, 1.6)),
                                   label=P4B_LAB[q] + ("" if P4B_DEX[q] else "   (right axis)")) for q in P4B_QTY],
                   loc="upper center", bbox_to_anchor=(0.5, 0.935), frameon=False, fontsize=9,
                   ncol=min(len(P4B_QTY), 5))
        fig.suptitle("Part 4b — when the difference appears: the gap between the class medians on the truncation clock. A flat "
                     "non-zero line is inherited, a line that leaves zero at $t_{\\rm SFT}$ is made there.", fontsize=11.5)
        fig.tight_layout(rect=(0, 0, 1, 0.88))
        paper_save(fig, "kappa_sigma_when")
        plt.show()
    else:
        print("  no box carries the kinematic branch — nothing to draw. Run Part 3 with P3_KIN_BUILD = True on the cluster, "
              "where the caesar catalogues are.")
    if ROWS4b:
        pd.DataFrame(ROWS4b).to_csv(os.path.join(OUT, "kappa_sigma.csv"), index=False)
        print(f"\n  every curve, distribution and epoch of this part -> {os.path.join(OUT, 'kappa_sigma.csv')}")
```

## Part 5 — the 25 Mpc close-up: the critical points, the radii, and the K-S plane

The 25 Mpc box is the only one with particle-level products, so it is the only place the selection can be checked against
something it was not made from. Three blocks, each gating on its own files:

* **(A) the critical points.** $\kappa_{\rm rot}^{\rm H_2}$, $\kappa_{\rm rot}^{\star}$, sSFR, stellar age, $f_{\rm dust}$ and
  $f_{\rm H_2}$ at ignition → jet → SFT → QT → end, in the two spherical zones of the K-S products (core $r<3.2$ kpc,
  outskirt 3.2–10 kpc), with the ALMA-C11 detections as bands. **Read it as the timing question**: if the kinematic class were
  a property the galaxy always had, the two curves would already be apart at ignition. They are not — they are on top of each
  other until QT and separate at `end`, in $\kappa_{\rm rot}^{\rm H_2}$ *and* in the dust, which is Part 3's statement seen in
  the resolved products.
* **(B) the radii.** $A_V$, $M_{\rm dust}/M_\star$, $M_{\rm H_2}/M_\star$, stellar age and $\kappa_{\rm rot}$ per annulus
  (0–1, 1–3.2, 3.2–10, 10–32 kpc), per-galaxy sightline medians with the m25 star-forming galaxies as the grey reference. The
  offset between the classes is present **at every radius**, in the dust and in $A_V$, while the stellar age and
  $\kappa_{\rm rot}^{\star}$ are identical — the division is in the ISM, not in the stellar structure.
* **(C) the K-S plane.** The $R_{50}({\rm H_2})$ tracks SFT → QT → end against Bigiel+08, with the track-end regions counted.

Nothing here is circular: the classes are cut on the **catalogue's whole-galaxy** $\kappa_{\rm gas}$, and every quantity drawn
is measured from the particles, in apertures, by a different notebook.

```python
# ── Part 5 — the 25 Mpc close-up of the two classes: the critical points, the radii, and the K-S plane ──
# The only part that reaches outside the ism_prediction products: it reads the m25 notebook's own particle-level tables and the
# K-S track products. Each block gates on its own file and prints what is missing, so the cell runs unchanged off the cluster.
#   (A) THE CRITICAL POINTS   kappa_rot(H2) / kappa_rot(*) / sSFR / stellar age / f_dust / f_H2 at ignition -> jet -> SFT -> QT ->
#       end, in the two spherical zones of the K-S products (core r < 3.2 kpc, outskirt 3.2-10 kpc), with the ALMA-C11
#       detections as bands. The question the sequence answers: is the kinematic class already there at ignition, or is it made?
#   (B) THE RADII             A_V, M_dust/M*, M_H2/M*, stellar age and kappa_rot per annulus (0-1, 1-3.2, 3.2-10, 10-32 kpc),
#       per-galaxy sightline medians, with the m25 star-forming galaxies as the grey reference line.
#   (C) THE K-S PLANE         the R50(H2) tracks SFT -> QT -> end of the two classes against Bigiel+08.
P5_BOX    = "cis25"
P5_NMIN   = 3               # galaxies a drawn point needs — deliberately below the notebook's NMIN: m25 is tens of objects
P5_M25    = globals().get("P5_M25_OVERRIDE", os.path.join(os.getcwd(), "output", P5_BOX))
ANN5  = os.path.join(P5_M25, "tables", "annulus_ism_truth.fits")            # m25 Part 8a: M_dust / M_gas / M_H2 per (annulus, sightline)
APT5  = os.path.join(P5_M25, "tables", "aperture_truth.fits")               # m25 Part 7e: M*, age per (annulus, sightline)
AV5   = os.path.join(P5_M25, "tables", "annulus_av_allincl.fits")           # m25 Part 8b: A_V per (annulus, sightline) — optional panel
KIN5  = os.path.join(P5_M25, "tables", "annulus_kinematics.fits")           # m25 Part 8j0: kappa_rot per spherical zone — optional panels
SEL5  = os.path.join(P5_M25, "tables", "powderday_quenched_selection.fits") # m25 Part 3: pop -> the SF-controls line — optional
KS5   = os.path.join(P5_M25, "ks_tracks_pt", "ks_tracks.fits")              # ks_tracks_quenched_m25 Part 4a (run tag 'pt')
KINST5 = os.path.join(P5_M25, "ks_tracks", "ks_stage_kinematics.fits")      # its Part 3b: kappa + stellar age per spherical zone
OBS5  = os.path.join(os.getcwd(), "obs_data", "almac11", "almac11_gas_dust.csv")
ANNULI5, NSTAR_AP_MIN5 = [("ap1kpc", "0-1"), ("ann3kpc", "1-3.2"), ("ann10kpc", "3.2-10"), ("ann32kpc", "10-32")], 20
P5_STAGES_ALL = ["agn_ign", "jet_on", "sft", "qt", "end"]
P5_ZONES  = [("ap3kpc", "core, $r < 3.2$ kpc"), ("ann10kpc", "outskirt, 3.2-10 kpc")]
P5_QUANT  = [("kappa_H2", r"$\kappa_{\rm rot}^{\rm H_2}$", "kappa"), ("kappa_star", r"$\kappa_{\rm rot}^{\star}$", "kappa"),
             ("ssfr", r"sSFR [yr$^{-1}$]", "log"), ("age_mw_gyr", "stellar age [Gyr]" + "\n" + r"(sim: MASS-weighted)", "lin"),
             ("fdust", r"$M_{\rm dust}/M_\star$", "log"), ("fH2", r"$M_{\rm H_2}/M_\star$", "log")]
P5_NSTAR_MIN, P5_NBOOT = 20, 400
P5_TICK   = {"agn_ign": "ign.", "jet_on": "jet", "sft": "SFT", "qt": "QT", "end": "end"}
# THE AGE AXIS IS NOT LIKE FOR LIKE, and the observed band on it is drawn hatched to say so: the simulation's age is caesar's
# mass-weighted age, the ALMA-C11 age is CIGALE's sfhdelayedbq ONSET age. For a delayed-tau SFH the mass-weighted age sits
# roughly 2 tau below the onset age, and the table carries no tau, so the conversion cannot be done here.
P5_OBS_AGE_LIKE = False

G5 = KC
C5 = {c: KCLS[c]["color"] for c in KC}
L5 = {c: KCLS[c]["short"] for c in KC}
KEY5 = {(s, g): v for (b_, s, g), v in KCLS_OF.items() if b_ == P5_BOX}
rows5 = []


def _fr5(path, cols):
    """A FITS table -> a DataFrame of the named columns, masked values as NaN and byte strings decoded."""
    t = Table.read(path); d = {}
    for c in cols:
        v = t[c]; v = v.filled(np.nan) if hasattr(v, "filled") else v
        v = np.asarray(v)
        d[c] = np.array([x.decode() if isinstance(x, bytes) else str(x) for x in v]) if v.dtype.kind in "SU" else v.astype(float)
    return pd.DataFrame(d)


def _bmed5(v, n=P5_NBOOT, seed=5):
    v = np.asarray(v, float); v = v[np.isfinite(v)]
    if len(v) < P5_NMIN:
        return np.nan, np.nan, np.nan, len(v)
    bs = np.median(v[np.random.default_rng(seed).integers(0, len(v), (n, len(v)))], axis=1)
    return float(np.median(v)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), len(v)


if not KEY5:
    print(f"Part 5: no {P5_BOX} galaxy carries a class — Part 2 must have run on a sample that includes it.")
else:
    print(f"Part 5 — {BOX_LABEL[P5_BOX]}: {len(KEY5)} classified anchors — "
          + ", ".join(f"{L5[c]} {sum(v == c for v in KEY5.values())}" for c in G5))

    # ── (A) the critical-point sequences ──
    if not (os.path.exists(KS5) and os.path.exists(KINST5)):
        print(f"\n(A) critical points: {KS5 if not os.path.exists(KS5) else KINST5} missing — ks_tracks_quenched_m25 "
              "Part 4a (run tag 'pt') and Part 3b build them (cluster); skipped")
    else:
        T5 = Table.read(KS5)
        _s5 = lambda a: np.array([x.decode() if isinstance(x, bytes) else str(x) for x in np.asarray(a)])
        stg, pop, ap, gk = _s5(T5["stage"]), _s5(T5["pop"]), _s5(T5["ap_label"]), _s5(T5["gkey"])
        P5_STAGES = [st for st in P5_STAGES_ALL if st in set(stg[pop == "Q"])]
        _q5 = (pop == "Q") & np.isin(stg, P5_STAGES)
        EP5 = pd.DataFrame(dict(gkey=gk[_q5], stage=stg[_q5], snap=np.asarray(T5["snap"], float)[_q5].astype(int),
                                gx=np.asarray(T5["gx"], float)[_q5].astype(int))).drop_duplicates(["gkey", "stage"])
        _kin5 = Table.read(KINST5)
        K5 = pd.DataFrame({c: (_s5(_kin5[c]) if c == "zone" else np.asarray(_kin5[c], float))
                           for c in ("snap", "gx", "zone", "kappa_H2", "kappa_star", "age_mw_gyr")})
        K5["snap"], K5["gx"] = K5["snap"].astype(int), K5["gx"].astype(int)
        K5 = EP5.merge(K5[K5["zone"].isin([z for z, _ in P5_ZONES])], on=["snap", "gx"], how="inner")
        # the zone ISM at every critical point, from the tracks' projected apertures: core = ap3kpc, outskirt = ap10kpc - ap3kpc
        _m5 = _q5 & np.isin(ap, ["ap3kpc", "ap10kpc"])
        TD5 = pd.DataFrame(dict(gkey=gk[_m5], stage=stg[_m5], ap=ap[_m5],
                                m_dust=np.asarray(T5["m_dust"], float)[_m5], m_H2=np.asarray(T5["m_H2"], float)[_m5],
                                sfr=np.asarray(T5["sfr100"], float)[_m5], m_star=np.asarray(T5["m_star"], float)[_m5],
                                n_star=np.asarray(T5["n_star"], float)[_m5])).drop_duplicates(["gkey", "stage", "ap"])
        _pv5 = TD5.pivot_table(index=["gkey", "stage"], columns="ap", values=["m_dust", "m_H2", "sfr", "m_star", "n_star"], aggfunc="first")
        _ism5 = []
        for (gk_, st_), r in _pv5.iterrows():
            for zone, sel in (("ap3kpc", lambda c: r[(c, "ap3kpc")]), ("ann10kpc", lambda c: r[(c, "ap10kpc")] - r[(c, "ap3kpc")])):
                ms_, ns_ = sel("m_star"), sel("n_star")
                ok_ = np.isfinite(ms_) and ms_ > 0 and ns_ >= P5_NSTAR_MIN
                d = dict(gkey=gk_, stage=st_, zone=zone)
                for c, num in (("fdust", "m_dust"), ("fH2", "m_H2")):
                    v = sel(num)
                    d[c] = (v / ms_) if (ok_ and np.isfinite(v) and v > 0) else np.nan       # zero dust / H2 -> NaN, never a floor
                v = sel("sfr")
                d["ssfr"] = max(v / ms_, SSFR_FLOOR) if (ok_ and np.isfinite(v) and v >= 0) else np.nan
                _ism5.append(d)
        K5 = K5.merge(pd.DataFrame(_ism5), on=["gkey", "stage", "zone"], how="left")
        K5["grp"] = [KEY5.get(tuple(int(v) for v in g_.split("_"))) for g_ in K5["gkey"]]
        K5 = K5[K5["grp"].notna()].reset_index(drop=True)
        print(f"  (A) {K5['gkey'].nunique()} of them are in the K-S tracks; the critical points " + " -> ".join(P5_STAGES) + ": "
              + ", ".join(f"{L5[c]} {K5.loc[(K5['grp'] == c) & (K5['zone'] == 'ap3kpc'), 'gkey'].nunique()}" for c in G5))
        # the observed bands: ALMA-C11 dust detections -> age / f_dust / sSFR, CO detections -> f_H2 (/1.36, bare H2 masses)
        B5 = {}
        if os.path.exists(OBS5):
            _o = pd.read_csv(OBS5)
            _d = _o[_o["det_dust"].astype(bool)]
            if len(_d):
                B5["age_mw_gyr"] = (float((10 ** _d["logage"] / 1e9).min()), float((10 ** _d["logage"] / 1e9).max()))
                B5["fdust"] = (float((10 ** (_d["logMdust"] - _d["logMstar"])).min()), float((10 ** (_d["logMdust"] - _d["logMstar"])).max()))
                B5["ssfr"] = (float((_d["SFR"] / 10 ** _d["logMstar"]).min()), float((_d["SFR"] / 10 ** _d["logMstar"]).max()))
            _c = _o[_o["det_co"].astype(bool)]
            if len(_c):
                B5["fH2"] = (float((_c["MH2_fid"] / 1.36 / 10 ** _c["logMstar"]).min()),
                             float((_c["MH2_fid"] / 1.36 / 10 ** _c["logMstar"]).max()))
        else:
            print(f"  (A) the ALMA-C11 bands need {OBS5} — drawn without them")
        fig, axs = plt.subplots(len(P5_ZONES), len(P5_QUANT), figsize=(2.95 * len(P5_QUANT), 3.4 * len(P5_ZONES)), squeeze=False)
        for zi, (zone, zlab) in enumerate(P5_ZONES):
            for qi, (col, lab, kind) in enumerate(P5_QUANT):
                ax = axs[zi, qi]
                for gi, g in enumerate(G5):
                    off = (gi - (len(G5) - 1) / 2) * 0.16
                    xx, yy, lo, hi = [], [], [], []
                    for i, st_ in enumerate(P5_STAGES):
                        med, l16, h84, n = _bmed5(K5.loc[(K5["grp"] == g) & (K5["zone"] == zone) & (K5["stage"] == st_), col])
                        rows5.append(dict(block="stage", zone=zone, quantity=col, kcls=g, x=st_, n_gal=n, med=med, lo=l16, hi=h84))
                        if np.isfinite(med):
                            xx.append(i + off); yy.append(med); lo.append(med - l16); hi.append(h84 - med)
                    if len(xx):
                        ax.errorbar(xx, yy, yerr=np.array([lo, hi]), color=C5[g], lw=2.2, elinewidth=1.3, capsize=2.5,
                                    marker="o", ms=5.5, mec="k", mew=0.5, ls="-", zorder=5,
                                    label=KCLS[g]["label"] if (zi == 0 and qi == 0) else None)
                _mm5 = (col == "age_mw_gyr") and not P5_OBS_AGE_LIKE
                if col in B5 and np.all(np.isfinite(B5[col])):
                    ax.axhspan(B5[col][0], B5[col][1], facecolor="#e7298a", alpha=0.12, lw=0, zorder=0,
                               hatch="///" if _mm5 else None, edgecolor="#e7298a")
                    if _mm5 and zi == 0:
                        ax.text(0.5, 0.985, "obs = CIGALE ONSET age, not mass-weighted", transform=ax.transAxes,
                                ha="center", va="top", fontsize=6.6, color="#8c1b52", fontweight="bold")
                if kind == "log":
                    ax.set_yscale("log")
                if kind == "kappa":
                    ax.axhline(KAPPA_ISO, color="0.3", lw=1.0, ls="--"); ax.set_ylim(0, 1)
                if col == "ssfr":
                    ax.axhline(1e-11, color="0.3", lw=0.9, ls="--")
                if col == "fdust":
                    ax.axhline(10 ** DUSTY_CUT, color="0.4", lw=0.9, ls=":")
                ax.set_xlim(-0.6, len(P5_STAGES) - 0.4)
                ax.set_xticks(range(len(P5_STAGES))); ax.set_xticklabels([P5_TICK.get(s_, s_) for s_ in P5_STAGES], fontsize=8.5)
                ax.set_ylabel((zlab + "\n" if qi == 0 else "") + lab, fontsize=9)
                ax.grid(alpha=0.25, lw=0.5)
        axs[0, 0].legend(frameon=False, fontsize=7.5, loc="best")
        if B5:
            axs[0, 2].legend(handles=[Line2D([], [], color="#e7298a", lw=6, alpha=0.25)], labels=["ALMA-C11 detections (min-max)"],
                             frameon=False, fontsize=7.5, loc="best")
        fig.suptitle(f"{BOX_LABEL[P5_BOX]}: the two classes at the critical points (median + galaxy-bootstrap 16-84; the core / "
                     f"outskirt spherical zones of the K-S products, the ISM from the tracks' projected apertures)", fontsize=10)
        fig.tight_layout(rect=(0, 0, 1, 0.955))
        paper_save(fig, "kappa_m25_sequences")
        plt.show()
        print("  READ IT LIKE THIS: if the kinematic class is a PROPERTY the galaxy always had, the curves separate at ignition. "
              "If they only separate at QT / end, the class is MADE during the quenching.")

    # ── (B) the radii ──
    if not (os.path.exists(ANN5) and os.path.exists(APT5)):
        print(f"\n(B) the radial anatomy: {ANN5} or {APT5} missing — the m25 notebook's Parts 8a / 7e (cluster)")
    else:
        K4 = ["snap", "gal_id", "incl", "aperture"]
        RF5 = _fr5(ANN5, K4 + ["M_dust", "M_H2"]).merge(_fr5(APT5, K4 + ["nstar_ap", "mstar", "age_m_star_myr"]), on=K4, how="inner")
        if os.path.exists(AV5):
            RF5 = RF5.merge(_fr5(AV5, K4 + ["A_V"]), on=K4, how="left")
        RF5 = RF5[RF5["aperture"].isin([a for a, _ in ANNULI5])].reset_index(drop=True)
        SF5 = set()
        if os.path.exists(SEL5):
            _s5b = _fr5(SEL5, ["snap", "gal_id", "pop"])
            SF5 = {(int(s_), int(i_)) for s_, i_, p_ in zip(_s5b["snap"], _s5b["gal_id"], _s5b["pop"]) if p_ == "SF"}
        RF5["grp"] = [KEY5.get((int(s_), int(i_)), "star_forming" if (int(s_), int(i_)) in SF5 else None)
                      for s_, i_ in zip(RF5["snap"], RF5["gal_id"])]
        RF5 = RF5[RF5["grp"].notna()].reset_index(drop=True)
        if os.path.exists(KIN5):
            RF5 = RF5.merge(_fr5(KIN5, ["snap", "gal_id", "aperture", "kappa_H2", "kappa_star"]),
                            on=["snap", "gal_id", "aperture"], how="left")
        for _f, _n in ((AV5, "A_V (m25 8b)"), (KIN5, "kappa (m25 8j0)"), (SEL5, "the SF-controls line (m25 Part 3)")):
            if not os.path.exists(_f):
                print(f"  (B) {_n}: {_f} missing — panel / line skipped")
        _ms5 = np.where(RF5["nstar_ap"].to_numpy(float) >= NSTAR_AP_MIN5, RF5["mstar"].to_numpy(float), np.nan)
        with np.errstate(divide="ignore", invalid="ignore"):
            for c, num in (("fdust_ap", "M_dust"), ("fH2_ap", "M_H2")):
                v = RF5[num].to_numpy(float)
                RF5[c] = np.where(np.isfinite(v) & (v > 0) & (_ms5 > 0), v / _ms5, np.nan)
            RF5["age_ap"] = np.where(np.isfinite(_ms5), RF5["age_m_star_myr"].to_numpy(float) / 1e3, np.nan)
            if "A_V" in RF5.columns:
                _av5 = RF5["A_V"].to_numpy(float)
                RF5["A_V"] = np.where(np.isfinite(_av5) & (_av5 > 0), _av5, np.nan)
        QUANT5 = ([("A_V", r"$A_V$ [mag]", "log")] if "A_V" in RF5.columns else []) \
            + [("fdust_ap", r"$M_{\rm dust}/M_\star$", "log"), ("fH2_ap", r"$M_{\rm H_2}/M_\star$", "log"),
               ("age_ap", "stellar age [Gyr]", "lin")] \
            + ([(k, lab, "kappa") for k, lab in (("kappa_H2", r"$\kappa_{\rm rot}^{\rm H_2}$"),
                                                 ("kappa_star", r"$\kappa_{\rm rot}^{\star}$"))] if "kappa_H2" in RF5.columns else [])
        GRP5B = (["star_forming"] if len(SF5) else []) + G5
        C5B = dict(C5, star_forming="0.6")
        L5B = dict(L5, star_forming="m25 star-forming controls")
        _x5 = np.arange(len(ANNULI5))
        _nc5 = (len(QUANT5) + 1) // 2
        rng5 = np.random.default_rng(5)
        NMAX5 = {g: 0 for g in GRP5B}
        fig, axs = plt.subplots(2, _nc5, figsize=(3.5 * _nc5, 7.1), squeeze=False)
        for qi, (col, lab, kind) in enumerate(QUANT5):
            ax = axs[qi // _nc5, qi % _nc5]
            for gi, g in enumerate(GRP5B):
                m_, l_, h_, n_ = [], [], [], []
                for a_, _ in ANNULI5:
                    sub = RF5[(RF5["grp"] == g) & (RF5["aperture"] == a_)]
                    v = sub[col].to_numpy(float); ok = np.isfinite(v)
                    per = {}
                    for s_, i_, vv in zip(sub["snap"].to_numpy(float)[ok], sub["gal_id"].to_numpy(float)[ok], v[ok]):
                        per.setdefault((int(s_), int(i_)), []).append(vv)              # the sightline rows -> one value per galaxy
                    pg = np.array([np.median(vv) for vv in per.values()])
                    if len(pg) >= P5_NMIN:
                        bs = np.median(pg[rng5.integers(0, len(pg), (NBOOT, len(pg)))], axis=1)
                        m_.append(float(np.median(pg))); l_.append(float(np.percentile(bs, 16))); h_.append(float(np.percentile(bs, 84)))
                    else:
                        m_.append(np.nan); l_.append(np.nan); h_.append(np.nan)
                    n_.append(len(pg))
                    rows5.append(dict(block="profile", zone="", quantity=col, kcls=g, x=a_, n_gal=len(pg), med=m_[-1], lo=l_[-1], hi=h_[-1]))
                NMAX5[g] = max(NMAX5[g], max(n_))
                off = (gi - (len(GRP5B) - 1) / 2) * 0.08
                m_, l_, h_ = np.array(m_), np.array(l_), np.array(h_)
                ok_ = np.isfinite(m_)
                if ok_.any():
                    ax.fill_between(_x5[ok_] + off, l_[ok_], h_[ok_], color=C5B[g], alpha=0.15, lw=0)
                    ax.plot(_x5[ok_] + off, m_[ok_], "o-", color=C5B[g], lw=1.3 if g == "star_forming" else 2.1, ms=4.5,
                            mec="k", mew=0.4)
            if kind == "log":
                ax.set_yscale("log")
            if kind == "kappa":
                ax.axhline(KAPPA_ISO, color="0.3", lw=1.0, ls="--"); ax.set_ylim(0, 1)
            if col == "fdust_ap":
                ax.axhline(10 ** DUSTY_CUT, color="0.4", lw=0.9, ls=":")
            ax.set_xticks(_x5); ax.set_xticklabels([r_ for _, r_ in ANNULI5], fontsize=8)
            if qi // _nc5 == 1:
                ax.set_xlabel("annulus [kpc]")
            ax.set_ylabel(lab); ax.grid(alpha=0.25, lw=0.5)
        for qi in range(len(QUANT5), 2 * _nc5):
            axs[qi // _nc5, qi % _nc5].axis("off")
        axs[0, 0].legend(handles=[Line2D([], [], color=C5B[g], marker="o", ms=4.5, mec="k", mew=0.4,
                                         lw=1.3 if g == "star_forming" else 2.1, label=f"{L5B[g]} (N <= {NMAX5[g]})")
                                  for g in GRP5B], frameon=False, fontsize=8)
        fig.suptitle(f"{BOX_LABEL[P5_BOX]}: the radial anatomy of the two classes (+ the SF controls) — per-galaxy sightline medians, "
                     f"bootstrap 16-84 of the class median"
                     + ("" if "kappa_H2" not in RF5.columns else "; kappa in the spherical zones (sightline-independent)"), fontsize=10.5)
        fig.tight_layout()
        paper_save(fig, "kappa_m25_profiles")
        plt.show()

    # ── (C) the K-S plane ──
    kl5 = None
    if not os.path.exists(KS5):
        print(f"\n(C) the K-S plane: {KS5} missing — ks_tracks_quenched_m25 Part 4 (run tag 'pt', cluster)")
    else:
        try:
            import ks_tracks_lib as kl5                                  # the B08 / K98 relations and the He factor, verbatim
        except ImportError as e:
            print(f"\n(C) the K-S plane: {KS5} is there but ks_tracks_lib is not importable ({e}) — run the notebook from the repo root")
    if kl5 is not None:
        TR5 = Table.read(KS5)
        _s5c = lambda a: np.array([x.decode() if isinstance(x, bytes) else str(x) for x in np.asarray(a)])
        ap5, pop5, stg5, gk5 = _s5c(TR5["ap_label"]), _s5c(TR5["pop"]), _s5c(TR5["stage"]), _s5c(TR5["gkey"])
        x5, y5, ul5 = np.asarray(TR5["logSigmaH2"], float), np.asarray(TR5["logSigmaSFR"], float), np.asarray(TR5["is_ul"], bool)
        grp5 = np.array([KEY5.get(tuple(int(v) for v in g_.split("_"))) for g_ in gk5], dtype=object)
        STAGES5, MRK5 = ["sft", "qt", "end"], {"sft": "^", "qt": "s", "end": "o"}
        base5 = ap5 == "R50_H2"
        band5 = float(kl5.RELATIONS["B08"]["sig"])
        fig, ax = plt.subplots(figsize=(6.8, 6.3))
        sf_ = base5 & (pop5 == "SF")
        ax.scatter(x5[sf_], y5[sf_], s=7, c="0.65", alpha=0.3, lw=0, zorder=1, rasterized=True)
        xs5 = np.linspace(-1.0, 4.1, 60)
        for rk, lw_, alpha_ in (("B08", 1.4, 1.0), ("K98", 0.9, 0.55)):
            rel = kl5.RELATIONS[rk]
            if rk == "B08":
                ax.fill_between(xs5, rel["A"] + rel["N"] * xs5 - rel["sig"], rel["A"] + rel["N"] * xs5 + rel["sig"],
                                color=rel["color"], alpha=0.10, lw=0, zorder=0)
            ax.plot(xs5, rel["A"] + rel["N"] * xs5, color=rel["color"], lw=lw_, ls=rel["ls"], alpha=alpha_, zorder=2)
        print(f"\n(C) the track-end regions w.r.t. B08 +- {band5:g} dex (the R50_H2 aperture):")
        for g in G5:
            mk = base5 & (pop5 == "Q") & (grp5 == g) & np.isin(stg5, STAGES5)
            for g_ in np.unique(gk5[mk]):
                r_ = np.flatnonzero(mk & (gk5 == g_))
                r_ = r_[np.argsort([STAGES5.index(s_) for s_ in stg5[r_]])]
                ok = np.isfinite(x5[r_]) & np.isfinite(y5[r_])
                if ok.sum() >= 2:
                    ax.plot(x5[r_][ok], y5[r_][ok], "-", color=C5[g], lw=0.8, alpha=0.4, zorder=3)
            medx, medy = [], []
            for st_ in STAGES5:
                ms_ = mk & (stg5 == st_) & ~ul5 & np.isfinite(x5) & np.isfinite(y5)
                if ms_.sum() >= P5_NMIN:
                    medx.append(float(np.median(x5[ms_]))); medy.append(float(np.median(y5[ms_])))
                    ax.scatter(medx[-1], medy[-1], s=95, marker=MRK5[st_], c=C5[g], edgecolors="k", linewidths=0.8, zorder=6)
                    rows5.append(dict(block="ks_track", zone="", quantity="stage_median", kcls=g, x=st_, n_gal=int(ms_.sum()),
                                      med=medy[-1], lo=np.nan, hi=np.nan, logSigmaH2=medx[-1]))
            if len(medx) >= 2:
                ax.plot(medx, medy, "-", color=C5[g], lw=2.6, zorder=5)
            me_ = mk & (stg5 == "end")
            with np.errstate(invalid="ignore"):
                offr = y5[me_] - kl5.relation_y("B08", x5[me_])
            fin_ = np.isfinite(offr) & ~ul5[me_]
            cnt = dict(n=int(len(np.unique(gk5[mk]))), below=int((fin_ & (offr < -band5)).sum()),
                       within=int((fin_ & (np.abs(offr) <= band5)).sum()), above=int((fin_ & (offr > band5)).sum()),
                       censored=int((~fin_).sum()))
            rows5.append(dict(block="ks_region", zone="", quantity="end_vs_B08", kcls=g, x="", n_gal=cnt["n"], med=np.nan,
                              lo=np.nan, hi=np.nan, **{c: cnt[c] for c in ("below", "within", "above", "censored")}))
            print(f"  {L5[g]:26s} N={cnt['n']:3d}: below {cnt['below']}, on {cnt['within']}, above {cnt['above']}, "
                  f"censored/undefined {cnt['censored']}")
        _h5 = [Line2D([], [], color=C5[g], lw=2.4, label=f"{L5[g]} — median track") for g in G5]
        _h5 += [Line2D([], [], ls="", marker=MRK5[st_], ms=8, mfc="0.5", mec="k",
                       label={"sft": "SFT (quench start)", "qt": "QT (quench end)", "end": "track end"}[st_]) for st_ in STAGES5]
        _h5 += [Line2D([], [], ls="", marker="o", ms=5, mfc="0.65", mec="none", label="m25 SF controls"),
                Line2D([], [], color=kl5.RELATIONS["B08"]["color"], lw=1.4, ls=kl5.RELATIONS["B08"]["ls"],
                       label=kl5.RELATIONS["B08"]["label"] + r" $\pm$ scatter"),
                Line2D([], [], color=kl5.RELATIONS["K98"]["color"], lw=0.9, ls=kl5.RELATIONS["K98"]["ls"], alpha=0.55,
                       label=kl5.RELATIONS["K98"]["label"] + " (reference)")]
        ax.legend(handles=_h5, loc="lower right", frameon=False, fontsize=8)
        ax.set_xlabel(r"$\log(\Sigma_{\rm H_2}/M_\odot\,{\rm pc}^{-2})$  [$r<R_{50}({\rm H_2})$, $\times$" + f"{kl5.HE_FACTOR:g} He]")
        ax.set_ylabel(r"$\log(\Sigma_{\rm SFR}/M_\odot\,{\rm yr}^{-1}\,{\rm kpc}^{-2})$")
        ax.set_xlim(-0.5, 3.6); ax.set_ylim(-4.6, 1.1)
        ax.tick_params(direction="in", top=True, right=True)
        ax.set_title(f"{BOX_LABEL[P5_BOX]}: the two classes on the K-S plane", fontsize=10.5)
        fig.tight_layout()
        paper_save(fig, "kappa_m25_ks")
        plt.show()

    if rows5:
        _p5 = os.path.join(OUT, "kappa_m25_closeup.csv")
        pd.DataFrame(rows5).to_csv(_p5, index=False)
        print(f"\nthe m25 close-up tables -> {_p5}")
```

## Part 6 — the picture: the two classes, the gas that is left, and the dust

Parts 3–5 follow the two classes in time. This part draws the population they come from. **There is one division in it, and
it is Part 2's — both axes**: the kinematic significance cut at `P6_GCUT = ROT_SIG` and the stellar mass cut at `ROT_MASS`,
applied at the *same physical thresholds in every box* rather than at each box's own median. The AGN dose is **not** a
division here or anywhere else in the notebook: `f_x`, `f_x_post` and `E_x` are computed and written to the CSV as per-galaxy
numbers, and nothing is ever split by them.

This part, and Parts 7 and 8, are the previous notebook's Parts 22–24 with that one change. Everything else — the fan, the
walked plane, the rejuvenation test, the resolved anatomy, the gallery — is unchanged, so the two notebooks can still be
compared panel by panel.

**Figure 1.** (a) the fan: the dust on each galaxy's own quenching clock, one curve before $t_{\rm QT}$ and the two classes
after it. (b) **the rule drawn**: the dust against the kinematic significance in `P6_GBINS` equal-count bins, one line per
side of the mass cut, with the significance cut as a vertical line. The two axes of the rule are then readable separately —
the dust rises along the significance *within* each mass half, and the two halves are offset from each other — and the cell
prints both correlations ($\rho$(significance, dust), $\rho(\log M_\star$, dust)) alongside the dex the full division moves.

**Figure 2.** How much H$_2$ the two sides of the division have at each critical point, how that gas moves, and the dust–gas
plane walked in bins of time since quenching. Note that panel (b) walks a *resolved* version of the axis
(`P6_GEOM_STAGE = "kappa_z_H2"`, the same construction on the core H$_2$ with its own member count): the catalogue
significance exists only at the anchor and cannot be walked through the stages. The cell says so when it substitutes.

**Figure 3.** The rejuvenation test: the same plane stepped through the critical points, one panel per bin of the galaxy's
sSFR **at the anchor**. If the dusty galaxies were simply the ones still (or again) forming stars, the two sides of the
division would separate across the panels and coincide within them.

**Sampling note.** Figure 1's m25 column is Part 0's sample (`P6_LIKE = True`, so all three columns are cut on the same
catalogue axis), while Figures 2–3 and Parts 7–8 use the larger K-S-track frame — every quenched anchor the m25 products
carry, classified by the same `rot_rule()` read straight off the catalogue cache. The counts are printed for both, and with
the mass axis on the selected class in that frame is small (~20 of ~175), so some Figure 3 panels fall below `P6_NMIN3` and
simply do not draw.

```python
# ── Part 6 — THE PICTURE: the two classes, the gas that is left, and the dust, drawn instead of scored ──
# Parts 3-5 follow the two classes in time. This part draws the population they come from.  Nothing new is measured, and there
# is only ONE division anywhere in it — Part 2's, both of its axes: the kinematic significance and the stellar mass.
#   FIG 1  ONE COLUMN PER BOX, two rows —
#          (a) THE FAN     the dust on each galaxy's own quenching clock, the sample split by THE SELECTION: one curve before
#                          t_QT, two after it;
#          (b) THE RULE, DRAWN   the dust against the kinematic significance in equal-count bins, one line per side of the mass
#                          cut, with the significance cut as a vertical line — the two axes of the rule, and what each is worth.
#   FIG 2  (a) how much H2 the two SIDES OF THE DIVISION have at each critical point, (b) how that H2 moves — binned
#          points, because the stages are epochs of a sample, not a curve (P6_STAGES; sf_peak is out) — and (c) the
#          dust-gas plane: EVERY point of the sample as a scatter (every critical point and the anchor of every galaxy, both
#          classes), the two walked tracks in bins of TIME SINCE QUENCHING over it, and at the bottom right of each
#          selected-class point a box with the time since quenching and the number of galaxies behind the pair of points.
#   FIG 3  THE REJUVENATION TEST: the SAME walked plane as (c), but stepped through the CRITICAL POINTS
#          (jet_on -> sft -> qt -> anchor) instead of bins of time, one panel per bin of the galaxy's sSFR AT THE
#          ANCHOR -- the panel is a property of the GALAXY, not of the epoch, so a whole track stays inside one panel.
#          If the dusty galaxies were simply the ones still (or again) forming stars, the two sides of the division
#          would separate ACROSS the panels and lie on top of each other inside them.
# THE DIVISION IS PART 2'S SELECTION, both axes: "kappa_z_cat", the catalogue significance (kappa_gas - 1/3)/s_iso(N_gas), cut
# at P6_GCUT = ROT_SIG in every box — the same physical threshold everywhere, not a per-box median — ANDed with Part 0's mass
# axis ROT_MASS wherever a mass is available. THE AGN DOSE IS NOT A DIVISION here or anywhere else in this notebook: it is
# carried in the CSV as a per-galaxy number and nothing is ever split by it. "kappa_z_H2" is the same
# construction on the m25 K-S products (kappa_rot of the H2 in the core, with its own member count); the other axes of the
# notebook this part comes from are kept as options and are described at their knob.
# ONE TRACK PER GALAXY: a galaxy that is a quenched anchor at several snapshots contributes several overlapping chains, so the
# chains are de-duplicated on the main branch itself (M*(t) at three fixed times) and only the LATEST anchor of each is kept.
P6_BOX    = "cis25"           # the box of Figs 2-3 and of Part 7: the one with KS-track products
P6_BOXES  = ["cis100", "cis50", "cis25"]   # Fig 1: one column per box, in this order (a box without a histories cache is skipped)
P6_KSDIR  = globals().get("P6_KSDIR_OVERRIDE", os.path.join(os.getcwd(), "output", "cis25", "ks_tracks"))
P6_CORE   = "ap3kpc"          # the core aperture the geometry is measured in (the disc annulus is empty for a quarter of them)
P6_GEOM   = "kappa_z_cat"     # THE AXIS OF THIS NOTEBOOK: "kappa_z_cat" = Part 2's significance (kappa_gas - 1/3)/s_iso(N_gas) from the
                               # catalogue, whole galaxy, the same number in every box. The alternatives are the axes of the notebook
                               # this part comes from, all measured on the m25 K-S products inside P6_CORE: "kappa_z_H2" (the same
                               # construction on the core H2, with its own member count) | "kappa_H2" | "kappa_gas" (raw, no null) |
                               # "sigma_H2" | "sigma_R50" (the K-S abscissa) | "compactness" (= -log R50(H2)/R50(*))
P6_GCUT   = ROT_SIG           # where the axis is split. With a significance axis this is A NUMBER OF SIGMA and Part 2's cut is the
                               # default, so every box is cut at the SAME physical threshold; "median" restores the per-box median
                               # split the other axes use (for kappa: 1/3 = isotropic, 0.5 = rotation-supported)
P6_NH2MIN = 0                 # a galaxy needs this many member H2 particles for its geometry to count in the split (0 = every galaxy);
                               # the cell prints the picture at P6_NH2CHK regardless
P6_NH2CHK = 100               # the resolution check printed for the chosen axis: the same split inside n_H2 >= this
P6_GEOM_STAGE = "kappa_z_H2"  # Fig 2 (b) walks the geometry axis THROUGH the critical points, which a catalogue quantity cannot do
                               # (it exists only at the anchor). Where P6_GEOM is not measured along the track, this resolved
                               # counterpart is drawn there instead and the panel says so; "kappa_H2" is the un-normalised version
P6_GEOM_CAT = "kappa_z_cat"   # Fig 1, the boxes WITHOUT KS-track products: the catalogue axis ("kappa_z_cat" | "kappa_gas" | "kappa_star")
P6_LIKE   = True              # Fig 1: True = the m25 column too is cut on P6_GEOM_CAT over Part 0's sample, so the three columns are
                               # strictly like for like — which is the point here, since P6_GEOM is itself the catalogue axis;
                               # False = m25 keeps the KS-track geometry (P6_GEOM in P6_CORE) and sample of Figs 2-3 / Part 7
P6_DUST   = "dh2"             # the dust axis of (a) and (b): "dh2" = log M_dust/M_H2 | "lfd" = log M_dust/M* | "dg" = log M_dust/M_gas
P6_STAGES = ["agn_ign", "jet_on", "sft", "qt", "anchor"]   # the critical points of Fig 2 (a) and (b) (sf_peak dropped; put it back to see it)
P6_PLANE  = "dust"            # Fig 2 (c): "dust" = the Sigma_dust-Sigma_H2 plane | "ks" = the Kennicutt-Schmidt plane
P6_SCAT_STAGES = None         # Fig 2 (c): the stage rows drawn as the scatter (None = every critical point and the anchor of every galaxy)
P6_NMIN   = 8                 # an epoch / a bin is drawn only where this many galaxies contribute
P6_NMIN3  = 5                 # Fig 3's own floor: a panel x half x critical point is a thinner cell than a Fig 2 epoch
P6_TG     = np.arange(-2.0, 2.01, 0.2)                  # the quenching clock of Fig 1 (a), t - t_QT [Gyr]
P6_GBINS  = 6                 # Fig 1 (b): equal-count bins of the significance axis, inside each side of the mass cut (fewer
                               # where a box cannot fill that many at P6_NMIN galaxies a bin)
P6_ZLIM   = (-25.0, 25.0)     # Fig 1 (b): the significance is drawn on a LINEAR axis clipped to this, as in Part 2 (b) — the
                               # axis runs to +-100 and a log axis makes an unreadable panel out of a monotonic trend
P6_TQT    = [-2.0, -1.0, -0.5, 0.0, 0.5, 1.0, 2.0, 4.0]  # Fig 2 (c): the bin edges of t - t_QT [Gyr] the plane is walked in
P6_SSTAGE = ["jet_on", "sft", "qt", "anchor"]      # Fig 3: the critical points the plane is walked through, in time order
P6_SEDGES = None              # Fig 3: the log sSFR panel edges (the ANCHOR sSFR, per galaxy); None = terciles of the galaxies
P6_NBOOT  = 300
P6_GLAB   = {"kappa_z_cat": (r"$(\kappa_{\rm gas} - 1/3)\,/\,s_{\rm iso}(N_{\rm gas})$", "rotating gas", "no resolved rotation"),
              "kappa_z_H2": (r"$(\kappa_{\rm H_2} - 1/3)\,/\,s_{\rm iso}(N_{\rm H_2})$, core", "rotating H$_2$", "no resolved rotation"),
              "kappa_H2": (r"$\kappa_{\rm rot}$(H$_2$), core", "higher-$\kappa$ H$_2$", "lower-$\kappa$ H$_2$"),
              "kappa_gas": (r"$\kappa_{\rm rot}$(gas), core", "higher-$\kappa$ gas", "lower-$\kappa$ gas"),
              "sigma_H2": (r"$\log \Sigma_{\rm H_2}$, core", "dense H$_2$", "diffuse H$_2$"),
              "sigma_R50": (r"$\log \Sigma_{\rm H_2}(<R_{50})$", "dense H$_2$", "diffuse H$_2$"),
              "compactness": (r"$-\log\,R_{50}({\rm H_2})/R_{50}(\star)$", "compact H$_2$", "extended H$_2$")}
P6_GTITLE = {"kappa_z_cat": "(b) how that gas moves", "kappa_z_H2": "(b) how that gas moves", "kappa_H2": "(b) how that gas moves", "kappa_gas": "(b) how that gas moves", "sigma_H2": "(b) how dense that gas is",
              "sigma_R50": "(b) how dense that H$_2$ is inside its own $R_{50}$",
              "compactness": "(b) how compact that H$_2$ is, relative to the stars"}

def _massify6(lab3):
    """A (axis label, upper-side name, lower-side name) triple with Part 0's mass axis folded into the names, so that every
    legend in Parts 6-8 says what the rule actually is."""
    lab, hi, lo = lab3
    return (lab, hi + rf", $\log M_\star > {ROT_MASS:g}$", "the rest") if ROT_MASS is not None else lab3


def _glab6(geom, gcut, vals):
    """The axis label and the names of its two sides. A kappa axis names the cut itself — the split is NOT 'rotating' against
    'dispersion-supported': kappa_rot = 1/3 is an isotropic velocity field, a disc sits above ~0.5 — the other axes keep their words."""
    lab, hi, lo = P6_GLAB[geom]
    if geom.startswith("kappa_z"):                 # the significance axis: the two sides ARE Part 2's classes, and keep its
        _, hi, lo = _massify6((lab, hi, lo))       # words — including the mass axis, which is ANDed into the rule (Part 0)
    elif geom.startswith("kappa"):
        c = float(np.nanmedian(vals)) if gcut == "median" else float(gcut)
        what = "H$_2$" if geom == "kappa_H2" else "gas"
        hi, lo = what + rf" $\kappa_{{\rm rot}} > {c:.2f}$", what + rf" $\kappa_{{\rm rot}} \leq {c:.2f}$"
    return lab, hi, lo
P6_GLAB_CAT = {"kappa_z_cat": P6_GLAB["kappa_z_cat"],
                "kappa_gas": (r"$\kappa_{\rm rot}$(gas), whole galaxy", "rotating gas", "dispersion-supported gas"),
                "kappa_star": (r"$\kappa_{\rm rot}$(stars), whole galaxy", "rotating stars", "dispersion-supported stars")}
P6_ULAB   = {"lfd": r"$\log\, M_{\rm dust}/M_\star$", "dg": r"$\log\, M_{\rm dust}/M_{\rm gas}$",
              "dh2": r"$\log\, M_{\rm dust}/M_{\rm H_2}$"}

_need6 = [os.path.join(P6_KSDIR, f) for f in ("ks_track_measurements.fits", "ks_stage_kinematics.fits", "ks_track_epochs.fits")]
if "CAT" not in globals():
    print("Part 6: needs Part 0 — run it first.")
elif not all(os.path.exists(p) for p in _need6):
    print("Part 6: the m25 KS-track products are missing — ks_tracks_quenched_m25.ipynb builds them:\n  "
          + "\n  ".join(p for p in _need6 if not os.path.exists(p)))
elif not os.path.exists(HIST_FILE.format(box=P6_BOX)):
    print(f"Part 6: {HIST_FILE.format(box=P6_BOX)} is missing — paper_ism_prediction_boxes Part 6a builds it.")
else:
    def _s6(t):
        d = t.to_pandas() if hasattr(t, "to_pandas") else t
        for c in d.columns:
            if d[c].dtype == object:
                d[c] = d[c].map(lambda v: v.decode() if isinstance(v, bytes) else v)
        return d

    def _band6(A, nmin=None):
        """A (n_gal, n_x) -> median track, its 16-84 bootstrap band over GALAXIES, and the per-column count."""
        A = np.asarray(A, float)
        nmin = P6_NMIN if nmin is None else nmin
        if A.ndim != 2 or not len(A):
            return (np.full(A.shape[-1] if A.ndim == 2 else 1, np.nan),) * 3 + (np.zeros(1, int),)
        n = np.isfinite(A).sum(axis=0)
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", RuntimeWarning)
            cent = np.nanmedian(A, axis=0)
            rng = np.random.default_rng(22)
            bs = np.array([np.nanmedian(A[rng.integers(0, len(A), len(A))], axis=0) for _ in range(P6_NBOOT)])
            lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
        ok = n >= nmin
        return np.where(ok, cent, np.nan), np.where(ok, lo, np.nan), np.where(ok, hi, np.nan), n

    def _pt6(v, nmin=None):
        """One bin of per-galaxy values -> median and half the 16-84 bootstrap width (nan below the floor)."""
        v = np.asarray(v, float); v = v[np.isfinite(v)]
        if len(v) < (P6_NMIN if nmin is None else nmin):
            return np.nan, np.nan, len(v)
        rng = np.random.default_rng(22)
        bs = np.array([np.median(v[rng.integers(0, len(v), len(v))]) for _ in range(P6_NBOOT)])
        return float(np.median(v)), float(0.5 * (np.percentile(bs, 84) - np.percentile(bs, 16))), len(v)

    MEA6, KIN6, EPO6 = (_s6(Table.read(p)) for p in _need6)
    EPO6 = EPO6[EPO6["pop"] == "Q"].copy()
    EPO6["key"] = list(zip(EPO6["anchor_snap"].astype(int), EPO6["gal_id"].astype(int)))
    _C6 = MEA6[MEA6["ap_label"] == P6_CORE].set_index(["snap", "gx"])
    _K6 = KIN6[KIN6["zone"] == P6_CORE].set_index(["snap", "gx"])

    def _geom6(snaps, gxs):
        """The structural block of a set of (snap, gx) rows, in the core aperture."""
        ix = list(zip(np.asarray(snaps, int), np.asarray(gxs, int)))
        g = lambda df, c: df[c].reindex(ix).to_numpy(float) if c in df.columns else np.full(len(ix), np.nan)
        with np.errstate(all="ignore"):
            mh2, md, mst, ar, mgs = g(_C6, "m_H2"), g(_C6, "m_dust"), g(_C6, "m_star"), g(_C6, "area_kpc2"), g(_C6, "m_gas")
            sf100, sf_tot, ms_tot = g(_C6, "sfr100"), g(_C6, "sfr100_tot"), g(_C6, "m_star_tot")
            out = dict(sigma_H2=np.log10(np.where(mh2 > 0, mh2 / ar, np.nan)),
                       sigma_SFR=np.log10(np.where(sf100 > 0, sf100 / ar, np.nan)),
                       sigma_dust=np.log10(np.where(md > 0, md / ar, np.nan)),
                       lfd_core=np.log10(np.where((md > 0) & (mst > 0), md / np.where(mst > 0, mst, np.nan), np.nan)),
                       dg_core=np.log10(np.where((md > 0) & (mgs > 0), md / np.where(mgs > 0, mgs, np.nan), np.nan)),
                       dh2_core=np.log10(np.where((md > 0) & (mh2 > 0), md / np.where(mh2 > 0, mh2, np.nan), np.nan)),
                       kappa_H2=g(_K6, "kappa_H2"), kappa_gas=g(_K6, "kappa_gas"), kappa_star=g(_K6, "kappa_star"),
                       kappa_z_H2=kappa_z(g(_K6, "kappa_H2"), g(_C6, "n_H2_tot")),      # Part 2's construction, on the core H2
                       lssfr_tot=np.log10(np.where((sf_tot > 0) & (ms_tot > 0), sf_tot / np.where(ms_tot > 0, ms_tot, np.nan), np.nan)),
                       compactness=-np.log10(g(_C6, "r50_H2") / np.where(g(_C6, "r50_star") > 0, g(_C6, "r50_star"), np.nan)),
                       sigma_R50=np.log10(np.where((g(_C6, "m_H2_tot") > 0) & (g(_C6, "r50_H2") > 0),          # the K-S abscissa:
                                                   1.36 * 0.5 * g(_C6, "m_H2_tot") / (np.pi * g(_C6, "r50_H2") ** 2) / 1e6, np.nan)),
                       n_H2_tot=g(_C6, "n_H2_tot"))                                                              # He x 0.5 M / pi R50^2 [Msun/pc^2]
        return pd.DataFrame(out)

    # ── the chains, the anchor block and the de-duplication: the same three steps for every box ──
    def _chains6(box, want=None):
        """The main-branch chains of one box's histories cache, keyed (anchor_snap, gal_id): the AGN clock, the dust ratio, the
        sSFR and f_gas along the branch, plus the branch fingerprint (log M* at three fixed times) the de-duplication uses.
        `want` restricts to a set of keys — the coarse boxes hold thousands of chains per anchor."""
        CH, FP = {}, {}
        with h5py.File(HIST_FILE.format(box=box), "r") as f:
            for k in f.keys():
                g, a = f[k], int(k[4:])
                t = g["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
                gids = np.asarray(g["gal_ids"][:], dtype=np.int64)
                js = [j for j, gid in enumerate(gids) if want is None or (a, int(gid)) in want]
                if not js:
                    continue
                ms, mg, mb, fe = (g[c][:][o] for c in ("mstar", "mgas", "mbh", "fedd"))
                sf = g["sfr"][:][o] if "sfr" in g else np.full_like(ms, np.nan)
                md = g["mdust"][:][o] if "mdust" in g else np.full_like(ms, np.nan)
                mh = g["mh2"][:][o] if "mh2" in g else np.full_like(ms, np.nan)
                S, _, X = agn_states(ms, mg, mb, fe)
                dt = np.gradient(t)
                for j in js:
                    gid, ex = int(gids[j]), X["exist"][:, j]
                    if ex.sum() < 3:
                        continue
                    xr = (np.nan_to_num(S["xray"][:, j]) > 0.5) & ex
                    with np.errstate(all="ignore"):
                        rat = lambda num, den: np.log10(np.where((num > 0) & (den > 0), num / np.where(den > 0, den, np.nan), np.nan))
                        fd = {"lfd": rat(md[:, j], ms[:, j]), "dg": rat(md[:, j], mg[:, j]), "dh2": rat(md[:, j], mh[:, j])}[P6_DUST]
                        ss = np.log10(np.where((sf[:, j] > 0) & (ms[:, j] > 0), sf[:, j] / np.where(ms[:, j] > 0, ms[:, j], np.nan), np.nan))
                    last = np.maximum.accumulate(np.where(xr, t, -np.inf))       # the last X-ray snapshot up to each step
                    tsx = np.where(np.isfinite(last), t - last, np.nan)          # time since the last X-ray episode, along the branch
                    CH[(a, gid)] = dict(t=t, ex=ex, xr=xr, fd=fd, dt=dt, tsx=tsx, ssfr=ss, fgas=X["fgas"][:, j])
                    FP[(a, gid)] = tuple(np.round(np.log10(np.interp([0.8, 1.5, 2.5], t[ex], ms[ex, j],
                                                                     left=np.nan, right=np.nan)), 5))   # the branch fingerprint
        return CH, FP

    def _anchor6(CH, keys, t_sft):
        """The per-anchor block of a set of galaxies, from their chains: the dust ratio and f_gas AT THE ANCHOR (what the
        figures are made of), plus the AGN exposure E_x, f_x, f_x_post, the state at the anchor and t_quiet. The exposure is
        CARRIED, never split on: this notebook has one division and it is Part 2's."""
        rows = []
        for k, ts in zip(keys, np.asarray(t_sft, float)):
            d = CH[k]; t, ex, xr, dtc = d["t"], d["ex"], d["xr"], d["dt"]
            ka = int(np.flatnonzero(ex)[-1])
            post = ex & (t >= ts) if np.isfinite(ts) else ex
            Tp, Ta = float(np.sum(dtc[post])), float(np.sum(dtc[ex]))
            rows.append(dict(e_x=float(np.sum(dtc[xr])), f_x=float(np.sum(dtc[xr]) / Ta) if Ta > 0 else np.nan,
                             f_x_post=float(np.sum(dtc[xr & post]) / Tp) if Tp > 0 else np.nan,
                             xray_now=float(xr[ka]), t_quiet=float(t[ka] - t[xr][-1]) if xr.any() else np.nan,
                             f_gas_anchor=float(d["fgas"][ka]), dust=float(d["fd"][ka])))
        return pd.DataFrame(rows)

    def _hi6(geom, gcut, lm):
        """PART 2'S RULE applied to whichever axis P6_GEOM names, on rows that are not Part 0's sample (the K-S products carry
        their own, larger anchor set): above the cut AND — when Part 0's mass axis is on — above ROT_MASS. Returns the two
        sides; both are False where either number is missing, so a galaxy without a geometry is in neither group."""
        g, l = np.asarray(geom, float), np.asarray(lm, float)
        ok = np.isfinite(g) & (np.isfinite(l) if ROT_MASS is not None else True)
        hi = ok & (g > float(gcut))
        if ROT_MASS is not None:
            hi = hi & (l > float(ROT_MASS))
        return hi, ok & ~hi

    def _dedup6(A, FP):
        """One chain per galaxy: anchors sharing a main-branch fingerprint are the same galaxy — keep the LATEST anchor."""
        A = A.copy(); A["fp"] = [FP[k] for k in A["key"]]
        return A.sort_values("anchor_snap").groupby("fp", as_index=False).tail(1).reset_index(drop=True)

    def _fan6(A, CH):
        """The dust track of every anchor on its own quenching clock t - t_QT, interpolated onto P6_TG."""
        TR = []
        for _, r in A.iterrows():
            d = CH[r["key"]]; ok = d["ex"] & np.isfinite(d["fd"])
            TR.append(np.interp(P6_TG, d["t"][ok] - r["t_qt_gyr"], d["fd"][ok], left=np.nan, right=np.nan)
                      if (ok.sum() >= 3 and np.isfinite(r["t_qt_gyr"])) else np.full(len(P6_TG), np.nan))
        return np.asarray(TR, float).reshape(len(A), len(P6_TG))

    # ── the m25 frame Figs 2-3 and Part 7 build on: the KS-track anchors, the KS-track geometry, one chain per galaxy ──
    _CH6, _fp6 = _chains6(P6_BOX)
    ANC6 = EPO6[EPO6["stage"] == "anchor"][["key", "snap", "gx", "gal_id", "anchor_snap", "z", "t_sft_gyr", "t_qt_gyr",
                                               "log_mstar_anchor"]].copy()
    ANC6 = ANC6[[k in _CH6 for k in ANC6["key"]]].copy()
    _n_rows6 = len(ANC6)
    ANC6 = _dedup6(ANC6, _fp6)                                                       # one per galaxy
    ANC6 = pd.concat([ANC6.reset_index(drop=True), _geom6(ANC6["snap"], ANC6["gx"]).reset_index(drop=True)], axis=1)
    # Part 2's axis for EVERY anchor of the KS products, not only for those inside Part 0's sample: the KS-track frame is its own
    # (bigger) sample, and the rule is a rule — it is read straight off the catalogue cache with the same two columns.
    _CZ6 = CAT[CAT["box"] == P6_BOX]
    _KZ6 = {(int(s_), int(i_)): float(v_) for s_, i_, v_ in
            zip(_CZ6["snap"], _CZ6["gal_id"], kappa_z(_CZ6[KAPPA_COL].to_numpy(float), _CZ6[NGAS_COL].to_numpy(float)))}
    ANC6["kappa_z_cat"] = [_KZ6.get((int(a_), int(g_)), np.nan) for a_, g_ in zip(ANC6["anchor_snap"], ANC6["gal_id"])]
    ANC6 = pd.concat([ANC6, _anchor6(_CH6, ANC6["key"], ANC6["t_sft_gyr"])], axis=1)
    _geom_v = ANC6[P6_GEOM].to_numpy(float)
    _nh2_v = ANC6["n_H2_tot"].to_numpy(float)
    if P6_NH2MIN > 0:                                   # the resolution floor: too few H2 particles -> no geometry, no group
        _geom_v = np.where(_nh2_v >= P6_NH2MIN, _geom_v, np.nan)
    _dust_v = ANC6["dust"].to_numpy(float)
    _mass_v = ANC6["log_mstar_anchor"].to_numpy(float)
    _glab, _ghi, _glo = _glab6(P6_GEOM, P6_GCUT, _geom_v)
    _gcut = float(np.nanmedian(_geom_v)) if P6_GCUT == "median" else float(P6_GCUT)
    ANC6["hi_geom"], ANC6["lo_geom"] = _hi6(_geom_v, _gcut, _mass_v)              # both False where either number is missing
    print(f"Part 6 — the picture, {BOX_LABEL.get(P6_BOX, P6_BOX)}: {_n_rows6} quenched anchor rows -> {len(ANC6)} distinct "
          f"galaxies (one chain each, the latest anchor).  THE ONE DIVISION: {P6_GEOM}" + (f" in {P6_CORE}" if P6_GEOM in ("kappa_H2", "kappa_gas", "sigma_H2") else " (galaxy-wide)")
          + f", split at {_gcut:.2f}" + (" (the median)" if P6_GCUT == "median" else "")
          + (f" AND log M* > {ROT_MASS:g}" if ROT_MASS is not None else "")
          + (f", geometry only above {P6_NH2MIN} H2 particles ({int(np.isfinite(_geom_v).sum())} galaxies)" if P6_NH2MIN > 0 else "") + ".")
    # ── the resolution check of the geometry axis (2026-09-04): the old kappa signal lived in the few-particle galaxies ──
    _k6 = ANC6["kappa_H2"].to_numpy(float)
    print(f"  kappa_rot(H2) in {P6_CORE}: median {np.nanmedian(_k6):.2f}; {100 * np.nanmean(_k6 > 1 / 3):.0f} % above the isotropic "
          f"1/3 (net rotation), {100 * np.nanmean(_k6 > 0.5):.0f} % above 0.5 (rotation-supported). Member H2 particles: median "
          f"{np.nanmedian(_nh2_v):.0f}, {100 * np.nanmean(_nh2_v < P6_NH2CHK):.0f} % below {P6_NH2CHK}.")
    for _nm6, _sel6 in (("every galaxy", np.ones(len(ANC6), bool)), (f"n_H2 >= {P6_NH2CHK}", _nh2_v >= P6_NH2CHK)):
        _a6 = _sel6 & ANC6["hi_geom"].to_numpy(bool) & np.isfinite(_dust_v)
        _b6 = _sel6 & ANC6["lo_geom"].to_numpy(bool) & np.isfinite(_dust_v)
        _gap6 = ((np.nanmedian(_dust_v[_a6]) - np.nanmedian(_dust_v[_b6]))
                 if (_a6.sum() >= P6_NMIN and _b6.sum() >= P6_NMIN) else np.nan)
        _ok6 = _sel6 & np.isfinite(_geom_v) & np.isfinite(_dust_v) & (_nh2_v > 0)
        print(f"  {P6_GEOM} vs {P6_DUST}, {_nm6:14s} (N {int(_ok6.sum()):3d}): Spearman "
              f"{spearmanr(_geom_v[_ok6], _dust_v[_ok6])[0]:+.2f}, with log n_H2 {spearmanr(_geom_v[_ok6], np.log10(_nh2_v[_ok6]))[0]:+.2f}; "
              f"the selected class minus the rest: {_gap6:+.2f} dex ({int(_a6.sum())} vs {int(_b6.sum())} galaxies)")

    GRP6 = [("sel", ANC6["hi_geom"], KCLS["rot"]["color"], "-", _ghi),
            ("rest", ANC6["lo_geom"], KCLS["iso"]["color"], "-", _glo)]
    _SPLITCOL = "hi_geom"                        # THE column Figs 2-3 walk their two tracks in: there is only one division

    # ── FIGURE 1: the fan (a) and the rule (b), one column per box ──
    # The m25 column is ANC6 itself (the KS-track geometry, Fig 2-3's frame) unless P6_LIKE. The coarse boxes have no KS-track
    # products, so their anchors are Part 0's sample (QM, with t_SFT / t_QT from the boxes notebook's classes) and their geometry
    # axis is the catalogue's whole-galaxy kappa_rot (P6_GEOM_CAT) — de-duplicated exactly as the m25 frame, and cut at the
    # SAME two thresholds (P6_GCUT in units of s_iso, and ROT_MASS), which is what makes the three columns comparable.
    FIG1_6 = {}
    for _b in P6_BOXES:
        if _b == P6_BOX and not P6_LIKE:
            _A = ANC6.assign(geom=_geom_v, lm=_mass_v); _CHb = _CH6         # the floored axis, the same split as the frame
            _lab3, _gsrc = (_glab, _ghi, _glo), (f"{P6_GEOM} in {P6_CORE}" if P6_GEOM in ("kappa_H2", "kappa_gas", "sigma_H2")
                                                  else f"{P6_GEOM}, galaxy-wide") + " (KS-track products)"
            print(f"  Fig 1 {_b}: the frame above; geometry = {_gsrc}")
        else:
            if not os.path.exists(HIST_FILE.format(box=_b)):
                print(f"  Fig 1 {_b}: no histories cache ({HIST_FILE.format(box=_b)}) — column skipped"); continue
            if not (HAVE_MD.get(_b, False) and (P6_DUST != "dh2" or HAVE_MH2.get(_b, False))):
                print(f"  Fig 1 {_b}: the histories carry no mdust{' / mh2' if P6_DUST == 'dh2' else ''} — no dust axis, column skipped"); continue
            if not {"t_sft", "t_qt", P6_GEOM_CAT} <= set(QM.columns) or not (QM["box"] == _b).any():
                print(f"  Fig 1 {_b}: QM has no rows for it or lacks t_sft / t_qt / {P6_GEOM_CAT} — column skipped"); continue
            _Q = QM.loc[QM["box"] == _b, ["snap", "gal_id", "z", "t_sft", "t_qt", "age", "log_mstar", P6_GEOM_CAT]].copy()
            _Q["key"] = list(zip(_Q["snap"].astype(int), _Q["gal_id"].astype(int)))
            _CHb, _FPb = (_CH6, _fp6) if _b == P6_BOX else _chains6(_b, want=set(_Q["key"]))
            _A = _Q[[k in _CHb for k in _Q["key"]]].rename(columns={"t_sft": "t_sft_gyr", "t_qt": "t_qt_gyr"}).copy()
            _A["anchor_snap"] = _A["snap"].astype(int)
            _nrow = len(_A)
            _A = _dedup6(_A, _FPb)
            _A = pd.concat([_A.reset_index(drop=True), _anchor6(_CHb, _A["key"], _A["t_sft_gyr"])], axis=1)
            _A["geom"] = _A[P6_GEOM_CAT].to_numpy(float)
            _A["lm"] = _A["log_mstar"].to_numpy(float)                      # the rule's second axis, from Part 0's own frame
            _lab3 = _massify6(P6_GLAB_CAT[P6_GEOM_CAT]) if P6_GEOM_CAT.startswith("kappa_z") else P6_GLAB_CAT[P6_GEOM_CAT]
            _gsrc = f"catalogue {P6_GEOM_CAT} (whole galaxy)"
            print(f"  Fig 1 {_b}: {_nrow} quenched anchor rows with a chain -> {len(_A)} distinct galaxies "
                  f"(Part 0's sample); geometry = {_gsrc}")
        _gv, _uv, _lv = _A["geom"].to_numpy(float), _A["dust"].to_numpy(float), _A["lm"].to_numpy(float)
        if np.isfinite(_gv).sum() < P6_NMIN or np.isfinite(_uv).sum() < P6_NMIN:
            print(f"  Fig 1 {_b}: fewer than {P6_NMIN} galaxies with a geometry and a dust value — column skipped"); continue
        _gc = (_gcut if (_b == P6_BOX and not P6_LIKE)                                 # the m25 frame keeps its own cut (P6_GCUT)
               else (float(np.nanmedian(_gv)) if P6_GCUT == "median" else float(P6_GCUT)))   # ... and a NUMBER is the same in every box
        _hi1, _lo1 = _hi6(_gv, _gc, _lv)
        FIG1_6[_b] = dict(A=_A, TR=_fan6(_A, _CHb), geom=_gv, dust=_uv, lm=_lv, gcut=_gc,
                           hi_geom=_hi1, lo_geom=_lo1, lab=_lab3, src=_gsrc)
        print(f"           the significance split at {_gc:.2f}"
              + (f", the mass at log M* = {ROT_MASS:g}" if ROT_MASS is not None else "")
              + f"  ->  {int(_hi1.sum())} selected, {int(_lo1.sum())} the rest")
        _fin6 = np.isfinite(_uv) & np.isfinite(_gv)
        if _fin6.sum() > 20:                       # WHAT THE FIGURE SHOWS, as a number: how far the division moves the dust
            _gapK = np.nanmedian(_uv[_fin6 & _hi1]) - np.nanmedian(_uv[_fin6 & _lo1])
            _rho_g = spearmanr(_gv[_fin6], _uv[_fin6])[0]
            _rho_m = (spearmanr(_lv[_fin6 & np.isfinite(_lv)], _uv[_fin6 & np.isfinite(_lv)])[0]
                      if np.isfinite(_lv[_fin6]).sum() > 10 else np.nan)
            print(f"           the division moves {P6_DUST} by {_gapK:+.2f} dex;  rho(significance, {P6_DUST}) = "
                  f"{_rho_g:+.2f}, rho(log M*, {P6_DUST}) = {_rho_m:+.2f} — the two axes of the rule, separately")
    _bx6 = [b for b in P6_BOXES if b in FIG1_6]
    if _bx6:
        fig, axs = plt.subplots(2, len(_bx6), figsize=(5.5 * len(_bx6), 9.2), sharey="row", squeeze=False,
                                constrained_layout=True)
        for _j, _b in enumerate(_bx6):
            F1 = FIG1_6[_b]; _gl1, _gh1, _glo1 = F1["lab"]
            _grp = [(F1["hi_geom"], KCLS["rot"]["color"], "-", _gh1), (F1["lo_geom"], KCLS["iso"]["color"], "-", _glo1)]
            # (a) the fan
            ax = axs[0, _j]
            for m, c, ls, lab in _grp:
                cen, lo, hi, n = _band6(F1["TR"][np.asarray(m, bool)])
                ax.fill_between(P6_TG, lo, hi, color=c, alpha=0.18, lw=0, zorder=3)
                ax.plot(P6_TG, cen, ls, color=c, lw=2.8, zorder=4, label=f"{lab}  ({int(np.sum(m))})")
            ax.axvline(0, color="k", lw=1.4, ls=":")
            ax.annotate("quenched", xy=(0, 1.0), xycoords=("data", "axes fraction"), xytext=(3, -11), textcoords="offset points",
                        fontsize=8.5, color="0.35")
            ax.set_xlabel(r"$t - t_{\rm QT}$   [Gyr]")
            if _j == 0:
                ax.set_ylabel(P6_ULAB[P6_DUST])
            ax.legend(frameon=False, fontsize=8.2, loc="lower left", bbox_to_anchor=(-0.01, -0.02)); ax.grid(alpha=0.22, lw=0.5)
            ax.set_title(f"(a) {BOX_LABEL.get(_b, _b)}\n{len(F1['A'])} quenched galaxies, geometry = {_gl1}", fontsize=10)
            # (b) THE RULE DRAWN: the dust against the significance, one line per side of the mass cut
            ax = axs[1, _j]
            _lines6 = ([(F1["lm"] > ROT_MASS, KCLS["rot"]["color"], rf"$\log M_\star > {ROT_MASS:g}$"),
                        (F1["lm"] <= ROT_MASS, KCLS["iso"]["color"], rf"$\log M_\star \leq {ROT_MASS:g}$")]
                       if ROT_MASS is not None else [(np.isfinite(F1["lm"]) | True, "k", "every galaxy")])
            _gx6 = np.clip(F1["geom"], *P6_ZLIM)          # clipped for DRAWING only; every cut and split uses the raw value
            for _sel6, _col6, _lab6 in _lines6:
                _gm = np.asarray(_sel6, bool) & np.isfinite(F1["geom"]) & np.isfinite(F1["dust"])
                if _gm.sum() < P6_NMIN:
                    continue
                _nb6 = int(np.clip(int(_gm.sum()) // P6_NMIN, 2, P6_GBINS))     # a bin the box cannot fill draws nothing at all
                _q = np.nanquantile(_gx6[_gm], np.linspace(0, 1, _nb6 + 1)); _q[0] -= 1e-9
                _x, _y, _e = [], [], []
                for _i in range(_nb6):
                    _m = _gm & (_gx6 > _q[_i]) & (_gx6 <= _q[_i + 1])
                    _c, _s, _n = _pt6(F1["dust"][_m])
                    if np.isfinite(_c):
                        _x.append(float(np.nanmedian(_gx6[_m]))); _y.append(_c); _e.append(_s)
                ax.errorbar(_x, _y, yerr=_e, marker="o", ms=6.5, lw=2.2, capsize=3, color=_col6,
                            label=_lab6 + f"  ({int(_gm.sum())}, {_nb6} bins)")
            ax.axvline(F1["gcut"], color="k", lw=1.6, ls="--")
            ax.annotate("the cut", xy=(F1["gcut"], 0.02), xycoords=("data", "axes fraction"), xytext=(4, 0),
                        textcoords="offset points", fontsize=8.5, color="0.3")
            ax.set_xlim(*P6_ZLIM)
            ax.set_xlabel(_gl1 + f"   (clipped to $\pm{P6_ZLIM[1]:g}$)")
            if _j == 0:
                ax.set_ylabel("median " + P6_ULAB[P6_DUST])
            ax.legend(frameon=False, fontsize=8.4, loc="upper left"); ax.grid(alpha=0.22, lw=0.5)
            ax.set_title(f"(b) {BOX_LABEL.get(_b, _b)}", fontsize=10)
        fig.suptitle("Part 6 — (a) the dust on each galaxy's own quenching clock, split by THE SELECTION: one galaxy before "
                     "quenching, two classes after it;\n(b) THE RULE DRAWN — the dust against the kinematic significance in "
                     "equal-count bins (as many as each box can fill), one line per side of the mass cut, the cut vertical",
                     fontsize=11.5)
        paper_save(fig, "kappa_dustclock_boxes")
        plt.show()
        for _b in _bx6:
            FIG1_6[_b]["A"].drop(columns=[c for c in ("key", "fp") if c in FIG1_6[_b]["A"].columns]).to_csv(
                os.path.join(OUT, f"kappa_dustclock_{_b}.csv"), index=False)
    else:
        print("  Fig 1: no box could be drawn.")

    # ── FIGURE 2: the two sides of the division at the critical points, and the plane they walk in time ──
    _SG6 = EPO6[EPO6["key"].isin(set(ANC6["key"]))].copy()
    _SG6 = pd.concat([_SG6.reset_index(drop=True), _geom6(_SG6["snap"], _SG6["gx"]).reset_index(drop=True)], axis=1)
    _hd6 = {k: v for k, v in zip(ANC6["key"], ANC6["hi_geom"])}
    _qt6 = {k: v for k, v in zip(ANC6["key"], ANC6["t_qt_gyr"])}
    _SG6["hi_geom"] = [_hd6.get(k, np.nan) for k in _SG6["key"]]
    _SG6["dtq"] = _SG6["t_stage_gyr"].to_numpy(float) - np.array([_qt6.get(k, np.nan) for k in _SG6["key"]], float)
    _st_ok = [s for s in P6_STAGES if (_SG6["stage"] == s).sum() >= P6_NMIN]
    _halves = [(True, KCLS["rot"]["color"], _ghi), (False, KCLS["iso"]["color"], _glo)]
    fig = plt.figure(figsize=(14.6, 7.6), constrained_layout=True)
    gs = fig.add_gridspec(2, 2, width_ratios=[1.0, 1.25])
    _gstage6 = P6_GEOM if P6_GEOM in _SG6.columns else P6_GEOM_STAGE      # the axis that CAN be walked through the stages
    _glab_st6 = _glab6(_gstage6, P6_GCUT, _SG6[_gstage6].to_numpy(float))[0] if _gstage6 in P6_GLAB else _gstage6
    if _gstage6 != P6_GEOM:
        print(f"  Fig 2 (b): {P6_GEOM} is an anchor quantity, so the panel walks {_gstage6} — the same construction on the "
              f"resolved core H2 — through the critical points instead.")
    for _p, (_c6, _yl, _tt) in enumerate([("sigma_H2", r"$\log\, \Sigma_{\rm H_2}$  [$M_\odot$ kpc$^{-2}$]",
                                            "(a) how much H$_2$ is in the core"),
                                           (_gstage6, _glab_st6, P6_GTITLE.get(_gstage6, "(b) the geometry axis"))]):
        ax = fig.add_subplot(gs[_p, 0])
        for _k3, (_hd, _col, _lab) in enumerate(_halves):
            _keys = set(ANC6.loc[ANC6[_SPLITCOL] == _hd, "key"])
            _x, _y, _e = [], [], []
            for _i6, s_ in enumerate(_st_ok):
                _v = _SG6.loc[(_SG6["stage"] == s_) & (_SG6["key"].isin(_keys)), _c6].to_numpy(float)
                _c, _s, _n = _pt6(_v)
                _x.append(_i6 + (0.09 if _k3 else -0.09)); _y.append(_c); _e.append(_s)
            ax.errorbar(_x, _y, yerr=_e, ls="none", marker="o", ms=8, capsize=3.5, lw=2.0, color=_col,
                        label=f"{_lab}  ({int((ANC6[_SPLITCOL] == _hd).sum())})")
        ax.set_xticks(range(len(_st_ok))); ax.set_xticklabels([s.replace("_", "\n") for s in _st_ok], fontsize=9)
        ax.set_xlim(-0.5, len(_st_ok) - 0.5)
        if "qt" in _st_ok:
            ax.axvline(_st_ok.index("qt"), color="k", lw=1.2, ls=":")
        ax.set_ylabel(_yl); ax.grid(alpha=0.22, lw=0.5, axis="y"); ax.set_title(_tt)
        ax.margins(y=0.14)
        if _p == 0:
            ax.legend(frameon=False, fontsize=9, loc="upper left")
    # (c) the plane: every point of the sample behind, the two walked tracks over it
    ax = fig.add_subplot(gs[:, 1])
    _PL = {"dust": ("sigma_H2", "sigma_dust", r"$\log\, \Sigma_{\rm H_2}$  [$M_\odot$ kpc$^{-2}$]",
                    r"$\log\, \Sigma_{\rm dust}$  [$M_\odot$ kpc$^{-2}$]", "(c) the dust-gas plane, walked in time since quenching"),
           "ks":   ("sigma_H2", "sigma_SFR", r"$\log\, \Sigma_{\rm H_2}$  [$M_\odot$ kpc$^{-2}$]",
                    r"$\log\, \Sigma_{\rm SFR}$  [$M_\odot$ yr$^{-1}$ kpc$^{-2}$]",
                    "(c) the KS plane, walked in time since quenching")}[P6_PLANE]
    _cent = [0.5 * (P6_TQT[_i] + P6_TQT[_i + 1]) for _i in range(len(P6_TQT) - 1)]
    _SC6 = (_SG6 if P6_SCAT_STAGES is None else _SG6[_SG6["stage"].isin(P6_SCAT_STAGES)]).dropna(subset=[_PL[0], _PL[1]])
    _nsc = {}
    for _hd, _col, _lab in _halves:                     # the scatter: every critical point and the anchor of every galaxy in the half
        _s = _SC6[_SC6["key"].isin(set(ANC6.loc[ANC6[_SPLITCOL] == _hd, "key"]))]
        _nsc[_hd] = len(_s)
        ax.scatter(_s[_PL[0]], _s[_PL[1]], s=12, color=_col, alpha=0.30, lw=0, zorder=1, rasterized=True)
    _walk = {}
    for _hd, _col, _lab in _halves:
        _keys = set(ANC6.loc[ANC6[_SPLITCOL] == _hd, "key"])
        _w = {}
        for _i6 in range(len(P6_TQT) - 1):
            _sub = _SG6[(_SG6["key"].isin(_keys)) & (_SG6["dtq"] > P6_TQT[_i6]) & (_SG6["dtq"] <= P6_TQT[_i6 + 1])]
            _per = _sub.groupby("key")[[_PL[0], _PL[1]]].median()             # one value per galaxy first
            _per = _per[np.isfinite(_per[_PL[0]]) & np.isfinite(_per[_PL[1]])]
            if len(_per) >= P6_NMIN:
                _w[_i6] = (float(_per[_PL[0]].median()), float(_per[_PL[1]].median()), int(len(_per)))
        _walk[_hd] = _w
    _common = sorted(set(_walk[False]) & set(_walk[True]))      # the same time bins for both halves, or the tracks are not comparable
    _all = np.array([_walk[h][i][:2] for h in (False, True) for i in _common], float).reshape(-1, 2)
    if len(_SC6):                       # the frame FIRST (the boxes below are placed in display space, so the limits must be final):
        _px = np.nanpercentile(_SC6[_PL[0]].to_numpy(float), [1, 99]); _py = np.nanpercentile(_SC6[_PL[1]].to_numpy(float), [1, 99])
        if len(_all):                    # the bulk of the scatter, never tighter than the tracks
            _px = [min(_px[0], _all[:, 0].min() - 0.3), max(_px[1], _all[:, 0].max() + 0.3)]
            _py = [min(_py[0], _all[:, 1].min() - 0.3), max(_py[1], _all[:, 1].max() + 0.3)]
        ax.set_xlim(_px[0] - 0.15, _px[1] + 0.15); ax.set_ylim(_py[0] - 0.15, _py[1] + 0.15)
        _nout = int(((_SC6[_PL[0]] < _px[0] - 0.15) | (_SC6[_PL[0]] > _px[1] + 0.15) | (_SC6[_PL[1]] < _py[0] - 0.15)
                     | (_SC6[_PL[1]] > _py[1] + 0.15)).sum())
        if _nout:
            print(f"  Fig 2 (c): {_nout} of {len(_SC6)} scatter points lie outside the frame (1-99 % of the scatter, padded)")
    if len(_all):
        _gx = np.linspace(np.nanmin(_all[:, 0]) - 0.15, np.nanmax(_all[:, 0]) + 0.15, 8)
        _off = np.nanmedian(_all[:, 1] - _all[:, 0])
        ax.plot(_gx, _gx + _off, ls=(0, (5, 4)), color="0.72", lw=1.1, zorder=0)
        ax.annotate("constant dust-to-gas", xy=(_gx[0], _gx[0] + _off), xytext=(4, 6), textcoords="offset points",
                    fontsize=8, color="0.55", ha="left")
    for _hd, _col, _lab in _halves:
        _xx = [_walk[_hd][_i6][0] for _i6 in _common]; _yy = [_walk[_hd][_i6][1] for _i6 in _common]
        ax.plot(_xx, _yy, "-", color=_col, lw=2.2, alpha=0.9, zorder=3)
        ax.scatter(_xx, _yy, s=np.linspace(150, 45, len(_xx)), color=_col, edgecolor="k", lw=0.6, zorder=4,
                   label=f"{_lab}  ({_nsc[_hd]} points, {int((ANC6[_SPLITCOL] == _hd).sum())} galaxies)")
        for _i6 in range(len(_xx) - 1):
            ax.annotate("", xy=(_xx[_i6 + 1], _yy[_i6 + 1]), xytext=(_xx[_i6], _yy[_i6]), zorder=3,
                        arrowprops=dict(arrowstyle="-|>", color=_col, lw=1.5, alpha=0.9))
    # the boxes: the time since quenching and the galaxies behind the pair, at the bottom right of each SELECTED point. Two
    # points a fraction of a dex apart would put their boxes on top of each other, so a box that lands on an earlier one is
    # stacked underneath it (worked out in points, after the layout is final).
    if _common:
        fig.canvas.draw()
        _xxh = [_walk[True][i][0] for i in _common]; _yyh = [_walk[True][i][1] for i in _common]
        _disp = ax.transData.transform(np.column_stack([_xxh, _yyh])) * 72.0 / fig.dpi          # the points, in points
        _BOXW, _BOXH, _GAP = 48.0, 22.0, 2.5                    # the footprint of one box (two lines of 7 pt text + padding)
        _placed = []                                            # (left, right, top, bottom) of every box drawn so far
        for _i6, _b6 in enumerate(_common):
            _left, _top = _disp[_i6, 0] + 9.0, _disp[_i6, 1] - 9.0
            for _ in range(12):                                 # push down until it overlaps none of the boxes already placed
                _hit = [b for b in _placed if _left < b[1] and _left + _BOXW > b[0] and _top > b[3] - _GAP and _top - _BOXH < b[2] + _GAP]
                if not _hit:
                    break
                _top = min(b[3] for b in _hit) - _GAP
            ax.annotate(f"{_cent[_b6]:+.1f} Gyr\nN {_walk[False][_b6][2]} / {_walk[True][_b6][2]}",
                        xy=(_xxh[_i6], _yyh[_i6]), xytext=(9.0, _top - _disp[_i6, 1]), textcoords="offset points",
                        ha="left", va="top", fontsize=7.0, color="#2166ac", zorder=7,
                        bbox=dict(boxstyle="round,pad=0.28", fc="white", ec="#2166ac", lw=0.7, alpha=0.92))
            _placed.append((_left, _left + _BOXW, _top, _top - _BOXH))
    ax.set_xlabel(_PL[2]); ax.set_ylabel(_PL[3])
    ax.legend(frameon=False, fontsize=9, loc="upper left",
              title=r"boxes: $t - t_{\rm QT}$ [Gyr] at the bin centre, N galaxies " + " / ".join(l for _, _, l in _halves),
              title_fontsize=8)
    ax.grid(alpha=0.22, lw=0.5); ax.set_title(_PL[4])
    fig.suptitle(rf"Part 6 — the two sides of the division through their own quenching ({BOX_LABEL.get(P6_BOX, P6_BOX)}, "
                 rf"core = {P6_CORE})", fontsize=12.5)
    paper_save(fig, f"kappa_dustsplit_{P6_BOX}")
    plt.show()

    # ── FIGURE 3: the rejuvenation test — the SAME walked plane as Fig 2 (c), stepped through the CRITICAL POINTS ──
    # Fig 2 (c) walks the dust-gas plane in bins of t - t_QT; this one walks the very same plane through the critical points
    # jet_on -> sft -> qt -> anchor (P6_SSTAGE, in time order), one panel per bin of the galaxy's OWN global sSFR
    # (sfr100_tot / m_star_tot) AT THAT EPOCH, so a galaxy moves between panels as it quenches. If the dusty galaxies were
    # simply the ones still (or again) forming stars, the two sides of the division would separate ACROSS the panels and lie on
    # top of each other INSIDE them. Each point is the median galaxy at that critical point (one value per galaxy first) with
    # its bootstrap error, the arrows give the direction of time, and the grey dashed line — the SAME line in every panel —
    # is constant dust-to-gas, so the vertical distance from it is the dust-to-gas ratio itself.
    S3_6 = _SG6[_SG6["stage"].isin(P6_SSTAGE)].dropna(subset=["sigma_H2", "sigma_dust"]).copy()
    # the panel is a property of the GALAXY, not of the epoch: its global sSFR AT THE ANCHOR, so one track stays in one panel.
    # A galaxy with a measured M* but no star formed in the last 100 Myr has no log sSFR at all — it is not missing data, it is
    # the deadest galaxy there is, so it goes to the bottom panel (-inf) instead of being dropped; the edges are quantiles of
    # the galaxies that DO have a measurable sSFR, so they stay finite whatever fraction of the sample is undetected.
    _ixa6 = list(zip(ANC6["snap"].astype(int), ANC6["gx"].astype(int)))
    _gcol6 = lambda c: (_C6[c].reindex(_ixa6).to_numpy(float) if c in _C6.columns else np.full(len(ANC6), np.nan))
    _sfa6, _msa6 = _gcol6("sfr100_tot"), _gcol6("m_star_tot")
    _lsa6 = ANC6["lssfr_tot"].to_numpy(float).copy()
    _nod6 = (~np.isfinite(_lsa6)) & (_msa6 > 0) & np.isfinite(_sfa6) & (_sfa6 <= 0)
    _lsa6[_nod6] = -np.inf
    _fin6 = np.isfinite(_lsa6)
    _sed6 = np.asarray(P6_SEDGES, float) if P6_SEDGES is not None else np.nanquantile(_lsa6[_fin6], [1 / 3, 2 / 3])
    _sed6 = np.atleast_1d(_sed6)[np.isfinite(np.atleast_1d(_sed6))] if _fin6.sum() >= 3 else np.array([])
    _drw6 = _fin6 | np.isneginf(_lsa6)                       # the galaxies that get a panel at all
    _pan6 = np.where(_drw6, np.digitize(np.where(_drw6, _lsa6, 0.0), _sed6), -1)
    _PAN6 = {k: int(p) for k, p in zip(ANC6["key"], _pan6)}
    S3_6["panel"] = [_PAN6.get(k, -1) for k in S3_6["key"]]
    _pe6 = np.concatenate([[-np.inf], _sed6, [np.inf]])
    MRK3 = {"sf_peak": "*", "agn_ign": "^", "jet_on": "v", "sft": "s", "qt": "D", "post_quench": "P", "gas_min": "X",
            "anchor": "o"}
    LAB3 = {"sf_peak": "SF peak", "agn_ign": "ignition", "jet_on": "jet", "sft": "SFT", "qt": "QT",
            "post_quench": "post-q", "gas_min": "gas min", "anchor": "anchor"}
    _np6 = len(_pe6) - 1
    _W3 = {}
    for _p6 in range(_np6):
        for _hd, _col, _lab in _halves:
            _keys = set(ANC6.loc[ANC6[_SPLITCOL] == _hd, "key"])
            _g6 = S3_6[(S3_6["panel"] == _p6) & (S3_6["key"].isin(_keys))]
            _w = {}
            for _st6 in P6_SSTAGE:                                  # one row per galaxy per stage, so the median is per galaxy
                _gs6 = _g6[_g6["stage"] == _st6]
                _cx, _ex6, _n6 = _pt6(_gs6.groupby("key")["sigma_H2"].median().to_numpy(float), P6_NMIN3)
                _cy, _ey6, _ = _pt6(_gs6.groupby("key")["sigma_dust"].median().to_numpy(float), P6_NMIN3)
                if np.isfinite(_cx) and np.isfinite(_cy):
                    _w[_st6] = (_cx, _cy, _ex6, _ey6, _n6)
            _W3[(_p6, _hd)] = _w
    _pts3 = np.array([v[:2] for _k3 in _W3 for v in _W3[_k3].values()], float).reshape(-1, 2)
    fig, axs = plt.subplots(1, _np6, figsize=(5.0 * _np6, 5.0), sharex=True, sharey=True, constrained_layout=True)
    axs = np.atleast_1d(axs)
    if len(_pts3) >= 2:                                               # ONE constant-dust-to-gas line for all panels: they compare
        _gx3 = np.linspace(_pts3[:, 0].min() - 0.35, _pts3[:, 0].max() + 0.35, 8)
        _off3 = float(np.median(_pts3[:, 1] - _pts3[:, 0]))
    else:
        _gx3, _off3 = np.linspace(0, 1, 8), 0.0
    print(f"\n  the rejuvenation test — the dust-gas plane walked through the critical points, the galaxies split by their own "
          f"sSFR AT THE ANCHOR\n  ({int(_nod6.sum())} of {len(ANC6)} formed no star in the last 100 Myr there and sit in "
          f"panel a; {int((_pan6 < 0).sum())} have no anchor measurement and are dropped).\n"
          f"  log Sigma_dust/Sigma_H2 of the median galaxy, N in brackets:")
    for _p6 in range(_np6):
        ax = axs[_p6]
        ax.plot(_gx3, _gx3 + _off3, ls=(0, (5, 4)), color="0.72", lw=1.1, zorder=0)
        for _hd, _col, _lab in _halves:
            _w = _W3[(_p6, _hd)]
            _st3 = [s_ for s_ in P6_SSTAGE if s_ in _w]
            _xx = np.array([_w[s_][0] for s_ in _st3], float); _yy = np.array([_w[s_][1] for s_ in _st3], float)
            if len(_st3) >= 2:
                ax.plot(_xx, _yy, "-", color=_col, lw=2.2, alpha=0.85, zorder=4)
                for _i6 in range(len(_st3) - 1):            # the head at the MIDPOINT: the markers are large and would hide it
                    ax.annotate("", xy=(0.5 * (_xx[_i6] + _xx[_i6 + 1]), 0.5 * (_yy[_i6] + _yy[_i6 + 1])),
                                xytext=(_xx[_i6], _yy[_i6]), zorder=5,
                                arrowprops=dict(arrowstyle="-|>", color=_col, lw=1.6, alpha=0.9, mutation_scale=16,
                                                shrinkA=6, shrinkB=0))
            for _i6, _s3 in enumerate(_st3):
                ax.errorbar(_xx[_i6], _yy[_i6], xerr=_w[_s3][2], yerr=_w[_s3][3], marker=MRK3.get(_s3, "o"), ms=11,
                            color=_col, mec="k", mew=0.7, capsize=3, lw=1.5, zorder=6)
                if not _hd:                                           # the stage order is the same on both tracks: label one
                    ax.annotate(LAB3.get(_s3, _s3), xy=(_xx[_i6], _yy[_i6]), xytext=(8, -12), textcoords="offset points",
                                fontsize=8.2, color=_col)
            print(f"    panel {chr(97 + _p6)}  {_lab:9s}: "
                  + (",  ".join(f"{LAB3.get(s_, s_)} {_w[s_][1] - _w[s_][0]:+.2f} ({_w[s_][4]})" for s_ in _st3)
                     if _st3 else f"nothing above N = {P6_NMIN3}"))
        _lo6 = "" if not np.isfinite(_pe6[_p6]) else f"{_pe6[_p6]:.2f} < "
        _hi_tex = "" if not np.isfinite(_pe6[_p6 + 1]) else r" $\leq$ " + f"{_pe6[_p6 + 1]:.2f}"
        _ngp6 = int((_pan6 == _p6).sum())
        _nod_t = f", {int((_nod6 & (_pan6 == _p6)).sum())} of them with no detected SF" if _p6 == 0 else ""
        ax.set_title(f"({chr(97 + _p6)})  {_lo6}" + r"$\log$ sSFR$_{\rm anchor}$" + f"{_hi_tex}"
                     + f"\n{_ngp6} galaxies{_nod_t}", fontsize=10.5)
        ax.set_xlabel(r"$\log\, \Sigma_{\rm H_2}$  [$M_\odot$ kpc$^{-2}$]"); ax.grid(alpha=0.22, lw=0.5)
    axs[0].set_ylabel(r"$\log\, \Sigma_{\rm dust}$  [$M_\odot$ kpc$^{-2}$]")
    _h3 = [Line2D([], [], color=c, lw=2.2, marker="o", ms=8, mec="k", mew=0.6, label=l) for _, c, l in _halves]
    _h3 += [Line2D([], [], ls=(0, (5, 4)), color="0.72", lw=1.1, label="constant dust-to-gas")]
    axs[0].legend(handles=_h3, frameon=False, fontsize=8.6, loc="upper left")
    _h3b = [Line2D([], [], ls="", marker=MRK3[s_], ms=8, mfc="0.6", mec="k", mew=0.5, label=LAB3.get(s_, s_))
            for s_ in P6_SSTAGE if s_ in MRK3]
    axs[-1].legend(handles=_h3b, frameon=False, fontsize=8, loc="lower right", ncol=2, title="critical point",
                   title_fontsize=8)
    fig.suptitle(r"Part 6 — is the dust just rejuvenation? the dust-gas plane walked through the critical points, the "
                 rf"galaxies split by their sSFR at the anchor ({BOX_LABEL.get(P6_BOX, P6_BOX)}, core = {P6_CORE})",
                 fontsize=12.5)
    paper_save(fig, f"kappa_dustgas_ssfr_{P6_BOX}")
    plt.show()

    ANC6.drop(columns=["key", "fp"]).to_csv(os.path.join(OUT, f"kappa_dustclock_{P6_BOX}.csv"), index=False)
    _fg6 = ANC6["f_gas_anchor"].to_numpy(float)
    _n_out, _n_all, _n_rot = 0, 0, 0                     # why "time since the last X-ray episode" is not an axis anywhere here
    for _i6, _r6 in ANC6.iterrows():
        _d = _CH6[_r6["key"]]
        if not _d["xr"].any() or not np.isfinite(_r6["t_qt_gyr"]):
            continue
        _m = _d["ex"] & np.isfinite(_d["tsx"]) & (_d["t"] >= _r6["t_qt_gyr"])
        _n_all += int(_m.sum()); _n_out += int((_m & (_d["tsx"] > 0)).sum())
        _n_rot += int((_m & (_d["tsx"] > 0)).sum()) if bool(ANC6["hi_geom"].to_numpy(bool)[_i6]) else 0
    print(f"  \"time since the last X-ray episode\" is thin by construction: {100 * np.nanmean(ANC6['xray_now']):.0f} % of them "
          f"are IN the X-ray state at the anchor, and over the post-quenching snapshots only "
          f"{100 * _n_out / max(1, _n_all):.0f} % are out of it — {100 * _n_rot / max(1, _n_out):.0f} % of those belong to the "
          f"SELECTED class, so getting OUT of the state is something that class manages.")
    print(f"  the X-ray gate is not what separates them: {100 * np.nanmean(_fg6 < XRAY_FGAS_MAX):.0f} % of these quenched "
          f"galaxies are already below f_gas = {XRAY_FGAS_MAX:g} at the anchor (median {np.nanmedian(_fg6):.3f}) — the channel "
          "has fired on essentially all of them, which is why its exposure is carried here but never used as a division.")
    _lo_end = np.nanmedian(ANC6.loc[ANC6["lo_geom"], "dust"]); _hi_end = np.nanmedian(ANC6.loc[ANC6["hi_geom"], "dust"])
    print(f"  THE DIVISION, as one number: the selected class ends {_hi_end - _lo_end:+.2f} dex above the rest in {P6_DUST} "
          f"({_hi_end:.2f} vs {_lo_end:.2f});  per galaxy -> {os.path.join(OUT, f'kappa_dustclock_{P6_BOX}.csv')}")
```

## Part 7 — the same division in the m25 particle-level products

Part 6 divides the galaxies by Part 2's selection — the kinematic significance and the stellar mass, the same two cuts, not
re-medianed. This part asks the same question of the resolved products, which are the ones a telescope would see:

* **(A) the radial anatomy** per annulus — $A_V$, $M_{\rm dust}/M_\star$, $M_{\rm H_2}/M_\star$, sSFR, $M_{\rm dust}/M_{\rm H_2}$,
  stellar age and $\kappa_{\rm rot}$ — with the m25 star-forming galaxies as the grey reference. Because the groups are split
  on the **galaxy-wide catalogue** axis, no annulus panel here is circular; the $\kappa_{\rm rot}^{\rm H_2}$ row is the
  independent check, and it separates by 0.3–0.4 at every radius.
* **(B) the K-S plane** at $r < R_{50}({\rm H_2})$ with the ALMA-C11 quiescent galaxies on it, cropped to where the tracks and
  the observations are, and with the beam-matched "as ALMA-C11 would measure it" positions overlaid (`P7_BEAM_KPC`,
  `P7_UV_MIN_KPC` — the two-tier rule described at the knobs).
* **(C)** the offset from the reference relation at each critical point, printed rather than drawn: the post-starburst
  question the fan raises has a testable answer on this plane.

```python
# ── Part 7 — the same division in the m25 PARTICLE-LEVEL products: the radial anatomy and the K-S plane ──
# Part 6 divides the quenched galaxies by two catalogue-level numbers — the X-ray duty fraction since SFT and Part 2's kinematic
# significance — and shows the dust follows that division. This part asks the same question of the resolved products, which are
# the ones a telescope would see: (A) the radial anatomy per annulus (A_V, M_dust/M*, M_H2/M*, sSFR, D/H2, stellar age,
# kappa_rot) with the m25 star-forming galaxies as the grey reference, and (B) the K-S plane at r < R50(H2) with the ALMA-C11
# quiescent galaxies on it, ONE panel cropped to where the tracks and the observations are (P7_KS_XLIM / P7_KS_YLIM). (C)
# answers the question the fan raises: the galaxies with no resolved rotation carry MORE dust just before quenching and lose it
# fastest — were they post-starbursts? On the K-S plane that is a testable statement, so (C) measures the offset from the
# reference relation at each critical point (a burst sits ABOVE it before quenching) — PRINTED and written to the CSV.
# Both blocks step through the SAME critical points as Part 6's Fig 3 (jet_on -> sft -> qt -> end), and the tracks carry the
# direction of time as arrows.
# NOT a pure read of the ism_prediction products: this needs the m25 notebook's own tables and the ks_tracks_quenched_m25 run.
# Every block gates on its file and prints what is missing, so the cell runs unchanged off the cluster.
P7_BOX     = "cis25"
P7_M25     = globals().get("P7_M25_OVERRIDE", os.path.join(os.getcwd(), "output", P7_BOX))
P7_OBS     = globals().get("P7_OBS_OVERRIDE", os.path.join(os.getcwd(), "obs_data", "almac11", "ks_table.csv"))   # the ALMA-C11 K-S table (see obs_data/almac11/README)
P7_NMIN    = 5             # galaxies a drawn point needs (below the notebook's NMIN: a class x stage cell is thin by itself)
P7_ANNULI  = [("ap1kpc", "0-1"), ("ann3kpc", "1-3.2"), ("ann10kpc", "3.2-10"), ("ann32kpc", "10-32")]
P7_NSTAR   = 20            # the m25 Part 7e sampling floor on an annulus
P7_SFRCOL  = "sfr100"      # the SFR column of aperture_truth.fits behind the resolved sSFR panel ("sfr25" is the
                            # other one); sSFR = that SFR over the SAME region's M*
P7_SSFRMIN = 1e-13         # an annulus with no star formation sits ON this floor (drawn as a dotted line) instead
                            # of vanishing off a log axis; Part 0's SSFR_FLOOR = 1e-15 would stretch the panel over
                            # five decades to show one dead annulus
P7_KS_AP   = "R50_H2"      # the K-S aperture: 0.5 M / pi R50(H2)^2, the convention the observed points are measured in
P7_KS_STG  = ["jet_on", "sft", "qt", "end"]                 # the critical points of the K-S track — Part 6's Fig 3 set ('end' is the
                                                             # anchor as the K-S products define it: the last snapshot still holding H2).
                                                             # Add "sf_peak" / "agn_ign" back to see the pre-quenching burst itself
P7_KS_SFR  = "logSigmaSFR"                                  # the SFR surface density column of ks_tracks.fits (100 Myr, observed convention)
P7_REL     = "B08"         # the ONLY relation drawn (the H2 one). K98 is a total-gas relation and is deliberately not shown
P7_KS_XLIM = (0.0, 4.0)    # the K-S frame, log Sigma_H2: where the beam-matched tracks and the observed points are
P7_KS_YLIM = (-3.2, 2.0)   # ... and log Sigma_SFR
P7_SFCLOUD = False         # True re-adds the m25 star-forming galaxies as a grey point cloud behind the tracks
P7_ALLANCH = True          # True = every quenched anchor row gets a group (the resolved tables are per anchor); False = Part 6's one-per-galaxy set
P7_BEAM_KPC = 3.8          # the ALMA-C11 beam half-width [kpc]: the size limit (R < b_maj/2) an UNRESOLVED source gets, and with it the Sigma
                            # lower limit 0.5 M / pi (b_maj/2)^2 — exactly what the open observed points are. 3.8 = the median b_maj/2 of
                            # the three unresolved ALMA-C11 sources (3.29 / 3.80 / 4.52 kpc, i.e. beams of 1.3-1.6 arcsec at z = 0.35-0.42,
                            # obs_data/almac11/README). None = no "as observed" positions at all
P7_UV_MIN_KPC = 1.6        # the smallest size the ALMA-C11 uv-plane fits recovered (hc9, 1.6 kpc): a uv fit measures BELOW the beam, so a
                            # simulated galaxy with R50(H2) at or above this keeps its own size; below it the fit would fail and the
                            # observation returns the P7_BEAM_KPC limit (two-tier rule). None = the one-tier rule R = max(R50, beam),
                            # which treats every galaxy as if its fit had failed (too conservative — the 2026-09-04 first version)
P7_KS_MAIN = "intrinsic"   # the track drawn solid: "intrinsic" (R = R50(H2), the physics) | "observed" (where ALMA-C11 would place it)
P7_KS_OVERLAY = True       # draw the other representation too, faded and dashed
P7_TAIL    = True          # the extended-H2 tail: every quenched galaxy LARGER than the beam at 'end' (R50(H2) > P7_BEAM_KPC), its own point

ANN7 = os.path.join(P7_M25, "tables", "annulus_ism_truth.fits")            # m25 Part 8a: M_dust / M_gas / M_H2 per (annulus, sightline)
APT7 = os.path.join(P7_M25, "tables", "aperture_truth.fits")               # m25 Part 7e: M*, age per (annulus, sightline)
AV7  = os.path.join(P7_M25, "tables", "annulus_av_allincl.fits")           # m25 Part 8b: A_V per (annulus, sightline) — optional panel
KIN7 = os.path.join(P7_M25, "tables", "annulus_kinematics.fits")           # m25 Part 8j0: kappa_rot per spherical zone — optional panels
SEL7 = os.path.join(P7_M25, "tables", "powderday_quenched_selection.fits") # m25 Part 3: pop -> the SF-controls line — optional
KS7  = os.path.join(P7_M25, "ks_tracks_pt", "ks_tracks.fits")              # ks_tracks_quenched_m25 Part 4a (run tag 'pt')

if "ANC6" not in globals():
    print("Part 7: needs Part 6 — run it first (it defines the division this part carries into the resolved products).")
else:
    from matplotlib.lines import Line2D as _L7
    import re as _re7
    # ── the division, re-derived for EVERY quenched anchor row (the resolved tables are keyed on the anchor) ──
    A7 = EPO6[EPO6["stage"] == "anchor"][["key", "snap", "gx", "gal_id", "anchor_snap", "z", "t_sft_gyr", "t_qt_gyr",
                                             "tau_q_gyr", "log_mstar_anchor"]].copy()
    A7 = A7[[k in _CH6 for k in A7["key"]]].reset_index(drop=True)
    if not P7_ALLANCH:
        A7 = A7[A7["key"].isin(set(ANC6["key"]))].reset_index(drop=True)
    A7 = pd.concat([A7, _geom6(A7["snap"], A7["gx"]).reset_index(drop=True)], axis=1)
    A7["kappa_z_cat"] = [_KZ6.get((int(a_), int(g_)), np.nan) for a_, g_ in zip(A7["anchor_snap"], A7["gal_id"])]   # Part 2's axis
    _d7 = []
    for _, _r7 in A7.iterrows():
        _d = _CH6[_r7["key"]]; _t, _ex, _xr, _dtc = _d["t"], _d["ex"], _d["xr"], _d["dt"]
        _post = _ex & (_t >= _r7["t_sft_gyr"]) if np.isfinite(_r7["t_sft_gyr"]) else _ex
        _Tp = float(np.sum(_dtc[_post]))
        _d7.append(float(np.sum(_dtc[_xr & _post]) / _Tp) if _Tp > 0 else np.nan)
    A7["f_x_post"] = _d7                                             # the AGN exposure, carried into the CSV, never a division
    _g7v = A7[P6_GEOM].to_numpy(float)
    if P6_NH2MIN > 0:                                                  # the SAME cuts as Part 6, not re-medianed, and its floor
        _g7v = np.where(A7["n_H2_tot"].to_numpy(float) >= P6_NH2MIN, _g7v, np.nan)
    A7["hi_geom"], A7["lo_geom"] = _hi6(_g7v, _gcut, A7["log_mstar_anchor"].to_numpy(float))
    _plain = lambda t: _re7.sub(r"\\[a-zA-Z]+|[{}$_]", "", t.replace("$_2$", "2")).replace("  ", " ").strip()   # a legend label -> a plain group name
    G7 = [(_plain(_ghi), A7["hi_geom"], KCLS["rot"]["color"], _ghi),      # the notebook's one division, as in Part 6
          (_plain(_glo), A7["lo_geom"], KCLS["iso"]["color"], _glo)]
    NAM7 = [n for n, _, _, _ in G7]; COL7 = {n: c for n, _, c, _ in G7}; LAB7 = {n: l for n, _, _, l in G7}
    A7["grp"] = None
    for _n7, _m7, _, _ in G7:
        A7.loc[_m7.to_numpy(), "grp"] = _n7
    KEY7 = {(int(a_), int(g_)): n_ for a_, g_, n_ in zip(A7["anchor_snap"], A7["gal_id"], A7["grp"]) if n_ is not None}
    print(f"Part 7 — the selection in the m25 particle-level products: {len(KEY7)} quenched anchors "
          + ("(every anchor row" if P7_ALLANCH else "(Part 6's one-per-galaxy set") + "), "
          + ", ".join(f"{n} {int(m.sum())}" for n, m, _, _ in G7))

    def _fr7(path, cols):
        """A FITS table -> a DataFrame of the named columns, masked values as NaN and byte strings decoded."""
        t = Table.read(path); d = {}
        for c in cols:
            v = t[c]; v = v.filled(np.nan) if hasattr(v, "filled") else v
            v = np.asarray(v)
            d[c] = np.array([x.decode() if isinstance(x, bytes) else str(x) for x in v]) if v.dtype.kind in "SU" else v.astype(float)
        return pd.DataFrame(d)

    def _gband7(vals, seed=23):
        """Per-galaxy values -> median and its bootstrap 16-84 (nan below P7_NMIN)."""
        v = np.asarray(vals, float); v = v[np.isfinite(v)]
        if len(v) < P7_NMIN:
            return np.nan, np.nan, np.nan, len(v)
        rng = np.random.default_rng(seed)
        bs = np.median(v[rng.integers(0, len(v), (P6_NBOOT, len(v)))], axis=1)
        return float(np.median(v)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), len(v)

    R7 = []
    # ── (A) the radial anatomy per annulus ──
    if not (os.path.exists(ANN7) and os.path.exists(APT7)):
        print(f"\n(A) radial profiles: {ANN7} or {APT7} missing — the m25 notebook's Parts 8a / 7e (cluster)")
    else:
        K4 = ["snap", "gal_id", "incl", "aperture"]
        _APC7 = set(Table.read(APT7).colnames)
        _sfr7 = P7_SFRCOL if P7_SFRCOL in _APC7 else None      # the resolved sSFR panel needs an SFR column
        if _sfr7 is None:
            print(f"  (A) '{P7_SFRCOL}' is not in {os.path.basename(APT7)} "
                  f"({', '.join(sorted(c for c in _APC7 if 'sfr' in c.lower())) or 'no SFR column at all'}) "
                  f"— the sSFR panel is skipped")
        RF = _fr7(ANN7, K4 + ["M_dust", "M_H2", "M_gas"]).merge(
             _fr7(APT7, K4 + ["nstar_ap", "mstar", "age_m_star_myr"] + ([_sfr7] if _sfr7 else [])),
             on=K4, how="inner")
        if os.path.exists(AV7):
            RF = RF.merge(_fr7(AV7, K4 + ["A_V"]), on=K4, how="left")
        RF = RF[RF["aperture"].isin([a for a, _ in P7_ANNULI])].reset_index(drop=True)
        SFK = set()
        if os.path.exists(SEL7):
            _s7 = _fr7(SEL7, ["snap", "gal_id", "pop"])
            SFK = {(int(s_), int(i_)) for s_, i_, p_ in zip(_s7["snap"], _s7["gal_id"], _s7["pop"]) if p_ == "SF"}
        RF["grp"] = [KEY7.get((int(s_), int(i_)), "star-forming controls" if (int(s_), int(i_)) in SFK else None)
                     for s_, i_ in zip(RF["snap"], RF["gal_id"])]
        RF = RF[RF["grp"].notna()].reset_index(drop=True)
        if os.path.exists(KIN7):
            RF = RF.merge(_fr7(KIN7, ["snap", "gal_id", "aperture", "kappa_H2", "kappa_star"]),
                          on=["snap", "gal_id", "aperture"], how="left")
        for _f7, _n7 in ((AV7, "A_V (m25 8b)"), (KIN7, "kappa (m25 8j0)"), (SEL7, "the SF-controls line (m25 Part 3)")):
            if not os.path.exists(_f7):
                print(f"  (A) {_n7}: {_f7} missing — panel / line skipped")
        _ms7 = np.where(RF["nstar_ap"].to_numpy(float) >= P7_NSTAR, RF["mstar"].to_numpy(float), np.nan)
        with np.errstate(all="ignore"):
            for _c7, _num in (("fdust_ap", "M_dust"), ("fH2_ap", "M_H2")):
                _v7 = RF[_num].to_numpy(float)
                RF[_c7] = np.where(np.isfinite(_v7) & (_v7 > 0) & (_ms7 > 0), _v7 / _ms7, np.nan)
            _mh7 = RF["M_H2"].to_numpy(float); _md7 = RF["M_dust"].to_numpy(float)
            RF["dh2_ann"] = np.where((_mh7 > 0) & (_md7 > 0), _md7 / np.where(_mh7 > 0, _mh7, np.nan), np.nan)
            RF["age_ap"] = np.where(np.isfinite(_ms7), RF["age_m_star_myr"].to_numpy(float) / 1e3, np.nan)
            if _sfr7:                                              # the RESOLVED sSFR: this region's SFR over its M*
                _sv7 = RF[_sfr7].to_numpy(float)
                RF["ssfr_ap"] = np.where(np.isfinite(_sv7) & (_ms7 > 0),
                                         np.clip(_sv7 / np.where(_ms7 > 0, _ms7, np.nan), P7_SSFRMIN, None),
                                         np.nan)
            if "A_V" in RF.columns:
                _av7 = RF["A_V"].to_numpy(float)
                RF["A_V"] = np.where(np.isfinite(_av7) & (_av7 > 0), _av7, np.nan)
        QUANT7 = ([("A_V", r"$A_V$ [mag]", "log")] if "A_V" in RF.columns else []) \
                + [("fdust_ap", r"$M_{\rm dust}/M_\star$", "log"), ("fH2_ap", r"$M_{\rm H_2}/M_\star$", "log")] \
                + ([("ssfr_ap", rf"sSFR$_{{{P7_SFRCOL[3:]}}}$ [yr$^{{-1}}$]", "log")] if _sfr7 else []) \
                + [("dh2_ann", r"$M_{\rm dust}/M_{\rm H_2}$", "log"), ("age_ap", "stellar age [Gyr]", "lin")] \
                + ([(k, lab, "kappa") for k, lab in (("kappa_H2", r"$\kappa_{\rm rot}^{\rm H_2}$"),
                                                     ("kappa_star", r"$\kappa_{\rm rot}^{\star}$"))] if "kappa_H2" in RF.columns else [])
        GRPA = (["star-forming controls"] if len(SFK) else []) + NAM7
        COLA = dict(COL7, **{"star-forming controls": "0.62"})
        _x7 = np.arange(len(P7_ANNULI)); _nc7 = (len(QUANT7) + 1) // 2
        NMAX7 = {g: 0 for g in GRPA}
        fig, axs = plt.subplots(2, _nc7, figsize=(3.6 * _nc7, 7.3), squeeze=False)
        for _qi, (_col7, _lab7, _kind) in enumerate(QUANT7):
            ax = axs[_qi // _nc7, _qi % _nc7]
            for _gi, _g7 in enumerate(GRPA):
                _m, _l, _h, _n = [], [], [], []
                for _a7, _ in P7_ANNULI:
                    _sub = RF[(RF["grp"] == _g7) & (RF["aperture"] == _a7)]
                    _v = _sub[_col7].to_numpy(float); _ok = np.isfinite(_v)
                    _per = {}
                    for _s_, _i_, _vv in zip(_sub["snap"].to_numpy(float)[_ok], _sub["gal_id"].to_numpy(float)[_ok], _v[_ok]):
                        _per.setdefault((int(_s_), int(_i_)), []).append(_vv)          # the sightlines -> one value per galaxy
                    _c, _lo, _hi, _nn = _gband7([np.median(vv) for vv in _per.values()])
                    _m.append(_c); _l.append(_lo); _h.append(_hi); _n.append(_nn)
                    R7.append(dict(block="profile", quantity=_col7, group=_g7, x=_a7, n_gal=_nn, med=_c, lo=_lo, hi=_hi))
                NMAX7[_g7] = max(NMAX7[_g7], max(_n))
                _off7 = (_gi - (len(GRPA) - 1) / 2) * 0.075
                _m, _l, _h = np.array(_m), np.array(_l), np.array(_h)
                _okm = np.isfinite(_m)
                if _okm.any():
                    ax.fill_between(_x7[_okm] + _off7, _l[_okm], _h[_okm], color=COLA[_g7], alpha=0.15, lw=0)
                    ax.plot(_x7[_okm] + _off7, _m[_okm], "o-", color=COLA[_g7], ms=4.5, mec="k", mew=0.4,
                            lw=1.3 if _g7.startswith("star-forming") else 1.9)
            if _kind == "log":
                ax.set_yscale("log")
            if _col7 == "ssfr_ap":                                  # say where the floor is, and stop above it
                ax.axhline(P7_SSFRMIN, color="0.45", lw=0.9, ls=":")
                ax.annotate("no SF", xy=(0.02, P7_SSFRMIN), xycoords=("axes fraction", "data"), xytext=(0, 2),
                            textcoords="offset points", fontsize=7, color="0.45", va="bottom")
                ax.set_ylim(bottom=0.55 * P7_SSFRMIN)
            if _kind == "kappa":                                     # the physical reference lines, not an arbitrary 0.3
                ax.axhline(1 / 3, color="0.3", lw=1.0, ls="--"); ax.axhline(0.5, color="0.3", lw=0.8, ls=":"); ax.set_ylim(0, 1)
                ax.annotate("isotropic (1/3)", xy=(0.98, 1 / 3), xycoords=("axes fraction", "data"), xytext=(0, 2),
                            textcoords="offset points", fontsize=6.5, color="0.3", ha="right", va="bottom")
                ax.annotate("rotation-supported (0.5)", xy=(0.98, 0.5), xycoords=("axes fraction", "data"), xytext=(0, 2),
                            textcoords="offset points", fontsize=6.5, color="0.3", ha="right", va="bottom")
            ax.set_xticks(_x7); ax.set_xticklabels([r_ for _, r_ in P7_ANNULI], fontsize=8)
            if _qi // _nc7 == 1:
                ax.set_xlabel("annulus [kpc]")
            ax.set_ylabel(_lab7); ax.grid(alpha=0.25, lw=0.5)
        for _qi in range(len(QUANT7), 2 * _nc7):
            axs[_qi // _nc7, _qi % _nc7].axis("off")
        axs[0, 0].legend(handles=[_L7([], [], color=COLA[g], marker="o", ms=4.5, mec="k", mew=0.4,
                                       lw=1.3 if g.startswith("star-forming") else 1.9, label=f"{g} (N $\\leq$ {NMAX7[g]})")
                                  for g in GRPA], frameon=False, fontsize=7.6)
        fig.suptitle(f"Part 7 — {BOX_LABEL.get(P7_BOX, P7_BOX)}: the radial anatomy of the two classes "
                     f"(+ the SF controls); per-galaxy sightline medians, bootstrap 16-84 of the group median", fontsize=11)
        if P6_GEOM in ("kappa_H2", "kappa_gas"):
            print(f"  (A) NOTE the {P6_GEOM} panel is PARTLY CIRCULAR inside {P6_CORE}: the groups are split on it there. "
                  f"The outer annuli, kappa_star and every other panel are not.")
        else:
            print(f"  (A) the groups are split on {P6_GEOM} (galaxy-wide), so no annulus panel here is circular. In the kappa panels the "
                  f"dashed line is the isotropic value 1/3 (no net rotation below it), the dotted one 0.5 (rotation-supported).")
        fig.tight_layout()
        paper_save(fig, f"kappa_resolved_profiles_{P7_BOX}")
        plt.show()

    # ── (B) the K-S plane, one panel; (C) the offset from the relation at each critical point, printed ──
    if not os.path.exists(KS7):
        print(f"\n(B) K-S plane: {KS7} missing — ks_tracks_quenched_m25 Part 4 (run tag 'pt', cluster)")
    else:
        try:
            import ks_tracks_lib as _kl7
            REL7 = dict(_kl7.RELATIONS[P7_REL]); _HE7 = float(getattr(_kl7, "HE_FACTOR", 1.36))
        except Exception:                                            # the relation, verbatim, so the cell runs outside the repo root
            REL7 = dict(A=-2.06 - 0.96, N=0.96, sig=0.20, color="#1b9e77", ls="--", label="Bigiel+08 (H$_2$)"); _HE7 = 1.36
            print("  (B) ks_tracks_lib not importable — the reference relation is used from its literal definition")
        T7 = Table.read(KS7)
        _sa = lambda a: np.array([x.decode() if isinstance(x, bytes) else str(x) for x in np.asarray(a)])
        ap7, pop7, stg7, gk7 = _sa(T7["ap_label"]), _sa(T7["pop"]), _sa(T7["stage"]), _sa(T7["gkey"])
        x7 = np.asarray(T7["logSigmaH2"], float); y7 = np.asarray(T7[P7_KS_SFR], float)
        ul7 = np.asarray(T7["is_ul"], bool)
        # ── the beam-matched coordinates: R = max(R50(H2), beam); an unresolved galaxy gets the lower limit the observation would give ──
        _sfrtot7 = {"logSigmaSFR": "sfr100_tot", "logSigmaSFR_sfr25": "sfr25_tot", "logSigmaSFR_inst": "sfr_inst_tot"}.get(P7_KS_SFR, "sfr100_tot")
        r50_7 = np.asarray(T7["r50_H2"], float); mh2t7 = np.asarray(T7["m_H2_tot"], float); sft7 = np.asarray(T7[_sfrtot7], float)
        if P7_BEAM_KPC and P7_UV_MIN_KPC:                          # two-tier: a uv fit measures down to P7_UV_MIN_KPC, below it the limit
            _Rb7 = np.where(r50_7 >= P7_UV_MIN_KPC, r50_7, P7_BEAM_KPC)
            unres7 = np.isfinite(r50_7) & (r50_7 < P7_UV_MIN_KPC)
        elif P7_BEAM_KPC:                                           # one-tier: every galaxy below the beam is unresolved
            _Rb7 = np.maximum(r50_7, P7_BEAM_KPC)
            unres7 = np.isfinite(r50_7) & (r50_7 < P7_BEAM_KPC)
        else:
            _Rb7 = r50_7; unres7 = np.zeros(len(r50_7), bool)
        with np.errstate(all="ignore"):
            xb7 = np.where((mh2t7 > 0) & (_Rb7 > 0), np.log10(_HE7 * 0.5 * mh2t7 / (np.pi * _Rb7 ** 2) / 1e6), np.nan)
            yb7 = np.where((sft7 > 0) & (_Rb7 > 0), np.log10(0.5 * sft7 / (np.pi * _Rb7 ** 2)), np.nan)
        if not P7_BEAM_KPC:
            xb7, yb7 = x7, y7                                    # no observed positions: both representations are the intrinsic one
        _RULE7 = (rf"$R = R_{{50}}$ if $\geq {P7_UV_MIN_KPC:g}$ kpc (uv fit), else the ${P7_BEAM_KPC:g}$ kpc limit"
                   if (P7_BEAM_KPC and P7_UV_MIN_KPC) else (rf"$R = \max(R_{{50}}, {P7_BEAM_KPC:g}\,$kpc$)$" if P7_BEAM_KPC else ""))
        _main_obs = (P7_KS_MAIN == "observed") and bool(P7_BEAM_KPC)
        _XM, _YM = (xb7, yb7) if _main_obs else (x7, y7)         # the solid track
        _XO, _YO = (x7, y7) if _main_obs else (xb7, yb7)         # the overlay
        grp7 = np.array([KEY7.get(tuple(int(v) for v in g_.split("_"))) for g_ in gk7], dtype=object)
        base7 = (ap7 == P7_KS_AP)
        MRK7 = {"sf_peak": "*", "agn_ign": "^", "jet_on": "v", "sft": "s", "qt": "D", "end": "o", "anchor": "o",
                 "gas_min": "X"}                                 # the same marker per epoch as Part 6's Fig 3
        LAB7S = {"sf_peak": "SF peak", "agn_ign": "ignition", "jet_on": "jet", "sft": "SFT", "qt": "QT",
                  "post_quench": "post-q", "gas_min": "gas min", "anchor": "anchor", "end": "end"}
        OBS7 = None
        if os.path.exists(P7_OBS):
            _o7 = pd.read_csv(P7_OBS)
            OBS7 = _o7[_o7["co_det"].astype(bool) & np.isfinite(_o7["logSigmaH2"])].reset_index(drop=True)
        else:
            print(f"  (B) {P7_OBS} missing — the observed points are skipped")

        fig, ax = plt.subplots(figsize=(9.6, 7.0), constrained_layout=True)
        _xs7 = np.linspace(P7_KS_XLIM[0] - 1.0, P7_KS_XLIM[1] + 1.0, 60)
        ax.fill_between(_xs7, REL7["A"] + REL7["N"] * _xs7 - REL7["sig"], REL7["A"] + REL7["N"] * _xs7 + REL7["sig"],
                        color=REL7["color"], alpha=0.10, lw=0, zorder=0)
        ax.plot(_xs7, REL7["A"] + REL7["N"] * _xs7, color=REL7["color"], lw=1.5, ls=REL7.get("ls", "--"), zorder=2)
        if P7_SFCLOUD:
            _sf7 = base7 & (pop7 == "SF") & np.isfinite(x7) & np.isfinite(y7)
            ax.scatter(x7[_sf7], y7[_sf7], s=7, c="0.72", alpha=0.30, lw=0, zorder=1, rasterized=True)
        _END7 = {}                                                  # per group at 'end': the numbers printed under the figure
        _lim7 = lambda ax_, x_, y_, c_, a_, s_: ax_.annotate("", xy=(x_ + 0.22, y_ + 0.22), xytext=(x_, y_), zorder=6,   # a slope-1 lower-limit arrow
                                                              arrowprops=dict(arrowstyle="-|>", color=c_, lw=1.1, alpha=a_, mutation_scale=9,
                                                                              shrinkA=s_, shrinkB=0))
        for _n7, _, _c7, _ in G7:
            _mk = base7 & (pop7 == "Q") & (grp7 == _n7) & np.isin(stg7, P7_KS_STG)
            _mx, _my, _ox, _oy, _fu = [], [], [], [], []
            for _st7 in P7_KS_STG:
                _ms7b = _mk & (stg7 == _st7) & ~ul7 & np.isfinite(_XM) & np.isfinite(_YM)
                if _ms7b.sum() >= P7_NMIN:
                    _mx.append(float(np.median(_XM[_ms7b]))); _my.append(float(np.median(_YM[_ms7b])))
                    _oo = _mk & (stg7 == _st7) & ~ul7 & np.isfinite(_XO) & np.isfinite(_YO)
                    _ox.append(float(np.median(_XO[_oo])) if _oo.sum() >= P7_NMIN else np.nan)
                    _oy.append(float(np.median(_YO[_oo])) if _oo.sum() >= P7_NMIN else np.nan)
                    _fu.append(float(np.mean(unres7[_ms7b])))
                    ax.scatter(_mx[-1], _my[-1], s=110 if _st7 != "sf_peak" else 200, marker=MRK7.get(_st7, "o"),
                               c=_c7, edgecolors="k", linewidths=0.8, zorder=6)
                    if _main_obs and _fu[-1] > 0.5:                  # most of the group unresolved: the observed point is a lower limit
                        _lim7(ax, _mx[-1], _my[-1], _c7, 0.9, 7)
                    _xi7, _yi7 = (_ox[-1], _oy[-1]) if _main_obs else (_mx[-1], _my[-1])
                    _xb7, _yb7 = (_mx[-1], _my[-1]) if _main_obs else (_ox[-1], _oy[-1])
                    R7.append(dict(block="ks_track", quantity="stage_median", group=_n7, x=_st7,
                                    n_gal=int(_ms7b.sum()), med=_yi7, lo=np.nan, hi=np.nan, logSigmaH2=_xi7,
                                    logSigmaH2_observed=_xb7, logSigmaSFR_observed=_yb7, f_unresolved=_fu[-1],
                                    r50_H2_kpc=float(np.median(r50_7[_ms7b]))))
                    if _st7 == "end":
                        _END7[_n7] = (_xb7, _xi7, _fu[-1], float(np.median(r50_7[_ms7b])), int(_ms7b.sum()))
            if len(_mx) >= 2:
                ax.plot(_mx, _my, "-", color=_c7, lw=2.6, zorder=5)
                for _ai7 in range(len(_mx) - 1):                    # the direction of time, as in Part 6's walked planes
                    ax.annotate("", xy=(0.5 * (_mx[_ai7] + _mx[_ai7 + 1]), 0.5 * (_my[_ai7] + _my[_ai7 + 1])),
                                xytext=(_mx[_ai7], _my[_ai7]), zorder=5,     # the head at the midpoint, clear of the markers
                                arrowprops=dict(arrowstyle="-|>", color=_c7, lw=1.6, alpha=0.9, mutation_scale=16,
                                                shrinkA=6, shrinkB=0))
                if P7_KS_OVERLAY and P7_BEAM_KPC:                  # the other representation, faded: the size of the resolution effect
                    _okO = np.isfinite(_ox) & np.isfinite(_oy)
                    ax.plot(np.asarray(_ox)[_okO], np.asarray(_oy)[_okO], "--", color=_c7, lw=1.2, alpha=0.45, zorder=3)
                    ax.scatter(np.asarray(_ox)[_okO], np.asarray(_oy)[_okO], s=26, marker="o", c=_c7, alpha=0.45, lw=0, zorder=3)
                    if not _main_obs:                                # the overlay IS the observed one: its lower-limit arrows, faded
                        for _xo_, _yo_, _fu_ in zip(_ox, _oy, _fu):
                            if np.isfinite(_xo_) and _fu_ > 0.5:
                                _lim7(ax, _xo_, _yo_, _c7, 0.45, 3)
        # ── the extended tail: the quenched galaxies the beam WOULD resolve at 'end', each its own point ──
        _tm7 = np.zeros(len(r50_7), bool)
        if P7_TAIL and P7_BEAM_KPC:
            _tm7 = base7 & (pop7 == "Q") & (stg7 == "end") & np.isfinite(r50_7) & (r50_7 > P7_BEAM_KPC) & np.isfinite(xb7)
            _KAP7 = {(int(a_), int(g_)): (k_, c_) for a_, g_, k_, c_ in zip(A7["anchor_snap"], A7["gal_id"], A7["kappa_H2"],
                                                                             A7[P6_GEOM])}
            for _i7 in np.flatnonzero(_tm7):
                _yt = yb7[_i7] if np.isfinite(yb7[_i7]) else P7_KS_YLIM[0] + 0.15        # no SF at all: parked on the frame's floor
                ax.scatter(xb7[_i7], _yt, s=75, marker="P", c=COL7.get(grp7[_i7], "0.5") if grp7[_i7] else "0.5",
                           edgecolors="k", linewidths=0.7, zorder=7)
                _k7, _c7g = _KAP7.get(tuple(int(v) for v in gk7[_i7].split("_")), (np.nan, np.nan))
                R7.append(dict(block="ks_tail", quantity="galaxy", group=grp7[_i7] if grp7[_i7] else "ungrouped", x=gk7[_i7],
                                n_gal=1, med=float(yb7[_i7]), lo=np.nan, hi=np.nan, logSigmaH2=float(xb7[_i7]),
                                r50_H2_kpc=float(r50_7[_i7]), kappa_H2=float(_k7), geom=float(_c7g)))
        _leg7 = []
        if OBS7 is not None:
            _CO, _CE = "#e7298a", "#3b0f2a"
            for _, _r7 in OBS7.iterrows():
                _xo, _yo = float(_r7["logSigmaH2"]), float(_r7["logSigmaSFR"])
                if bool(_r7["sigma_is_ll"]):
                    ax.annotate("", xy=(_xo + 0.3, _yo + 0.3), xytext=(_xo, _yo),
                                arrowprops=dict(arrowstyle="-|>", color=_CE, lw=1.2, mutation_scale=10), zorder=8)
                    ax.scatter(_xo, _yo, s=110, marker="o", facecolors="white", edgecolors=_CE, linewidths=1.4, zorder=9)
                    ax.scatter(_xo, _yo, s=40, marker="o", c=_CO, edgecolors="none", zorder=10)
                else:
                    _elo = float(_r7["logSigmaSFR_elo"]) if np.isfinite(_r7["logSigmaSFR_elo"]) else 0.0
                    _ehi = float(_r7["logSigmaSFR_ehi"]) if np.isfinite(_r7["logSigmaSFR_ehi"]) else 0.0
                    _xe = float(_r7["logSigmaH2_err"]) if np.isfinite(_r7["logSigmaH2_err"]) else 0.0
                    ax.errorbar(_xo, _yo, xerr=_xe, yerr=[[min(_elo, _yo + 4.5)], [_ehi]], fmt="none", ecolor=_CE,
                                elinewidth=1.0, zorder=8)
                    ax.scatter(_xo, _yo, s=110, marker="o", c=_CO, edgecolors=_CE, linewidths=1.4, zorder=9)
            _leg7 += [_L7([], [], marker="o", ls="", ms=9, mfc=_CO, mec=_CE,
                            label=rf"ALMA-C11 QGs, $z\approx0.4$ (N={len(OBS7)})"),
                       _L7([], [], marker="o", ls="", ms=9, mfc="white", mec=_CE, label=r"unresolved: $\Sigma$ lower limits")]
        _leg7 = [_L7([], [], color=_c7, lw=2.4, label=LAB7[_n7]) for _n7, _, _c7, _ in G7] \
               + [_L7([], [], ls="", marker=MRK7.get(s_, "o"), ms=8, mfc="0.55", mec="k",
                        label=LAB7S.get(s_, s_.replace("_", " "))) for s_ in P7_KS_STG] \
               + [_L7([], [], color=REL7["color"], lw=1.5, ls=REL7.get("ls", "--"),
                       label=REL7.get("label", P7_REL) + r" $\pm$ scatter")] \
               + ([_L7([], [], color="0.4", lw=2.0 if _main_obs else 1.2, ls="-" if _main_obs else "--", alpha=1.0 if _main_obs else 0.5,
                         marker="o", ms=6 if _main_obs else 4, mec="k" if _main_obs else "none",
                         label="as ALMA-C11 would measure it (arrow: mostly unresolved)")]
                  + ([_L7([], [], color="0.4", lw=1.2 if _main_obs else 2.0, ls="--" if _main_obs else "-", alpha=0.5 if _main_obs else 1.0,
                            marker="o", ms=4 if _main_obs else 6, mec="none" if _main_obs else "k", label=r"intrinsic: $R = R_{50}({\rm H_2})$")]
                     if (P7_KS_OVERLAY or not _main_obs) else [])
                  + ([_L7([], [], ls="", marker="P", ms=8, mfc="0.6", mec="k",
                            label=rf"larger than the beam: $R_{{50}}({{\rm H_2}}) > {P7_BEAM_KPC:g}$ kpc (N={int(_tm7.sum())})")]
                     if P7_TAIL else [])
                  if P7_BEAM_KPC else []) + _leg7
        ax.legend(handles=_leg7, loc="upper left", frameon=False, fontsize=7.6, ncol=2)
        ax.set_xlabel(r"$\log(\Sigma_{\rm H_2}/M_\odot\,{\rm pc}^{-2})$   [$0.5M/\pi R^2$, He included; solid: "
                      + ("as observed" if _main_obs else r"$R = R_{50}({\rm H_2})$") + "]")
        ax.set_ylabel(r"$\log(\Sigma_{\rm SFR}/M_\odot\,{\rm yr}^{-1}\,{\rm kpc}^{-2})$")
        ax.set_xlim(*P7_KS_XLIM); ax.set_ylim(*P7_KS_YLIM); ax.tick_params(direction="in", top=True, right=True)
        ax.grid(alpha=0.15, lw=0.6)
        ax.set_title(("solid: as observed; faded: intrinsic" if _main_obs else "solid: intrinsic $R_{50}$; faded: as observed")
                     + (" — " + _RULE7 + r"; $+$: larger than the beam" if P7_BEAM_KPC else ""), fontsize=9.5)
        fig.suptitle(f"Part 7 — {BOX_LABEL.get(P7_BOX, P7_BOX)}: the two classes on the K-S plane, against the ALMA-C11 QGs",
                     fontsize=11)
        paper_save(fig, f"kappa_resolved_ks_{P7_BOX}")
        plt.show()
        if P7_BEAM_KPC and _END7:
            print(f"\n(B) at 'end', as ALMA-C11 would measure it ("
                  + (f"uv floor {P7_UV_MIN_KPC:g} kpc, limit {P7_BEAM_KPC:g} kpc" if P7_UV_MIN_KPC else f"R = max(R50, {P7_BEAM_KPC:g} kpc)")
                  + "): log Sigma_H2 observed | intrinsic; unresolved fraction; median R50(H2):")
            for _n7 in NAM7:
                if _n7 in _END7:
                    _e = _END7[_n7]
                    print(f"    {_n7:34s} {_e[0]:5.2f} | {_e[1]:5.2f};  {100 * _e[2]:3.0f} % unresolved;  R50 {_e[3]:.2f} kpc  (N={_e[4]})")
            if P7_TAIL:
                _tl = [r for r in R7 if r["block"] == "ks_tail"]
                print(f"    the galaxies larger than the beam (R50(H2) > {P7_BEAM_KPC:g} kpc at 'end'): {len(_tl)} of "
                      f"{int((base7 & (pop7 == 'Q') & (stg7 == 'end') & np.isfinite(r50_7)).sum())} anchor rows"
                      + (":" if _tl else ""))
                for r in sorted(_tl, key=lambda r: -r["r50_H2_kpc"]):
                    print(f"      {r['x']:>10s}  {r['group']:34s} R50 {r['r50_H2_kpc']:5.2f} kpc  kappa_H2 {r['kappa_H2']:.2f}  "
                          f"{P6_GEOM} {r['geom']:+.2f}  log Sigma_H2 {r['logSigmaH2']:.2f}  log Sigma_SFR "
                          + (f"{r['med']:.2f}" if np.isfinite(r['med']) else "no SF"))

        # ── (C) were they starbursts? the offset from the relation at each critical point — printed + tabulated, not drawn ──
        _off7 = y7 - (REL7["A"] + REL7["N"] * x7)
        print(f"\n(C) the offset from {REL7.get('label', P7_REL)} at each critical point (median over galaxies, "
              f"the censored SFRs dropped; intrinsic R50 apertures — a slope-1 beam shift moves the offset from a slope-{REL7['N']:.2f} "
              f"relation by {100 * abs(1 - REL7['N']):.0f} % of the shift):")
        for _gi, (_n7, _, _c7, _) in enumerate(G7):
            _yy = []
            for _si, _st7 in enumerate(P7_KS_STG):
                _m7 = base7 & (pop7 == "Q") & (grp7 == _n7) & (stg7 == _st7) & ~ul7 & np.isfinite(_off7)
                _c, _l, _h, _n = _gband7(_off7[_m7])
                _yy.append(_c)
                R7.append(dict(block="ks_offset", quantity="dlog_rel", group=_n7, x=_st7, n_gal=_n, med=_c, lo=_l, hi=_h))
            print(f"    {_n7:34s} " + "  ".join(f"{LAB7S.get(s_, s_)} {v:+.2f}" if np.isfinite(v)
                                                 else f"{LAB7S.get(s_, s_)} n/a" for s_, v in zip(P7_KS_STG, _yy)))
        print(f"    (above the relation = a burst; below = a dying disc; +- {REL7['sig']:.2f} dex is the relation's own scatter)")
        _tq7 = A7.groupby("grp")["tau_q_gyr"].median()
        _bst = {}                       # the burst itself: peak sSFR in the Gyr before SFT over the median of the 3 Gyr before that
        for _, _r7 in A7.iterrows():
            _d = _CH6.get(_r7["key"]); _ts = _r7["t_sft_gyr"]
            if _d is None or not np.isfinite(_ts) or "ssfr" not in _d:
                continue
            _late = _d["ex"] & (_d["t"] > _ts - 1.0) & (_d["t"] <= _ts)
            _base = _d["ex"] & (_d["t"] > _ts - 4.0) & (_d["t"] <= _ts - 1.0)
            if _late.sum() >= 2 and _base.sum() >= 3:
                with np.errstate(all="ignore"):
                    _bst.setdefault(_r7["grp"], []).append(float(np.nanmax(_d["ssfr"][_late]) - np.nanmedian(_d["ssfr"][_base])))
        print("  quenching time tau_q [Gyr], median per group: "
              + ",  ".join(f"{n} {_tq7.get(n, np.nan):.2f}" for n in NAM7))
        if _bst:
            print("  the burst before SFT (peak log sSFR in the last Gyr minus the median of the 3 Gyr before it, dex): "
                  + ",  ".join(f"{n} {np.nanmedian(_bst[n]):+.2f}" for n in NAM7 if n in _bst))

    if R7:
        _p7 = os.path.join(OUT, f"kappa_resolved_{P7_BOX}.csv")
        pd.DataFrame(R7).to_csv(_p7, index=False)
        print(f"\n  the tables behind both figures -> {_p7}")
```

## Part 8 — the pictures: one galaxy per class, at each critical point

Parts 6 and 7 score the division. This part shows it: one representative galaxy per class, rendered at
jet_on → SFT → QT → anchor, from the **reduced particle cutouts** — the very particles the K-S and profile products are
measured from — through `simbanator.visualization.rendering.ReducedRender`.

The composite is stars + gas, Screen-blended and attenuated by the dust column; the physical maps
(`P8_MAPS`: $\Sigma_{\rm dust}$, $\Sigma_{\rm H_2}$, temperature) share a log colour scale and carry the mass-weighted gas
**velocity field as streamlines**. Face-on, a rotating disc closes its streamlines into circles and one with no resolved
rotation does not — which is Part 2's selection drawn instead of scored.

The representative of a class is the galaxy closest to its class median in the axes that define the picture (the kinematic
significance and the dust) among those whose four epochs all have a reduced file; the set is seeded on the class that has to
carry the low-dust end (`P8_SEEDCLS = "rest"`) and the other is matched to it in age and mass, with every cost printed.
`P8_SIZE` and `P8_FS` are the two dials worth touching; `P8_PICK` overrides the choice by hand.

This is the one part that needs the reduced cutouts and a working `simbanator` import — it prints what is missing and skips
cleanly otherwise.

```python
# ── Part 8 — THE PICTURES: one representative galaxy per class, at each critical point ──
# Parts 6 and 7 divide the quenched galaxies in two, by Part 2's selection, and score the division. This part SHOWS it: one
# representative galaxy per class, rendered at jet_on -> sft -> qt -> anchor, the same four critical points the tracks are
# walked through.
#   FIG 1  the RGB composite: stars + gas, Screen-blended and attenuated by the dust column
#   FIG 2+ one physical map per P8_MAPS field (Sigma_dust, Sigma_H2 by default) on a SHARED log colour scale,
#          with the mass-weighted GAS VELOCITY FIELD as streamlines: face-on, a rotating disc closes its
#          streamlines into circles and one with no resolved rotation does not — the selection drawn instead of scored.
# The renders read the REDUCED PARTICLE CUTOUTS (build_reduced_particles_job.py), i.e. the very particles the KS
# and profile products are measured from, through simbanator.visualization.rendering.ReducedRender — no snapshot
# and no Caesar catalogue, so the whole grid is seconds rather than one full-box load per panel.
# The representative of a class is the galaxy CLOSEST TO ITS CLASS MEDIAN in the axes that define the picture (the kinematic
# significance and the dust), among those whose four epochs all have a reduced file; P8_PICK overrides it.
P8_SIZE    = 15.0             # ◄── HOW BIG EACH IMAGE IS, in kpc a side (physical).  60 = a 60 x 60 kpc box around
                               #     each galaxy.  The cutouts hold particles out to 100 kpc from the centre, so
                               #     anything past ~200 has empty corners and the cell says so.  None = derive it
                               #     from the galaxies themselves (P8_EXTFAC x the median R50(*), inside P8_EXTLIM).
P8_FS      = 1.5              # ◄── HOW BIG THE TEXT IS: one multiplier for every label, title and number drawn here.
# ════════════════════════════════════════════════════════════════════════════════════════════════════════════════
P8_BOX     = "cis25"
P8_REDUCED = globals().get("P8_REDUCED_OVERRIDE", os.path.join(os.getcwd(), "output", P8_BOX, "reduced_particles"))
P8_PREFIX  = "m25n512"        # the reduced files' name stem: <prefix>_snapNNN_galGGGGGG.h5
P8_STAGES  = ["jet_on", "sft", "qt", "anchor"]     # the columns, in time order
P8_SPOS    = "faceon"         # "faceon" (looking down the spin axis of P8_FRAME) | "edgeon" (along its disc)
P8_FRAME   = "H2_L"           # what faceon / edgeon refer to: "H2_L" (the angular momentum of the member H2 inside P8_FRAME_RMAX —
                               # the ISM's own spin, 2026-09-04) | "gas_L" | "dust_L" | "star_L" | "stellar" (the stored stellar
                               # principal frame: the inertia tensor of the member stars, nearly degenerate for a spheroid, which is
                               # why faceon / edgeon did nothing visible before). A file without velocities, or with fewer than 20
                               # weighted particles inside the radius, falls back to "stellar" and the cell says so.
P8_FRAME_RMAX = 4.0          # [kpc] the radius the frame's angular momentum is summed inside
P8_FRAME_MEMBERS = True       # ... over the CAESAR members only, whatever P8_MEMBERS says: the CGM must not set the orientation
P8_FRAME_STAGE = None        # ONE frame per row: the spin measured at this critical point orients every panel of the row (None =
                               # each panel its own spin, which jumps from panel to panel when the gas barely rotates). The report
                               # prints each spin's COHERENCE |sum l| / sum |l| (1 = a disc, ~1/sqrt(n) = random): below ~0.3 there
                               # is no disc to orient and any frame is a random tilt — the quenched H2 of most of the sample.
# the frame, in full: P8_SIZE above is the whole story unless it is None, in which case the half-width becomes
# P8_EXTFAC x the median R50(*) of the drawn epochs and P8_EXTLIM is a GUARD ON THAT AUTOMATIC VALUE — raising its
# upper bound does nothing unless the automatic value was already being clipped by it.  The cell prints which decided.
P8_EXTFAC  = 3.5              # P8_SIZE is None: the half-width in units of the median R50(*) of the drawn epochs
P8_EXTLIM  = (8.0, 20.0)      # P8_SIZE is None: the bounds that automatic half-width is clipped to [kpc]
P8_EXT     = None if P8_SIZE is None else 0.5 * float(P8_SIZE)      # the half-width the code works in
P8_SIDE    = "right"          # which margin carries the row labels + numbers: "right" | "left"
P8_MARGIN  = 0.21             # the fraction of the figure width that margin gets (the class names are long)
P8_WSPACE  = 0.015            # the gap between panels, as a fraction of a panel (both directions)
P8_XSIZE   = 400              # render size [pixels] of the composites and maps
P8_MEMBERS = False            # False = the WHOLE 100 kpc cutout (CGM + neighbours), which is also what makes the
                               # smoothing lengths right: with members only, hsml is estimated from the members alone
                               # and the ISM comes out blobbier and smaller than it is. True = catalogue members only.
P8_CHAN    = ("star", "gas", "dust")   # the composite's channels, each painted with its own black-to-hue colormap
P8_DYN     = (1.0, 2.5, 2.5)  # decades shown per channel (log_stretch): more = fainter outskirts visible
P8_PMAX    = 99.9             # the percentile of the POSITIVE pixels that saturates to white, per channel
P8_STRETCH = "dyn"            # "dyn" = log_stretch (fixed dynamic range); "percentile" = RenderRGB's clip, which
                               # saturates a small source into a uniform blob — kept only for comparison
# THE PAIR: seeded on the class that has to carry the low-dust end, then matched to it in age and mass.
P8_SEEDCLS = "rest"           # the class the set is seeded from ("rest" = the lower side of the division, which is the one
                               # that has to carry the low-dust end). A name Part 6 did not build falls back to its LAST group,
                               # which is that same class
P8_DGR     = "dh2"            # THE ratio, for the seed cut and for every number reported: "dh2" = M_dust/M_H2 (what
                               # P6_DUST means), "dg" = M_dust/M_gas
P8_SEEDMAX = -3.0             # the seed must be below this in log of that ratio — the low-dust end of its class;
                               # None = no cut.  Among those, the best-resolved galaxy wins (the only place resolution
                               # decides anything)
P8_SAMEANCHOR = True          # PREFER the seed's own anchor for the other three — but only as a preference: a class
P8_COSTMAX = 1.0              # that cannot be matched there to within P8_COSTMAX is taken from whichever anchor
                               # matches best, and the cell says so.  A matched pair at two redshifts says more about
                               # the classes than a mismatched pair at one
P8_SEEDTRY = 5                # how far down the seed's resolution order to walk looking for one the other three CAN
                               # be matched to; the least-bad seed is kept if none of them can
P8_ZMIN    = 0.5              # the anchor must be above this redshift
P8_TOLAGE  = 0.4              # Gyr:  how age trades against mass in the match; the cost is in units of it, so a cost
P8_TOLM    = 0.12             # dex:  of 1 means "off by one tolerance" — the cell prints each galaxy's cost
P8_RESW    = 0.0              # the weight resolution carries when matching the other three: 0 = matching only, which
                               # is the point — a well-matched galaxy beats a better-sampled one
P8_ALLANCH = True             # draw from every anchor row (Part 7's A7, a bigger pool) when Part 7 has run;
                               # False, or Part 7 not run, = Part 6's one-per-galaxy set
P8_MAPS    = ["dust", "H2", "temp"]     # the physical maps drawn, one figure each: any key of P8_MAPDEF
#      name -> (kind, label, unit, log?, cmap, symmetric?).  "sigma" = a surface density through ReducedRender.sigma_map;
#      "weighted" = the mass-weighted line-of-sight mean of a per-particle gas quantity through .weighted_map, which is
#      what a temperature (or a velocity, or a metallicity) needs — a temperature does not project as a column.
P8_MAPDEF  = {
    "dust": ("sigma",    r"\Sigma_{\rm dust}",       r"M_\odot\,{\rm kpc^{-2}}",    True,  "magma",   False),
    "H2":   ("sigma",    r"\Sigma_{\rm H_2}",        r"M_\odot\,{\rm kpc^{-2}}",    True,  "viridis", False),
    "HI":   ("sigma",    r"\Sigma_{\rm HI}",         r"M_\odot\,{\rm kpc^{-2}}",    True,  "cividis", False),
    "gas":  ("sigma",    r"\Sigma_{\rm gas}",        r"M_\odot\,{\rm kpc^{-2}}",    True,  "bone",    False),
    "star": ("sigma",    r"\Sigma_\star",            r"M_\odot\,{\rm kpc^{-2}}",    True,  "afmhot",  False),
    "sfr":  ("sigma",    r"\Sigma_{\rm SFR}",        r"M_\odot\,{\rm yr^{-1}\,kpc^{-2}}", True, "inferno", False),
    "temp": ("weighted", r"\langle T \rangle_M",     r"{\rm K}",                    True,  "inferno", False),
    "zgas": ("weighted", r"\langle Z_{\rm gas} \rangle_M", "",                      True,  "cubehelix", False),
    "vlos": ("weighted", r"\langle v_{\rm los} \rangle_M", r"{\rm km\,s^{-1}}",     False, "RdBu_r",  True),
}
P8_STREAM  = False             # overlay the gas velocity field as streamlines
P8_NGRID   = 8               # the velocity field grid (per side)
P8_STREAM_DENSITY = 1.0       # HOW MANY streamlines: matplotlib seeds them on its own 30 x 30 base grid scaled by this
                               #   (0.5 = half as many lines, 2 = twice) — P8_NGRID only sets the RESOLUTION of the
                               #   velocity field the lines follow, not their number
P8_STREAM_SOURCE = "slab"     # WHICH gas feeds the velocity field: "slab" = every gas particle of the loaded region within
                               #   +-P8_STREAM_DEPTH kpc of the galaxy's mid-plane along the line of sight (enough particles for
                               #   smooth flows, without the CGM in front and behind) | "members" = the CAESAR members only (too few
                               #   for nice flows) | "region" = everything loaded (the CGM and every neighbour in the frame too)
P8_STREAM_DEPTH  = 2.0        # [kpc] the half-thickness of that slab along the line of sight (face-on: the disc plane +- this)
P8_STREAM_BULK   = "members"  # the velocity subtracted before drawing: "members" = the galaxy's own systemic velocity (its member gas,
                               #   its member stars if it has no gas) | "slab" = the mass-weighted mean of the particles used
P8_STREAM_WEIGHT = "m_gas"    # what weights the field: "m_gas" (all the gas) | "m_H2" (the molecular ISM only) | "m_dust"
P8_STREAM_LW = 0.85           # the streamline width
P8_STREAM_ARROW = 0.8         # the arrow size
P8_TAU     = 0.0              # >0 also attenuates every channel by the dust column; with dust drawn as its own
                               # channel that double-counts it, so it is off
P8_CMAP    = {}               # per-field colormap overrides of P8_MAPDEF, e.g. {"temp": "turbo"}
P8_PICK    = {}               # {"sel": (anchor_snap, gal_id), "rest": (...)} names the representatives by hand; {} = the seeded
                               # match below.
# GALAXIES WORTH LOOKING AT, from the runs behind this notebook (they are coordinates, not groups — the four-way dose x class
# grid they were chosen under is gone):
#   the ROTATING set (2026-09-04), the only one whose H2 has a disc to orient — (116, 61), (116, 119), (116, 42), (110, 71):
#   kappa_rot(H2) 0.77 / 0.64 / 0.69 / 0.49 (the median quenched galaxy sits at 0.2, below the isotropic 1/3, and no frame can
#   align a spin that is noise); age 4.01 / 3.62 / 4.02 / 3.48 Gyr, log M* 10.46 / 10.17 / 10.64 / 10.39, log M_dust/M*
#   -3.38 / -4.74 / -4.26 / -5.31, 201 / 527 / 405 / 853 gas particles, three of them at anchor 116 (z 0.70).
#   the DUST-CONTRAST set, tighter-matched but NOT rotating (kappa 0.1-0.2, age 1.29-1.55 Gyr, log M* 10.34-10.53, dust
#   -3.40 / -4.05 / -5.77 / -4.24) — (100, 45), (90, 43), (95, 60), (90, 31).
P8_PCT     = (2.0, 99.5)      # the shared colour scale of a map grid, as percentiles of its finite pixels
P8_MAPDYN  = 4.0              # ... but a LOG map shows at most this many decades below its top: with the whole region
                               # loaded the empty outskirts drag the low percentile down and wash the galaxy out.
                               # None = the percentile alone

if "ANC6" not in globals():
    print("Part 8: needs Part 6 — run it first (it defines the two classes and the galaxy set).")
elif not os.path.isdir(P8_REDUCED):
    print(f"Part 8: {P8_REDUCED} is missing — build_reduced_particles_job.py writes the cutouts (cluster).")
else:
    _fs8 = lambda v: float(v) * float(P8_FS)     # one dial for every label in this part
    import matplotlib.cm as cm
    import matplotlib.colors as mcolors
    import matplotlib.patheffects as pe
    try:
        from simbanator.visualization.rendering import ReducedRender
    except Exception as _e8:
        ReducedRender = None
        print(f"Part 8: simbanator.visualization.rendering.ReducedRender not importable ({_e8}) — "
              f"it needs py-sphviewer + h5py.")
    if ReducedRender is not None:
        # ── where every galaxy is at every critical point, and which of those have a cutout on disk ──
        _EP8 = {(k, s): (int(sn), int(gx)) for k, s, sn, gx
                 in zip(EPO6["key"], EPO6["stage"], EPO6["snap"], EPO6["gx"])}
        _path8 = lambda sn, gx: ReducedRender.epoch_path(P8_REDUCED, sn, gx, P8_PREFIX)

        def _stream8(rr):
            """The velocity field behind the streamlines, from the gas P8_STREAM_SOURCE selects (in rr's CURRENT frame), the bulk
            velocity P8_STREAM_BULK names subtracted -> ((X, Y, VX, VY, den), (n_used, n_in_frame))."""
            g = rr.parts.get("gas", {})
            if "pos" not in g or "vel" not in g:
                return None, (0, 0)
            xyz = rr.project("gas", P8_SPOS); vel = rr.project("gas", P8_SPOS, what="vel")
            w = np.asarray(g.get(P8_STREAM_WEIGHT, g.get("m_gas")), float)
            inframe = (np.abs(xyz[:, 0]) <= _EXT8) & (np.abs(xyz[:, 1]) <= _EXT8)
            memb = np.asarray(g["member"], bool) if "member" in g else np.ones(len(w), bool)
            keep = (memb if P8_STREAM_SOURCE == "members" else
                    (np.abs(xyz[:, 2]) <= float(P8_STREAM_DEPTH)) if P8_STREAM_SOURCE == "slab" else np.ones(len(w), bool))
            keep &= np.isfinite(w) & (w > 0) & np.all(np.isfinite(vel), axis=1)
            # the systemic velocity: the galaxy's own (member gas, else member stars), or the slab's mean
            v0 = None
            if P8_STREAM_BULK == "members":
                mm = memb & np.isfinite(w) & (w > 0)
                if mm.sum() >= 5:
                    v0 = np.average(vel[mm], axis=0, weights=w[mm])
                elif "star" in rr.parts and "vel" in rr.parts["star"]:
                    sv = rr.project("star", P8_SPOS, what="vel"); sw = np.asarray(rr.parts["star"]["m_star"], float)
                    sm = (np.asarray(rr.parts["star"]["member"], bool) if "member" in rr.parts["star"] else np.ones(len(sw), bool)) & (sw > 0)
                    if sm.sum() >= 5:
                        v0 = np.average(sv[sm], axis=0, weights=sw[sm])
            if v0 is None:
                v0 = np.average(vel[keep], axis=0, weights=w[keep]) if keep.sum() else np.zeros(3)
            saved = rr.parts["gas"]
            rr.parts["gas"] = {k: np.asarray(v)[keep] for k, v in saved.items()}          # the selection, in place, briefly
            try:
                X, Y, VX, VY, den = rr.velocity_field(ex=_EXT8, spos=P8_SPOS, weight=P8_STREAM_WEIGHT, ngrid=P8_NGRID,
                                                      subtract_bulk=False)
            finally:
                rr.parts["gas"] = saved
            VX = np.where(den > 0, VX - v0[0], 0.0); VY = np.where(den > 0, VY - v0[1], 0.0)
            return (X, Y, VX, VY, den), (int(np.sum(keep & inframe)), int(np.sum(inframe)))

        def _chain8(key):
            """The (snap, gx) of `key` at every P8_STAGES epoch, or None if one is missing on disk."""
            out = {}
            for s_ in P8_STAGES:
                sg = _EP8.get((key, s_))
                if sg is None or not os.path.exists(_path8(*sg)):
                    return None
                out[s_] = sg
            return out

        # ── the four galaxies: SEED ON THE LOW-DUST END, then MATCH THE OTHER THREE IN AGE AND MASS ──
        # A gallery only says something about the classes if the four galaxies are otherwise alike, so they are chosen
        # in one dependent pass rather than class by class:
        #   1. the SEED is the class P8_SEEDCLS (the lower side of the division by default) — the one that has to
        #      carry the low-dust end of the argument — restricted to log M_dust/M_H2 < P8_SEEDMAX, and among those
        #      the BEST-RESOLVED galaxy (most gas + star particles), because only there does resolution decide;
        #   2. the other three are then whichever member of their class is CLOSEST TO THE SEED in mass-weighted stellar
        #      age and log M*, from the seed's own anchor when P8_SAMEANCHOR.  Matching is all that decides them:
        #      P8_TOLAGE / P8_TOLM only set how age trades against mass (the cost is in units of them, so 1 = off by
        #      one tolerance), and P8_RESW, the weight resolution carries in that choice, is 0 by default.
        _ANCROW8 = EPO6[EPO6["stage"] == "anchor"].drop_duplicates("key").set_index("key")
        _CAND8 = (A7 if (P8_ALLANCH and "A7" in globals() and "hi_geom" in getattr(A7, "columns", []))
                   else ANC6).copy()
        _pool8 = "every anchor row (Part 7's set)" if _CAND8 is not ANC6 else "Part 6's one-per-galaxy set"
        if P8_ALLANCH and _CAND8 is ANC6:
            _pool8 += " — run Part 7 first for the full per-anchor pool"
        # the numbers the match, the seed cut and the figure are made of, all at the anchor
        # the age comes from the FULL Part 0 catalogue (CAT: every galaxy above log M* 9.5 per box and anchor), not
        # from QM — QM is CAT after the quenched rule and the mass / age / history cuts, and those are not the K-S
        # products' cuts, so a perfectly drawable anchor row can be missing from QM and have no age at all.
        _SRC8 = ("CAT" if "CAT" in globals() else ("QM" if "QM" in globals() else None))
        _Q8 = globals()[_SRC8][globals()[_SRC8]["box"] == P8_BOX].copy() if _SRC8 else None
        if _Q8 is not None:
            with np.errstate(all="ignore"):
                if "log_mstar" not in _Q8.columns:
                    _mst8 = _Q8["mstar"].to_numpy(float)
                    _Q8["log_mstar"] = np.log10(np.where(_mst8 > 0, _mst8, np.nan))
            _Q8 = _Q8[~_Q8.duplicated(subset=["snap", "gal_id"])]
            _Q8 = _Q8.set_index([_Q8["snap"].astype(int), _Q8["gal_id"].astype(int)])
        for _c8, _src8 in (("_age", "age"), ("_lm", "log_mstar")):
            _CAND8[_c8] = (_Q8[_src8].reindex(_CAND8["key"]).to_numpy(float)
                             if (_Q8 is not None and _src8 in _Q8.columns) else np.full(len(_CAND8), np.nan))
        if "log_mstar_anchor" in _ANCROW8.columns:            # the KS table always has the mass; QM does not always
            _CAND8["_lm"] = np.where(np.isfinite(_CAND8["_lm"].to_numpy(float)), _CAND8["_lm"].to_numpy(float),
                                      _ANCROW8["log_mstar_anchor"].reindex(_CAND8["key"]).to_numpy(float))
        for _c8, _src8 in (("_ngas", "ngas_cat"), ("_nstar", "nstar_cat"), ("_z", "z"),
                             ("_mdust", "mdust_cat"), ("_mh2", "mh2_cat"), ("_mgas", "mgas_cat")):
            _CAND8[_c8] = (_ANCROW8[_src8].reindex(_CAND8["key"]).to_numpy(float)
                             if _src8 in _ANCROW8.columns else np.full(len(_CAND8), np.nan))
        with np.errstate(all="ignore"):                          # THE ratio: dust over H2 by default, as in P6_DUST
            _num8 = _CAND8["_mdust"].to_numpy(float)
            _den8 = _CAND8["_mh2" if P8_DGR == "dh2" else "_mgas"].to_numpy(float)
            _CAND8["_dgr"] = np.log10(np.where((_num8 > 0) & (_den8 > 0),
                                                _num8 / np.where(_den8 > 0, _den8, np.nan), np.nan))
        _CAND8["_anchor"] = [int(k[0]) for k in _CAND8["key"]]
        _CAND8["_ok"] = [(_chain8(k) is not None) for k in _CAND8["key"]]       # all four epochs on disk
        _CAND8["_rank"] = 0.5 * (_CAND8["_ngas"].rank(pct=True) + _CAND8["_nstar"].rank(pct=True))

        _GNAME8 = [_n8 for _n8, _, _, _, _ in GRP6]                            # "sel" and "rest", the way Part 6 names GRP6
        _GINFO8 = {_n8: (_c8, _lab8) for _n8, _, _c8, _, _lab8 in GRP6}
        def _inclass8(df, name):
            return df["hi_geom"].to_numpy(bool) == (name == "sel")

        _base8 = _CAND8[_CAND8["_ok"].to_numpy(bool)].copy()
        _zok8 = _base8[_base8["_z"].to_numpy(float) > float(P8_ZMIN)]
        if len(_zok8) == 0:
            print(f"  no candidate anchor is above z = {P8_ZMIN:g} — the redshift floor is ignored")
            _zok8 = _base8
        # a galaxy with no age cannot be matched, so it is kept out of the pool rather than silently drawn blank
        _nz8 = len(_zok8)                                        # above the redshift floor, before the age cut
        _hasam8 = np.isfinite(_zok8["_age"].to_numpy(float)) & np.isfinite(_zok8["_lm"].to_numpy(float))
        _nomatch8 = _zok8[~_hasam8]
        if int((~_hasam8).sum()):
            print(f"  {int((~_hasam8).sum())} of {len(_zok8)} drawable candidates have no age in Part 0's "
                  f"{_SRC8} — they are kept out of the match")
        _zok8 = _zok8[_hasam8]

        def _hand8(name):
            """The galaxy P8_PICK names for this class, if it named one and it is drawable."""
            if name not in P8_PICK:
                return None
            _w = tuple(P8_PICK[name])
            _h = _base8[[k == _w for k in _base8["key"]]]
            if not len(_h):
                print(f"  P8_PICK['{name}'] = {_w} has no cutout at every critical point — ignored")
                return None
            return _h.iloc[0]

        REP8, _miss8, _pick8, _note8 = [], [], {}, ""
        _seedcls8 = P8_SEEDCLS if P8_SEEDCLS in _GNAME8 else _GNAME8[-1]

        def _cost8(row, age, lm):
            """How far a candidate is from (age, log M*), in units of the two tolerances."""
            _d = np.hypot((float(row["_age"]) - age) / float(P8_TOLAGE),
                          (float(row["_lm"]) - lm) / float(P8_TOLM))
            return float(_d) if np.isfinite(_d) else np.inf

        def _others8(seed):
            """The best match to `seed` in each of the other classes -> (picks, costs, crossed, unmatched)."""
            _a0, _m0 = float(seed["_age"]), float(seed["_lm"])
            _ok0 = np.isfinite(_a0) and np.isfinite(_m0)
            _p, _c, _x, _u = {}, {}, [], []
            for _n in _GNAME8:
                if _n == _seedcls8:
                    continue
                _h = _hand8(_n)
                if _h is not None:
                    _p[_n], _c[_n] = _h, (_cost8(_h, _a0, _m0) if _ok0 else np.nan)
                    continue
                _all = _zok8[_inclass8(_zok8, _n)]
                if not len(_all):                                  # nobody in this class has an age at all
                    _fb = _nomatch8[_inclass8(_nomatch8, _n)]
                    if len(_fb):
                        _p[_n], _c[_n] = _fb.iloc[int(np.argmax(np.nan_to_num(_fb["_rank"].to_numpy(float))))], np.nan
                        _u.append(_n)
                    continue
                if not _ok0:
                    _p[_n], _c[_n] = _all.iloc[int(np.argmax(np.nan_to_num(_all["_rank"].to_numpy(float))))], np.nan
                    continue
                _pool = _all
                if P8_SAMEANCHOR:
                    _same = _all[_all["_anchor"].to_numpy(int) == int(seed["_anchor"])]
                    _pool = _same if len(_same) else _all
                _score = lambda df: (np.array([_cost8(r, _a0, _m0) for _, r in df.iterrows()])
                                     - float(P8_RESW) * np.nan_to_num(df["_rank"].to_numpy(float)))
                _best = _pool.iloc[int(np.argmin(_score(_pool)))]
                # the seed's own anchor is a preference, not a cage: if it cannot match this class to within
                # P8_COSTMAX and another anchor can do better, take that and say so — a matched pair at two
                # redshifts says more about the classes than a mismatched pair at one.
                if P8_SAMEANCHOR and len(_pool) < len(_all) and _cost8(_best, _a0, _m0) > float(P8_COSTMAX):
                    _alt = _all.iloc[int(np.argmin(_score(_all)))]
                    if _cost8(_alt, _a0, _m0) < _cost8(_best, _a0, _m0):
                        _best, _ = _alt, _x.append(_n)
                _p[_n], _c[_n] = _best, _cost8(_best, _a0, _m0)
            return _p, _c, _x, _u

        _seed8, _why8 = _hand8(_seedcls8), "named by P8_PICK"
        _cost8s, _cross8, _unm8 = {}, [], []
        if _seed8 is not None:
            _pick8[_seedcls8] = _seed8
            _o, _cost8s, _cross8, _unm8 = _others8(_seed8)
            _pick8.update(_o)
        else:
            _sp8 = _zok8[_inclass8(_zok8, _seedcls8)]                 # the seed must be matchable itself
            _cut8 = _sp8[_sp8["_dgr"].to_numpy(float) < float(P8_SEEDMAX)] if P8_SEEDMAX is not None else _sp8
            _why8 = (f"log M_dust/M_{'H2' if P8_DGR == 'dh2' else 'gas'} < {P8_SEEDMAX:g}, "
                      f"{len(_cut8)} of {len(_sp8)}")
            if not len(_cut8) and len(_sp8):
                _cut8 = _sp8.nsmallest(3, "_dgr")
                _why8 = (f"NO member of this class is below {P8_SEEDMAX:g} — the three lowest were used instead "
                          f"(lowest {np.nanmin(_sp8['_dgr']):.2f})")
            # the best-resolved low-dust galaxy is the FIRST choice, but a seed the other three cannot be matched to
            # is a bad seed: walk down the resolution order until one of them can be matched to within P8_COSTMAX,
            # and if none can, keep the one whose worst match is least bad.
            _cut8 = _cut8.sort_values("_rank", ascending=False)
            _try8, _best8 = min(int(P8_SEEDTRY), len(_cut8)), None
            for _i8 in range(_try8):
                _cand8 = _cut8.iloc[_i8]
                _o, _c, _x, _u = _others8(_cand8)
                _worst = max([v for v in _c.values() if np.isfinite(v)], default=np.inf)
                if _best8 is None or _worst < _best8[0]:
                    _best8 = (_worst, _i8, _cand8, _o, _c, _x, _u)
                if _worst <= float(P8_COSTMAX):
                    break
            if _best8 is not None:
                _worst, _i8, _seed8, _o, _cost8s, _cross8, _unm8 = _best8
                _pick8[_seedcls8] = _seed8
                _pick8.update(_o)
                _why8 += ("; best resolved of those" if _i8 == 0 else
                           f"; #{_i8 + 1} in resolution order — the {_i8} better-resolved ones could not be "
                           f"matched to within {P8_COSTMAX:g}")
                if not np.isfinite(_worst) or _worst > float(P8_COSTMAX):
                    _why8 += f", and neither could this one (worst match {_worst:.2f})"
        for _n8 in _cross8:
            print(f"  '{_GINFO8[_n8][1]}' matches better at another anchor than the seed's — taken from there "
                  f"(P8_SAMEANCHOR is a preference, P8_COSTMAX the threshold)")
        for _n8 in _unm8:
            print(f"  '{_GINFO8[_n8][1]}' has no member with an age — drawn unmatched, best resolved")

        for _n8 in _GNAME8:
            if _n8 not in _pick8:
                _miss8.append(_GINFO8[_n8][1])
                continue
            _r8 = _pick8[_n8]
            _col8, _lab8 = _GINFO8[_n8]
            REP8.append(dict(group=_n8, label=_lab8, color=_col8, key=_r8["key"], chain=_chain8(_r8["key"]),
                              seed=(_n8 == _seedcls8), geom=float(_r8[P6_GEOM]),
                              z=float(_r8["_z"]), age=float(_r8["_age"]), lmstar=float(_r8["_lm"]),
                              ldgr=float(_r8["_dgr"]), ngas=float(_r8["_ngas"]), nstar=float(_r8["_nstar"]),
                              cost=(0.0 if _n8 == _seedcls8 else float(_cost8s.get(_n8, np.nan)))))
        for _lb in _miss8:
            print(f"  no galaxy of '{_lb}' survives the cuts with all of {', '.join(P8_STAGES)} on disk "
                  f"— that row is dropped")
        if not REP8:
            print("Part 8: nothing to draw — no class has a galaxy with a cutout at every critical point.")
        _DGLAB8 = (r"\log\,M_{\rm dust}/M_{\rm H_2}" if P8_DGR == "dh2" else r"\log\,M_{\rm dust}/M_{\rm gas}")
        _DGSHORT8 = (r"\log\,M_{\rm d}/M_{\rm H_2}" if P8_DGR == "dh2" else r"\log\,M_{\rm d}/M_{\rm g}")
        _fignote8 = ""
        if REP8:
            _ages8 = [_r["age"] for _r in REP8 if np.isfinite(_r["age"])]
            _lms8 = [_r["lmstar"] for _r in REP8 if np.isfinite(_r["lmstar"])]
            _anc8 = sorted({_r["key"][0] for _r in REP8})
            _note8 = ("seeded on " + _GINFO8[_seedcls8][1].replace("$", "").replace("_", "")
                       + f", then matched in age and mass;  anchor"
                       + (f" snap {_anc8[0]} (z = {REP8[0]['z']:.2f})" if len(_anc8) == 1
                          else "s " + ", ".join(str(a) for a in _anc8))
                       + (f";  age {min(_ages8):.2f}-{max(_ages8):.2f} Gyr (spread {max(_ages8) - min(_ages8):.2f}, "
                          f"tolerance {P8_TOLAGE:g})" if _ages8 else "")
                       + (f", log M* {min(_lms8):.2f}-{max(_lms8):.2f} (spread {max(_lms8) - min(_lms8):.2f}, "
                          f"tolerance {P8_TOLM:g})" if _lms8 else ""))
            _fignote8 = ("seeded on " + _GINFO8[_seedcls8][1] + r", matched in age and $M_\star$"
                          + (rf";  anchor {_anc8[0]}, $z = {REP8[0]['z']:.2f}$" if len(_anc8) == 1 else "")
                          + (rf";  spread {max(_ages8) - min(_ages8):.2f} Gyr" if _ages8 else "")
                          + (rf", {max(_lms8) - min(_lms8):.2f} dex" if _lms8 else ""))
        print(f"Part 8 — the pictures, {BOX_LABEL.get(P8_BOX, P8_BOX)}: {len(REP8)} classes x {len(P8_STAGES)} "
              f"critical points, {P8_SPOS}, "
              f"{'catalogue members only' if P8_MEMBERS else 'the whole 100 kpc cutout'}.")
        print(f"  the pool: {_pool8}, {len(_base8)} with all four cutouts, {_nz8} of them above z = "
              f"{P8_ZMIN:g}, {len(_zok8)} of those with an age to match on.  "
              f"The seed is '{_GINFO8[_seedcls8][1]}' ({_why8}).")
        if _note8:
            print("  " + _note8 + ".")
        for _r in REP8:
            _f24n = lambda v, f="{:.2f}": (f.format(v) if np.isfinite(v) else "n/a")
            print(f"    {'SEED' if _r['seed'] else '    '} {_r['label']:42s} anchor_snap {_r['key'][0]:3d} "
                  f"gal {_r['key'][1]:5d}  z {_r['z']:.2f}  age {_f24n(_r['age'])} Gyr  "
                  f"log M* {_f24n(_r['lmstar'])}  "
                  f"{'dh2' if P8_DGR == 'dh2' else 'D/G'} {_f24n(_r['ldgr'])}  "
                  + ("" if not np.isfinite(_r["cost"]) else f"match cost {_r['cost']:.2f}  ")
                  + f"({_r['ngas']:.0f} gas + {_r['nstar']:.0f} star particles)")
        if REP8 and any(np.isfinite(_r["cost"]) and _r["cost"] > 1 for _r in REP8):
            print("  NOTE a match cost above 1 means that galaxy is more than one tolerance from the seed in the "
                  "(age, log M*) plane — loosen P8_ZMIN, raise P8_SEEDMAX, or accept the spread.")

        # ── the frame: one half-width for the whole grid, and a line saying which knob set it ──
        if P8_EXT is not None:
            _EXT8 = float(P8_EXT)
            _why8 = (f"P8_SIZE = {2 * _EXT8:g} kpc a side (P8_EXTFAC and P8_EXTLIM are not used)")
        else:
            _r50_8 = (EPO6.groupby(["snap", "gx"])["r50star_cat"].first()      # (snap, gx) repeats across anchors
                       if "r50star_cat" in EPO6.columns else None)
            _rr8 = [float(_r50_8.get(sg, np.nan)) for r in REP8 for sg in r["chain"].values()] \
                if _r50_8 is not None else []
            _rr8 = [v for v in _rr8 if np.isfinite(v) and v > 0]
            _raw8 = round(P8_EXTFAC * float(np.median(_rr8))) if _rr8 else 15.0
            _EXT8 = float(np.clip(_raw8, *P8_EXTLIM))
            _why8 = (f"P8_EXTFAC {P8_EXTFAC:g} x median R50(*) "
                      + (f"{np.median(_rr8):.1f} kpc = {_raw8:g}" if _rr8 else "unavailable -> 15")
                      + (f", CLIPPED by P8_EXTLIM {tuple(P8_EXTLIM)}" if _EXT8 != _raw8 else
                         f"; P8_EXTLIM {tuple(P8_EXTLIM)} did not bind — SET P8_SIZE (kpc a side) for a larger "
                         f"region"))
        print(f"  every panel is {2 * _EXT8:.0f} kpc across (half-width {_EXT8:g} kpc from {_why8}).")

        # ── the renders: one file open per panel, every product taken from it ──
        REN8, ROW8, _warn8 = {}, [], False
        _frames8 = []                                            # (class, stage, frame_info) per panel: which frame was used
        for _r in REP8:
            _rowframe = None                                      # the row's common frame: the spin at P8_FRAME_STAGE
            if P8_FRAME_STAGE and P8_FRAME != "stellar" and P8_FRAME_STAGE in _r["chain"]:
                _rf = ReducedRender.from_epoch(P8_REDUCED, *_r["chain"][P8_FRAME_STAGE], P8_PREFIX, members_only=P8_MEMBERS,
                                               frame=P8_FRAME, frame_rmax=P8_FRAME_RMAX, frame_members=P8_FRAME_MEMBERS)
                if _rf.frame != "stellar":
                    _rowframe = (_rf.evecs.copy(), dict(_rf.frame_info))
                del _rf
            for _s8 in P8_STAGES:
                _sn, _gx = _r["chain"][_s8]
                _rr = ReducedRender.from_epoch(P8_REDUCED, _sn, _gx, P8_PREFIX, members_only=P8_MEMBERS,
                                               frame=P8_FRAME, frame_rmax=P8_FRAME_RMAX, frame_members=P8_FRAME_MEMBERS)
                _own = dict(getattr(_rr, "frame_info", {}))
                if _rowframe is not None:
                    _rr.set_evecs(_rowframe[0], label=f"{P8_FRAME}@{P8_FRAME_STAGE}", info=_rowframe[1])
                    _own = dict(_rr.frame_info, own_coherence=_own.get("coherence", np.nan), own_n=_own.get("n_used", 0))
                _frames8.append((_r.get("name", _r.get("cls", "")) if isinstance(_r, dict) else "", _s8, _own))
                if np.isfinite(_rr.rmax_kpc) and _EXT8 > _rr.rmax_kpc and not _warn8:
                    print(f"  NOTE the frame half-width {_EXT8:g} kpc runs past the cutout's own rmax_kpc = "
                          f"{_rr.rmax_kpc:g}: the corners are empty because there are no particles there, not because "
                          f"the sky is.")
                    _warn8 = True
                _cell = dict(z=_rr.redshift, ext=[-_EXT8, _EXT8, -_EXT8, _EXT8],
                             rgb=_rr.rgb(ex=_EXT8, spos=P8_SPOS, channels=P8_CHAN, dyn=P8_DYN, pmax=P8_PMAX,
                                         stretch=P8_STRETCH, tau_max=P8_TAU, xsize=P8_XSIZE)[0],
                             tot=_rr.totals(ex=_EXT8, spos=P8_SPOS),
                             memb=_rr.totals(ex=None, spos=P8_SPOS, members=True))
                for _f8 in P8_MAPS:
                    _cell[f"map_{_f8}"] = (
                        _rr.sigma_map(_f8, ex=_EXT8, spos=P8_SPOS, xsize=P8_XSIZE)[0]
                        if P8_MAPDEF[_f8][0] == "sigma" else
                        _rr.weighted_map(_f8, ex=_EXT8, spos=P8_SPOS, xsize=P8_XSIZE)[0])
                if P8_STREAM:
                    _cell["vel"], _cell["vel_n"] = _stream8(_rr)
                REN8[(_r["group"], _s8)] = _cell
                _row = dict(group=_r["group"], label=_r["label"], stage=_s8, anchor_snap=_r["key"][0],
                            gal_id=_r["key"][1], snap=_sn, gx=_gx, z=_rr.redshift, ext_kpc=_EXT8,
                            loaded="members" if P8_MEMBERS else "region",
                            stream_gas=(f"slab +-{P8_STREAM_DEPTH:g} kpc" if P8_STREAM_SOURCE == "slab" else P8_STREAM_SOURCE) if P8_STREAM else "",
                            stream_n_used=_cell.get("vel_n", (0, 0))[0], stream_n_in_frame=_cell.get("vel_n", (0, 0))[1])
                _row.update({f"frame_{k}": v for k, v in _cell["tot"].items()})       # what is drawn in the panel
                _row.update({f"member_{k}": v for k, v in _cell["memb"].items()})     # the catalogue galaxy itself
                ROW8.append(_row)

        _stlab8 = {"jet_on": "Jet start", "sft": "Quenching", "qt": "Quenched", "anchor": "Observed", "agn_ign": "ignition",
                    "sf_peak": "SF peak", "post_quench": "post-q", "gas_min": "gas min", "end": "end"}

        def _grid8(draw, title, fname, cbar=None, cbar_label=None, fg="w", bg="k"):
            """The 4 x 4 frame every figure of this part shares: classes down, critical points across."""
            if not REP8:
                return
            if P8_STREAM and REN8 and "vel_n" in next(iter(REN8.values())):
                _nu = [c["vel_n"] for c in REN8.values()]
                print(f"  streamlines: the {P8_STREAM_WEIGHT}-weighted velocity field of "
                      + {"slab": f"the gas within +-{P8_STREAM_DEPTH:g} kpc of the mid-plane along the line of sight (P8_STREAM_SOURCE 'slab')",
                         "members": "the galaxy's CAESAR member gas only", "region": "EVERY gas particle in the frame (CGM and neighbours included)"}[P8_STREAM_SOURCE]
                      + f", the {'galaxy' if P8_STREAM_BULK == 'members' else 'selection'}'s bulk velocity subtracted"
                      + f" — per panel {min(n for n, _ in _nu)}-{max(n for n, _ in _nu)} particles used of {min(m for _, m in _nu)}-{max(m for _, m in _nu)} in the frame")
            if _frames8:                                             # the frame every panel was drawn in
                _nfb = sum(1 for _, _, fi in _frames8 if fi.get("frame") == "stellar")
                print(f"  projection frame P8_FRAME = {P8_FRAME}" + (f" measured at {P8_FRAME_STAGE} for the whole row" if P8_FRAME_STAGE else "")
                      + f": {len(_frames8) - _nfb} of {len(_frames8)} panels in that frame"
                      + (f", {_nfb} fell back to the stellar frame" if _nfb else "") + " (tilt = angle between that spin and the stellar minor axis):")
                for _lb8, _st8, fi in _frames8:
                    print(f"    {str(_lb8)[:34]:34s} {_st8:7s} {fi.get('frame', '?'):10s} n={fi.get('n_used', 0):5d}"
                          + (f"  coherence {fi['coherence']:.2f}" if np.isfinite(fi.get("coherence", np.nan)) else "")
                          + (f"  (this epoch's own spin: coherence {fi['own_coherence']:.2f}, n={fi['own_n']})" if "own_coherence" in fi else "")
                          + (f"  tilt {fi['tilt_vs_stellar_deg']:5.1f} deg" if "tilt_vs_stellar_deg" in fi else f"  ({fi.get('reason', '')})"))
            fig, axs = plt.subplots(len(REP8), len(P8_STAGES),
                                    figsize=(2.9 * len(P8_STAGES) / (1 - P8_MARGIN), 3.05 * len(REP8) + 0.9),
                                    squeeze=False)
            for _i, _r in enumerate(REP8):
                for _j, _s8 in enumerate(P8_STAGES):
                    ax = axs[_i][_j]
                    _cell = REN8[(_r["group"], _s8)]
                    draw(ax, _cell, _r, _s8)
                    ax.set_xticks([]); ax.set_yticks([])
                    for _sp in ax.spines.values():
                        _sp.set_color(_r["color"]); _sp.set_linewidth(2.0)
                    _dg8 = (_cell["memb"].get("m_dust", np.nan)
                             / max(_cell["memb"].get("m_H2" if P8_DGR == "dh2" else "m_gas", np.nan), 1e-30))
                    _ann8 = f"$z = {_cell['z']:.2f}$" + (f"\n${_DGSHORT8} = {np.log10(_dg8):.2f}$"
                                                         if np.isfinite(_dg8) and _dg8 > 0 else "")
                    ax.text(0.03, 0.97, _ann8, transform=ax.transAxes, ha="left", va="top", linespacing=1.5,
                            fontsize=_fs8(8.5), color="0.1", zorder=12,
                            bbox=dict(boxstyle="round,pad=0.32", facecolor="white", alpha=0.82, edgecolor="none"))
                    if _i == 0:
                        ax.set_title(_stlab8.get(_s8, _s8), fontsize=_fs8(12))
                    if _j == 0:
                        _bar = 5.0 if _EXT8 <= 25 else (10.0 if _EXT8 <= 60 else 25.0)   # once per row
                        _f24b = _bar / (2 * _EXT8)                   # its length as a fraction of the panel
                        ax.plot([0.06, 0.06 + _f24b], [0.065, 0.065], "-", transform=ax.transAxes, color=fg, lw=2.6,
                                zorder=12, path_effects=[pe.withStroke(linewidth=4.2, foreground=bg)])
                        ax.text(0.06 + 0.5 * _f24b, 0.105, f"{_bar:.0f} kpc", transform=ax.transAxes, color=fg,
                                ha="center", va="bottom", fontsize=_fs8(8), zorder=12,
                                path_effects=[pe.withStroke(linewidth=2.2, foreground=bg)])
            #fig.suptitle(title, fontsize=_fs8(13) if "\n" not in title else 11.5)
            _rt8 = P8_SIDE == "right"
            fig.tight_layout(rect=(0 if _rt8 else P8_MARGIN, 0.075 if cbar is not None else 0,
                                   1 - P8_MARGIN if _rt8 else 1, 0.985))
            fig.subplots_adjust(wspace=P8_WSPACE, hspace=P8_WSPACE)     # images: the grid wants almost no gutter
            for _i, _r in enumerate(REP8):                      # written after the layout, so nothing is clipped
                _bb = (axs[_i][-1] if _rt8 else axs[_i][0]).get_position()
                _x, _ha = (_bb.x1 + 0.010, "left") if _rt8 else (_bb.x0 - 0.010, "right")
                _y = _bb.y0 + 0.5 * _bb.height
                _v8 = lambda k: (f"{_r[k]:.2f}" if np.isfinite(_r.get(k, np.nan)) else "n/a")
                _num8 = [rf"$\log M_\star = $ {_v8('lmstar')}", rf"age $= $ {_v8('age')} Gyr",
                          "$" + _DGLAB8 + "$ $=$ " + _v8("ldgr")]
                # if np.isfinite(_r.get("cost", np.nan)) and not _r.get("seed"):
                #     _num8.append(rf"match cost {_r['cost']:.2f}")
                fig.text(_x, _y + 0.010, _r["label"].replace(" $+$ ", "\n"), ha=_ha, va="bottom", fontsize=_fs8(12),
                         color=_r["color"], linespacing=1.35, fontweight="bold")
                fig.text(_x, _y - 0.012, "\n".join(_num8), ha=_ha,
                         va="top", fontsize=_fs8(9), color="0.2", linespacing=1.6)
            if cbar is not None:
                _cax = fig.add_axes([0.30, 0.022, 0.42, 0.013])
                _cb8 = fig.colorbar(cbar, cax=_cax, orientation="horizontal")
                _cb8.ax.tick_params(labelsize=_fs8(9))
                _cax.set_title(cbar_label, fontsize=_fs8(9.5), pad=4)
            paper_save(fig, fname)
            plt.show()

        # ── FIGURE 1: the composite ──
        def _drawrgb8(ax, cell, r, s):
            ax.imshow(cell["rgb"], extent=cell["ext"], origin="lower")
            if r is REP8[0] and s == P8_STAGES[0]:                 # name the channels once, in their own colours
                for _k8, _nm8 in enumerate(P8_CHAN):
                    ax.text(0.965, 0.955 - 0.08 * _k8, _nm8, transform=ax.transAxes, ha="right", va="top",
                            fontsize=_fs8(9.5), weight="bold",
                            color=ReducedRender.CHANNEL_CMAPS.get(_nm8, cm.Greys_r)(0.88),
                            path_effects=[pe.withStroke(linewidth=2.2, foreground="k")])

        _grid8(_drawrgb8,
                rf"Part 8 — one galaxy per class through its own quenching ({BOX_LABEL.get(P8_BOX, P8_BOX)}, "
                rf"{P8_SPOS}, {' $+$ '.join(P8_CHAN)}, {P8_STRETCH} stretch)"
                + ("\n" + _fignote8 if _fignote8 else ""),
                f"kappa_gallery_rgb_{P8_BOX}")

        # ── FIGURE 2+: the physical maps, one per field, shared colour scale, velocity field on top ──
        for _f8 in (P8_MAPS if REP8 else []):
            _kind8, _lab8, _unit8, _log8, _cm8, _sym8 = P8_MAPDEF[_f8]
            _cm8 = P8_CMAP.get(_f8, _cm8)
            _all8 = np.concatenate([REN8[(r["group"], s)][f"map_{_f8}"].ravel()
                                     for r in REP8 for s in P8_STAGES])
            _all8 = _all8[np.isfinite(_all8)]
            if _log8:
                _all8 = _all8[_all8 > 0]
            if not len(_all8):
                print(f"  the {_f8} maps are empty everywhere — figure skipped")
                continue
            if _sym8:                                     # a diverging quantity: one symmetric scale about zero
                _q8 = float(np.percentile(np.abs(_all8), P8_PCT[1])); _vmn, _vmx = -_q8, _q8
            else:
                _vmn, _vmx = (float(np.percentile(_all8, P8_PCT[0])), float(np.percentile(_all8, P8_PCT[1])))
                if _log8:
                    _vmn, _vmx = np.log10(_vmn), np.log10(_vmx)
                    if P8_MAPDYN is not None:
                        _vmn = max(_vmn, _vmx - float(P8_MAPDYN))
            _pre8 = r"\log\," if _log8 else ""
            _cbl8 = "$" + _pre8 + _lab8 + "$" + ("  [$" + _unit8 + "$]" if _unit8 else "")
            _sm8 = cm.ScalarMappable(norm=mcolors.Normalize(_vmn, _vmx), cmap=_cm8)

            def _draw8(ax, cell, r, s, _f=_f8, _lg=_log8, _c=_cm8, _lo=_vmn, _hi=_vmx):
                ax.set_facecolor(plt.get_cmap(_c)(0.0))    # empty pixels take the colormap's own low end, so the
                _m = cell[f"map_{_f}"]                     # white streamlines and labels read everywhere
                with np.errstate(divide="ignore", invalid="ignore"):
                    _d = np.log10(np.where(_m > 0, _m, np.nan)) if _lg else np.where(np.isfinite(_m), _m, np.nan)
                ax.imshow(_d, extent=cell["ext"], origin="lower", vmin=_lo, vmax=_hi, cmap=_c)
                if P8_STREAM and cell.get("vel") is not None:
                    _X, _Y, _VX, _VY, _den = cell["vel"]
                    if np.any(np.hypot(_VX, _VY) > 0):
                        ax.streamplot(_X, _Y, _VX, _VY, color="w", density=P8_STREAM_DENSITY, linewidth=P8_STREAM_LW, arrowsize=P8_STREAM_ARROW,
                                      zorder=4)
                ax.set_xlim(cell["ext"][0], cell["ext"][1]); ax.set_ylim(cell["ext"][2], cell["ext"][3])

            _grid8(_draw8,
                    "Part 8 — $" + _lab8 + "$ through the critical points, one galaxy per class "
                    + f"({BOX_LABEL.get(P8_BOX, P8_BOX)}, {P8_SPOS}"
                    + ((f", the gas velocity field within $\\pm${P8_STREAM_DEPTH:g} kpc of the mid-plane as streamlines)" if P8_STREAM_SOURCE == "slab" else
                        ", the member gas velocity field as streamlines)" if P8_STREAM_SOURCE == "members" else ", the region's gas velocity field as streamlines)") if P8_STREAM else ")")
                    + ("\n" + _fignote8 if _fignote8 else ""),
                    f"kappa_gallery_{_f8}_{P8_BOX}", cbar=_sm8, cbar_label=_cbl8,
                    fg="w", bg="k")

        if ROW8:
            _p8 = os.path.join(OUT, f"kappa_gallery_{P8_BOX}.csv")
            pd.DataFrame(ROW8).to_csv(_p8, index=False)
            print(f"  what was drawn, panel by panel (ids, epochs, masses inside the frame) -> {_p8}")
```

## Part 9 — the m25 branch on the clock: the X-ray mode, and what the dust is divided by

Parts 3–8 read the dust as $M_{\rm dust}/M_\star$ — one numerator over one denominator. This part keeps the numerator and
changes the denominator, on the 25 Mpc box only and on the same truncation clock, because *which ratio moves* is a physical
statement and $M_{\rm dust}/M_\star$ alone cannot make it:

| panel | ratio | what a fall in it means |
|---|---|---|
| (b) | $M_{\rm dust}/M_\star$ | the dust per unit of the stellar mass that is already there — the observable, and the label of Part 0 |
| (c) | $M_{\rm dust}/M_{\rm H_2}$ | the grains per unit **cold gas**: falls only if the grains go *and the H₂ stays* |
| (d) | $M_{\rm dust}/(M_{\rm dust} + Z_{\rm gas} M_{\rm gas})$ | the share of the **metals** that is in grains: falls only under grain **destruction**, and is flat under dilution, outflow or the accretion of pristine gas, which remove metals and grains together |

Panel (a) is the third clock and the reason the other three are read next to it: the **fraction of each class whose branch
has SIMBA's X-ray channel firing** at that point — an occupancy, not a dose. The channel is the mechanism Part 4 blames for
the dust destruction, and this says when it switches on relative to the truncation, class by class.

The dust-to-metal ratio is the bounded *metal-share-in-grains* form. It needs no assumption about whether caesar's `zgas`
already carries the grain metals, and it cannot exceed 1. **Neither ratio is floored** (Part 3's `lfd` is, at `P9_FLOOR`): a
step with no dust or no H₂ left is dropped rather than pinned to a floor whose value would then set the median — the cell
prints what that costs (it is well under 1 % of the branch steps).

This part also builds **the branch index** Parts 10–11 need: the most-massive-progenitor walk of every m25 galaxy of Part 0's
sample, $(\text{anchor snap}, \text{gal id}) \rightarrow g_x(\text{snap})$, from the same merger-tree pointer the histories
were built from — caesar's `tree_data/progen_galaxy_star`, and for m25, whose shared catalogues never got a tree, the
`progen_links` sidecar `find_progen_m25_job.py` wrote. It is cached (`P9_IDX`; delete it to rebuild).

The index makes the **de-duplication exact**. A galaxy that is a quenched anchor at several snapshots is in Part 0's sample
several times and its branches then overlap; two anchors are the same object when one *lies on* the other's branch, and only
the latest is kept. In m25 that is 75 anchor rows → 44 galaxies (11 rotating, 33 the rest). `P9_DEDUP = False` restores every
anchor row, at the price of counting the same object several times in the same bootstrap.

```python
# ── Part 9 — the m25 branch index, and the CLOCKS the catalogue chains already carry ──
# Every part above this one reads the m25 galaxies at their anchor, or at the K-S products' handful of critical epochs. This
# part opens the branch itself and puts THREE clocks on it, all in the 25 Mpc box only and all split by Part 2's selection:
#   (a) THE X-RAY MODE, as an occupancy:  the FRACTION of the galaxies of each class whose branch has SIMBA's X-ray channel
#       firing at that point of the clock (Wilson 68 % on the count, and the count itself drawn behind) — not a dose, not an
#       integral: what share of each class is in the mode right now.
#   (b) log M_dust/M_H2 — the dust per unit molecular gas: the grain-to-cold-gas ratio the sub-mm continuum-to-CO ratio measures.
#   (c) log M_dust/(M_dust + Z_gas M_gas) — the DUST-TO-METAL ratio in the bounded metal-share-in-grains form: what fraction of
#       the metals sits in grains. It needs no assumption about whether caesar's zgas already carries the grain metals, and it
#       cannot exceed 1. This is the ratio that separates DESTRUCTION (metals stay, grains go: DTM falls) from DILUTION or
#       accretion of pristine gas (both metals and grains go, or arrive: DTM flat).
#   ... with (d) log M_dust/M*, the clock Part 3 already draws, on the same figure so the three ratios can be read against
#       each other: the same numerator, three different denominators, and the one that moves is the answer.
# NEITHER RATIO IS FLOORED here (Part 3's lfd is, at P9_FLOOR): a step with no dust or no H2 left is dropped from the ratio
# rather than pinned to a floor whose value would then set the median. The share of steps that costs is PRINTED.
# THE BRANCH INDEX is built here too, because Parts 10-12 need it: the most-massive-progenitor walk of every m25 galaxy of
# Part 0's sample, snapshot by snapshot, from its anchor back to the start of its chain — (anchor_snap, gal_id) -> gx(snap).
# It comes from the same merger-tree pointer the histories were built from (caesar tree_data/progen_galaxy_star, and for m25 —
# whose shared catalogues never got a tree — the progen_links sidecar find_progen_m25_job.py wrote), and it is cached.
# ONE CHAIN PER GALAXY: a galaxy that is a quenched anchor at several snapshots is in Part 0's sample several times, and its
# branches then overlap. The index makes the de-duplication EXACT (two anchors are the same object when one lies on the
# other's branch, not a fingerprint match) and ONE anchor row per galaxy is kept — the latest, its quenched end state.
# That de-duplication is not a loss of statistics: without it the same object enters a median up to seven times. A TRUNCATED
# INDEX is one, and it is silent, so the cache is now incremental and self-checking — a chain the cache does not carry is
# walked and appended, and the coverage against the sample is printed every run (see (B)). What the de-duplication does cost
# is the CLASS: kappa_rot(gas) is re-measured at every anchor and falls as the gas goes, so the latest anchor is the most
# isotropic reading of the same object — P9_CLASS chooses which of the galaxy's own anchors Part 2's call is taken from, and
# all three answers are printed whichever is in force.
# The per-step AGN STATE CODE and the BHAR are kept on every chain as well (st, mdot): Part 12 reads the critical epochs —
# the AGN ignition and the first X-ray step — off them, so the stages and the clocks come from one and the same history.
P9_BOX     = "cis25"        # this part and Parts 10-11 are the 25 Mpc box only
P9_DEDUP   = True           # one chain per galaxy (the latest anchor); False = every anchor row of Part 0's sample
P9_CLASS   = "latest"       # which anchor of a de-duplicated galaxy Part 2's class is read at: "latest" = the anchor kept
                            # (the quenched end state) | "ever" = rotating at ANY anchor it is in the sample at | "sft" =
                            # at the anchor nearest its own truncation (the epoch these clocks are drawn on)
P9_NMIN    = 3              # an epoch is drawn only where this many galaxies contribute (Part 5's floor: m25 is tens of objects)
P9_MARKS   = [-1.0, 0.0, 1.0, 2.0]     # the epochs the table under the figure reports [Gyr from SFT]
P9_SPAG    = True           # the individual galaxy tracks behind the medians
P9_LINKDIR = os.path.join(os.getcwd(), "output", "{box}", "progen_links")   # the m25 merger-tree sidecars
P9_IDX     = os.path.join(OUT, "kappa_m25_branch_ids.csv")                  # the branch index cache (delete to rebuild)
P9_SMIN    = 10             # the walk stops here (below the sidecars' range; the branch simply ends)
P9_PANELS  = [("xrf",  "(a) the X-ray mode: the share of each class with the channel on", "X-ray channel on  [fraction of the class]"),
              ("lfd",  r"(b) the dust per unit stellar mass", r"$\log\, M_{\rm dust}/M_\star$"),
              ("ldh2", r"(c) the dust per unit H$_2$", r"$\log\, M_{\rm dust}/M_{\rm H_2}$"),
              ("ldtm", r"(d) the metals that are in grains", r"$\log\, M_{\rm dust}/(M_{\rm dust} + Z_{\rm gas} M_{\rm gas})$")]

if "KROT" not in globals():
    print("Part 9: needs Parts 0 and 2 — run them first.")
elif not os.path.exists(HIST_FILE.format(box=P9_BOX)):
    print(f"Part 9: {HIST_FILE.format(box=P9_BOX)} is missing — paper_ism_prediction_boxes Part 6a builds it.")
else:
    S9 = QM[QM["box"] == P9_BOX].copy()
    WANT9 = {(int(a_), int(g_)) for a_, g_ in zip(S9["snap"], S9["gal_id"])}

    def _st9(S, j, ex):
        """One galaxy's AGN state at every step as a single CODE — 0 dormant, 1 radiative, 2 X-ray choked, 3 X-ray on
        (NaN where the galaxy does not exist yet). The four states are exclusive, so nothing is lost by collapsing them,
        and Part 12 reads the stage times (ignition, the first X-ray step) straight off this array."""
        out = np.full(len(ex), np.nan)
        for _ci, _sn in enumerate(("dormant", "radiative", "choked", "xray")):
            out = np.where(np.nan_to_num(S[_sn][:, j]) > 0.5, float(_ci), out)
        return np.where(ex, out, np.nan)

    # ── (A) the branch: one pass over the histories for the tracks and the snapshot -> time ladder ──
    T9, CH9 = {}, {}
    with h5py.File(HIST_FILE.format(box=P9_BOX), "r") as _f9:
        for _k9 in _f9:
            _g9 = _f9[_k9]; _a9 = int(_k9[4:])
            _t = _g9["t_yr"][:] / 1e9; _o = np.argsort(_t); _t = _t[_o]
            for _s_, _tt in zip(np.asarray(_g9["snaps"][:])[_o].astype(int), _t):
                T9[int(_s_)] = float(_tt)                         # the snapshot -> cosmic time ladder, shared by Parts 10-11
            _ids = np.asarray(_g9["gal_ids"][:]).astype(np.int64)
            if not any((_a9, int(v)) in WANT9 for v in _ids):
                continue
            _ms, _sfr, _mg, _mbh, _fe = (_g9[c][:][_o] for c in ("mstar", "sfr", "mgas", "mbh", "fedd"))
            _md = _g9["mdust"][:][_o] if "mdust" in _g9 else np.full_like(_ms, np.nan)
            _mh = _g9["mh2"][:][_o] if "mh2" in _g9 else np.full_like(_ms, np.nan)
            _zg = _g9["zgas"][:][_o] if "zgas" in _g9 else np.full_like(_ms, np.nan)
            _St, _Dg, _X = agn_states(_ms, _mg, _mbh, _fe)
            _snaps = np.asarray(_g9["snaps"][:])[_o].astype(int)
            for _j, _gid in enumerate(_ids):
                _key = (_a9, int(_gid))
                if _key not in WANT9:
                    continue
                _ex = _X["exist"][:, _j]
                if _ex.sum() < 3:
                    continue
                with np.errstate(all="ignore"):
                    _den = np.where(_ex, _ms[:, _j], np.nan)
                    _met = _zg[:, _j] * _mg[:, _j]                        # the metal mass in the gas phase
                    _r = lambda num, den: np.log10(np.where((num > 0) & (den > 0) & _ex,
                                                            num / np.where(den > 0, den, np.nan), np.nan))
                    CH9[_key] = dict(
                        t=_t, snaps=_snaps, ex=_ex,
                        xrf=np.where(_ex, ((np.nan_to_num(_St["xray"][:, _j]) > 0.5) & _ex).astype(float), np.nan),
                        st=_st9(_St, _j, _ex), mdot=np.where(_ex, _X["mdot"][:, _j], np.nan),   # for Part 12's stages
                        lfd=_r(_md[:, _j], _den),                          # NOT floored: see the header
                        ldh2=_r(_md[:, _j], _mh[:, _j]),
                        ldtm=_r(_md[:, _j], _md[:, _j] + _met),
                        lfh2=_r(HE * _mh[:, _j], _den), lfgas=_r(_mg[:, _j], _den),
                        lz=np.log10(np.where((_zg[:, _j] > 0) & _ex, _zg[:, _j], np.nan)))
    print(f"Part 9 — the {BOX_LABEL[P9_BOX]} branch: {len(CH9)} of {len(WANT9)} anchor rows of Part 0's sample have a chain; "
          f"the snapshot ladder covers {min(T9)}-{max(T9)}")

    # ── (B) the branch INDEX: the most-massive-progenitor walk, cached ──
    def _prog9(box, snap):
        """The most-massive-progenitor pointer of snapshot `snap` (index at snap-1, -1 = none): the progen_links sidecar
        first (the m25 catalogues carry no tree), the catalogue's own tree_data otherwise. None if neither is there."""
        _pick = lambda a: (a[:, 0] if np.ndim(a) == 2 else a).astype(np.int64)
        _cdir, _fmt = (_catcfg3(box) if "_catcfg3" in globals()
                       else (None, "%s_{snap:03d}.hdf5" % {"cis25": "m25n512", "cis50": "m50n512",
                                                           "cis50nox": "m50n512", "cis100": "m100n1024"}.get(box, box)))
        for _p in (os.path.join(P9_LINKDIR.format(box=box), _fmt.format(snap=int(snap))),
                   os.path.join(_cdir or "", _fmt.format(snap=int(snap)))):
            if os.path.exists(_p):
                try:
                    with h5py.File(_p, "r") as _f:
                        if "tree_data/progen_galaxy_star" in _f:
                            return _pick(_f["tree_data/progen_galaxy_star"][:])
                except OSError:
                    continue
        return None

    def _walk9(rows):
        """The most-massive-progenitor walk of the given (anchor_snap, gal_id) rows -> the index rows, and the snapshots
        that had no pointer. Every anchor JOINS the walk at its own snapshot, so a snapshot where all the chains alive die
        must not end the pass while anchors below it are still waiting to join — it ended it there before, which is why a
        cache written in one go could carry only the anchors above the first gap."""
        _anch = {}
        for _a, _g in rows:
            _anch.setdefault(int(_a), []).append(int(_g))
        _cur, _out, _miss, _lo = {}, [], [], min(_anch)
        for _s in range(max(_anch), P9_SMIN - 1, -1):
            for _g in _anch.get(_s, []):
                _cur[(_s, _g)] = _g                               # a galaxy joins the walk at its own anchor
            for _k, _gx in _cur.items():
                _out.append((_k[0], _k[1], _s, _gx))
            _P = _prog9(P9_BOX, _s)
            if _P is None and _cur:
                _miss.append(_s)
            _nxt = {}
            for _k, _gx in _cur.items():
                _v = int(_P[_gx]) if (_P is not None and 0 <= _gx < len(_P)) else -1
                if _v >= 0:
                    _nxt[_k] = _v
            _cur = _nxt
            if not _cur and _s <= _lo:                            # nothing alive AND no anchor left to join: done
                break
        return _out, _miss

    # THE CACHE IS INCREMENTAL: it is kept for the chains it already carries and the ones it does not are walked and
    # appended, so a cache written for a NARROWER sample (a different Z_MAX / AGE_MIN / mass cut, or the walk stopping at
    # its first gap) can never silently truncate this part and Parts 10-13. What it covers is printed every run.
    BR9 = pd.DataFrame(columns=["anchor_snap", "gal_id", "snap", "gx"])
    if os.path.exists(P9_IDX):
        BR9 = pd.read_csv(P9_IDX)
        BR9 = BR9[[(int(a_), int(g_)) in CH9 for a_, g_ in zip(BR9["anchor_snap"], BR9["gal_id"])]].reset_index(drop=True)
    _todo9 = sorted(set(CH9) - {(int(a_), int(g_)) for a_, g_ in zip(BR9["anchor_snap"], BR9["gal_id"])})
    if _todo9:
        _rows9, _miss9 = _walk9(_todo9)
        _new9 = pd.DataFrame(_rows9, columns=["anchor_snap", "gal_id", "snap", "gx"])
        BR9 = (pd.concat([BR9, _new9], ignore_index=True) if len(BR9) else _new9).astype(int)
        BR9.to_csv(P9_IDX, index=False)
        print(f"  the branch index -> {P9_IDX}  ({len(BR9)} nodes, "
              f"{BR9.groupby(['anchor_snap', 'gal_id']).ngroups} chains, median "
              f"{int(BR9.groupby(['anchor_snap', 'gal_id']).size().median())} snapshots each; "
              f"{len(_todo9)} chain{'s' if len(_todo9) != 1 else ''} walked and appended this run)"
              + (f";  no merger-tree pointer at snapshots {_miss9[:6]}{'...' if len(_miss9) > 6 else ''} — those branches "
                 "stop there" if _miss9 else ""))
    else:
        print(f"  the branch index, cached -> {P9_IDX}  ({len(BR9)} nodes, complete for the sample)")
    _ncov9 = BR9.groupby(["anchor_snap", "gal_id"]).ngroups if len(BR9) else 0
    if _ncov9 < len(CH9):
        print(f"  WARNING: {len(CH9) - _ncov9} of the {len(CH9)} anchor rows still have no chain — the walk found no "
              "progenitor pointer for them; Parts 9-13 run on the rest")
    BR9["t"] = BR9["snap"].map(T9)
    NODE9 = {k: set(zip(d["snap"].astype(int), d["gx"].astype(int))) for k, d in BR9.groupby(["anchor_snap", "gal_id"])}

    # ── (C) one chain per galaxy: EXACT de-duplication on the index (an anchor that lies on a later anchor's branch is it) ──
    _ord9 = sorted(NODE9, key=lambda k: -k[0])
    KEEP9, _dupe9, GRP9 = [], set(), {}
    for _k9 in _ord9:
        if _k9 in _dupe9:
            continue
        KEEP9.append(_k9)
        GRP9[_k9] = [_k9]                                     # every anchor row of Part 0's sample that IS this galaxy
        for _k2 in _ord9:
            if _k2 != _k9 and (_k2[0], _k2[1]) in NODE9[_k9]:
                _dupe9.add(_k2); GRP9[_k9].append(_k2)
    if not P9_DEDUP:
        KEEP9 = list(_ord9); GRP9 = {k: [k] for k in _ord9}
    KEEP9 = [k for k in KEEP9 if k in CH9]
    _tsft9 = {(int(a_), int(g_)): float(v_) for a_, g_, v_ in zip(S9["snap"], S9["gal_id"], S9["t_sft"])}

    def _cls9(k, how):
        """THE CLASS OF A DE-DUPLICATED GALAXY. Part 2 calls kappa_rot(gas) at every anchor the galaxy is a quenched
        member at, and those calls need not agree: the rotation is measured on the gas that is left, and it goes. So the
        latest anchor — the end state — is systematically the most isotropic reading of one and the same object."""
        _rows = GRP9.get(k, [k])
        if how == "ever":                                     # rotating at ANY anchor it is in the sample at
            return "rot" if any(KROT.get((P9_BOX, a_, g_), False) for a_, g_ in _rows) else "iso"
        if how == "sft":                                      # the anchor closest in time to its own truncation
            _t0 = _tsft9.get(k, np.nan)
            _c9 = [(abs(T9[a_] - _t0), a_, g_) for a_, g_ in _rows if a_ in T9]
            if np.isfinite(_t0) and _c9:
                _, _a9, _g9 = min(_c9)
                return "rot" if KROT.get((P9_BOX, _a9, _g9), False) else "iso"
        return "rot" if KROT.get((P9_BOX, k[0], k[1]), False) else "iso"      # "latest": the anchor that was kept

    CLS9 = {k: _cls9(k, P9_CLASS) for k in KEEP9}
    print(f"  one chain per galaxy: {len(_ord9)} anchor rows -> {len(KEEP9)} "
          + ("galaxies" if P9_DEDUP else "chains (P9_DEDUP is off)")
          + f";  {sum(v == 'rot' for v in CLS9.values())} {KCLS['rot']['short']} / "
            f"{sum(v == 'iso' for v in CLS9.values())} {KCLS['iso']['short']}"
          + f";  {sum(np.isfinite(_tsft9.get(k, np.nan)) for k in KEEP9)} of them have an SFT")
    _nan9 = [len(GRP9.get(k, [k])) for k in KEEP9] or [0]
    print(f"  the class is read at P9_CLASS = '{P9_CLASS}' (a galaxy is a quenched anchor {np.median(_nan9):.0f} times in "
          f"the median, {max(_nan9)} at most):  "
          + ",  ".join(f"'{_h9}' {sum(_cls9(k, _h9) == 'rot' for k in KEEP9)} {KCLS['rot']['short']}"
                       for _h9 in ("latest", "ever", "sft")))

    # ── (D) the clocks ──
    _TRK9 = {c: {} for c, _, _ in P9_PANELS}
    _drop9 = {c: [0, 0] for c, _, _ in P9_PANELS}
    for _k9 in KEEP9:
        _d9, _ts9 = CH9[_k9], _tsft9.get(_k9, np.nan)
        for _c9, _, _ in P9_PANELS:
            _y9 = _d9[_c9]
            _drop9[_c9][0] += int(np.sum(_d9["ex"] & ~np.isfinite(_y9))); _drop9[_c9][1] += int(np.sum(_d9["ex"]))
            _TRK9[_c9][_k9] = (to_grid(_d9["t"] - _ts9, _y9) if np.isfinite(_ts9) else np.full(len(TG), np.nan))
    ROW9 = []
    fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
    for _pi, (_c9, _ttl, _ylab) in enumerate(P9_PANELS):
        ax = axs[_pi // 2, _pi % 2]
        axb = ax.twinx(); axb.set_zorder(ax.get_zorder() - 1); ax.patch.set_visible(False); axb.set_yticks([])
        _top = 1.0
        for _cl in KC:
            _ks = [k for k in KEEP9 if CLS9[k] == _cl]
            if len(_ks) < 2:
                continue
            A9 = np.asarray([_TRK9[_c9][k] for k in _ks], float)
            if _c9 == "xrf":                       # an OCCUPANCY, not a level: the count of galaxies in the mode, Wilson 68 %
                _fin = np.isfinite(A9); _n9 = _fin.sum(axis=0); _k_ = (_fin & (A9 >= 0.5)).sum(axis=0)
                _p = np.array([wilson(k_, n_) for k_, n_ in zip(_k_, _n9)], float)
                cen, lo, hi = (np.where(_n9 >= P9_NMIN, _p[:, i], np.nan) for i in range(3))
                n_t = _n9
            else:
                cen, lo, hi, n_t = med_band(A9, nmin=P9_NMIN)
                if P9_SPAG:
                    _al = float(np.clip(20.0 / max(len(_ks), 1), 0.05, 0.35))
                    for _row in A9:
                        ax.plot(TG, _row, color=KCLS[_cl]["color"], lw=0.4, alpha=_al, zorder=2)
            _top = max(_top, float(np.nanmax(n_t)) if len(n_t) else 1.0)
            axb.bar(TG, n_t, width=0.9 * (TG[1] - TG[0]), color=KCLS[_cl]["color"], alpha=0.10, lw=0, zorder=1)
            ax.fill_between(TG, lo, hi, color=KCLS[_cl]["color"], alpha=0.20, lw=0, zorder=3)
            ax.plot(TG, cen, KCLS[_cl]["ls"], color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], zorder=5,
                    label=f"{KCLS[_cl]['short']}  ({len(_ks)})")
            for _x, _m, _l, _h, _n in zip(TG, cen, lo, hi, n_t):
                ROW9.append(dict(block="clock", quantity=_c9, kcls=_cl, x=float(_x), n_gal=int(_n),
                                 med=float(_m), lo=float(_l), hi=float(_h)))
        axb.set_ylim(0, _top * 3.0)
        ax.axvline(0, color="k", lw=1.4, ls="--", alpha=0.9, zorder=1)
        ax.annotate("SFT", xy=(0, 1.0), xycoords=("data", "axes fraction"), xytext=(3, -12), textcoords="offset points",
                    fontsize=8.5, color="0.3")
        if _c9 == "xrf":
            ax.set_ylim(0, 1.02)
        if _c9 == "lfd":
            ax.axhline(DUSTY_CUT, color="0.35", lw=1.1, ls=":", zorder=1)
            ax.annotate("the dust label", xy=(0.99, DUSTY_CUT), xycoords=("axes fraction", "data"), xytext=(0, 3),
                        textcoords="offset points", fontsize=7, color="0.35", ha="right", va="bottom")
        ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(_ylab); ax.set_title(_ttl, loc="left")
        ax.grid(alpha=0.22, lw=0.5)
        if _pi // 2 == 1:
            ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
        if _pi == 0:
            ax.legend(loc="upper left", frameon=False, fontsize=8.5)
    fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the quenching clock of the two classes: the X-ray mode, and what the dust is "
                 f"divided by", y=0.99, fontsize=12, fontweight="bold")
    fig.tight_layout(rect=(0, 0, 1, 0.965))
    paper_save(fig, "kappa_m25_clocks"); plt.show()

    # ── what the clocks say, at the epochs the eye reads them at ──
    print(f"\nthe clocks at {', '.join(f'{m:+g}' for m in P9_MARKS)} Gyr from the truncation "
          f"(median [16-84 of the bootstrap], N; AUC = P(rotating > the rest) at that epoch):")
    for _c9, _, _ylab in P9_PANELS:
        print(f"  {_c9}:")
        for _m9 in P9_MARKS:
            _i9 = int(np.argmin(np.abs(TG - _m9)))
            _txt, _v = [], {}
            for _cl in KC:
                _ks = [k for k in KEEP9 if CLS9[k] == _cl]
                _col = np.array([_TRK9[_c9][k][_i9] for k in _ks], float)
                _v[_cl] = _col[np.isfinite(_col)]
                _txt.append(f"{KCLS[_cl]['short']} " + (f"{np.median(_v[_cl]):+.2f} (N={len(_v[_cl])})"
                                                        if len(_v[_cl]) >= P9_NMIN else f"— (N={len(_v[_cl])})"))
            _all = np.concatenate([_v["rot"], _v["iso"]]) if len(_v["rot"]) and len(_v["iso"]) else np.array([])
            _auc = mw_auc(_all, np.concatenate([np.ones(len(_v["rot"]), bool), np.zeros(len(_v["iso"]), bool)]))[0] \
                if len(_all) else np.nan
            print(f"    {_m9:+.1f} Gyr:  " + ";  ".join(_txt) + (f";  AUC {_auc:.2f}" if np.isfinite(_auc) else ""))
            ROW9.append(dict(block="epoch", quantity=_c9, kcls="both", x=float(_m9),
                             n_gal=len(_v["rot"]) + len(_v["iso"]),
                             med=float(np.median(_v["rot"])) if len(_v["rot"]) else np.nan,
                             lo=float(np.median(_v["iso"])) if len(_v["iso"]) else np.nan, hi=float(_auc)))
    print("\nthe cost of not flooring the ratios (branch steps with the galaxy alive but the ratio undefined):")
    for _c9, _, _ in P9_PANELS:
        _a, _b = _drop9[_c9]
        print(f"  {_c9}: {_a}/{_b} ({100 * _a / max(_b, 1):.1f} %)"
              + ("  — no dust, or no H2, left at that step" if _c9 in ("ldh2", "ldtm", "lfd") else ""))
    pd.DataFrame(ROW9).to_csv(os.path.join(OUT, "kappa_m25_clocks.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_m25_clocks.csv')}")
```

## Part 10 — where the dust is, and how it moves: the ISM and the CGM, from the particles

Every part above measures the dust as a catalogue number: $M_{\rm dust}$ of the caesar galaxy, i.e. of its member list. That
number cannot tell **destruction** from **transport**, and once a grain leaves the member list it cannot see it at all. The
reduced particle files can: they hold every gas and star particle within $100$ physical kpc of the centre — the ISM *and* the
CGM — with the membership flag, the dust mass and the peculiar velocity of each one.

**The partition.** Three zones, every particle in exactly one:

* **core** — the inner half of the member gas ($r < R_{50}^{\rm gas}$);
* **outskirt** — the outer half ($r \geq R_{50}^{\rm gas}$);
* **CGM** — gas inside the aperture that is **not** on the member list.

The ISM boundary is measured **both** ways and cached, so `P10_SPLIT` flips between the half-mass radius (the default) and a
fixed `P10_RCORE` = 3 kpc (Parts 6–7's `ap3kpc`) without re-reading a file. The default is the scaled one because these
galaxies are compact — the member gas half-mass radius is ~1.7 kpc around the truncation — so a fixed 3 kpc boundary puts
almost the whole ISM in the "core" and leaves the outskirt empty in a fifth of the nodes. The cell prints both numbers.

**The velocity** is radial, in the galaxy's own rest frame: $v_r = (\mathbf{v}_{\rm pec} - \mathbf{v}_{\rm sys})\cdot\hat r +
H(z)\,r$, positive outward. $\mathbf{v}_{\rm sys}$ is the mass-weighted velocity of the **member stars** — the gas is what may
be flowing, so it cannot define the frame — and the Hubble term restores the expansion the peculiar velocities have had taken
out (`P10_HUBFLOW`; it is $\sim 10$ km s⁻¹ at 100 kpc).

**Everything is measured twice over the same particles**: weighted by $m_{\rm dust}$ — the dust-hosting gas, which is what a
sub-mm map traces — and by $m_{\rm gas}$ — every particle in the zone. Fig 2 against Fig 3 *is* the question "is the dense,
dusty gas an exception, or does it just do what all the gas does?".

| figure | what it shows |
|---|---|
| **1 the budget** | the share of the dust that is already outside the galaxy, with the same share of the gas as the control, and the three zones' dust masses |
| **2 the dust flow** | $v_r$ (dust-weighted) and the dust fraction moving out faster than `P10_VFAST`, per zone, per class |
| **3 the gas flow** | the same figure for every gas particle in the same zones |
| **4 the flux** | $\dot M^{\rm out}$ across a fixed spherical shell for the dust and the gas, and the **dust-to-gas ratio of what leaves** against the ratio of the ISM it comes from |

The clock is **binned, not interpolated**: a node exists only where a snapshot of the branch happens to have a particle file,
so each galaxy contributes its own median inside each bin (a galaxy with many snapshots in a bin does not outvote one with a
single snapshot) and the class median is taken over galaxies. The census is **cached** (`P10_CACHE`) — the first run reads a
few GiB and the second is instant; delete the file to rebuild it.

```python
# ── Part 10 — WHERE THE DUST IS AND HOW IT MOVES: the m25 particle clock, in the ISM and in the CGM ──
# Everything above this line measures the dust as a catalogue number — M_dust of the caesar galaxy, i.e. of the member list.
# That number cannot say whether the dust that disappears after the truncation is DESTROYED or CARRIED OUT, and it cannot see
# the grains once they leave the member list at all. The reduced particle files can: they hold every gas and star particle
# within P10_RMAX physical kpc of the galaxy centre — the ISM *and* the CGM — with the caesar membership flag, the dust mass
# and the peculiar velocity of each one (build_reduced_particles_job.py).
# THE PARTITION, three zones, one particle in exactly one of them:
#     core      the INNER half of the member gas           (the ISM where the H2 and the star formation are)
#     outskirt  the OUTER half of it                       (the rest of the ISM)
#     CGM       gas inside the aperture that is NOT on the member list  (the halo gas caesar does not call the galaxy)
# The ISM boundary is drawn TWICE and both are measured, so P10_SPLIT switches between them with no re-read:
#   "r50"  (the default) at the member gas half-mass radius — each galaxy's own size. These galaxies are compact (R50 of the
#          member gas is ~1.7 kpc around the truncation in m25), so a fixed 3 kpc boundary puts almost the whole ISM in the
#          "core" and leaves the outskirt empty in a fifth of the nodes; the cell prints both numbers so this can be checked.
#   "kpc"  at P10_RCORE, the fixed aperture Parts 6-7 use (ap3kpc) — comparable across galaxies in physical kpc, which is what
#          a flow measurement wants, at the cost of meaning something different for a large and a small galaxy.
# THE VELOCITY is the radial one in the galaxy's own rest frame: v_r = (v_pec - v_sys) . r_hat + H(z) r, with v_sys the
# mass-weighted velocity of the MEMBER STARS (the gas is what may be flowing, so it cannot define the frame) and the Hubble
# term restoring the expansion the peculiar velocities have had taken out (P10_HUBFLOW; it is ~10 km/s at 100 kpc).
# Positive is OUTWARD. Every statistic is MASS-WEIGHTED, and computed TWICE over the same particles: once weighted by
# m_dust — the dust-hosting gas, which is what a sub-mm map traces — and once by m_gas — every particle in the zone. The two
# together are the question "is the dense, dusty gas an exception, or does it just do what all the gas does?".
#   FIG 1  THE BUDGET     what share of the dust is outside the galaxy, on the clock, with the same share of the gas as the
#                         control, and the three zones' dust masses themselves.
#   FIG 2  THE DUST FLOW  v_r (dust-weighted) and the dust fraction moving out faster than P10_VFAST, per zone, per class.
#   FIG 3  THE GAS FLOW   the same figure for every gas particle in the same zones — Fig 2's control.
#   FIG 4  THE FLUX       the mass crossing fixed spherical shells: Mdot_out for the dust and for the gas, and the
#                         DUST-TO-GAS RATIO OF THE OUTFLOW against the ratio of the material it comes from — is what leaves
#                         dust-enriched, dust-poor, or the same stuff?
# NOT a pure read of the ism_prediction products: this reads the m25 reduced particle files, and it CACHES what it measures
# (P10_CACHE) — the second run is instant, and deleting the cache rebuilds it. Nodes are looked up through Part 9's branch
# index, so the clock is the branch's own, not the K-S products' handful of critical epochs.
P10_DIR    = os.path.join(os.getcwd(), "output", P9_BOX, "reduced_particles")
P10_PREFIX = {"cis25": "m25n512", "cis50": "m50n512", "cis50nox": "m50n512", "cis100": "m100n1024"}.get(P9_BOX, P9_BOX)
P10_CACHE  = os.path.join(OUT, f"kappa_{P9_BOX}_particle_zones.fits")     # the census cache (delete to rebuild; it is appended to)
P10_SPLIT  = "r50"         # where the ISM is cut in two: "r50" = at the member gas half-mass radius | "kpc" = at P10_RCORE.
                            # BOTH are measured and cached, so this can be flipped without re-reading a single file
P10_RCORE  = 3.0           # the fixed core / outskirt boundary [physical kpc] used by P10_SPLIT = "kpc" — Parts 6-7's ap3kpc
P10_CGM_RMIN = 0.0         # the CGM's inner edge [kpc]: 0 = pure membership (every non-member particle in the aperture, which is
                            # what "not the galaxy" means operationally). Raise it to exclude the hot gas that sits inside the disc
P10_RMAX   = 100.0         # the aperture the files were built with [kpc] — do not raise it above build_reduced_particles' REDUCED_RMAX_KPC
P10_VFAST  = 100.0         # "moving out fast" [km/s]: the threshold of the second row of Figs 2-3
P10_SHELLS = [(10.0, 2.0), (30.0, 5.0), (60.0, 10.0)]     # (radius, half-width) of Fig 4's spherical shells [kpc]
P10_HUBFLOW = True         # add H(z) r to the radial velocity (the peculiar velocities carry no expansion)
P10_OM     = 0.3           # SIMBA's Omega_m, for H(z) = 100 h sqrt(Om (1+z)^3 + 1 - Om); h comes from the file itself
P10_TB     = np.arange(-3.0, 4.001, 0.5)      # THE CLOCK of this part: bin EDGES of t - t_SFT [Gyr]. The nodes are the snapshots
                            # that happen to have a particle file, so the clock is binned, not interpolated
P10_NMIN   = 3             # galaxies a drawn bin needs
P10_NBOOT  = 300
P10_MAXNODE = None         # a cap on the number of files the census reads (None = all) — set it to smoke-test the cell
P10_ZONES  = ([("core_h", "core"), ("outskirt_h", "outskirt"), ("cgm", "cgm")] if P10_SPLIT == "r50"
              else [("core", "core"), ("outskirt", "outskirt"), ("cgm", "cgm")])   # (the census column suffix, the name)
KMS_KPC_PER_YR = 1.0227121650537077e-9        # (Msun km/s / kpc) -> Msun/yr

if "BR9" not in globals():
    print("Part 10: needs Part 9 — run it first (it builds the branch index this part looks the particle files up through).")
elif not os.path.isdir(P10_DIR):
    print(f"Part 10: {P10_DIR} is missing — build_reduced_particles_job.py writes it (cluster).")
else:
    import glob as _glob10, re as _re10, time as _time10

    # ── (A) THE CENSUS: one row per (snap, gx) that is on a retained branch and has a file ──
    _inv10 = {}
    for _p10 in _glob10.glob(os.path.join(P10_DIR, "snap_*", f"{P10_PREFIX}_snap*_gal*.h5")):
        _m10 = _re10.search(r"_snap(\d+)_gal(\d+)\.h5$", _p10)
        if _m10:
            _inv10[(int(_m10.group(1)), int(_m10.group(2)))] = _p10
    _keys10 = set(KEEP9)
    NOD10 = BR9[[(int(a_), int(g_)) in _keys10 for a_, g_ in zip(BR9["anchor_snap"], BR9["gal_id"])]].copy()
    NOD10 = NOD10[[(int(s_), int(x_)) in _inv10 for s_, x_ in zip(NOD10["snap"], NOD10["gx"])]].reset_index(drop=True)
    _uniq10 = sorted({(int(s_), int(x_)) for s_, x_ in zip(NOD10["snap"], NOD10["gx"])})
    print(f"Part 10 — {len(_inv10)} reduced particle files in {P10_DIR}; {len(NOD10)} of them are nodes of the "
          f"{len(_keys10)} retained {P9_BOX} branches ({len(_uniq10)} distinct files, median "
          f"{int(NOD10.groupby(['anchor_snap', 'gal_id']).size().median()) if len(NOD10) else 0} per galaxy)")

    def _wst10(v, w, vfast):
        """Mass-weighted statistics of v under the weights w: the weighted 10/50/90 quantiles, the weighted mean, the
        weighted scatter about it, and the share of the WEIGHT (not of the particles) moving out faster than vfast."""
        ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
        if not ok.any():
            return dict(n=0, q10=np.nan, q50=np.nan, q90=np.nan, mean=np.nan, sig=np.nan, ffast=np.nan)
        v, w = v[ok], w[ok]
        o = np.argsort(v); v, w = v[o], w[o]
        c = (np.cumsum(w) - 0.5 * w) / w.sum()                       # the weighted plotting positions
        m = float(np.average(v, weights=w))
        return dict(n=int(len(v)), q10=float(np.interp(0.10, c, v)), q50=float(np.interp(0.50, c, v)),
                    q90=float(np.interp(0.90, c, v)), mean=m,
                    sig=float(np.sqrt(np.average((v - m) ** 2, weights=w))),
                    ffast=float(w[v > vfast].sum() / w.sum()))

    def _census10(path):
        """ONE reduced particle file -> the zone census: the dust and gas masses of the three zones, the mass-weighted radial
        kinematics of each under both weights, the two half-mass radii, and the shell fluxes."""
        with h5py.File(path, "r") as h:
            A = dict(h.attrs); G, St = h["gas"], h["star"]
            gp = G["pos"][:].astype(float); gmem = G["member"][:].astype(bool)
            gm = G["m_gas"][:].astype(float); gd = G["m_dust"][:].astype(float)
            gh2 = G["m_H2"][:].astype(float) if "m_H2" in G else np.full(len(gm), np.nan)
            gT = G["temp"][:].astype(float) if "temp" in G else np.full(len(gm), np.nan)
            gv = G["vel"][:].astype(float) if "vel" in G else np.full((len(gm), 3), np.nan)
            sm = St["m_star"][:].astype(float); smem = St["member"][:].astype(bool)
            sp = St["pos"][:].astype(float)
            sv = St["vel"][:].astype(float) if "vel" in St else np.full((len(sm), 3), np.nan)
        z, hub = float(A["redshift"]), float(A["hub"])
        r, rs = np.linalg.norm(gp, axis=1), np.linalg.norm(sp, axis=1)
        ref = smem & np.isfinite(sv).all(1) & (sm > 0)               # THE REST FRAME: the member stars
        if ref.sum() >= 10:
            vsys = np.average(sv[ref], axis=0, weights=sm[ref])
        else:
            ref = gmem & np.isfinite(gv).all(1) & (gm > 0)
            vsys = np.average(gv[ref], axis=0, weights=gm[ref]) if ref.sum() >= 10 else np.full(3, np.nan)
        vr = ((gv - vsys) * (gp / np.clip(r, 1e-9, None)[:, None])).sum(1)
        if P10_HUBFLOW:
            vr = vr + (100.0 * hub * np.sqrt(P10_OM * (1 + z) ** 3 + 1 - P10_OM)) * r / 1e3
        _rh = np.nan                                        # the member gas half-mass radius: the galaxy's own ISM scale
        if gmem.sum() >= 5 and np.nansum(gm[gmem]) > 0:
            _o = np.argsort(r[gmem])
            _rh = float(np.interp(0.5, np.cumsum(gm[gmem][_o]) / np.nansum(gm[gmem]), r[gmem][_o]))
        Z = {"core": gmem & (r < P10_RCORE), "outskirt": gmem & (r >= P10_RCORE),
             "core_h": gmem & (r < _rh), "outskirt_h": gmem & (r >= _rh),
             "cgm": (~gmem) & (r >= P10_CGM_RMIN) & (r <= P10_RMAX)}
        row = dict(snap=int(A["snap"]), gx=int(A["gx"]), z=z, has_vel=bool(np.isfinite(vsys).all()),
                   n_gas_file=int(len(gm)), n_star_file=int(len(sm)), v_sys=float(np.linalg.norm(vsys)),
                   m_star_mem=float(np.nansum(sm[smem])),
                   mdust_ism=float(np.nansum(gd[gmem])), mdust_cgm_all=float(np.nansum(gd[~gmem])),
                   mgas_ism=float(np.nansum(gm[gmem])), mgas_cgm_all=float(np.nansum(gm[~gmem])),
                   mh2_ism=float(np.nansum(gh2[gmem])))
        row["r50_gasmem"] = _rh
        for _nm, _msk, _w, _rr in (("star", smem, sm, rs), ("H2", gmem & (gh2 > 0), gh2, r)):
            _tot = float(np.nansum(_w[_msk])) if _msk.any() else 0.0
            if _msk.sum() >= 5 and _tot > 0:
                _o = np.argsort(_rr[_msk])
                row[f"r50_{_nm}"] = float(np.interp(0.5, np.cumsum(_w[_msk][_o]) / _tot, _rr[_msk][_o]))
            else:
                row[f"r50_{_nm}"] = np.nan
        for _zn, _msk in Z.items():
            row[f"n_{_zn}"] = int(_msk.sum())
            row[f"mdust_{_zn}"] = float(np.nansum(gd[_msk])); row[f"mgas_{_zn}"] = float(np.nansum(gm[_msk]))
            row[f"mh2_{_zn}"] = float(np.nansum(gh2[_msk]))
            row[f"T_{_zn}"] = _wst10(gT[_msk], gm[_msk], np.inf)["q50"]
            for _wn, _wv in (("d", gd), ("g", gm)):
                _st = _wst10(vr[_msk], _wv[_msk], P10_VFAST)
                for _q in ("q10", "q50", "q90", "mean", "sig", "ffast", "n"):
                    row[f"{_q}_{_wn}_{_zn}"] = _st[_q]
        for _R, _dR in P10_SHELLS:
            _sh = np.abs(r - _R) < _dR
            row[f"n_shell{int(_R)}"] = int(_sh.sum())
            for _wn, _wv in (("d", gd), ("g", gm)):
                _w = np.where(_sh & np.isfinite(vr) & np.isfinite(_wv), _wv, 0.0)
                row[f"mdot_out_{_wn}{int(_R)}"] = float(np.sum(_w * np.clip(vr, 0, None)) / (2 * _dR) * KMS_KPC_PER_YR)
                row[f"mdot_in_{_wn}{int(_R)}"] = float(np.sum(_w * np.clip(vr, None, 0)) / (2 * _dR) * KMS_KPC_PER_YR)
        return row

    Z10 = pd.DataFrame()
    if os.path.exists(P10_CACHE):
        Z10 = Table.read(P10_CACHE).to_pandas()
        print(f"  the census, cached -> {P10_CACHE}  ({len(Z10)} files already measured)")
    _done10 = set(zip(Z10["snap"].astype(int), Z10["gx"].astype(int))) if len(Z10) else set()
    _todo10 = [k for k in _uniq10 if k not in _done10]
    if P10_MAXNODE is not None:
        _todo10 = _todo10[:int(P10_MAXNODE)]
    if _todo10:
        print(f"  measuring {len(_todo10)} files (~{sum(os.path.getsize(_inv10[k]) for k in _todo10) / 2 ** 30:.1f} GiB "
              "to read; the result is cached)", flush=True)
        _new10, _t010, _bad10 = [], _time10.time(), []
        for _i10, _k10 in enumerate(_todo10):
            try:
                _new10.append(_census10(_inv10[_k10]))
            except Exception as _e10:
                _bad10.append(f"{os.path.basename(_inv10[_k10])}: {type(_e10).__name__}: {_e10}")
            if (_i10 + 1) % 200 == 0:
                _el = _time10.time() - _t010
                print(f"     {_i10 + 1}/{len(_todo10)} files, {_el:.0f} s "
                      f"(~{_el / (_i10 + 1) * (len(_todo10) - _i10 - 1):.0f} s left)", flush=True)
        for _b10 in _bad10[:5]:
            print(f"     file not read, {_b10}")
        Z10 = pd.concat([Z10, pd.DataFrame(_new10)], ignore_index=True) if len(_new10) else Z10
        if len(Z10):
            Table.from_pandas(Z10).write(P10_CACHE, overwrite=True)
            print(f"  the census -> {P10_CACHE}  ({len(Z10)} files, {_time10.time() - _t010:.0f} s)")
    if not len(Z10):
        print("  nothing measured — Part 10 stops here.")
    else:
        _nov10 = int((~Z10["has_vel"].astype(bool)).sum())
        if _nov10:
            print(f"  {_nov10} of {len(Z10)} files carry no velocities (built before build_reduced_particles' 'vel' field) — "
                  "their kinematic columns are NaN; re-submit that plan's sbatch to backfill them")

        # ── (B) the census on the clock: one value per galaxy per bin, then the class median ──
        D10 = NOD10.merge(Z10, on=["snap", "gx"], how="inner")
        D10["kcls"] = [CLS9.get((int(a_), int(g_))) for a_, g_ in zip(D10["anchor_snap"], D10["gal_id"])]
        D10["t_sft"] = [_tsft9.get((int(a_), int(g_)), np.nan) for a_, g_ in zip(D10["anchor_snap"], D10["gal_id"])]
        D10["dt"] = D10["t"].to_numpy(float) - D10["t_sft"].to_numpy(float)
        with np.errstate(all="ignore"):
            _dt10 = D10["mdust_ism"].to_numpy(float) + D10["mdust_cgm_all"].to_numpy(float)
            _gt10 = D10["mgas_ism"].to_numpy(float) + D10["mgas_cgm_all"].to_numpy(float)
            D10["f_dust_cgm"] = np.where(_dt10 > 0, D10["mdust_cgm_all"].to_numpy(float) / _dt10, np.nan)
            D10["f_gas_cgm"] = np.where(_gt10 > 0, D10["mgas_cgm_all"].to_numpy(float) / _gt10, np.nan)
            for _zn, _ in P10_ZONES:                       # the zone dust masses, and each zone's own dust-to-gas ratio
                _md, _mg = D10[f"mdust_{_zn}"].to_numpy(float), D10[f"mgas_{_zn}"].to_numpy(float)
                D10[f"lmdust_{_zn}"] = np.log10(np.where(_md > 0, _md, np.nan))
                D10[f"ldg_{_zn}"] = np.log10(np.where((_md > 0) & (_mg > 0), _md / np.where(_mg > 0, _mg, np.nan), np.nan))
            for _R, _ in P10_SHELLS:
                _od, _og = D10[f"mdot_out_d{int(_R)}"].to_numpy(float), D10[f"mdot_out_g{int(_R)}"].to_numpy(float)
                D10[f"lmdot_d{int(_R)}"] = np.log10(np.where(_od > 0, _od, np.nan))
                D10[f"lmdot_g{int(_R)}"] = np.log10(np.where(_og > 0, _og, np.nan))
                D10[f"ldg_out{int(_R)}"] = np.log10(np.where((_od > 0) & (_og > 0), _od / np.where(_og > 0, _og, np.nan), np.nan))
        D10 = D10[np.isfinite(D10["dt"].to_numpy(float))].reset_index(drop=True)
        _XB10 = 0.5 * (P10_TB[1:] + P10_TB[:-1])
        D10["tbin"] = np.digitize(D10["dt"].to_numpy(float), P10_TB) - 1
        print(f"  the clock: {len(D10)} galaxy-nodes on t - t_SFT in [{P10_TB[0]:g}, {P10_TB[-1]:g}] Gyr, "
              + ", ".join(f"{KCLS[c]['short']} {int((D10['kcls'] == c).sum())} nodes / "
                          f"{D10[D10['kcls'] == c].groupby(['anchor_snap', 'gal_id']).ngroups} galaxies" for c in KC))
        _rh10 = D10["r50_gasmem"].to_numpy(float)
        print(f"  the ISM boundary (P10_SPLIT = '{P10_SPLIT}'): the member gas half-mass radius is "
              f"{np.nanmedian(_rh10):.2f} kpc in the median and "
              f"{np.nanmedian(D10[np.abs(D10['dt'].to_numpy(float)) < 0.5]['r50_gasmem'].to_numpy(float)):.2f} kpc around the "
              f"truncation;  at the fixed {P10_RCORE:g} kpc the outskirt would be EMPTY in "
              f"{100 * np.mean(D10['n_outskirt'].to_numpy(float) == 0):.0f} % of the nodes, at the half-mass radius in "
              f"{100 * np.mean(D10['n_outskirt_h'].to_numpy(float) == 0):.0f} %")
        ROW10 = []

        def _track10(col, cls, agg="median"):
            """One column -> (centre, lo, hi, n_gal) per clock bin for one class: the galaxy's own median inside the bin
            first, so a galaxy with many snapshots in a bin does not outvote one with a single snapshot."""
            cen = np.full(len(_XB10), np.nan); lo = cen.copy(); hi = cen.copy(); n = np.zeros(len(_XB10), int)
            sub = D10[(D10["kcls"] == cls) & np.isfinite(D10[col].to_numpy(float))]
            rng = np.random.default_rng(31)
            for _b in range(len(_XB10)):
                _s = sub[sub["tbin"] == _b]
                if not len(_s):
                    continue
                _v = _s.groupby(["anchor_snap", "gal_id"])[col].median().to_numpy(float)
                _v = _v[np.isfinite(_v)]
                n[_b] = len(_v)
                if len(_v) < P10_NMIN:
                    continue
                _f = np.median if agg == "median" else np.mean
                cen[_b] = _f(_v)
                _bs = _f(_v[rng.integers(0, len(_v), (P10_NBOOT, len(_v)))], axis=1)
                lo[_b], hi[_b] = np.percentile(_bs, 16), np.percentile(_bs, 84)
            return cen, lo, hi, n

        def _panel10(ax, col, ylab, title, agg="median", block="", pct=False):
            for _cl in KC:
                cen, lo, hi, n = _track10(col, _cl, agg)
                _ok = np.isfinite(cen)
                if not _ok.any():
                    continue
                ax.fill_between(_XB10[_ok], lo[_ok], hi[_ok], color=KCLS[_cl]["color"], alpha=0.20, lw=0)
                ax.plot(_XB10[_ok], cen[_ok], "o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=4,
                        mec="k", mew=0.4, label=f"{KCLS[_cl]['short']}  ({int(np.nanmax(n))} gal max)")
                for _x, _c, _l, _h, _n in zip(_XB10, cen, lo, hi, n):
                    ROW10.append(dict(block=block, quantity=col, kcls=_cl, x=float(_x), n_gal=int(_n),
                                      med=float(_c), lo=float(_l), hi=float(_h)))
            ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9)
            ax.set_xlim(P10_TB[0], P10_TB[-1]); ax.set_ylabel(ylab); ax.set_title(title, loc="left", fontsize=10)
            ax.grid(alpha=0.22, lw=0.5)
            if pct:
                ax.set_ylim(0, 1.02)

        _ZL10 = {"core": rf"core: member gas, $r < {P10_RCORE:g}$ kpc",
                 "outskirt": rf"outskirt: member gas, $r \geq {P10_RCORE:g}$ kpc",
                 "core_h": r"core: the inner half of the member gas  ($r < R_{50}^{\rm gas}$)",
                 "outskirt_h": r"outskirt: the outer half  ($r \geq R_{50}^{\rm gas}$)",
                 "cgm": rf"CGM: non-member gas, $r \leq {P10_RMAX:g}$ kpc"}

        # ── FIG 1: the budget — how much of the dust is already outside the galaxy ──
        fig, axs = plt.subplots(1, 3, figsize=(13.4, 3.9))
        _panel10(axs[0], "f_dust_cgm", r"$M_{\rm dust}^{\rm CGM} / M_{\rm dust}^{\rm total}$",
                 "(a) the share of the dust that is NOT in the galaxy", block="budget", pct=True)
        _panel10(axs[1], "f_gas_cgm", r"$M_{\rm gas}^{\rm CGM} / M_{\rm gas}^{\rm total}$",
                 "(b) the same share of the GAS — the control", block="budget", pct=True)
        for _zn, _ in P10_ZONES:
            for _cl in KC:
                cen, lo, hi, n = _track10(f"lmdust_{_zn}", _cl)
                _ok = np.isfinite(cen)
                if _ok.any():
                    axs[2].plot(_XB10[_ok], cen[_ok], dict(zip([z for z, _ in P10_ZONES], ("-", "--", ":")))[_zn],
                                color=KCLS[_cl]["color"], lw=2.0)
                for _x, _c, _l, _h, _n in zip(_XB10, cen, lo, hi, n):
                    ROW10.append(dict(block="budget", quantity=f"lmdust_{_zn}", kcls=_cl, x=float(_x), n_gal=int(_n),
                                      med=float(_c), lo=float(_l), hi=float(_h)))
        axs[2].axvline(0, color="k", lw=1.3, ls="--", alpha=0.9); axs[2].grid(alpha=0.22, lw=0.5)
        axs[2].set_xlim(P10_TB[0], P10_TB[-1]); axs[2].set_ylabel(r"$\log M_{\rm dust}$  [$M_\odot$]")
        axs[2].set_title("(c) the dust in each zone", loc="left", fontsize=10)
        axs[2].legend(handles=[Line2D([], [], color="0.3", ls=l_, lw=2.0, label=_ZL10[z_])
                               for z_, l_ in zip([z for z, _ in P10_ZONES], ("-", "--", ":"))]
                              + [Line2D([], [], color=KCLS[c]["color"], lw=2.4, label=KCLS[c]["short"]) for c in KC],
                      loc="best", frameon=True, framealpha=0.75, edgecolor="none", fontsize=7.5)
        for ax in axs:
            ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
        axs[0].legend(loc="upper left", frameon=False, fontsize=8)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the dust budget inside {P10_RMAX:g} kpc: the ISM, and the CGM",
                     y=1.02, fontsize=12, fontweight="bold")
        fig.tight_layout(); paper_save(fig, "kappa_m25_dust_budget"); plt.show()

        # ── FIGS 2-3: how the dust moves, and how all the gas moves ──
        for _wn, _wlab, _stem in (("d", "dust-mass-weighted", "kappa_m25_flow_dust"),
                                  ("g", "gas-mass-weighted (every particle in the zone)", "kappa_m25_flow_gas")):
            fig, axs = plt.subplots(2, 3, figsize=(13.4, 7.2), sharex=True)
            for _zi, (_zn, _) in enumerate(P10_ZONES):
                _panel10(axs[0, _zi], f"q50_{_wn}_{_zn}", r"$v_r$  [km s$^{-1}$]",
                         f"({'abc'[_zi]}) {_ZL10[_zn]}", block=f"flow_{_wn}")
                axs[0, _zi].axhline(0, color="0.4", lw=0.9, ls=":")
                _panel10(axs[1, _zi], f"ffast_{_wn}_{_zn}",
                         rf"mass fraction with $v_r > {P10_VFAST:g}$ km s$^{{-1}}$", "", block=f"flow_{_wn}", pct=True)
                axs[1, _zi].set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
                for _r in (0, 1):
                    if _zi:
                        axs[_r, _zi].set_ylabel("")
            axs[0, 0].legend(loc="upper left", frameon=False, fontsize=8)
            axs[0, 2].annotate("outward", xy=(0.985, 0.97), xycoords="axes fraction", fontsize=7.5, color="0.35",
                               ha="right", va="top")
            axs[0, 2].annotate("inward", xy=(0.985, 0.03), xycoords="axes fraction", fontsize=7.5, color="0.35",
                               ha="right", va="bottom")
            fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the radial flow on the quenching clock, {_wlab}: "
                         "the net motion (top) and the fast outward tail (bottom)", y=0.995, fontsize=12, fontweight="bold")
            fig.tight_layout(rect=(0, 0, 1, 0.97)); paper_save(fig, _stem); plt.show()

        # ── FIG 4: the flux through fixed shells, and whether what leaves is dust-enriched ──
        fig, axs = plt.subplots(1, 3, figsize=(13.4, 3.9))
        _R0 = int(P10_SHELLS[min(1, len(P10_SHELLS) - 1)][0])          # the middle shell is the one drawn
        _panel10(axs[0], f"lmdot_d{_R0}", r"$\log \dot M_{\rm dust}^{\rm out}$  [$M_\odot$ yr$^{-1}$]",
                 rf"(a) the dust crossing $r = {_R0}$ kpc outward", block="flux")
        _panel10(axs[1], f"lmdot_g{_R0}", r"$\log \dot M_{\rm gas}^{\rm out}$  [$M_\odot$ yr$^{-1}$]",
                 rf"(b) the gas crossing $r = {_R0}$ kpc outward", block="flux")
        _panel10(axs[2], f"ldg_out{_R0}", r"$\log \dot M_{\rm dust}^{\rm out} / \dot M_{\rm gas}^{\rm out}$",
                 "(c) the dust-to-gas ratio of what leaves", block="flux")
        for _cl in KC:                                                  # the material it comes from, as the reference
            cen, lo, hi, n = _track10(f"ldg_{P10_ZONES[1][0]}", _cl)
            _ok = np.isfinite(cen)
            if _ok.any():
                axs[2].plot(_XB10[_ok], cen[_ok], ls="--", color=KCLS[_cl]["color"], lw=1.4, alpha=0.8)
            for _x, _c, _l, _h, _n in zip(_XB10, cen, lo, hi, n):
                ROW10.append(dict(block="flux", quantity=f"ldg_{P10_ZONES[1][0]}", kcls=_cl, x=float(_x), n_gal=int(_n),
                                  med=float(_c), lo=float(_l), hi=float(_h)))
        axs[2].legend(handles=[Line2D([], [], color="0.3", lw=2.0, label="the outflow"),
                               Line2D([], [], color="0.3", lw=1.4, ls="--", label="the outskirt ISM it comes from")],
                      loc="lower left", frameon=False, fontsize=7.5)
        for ax in axs:
            ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
        axs[0].legend(loc="lower left", frameon=False, fontsize=8)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the flux through the $r = {_R0}$ kpc shell "
                     f"(half-width {P10_SHELLS[min(1, len(P10_SHELLS) - 1)][1]:g} kpc)",
                     y=1.02, fontsize=12, fontweight="bold")
        fig.tight_layout(); paper_save(fig, "kappa_m25_flux_shell"); plt.show()

        # ── what it says ──
        _MK10 = [m for m in (-1.0, 0.0, 1.0, 2.0) if P10_TB[0] <= m <= P10_TB[-1]]
        print(f"\nthe zones at {', '.join(f'{m:+g}' for m in _MK10)} Gyr from the truncation "
              "(the class median of the per-galaxy medians; AUC = P(rotating > the rest)):")
        _QT10 = ([("f_dust_cgm", "the CGM share of the dust"), ("f_gas_cgm", "the CGM share of the gas")]
                 + [(f"q50_{w}_{z}", f"v_r [km/s], {w == 'd' and 'dust' or 'gas'}-weighted, {n}")
                    for w in ("d", "g") for z, n in P10_ZONES]
                 + [(f"ffast_{w}_{z}", f"mass fraction v_r > {P10_VFAST:g} km/s, {w == 'd' and 'dust' or 'gas'}-weighted, {n}")
                    for w in ("d", "g") for z, n in P10_ZONES]
                 + [(f"lmdot_d{_R0}", f"log Mdot_dust,out at {_R0} kpc"), (f"ldg_out{_R0}", "log D/G of the outflow")])
        for _col, _nm in _QT10:
            if _col not in D10.columns:
                continue
            _line = []
            for _m10 in _MK10:
                _b = int(np.clip(np.digitize([_m10], P10_TB)[0] - 1, 0, len(_XB10) - 1))
                _s = D10[(D10["tbin"] == _b) & np.isfinite(D10[_col].to_numpy(float))]
                _pg = _s.groupby(["anchor_snap", "gal_id", "kcls"])[_col].median().reset_index()
                _a = _pg[_pg["kcls"] == "rot"][_col].to_numpy(float); _b2 = _pg[_pg["kcls"] == "iso"][_col].to_numpy(float)
                _auc = mw_auc(np.concatenate([_a, _b2]),
                              np.concatenate([np.ones(len(_a), bool), np.zeros(len(_b2), bool)]))[0] if len(_a) and len(_b2) else np.nan
                _line.append(f"{_m10:+.0f}: " + (f"{np.median(_a):.3g}" if len(_a) >= P10_NMIN else "—")
                             + "/" + (f"{np.median(_b2):.3g}" if len(_b2) >= P10_NMIN else "—")
                             + (f" [{_auc:.2f}]" if np.isfinite(_auc) else "") + f" (N {len(_a)},{len(_b2)})")
                ROW10.append(dict(block="epoch", quantity=_col, kcls="both", x=float(_m10), n_gal=len(_a) + len(_b2),
                                  med=float(np.median(_a)) if len(_a) else np.nan,
                                  lo=float(np.median(_b2)) if len(_b2) else np.nan, hi=float(_auc)))
            print(f"  {_nm:58s} " + "   ".join(_line))
        print("  (each cell is  rotating/the rest [AUC] (N_rot, N_rest);  the classes are Part 2's, defined at the anchor)")
        pd.DataFrame(ROW10).to_csv(os.path.join(OUT, "kappa_m25_particle_clock.csv"), index=False)
        D10.to_csv(os.path.join(OUT, "kappa_m25_particle_nodes.csv"), index=False)
        print(f"\ntables -> {os.path.join(OUT, 'kappa_m25_particle_clock.csv')} (the drawn points), "
              f"{os.path.join(OUT, 'kappa_m25_particle_nodes.csv')} (every node)")
```

## Part 11 — were the anchor-rotators already denser in H₂ at the truncation?

The classes are defined **at the anchor**, which is the end of the story: Part 2 selects on $\kappa_{\rm rot}$(gas) measured
in the quenched galaxy we observe. Everything upstream is a question about the past, and this is its sharpest form — at
$t_{\rm SFT}$, long before the anchor, is the H₂ of the *future* rotators already denser?

* **If it is**, the selection picks galaxies that entered quenching with a different ISM, and the dust that survives in them
  is a property they brought in.
* **If it is not**, the two classes started from the same ISM and whatever separates them is made afterwards — which is what
  Parts 9 and 10 are then about.

The measurement is the **K-S abscissa**, the convention the observations are in:
$\Sigma_{\rm H_2} = \mathrm{He} \times 0.5\,M_{\rm H_2} / (\pi R_{50}({\rm H_2})^2)$ — half the molecular mass inside its own
half-mass radius, from the m25 K-S track products (particles, not catalogue numbers), at every critical point of the branch
so the answer can be read as a sequence and not only at one epoch. `P11_AP = "ap3kpc"` swaps it for the fixed core aperture.

**The mass and the size are reported separately**, because a density is a ratio: more H₂, and the same H₂ in a smaller radius,
are different physical statements.

**The stellar mass is controlled for.** Part 2's rule contains $\log M_\star > 10.5$, so any raw difference is partly the mass
axis restating itself. Every AUC is printed twice — raw, and pooled *within* terciles of $\log M_\star$ at the anchor
(`P11_MTERC`); the bracketed number is the one to quote.

Where Part 10's census exists, the same density is measured **independently from the member H₂ particles** at the branch node
nearest each critical point (within 0.25 Gyr) and drawn dashed in panel (a) — a check that the answer is not an artefact of
how the K-S products define their aperture.

```python
# ── Part 11 — WERE THE ANCHOR-ROTATORS ALREADY DENSER IN H2 AT THE TRUNCATION? ──
# The classes are defined at the ANCHOR, which is the end of the story: Part 2 selects on kappa_rot(gas) measured in the
# quenched galaxy we observe. Everything upstream of that is a question about the past, and this is the sharpest form of it —
# at t_SFT, when the star formation is cut off and long before the anchor, is the H2 of the future rotators ALREADY denser?
# If it is, the selection is picking galaxies that entered quenching with a different ISM, and the dust that survives in them
# is a property they brought in. If it is not, the two classes started from the same ISM and the difference is made after.
# THE MEASUREMENT is the K-S abscissa, the convention the observations are in: Sigma_H2 = He x 0.5 M_H2 / (pi R50(H2)^2),
# half the molecular mass inside its own half-mass radius, from the m25 K-S track products (particles, not catalogue numbers).
# It is taken at each CRITICAL POINT of the branch, so the question can be read as a sequence and not only at one epoch.
# THE SIZE AND THE MASS ARE SHOWN SEPARATELY, because a density is a ratio and the answer is different depending on which of
# the two moves: more H2, or the same H2 in a smaller radius, are different physical statements.
# THE MASS IS CONTROLLED FOR. Part 2's rule contains log M* > 10.5, so any raw difference is partly the mass axis restating
# itself; the AUC is therefore printed twice — raw, and pooled WITHIN terciles of log M* at the anchor, which is the number
# that says whether the H2 density is its own axis.
# A pure read of the K-S products, plus — where Part 10's census exists — the same density measured independently from the
# member H2 PARTICLES at the branch node nearest each critical point.
P11_KSDIR  = globals().get("P6_KSDIR_OVERRIDE", os.path.join(os.getcwd(), "output", "cis25", "ks_tracks"))
P11_STAGES = ["sf_peak", "sft", "qt", "end"]     # the critical points, in time order (add "agn_ign" / "jet_on" to see them)
P11_FOCUS  = "sft"                               # the epoch panels (a) and (c) ask the question at
P11_AP     = "R50_H2"                            # the aperture Sigma_H2 is measured in: "R50_H2" (the K-S convention) | "ap3kpc" (the core)
P11_NMIN   = 3
P11_ZLIM   = (-25.0, 25.0)                       # panel (c) draws the significance on a LINEAR axis clipped to this, as Part 2 does
P11_NBOOT  = 500
P11_MTERC  = 3                                   # the number of log M* cells the controlled AUC pools over
P11_PART   = True                                # add the particle-census cross-check where Part 10 has run

_NEED11 = [os.path.join(P11_KSDIR, f) for f in ("ks_track_epochs.fits", "ks_track_measurements.fits")]
if "KROT" not in globals():
    print("Part 11: needs Parts 0 and 2 — run them first.")
elif not all(os.path.exists(p) for p in _NEED11):
    print("Part 11: the m25 K-S track products are missing — ks_tracks_quenched_m25.ipynb builds them:\n  "
          + "\n  ".join(p for p in _NEED11 if not os.path.exists(p)))
else:
    _APLAB11 = {"R50_H2": r"$R_{50}({\rm H_2})$", "ap3kpc": r"$r < 3$ kpc", "ap1kpc": r"$r < 1$ kpc",
                "ap10kpc": r"$r < 10$ kpc", "R50_star": r"$R_{50}(\star)$", "R50_SFR": r"$R_{50}({\rm SFR})$"}.get(P11_AP, P11_AP)

    def _s11(t):
        d = t.to_pandas() if hasattr(t, "to_pandas") else t
        for c in d.columns:
            if d[c].dtype == object:
                d[c] = d[c].map(lambda v: v.decode().strip() if isinstance(v, bytes) else (v.strip() if isinstance(v, str) else v))
        return d

    EPO11, MEA11 = (_s11(Table.read(p)) for p in _NEED11)
    EPO11 = EPO11[EPO11["pop"] == "Q"].copy()
    _M11 = MEA11[MEA11["ap_label"] == P11_AP].set_index(["snap", "gx"])
    _g11 = lambda c: _M11[c].reindex(list(zip(EPO11["snap"].astype(int), EPO11["gx"].astype(int)))).to_numpy(float) \
        if c in _M11.columns else np.full(len(EPO11), np.nan)
    with np.errstate(all="ignore"):
        if P11_AP.startswith("R50"):                       # the K-S convention: half the mass inside the half-mass radius
            _mh11, _r11 = _g11("m_H2_tot"), _g11("r50_H2")
            EPO11["sigma_H2"] = np.log10(np.where((_mh11 > 0) & (_r11 > 0),
                                                  HE * 0.5 * _mh11 / (np.pi * _r11 ** 2) / 1e6, np.nan))
            EPO11["r_H2"] = _r11; EPO11["m_H2"] = _mh11; EPO11["n_H2"] = _g11("n_H2_tot")
        else:                                              # a fixed aperture: the mass in it over its own area
            _mh11, _ar11 = _g11("m_H2"), _g11("area_kpc2")
            EPO11["sigma_H2"] = np.log10(np.where((_mh11 > 0) & (_ar11 > 0), HE * _mh11 / _ar11 / 1e6, np.nan))
            EPO11["r_H2"] = _g11("r50_H2"); EPO11["m_H2"] = _mh11; EPO11["n_H2"] = _g11("n_H2")
        EPO11["lm_H2"] = np.log10(np.where(EPO11["m_H2"].to_numpy(float) > 0, EPO11["m_H2"].to_numpy(float), np.nan))
        EPO11["lr_H2"] = np.log10(np.where(EPO11["r_H2"].to_numpy(float) > 0, EPO11["r_H2"].to_numpy(float), np.nan))
    # ── the sample: Part 0's m25 rows, Part 2's class, and — where Part 9 has run — one chain per galaxy ──
    EPO11["key"] = list(zip(EPO11["anchor_snap"].astype(int), EPO11["gal_id"].astype(int)))
    _keep11 = set(KEEP9) if ("KEEP9" in globals() and P9_DEDUP) else None
    EPO11["kcls"] = [KCLS_OF.get((P9_BOX if "P9_BOX" in globals() else "cis25", int(a_), int(g_)))
                     for a_, g_ in EPO11["key"]]
    EPO11 = EPO11[EPO11["kcls"].notna()]
    if _keep11 is not None:
        EPO11 = EPO11[[k in _keep11 for k in EPO11["key"]]]
    EPO11 = EPO11.reset_index(drop=True)
    _KZ11 = {(int(s_), int(g_)): float(v_) for s_, g_, v_ in zip(QM[QM["box"] == "cis25"]["snap"],
                                                                QM[QM["box"] == "cis25"]["gal_id"],
                                                                QM[QM["box"] == "cis25"]["kz"])}
    EPO11["kz"] = [_KZ11.get(k, np.nan) for k in EPO11["key"]]
    print(f"Part 11 — Sigma_H2 in {P11_AP} at the critical points, {EPO11['key'].nunique()} m25 galaxies"
          + (" (one chain each)" if _keep11 is not None else "")
          + ": " + ", ".join(f"{KCLS[c]['short']} {EPO11[EPO11['kcls'] == c]['key'].nunique()}" for c in KC))

    # ── the particle-census cross-check: the same density from the member H2 particles at the nearest branch node ──
    PC11 = {}
    if P11_PART and "Z10" in globals() and len(globals().get("Z10", [])) and "BR9" in globals():
        _zi11 = Z10.set_index(["snap", "gx"])
        for _k11, _sub in EPO11.groupby("key"):
            _br = BR9[(BR9["anchor_snap"] == _k11[0]) & (BR9["gal_id"] == _k11[1])]
            if not len(_br):
                continue
            for _st in P11_STAGES:
                _r = _sub[_sub["stage"] == _st]
                if not len(_r) or not np.isfinite(_r["t_stage_gyr"].to_numpy(float)[0]):
                    continue
                _ts = float(_r["t_stage_gyr"].to_numpy(float)[0])
                _j = int(np.argmin(np.abs(_br["t"].to_numpy(float) - _ts)))
                _nd = (int(_br["snap"].to_numpy(int)[_j]), int(_br["gx"].to_numpy(int)[_j]))
                if _nd not in _zi11.index or abs(float(_br["t"].to_numpy(float)[_j]) - _ts) > 0.25:
                    continue
                _row = _zi11.loc[_nd]
                _mh, _rh = float(np.atleast_1d(_row["mh2_ism"])[0]), float(np.atleast_1d(_row["r50_H2"])[0])
                if _mh > 0 and _rh > 0:
                    PC11[(_k11, _st)] = np.log10(HE * 0.5 * _mh / (np.pi * _rh ** 2) / 1e6)
        print(f"  the particle cross-check: {len(PC11)} (galaxy, stage) cells measured from the member H2 particles "
              "of the nearest branch node (within 0.25 Gyr of the stage)")

    def _bt11(v, seed=41):
        v = np.asarray(v, float); v = v[np.isfinite(v)]
        if len(v) < P11_NMIN:
            return np.nan, np.nan, np.nan, len(v)
        rng = np.random.default_rng(seed)
        bs = np.median(v[rng.integers(0, len(v), (P11_NBOOT, len(v)))], axis=1)
        return float(np.median(v)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), len(v)

    def _auc11(sub, col):
        """The raw AUC toward the class, and the AUC pooled inside terciles of log M* at the anchor — what survives the
        mass axis that is already inside Part 2's rule."""
        v = sub[col].to_numpy(float); m = (sub["kcls"] == "rot").to_numpy(bool)
        raw = mw_auc(v, m)[0]
        lm = sub["log_mstar_anchor"].to_numpy(float)
        ok = np.isfinite(v) & np.isfinite(lm)
        if ok.sum() < 8:
            return raw, np.nan, int(ok.sum())
        _e = np.nanpercentile(lm[ok], np.linspace(0, 100, P11_MTERC + 1))
        _c = np.clip(np.digitize(lm, _e) - 1, 0, P11_MTERC - 1)
        num, den = 0.0, 0.0                                # the cell-weighted AUC: within-cell comparisons only
        for _q in range(P11_MTERC):
            _s = ok & (_c == _q)
            a, b = v[_s & m], v[_s & ~m]
            if len(a) < 2 or len(b) < 2:
                continue
            w = len(a) * len(b)
            num += mw_auc(np.concatenate([a, b]),
                          np.concatenate([np.ones(len(a), bool), np.zeros(len(b), bool)]))[0] * w
            den += w
        return raw, (num / den if den else np.nan), int(ok.sum())

    ROW11, _ST11 = [], [s for s in P11_STAGES if (EPO11["stage"] == s).any()]
    fig, axs = plt.subplots(1, 3, figsize=(13.4, 4.0))
    # (a) the distributions at the focus epoch
    _F11 = EPO11[EPO11["stage"] == P11_FOCUS]
    for _cl in KC:
        _v = np.sort(_F11[_F11["kcls"] == _cl]["sigma_H2"].to_numpy(float))
        _v = _v[np.isfinite(_v)]
        if len(_v) < 2:
            continue
        axs[0].step(_v, np.arange(1, len(_v) + 1) / len(_v), where="post", color=KCLS[_cl]["color"],
                    lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_v)})")
        axs[0].axvline(np.median(_v), color=KCLS[_cl]["color"], lw=1.0, ls=":")
    if PC11:
        for _cl in KC:
            _v = np.sort(np.array([v for (k_, s_), v in PC11.items() if s_ == P11_FOCUS
                                   and KCLS_OF.get(("cis25", k_[0], k_[1])) == _cl], float))
            if len(_v) >= 2:
                axs[0].step(_v, np.arange(1, len(_v) + 1) / len(_v), where="post", color=KCLS[_cl]["color"],
                            lw=1.1, ls="--", alpha=0.75)
    axs[0].set_xlabel(r"$\log \Sigma_{\rm H_2}$  [$M_\odot$ pc$^{-2}$]"); axs[0].set_ylabel("cumulative fraction")
    axs[0].set_title(f"(a) at {P11_FOCUS}: the whole distribution", loc="left", fontsize=10)
    axs[0].grid(alpha=0.22, lw=0.5); axs[0].legend(loc="lower right", frameon=False, fontsize=8.5)
    if PC11:
        axs[0].annotate("dashed: the same density from the H$_2$ particles", xy=(0.03, 0.95), xycoords="axes fraction",
                        fontsize=7, color="0.35", va="top")
    # (b) the sequence of critical points
    _x11 = np.arange(len(_ST11))
    for _cl in KC:
        _m, _l, _h, _n = [], [], [], []
        for _st in _ST11:
            _c, _lo, _hi, _nn = _bt11(EPO11[(EPO11["stage"] == _st) & (EPO11["kcls"] == _cl)]["sigma_H2"])
            _m.append(_c); _l.append(_lo); _h.append(_hi); _n.append(_nn)
            ROW11.append(dict(block="stage", quantity="sigma_H2", kcls=_cl, stage=_st, n_gal=_nn, med=_c, lo=_lo, hi=_hi))
        _m, _l, _h = np.array(_m), np.array(_l), np.array(_h)
        _ok = np.isfinite(_m)
        if _ok.any():
            axs[1].fill_between(_x11[_ok], _l[_ok], _h[_ok], color=KCLS[_cl]["color"], alpha=0.20, lw=0)
            axs[1].plot(_x11[_ok], _m[_ok], "o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=5, mec="k", mew=0.4)
        for _xx, _nn in zip(_x11, _n):
            axs[1].annotate(str(_nn), xy=(_xx, 0.02 + 0.05 * (_cl == "iso")), xycoords=("data", "axes fraction"),
                            fontsize=6.5, color=KCLS[_cl]["color"], ha="center")
    axs[1].set_xticks(_x11); axs[1].set_xticklabels(_ST11, fontsize=9)
    axs[1].axvline(_ST11.index(P11_FOCUS) if P11_FOCUS in _ST11 else 0, color="k", lw=1.2, ls="--", alpha=0.7)
    axs[1].set_ylabel(r"$\log \Sigma_{\rm H_2}$  [$M_\odot$ pc$^{-2}$]")
    axs[1].set_title("(b) through the critical points", loc="left", fontsize=10); axs[1].grid(alpha=0.22, lw=0.5)
    # (c) the density at the focus epoch against the selection axis itself
    _v11, _k11c = _F11["sigma_H2"].to_numpy(float), _F11["kz"].to_numpy(float)
    _co = np.where((_F11["kcls"] == "rot").to_numpy(bool), KCLS["rot"]["color"], KCLS["iso"]["color"])
    axs[2].scatter(np.clip(_k11c, *P11_ZLIM), _v11, s=26, c=_co, edgecolor="k", linewidth=0.4, alpha=0.85, zorder=3)
    if np.nanmax(np.abs(_k11c)) > P11_ZLIM[1]:
        axs[2].annotate(f"the axis runs past $\\pm${P11_ZLIM[1]:g}; points outside sit on the edge", xy=(0.02, 0.97),
                        xycoords="axes fraction", fontsize=6.5, color="0.4", va="top")
    axs[2].axvline(ROT_SIG, color="k", lw=1.3, ls="--")
    axs[2].annotate(rf"Part 2's cut, {ROT_SIG:g}$\,s_{{\rm iso}}$", xy=(ROT_SIG, 0.02), xycoords=("data", "axes fraction"),
                    xytext=(4, 0), textcoords="offset points", fontsize=7.5, color="0.3")
    _ok11 = np.isfinite(_v11) & np.isfinite(_k11c)
    if _ok11.sum() >= 6:
        _rho, _p = spearmanr(_k11c[_ok11], _v11[_ok11])
        axs[2].set_title(rf"(c) at {P11_FOCUS} against the selection axis  ($\rho$ = {_rho:+.2f}, p = {_p:.2g})",
                         loc="left", fontsize=10)
    else:
        axs[2].set_title(f"(c) at {P11_FOCUS} against the selection axis", loc="left", fontsize=10)
    axs[2].set_xlabel(r"$(\kappa_{\rm gas} - 1/3)\,/\,s_{\rm iso}(N_{\rm gas})$  at the anchor")
    axs[2].set_ylabel(r"$\log \Sigma_{\rm H_2}$  [$M_\odot$ pc$^{-2}$]"); axs[2].grid(alpha=0.22, lw=0.5)
    fig.suptitle(rf"{BOX_LABEL.get('cis25')} — was the H$_2$ of the future rotators already denser at the truncation? "
                 f"($\\Sigma_{{\\rm H_2}}$ inside {_APLAB11})", y=1.02, fontsize=12, fontweight="bold")
    fig.tight_layout(); paper_save(fig, "kappa_m25_sigma_h2_at_sft"); plt.show()

    # ── the answer, in numbers ──
    print(f"\nthe three quantities behind the density, per critical point "
          f"(rotating / the rest;  AUC raw [AUC within {P11_MTERC} log M* cells];  MW p):")
    for _q11, _lab11 in (("sigma_H2", "log Sigma_H2 [Msun/pc2]"), ("lm_H2", "log M_H2 [Msun]"),
                         ("lr_H2", "log R50(H2) [kpc]"), ("n_H2", "N_H2 particles")):
        print(f"  {_lab11}:")
        for _st in _ST11:
            _s = EPO11[EPO11["stage"] == _st]
            _a = _s[_s["kcls"] == "rot"][_q11].to_numpy(float); _b = _s[_s["kcls"] == "iso"][_q11].to_numpy(float)
            _a, _b = _a[np.isfinite(_a)], _b[np.isfinite(_b)]
            _raw, _ctl, _n = _auc11(_s, _q11)
            _p = mw_auc(_s[_q11].to_numpy(float), (_s["kcls"] == "rot").to_numpy(bool))[1]
            print(f"    {_st:9s} " + (f"{np.median(_a):+7.2f}" if len(_a) >= P11_NMIN else "      —")
                  + " / " + (f"{np.median(_b):+7.2f}" if len(_b) >= P11_NMIN else "      —")
                  + f"   (N {len(_a)}, {len(_b)})   AUC {_raw:.2f}"
                  + (f" [{_ctl:.2f}]" if np.isfinite(_ctl) else " [—]") + (f"   p {_p:.3f}" if np.isfinite(_p) else ""))
            ROW11.append(dict(block="score", quantity=_q11, kcls="both", stage=_st, n_gal=len(_a) + len(_b),
                              med=float(np.median(_a)) if len(_a) else np.nan,
                              lo=float(np.median(_b)) if len(_b) else np.nan, hi=float(_raw), auc_ctl=float(_ctl)))
    if PC11:
        _pa = [v for (k_, s_), v in PC11.items() if s_ == P11_FOCUS and KCLS_OF.get(("cis25", k_[0], k_[1])) == "rot"]
        _pb = [v for (k_, s_), v in PC11.items() if s_ == P11_FOCUS and KCLS_OF.get(("cis25", k_[0], k_[1])) == "iso"]
        if len(_pa) >= P11_NMIN and len(_pb) >= P11_NMIN:
            _pauc = mw_auc(np.array(_pa + _pb), np.concatenate([np.ones(len(_pa), bool), np.zeros(len(_pb), bool)]))[0]
            print(f"\n  the particle cross-check at {P11_FOCUS} (member H2 particles of the nearest node): "
                  f"{np.median(_pa):+.2f} / {np.median(_pb):+.2f}  (N {len(_pa)}, {len(_pb)})   AUC {_pauc:.2f}")
    print("\n  a raw AUC of 0.5 is no difference; the bracket is the same score computed only between galaxies of similar\n"
          "  stellar mass, which is the number to quote — Part 2's rule already contains log M* > 10.5.")
    pd.DataFrame(ROW11).to_csv(os.path.join(OUT, "kappa_m25_sigma_h2_at_sft.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_m25_sigma_h2_at_sft.csv')}")
```

## Part 12 — the radial profiles at the critical points, the central hole, and what the hole does to the dust

Parts 5 and 7 already read radial information, but through the m25 notebook's four coarse annuli (0–1, 1–3.2, 3.2–10, 10–32
kpc), for the handful of galaxies the K-S products cover, at the stages those products define. This part goes back to the
particles — the same reduced files Part 10 reads — and asks the question those annuli are too coarse to answer: **where does
the gas actually sit**, in fourteen logarithmic bins from 0.25 to 32 kpc, at the four epochs that matter.

**The four critical points**, all taken off the *same* chain Part 9 built, so the stages and the clocks cannot disagree:

| stage | definition |
|---|---|
| `agn_ign` | the AGN ignites — the first step whose BHAR reaches `P12_AGNFRAC` of its own pre-truncation peak (the convention of the m25 notebook's Part 2d windows table, rebuilt here from the histories themselves) |
| `xray_on` | the first step with SIMBA's X-ray channel firing — Part 0's `agn_states`: full-velocity jets **and** $f_{\rm gas} <$ `XRAY_FGAS_MAX` |
| `sft` | the truncation, $t_{\rm SFT}$ |
| `anchor` | the epoch the galaxy enters Part 0's sample at — the quenched end state |

**The geometry.** A rotating gas disc has a plane, so the profile is measured **face-on in that plane**: the particles are
rotated into the frame of the member gas's own angular momentum, $R$ is the cylindrical radius there, and only the slab
$|z| <$ `P12_ZMAX` counts. The misalignment between that plane and the stellar body is measured and printed — this notebook's
whole premise is that the gas can rotate where the stars do not, so the two frames are never assumed to be the same one.
`P12_FRAME` switches to the stellar frame, or off.

**The hole.**

$$\mathrm{hole} \;=\; \log_{10}\frac{\Sigma_{\rm peak}}{\Sigma(<R_{\rm in})}$$

the peak of the profile outside `P12_RIN` against the mean surface density **inside** it. A galaxy whose gas sits in a ring
has a large positive value; a centrally concentrated one has ~0 or below (the inner aperture is a cumulative mean and the
peak is an annulus, so a declining profile scores negative — the two regimes do not overlap). Two guards make it a
measurement rather than an artefact of sparseness:

* $\Sigma(<R_{\rm in})$ is **floored at one particle** in the aperture, so an empty centre gives a large but finite number;
* a galaxy is classified only where the emptiness could have been seen — `n_exp`, the count the peak surface density would
  put inside the aperture, must reach `P12_NEXP`. With ~500–5000 member gas particles per node in m25 (Part 10's census)
  this is a real measurement and not the few-particle trap $\kappa_{\rm rot}$ falls into; the counts are printed and the
  individual galaxies are drawn behind every median.

**Then the point of it all.** The galaxies are split by the hole at `P12_HOLE_AT` and their **dust clocks** drawn against
each other — $M_{\rm dust}/M_\star$, the dust-to-metal ratio and $M_{\rm H_2}/M_\star$ on $t - t_{\rm SFT}$ — with the
hole × class cross in panel (b), so it can be read off whether the dust is following the hole or the rotation.

*Not a pure read.* It reads the m25 reduced particle files and **caches** the profiles (`P12_CACHE`, an `.npz` keyed on the
bins and the geometry — change `P12_RB`, `P12_ZMAX` or `P12_FRAME` and it rebuilds itself). Only the nodes of the four stages
are read by default (~4 files per galaxy, a few hundred MB); `P12_ALLNODES = True` measures every node of every branch
instead — the cache is incremental, so that only ever reads the files it does not already hold.

**One scale to keep in mind.** The hole is measured at 1 kpc. The m25 gravitational softening is well below it (the band
shaded on every profile panel), so this is resolved in the simulation — but it is a sub-kpc structure, and only a map with
sub-kpc resolution could test it against a real quiescent galaxy.


```python
# ── Part 12 — THE RADIAL PROFILES at the critical points, the central HOLE, and what the hole does to the dust ──
# Parts 5 and 7 already read radial information, but through the m25 notebook's four coarse annuli (0-1, 1-3.2, 3.2-10,
# 10-32 kpc), for the handful of galaxies the K-S products cover, and at the stages those products define. This part goes back
# to the particles themselves — the same reduced files Part 10 reads — and asks the question those annuli are too coarse for:
# WHERE the gas actually sits, in 14 logarithmic bins from 0.25 to 32 kpc, at the four epochs that matter.
#   THE FOUR CRITICAL POINTS, all taken off the SAME chain Part 9 built, so the stages and the clocks cannot disagree:
#     agn_ign   the AGN ignites — the first step whose BHAR is at least P12_AGNFRAC of its own pre-truncation peak
#               (the convention of the m25 notebook's Part 2d windows table, rebuilt here from the histories)
#     xray_on   the first step with SIMBA's X-ray channel firing (Part 0's agn_states: full jets AND f_gas < XRAY_FGAS_MAX)
#     sft       the truncation itself (t_SFT, the ism_prediction classes table)
#     anchor    the epoch the galaxy enters Part 0's sample at — the quenched end state
#   THE GEOMETRY. A rotating gas disc has a plane, so the profile is measured FACE-ON in that plane: the particles are rotated
#   into the frame of the member gas's own angular momentum, R is the cylindrical radius there and only the slab |z| < P12_ZMAX
#   is counted. The misalignment between that plane and the stellar body is measured and printed — this notebook's whole point
#   is that the gas rotates when the stars need not, so the two frames are not assumed to be the same one.
#   THE HOLE. hole = log10( Sigma_peak / Sigma_in ), the peak of the profile outside P12_RIN against the mean surface density
#   INSIDE it: a galaxy whose gas sits in a ring has a large positive value, a centrally concentrated one has ~0. An empty
#   centre would give infinity, so Sigma_in is floored at ONE PARTICLE in the aperture, and a galaxy is only classified when
#   the profile is dense enough for the emptiness to mean something: n_exp = the count the peak surface density would put
#   inside the aperture must be at least P12_NEXP. With ~500-5000 member gas particles per node in m25 (Part 10's census) that
#   is a real measurement and not the few-particle trap of kappa_rot — the count itself is drawn behind every profile.
# THEN THE POINT OF IT ALL: the galaxies are split by the hole at P12_HOLE_AT and their DUST CLOCKS are drawn against each
# other — f_dust, the dust-to-metal ratio and f_H2 on t - t_SFT — with the hole x class cross, so it can be seen whether the
# hole or the rotation is what the dust is actually following.
# NOT a pure read: it reads the m25 reduced particle files and CACHES the profiles (P12_CACHE, an .npz keyed on the bins —
# change P12_RB / P12_ZMAX / P12_FRAME and it rebuilds itself). Only the nodes of the four stages are read by default
# (~4 files per galaxy); P12_ALLNODES = True measures every node of every branch instead and unlocks the hole CLOCK.
P9_BOX = "cis25"
P12_DIR     = os.path.join(os.getcwd(), "output", P9_BOX, "reduced_particles")
P12_PREFIX  = {"cis25": "m25n512", "cis50": "m50n512", "cis50nox": "m50n512", "cis100": "m100n1024"}.get(P9_BOX, P9_BOX)
P12_CACHE   = os.path.join(OUT, f"kappa_{P9_BOX}_profiles.npz")
P12_STAGES  = ["agn_ign", "xray_on", "sft", "anchor"]     # the columns of Fig 1, in time order ("jet_on" can be added)
P12_TOL     = 0.4          # Gyr: how close a branch node with a particle file must be to a stage for that stage to be measured
P12_AGNFRAC = 0.5          # "the AGN ignites" = the BHAR first reaches this fraction of its own pre-SFT peak
P12_RB      = 2.0 ** np.arange(-2.0, 5.001, 0.5)     # the radial bin EDGES [physical kpc]: 0.25 -> 32, sqrt(2) apart, so
                            # 0.25, 0.5, 1, 2, 4, 8, 16, 32 are all edges and P12_RIN / P12_ROUT can sit exactly on one
P12_RB_RE   = 2.0 ** np.arange(-3.0, 3.001, 0.5)     # the same in units of the galaxy's own R50(*): 0.125 -> 8
P12_XAXIS   = "kpc"        # which of the two grids Fig 1 draws: "kpc" (physical) | "re" (R / R50*). BOTH are cached
P12_ZMAX    = 10.0         # the half-thickness of the face-on slab [kpc] (None = the whole 100 kpc aperture)
P12_FRAME   = "gas"        # the disc plane: "gas" = the member gas's angular momentum | "star" = the stellar minor axis
                            # (the file's own evecs) | "none" = no rotation, R is the 3D radius and Sigma a shell density
P12_RSOFT   = 0.5          # [kpc] the gravitational-softening scale of the box, shaded on every profile panel
P12_QUANT   = [("gas",  r"$\Sigma_{\rm gas}$  [$M_\odot$ pc$^{-2}$]", -2.0),
               ("h2",   r"$\Sigma_{\rm H_2}$  [$M_\odot$ pc$^{-2}$]", -3.0),
               ("dust", r"$\Sigma_{\rm dust}$  [$M_\odot$ pc$^{-2}$]", -5.0)]   # (the census suffix, the label, the log floor
                            # an EMPTY bin is drawn at — a hole is exactly where the surface density goes to zero, so the
                            # empty bins cannot be dropped without erasing the effect they are evidence for)
P12_MEMBER  = True        # the profile over the caesar MEMBER gas only (True), or over every gas particle in the slab
                            # (False — what a map of the galaxy would show, the CGM along the line of sight included)
P12_HQ      = "gas"        # the quantity the hole is defined on
P12_RIN     = 1.0          # [kpc] "the centre": the aperture Sigma_in is measured in (snapped to a bin edge)
P12_ROUT    = 8.0          # [kpc] the peak outside the centre is looked for within this
P12_HOLE    = 0.1          # dex: a HOLE = the centre is at least this far below the peak further out (a factor 2)
P12_HOLE_AT = "sft"        # the stage the hole is measured at, and the split Fig 3's clocks are drawn by
P12_NEXP    = 1            # the hole is only CLASSIFIED where the peak surface density would put this many particles inside
                            # the aperture — below it the centre is empty because the galaxy is sparse, which is not a hole
P12_NMIN    = 1            # galaxies a drawn point needs (Part 9's floor: m25 is tens of objects)
P12_NBOOT   = 300
P12_SPAG    = True         # the individual galaxies behind the medians
P12_ALLNODES = False       # measure EVERY node of every branch (~1200 files) instead of only the stage nodes (~4 per galaxy).
                            # The cache is incremental, so this can be flipped on later and only the new files are read
P12_MARKS   = [-1.0, 0.0, 1.0, 2.0]
HCLS = {"hole": dict(color="#7b3294", lw=2.6, ls="-", short="central hole",
                     label=rf"central hole  ($\log \Sigma_{{\rm peak}}/\Sigma_{{<{P12_RIN:g}\,\rm kpc}} \geq {P12_HOLE:g}$)"),
        "full": dict(color="#008837", lw=2.4, ls="-", short="filled centre",
                     label=rf"filled centre  ($< {P12_HOLE:g}$ dex)")}
HC = list(HCLS)

if "CH9" not in globals():
    print("Part 12: needs Part 9 — run it first (the chains, the branch index and the AGN state code come from it).")
elif not os.path.isdir(P12_DIR):
    print(f"Part 12: {P12_DIR} is missing — build_reduced_particles_job.py writes it (cluster).")
elif "st" not in next(iter(CH9.values())):
    print("Part 12: the chains carry no AGN state code — re-run Part 9 (its cell now stores 'st' and 'mdot').")
else:
    import glob as _glob12, re as _re12, time as _time12

    # ── (A) THE FOUR CRITICAL POINTS of every retained galaxy, off its own chain ──
    _SLAB12 = {"agn_ign": "AGN ignition", "jet_on": "jet onset", "xray_on": "X-ray channel on",
               "sft": "the truncation (SFT)", "anchor": "the anchor (quenched)"}
    _SSHORT12 = {"agn_ign": "AGN ignition", "jet_on": "jet onset", "xray_on": "X-ray on", "sft": "SFT", "anchor": "anchor"}

    def _leg12(ax, **kw):
        """A legend only where something was actually drawn — a thin stage x class cell draws nothing at all."""
        if ax.get_legend_handles_labels()[0]:
            ax.legend(**{"frameon": False, "fontsize": 8, **kw})
    ST12 = {}
    for _k12 in KEEP9:
        _d12, _ts12 = CH9[_k12], _tsft9.get(_k12, np.nan)
        _t12, _st12, _md12 = _d12["t"], _d12["st"], _d12["mdot"]
        _pre = np.isfinite(_md12) & (_md12 > 0) & (_t12 <= (_ts12 if np.isfinite(_ts12) else np.inf))
        _pk = float(np.nanmax(_md12[_pre])) if _pre.any() else np.nan
        _ign = _t12[np.isfinite(_md12) & (_md12 >= P12_AGNFRAC * _pk)] if np.isfinite(_pk) else np.array([])
        _jet = _t12[np.nan_to_num(_st12, nan=-1) >= 2]
        _xr = _t12[np.nan_to_num(_st12, nan=-1) >= 3]
        ST12[_k12] = {"agn_ign": float(_ign[0]) if len(_ign) else np.nan,
                      "jet_on": float(_jet[0]) if len(_jet) else np.nan,
                      "xray_on": float(_xr[0]) if len(_xr) else np.nan,
                      "sft": _ts12, "anchor": T9.get(int(_k12[0]), np.nan)}
    print(f"Part 12 — the critical points of the {len(KEEP9)} retained {P9_BOX} galaxies "
          "(median t - t_SFT [Gyr], and how many galaxies ever reach the stage):")
    for _s12 in P12_STAGES:
        _dt = np.array([ST12[k][_s12] - ST12[k]["sft"] for k in KEEP9], float)
        print(f"  {_s12:8s} {_SLAB12[_s12]:26s} {np.nanmedian(_dt):+6.2f}   "
              f"{int(np.isfinite(_dt).sum())}/{len(KEEP9)}")

    # ── (B) the node of each (galaxy, stage): the branch node nearest in time that HAS a particle file ──
    _inv12 = {}
    for _p12 in _glob12.glob(os.path.join(P12_DIR, "snap_*", f"{P12_PREFIX}_snap*_gal*.h5")):
        _m12 = _re12.search(r"_snap(\d+)_gal(\d+)\.h5$", _p12)
        if _m12:
            _inv12[(int(_m12.group(1)), int(_m12.group(2)))] = _p12
    _BRK12 = {k: d for k, d in BR9.groupby(["anchor_snap", "gal_id"])}
    NOD12, _cov12 = {}, {s: 0 for s in P12_STAGES}
    for _k12 in KEEP9:
        _b12 = _BRK12.get(_k12)
        if _b12 is None:
            continue
        _av = [(int(s_), int(x_), float(t_)) for s_, x_, t_ in zip(_b12["snap"], _b12["gx"], _b12["t"])
               if (int(s_), int(x_)) in _inv12 and np.isfinite(t_)]
        if not _av:
            continue
        _tt = np.array([a[2] for a in _av], float)
        for _s12 in P12_STAGES:
            _t0 = ST12[_k12][_s12]
            if not np.isfinite(_t0):
                continue
            _i = int(np.argmin(np.abs(_tt - _t0)))
            if abs(_tt[_i] - _t0) <= P12_TOL:
                NOD12[(_k12, _s12)] = (_av[_i][0], _av[_i][1], _av[_i][2] - _t0)
                _cov12[_s12] += 1
    _need12 = sorted({(v[0], v[1]) for v in NOD12.values()})
    if P12_ALLNODES:
        _kk12 = set(KEEP9)
        _need12 = sorted({(int(s_), int(x_)) for a_, g_, s_, x_ in
                          zip(BR9["anchor_snap"], BR9["gal_id"], BR9["snap"], BR9["gx"])
                          if (int(a_), int(g_)) in _kk12 and (int(s_), int(x_)) in _inv12})
    _off12 = {s: np.array([abs(v[2]) for (k_, s_), v in NOD12.items() if s_ == s], float) for s in P12_STAGES}
    print(f"  {len(_inv12)} particle files on disk;  matched within {P12_TOL:g} Gyr (median offset in brackets):  "
          + ",  ".join(f"{s} {_cov12[s]}/{len(KEEP9)}"
                       + (f" [{np.median(_off12[s]):.2f} Gyr]" if len(_off12[s]) else "") for s in P12_STAGES)
          + f"  ->  {len(_need12)} distinct files to measure" + ("  (P12_ALLNODES: every node)" if P12_ALLNODES else ""))

    # ── (C) THE PROFILE CENSUS, cached ──
    _QP12 = ["gas", "h2", "dust", "gasmem", "h2mem", "dustmem", "star"]
    _GR12 = {"kpc": P12_RB, "re": P12_RB_RE}

    def _basis12(n):
        """An orthonormal frame whose THIRD axis is n: the profile is then measured in the (1, 2) plane."""
        n = np.asarray(n, float); n = n / np.linalg.norm(n)
        a = np.array([0.0, 0.0, 1.0]) if abs(n[2]) < 0.9 else np.array([1.0, 0.0, 0.0])
        e1 = np.cross(a, n); e1 /= np.linalg.norm(e1)
        return np.column_stack([e1, np.cross(n, e1), n])

    def _prof12(path):
        """ONE reduced particle file -> the binned mass and count profiles of every quantity, on both radial grids."""
        with h5py.File(path, "r") as h:
            A = dict(h.attrs); G, St = h["gas"], h["star"]
            gp = G["pos"][:].astype(float); gmem = G["member"][:].astype(bool)
            gm = G["m_gas"][:].astype(float); gd = G["m_dust"][:].astype(float)
            gh2 = G["m_H2"][:].astype(float) if "m_H2" in G else np.full(len(gm), np.nan)
            gv = G["vel"][:].astype(float) if "vel" in G else np.full((len(gm), 3), np.nan)
            sp = St["pos"][:].astype(float); sm = St["m_star"][:].astype(float)
            smem = St["member"][:].astype(bool)
            sv = St["vel"][:].astype(float) if "vel" in St else np.full((len(sm), 3), np.nan)
        _ev = np.asarray(A.get("evecs", np.eye(3)), float)          # the STELLAR principal frame; column 2 = the minor axis
        _minor = _ev[:, 2] if _ev.shape == (3, 3) else np.array([0.0, 0.0, 1.0])
        _ok = smem & np.isfinite(sv).all(1) & (sm > 0)
        _gok = gmem & np.isfinite(gv).all(1) & (gm > 0)
        vsys = (np.average(sv[_ok], axis=0, weights=sm[_ok]) if _ok.sum() >= 10 else          # THE REST FRAME, as in Part 10
                np.average(gv[_gok], axis=0, weights=gm[_gok]) if _gok.sum() >= 10 else np.zeros(3))
        _Ls = np.cross(sp[_ok], (sv[_ok] - vsys) * sm[_ok, None]).sum(0) if _ok.sum() >= 20 else np.full(3, np.nan)
        nrm, frm = None, 3
        if P12_FRAME == "gas" and _gok.sum() >= 20:                 # the disc plane of the MEMBER GAS itself
            _L = np.cross(gp[_gok], (gv[_gok] - vsys) * gm[_gok, None]).sum(0)
            if np.isfinite(_L).all() and np.linalg.norm(_L) > 0:
                nrm, frm = _L / np.linalg.norm(_L), 0
        if nrm is None and P12_FRAME in ("gas", "star") and np.isfinite(_Ls).all() and np.linalg.norm(_Ls) > 0:
            nrm, frm = _Ls / np.linalg.norm(_Ls), 1
        if nrm is None and P12_FRAME != "none":
            nrm, frm = _minor / np.linalg.norm(_minor), 2
        if P12_FRAME == "none" or nrm is None:
            R, Rs, zg, zs, frm = (np.linalg.norm(gp, axis=1), np.linalg.norm(sp, axis=1),
                                  np.zeros(len(gm)), np.zeros(len(sm)), 3)
        else:
            E = _basis12(nrm)
            _qg, _qs = gp @ E, sp @ E
            R, Rs = np.hypot(_qg[:, 0], _qg[:, 1]), np.hypot(_qs[:, 0], _qs[:, 1])
            zg, zs = np.abs(_qg[:, 2]), np.abs(_qs[:, 2])
        _sl = (zg <= P12_ZMAX) if (P12_ZMAX and frm != 3) else np.ones(len(gm), bool)
        _sls = (zs <= P12_ZMAX) if (P12_ZMAX and frm != 3) else np.ones(len(sm), bool)
        row = {"snap": int(A["snap"]), "gx": int(A["gx"]), "z": float(A["redshift"]), "frame": int(frm),
               "misalign": float(np.degrees(np.arccos(np.clip(abs(nrm @ _minor), 0, 1)))) if nrm is not None else np.nan,
               "misalign_ls": (float(np.degrees(np.arccos(np.clip(nrm @ _Ls / np.linalg.norm(_Ls), -1, 1))))
                               if (nrm is not None and np.isfinite(_Ls).all() and np.linalg.norm(_Ls) > 0) else np.nan),
               "n_memgas": int(gmem.sum()), "m_part": float(np.nanmedian(gm[gm > 0])) if (gm > 0).any() else np.nan,
               "m_star_mem": float(np.nansum(sm[smem])), "mgas_mem": float(np.nansum(gm[gmem])),
               "mdust_mem": float(np.nansum(gd[gmem]))}
        for _nm, _msk, _w, _rr in (("star", smem, sm, Rs), ("gasmem", gmem, gm, R), ("h2", gmem & (gh2 > 0), gh2, R)):
            _tot = float(np.nansum(_w[_msk])) if _msk.any() else 0.0
            if _msk.sum() >= 5 and _tot > 0:
                _o = np.argsort(_rr[_msk])
                row[f"r50_{_nm}"] = float(np.interp(0.5, np.cumsum(_w[_msk][_o]) / _tot, _rr[_msk][_o]))
            else:
                row[f"r50_{_nm}"] = np.nan
        _SRC12 = {"gas": (gm, _sl, R), "h2": (gh2, _sl, R), "dust": (gd, _sl, R),
                  "gasmem": (gm, _sl & gmem, R), "h2mem": (gh2, _sl & gmem, R), "dustmem": (gd, _sl & gmem, R),
                  "star": (sm, _sls, Rs)}
        for _gn, _rb in _GR12.items():
            _sc = row["r50_star"] if _gn == "re" else 1.0
            for _qn in _QP12:
                _w, _mk, _rr = _SRC12[_qn]
                _nb = len(_rb) - 1
                if not np.isfinite(_sc) or _sc <= 0 or not np.isfinite(_w).any():
                    row[f"m_{_qn}_{_gn}"] = np.full(_nb, np.nan); row[f"n_{_qn}_{_gn}"] = np.full(_nb, np.nan)
                    row[f"m0_{_qn}_{_gn}"] = np.nan; row[f"n0_{_qn}_{_gn}"] = np.nan
                    continue
                _x = _rr[_mk] / _sc; _ww = np.nan_to_num(_w[_mk])
                row[f"m_{_qn}_{_gn}"] = np.histogram(_x, bins=_rb, weights=_ww)[0]
                row[f"n_{_qn}_{_gn}"] = np.histogram(_x, bins=_rb)[0].astype(float)
                _in = _x < _rb[0]
                row[f"m0_{_qn}_{_gn}"] = float(_ww[_in].sum()); row[f"n0_{_qn}_{_gn}"] = float(_in.sum())
        return row

    P12 = {}
    if os.path.exists(P12_CACHE):
        _z = np.load(P12_CACHE, allow_pickle=False)
        if (len(_z["rb_kpc"]) == len(P12_RB) and np.allclose(_z["rb_kpc"], P12_RB)
                and len(_z["rb_re"]) == len(P12_RB_RE) and np.allclose(_z["rb_re"], P12_RB_RE)
                and float(_z["zmax"]) == float(P12_ZMAX or -1) and str(_z["frame_mode"]) == P12_FRAME
                and bool(_z["member_only"]) == bool(P12_MEMBER)):
            P12 = {k: _z[k] for k in _z.files if k not in ("rb_kpc", "rb_re", "zmax", "frame_mode", "member_only")}
            print(f"  the profile census, cached -> {P12_CACHE}  ({len(P12['snap'])} files already measured)")
        else:
            print(f"  {P12_CACHE} was built with different bins / geometry — it is REBUILT from scratch")
    _done12 = set(zip(P12["snap"].astype(int), P12["gx"].astype(int))) if P12 else set()
    _todo12 = [k for k in _need12 if k not in _done12]
    if _todo12:
        print(f"  measuring {len(_todo12)} files "
              f"(~{sum(os.path.getsize(_inv12[k]) for k in _todo12) / 2 ** 30:.2f} GiB to read; cached afterwards)",
              flush=True)
        _rows12, _t012, _bad12 = [], _time12.time(), []
        for _i12, _k12 in enumerate(_todo12):
            try:
                _rows12.append(_prof12(_inv12[_k12]))
            except Exception as _e12:
                _bad12.append(f"{os.path.basename(_inv12[_k12])}: {type(_e12).__name__}: {_e12}")
            if (_i12 + 1) % 100 == 0:
                _el = _time12.time() - _t012
                print(f"     {_i12 + 1}/{len(_todo12)} files, {_el:.0f} s "
                      f"(~{_el / (_i12 + 1) * (len(_todo12) - _i12 - 1):.0f} s left)", flush=True)
        for _b12 in _bad12[:5]:
            print(f"     file not read, {_b12}")
        if _rows12:
            _new = {k: np.array([r[k] for r in _rows12]) for k in _rows12[0]}
            if P12 and set(P12) != set(_new):
                print("     the cached census has a different set of columns — it is replaced by this run's")
                P12 = {}
            P12 = {k: np.concatenate([P12[k], _new[k]]) for k in _new} if P12 else _new
            np.savez_compressed(P12_CACHE, rb_kpc=P12_RB, rb_re=P12_RB_RE, zmax=float(P12_ZMAX or -1),
                                frame_mode=P12_FRAME, member_only=bool(P12_MEMBER), **P12)
            print(f"  the profile census -> {P12_CACHE}  ({len(P12['snap'])} files, {_time12.time() - _t012:.0f} s)")
    if not P12:
        print("  nothing measured — Part 12 stops here.")
    else:
        IDX12 = {(int(s_), int(x_)): i for i, (s_, x_) in enumerate(zip(P12["snap"], P12["gx"]))}
        _FR12 = ["the member gas's angular momentum", "the member stars' angular momentum",
                 "the stellar minor axis", "no rotation (3D radius)"]
        _fc = np.asarray(P12["frame"], int)
        print("  the disc plane used: " + ", ".join(f"{_FR12[i]} {int((_fc == i).sum())}" for i in range(4) if (_fc == i).any())
              + f";  it is {np.nanmedian(P12['misalign']):.0f} deg from the stellar MINOR AXIS in the median (which is barely "
                f"defined in a spheroid) and {np.nanmedian(P12['misalign_ls']):.0f} deg from the stellar ANGULAR MOMENTUM "
                f"({100 * np.nanmean(np.asarray(P12['misalign_ls'], float) > 30):.0f} % of the nodes beyond 30 deg: the gas "
                "and the stars need not turn together, which is this notebook's premise)")

        # ── (D) the profiles and the hole metric, derived from the cache ──
        _MEM12 = "mem" if P12_MEMBER else ""
        _XC12 = np.sqrt(_GR12[P12_XAXIS][1:] * _GR12[P12_XAXIS][:-1])     # the geometric centre of each DRAWN bin
        _XK12 = np.sqrt(P12_RB[1:] * P12_RB[:-1])                         # ... and of the physical-kpc bins the hole uses

        def _sig12(i, q, grid=None):
            """Node i, quantity q -> the surface density profile [Msun/pc^2] (NaN where the node has no profile)."""
            g = P12_XAXIS if grid is None else grid
            _q = q + (_MEM12 if q in ("gas", "h2", "dust") else "")
            m = np.asarray(P12[f"m_{_q}_{g}"][i], float)
            _rb = _GR12[g] * (float(P12["r50_star"][i]) if g == "re" else 1.0)      # the "re" grid in physical kpc
            return m / (np.pi * (_rb[1:] ** 2 - _rb[:-1] ** 2) * 1e6)

        _iin12 = int(np.argmin(np.abs(P12_RB - P12_RIN)))            # the aperture, snapped to a bin edge
        _RIN12, _iout12 = float(P12_RB[_iin12]), int(np.argmin(np.abs(P12_RB - P12_ROUT)))

        def _hole12(i, q=None):
            """Node i -> (the hole in dex, the radius of the peak [kpc], the count inside the aperture, the count the peak
            surface density would put there). The centre is floored at ONE PARTICLE, so an empty centre is bounded."""
            q = P12_HQ if q is None else q
            _q = q + (_MEM12 if q in ("gas", "h2", "dust") else "")
            m = np.asarray(P12[f"m_{_q}_kpc"][i], float); n = np.asarray(P12[f"n_{_q}_kpc"][i], float)
            if not np.isfinite(m).any():
                return (np.nan,) * 4
            _ain = np.pi * _RIN12 ** 2 * 1e6
            _min = float(P12[f"m0_{_q}_kpc"][i]) + float(np.nansum(m[:_iin12]))
            _nin = float(P12[f"n0_{_q}_kpc"][i]) + float(np.nansum(n[:_iin12]))
            _mp = float(P12["m_part"][i])
            _sin = max(_min, _mp if np.isfinite(_mp) else 0.0) / _ain
            _band = slice(_iin12, max(_iin12 + 1, _iout12))
            _sb = _sig12(i, q, "kpc")[_band]
            if not np.isfinite(_sb).any() or np.nanmax(_sb) <= 0 or _sin <= 0:
                return (np.nan,) * 4
            _j = int(np.nanargmax(_sb))
            return (float(np.log10(np.nanmax(_sb) / _sin)), float(_XK12[_iin12 + _j]), _nin,
                    float(np.nanmax(_sb) * _ain / _mp) if _mp > 0 else np.nan)

        H12 = {}
        for (_k12, _s12), (_sn, _gx, _off) in NOD12.items():
            _i = IDX12.get((_sn, _gx))
            if _i is None:
                continue
            _h, _rp, _nin, _nexp = _hole12(_i)
            H12[(_k12, _s12)] = dict(i=_i, snap=_sn, gx=_gx, dt=_off, hole=_h, r_peak=_rp, n_in=_nin, n_exp=_nexp,
                                     ok=bool(np.isfinite(_h) and np.isfinite(_nexp) and _nexp >= P12_NEXP))
        print(f"  the hole, on {P12_HQ}{' (member gas only)' if P12_MEMBER else ''}: the centre is r < {_RIN12:g} kpc, the peak "
              f"is looked for out to {P12_RB[_iout12]:g} kpc; measurable (n_exp >= {P12_NEXP:g}) at "
              + ", ".join(f"{s} {sum(1 for (k_, s_), v in H12.items() if s_ == s and v['ok'])}/{_cov12[s]}"
                          for s in P12_STAGES))

        # ── FIG 1: the profiles at the four critical points ──
        _NS12 = len(P12_STAGES)
        fig, axs = plt.subplots(len(P12_QUANT), _NS12, figsize=(3.35 * _NS12, 3.0 * len(P12_QUANT)),
                                sharex=True, sharey="row", squeeze=False)
        ROW12 = []
        for _qi, (_qn, _ylab, _flo) in enumerate(P12_QUANT):
            for _si, _s12 in enumerate(P12_STAGES):
                ax = axs[_qi, _si]
                for _cl in KC:
                    _ii = [v["i"] for (k_, s_), v in H12.items() if s_ == _s12 and CLS9.get(k_) == _cl]
                    if not _ii:
                        continue
                    A12 = np.array([np.log10(np.maximum(_sig12(i, _qn), 10.0 ** _flo)) for i in _ii], float)
                    _n12 = np.isfinite(A12).sum(axis=0)
                    with warnings.catch_warnings():
                        warnings.simplefilter("ignore", RuntimeWarning)
                        _cen = np.nanmedian(A12, axis=0)
                        _rng = np.random.default_rng(7)
                        _bs = np.array([np.nanmedian(A12[_rng.integers(0, len(A12), len(A12))], axis=0)
                                        for _ in range(P12_NBOOT)])
                        _lo, _hi = np.nanpercentile(_bs, 16, axis=0), np.nanpercentile(_bs, 84, axis=0)
                    _ok = _n12 >= P12_NMIN
                    if P12_SPAG:
                        for _r12 in A12:
                            ax.plot(_XC12, _r12, color=KCLS[_cl]["color"], lw=0.5,
                                    alpha=float(np.clip(12.0 / max(len(A12), 1), 0.06, 0.35)), zorder=2)
                    ax.fill_between(_XC12[_ok], _lo[_ok], _hi[_ok], color=KCLS[_cl]["color"], alpha=0.22, lw=0, zorder=3)
                    ax.plot(_XC12[_ok], _cen[_ok], color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], zorder=5,
                            label=f"{KCLS[_cl]['short']}  ({len(_ii)})")
                    for _x, _c, _l, _h, _n in zip(_XC12, _cen, _lo, _hi, _n12):
                        ROW12.append(dict(block="profile", stage=_s12, quantity=_qn, kcls=_cl, x=float(_x),
                                          n_gal=int(_n), med=float(_c), lo=float(_l), hi=float(_h)))
                ax.set_xscale("log"); ax.axhline(_flo, color="0.6", lw=0.8, ls=":", zorder=1)
                if P12_XAXIS == "kpc":
                    ax.axvspan(_XC12[0] * 0.7, P12_RSOFT, color="0.85", alpha=0.5, lw=0, zorder=0)
                    ax.axvline(_RIN12, color="0.4", lw=1.0, ls="--", alpha=0.8, zorder=1)
                ax.grid(alpha=0.2, lw=0.5, which="both")
                if _qi == 0:
                    ax.set_title(f"({'abcdef'[_si]}) {_SLAB12[_s12]}", loc="left", fontsize=10)
                    _leg12(ax, loc="lower left", fontsize=7.5)
                if _si == 0:
                    ax.set_ylabel(r"$\log$ " + _ylab)
                if _qi == len(P12_QUANT) - 1:
                    ax.set_xlabel(r"$R$  [kpc]" if P12_XAXIS == "kpc" else r"$R / R_{50}^\star$")
        axs[0, -1].annotate(f"grey: below the {P12_RSOFT:g} kpc softening" + "\n" +
                            f"dashed: the centre of the hole metric ({_RIN12:g} kpc)" + "\n" +
                            "dotted: an empty bin",
                            xy=(0.98, 0.04), xycoords="axes fraction", ha="right", va="bottom", fontsize=7, color="0.35")
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the face-on radial profiles at the critical points"
                     + (f", $|z| < {P12_ZMAX:g}$ kpc" if P12_ZMAX else "")
                     + ("  [member gas only]" if P12_MEMBER else ""), y=0.995, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.97)); paper_save(fig, "kappa_m25_stage_profiles"); plt.show()

        # ── FIG 2: the hole — what it is, who has one, and when ──
        _AT12 = P12_HOLE_AT
        HOLE12 = {k_: ("hole" if v["hole"] >= P12_HOLE else "full")
                  for (k_, s_), v in H12.items() if s_ == _AT12 and v["ok"]}
        fig, axs = plt.subplots(1, 3, figsize=(13.4, 3.9))
        for _hc in HC:                                              # (a) what the split means: the two groups' profiles
            _ii = [H12[(k_, _AT12)]["i"] for k_ in HOLE12 if HOLE12[k_] == _hc]
            if len(_ii) < P12_NMIN:
                continue
            A12 = np.array([np.log10(np.maximum(_sig12(i, P12_HQ, "kpc"), 10.0 ** P12_QUANT[0][2])) for i in _ii], float)
            axs[0].plot(_XK12, np.nanmedian(A12, axis=0), color=HCLS[_hc]["color"], lw=HCLS[_hc]["lw"],
                        label=f"{HCLS[_hc]['short']}  ({len(_ii)})")
            axs[0].fill_between(_XK12, np.nanpercentile(A12, 16, axis=0), np.nanpercentile(A12, 84, axis=0),
                                color=HCLS[_hc]["color"], alpha=0.18, lw=0)
            for _x, _c in zip(_XK12, np.nanmedian(A12, axis=0)):
                ROW12.append(dict(block="hole_profile", stage=_AT12, quantity=P12_HQ, kcls=_hc, x=float(_x),
                                  n_gal=len(_ii), med=float(_c), lo=np.nan, hi=np.nan))
        axs[0].set_xscale("log"); axs[0].axvline(_RIN12, color="0.4", lw=1.0, ls="--")
        axs[0].axvspan(_XK12[0] * 0.7, P12_RSOFT, color="0.85", alpha=0.5, lw=0)
        axs[0].set_xlabel(r"$R$  [kpc]"); axs[0].set_ylabel(r"$\log\,\Sigma_{\rm " + P12_HQ + r"}$  [$M_\odot$ pc$^{-2}$]")
        axs[0].set_title(f"(a) the two groups at {_SLAB12[_AT12]}", loc="left", fontsize=10)
        _leg12(axs[0], loc="lower left"); axs[0].grid(alpha=0.2, lw=0.5, which="both")
        for _cl in KC:                                              # (b) who has one: the metric itself, by class
            _v = np.array([H12[(k_, _AT12)]["hole"] for k_ in HOLE12 if CLS9.get(k_) == _cl], float)
            if not len(_v):
                continue
            axs[1].plot(np.sort(_v), np.arange(1, len(_v) + 1) / len(_v), drawstyle="steps-post",
                        color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_v)})")
            axs[1].plot(np.median(_v), 0.5, "o", color=KCLS[_cl]["color"], ms=7, mec="k", mew=0.5)
        axs[1].axvline(P12_HOLE, color="0.3", lw=1.2, ls="--"); axs[1].set_ylim(0, 1.02)
        axs[1].annotate("hole  " + r"$\rightarrow$", xy=(P12_HOLE, 0.03), xycoords=("data", "axes fraction"),
                        xytext=(4, 0), textcoords="offset points", fontsize=8, color="0.3")
        axs[1].set_xlabel(rf"$\log\,\Sigma_{{\rm peak}} / \Sigma(<{_RIN12:g}\,$kpc$)$   [dex]")
        axs[1].set_ylabel("cumulative fraction of the class")
        axs[1].set_title(f"(b) the central depression at {_SLAB12[_AT12]}", loc="left", fontsize=10)
        _leg12(axs[1], loc="upper left"); axs[1].grid(alpha=0.2, lw=0.5)
        _xs12 = np.arange(_NS12)                                    # (c) when: the hole fraction stage by stage
        for _ci, _cl in enumerate(KC):
            _p, _l, _h, _nn = [], [], [], []
            for _s12 in P12_STAGES:
                _v = [v["hole"] for (k_, s_), v in H12.items() if s_ == _s12 and v["ok"] and CLS9.get(k_) == _cl]
                _w = wilson(sum(1 for x in _v if x >= P12_HOLE), len(_v)) if len(_v) >= P12_NMIN else (np.nan,) * 3
                _p.append(_w[0]); _l.append(_w[1]); _h.append(_w[2]); _nn.append(len(_v))
                ROW12.append(dict(block="hole_fraction", stage=_s12, quantity=P12_HQ, kcls=_cl, x=float(_xs12[len(_p) - 1]),
                                  n_gal=len(_v), med=_w[0], lo=_w[1], hi=_w[2]))
            _o = 0.06 * (_ci - 0.5)
            axs[2].errorbar(_xs12 + _o, _p, yerr=[np.array(_p) - np.array(_l), np.array(_h) - np.array(_p)],
                            fmt="o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=6, mec="k", mew=0.4, capsize=3,
                            label=KCLS[_cl]["short"])
            for _x, _y, _n in zip(_xs12 + _o, _p, _nn):
                if np.isfinite(_y):
                    axs[2].annotate(str(_n), xy=(_x, _y), xytext=(0, 7), textcoords="offset points", fontsize=7,
                                    ha="center", color=KCLS[_cl]["color"])
        axs[2].set_xticks(_xs12); axs[2].set_xticklabels([_SSHORT12[s] for s in P12_STAGES], fontsize=8.5)
        axs[2].set_ylim(0, 1.02); axs[2].set_ylabel("fraction of the class with a hole")
        axs[2].set_title("(c) when the hole is there", loc="left", fontsize=10)
        _leg12(axs[2], loc="upper left"); axs[2].grid(alpha=0.2, lw=0.5)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the central hole: the split, the two classes, and its epoch",
                     y=1.02, fontsize=12, fontweight="bold")
        fig.tight_layout(); paper_save(fig, "kappa_m25_hole"); plt.show()

        # ── FIG 3: what the hole does to the dust — the clocks, split by it ──
        _CLK12 = [("lfd", r"(a) $\log\,M_{\rm dust}/M_\star$", r"$\log\, M_{\rm dust}/M_\star$"),
                  ("ldtm", r"(c) the metals that are in grains", r"$\log\, M_{\rm dust}/(M_{\rm dust} + Z_{\rm gas}M_{\rm gas})$"),
                  ("lfh2", r"(d) $\log\,M_{\rm H_2}/M_\star$", r"$\log\, M_{\rm H_2}/M_\star$")]
        _TR12 = {c: {k: to_grid(CH9[k]["t"] - _tsft9.get(k, np.nan), CH9[k][c]) for k in HOLE12}
                 for c, _, _ in _CLK12}
        fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
        for _pi, (_c12, _ttl, _ylab) in enumerate(_CLK12):
            ax = axs[[0, 1, 1][_pi], [0, 0, 1][_pi]]
            for _hc in HC:
                _ks = [k for k in HOLE12 if HOLE12[k] == _hc]
                if len(_ks) < 2:
                    continue
                _cen, _lo, _hi, _n = med_band(np.array([_TR12[_c12][k] for k in _ks], float), nmin=P12_NMIN,
                                              nboot=P12_NBOOT)
                if P12_SPAG:
                    for _r12 in np.array([_TR12[_c12][k] for k in _ks], float):
                        ax.plot(TG, _r12, color=HCLS[_hc]["color"], lw=0.4,
                                alpha=float(np.clip(14.0 / max(len(_ks), 1), 0.06, 0.35)), zorder=2)
                ax.fill_between(TG, _lo, _hi, color=HCLS[_hc]["color"], alpha=0.20, lw=0, zorder=3)
                ax.plot(TG, _cen, color=HCLS[_hc]["color"], lw=HCLS[_hc]["lw"], zorder=5,
                        label=f"{HCLS[_hc]['short']}  ({len(_ks)})")
                for _x, _c, _l, _h, _n_ in zip(TG, _cen, _lo, _hi, _n):
                    ROW12.append(dict(block="clock", stage=_AT12, quantity=_c12, kcls=_hc, x=float(_x), n_gal=int(_n_),
                                      med=float(_c), lo=float(_l), hi=float(_h)))
            ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9, zorder=1)
            ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(_ylab); ax.set_title(_ttl, loc="left", fontsize=10)
            ax.grid(alpha=0.22, lw=0.5)
            if _pi:
                ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
        _leg12(axs[0, 0], loc="lower left", fontsize=8.5)
        ax = axs[0, 1]                                              # (b) the cross: is it the hole, or the rotation?
        for _hc in HC:
            for _cl in KC:
                _ks = [k for k in HOLE12 if HOLE12[k] == _hc and CLS9.get(k) == _cl]
                if len(_ks) < P12_NMIN:
                    continue
                _cen, _lo, _hi, _n = med_band(np.array([_TR12["lfd"][k] for k in _ks], float), nmin=P12_NMIN,
                                              nboot=P12_NBOOT)
                ax.plot(TG, _cen, color=HCLS[_hc]["color"], lw=2.2, ls="-" if _cl == "rot" else "--",
                        label=f"{HCLS[_hc]['short']}, {KCLS[_cl]['short']}  ({len(_ks)})")
                for _x, _c, _l, _h, _n_ in zip(TG, _cen, _lo, _hi, _n):
                    ROW12.append(dict(block="cross", stage=_AT12, quantity=f"lfd_{_cl}", kcls=_hc, x=float(_x),
                                      n_gal=int(_n_), med=float(_c), lo=float(_l), hi=float(_h)))
        ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9)
        ax.axhline(DUSTY_CUT, color="0.35", lw=1.0, ls=":")
        ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(r"$\log\, M_{\rm dust}/M_\star$")
        ax.set_title("(b) the hole $\\times$ the kinematic class", loc="left", fontsize=10)
        ax.grid(alpha=0.22, lw=0.5); _leg12(ax, loc="lower left", fontsize=7.5)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the dust clocks, split by whether the centre is empty at {_SLAB12[_AT12]}",
                     y=0.99, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_hole_clock"); plt.show()

        # ── what it says ──
        _nh = sum(1 for v in HOLE12.values() if v == "hole")
        print(f"\nthe hole at {_SLAB12[_AT12]}: {_nh}/{len(HOLE12)} galaxies "
              f"({100 * _nh / max(len(HOLE12), 1):.0f} %) have one; by class — "
              + ", ".join(f"{KCLS[c]['short']} {sum(1 for k in HOLE12 if HOLE12[k] == 'hole' and CLS9.get(k) == c)}"
                          f"/{sum(1 for k in HOLE12 if CLS9.get(k) == c)}" for c in KC))
        _hv = np.array([H12[(k, _AT12)]["hole"] for k in HOLE12], float)
        _hm = np.array([CLS9.get(k) == "rot" for k in HOLE12], bool)
        _a12 = mw_auc(_hv, _hm)
        print(f"  the depression itself: {KCLS['rot']['short']} {np.median(_hv[_hm]) if _hm.any() else np.nan:+.2f} dex vs "
              f"{np.median(_hv[~_hm]) if (~_hm).any() else np.nan:+.2f} dex"
              + (f", AUC {_a12[0]:.2f} (p {_a12[1]:.3f}, N {_a12[2]},{_a12[3]})" if np.isfinite(_a12[0])
                 else f"  (too few to score: N {_a12[2]},{_a12[3]})"))
        _nin12 = np.array([H12[(k, _AT12)]["n_in"] for k in HOLE12], float)
        _nex12 = np.array([H12[(k, _AT12)]["n_exp"] for k in HOLE12], float)
        _hh12 = np.array([HOLE12[k] == "hole" for k in HOLE12], bool)
        _f12 = lambda a: f"{np.nanmedian(a):.0f}" if len(a) and np.isfinite(a).any() else "—"
        print(f"  and it is not shot noise: inside {_RIN12:g} kpc the galaxies called HOLES hold {_f12(_nin12[_hh12])} gas "
              f"particles where the peak surface density would put {_f12(_nex12[_hh12])}; the filled centres hold "
              f"{_f12(_nin12[~_hh12])} against {_f12(_nex12[~_hh12])}")
        _rp = np.array([H12[(k, _AT12)]["r_peak"] for k in HOLE12], float)
        print(f"  the peak of the profile sits at {np.nanmedian(_rp[_hm]) if _hm.any() else np.nan:.2f} / "
              f"{np.nanmedian(_rp[~_hm]) if (~_hm).any() else np.nan:.2f} kpc "
              f"({KCLS['rot']['short']} / {KCLS['iso']['short']})")
        print(f"\nthe dust clocks split by the hole (median [N]; AUC = P(hole > filled) at that epoch):")
        for _c12, _, _ in _CLK12:
            _line = []
            for _m12 in P12_MARKS:
                _i12 = int(np.argmin(np.abs(TG - _m12)))
                _va = np.array([_TR12[_c12][k][_i12] for k in HOLE12 if HOLE12[k] == "hole"], float)
                _vb = np.array([_TR12[_c12][k][_i12] for k in HOLE12 if HOLE12[k] == "full"], float)
                _va, _vb = _va[np.isfinite(_va)], _vb[np.isfinite(_vb)]
                _au = mw_auc(np.concatenate([_va, _vb]),
                             np.concatenate([np.ones(len(_va), bool), np.zeros(len(_vb), bool)]))[0] \
                    if len(_va) and len(_vb) else np.nan
                _line.append(f"{_m12:+.0f}: " + (f"{np.median(_va):+.2f}" if len(_va) >= P12_NMIN else "—") + "/"
                             + (f"{np.median(_vb):+.2f}" if len(_vb) >= P12_NMIN else "—")
                             + (f" [{_au:.2f}]" if np.isfinite(_au) else "") + f" (N {len(_va)},{len(_vb)})")
                ROW12.append(dict(block="epoch", stage=_AT12, quantity=_c12, kcls="both", x=float(_m12),
                                  n_gal=len(_va) + len(_vb), med=float(np.median(_va)) if len(_va) else np.nan,
                                  lo=float(np.median(_vb)) if len(_vb) else np.nan, hi=float(_au)))
            print(f"  {_c12:5s} " + "   ".join(_line))
        print("  (each cell is  hole/filled centre [AUC] (N_hole, N_filled))")
        pd.DataFrame(ROW12).to_csv(os.path.join(OUT, "kappa_m25_hole.csv"), index=False)
        _G12 = []
        for _k12 in KEEP9:
            _r = dict(anchor_snap=_k12[0], gal_id=_k12[1], kcls=CLS9.get(_k12), hole_class=HOLE12.get(_k12),
                      t_sft=_tsft9.get(_k12, np.nan))
            for _s12 in P12_STAGES:
                _v = H12.get((_k12, _s12))
                _r[f"t_{_s12}"] = ST12[_k12][_s12]
                for _f in ("snap", "gx", "dt", "hole", "r_peak", "n_in", "n_exp"):
                    _r[f"{_f}_{_s12}"] = _v[_f] if _v else np.nan
            _G12.append(_r)
        pd.DataFrame(_G12).to_csv(os.path.join(OUT, "kappa_m25_hole_galaxies.csv"), index=False)
        print(f"\ntables -> {os.path.join(OUT, 'kappa_m25_hole.csv')} (the drawn points), "
              f"{os.path.join(OUT, 'kappa_m25_hole_galaxies.csv')} (one row per galaxy)")
        if not P12_ALLNODES:
            print("  set P12_ALLNODES = True to measure every node of every branch (the cache is incremental — only the "
                  "files not yet read are added) and the hole becomes a clock like Part 10's")
```

## Part 12b — the opening angle of the outflow: does the wind leave through the poles where the gas rotates?

Part 12 asked where the gas **sits**. This one asks where it **leaves**, and it is the mechanism test the rest of the
notebook has been circling. The two classes are identical for 3 Gyr before the truncation (Part 4b), they receive the same
AGN dose (Part 6 scored the dose as a divider and it failed), and they end 1–2 dex apart in dust. What differs can then only
be the **coupling** between the energy and the ISM, not the energy — and the geometric form of that claim is a statement
about *angles*, which nothing in the notebook has tested yet:

> if the wind is collimated about the gas's own rotation axis, a rotating galaxy hands the energy a low-density polar
> channel and keeps the plane it lives in; a galaxy whose gas has no plane offers no such channel, and the same energy
> intersects gas in every direction.

**The angle.** For every gas particle, $\mu = |\hat r \cdot \hat n|$ with $\hat n$ the angular-momentum axis of the *member
gas* (Part 12's frame): $\mu = 0$ in the disc plane, $\mu = 1$ along the pole. The bins are uniform in $\mu$ — and **solid
angle is uniform in $\mu$** as well ($d\Omega = 2\pi\,d\mu$ over a hemisphere) — so a profile divided by its own mean is
literally $dM/d\Omega$ in units of the isotropic expectation: 1 everywhere = isotropic, $>1$ = a polar excess. Nothing about
the shape of the galaxy enters that normalisation.

**The wind** is the gas with $v_r \ge$ `P12B_VFAST` inside `P12B_SHELL`, with $v_r = (\vec v_{\rm pec} - \vec v_{\rm sys})
\cdot \hat r + H(z)\,r$ — Part 10's definition verbatim, $\vec v_{\rm sys}$ the mass-weighted velocity of the member stars.
Measured gas-weighted **and** dust-weighted, as everywhere in Parts 10 and 13.

**The control is the reservoir itself, and it is not optional.** A disc trivially holds its gas at $\mu \sim 0$, so a wind
drawn from a disc *starts* planar whatever it then does: the raw angular profile of the wind cannot be read on its own. So
the statement of this part is the normalisation-free one,

$$f_{\rm esc}(\mu) \;=\; \frac{M(v_r \ge v_{\rm fast},\,\mu)}{M(\text{all gas},\,\mu)},
\qquad R_{\rm esc} \;=\; \frac{f_{\rm esc}(\mu > 0.5)}{f_{\rm esc}(\mu < 0.5)}$$

— the share of the gas *at that angle* that is leaving, and its ratio between the two halves of the sky. $R_{\rm esc} = 1$
means the gas leaves equally in every direction; $R_{\rm esc} > 1$ means it leaves through the poles. It needs no isotropic
reference and no assumption about the reservoir, which is why it is what the classes are compared on and what Fig 3 splits
the dust clocks by (`P12B_SPLITON`).

**And a frame-free measurement beside it**, because $\hat n$ is noisy exactly where this notebook says there is no disc —
the non-rotating class's wind would otherwise look isotropic by construction of that noise. The direction tensor
$T_{ij} = \sum_k w_k\,\hat r_i \hat r_j$ (the mass as the weight) needs no axis at all:

| the wind is | its eigenvalues |
|---|---|
| isotropic | $(1/3,\,1/3,\,1/3)$ |
| a collimated bipolar cone | $\to (1,\,0,\,0)$ |
| confined to a plane | $\to (1/2,\,1/2,\,0)$ |

The angle between its major axis and $\hat n$ is then the **alignment** (0° = the wind leaves along the rotation axis, 90° =
in the plane; a *random* axis gives a median of 60°, drawn on the figure as the null), and the mass-weighted median of
$|\cos|$ about the tensor's own axis is the wind's **half-opening angle** (60° for isotropic).

**What would falsify the picture:** if the two classes' winds have the same opening angle, the same alignment and the same
$R_{\rm esc}$, then the coupling is not geometric, and the difference has to be sought in the *density* of the ISM instead —
Part 4b's $\Sigma$ reading, where the classes do separate.

*Not a pure read.* It reads the same reduced particle files as Parts 10 and 12 and **caches** the angular census
(`P12B_CACHE`, an `.npz` keyed on the bins and on the wind definition — change `P12B_RB`, `P12B_MU`, `P12B_VE`,
`P12B_VFAST`, `P12B_SHELL` or `P12B_RMAX` and it rebuilds). By default only the stage nodes Part 12 already resolved are
measured, so it costs the same ~4 files per galaxy; the whole $(\mu \times r \times v_r)$ cube is stored, so every threshold
except the wind definition itself is recomputed at draw time. `P12B_ALLNODES = True` turns the opening angle into a clock.


```python
# ── Part 12b — THE OPENING ANGLE OF THE OUTFLOW: does the wind leave through the poles where the gas rotates? ──
# Part 12 asked where the gas SITS. This part asks where it LEAVES, and it is the mechanism test the rest of the notebook has
# been circling: the two classes are identical for 3 Gyr before the truncation (Part 4b), they receive the same AGN dose
# (Part 6 scored the dose as a divider and it failed), and they end 1-2 dex apart in dust. What differs can then only be the
# COUPLING between the energy and the ISM, not the energy — and the geometric form of that claim is measurable with the very
# files Parts 10 and 12 already read. If the wind is collimated about the gas's own rotation axis, a rotating galaxy offers
# the energy a low-density polar channel and keeps the plane it lives in; a galaxy whose gas has no plane offers no channel,
# and the same energy intersects gas in every direction. That is a prediction about ANGLES, and nothing else in the notebook
# has tested it.
#   THE ANGLE. For every gas particle, mu = |r_hat . n_hat|, with n_hat the angular-momentum axis of the MEMBER gas (Part 12's
#   frame): mu = 0 in the disc plane, mu = 1 along the pole. The bins are uniform in mu, and SOLID ANGLE IS UNIFORM IN mu
#   (dOmega = 2 pi dmu over a hemisphere), so a profile divided by its own mean is literally dM/dOmega in units of the
#   isotropic expectation — 1 everywhere = isotropic, > 1 = a polar excess, < 1 = a deficit. No assumption about the shape of
#   the galaxy enters that normalisation.
#   THE WIND. v_r = (v_pec - v_sys) . r_hat + H(z) r, Part 10's definition verbatim (v_sys = the mass-weighted velocity of the
#   MEMBER STARS), and the wind is the gas with v_r >= P12B_VFAST inside P12B_SHELL — outside the ISM, where material is on
#   its way out rather than turning in a disc. Measured gas-weighted AND dust-weighted, as everywhere in Parts 10 and 13.
#   THE CONTROL IS THE RESERVOIR ITSELF, and it is not optional. A disc trivially holds its gas at mu ~ 0, so a wind that is
#   drawn from a disc starts planar whatever it does afterwards: the raw angular profile of the wind CANNOT be read on its
#   own. Every panel therefore draws the same profile for ALL the gas beside it, and the statement of this part is the
#   NORMALISATION-FREE one between them —
#     f_esc(mu) = M(v_r >= VFAST, mu) / M(all gas, mu),
#   the share of the gas AT THAT ANGLE that is leaving. It needs no isotropic reference and no assumption about the shape of
#   the reservoir, and its ratio between the two halves of the sky,
#     R_esc = f_esc(mu > POL) / f_esc(mu < POL),
#   is one number per galaxy: 1 = the gas leaves equally in every direction, > 1 = it leaves preferentially through the
#   poles. That is the quantity the classes are compared on and the one Fig 3 splits the dust clocks by.
#   AND A FRAME-FREE MEASUREMENT BESIDE IT, because n_hat is noisy exactly where this notebook says there is no disc — the
#   non-rotating class's wind would look isotropic by construction of that noise. The direction tensor
#   T_ij = sum_k w_k r_hat_i r_hat_j (w = the mass, normalised) needs no axis at all: its eigenvalues are (1/3, 1/3, 1/3) for
#   an isotropic outflow, tend to (1, 0, 0) for a collimated bipolar one and to (1/2, 1/2, 0) for one confined to a plane.
#   The angle between its major axis and n_hat is then the ALIGNMENT (0 deg = the wind leaves along the rotation axis, 90 deg
#   = it leaves in the plane; a RANDOM axis gives a median of 60 deg, drawn on the figure), and the mass-weighted median of
#   |cos| about the tensor's own axis is the HALF-OPENING ANGLE of the wind.
# WHAT WOULD FALSIFY THE PICTURE: if the two classes' winds have the same opening angle and the same alignment, the coupling
# is not geometric and the difference must be sought in the density of the ISM instead (Part 4b's Sigma reading).
# NOT a pure read: it reads the same reduced particle files as Parts 10 and 12 and CACHES the angular census (P12B_CACHE, an
# .npz keyed on the bins and the wind definition — change P12B_RB / P12B_MU / P12B_VE / P12B_VFAST / P12B_SHELL / P12B_RMAX
# and it rebuilds). By default only the stage nodes Part 12 already resolved are measured, so it costs the same ~4 files per
# galaxy; the mu / radius / velocity cube is stored whole, so every threshold except the wind definition itself is
# recomputed at draw time.
P12B_CACHE  = os.path.join(OUT, f"kappa_{P9_BOX}_outflow_angles.npz")
P12B_STAGES = P12_STAGES         # the same four critical points, off the same chains
P12B_AT     = "xray_on"          # the stage the collimation split of Fig 3 is measured at
P12B_RB     = np.array([0.5, 3.0, 10.0, 30.0, 100.0])    # the radial shell edges [physical kpc]
P12B_MU     = np.linspace(0.0, 1.0, 11)                  # |cos(theta)| bin edges — UNIFORM IN SOLID ANGLE
P12B_VE     = np.array([0.0, 50.0, 100.0, 300.0])        # the INNER edges of the v_r classes [km/s] (5 classes with them)
P12B_VFAST  = 100.0              # "the wind": v_r at or above this [km/s] — must be one of P12B_VE
P12B_SHELL  = (3.0, 100.0)       # the radial range the wind is measured in [kpc] — must fall on P12B_RB edges
P12B_RMAX   = 100.0              # the aperture the files were built with [kpc]
P12B_POL    = 0.5                # mu above this = the polar half of the sky (exactly half the solid angle, so isotropic = 0.5)
P12B_FPOL   = 0.5                # the isotropic value of f_pol, and its cut if P12B_SPLITON is "fpol"
P12B_SPLITON = "resc"            # WHAT FIG 3 SPLITS ON: "resc" = the polar/plane ratio of the escaping fraction (the
                                 # reservoir-free statistic) | "fpol" = the raw polar mass share of the wind | "dfpol" =
                                 # the wind's polar share minus the reservoir's (the excess over what was available)
P12B_SCUT   = None               # the cut on it (None = its own isotropic value: 1.0 / P12B_FPOL / 0.0)
P12B_FESC_FLOOR = 1e-3           # the log floor an EMPTY escaping-fraction bin is drawn at (no wind at that angle at all)
P12B_HUBFLOW, P12B_OM = True, 0.3        # the expansion term of v_r, as in Part 10
P12B_NLAUNCH = 256               # SIMBA ejects its AGN winds along +-L of the gas elements NEAREST THE BLACK HOLE — the
                                 # 256 closest, typically the inner ~kpc (Dave+19) — so the LAUNCH-AXIS PROXY L_in is
                                 # measured over exactly that many nearest gas particles, NOT inside a fixed radius: where
                                 # Part 12's central hole has emptied the centre, those 256 sit far out, and the radius
                                 # they reach (r_launch) is recorded with the axis. Changing it rebuilds the census.
P12B_NLMIN  = 20                 # member gas particles the angular-momentum axis needs
P12B_NFAST  = 20                 # wind particles a node needs before its angles are measured at all
P12B_NMIN   = 3                  # galaxies a drawn point needs
P12B_NBOOT  = 300
P12B_SPAG   = True               # the individual galaxies behind the medians
P12B_MARKS  = P12_MARKS
P12B_ALLNODES = False            # measure every node of every branch instead of only the stage nodes (incremental cache)
_SPLAB12B = {"resc": (r"$R_{\rm esc}$", 1.0), "fpol": (r"$f_{\rm pol}$", P12B_FPOL),
             "dfpol": (r"$\Delta f_{\rm pol}$", 0.0)}
_SVAR12B, _SDEF12B = _SPLAB12B.get(P12B_SPLITON, _SPLAB12B["resc"])
_SCUT12B = float(_SDEF12B if P12B_SCUT is None else P12B_SCUT)
CCLS = {"pol": dict(color="#e66101", lw=2.6, ls="-", short="polar escape",
                    label=rf"polar escape  ({_SVAR12B} $\geq {_SCUT12B:g}$)"),
        "brd": dict(color="#018571", lw=2.4, ls="-", short="escape in all directions",
                    label=rf"no preferred direction  ({_SVAR12B} $< {_SCUT12B:g}$)")}
CC = list(CCLS)

if "NOD12" not in globals():
    print("Part 12b: needs Part 12 — run it first (the stages, the nodes and the particle-file inventory come from it).")
else:
    import time as _time12b

    _NMU12B, _NSH12B, _NV12B = len(P12B_MU) - 1, len(P12B_RB) - 1, len(P12B_VE) + 1
    _XMU12B = 0.5 * (P12B_MU[1:] + P12B_MU[:-1])                       # the bin centres of the drawn axis
    _VLO12B = np.concatenate([[-np.inf], P12B_VE])                     # the lower edge of each v_r class
    _KMS12B = 1.0227121650537077e-9                                    # (Msun km/s / kpc) -> Msun/yr, as in Part 10

    # ── (A) THE ANGULAR CENSUS, cached: one (mu x radius x v_r) cube per node, gas- and dust-weighted ──
    def _ang12b(path):
        """ONE reduced particle file -> the angular cube of its gas about the member gas's rotation axis, plus the
        frame-free direction tensor of the wind (eigenvalues, its axis against that rotation axis, its opening angle)."""
        with h5py.File(path, "r") as h:
            A = dict(h.attrs); G, St = h["gas"], h["star"]
            gp = G["pos"][:].astype(float); gmem = G["member"][:].astype(bool)
            gm = G["m_gas"][:].astype(float); gd = G["m_dust"][:].astype(float)
            gv = G["vel"][:].astype(float) if "vel" in G else np.full((len(gm), 3), np.nan)
            sp = St["pos"][:].astype(float); sm = St["m_star"][:].astype(float)
            smem = St["member"][:].astype(bool)
            sv = St["vel"][:].astype(float) if "vel" in St else np.full((len(sm), 3), np.nan)
        z, hub = float(A["redshift"]), float(A.get("hub", 0.68))
        _ev = np.asarray(A.get("evecs", np.eye(3)), float)             # the STELLAR principal frame; column 2 = the minor axis
        _minor = _ev[:, 2] if _ev.shape == (3, 3) else np.array([0.0, 0.0, 1.0])
        r = np.linalg.norm(gp, axis=1)
        _ok = smem & np.isfinite(sv).all(1) & (sm > 0)                 # THE REST FRAME: the member stars (Part 10 verbatim)
        if _ok.sum() >= 10:
            vsys = np.average(sv[_ok], axis=0, weights=sm[_ok])
        else:
            _ok = gmem & np.isfinite(gv).all(1) & (gm > 0)
            vsys = np.average(gv[_ok], axis=0, weights=gm[_ok]) if _ok.sum() >= 10 else np.zeros(3)
        vr = ((gv - vsys) * (gp / np.clip(r, 1e-9, None)[:, None])).sum(1)
        if P12B_HUBFLOW:
            vr = vr + (100.0 * hub * np.sqrt(P12B_OM * (1 + z) ** 3 + 1 - P12B_OM)) * r / 1e3
        _g = gmem & np.isfinite(gv).all(1) & (gm > 0) & (r > 0)        # THE AXIS: the member gas's angular momentum
        nrm, frm, coh = None, 3, np.nan
        if _g.sum() >= P12B_NLMIN:
            _j = np.cross(gp[_g], (gv[_g] - vsys))
            _L = (_j * gm[_g, None]).sum(0)
            _jsum = float(np.sum(gm[_g] * np.linalg.norm(_j, axis=1)))
            if np.isfinite(_L).all() and np.linalg.norm(_L) > 0:
                nrm, frm = _L / np.linalg.norm(_L), 0
                coh = float(np.linalg.norm(_L) / _jsum) if _jsum > 0 else np.nan   # ordered / total specific j
        if nrm is None:                                                # the stars' own angular momentum, then their minor axis
            _s = smem & np.isfinite(sv).all(1) & (sm > 0)
            _Ls = np.cross(sp[_s], (sv[_s] - vsys) * sm[_s, None]).sum(0) if _s.sum() >= 20 else np.full(3, np.nan)
            if np.isfinite(_Ls).all() and np.linalg.norm(_Ls) > 0:
                nrm, frm = _Ls / np.linalg.norm(_Ls), 1
            elif np.linalg.norm(_minor) > 0:
                nrm, frm = _minor / np.linalg.norm(_minor), 2
        # THE LAUNCH-AXIS PROXY, built the way the code builds it: the N nearest GAS elements to the centre, member or
        # not (the BH kernel does not know about caesar membership), and the radius they reach.
        _lin, _cohin, _nin, _rlau = np.full(3, np.nan), np.nan, 0.0, np.nan
        _cand = np.where(np.isfinite(gv).all(1) & (gm > 0) & (r > 0))[0]
        if len(_cand) >= 10:
            _sel = _cand[np.argsort(r[_cand])[:int(P12B_NLAUNCH)]]
            _nin, _rlau = float(len(_sel)), float(r[_sel].max())
            _ji = np.cross(gp[_sel], (gv[_sel] - vsys))
            _Li = (_ji * gm[_sel, None]).sum(0)
            _jsi = float(np.sum(gm[_sel] * np.linalg.norm(_ji, axis=1)))
            if np.isfinite(_Li).all() and np.linalg.norm(_Li) > 0:
                _lin = _Li / np.linalg.norm(_Li)
                _cohin = float(np.linalg.norm(_Li) / _jsi) if _jsi > 0 else np.nan
        row = {"snap": int(A["snap"]), "gx": int(A["gx"]), "z": z, "frame": float(frm), "coh": coh,
               "coh_in": _cohin, "n_gas_in": _nin, "r_launch": _rlau,
               "linx": float(_lin[0]), "liny": float(_lin[1]), "linz": float(_lin[2]),
               "n_memgas": float(gmem.sum()), "m_star_mem": float(np.nansum(sm[smem])),
               "mgas_ism": float(np.nansum(gm[gmem])), "mdust_ism": float(np.nansum(gd[gmem])),
               "misalign": (float(np.degrees(np.arccos(np.clip(abs(nrm @ _minor), 0, 1)))) if nrm is not None else np.nan)}
        _rh = np.nan
        if gmem.sum() >= 5 and np.nansum(gm[gmem]) > 0:
            _o = np.argsort(r[gmem])
            _rh = float(np.interp(0.5, np.cumsum(gm[gmem][_o]) / np.nansum(gm[gmem]), r[gmem][_o]))
        row["r50_gasmem"] = _rh
        _C = {k: np.zeros((_NMU12B, _NSH12B, _NV12B)) for k in ("g", "d", "n", "mvg", "mvd")}
        for _k in ("lam1", "lam2", "lam3", "align", "open_deg", "n_fast", "m_fast_g", "m_fast_d",
                   "mdot_g", "mdot_d", "vr_fast"):
            row[_k] = np.nan
        row["nx"], row["ny"], row["nz"] = ((float(nrm[0]), float(nrm[1]), float(nrm[2])) if nrm is not None
                                          else (np.nan, np.nan, np.nan))
        if nrm is None:
            return row, _C
        mu = np.abs((gp @ nrm) / np.clip(r, 1e-9, None))
        _in = np.isfinite(mu) & np.isfinite(vr) & (r > 0) & (r <= P12B_RMAX)
        _vc = np.digitize(np.where(_in, vr, 0.0), P12B_VE)             # 0 = infall, then the P12B_VE classes
        for _iv in range(_NV12B):
            _m = _in & (_vc == _iv)
            if not _m.any():
                continue
            _w = {"g": np.nan_to_num(gm[_m]), "d": np.nan_to_num(gd[_m]), "n": np.ones(int(_m.sum())),
                  "mvg": np.nan_to_num(gm[_m]) * np.clip(vr[_m], 0, None),
                  "mvd": np.nan_to_num(gd[_m]) * np.clip(vr[_m], 0, None)}
            for _k, _ww in _w.items():
                _C[_k][:, :, _iv] = np.histogram2d(mu[_m], r[_m], bins=(P12B_MU, P12B_RB), weights=_ww)[0]
        # THE WIND, and its frame-free geometry
        _fast = _in & (vr >= P12B_VFAST) & (r >= P12B_SHELL[0]) & (r <= P12B_SHELL[1]) & np.isfinite(gm) & (gm > 0)
        row["n_fast"] = float(_fast.sum())
        if _fast.sum() >= P12B_NFAST:
            _u = gp[_fast] / np.clip(r[_fast], 1e-9, None)[:, None]
            _w = gm[_fast] / float(np.sum(gm[_fast]))
            _T = np.einsum("k,ki,kj->ij", _w, _u, _u)
            _ev12b, _evec = np.linalg.eigh(_T)                         # ascending
            _ax = _evec[:, -1]
            _muo = np.abs(_u @ _ax)
            _o = np.argsort(_muo); _cw = (np.cumsum(_w[_o]) - 0.5 * _w[_o]) / max(_w.sum(), 1e-30)
            row["lam1"], row["lam2"], row["lam3"] = float(_ev12b[2]), float(_ev12b[1]), float(_ev12b[0])
            row["align"] = float(np.degrees(np.arccos(np.clip(abs(_ax @ nrm), 0, 1))))
            row["open_deg"] = float(np.degrees(np.arccos(np.clip(np.interp(0.5, _cw, _muo[_o]), 0, 1))))
            row["m_fast_g"] = float(np.nansum(gm[_fast])); row["m_fast_d"] = float(np.nansum(gd[_fast]))
            row["vr_fast"] = float(np.average(vr[_fast], weights=gm[_fast]))
            _dr = float(P12B_SHELL[1] - P12B_SHELL[0])
            row["mdot_g"] = float(np.sum(gm[_fast] * vr[_fast]) / _dr * _KMS12B)
            row["mdot_d"] = float(np.sum(np.nan_to_num(gd[_fast]) * vr[_fast]) / _dr * _KMS12B)
        return row, _C

    _SCA12B = ["snap", "gx", "z", "frame", "coh", "coh_in", "n_gas_in", "r_launch", "n_memgas", "m_star_mem", "mgas_ism",
               "mdust_ism", "misalign", "r50_gasmem", "lam1", "lam2", "lam3", "align", "open_deg", "n_fast",
               "m_fast_g", "m_fast_d", "mdot_g", "mdot_d", "vr_fast",
               "nx", "ny", "nz", "linx", "liny", "linz"]
    _need12b = sorted({(v[0], v[1]) for v in NOD12.values()})
    if P12B_ALLNODES:
        _kk12b = set(KEEP9)
        _need12b = sorted({(int(s_), int(x_)) for a_, g_, s_, x_ in
                           zip(BR9["anchor_snap"], BR9["gal_id"], BR9["snap"], BR9["gx"])
                           if (int(a_), int(g_)) in _kk12b and (int(s_), int(x_)) in _inv12})
    P12B = {}
    if os.path.exists(P12B_CACHE):
        _z12b = np.load(P12B_CACHE, allow_pickle=False)
        _same = (len(_z12b["mu"]) == len(P12B_MU) and np.allclose(_z12b["mu"], P12B_MU)
                 and len(_z12b["rb"]) == len(P12B_RB) and np.allclose(_z12b["rb"], P12B_RB)
                 and len(_z12b["ve"]) == len(P12B_VE) and np.allclose(_z12b["ve"], P12B_VE)
                 and float(_z12b["vfast"]) == float(P12B_VFAST) and float(_z12b["rmax"]) == float(P12B_RMAX)
                 and np.allclose(_z12b["shell"], np.asarray(P12B_SHELL, float))
                 and bool(_z12b["hubflow"]) == bool(P12B_HUBFLOW)
                 and "nlaunch" in _z12b.files and int(_z12b["nlaunch"]) == int(P12B_NLAUNCH))
        if _same:
            P12B = {k: _z12b[k] for k in _z12b.files
                    if k not in ("mu", "rb", "ve", "vfast", "rmax", "shell", "hubflow", "nlaunch")}
            print(f"  the angular census, cached -> {P12B_CACHE}  ({len(P12B['snap'])} files already measured)")
        else:
            print(f"  {P12B_CACHE} was built with different bins / a different wind definition — it is REBUILT")
    _done12b = set(zip(P12B["snap"].astype(int), P12B["gx"].astype(int))) if P12B else set()
    _todo12b = [k for k in _need12b if k not in _done12b]
    print(f"Part 12b — the wind is $v_r \\geq$ {P12B_VFAST:g} km/s in {P12B_SHELL[0]:g}-{P12B_SHELL[1]:g} kpc; "
          f"{len(_need12b)} nodes needed, {len(_todo12b)} to measure"
          + ("  (P12B_ALLNODES: every node)" if P12B_ALLNODES else ""))
    if _todo12b:
        _rows12b, _cub12b, _t012b, _bad12b = [], {k: [] for k in ("g", "d", "n", "mvg", "mvd")}, _time12b.time(), []
        for _i12b, _k12b in enumerate(_todo12b):
            try:
                _r12b, _c12b = _ang12b(_inv12[_k12b])
                _rows12b.append(_r12b)
                for _k in _cub12b:
                    _cub12b[_k].append(_c12b[_k])
            except Exception as _e12b:
                _bad12b.append(f"{os.path.basename(_inv12[_k12b])}: {type(_e12b).__name__}: {_e12b}")
            if (_i12b + 1) % 100 == 0:
                _el = _time12b.time() - _t012b
                print(f"     {_i12b + 1}/{len(_todo12b)} files, {_el:.0f} s "
                      f"(~{_el / (_i12b + 1) * (len(_todo12b) - _i12b - 1):.0f} s left)", flush=True)
        for _b12b in _bad12b[:5]:
            print(f"     file not read, {_b12b}")
        if _rows12b:
            _new = {k: np.array([r[k] for r in _rows12b], float) for k in _SCA12B}
            for _k, _v in _cub12b.items():
                _new[f"cube_{_k}"] = np.asarray(_v, float)
            P12B = {k: np.concatenate([P12B[k], _new[k]]) for k in _new} if P12B else _new
            np.savez_compressed(P12B_CACHE, mu=P12B_MU, rb=P12B_RB, ve=P12B_VE, vfast=float(P12B_VFAST),
                                rmax=float(P12B_RMAX), shell=np.asarray(P12B_SHELL, float),
                                hubflow=bool(P12B_HUBFLOW), nlaunch=int(P12B_NLAUNCH), **P12B)
            print(f"  the angular census -> {P12B_CACHE}  ({len(P12B['snap'])} files, {_time12b.time() - _t012b:.0f} s)")
    if not P12B:
        print("  nothing measured — Part 12b stops here.")
    else:
        IDX12B = {(int(s_), int(x_)): i for i, (s_, x_) in enumerate(zip(P12B["snap"], P12B["gx"]))}
        _ISH12B = [i for i in range(_NSH12B)
                   if P12B_RB[i] >= P12B_SHELL[0] - 1e-9 and P12B_RB[i + 1] <= P12B_SHELL[1] + 1e-9]
        _IVF12B = [i for i in range(_NV12B) if _VLO12B[i] >= P12B_VFAST - 1e-9]
        _IPOL12B = _XMU12B > P12B_POL

        def _prof12b(i, w="g", fast=True, shells=None):
            """Node i -> dM/dOmega against mu IN UNITS OF THE ISOTROPIC EXPECTATION (1 = isotropic everywhere)."""
            C = np.asarray(P12B[f"cube_{w}"][i], float)
            _sh = _ISH12B if shells is None else list(shells)
            _iv = _IVF12B if fast else list(range(_NV12B))
            p = C[:, _sh, :][:, :, _iv].sum(axis=(1, 2))
            _t = float(np.nansum(p))
            return p / _t * _NMU12B if _t > 0 else np.full(_NMU12B, np.nan)

        def _fpol12b(i, w="g", fast=True):
            """The share of the mass in the POLAR half of the sky — 0.5 for an isotropic distribution, by construction."""
            p = _prof12b(i, w, fast)
            return float(np.nansum(p[_IPOL12B]) / _NMU12B) if np.isfinite(p).any() else np.nan

        def _esc12b(i, w="g"):
            """Node i -> f_esc(mu): the share of the gas mass AT EACH ANGLE that is moving out faster than P12B_VFAST.
            The reservoir divides out, so nothing here depends on the shape of the galaxy it came from."""
            C = np.asarray(P12B[f"cube_{w}"][i], float)[:, _ISH12B, :]
            num, den = C[:, :, _IVF12B].sum(axis=(1, 2)), C.sum(axis=(1, 2))
            return np.where(den > 0, num / np.where(den > 0, den, np.nan), np.nan)

        def _resc12b(i, w="g"):
            """The polar / in-plane ratio of that fraction: 1 = the gas leaves equally in every direction, > 1 = through
            the poles. One number per node, and the reservoir-free version of f_pol."""
            C = np.asarray(P12B[f"cube_{w}"][i], float)[:, _ISH12B, :]
            out = []
            for _m in (_IPOL12B, ~_IPOL12B):
                _n, _d = C[_m][:, :, _IVF12B].sum(), C[_m].sum()
                out.append(_n / _d if _d > 0 else np.nan)
            return float(out[0] / out[1]) if (np.isfinite(out).all() and out[1] > 0) else np.nan

        def _band12b(A):
            """A (n_gal, n_mu) stack -> the median profile, its bootstrap 16-84 band and the per-bin count."""
            A = np.asarray(A, float)
            n = np.isfinite(A).sum(axis=0)
            if not len(A):
                return (np.full(_NMU12B, np.nan),) * 3 + (n,)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", RuntimeWarning)
                cen = np.nanmedian(A, axis=0)
                _rng = np.random.default_rng(23)
                bs = np.array([np.nanmedian(A[_rng.integers(0, len(A), len(A))], axis=0) for _ in range(P12B_NBOOT)])
                lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
            _ok = n >= P12B_NMIN
            return np.where(_ok, cen, np.nan), np.where(_ok, lo, np.nan), np.where(_ok, hi, np.nan), n

        def _leg12b(ax, **kw):
            if ax.get_legend_handles_labels()[0]:
                ax.legend(**{"frameon": False, "fontsize": 8, **kw})

        # the nodes of each (galaxy, stage), and which of them have a measurable wind
        W12B = {}
        for (_k12b, _s12b), (_sn, _gx, _off) in NOD12.items():
            _i = IDX12B.get((_sn, _gx))
            if _i is None:
                continue
            W12B[(_k12b, _s12b)] = dict(i=_i, snap=_sn, gx=_gx, dt=_off, n_fast=float(P12B["n_fast"][_i]),
                                        align=float(P12B["align"][_i]), open_deg=float(P12B["open_deg"][_i]),
                                        lam1=float(P12B["lam1"][_i]), coh=float(P12B["coh"][_i]),
                                        coh_in=float(P12B["coh_in"][_i]), n_gas_in=float(P12B["n_gas_in"][_i]),
                                        r_launch=float(P12B["r_launch"][_i]),
                                        fpol=_fpol12b(_i, "g", True), fpol_d=_fpol12b(_i, "d", True),
                                        fpol_all=_fpol12b(_i, "g", False), resc=_resc12b(_i, "g"),
                                        resc_d=_resc12b(_i, "d"), mdot_g=float(P12B["mdot_g"][_i]),
                                        mdot_d=float(P12B["mdot_d"][_i]), vr_fast=float(P12B["vr_fast"][_i]),
                                        dfpol=_fpol12b(_i, "g", True) - _fpol12b(_i, "g", False),
                                        ok=bool(np.isfinite(P12B["n_fast"][_i]) and P12B["n_fast"][_i] >= P12B_NFAST))
        print(f"  a wind of at least {P12B_NFAST:g} particles at: "
              + ", ".join(f"{s} {sum(1 for (k_, s_), v in W12B.items() if s_ == s and v['ok'])}/{_cov12[s]}"
                          for s in P12B_STAGES)
              + f";  the rotation axis is the member gas's own in {int((np.asarray(P12B['frame']) == 0).sum())} of the "
                f"{len(P12B['frame'])} nodes (L-coherence {np.nanmedian(P12B['coh']):.2f} in the median)")

        # ── FIG 1: the angular profiles — the reservoir, then the wind, gas- and dust-weighted ──
        _ROWS12B = [("all", "g", False, r"ALL the gas  $dM/d\Omega$"),
                    ("wind_g", "g", True, r"the wind  $dM/d\Omega$"),
                    ("fesc_g", "g", "esc", r"$f_{\rm esc}$: the share LEAVING")]
        ROW12B = []
        fig, axs = plt.subplots(len(_ROWS12B), len(P12B_STAGES), figsize=(3.35 * len(P12B_STAGES), 3.0 * len(_ROWS12B)),
                                sharex=True, sharey="row", squeeze=False)
        for _ri, (_rn, _wn, _fast, _ylab) in enumerate(_ROWS12B):
            for _si, _s12b in enumerate(P12B_STAGES):
                ax = axs[_ri, _si]
                for _cl in KC:
                    _ii = [v["i"] for (k_, s_), v in W12B.items()
                           if s_ == _s12b and CLS9.get(k_) == _cl and (v["ok"] or _fast is False)]
                    if not _ii:
                        continue
                    A12b = (np.array([np.maximum(_esc12b(i, _wn), P12B_FESC_FLOOR) for i in _ii], float)
                            if _fast == "esc" else np.array([_prof12b(i, _wn, _fast) for i in _ii], float))
                    _cen, _lo, _hi, _n = _band12b(A12b)
                    if P12B_SPAG:
                        for _r12b in A12b:
                            ax.plot(_XMU12B, _r12b, color=KCLS[_cl]["color"], lw=0.5,
                                    alpha=float(np.clip(12.0 / max(len(A12b), 1), 0.06, 0.35)), zorder=2)
                    ax.fill_between(_XMU12B, _lo, _hi, color=KCLS[_cl]["color"], alpha=0.22, lw=0, zorder=3)
                    ax.plot(_XMU12B, _cen, color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], zorder=5,
                            label=f"{KCLS[_cl]['short']}  ({len(_ii)})")
                    for _x, _c, _l, _h, _nn in zip(_XMU12B, _cen, _lo, _hi, _n):
                        ROW12B.append(dict(block=f"profile_{_rn}", stage=_s12b, kcls=_cl, x=float(_x), n_gal=int(_nn),
                                           med=float(_c), lo=float(_l), hi=float(_h)))
                if _fast != "esc":
                    ax.axhline(1.0, color="0.4", lw=1.0, ls="--", zorder=1)
                else:
                    ax.axhline(P12B_FESC_FLOOR, color="0.6", lw=0.8, ls=":", zorder=1)
                ax.axvline(P12B_POL, color="0.75", lw=0.9, ls=":", zorder=1)
                ax.set_yscale("log"); ax.grid(alpha=0.2, lw=0.5, which="both")
                if _ri == 0:
                    ax.set_title(f"({'abcdef'[_si]}) {_SLAB12[_s12b]}", loc="left", fontsize=10)
                    _leg12b(ax, loc="upper left", fontsize=7.5)
                if _si == 0:
                    ax.set_ylabel(_ylab + ("\n[$v_r \\geq$ %g km/s, %g-%g kpc]" % (P12B_VFAST, *P12B_SHELL)
                                           if _fast == "esc" else "\n[isotropic $= 1$]"))
                if _ri == len(_ROWS12B) - 1:
                    ax.set_xlabel(r"$\mu = |\cos\theta|$    (0 = plane, 1 = pole)")
        axs[0, -1].annotate("dashed: isotropic\ndotted: the polar half\nof the sky", xy=(0.97, 0.06),
                            xycoords="axes fraction", ha="right", va="bottom", fontsize=7, color="0.35")
        axs[-1, -1].annotate("row 3 divides row 2 by row 1:\nno reservoir left in it", xy=(0.98, 0.05),
                             xycoords="axes fraction", ha="right", va="bottom", fontsize=7, color="0.35")
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — where the gas is, where the wind goes, and what share of it leaves",
                     y=0.995, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.97)); paper_save(fig, "kappa_m25_outflow_angles"); plt.show()

        # ── FIG 2: the collimation — how polar, how aligned, how open ──
        fig, _AX12B = plt.subplots(2, 2, figsize=(11.0, 7.6))
        axs = _AX12B.ravel()
        _xs12b = np.arange(len(P12B_STAGES))
        for _ci, _cl in enumerate(KC):                                  # (a) the polar share, wind against reservoir
            for _fast, _ls, _mk, _lab in ((True, "-", "o", "the wind"), (False, "--", "s", "all the gas")):
                _p, _l, _h, _nn = [], [], [], []
                for _s12b in P12B_STAGES:
                    _v = np.array([v["fpol"] if _fast else v["fpol_all"] for (k_, s_), v in W12B.items()
                                   if s_ == _s12b and CLS9.get(k_) == _cl and (v["ok"] or not _fast)], float)
                    _v = _v[np.isfinite(_v)]
                    if len(_v) < P12B_NMIN:
                        _p.append(np.nan); _l.append(np.nan); _h.append(np.nan); _nn.append(len(_v)); continue
                    _rng = np.random.default_rng(5)
                    _bs = np.median(_v[_rng.integers(0, len(_v), (P12B_NBOOT, len(_v)))], axis=1)
                    _p.append(float(np.median(_v))); _l.append(float(np.percentile(_bs, 16)))
                    _h.append(float(np.percentile(_bs, 84))); _nn.append(len(_v))
                    ROW12B.append(dict(block="fpol" if _fast else "fpol_all", stage=_s12b, kcls=_cl,
                                       x=float(_xs12b[len(_p) - 1]), n_gal=len(_v), med=_p[-1], lo=_l[-1], hi=_h[-1]))
                _o = 0.06 * (_ci - 0.5)
                axs[0].errorbar(_xs12b + _o, _p, yerr=[np.array(_p) - np.array(_l), np.array(_h) - np.array(_p)],
                                fmt=_mk + _ls, color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"] if _fast else 1.4,
                                ms=6 if _fast else 4.5, mec="k", mew=0.4, capsize=3, alpha=1.0 if _fast else 0.65,
                                mfc=KCLS[_cl]["color"] if _fast else "white",
                                label=f"{KCLS[_cl]['short']}, {_lab}")
        axs[0].axhline(0.5, color="0.3", lw=1.2, ls=":")
        axs[0].annotate("isotropic", xy=(0.02, 0.5), xycoords=("axes fraction", "data"), xytext=(0, 4),
                        textcoords="offset points", fontsize=8, color="0.3")
        axs[0].set_xticks(_xs12b); axs[0].set_xticklabels([_SSHORT12[s] for s in P12B_STAGES], fontsize=8.5)
        axs[0].set_ylabel(rf"$f_{{\rm pol}}$: the mass share at $\mu > {P12B_POL:g}$")
        axs[0].set_title("(a) how polar the wind is", loc="left", fontsize=10)
        _leg12b(axs[0], loc="best", fontsize=7.5); axs[0].grid(alpha=0.2, lw=0.5)
        _th12b = np.linspace(0, 90, 181)                                # (b) the alignment against a random axis
        axs[1].plot(_th12b, 1 - np.cos(np.radians(_th12b)), color="0.6", lw=1.2, ls=":", label="a random axis")
        for _cl in KC:
            _v = np.array([v["align"] for (k_, s_), v in W12B.items()
                           if s_ == P12B_AT and v["ok"] and CLS9.get(k_) == _cl], float)
            _v = _v[np.isfinite(_v)]
            if not len(_v):
                continue
            axs[1].plot(np.sort(_v), np.arange(1, len(_v) + 1) / len(_v), drawstyle="steps-post",
                        color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_v)})")
            axs[1].plot(np.median(_v), 0.5, "o", color=KCLS[_cl]["color"], ms=7, mec="k", mew=0.5)
            ROW12B.append(dict(block="align", stage=P12B_AT, kcls=_cl, x=float(np.median(_v)), n_gal=len(_v),
                               med=float(np.median(_v)), lo=np.nan, hi=np.nan))
        axs[1].set_xlim(0, 90); axs[1].set_ylim(0, 1.02); axs[1].set_xticks([0, 15, 30, 45, 60, 75, 90])
        axs[1].set_xlabel("angle between the wind axis and the gas rotation axis  [deg]")
        axs[1].set_ylabel("cumulative fraction of the class")
        axs[1].set_title(f"(b) is it aligned, at {_SLAB12[P12B_AT]}?", loc="left", fontsize=10)
        _leg12b(axs[1], loc="lower right"); axs[1].grid(alpha=0.2, lw=0.5)
        for _ci, _cl in enumerate(KC):                                  # (c) the opening angle
            _p, _l, _h, _nn = [], [], [], []
            for _s12b in P12B_STAGES:
                _v = np.array([v["open_deg"] for (k_, s_), v in W12B.items()
                               if s_ == _s12b and v["ok"] and CLS9.get(k_) == _cl], float)
                _v = _v[np.isfinite(_v)]
                if len(_v) < P12B_NMIN:
                    _p.append(np.nan); _l.append(np.nan); _h.append(np.nan); _nn.append(len(_v)); continue
                _rng = np.random.default_rng(9)
                _bs = np.median(_v[_rng.integers(0, len(_v), (P12B_NBOOT, len(_v)))], axis=1)
                _p.append(float(np.median(_v))); _l.append(float(np.percentile(_bs, 16)))
                _h.append(float(np.percentile(_bs, 84))); _nn.append(len(_v))
                ROW12B.append(dict(block="open", stage=_s12b, kcls=_cl, x=float(_xs12b[len(_p) - 1]), n_gal=len(_v),
                                   med=_p[-1], lo=_l[-1], hi=_h[-1]))
            _o = 0.06 * (_ci - 0.5)
            axs[2].errorbar(_xs12b + _o, _p, yerr=[np.array(_p) - np.array(_l), np.array(_h) - np.array(_p)],
                            fmt="o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=6, mec="k", mew=0.4, capsize=3,
                            label=KCLS[_cl]["short"])
            for _x, _y, _n in zip(_xs12b + _o, _p, _nn):
                if np.isfinite(_y):
                    axs[2].annotate(str(_n), xy=(_x, _y), xytext=(0, 7), textcoords="offset points", fontsize=7,
                                    ha="center", color=KCLS[_cl]["color"])
        axs[2].axhline(60.0, color="0.3", lw=1.2, ls=":")
        axs[2].annotate("isotropic", xy=(0.02, 60.0), xycoords=("axes fraction", "data"), xytext=(0, 4),
                        textcoords="offset points", fontsize=8, color="0.3")
        axs[2].set_xticks(_xs12b); axs[2].set_xticklabels([_SSHORT12[s] for s in P12B_STAGES], fontsize=8.5)
        axs[2].set_ylabel("half-opening angle of the wind  [deg]")
        axs[2].set_title("(c) how collimated it is", loc="left", fontsize=10)
        _leg12b(axs[2], loc="best"); axs[2].grid(alpha=0.2, lw=0.5)
        ax = axs[3]                                                     # (d) does the DUST leave the same way as the gas?
        for _cl in KC:
            for _wn, _ls, _lab in (("g", "--", "gas"), ("d", "-", "dust")):
                _ii = [v["i"] for (k_, s_), v in W12B.items() if s_ == P12B_AT and v["ok"] and CLS9.get(k_) == _cl]
                if len(_ii) < P12B_NMIN:
                    continue
                A12b = np.array([np.maximum(_esc12b(i, _wn), P12B_FESC_FLOOR) for i in _ii], float)
                _cen, _lo, _hi, _n = _band12b(A12b)
                if _wn == "d":
                    ax.fill_between(_XMU12B, _lo, _hi, color=KCLS[_cl]["color"], alpha=0.18, lw=0)
                ax.plot(_XMU12B, _cen, color=KCLS[_cl]["color"], lw=2.3 if _wn == "d" else 1.4, ls=_ls,
                        label=f"{KCLS[_cl]['short']}, {_lab}  ({len(_ii)})")
                for _x, _c in zip(_XMU12B, _cen):
                    ROW12B.append(dict(block=f"fesc_{_wn}", stage=P12B_AT, kcls=_cl, x=float(_x), n_gal=len(_ii),
                                       med=float(_c), lo=np.nan, hi=np.nan))
        ax.set_yscale("log"); ax.axvline(P12B_POL, color="0.75", lw=0.9, ls=":")
        ax.set_xlabel(r"$\mu = |\cos\theta|$    (0 = plane, 1 = pole)")
        ax.set_ylabel(r"$f_{\rm esc}$: the share leaving")
        ax.set_title(f"(d) the dust against the gas, at {_SLAB12[P12B_AT]}", loc="left", fontsize=10)
        _leg12b(ax, loc="best", fontsize=7.5); ax.grid(alpha=0.2, lw=0.5, which="both")
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the geometry of the wind: polar share, alignment, opening angle",
                     y=0.995, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_outflow_collimation"); plt.show()

        # ── FIG 3: what the geometry costs the ISM — the dust clocks split by the collimation ──
        COL12B = {k_: ("pol" if v[P12B_SPLITON] >= _SCUT12B else "brd")
                  for (k_, s_), v in W12B.items() if s_ == P12B_AT and v["ok"] and np.isfinite(v[P12B_SPLITON])}
        _TR12B = {c: {k: to_grid(CH9[k]["t"] - _tsft9.get(k, np.nan), CH9[k][c]) for k in COL12B}
                  for c in ("lfd", "ldtm", "lfh2")}
        _CLK12B = [("lfd", r"(a) $\log\,M_{\rm dust}/M_\star$", r"$\log\, M_{\rm dust}/M_\star$"),
                   ("ldtm", r"(c) the metals that are in grains",
                    r"$\log\, M_{\rm dust}/(M_{\rm dust} + Z_{\rm gas}M_{\rm gas})$"),
                   ("lfh2", r"(d) $\log\,M_{\rm H_2}/M_\star$", r"$\log\, M_{\rm H_2}/M_\star$")]
        fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
        for _pi, (_c12b, _ttl, _ylab) in enumerate(_CLK12B):
            ax = axs[[0, 1, 1][_pi], [0, 0, 1][_pi]]
            for _cc in CC:
                _ks = [k for k in COL12B if COL12B[k] == _cc]
                if len(_ks) < 2:
                    continue
                _cen, _lo, _hi, _n = med_band(np.array([_TR12B[_c12b][k] for k in _ks], float), nmin=P12B_NMIN,
                                              nboot=P12B_NBOOT)
                if P12B_SPAG:
                    for _r12b in np.array([_TR12B[_c12b][k] for k in _ks], float):
                        ax.plot(TG, _r12b, color=CCLS[_cc]["color"], lw=0.4,
                                alpha=float(np.clip(14.0 / max(len(_ks), 1), 0.06, 0.35)), zorder=2)
                ax.fill_between(TG, _lo, _hi, color=CCLS[_cc]["color"], alpha=0.20, lw=0, zorder=3)
                ax.plot(TG, _cen, color=CCLS[_cc]["color"], lw=CCLS[_cc]["lw"], zorder=5,
                        label=f"{CCLS[_cc]['short']}  ({len(_ks)})")
                for _x, _c, _l, _h, _nn in zip(TG, _cen, _lo, _hi, _n):
                    ROW12B.append(dict(block="clock", stage=P12B_AT, quantity=_c12b, kcls=_cc, x=float(_x),
                                       n_gal=int(_nn), med=float(_c), lo=float(_l), hi=float(_h)))
            ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9, zorder=1)
            ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(_ylab); ax.set_title(_ttl, loc="left", fontsize=10)
            ax.grid(alpha=0.22, lw=0.5)
            if _pi:
                ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]")
        _leg12b(axs[0, 0], loc="lower left", fontsize=8.5)
        ax = axs[0, 1]                                                  # (b) the cross: the geometry, or the rotation?
        for _cc in CC:
            for _cl in KC:
                _ks = [k for k in COL12B if COL12B[k] == _cc and CLS9.get(k) == _cl]
                if len(_ks) < P12B_NMIN:
                    continue
                _cen, _lo, _hi, _n = med_band(np.array([_TR12B["lfd"][k] for k in _ks], float), nmin=P12B_NMIN,
                                              nboot=P12B_NBOOT)
                ax.plot(TG, _cen, color=CCLS[_cc]["color"], lw=2.2, ls="-" if _cl == "rot" else "--",
                        label=f"{CCLS[_cc]['short']}, {KCLS[_cl]['short']}  ({len(_ks)})")
                for _x, _c, _l, _h, _nn in zip(TG, _cen, _lo, _hi, _n):
                    ROW12B.append(dict(block="cross", stage=P12B_AT, quantity=f"lfd_{_cl}", kcls=_cc, x=float(_x),
                                       n_gal=int(_nn), med=float(_c), lo=float(_l), hi=float(_h)))
        ax.axvline(0, color="k", lw=1.3, ls="--", alpha=0.9)
        ax.axhline(DUSTY_CUT, color="0.35", lw=1.0, ls=":")
        ax.set_xlim(TG[0], TG[-1]); ax.set_ylabel(r"$\log\, M_{\rm dust}/M_\star$")
        ax.set_title(r"(b) the wind geometry $\times$ the kinematic class", loc="left", fontsize=10)
        ax.grid(alpha=0.22, lw=0.5); _leg12b(ax, loc="lower left", fontsize=7.5)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the dust clocks, split by how the wind left at {_SLAB12[P12B_AT]}",
                     y=0.99, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_outflow_clock"); plt.show()

        # ── what it says ──
        print(f"\nthe geometry of the wind, stage by stage (median over the galaxies of the stage):")
        print(f"  {'stage':10s} {'N':>4s} {'R_esc':>6s} {'f_pol':>6s} {'gas':>5s} {'d f_pol':>8s} {'align':>6s} "
              f"{'open':>5s} {'lam1':>5s}   AUC({KCLS['rot']['short']} > the rest): R_esc / align / open")
        for _s12b in P12B_STAGES:
            _ks = [k_ for (k_, s_), v in W12B.items() if s_ == _s12b and v["ok"]]
            if not _ks:
                print(f"  {_s12b:10s}    0"); continue
            _get = lambda f: np.array([W12B[(k_, _s12b)][f] for k_ in _ks], float)
            _m = np.array([CLS9.get(k_) == "rot" for k_ in _ks], bool)
            _au = [mw_auc(_get(f), _m) for f in ("resc", "align", "open_deg")]
            _f = lambda a, d=2: f"{np.nanmedian(a):.{d}f}" if np.isfinite(a).any() else "—"
            print(f"  {_s12b:10s} {len(_ks):4d} {_f(_get('resc')):>6s} {_f(_get('fpol')):>6s} "
                  f"{_f(_get('fpol_all')):>5s} {_f(_get('dfpol')):>8s} {_f(_get('align'), 0):>5s}d "
                  f"{_f(_get('open_deg'), 0):>4s}d {_f(_get('lam1')):>5s}   "
                  + " / ".join((f"{a[0]:.2f} (p {a[1]:.3f})" if np.isfinite(a[0]) else "—") for a in _au))
            for _cl in KC:
                _mm = np.array([CLS9.get(k_) == _cl for k_ in _ks], bool)
                if _mm.sum() >= P12B_NMIN:
                    print(f"      {KCLS[_cl]['short']:22s} N {int(_mm.sum()):3d}   R_esc {_f(_get('resc')[_mm]):>5s}"
                          f"   f_pol {_f(_get('fpol')[_mm]):>5s} (the gas it came from {_f(_get('fpol_all')[_mm]):>5s})"
                          f"   align {_f(_get('align')[_mm], 0)} deg"
                          f"   opening {_f(_get('open_deg')[_mm], 0)} deg"
                          f"   v_r {_f(_get('vr_fast')[_mm], 0)} km/s"
                          f"   Mdot_out {_f(_get('mdot_g')[_mm])} Msun/yr")
        print("  (R_esc = f_esc(pole)/f_esc(plane), THE reservoir-free statistic: 1 = the gas leaves in every direction, "
              "> 1 = through the poles. f_pol = the share of the wind in the polar half of the sky and 'gas' the same for "
              "the reservoir it is drawn from — 0.5 is isotropic for both, and a disc pushes BOTH down, which is why the "
              "difference d f_pol and the ratio R_esc are the ones to read. align = the angle between the wind's own axis "
              "and the gas rotation axis, 60 deg for a random axis. open = the wind's half-opening angle, 60 deg for "
              "isotropic. lam1 = the largest eigenvalue of the direction tensor, 1/3 for isotropic.)")
        _npol = sum(1 for v in COL12B.values() if v == "pol")
        print(f"\nthe split of Fig 3 at {_SLAB12[P12B_AT]} (on {P12B_SPLITON} at {_SCUT12B:g}): {_npol}/{len(COL12B)} "
              f"galaxies eject preferentially through the poles; by class — "
              + ", ".join(f"{KCLS[c]['short']} {sum(1 for k in COL12B if COL12B[k] == 'pol' and CLS9.get(k) == c)}"
                          f"/{sum(1 for k in COL12B if CLS9.get(k) == c)}" for c in KC))
        print("the dust clocks split by the wind geometry (median [N]; AUC = P(polar > broad) at that epoch):")
        for _c12b, _, _ in _CLK12B:
            _line = []
            for _m12b in P12B_MARKS:
                _i12b = int(np.argmin(np.abs(TG - _m12b)))
                _va = np.array([_TR12B[_c12b][k][_i12b] for k in COL12B if COL12B[k] == "pol"], float)
                _vb = np.array([_TR12B[_c12b][k][_i12b] for k in COL12B if COL12B[k] == "brd"], float)
                _va, _vb = _va[np.isfinite(_va)], _vb[np.isfinite(_vb)]
                _au = mw_auc(np.concatenate([_va, _vb]),
                             np.concatenate([np.ones(len(_va), bool), np.zeros(len(_vb), bool)]))[0] \
                    if len(_va) and len(_vb) else np.nan
                _line.append(f"{_m12b:+.0f}: " + (f"{np.median(_va):+.2f}" if len(_va) >= P12B_NMIN else "—") + "/"
                             + (f"{np.median(_vb):+.2f}" if len(_vb) >= P12B_NMIN else "—")
                             + (f" [{_au:.2f}]" if np.isfinite(_au) else "") + f" (N {len(_va)},{len(_vb)})")
                ROW12B.append(dict(block="epoch", stage=P12B_AT, quantity=_c12b, kcls="both", x=float(_m12b),
                                   n_gal=len(_va) + len(_vb), med=float(np.median(_va)) if len(_va) else np.nan,
                                   lo=float(np.median(_vb)) if len(_vb) else np.nan, hi=float(_au)))
            print(f"  {_c12b:5s} " + "   ".join(_line))
        print("  (each cell is  polar/broad [AUC] (N_polar, N_broad))")
        pd.DataFrame(ROW12B).to_csv(os.path.join(OUT, "kappa_m25_outflow.csv"), index=False)
        _G12B = []
        for _k12b in KEEP9:
            _r = dict(anchor_snap=_k12b[0], gal_id=_k12b[1], kcls=CLS9.get(_k12b), wind_class=COL12B.get(_k12b),
                      t_sft=_tsft9.get(_k12b, np.nan))
            for _s12b in P12B_STAGES:
                _v = W12B.get((_k12b, _s12b))
                _r[f"t_{_s12b}"] = ST12[_k12b][_s12b]
                for _f in ("snap", "gx", "dt", "n_fast", "resc", "resc_d", "fpol", "fpol_d", "fpol_all", "dfpol",
                           "align", "open_deg", "lam1", "coh"):
                    _r[f"{_f}_{_s12b}"] = _v[_f] if _v else np.nan
                _i = _v["i"] if _v else None
                for _f in ("mdot_g", "mdot_d", "vr_fast", "m_fast_g", "m_fast_d"):
                    _r[f"{_f}_{_s12b}"] = float(P12B[_f][_i]) if _i is not None else np.nan
            _G12B.append(_r)
        pd.DataFrame(_G12B).to_csv(os.path.join(OUT, "kappa_m25_outflow_galaxies.csv"), index=False)
        print(f"\ntables -> {os.path.join(OUT, 'kappa_m25_outflow.csv')} (the drawn points), "
              f"{os.path.join(OUT, 'kappa_m25_outflow_galaxies.csv')} (one row per galaxy)")
        if not P12B_ALLNODES:
            print("  set P12B_ALLNODES = True to measure every node of every branch (the cache is incremental) and the "
                  "opening angle becomes a clock like Part 10's")
```

## Part 12c — the null of Part 12b, and what SIMBA's own prescription already fixes

Two things are outstanding after Part 12b, and its figures answer neither.

### 1. Is the rotators' deviation from isotropy a small-number artefact?

The classes are 12–13 galaxies against 47–63, and $R_{\rm esc}$ is a ratio of two ratios. Three nulls are put to it, none
of which needs the particle files read again — the cube Part 12b cached carries the **counts** as well as the masses:

* **within a galaxy** — if leaving were independent of angle, the number of wind particles in the polar half of the sky
  would be $\mathrm{Binomial}(N_{\rm wind},\,p)$ with $p$ the polar share of the **reservoir**, a null that already
  contains the disc's own geometry. The $z$ of the observed count against it is printed per galaxy and drawn as an ECDF.
  It is an **upper bound** on the significance — a wind is launched in events, so its particles are not independent draws
  — which is why it is never the only test here;
* **counts against masses** — if a handful of heavy particles carried the effect, the count-weighted and mass-weighted
  $R_{\rm esc}$ would part company. They are printed side by side;
* **the controls** — $R_{\rm esc}$ against $N_{\rm wind}$ (an artefact grows as the counts fall) and against the
  **L-coherence** $|\sum \vec L| / \sum |\vec L|$ of the member gas (the structural variable a real effect should follow).

And because the class contrast is scored in a **grid** — 3 statistics × 4 stages — the p-values go through a hand-rolled
Benjamini–Hochberg at `P12C_FDR`. Twelve tests on thirteen galaxies is exactly the situation in which an isolated
$p = 0.03$ means nothing, and the cell says so. What replaces it is a **structured prediction, tested once**: the effect
should be *absent* at the AGN ignition and *present* at and after the truncation — in the same galaxies, with the same
particle counts.

### 2. How much of this does SIMBA put in by hand?

**All of the collimation, and none of the geometry.** SIMBA's kinetic AGN feedback is *always* bipolar: the wind is
ejected parallel and antiparallel to the angular momentum of the ~256 gas elements nearest the black hole, and that axis
is free to reorient as the inner gas does (Davé et al. 2019). So *"is the wind collimated?"* is not a question one can ask
of this simulation, and Part 12b never asks it. What is **not** prescribed — and is what Part 12b actually measures — is:

* whether that launch axis points **out of** the ISM the galaxy has to lose. The axis is set by the gas nearest the hole;
  the disc that must survive is the whole member gas, and Part 12 already measured 41–49° between the gas and the stars.
  `L_in` is built here the way the code builds it — over the `P12B_NLAUNCH` **nearest gas elements**, not inside a fixed
  radius — and the **radius those particles reach** is recorded with it. That is where Part 12's central hole re-enters:
  empty the middle, and the axis the jet follows is set by gas kiloparsecs out rather than by the nucleus;
* whether it **stays there**. A stable axis carves a channel and spares the plane; a wandering one sweeps every direction
  in turn and the plane is not spared at all. The angle between `L_in` at consecutive stages — and its rate in deg/Gyr —
  is the wander; pairs that land on the same snapshot are dropped, their angle being zero by construction.

### 3. What the answer turned out to be — 2026-09-10

**The wind is not aimed out of the disc**, and the earlier *"the coincidence between the cone and the disc is the
galaxy's"* is withdrawn. Read straight off the cached census:

| | rotating | the rest |
|---|---|---|
| wind axis vs the disc normal (a **random** axis gives 60°) | 60° / **75°** / 64° / 71° | 53° / 60° / 60° / 63° |
| $\lambda_1$ of the direction tensor (⅓ isotropic, ½ **planar**, 1 collimated) | 0.49–0.53 | 0.46–0.55 |
| polar half-sky share of the **fast mass** (0.5 = isotropic) | **0.44** | 0.48–0.51 |
| polar half-sky share of the **reservoir** | **0.26–0.31** | 0.49–0.52 |

At the X-ray stage the rotators' wind axis is significantly **in-plane** against the random-axis null (KS $p = 0.011$),
and only 8 % of them lie within 30° of the pole where a random axis gives 13 %. So $R_{\rm esc} > 1$ comes from the
**denominator**: their reservoir is planar and their wind is nearly isotropic — *the wind is more isotropic than the
medium it sits in*, and against a planar reservoir that reads as a polar channel. The other class's $R_{\rm esc}\approx 1$
is **forced rather than physical**: their reservoir share is already 0.49–0.52, so they have neither pole nor plane.

**The claim for the paper: the wind's geometry is decoupled from the disc's** — in a rotator that decoupling opens an
escape channel because the disc is planar; in the rest there is nothing to be decoupled from. Panels (c) and (f) are that
argument, and **(c) is also its control**: a rotationally supported plane has $v_r = 0$ in it *by construction*, so an
excess manufactured at disc radii would be a kinematic tautology rather than a wind. It is not — split shell by shell the
rotators' excess **peaks in the CGM**, 2.2–2.4 at 10–30 kpc against ~1.0 inside 3 kpc, and holds out to 100 kpc.

```python
# ── Part 12c — the null of Part 12b, and what SIMBA's own prescription already fixes ──
# Two questions are outstanding after Part 12b, and neither is answered by the figures it draws.
#   (1) IS THE ROTATORS' DEVIATION FROM ISOTROPY A SMALL-NUMBER ARTEFACT? The classes are 12-13 galaxies against 47-63, and
#       R_esc is a ratio of two ratios. Three separate nulls are put to it here, none of which needs the particle files to be
#       read again — the (mu x r x v_r) cube Part 12b cached carries the COUNTS as well as the masses:
#         (a) WITHIN a galaxy: if leaving were independent of angle, the number of wind particles in the polar half of the
#             sky would be Binomial(N_wind, p) with p the polar share of the RESERVOIR — a null that already contains the
#             disc's geometry. The z of the observed count against it is printed per galaxy and drawn as an ECDF. It is an
#             upper bound on the significance (a wind is launched in events, so the particles are not independent draws),
#             which is why it is never the only test here.
#         (b) THE COUNT VERSION of R_esc against the MASS-weighted one: if a handful of heavy particles carried the effect
#             the two would part company. They are printed side by side.
#         (c) THE CONTROLS: R_esc against N_wind (the artefact would grow as the counts fall) and against the L-COHERENCE
#             |sum L| / sum |L| of the member gas (the structural variable the effect should follow if it is real).
#       And BECAUSE THE CLASS CONTRAST IS SCORED IN A GRID (3 statistics x 4 stages), the p-values are put through a
#       hand-rolled Benjamini-Hochberg at P12C_FDR — 12 tests on 13 galaxies is exactly where an isolated p = 0.03 means
#       nothing. What replaces it is a STRUCTURED prediction, tested once: the effect should be ABSENT at the AGN ignition
#       and PRESENT at and after the truncation, in the same galaxies with the same particle counts.
#   (2) HOW MUCH OF THIS DOES SIMBA PUT IN BY HAND? All of the collimation, and none of the geometry. SIMBA's kinetic AGN
#       feedback is ALWAYS bipolar: the wind is ejected parallel and antiparallel to the angular momentum of the ~256 gas
#       elements nearest the black hole — the inner ~kpc — and that axis is free to reorient as the inner gas does
#       (Dave et al. 2019). So "is the wind collimated" is not a question one can ask of this simulation, and Part 12b never
#       asks it. What is NOT prescribed, and is what Part 12b actually measures, is
#         - whether that launch axis points OUT OF the ISM the galaxy has to lose. The axis is set by the inner kpc; the
#           disc that must survive is the whole member gas, and Part 12 already measured 41-49 deg between the gas and the
#           stars, so the two need not agree. The angle between L_in (the launch proxy: the P12B_NLAUNCH gas elements
#           nearest the centre, the code's own rule) and L_ISM is measured here, per class — together with the RADIUS
#           those 256 particles reach, which is where Part 12's central hole enters: empty the middle and the axis the
#           jet follows is set by gas kiloparsecs out, not by the nucleus.
#         - whether it STAYS THERE. A stable axis carves a channel and spares the plane; a wandering one sweeps every
#           direction in turn and the plane is not spared at all. The angle between L_in at consecutive stages — and its
#           rate in deg/Gyr — is the wander, and pairs that fall on the SAME snapshot are excluded, since their angle is
#           zero by construction.
#       That is the honest form of the claim. WHAT THE ANSWER TURNED OUT TO BE (2026-09-10): the wind is NOT aimed out
#       of the disc, and the "coincidence between the cone and the disc" phrasing is withdrawn. The wind's own
#       mass-weighted axis sits 60 / 75 / 64 / 71 deg from the disc normal over the four stages for the rotators, where
#       a RANDOM axis gives 60 (at xray_on the KS against that null gives p = 0.011, on the IN-PLANE side); lam1 =
#       0.49-0.50 is the PLANAR signature (1/2, 1/2, 0) rather than the collimated one (1, 0, 0); and the polar half of
#       the sky still takes only 0.44 of the fast mass. R_esc > 1 comes from the DENOMINATOR: the rotators' RESERVOIR
#       polar share is 0.26-0.31 against their wind's 0.44 - the wind is MORE ISOTROPIC THAN THE MEDIUM it sits in, and
#       measured against a planar reservoir that reads as a polar channel. The rest's R_esc ~ 1 is forced rather than
#       physical: their reservoir share is 0.49-0.52, so they have neither pole nor plane and the ratio is 1 by
#       construction. THE CLAIM, for the paper: the wind's geometry is DECOUPLED from the disc's - in a rotator that
#       decoupling opens an escape channel because the disc is planar, in the rest there is nothing to be decoupled
#       from. Panels (c) and (f) are that argument, and (c) is also its control: a rotationally supported plane has
#       v_r = 0 in it BY CONSTRUCTION, so an excess made at disc radii would be a kinematic tautology rather than a
#       wind. It is not - the excess peaks in the CGM, at 10-30 kpc, and holds out to 100.
# A PURE READ of Part 12b's cache and Part 9's chains — it reads no files and writes one figure and one table.
P12C_FDR   = 0.05          # the Benjamini-Hochberg level the class-contrast grid is controlled at
P12C_ZCUT  = 3.0           # "this galaxy's own polar excess is significant" on the binomial null
P12C_QTY   = [("resc", r"$R_{\rm esc}$"), ("align", "the wind-vs-rotation angle"), ("open_deg", "the opening angle")]
P12C_EARLY, P12C_LATE = ["agn_ign"], ["sft", "anchor"]      # the structured prediction: absent before, present after
P12C_NBOOT = 2000
P12C_SHSTAGE = None        # panel (c) is drawn at this stage; None = follow P12B_AT (the table carries every stage)

if "W12B" not in globals():
    print("Part 12c: needs Part 12b — run it first.")
elif "linx" not in P12B or "r_launch" not in P12B:
    print("Part 12c: the census predates the launch-axis proxy — delete P12B_CACHE and re-run Part 12b "
          "(it re-reads the same ~257 files in ~10 s).")
else:
    # ── (A) the per-galaxy null, straight off the cached COUNTS ──
    def _zbin12c(i):
        """The z of the polar wind COUNT against Binomial(N_wind, p_reservoir): the null in which leaving is independent
        of angle, with the reservoir's own geometry already inside p."""
        C = np.asarray(P12B["cube_n"][i], float)[:, _ISH12B, :]
        nf, na = C[:, :, _IVF12B].sum(axis=(1, 2)), C.sum(axis=(1, 2))
        N, _tot = nf.sum(), na.sum()
        if N < 5 or _tot <= 0:
            return np.nan
        p = na[_IPOL12B].sum() / _tot
        return float((nf[_IPOL12B].sum() - N * p) / np.sqrt(N * p * (1 - p))) if 0 < p < 1 else np.nan

    R12C = []
    for (_k12c, _s12c), _v in W12B.items():
        if not _v["ok"]:
            continue
        R12C.append(dict(gal=_k12c, stage=_s12c, kcls=CLS9.get(_k12c), i=_v["i"], resc=_v["resc"],
                         resc_n=_resc12b(_v["i"], "n"), align=_v["align"], open_deg=_v["open_deg"],
                         n_fast=_v["n_fast"], coh=_v["coh"], coh_in=_v["coh_in"], n_gas_in=_v["n_gas_in"],
                         zbin=_zbin12c(_v["i"])))
    D12C = pd.DataFrame(R12C)
    ROW12C = []
    print(f"Part 12c — the null of the polar excess, on {len(D12C)} galaxy-stage measurements "
          f"({D12C['gal'].nunique()} galaxies)\n")
    print("(a) WITHIN each galaxy: the wind's polar count against Binomial(N_wind, p_reservoir)")
    print(f"  {'stage':9s} {'class':24s} {'N':>3s} {'N_wind':>7s}  {'R_esc':>5s} {'(counts)':>8s}  {'median z':>8s}  "
          f"share with z > {P12C_ZCUT:g}")
    for _s12c in P12B_STAGES:
        for _cl in KC:
            _d = D12C[(D12C.stage == _s12c) & (D12C.kcls == _cl)]
            if not len(_d):
                continue
            _zz = _d.zbin.to_numpy(float); _zz = _zz[np.isfinite(_zz)]
            print(f"  {_s12c:9s} {KCLS[_cl]['short']:24s} {len(_d):3d} {np.nanmedian(_d.n_fast):7.0f}  "
                  f"{np.nanmedian(_d.resc):5.2f} {np.nanmedian(_d.resc_n):8.2f}  {np.median(_zz) if len(_zz) else np.nan:+8.1f}  "
                  f"{np.mean(_zz > P12C_ZCUT) * 100 if len(_zz) else np.nan:3.0f} %  ({int((_zz > P12C_ZCUT).sum())}/{len(_zz)})")
            ROW12C.append(dict(block="null", stage=_s12c, kcls=_cl, n_gal=len(_d),
                               resc=float(np.nanmedian(_d.resc)), resc_n=float(np.nanmedian(_d.resc_n)),
                               z_med=float(np.median(_zz)) if len(_zz) else np.nan,
                               f_sig=float(np.mean(_zz > P12C_ZCUT)) if len(_zz) else np.nan,
                               n_fast=float(np.nanmedian(_d.n_fast))))
    print("  (the count and mass versions of R_esc agreeing means no handful of heavy particles is carrying it; the z is an "
          "UPPER bound — wind particles are launched in events and are not independent draws)")

    print("\n(b) THE CONTROLS: does R_esc follow the particle count, or the structure?")
    for _s12c in P12B_STAGES:
        _d = D12C[D12C.stage == _s12c].dropna(subset=["resc"])
        _o = []
        for _c12c, _lab in (("n_fast", "N_wind"), ("coh", "L-coherence"), ("n_gas_in", "N_gas(<launch)")):
            _dd = _d.dropna(subset=[_c12c])
            if len(_dd) < 8:
                continue
            _r = spearmanr(_dd[_c12c].to_numpy(float), _dd.resc.to_numpy(float))
            _o.append(f"rho({_lab}) {_r.statistic:+.2f} (p {_r.pvalue:.3f})")
            ROW12C.append(dict(block="control", stage=_s12c, kcls="all", quantity=_c12c, n_gal=len(_dd),
                               med=float(_r.statistic), lo=float(_r.pvalue)))
        print(f"  {_s12c:9s} " + "   ".join(_o))
    print("  (an artefact of small numbers would make R_esc scale with N_wind; a real geometry makes it follow the "
          "coherence of the gas that sets the axis)")

    # ── (C) the class-contrast grid, with multiplicity controlled ──
    print(f"\n(c) THE CLASS CONTRAST, all {len(P12C_QTY) * len(P12B_STAGES)} cells, Benjamini-Hochberg at q = {P12C_FDR:g}")
    _P12C = []
    for _s12c in P12B_STAGES:
        for _q12c, _ in P12C_QTY:
            _d = D12C[D12C.stage == _s12c].dropna(subset=[_q12c])
            _a = _d[_d.kcls == "rot"][_q12c].to_numpy(float); _b = _d[_d.kcls == "iso"][_q12c].to_numpy(float)
            if len(_a) < 3 or len(_b) < 3:
                continue
            _u = mannwhitneyu(_a, _b, alternative="two-sided")
            _P12C.append((_s12c, _q12c, float(_u.statistic) / (len(_a) * len(_b)), float(_u.pvalue), len(_a), len(_b)))
    _P12C.sort(key=lambda t: t[3])
    _m12c = len(_P12C)
    _kmax = max([k for k, t in enumerate(_P12C, 1) if t[3] <= k / _m12c * P12C_FDR] or [0])
    for _k12c, (_s12c, _q12c, _auc, _p, _na, _nb) in enumerate(_P12C, 1):
        print(f"  {_s12c:9s} {_q12c:9s} AUC {_auc:5.2f}  p {_p:7.4f}  N {_na},{_nb}   rank {_k12c:2d}/{_m12c}  "
              f"BH threshold {_k12c / _m12c * P12C_FDR:.4f}   {'SURVIVES' if _k12c <= _kmax else 'no'}")
        ROW12C.append(dict(block="contrast", stage=_s12c, kcls="rot-vs-iso", quantity=_q12c, n_gal=_na + _nb,
                           med=_auc, lo=_p, hi=float(_k12c <= _kmax)))
    print(f"  -> {_kmax} of {_m12c} cells survive. With {D12C[D12C.kcls == 'rot'].gal.nunique()} rotating galaxies that is "
          "the honest reading of any SINGLE cell: suggestive, not established.")

    # ── (D) the structured prediction, tested once per class ──
    print(f"\n(d) THE PREDICTION THAT IS NOT A GRID: R_esc should be absent at {'/'.join(P12C_EARLY)} and present at "
          f"{'/'.join(P12C_LATE)} — same galaxies, same counts")
    for _cl in KC:
        _e = D12C[(D12C.kcls == _cl) & (D12C.stage.isin(P12C_EARLY))].resc.dropna().to_numpy(float)
        _l = D12C[(D12C.kcls == _cl) & (D12C.stage.isin(P12C_LATE))].resc.dropna().to_numpy(float)
        if len(_e) < 3 or len(_l) < 3:
            continue
        _u = mannwhitneyu(_l, _e, alternative="two-sided")
        print(f"  {KCLS[_cl]['short']:24s} {np.median(_e):.2f} (N {len(_e)})  ->  {np.median(_l):.2f} (N {len(_l)})   "
              f"AUC {float(_u.statistic) / (len(_e) * len(_l)):.2f}, p {float(_u.pvalue):.4f}")
        ROW12C.append(dict(block="prediction", stage="early_vs_late", kcls=_cl, n_gal=len(_e) + len(_l),
                           med=float(np.median(_l) - np.median(_e)), lo=float(_u.pvalue)))

    # ── (E) what SIMBA fixes, and what it does not: the launch axis and its wander ──
    def _ang12c(u, v):
        u, v = np.asarray(u, float), np.asarray(v, float)
        if not (np.isfinite(u).all() and np.isfinite(v).all()):
            return np.nan
        return float(np.degrees(np.arccos(np.clip(abs(u @ v), 0, 1))))
    _AX12C = {}
    for (_k12c, _s12c), _v in W12B.items():
        _i = _v["i"]
        _AX12C[(_k12c, _s12c)] = dict(snap=_v["snap"], gx=_v["gx"],
                                      n=np.array([P12B["nx"][_i], P12B["ny"][_i], P12B["nz"][_i]], float),
                                      lin=np.array([P12B["linx"][_i], P12B["liny"][_i], P12B["linz"][_i]], float),
                                      t=ST12[_k12c][_s12c])
    print(f"\n(e) WHAT SIMBA FIXES AND WHAT IT DOES NOT. The wind is bipolar BY CONSTRUCTION, along +-L of the gas nearest "
          f"the BH;\n    L_in here is that axis over the {P12B_NLAUNCH} nearest gas elements. Free are the two below.")
    print(f"  {'stage':9s} {'class':24s} {'N':>3s}  L_in vs the ISM disc   r of those {P12B_NLAUNCH}   coherence there")
    for _s12c in P12B_STAGES:
        for _cl in KC:
            _ks = [k for (k, s) in _AX12C if s == _s12c and CLS9.get(k) == _cl]
            if len(_ks) < P12B_NMIN:
                continue
            _a = np.array([_ang12c(_AX12C[(k, _s12c)]["lin"], _AX12C[(k, _s12c)]["n"]) for k in _ks], float)
            _n = np.array([W12B[(k, _s12c)]["r_launch"] for k in _ks], float)
            _c = np.array([W12B[(k, _s12c)]["coh_in"] for k in _ks], float)
            print(f"  {_s12c:9s} {KCLS[_cl]['short']:24s} {len(_ks):3d}  {np.nanmedian(_a):9.0f} deg        "
                  f"{np.nanmedian(_n):7.2f} kpc      {np.nanmedian(_c):.2f}")
            ROW12C.append(dict(block="launch", stage=_s12c, kcls=_cl, n_gal=len(_ks), med=float(np.nanmedian(_a)),
                               lo=float(np.nanmedian(_n)), hi=float(np.nanmedian(_c))))
    _W12C = {c: [] for c in KC}
    for _k12c in KEEP9:
        _st = [s for s in P12B_STAGES if (_k12c, s) in _AX12C]
        for _s1, _s2 in zip(_st[:-1], _st[1:]):
            _A, _B = _AX12C[(_k12c, _s1)], _AX12C[(_k12c, _s2)]
            if (_A["snap"], _A["gx"]) == (_B["snap"], _B["gx"]):        # the same node twice: the angle is zero by fiat
                continue
            _dt = abs(_B["t"] - _A["t"])
            for _w, _lab in ((_ang12c(_A["lin"], _B["lin"]), "lin"), (_ang12c(_A["n"], _B["n"]), "ism")):
                _W12C[CLS9.get(_k12c, "iso")].append(dict(pair=f"{_s1}->{_s2}", axis=_lab, ang=_w, dt=_dt,
                                                          rate=_w / _dt if _dt > 0 else np.nan))
    print("\n  THE WANDER of that axis between consecutive stages (pairs on the same snapshot dropped):")
    for _cl in KC:
        _d = pd.DataFrame(_W12C[_cl])
        if not len(_d):
            continue
        for _ax, _lab in (("lin", f"the launch axis  (L_in, {P12B_NLAUNCH} nearest)"), ("ism", "the ISM disc     (L_gas)")):
            _dd = _d[(_d.axis == _ax)].dropna(subset=["ang"])
            if not len(_dd):
                continue
            print(f"    {KCLS[_cl]['short']:24s} {_lab:38s} {np.median(_dd.ang):5.0f} deg over "
                  f"{np.median(_dd.dt):.2f} Gyr  ({np.median(_dd.rate):5.0f} deg/Gyr, N {len(_dd)})")
            ROW12C.append(dict(block="wander", stage="consecutive", kcls=_cl, quantity=_ax, n_gal=len(_dd),
                               med=float(np.median(_dd.ang)), lo=float(np.median(_dd.rate))))
    print("    (90 deg between two epochs = the cone has moved onto what the plane used to be; a random pair of axes "
          "gives 60 deg in the median)")

    # ── (F) R_esc SHELL BY SHELL — where in the halo the excess is made ──
    def _resc_sh12c(i, sh, w="g"):
        """R_esc restricted to ONE radial shell of the cached cube — the same ratio of ratios, per radius."""
        C = np.asarray(P12B[f"cube_{w}"][i], float)[:, list(sh), :]
        out = []
        for _m in (_IPOL12B, ~_IPOL12B):
            _n, _d = C[_m][:, :, _IVF12B].sum(), C[_m].sum()
            out.append(_n / _d if _d > 0 else np.nan)
        return float(out[0] / out[1]) if (np.isfinite(out).all() and out[1] > 0) else np.nan

    _RB12C = np.asarray(P12B_RB, float)                                # the knob, not the cache: P12B may be in memory
    _SHC12C = np.sqrt(_RB12C[:-1] * _RB12C[1:])                        # the geometric centre of each shell, kpc
    _SHL12C = [f"{_RB12C[j]:g}-{_RB12C[j + 1]:g}" for j in range(len(_RB12C) - 1)]
    S12C = {}
    for (_k12c, _s12c), _v in W12B.items():
        if not _v["ok"]:
            continue
        for _j in range(len(_SHL12C)):
            S12C.setdefault((CLS9.get(_k12c), _s12c, _j), []).append(_resc_sh12c(_v["i"], [_j]))
    _r50_12c = float(np.nanmedian(P12B["r50_gasmem"]))
    print(f"\n  R_esc SHELL BY SHELL — the disc plane has v_r = 0 in it BY CONSTRUCTION, so an excess made at disc "
          f"radii would be a tautology rather than a wind (median R50 of the member gas: {_r50_12c:.1f} kpc)")
    print("    " + f"{'stage':<10}{'class':<24}" + "".join(f"{_l + ' kpc':>14}" for _l in _SHL12C))
    for _s12c in P12B_STAGES:
        for _cl in KC:
            _cells = []
            for _j in range(len(_SHL12C)):
                _v = np.array(S12C.get((_cl, _s12c, _j), []), float); _v = _v[np.isfinite(_v)]
                _cells.append(f"{np.median(_v):.2f} ({len(_v)})" if len(_v) >= P12B_NMIN else "-")
                if len(_v) >= P12B_NMIN:
                    ROW12C.append(dict(block="shell", stage=_s12c, kcls=_cl, quantity=_SHL12C[_j], n_gal=len(_v),
                                       x=float(_SHC12C[_j]), med=float(np.median(_v))))
            print(f"    {_s12c:<10}{KCLS[_cl]['short']:<24}" + "".join(f"{_c:>14}" for _c in _cells))

    # ── FIG 4: the null, the control, the radius, the clock, the axis and the reservoir ──
    _xs12c = np.arange(len(P12B_STAGES))
    _SHAT12C = P12C_SHSTAGE or P12B_AT
    fig, _AX = plt.subplots(2, 3, figsize=(16.4, 7.6))
    ax = _AX[0, 0]                                                     # (a) the per-galaxy null
    for _cl in KC:
        _v = D12C[(D12C.stage == P12B_AT) & (D12C.kcls == _cl)].zbin.dropna().to_numpy(float)
        if not len(_v):
            continue
        ax.plot(np.sort(_v), np.arange(1, len(_v) + 1) / len(_v), drawstyle="steps-post",
                color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_v)})")
    ax.axvline(0, color="0.3", lw=1.2, ls="--"); ax.axvline(P12C_ZCUT, color="0.55", lw=1.0, ls=":")
    ax.annotate("isotropic", xy=(0, 0.03), xytext=(3, 0), textcoords="offset points", fontsize=8, color="0.3")
    ax.set_xscale("symlog", linthresh=3); ax.set_ylim(0, 1.02)
    ax.set_xlabel("z of the polar wind count against Binomial($N_{\\rm wind}$, $p_{\\rm reservoir}$)")
    ax.set_ylabel("cumulative fraction of the class")
    ax.set_title(f"(a) is ONE galaxy's excess significant? at {_SLAB12[P12B_AT]}", loc="left", fontsize=10)
    _leg12b(ax, loc="upper left"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX[0, 1]                                                     # (b) the control: counts vs structure
    for _cl in KC:
        _d = D12C[(D12C.stage == P12B_AT) & (D12C.kcls == _cl)].dropna(subset=["resc", "coh"])
        ax.scatter(_d.coh, _d.resc, s=18 + 12 * np.log10(np.clip(_d.n_fast, 10, None)), alpha=0.75,
                   color=KCLS[_cl]["color"], edgecolor="k", linewidth=0.3, label=f"{KCLS[_cl]['short']}  ({len(_d)})")
    _dd = D12C[D12C.stage == P12B_AT].dropna(subset=["resc", "coh"])
    if len(_dd) >= 8:
        _r = spearmanr(_dd.coh.to_numpy(float), _dd.resc.to_numpy(float))
        ax.annotate(rf"$\rho = {_r.statistic:+.2f}$  (p {_r.pvalue:.3f})", xy=(0.03, 0.95), xycoords="axes fraction",
                    va="top", fontsize=9, color="0.25")
    ax.axhline(1.0, color="0.3", lw=1.2, ls="--"); ax.set_yscale("log")
    ax.set_xlabel(r"L-coherence of the member gas,  $|\sum \vec L| / \sum |\vec L|$")
    ax.set_ylabel(r"$R_{\rm esc}$"); ax.set_title("(b) it follows the structure, not the counts", loc="left", fontsize=10)
    _leg12b(ax, loc="lower right"); ax.grid(alpha=0.2, lw=0.5, which="both")
    ax = _AX[0, 2]                                                     # (c) where in the halo it is made
    for _cl in KC:
        _p, _l, _h, _nn = [], [], [], []
        for _j in range(len(_SHL12C)):
            _v = np.array(S12C.get((_cl, _SHAT12C, _j), []), float); _v = _v[np.isfinite(_v)]
            if len(_v) < P12B_NMIN:
                _p.append(np.nan); _l.append(np.nan); _h.append(np.nan); _nn.append(len(_v)); continue
            _bs = np.median(np.random.default_rng(23).choice(_v, (P12C_NBOOT, len(_v))), axis=1)
            _p.append(float(np.median(_v))); _l.append(float(np.percentile(_bs, 16)))
            _h.append(float(np.percentile(_bs, 84))); _nn.append(len(_v))
        _p, _l, _h = np.array(_p), np.array(_l), np.array(_h)
        ax.fill_between(_SHC12C, _l, _h, color=KCLS[_cl]["color"], alpha=0.20, lw=0, zorder=2)
        ax.plot(_SHC12C, _p, "o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=5, mec="k", mew=0.4, zorder=4,
                label=f"{KCLS[_cl]['short']}  ({max(_nn) if _nn else 0})")
    if np.isfinite(_r50_12c):
        ax.axvline(_r50_12c, color="0.45", lw=1.0, ls="-.", zorder=1)
        ax.annotate(f"$R_{{50}}$(gas) = {_r50_12c:.1f} kpc", xy=(_r50_12c, 0.03), xycoords=("data", "axes fraction"),
                    xytext=(3, 0), textcoords="offset points", fontsize=7.5, color="0.35", rotation=90)
    ax.axhline(1.0, color="0.3", lw=1.2, ls="--", zorder=1)
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlabel("radius [kpc]  (shell centre)"); ax.set_ylabel(r"$R_{\rm esc}$ in that shell alone")
    ax.set_title(f"(c) made in the CGM, not in the plane — {_SLAB12[_SHAT12C]}", loc="left", fontsize=10)
    _leg12b(ax, loc="upper left"); ax.grid(alpha=0.2, lw=0.5, which="both")
    ax = _AX[1, 0]                                                     # (d) the clock of R_esc
    for _ci, _cl in enumerate(KC):
        _p, _l, _h, _nn = [], [], [], []
        for _s12c in P12B_STAGES:
            _v = D12C[(D12C.stage == _s12c) & (D12C.kcls == _cl)].resc.dropna().to_numpy(float)
            if len(_v) < P12B_NMIN:
                _p.append(np.nan); _l.append(np.nan); _h.append(np.nan); _nn.append(len(_v)); continue
            _bs = np.median(np.random.default_rng(11).choice(_v, (P12C_NBOOT, len(_v))), axis=1)
            _p.append(float(np.median(_v))); _l.append(float(np.percentile(_bs, 16)))
            _h.append(float(np.percentile(_bs, 84))); _nn.append(len(_v))
            ROW12C.append(dict(block="clock", stage=_s12c, kcls=_cl, n_gal=len(_v), med=_p[-1], lo=_l[-1], hi=_h[-1]))
        _o = 0.06 * (_ci - 0.5)
        ax.errorbar(_xs12c + _o, _p, yerr=[np.array(_p) - np.array(_l), np.array(_h) - np.array(_p)],
                    fmt="o-", color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], ms=6, mec="k", mew=0.4, capsize=3,
                    label=KCLS[_cl]["short"])
        for _x, _y, _n in zip(_xs12c + _o, _p, _nn):
            if np.isfinite(_y):
                ax.annotate(str(_n), xy=(_x, _y), xytext=(0, 7), textcoords="offset points", fontsize=7,
                            ha="center", color=KCLS[_cl]["color"])
    ax.set_yscale("log"); ax.axhline(1.0, color="0.3", lw=1.2, ls=":")
    ax.annotate("no preferred direction", xy=(0.02, 1.0), xycoords=("axes fraction", "data"), xytext=(0, 4),
                textcoords="offset points", fontsize=8, color="0.3")
    ax.set_xticks(_xs12c); ax.set_xticklabels([_SSHORT12[s] for s in P12B_STAGES], fontsize=8.5)
    ax.set_ylabel(r"$R_{\rm esc}$"); ax.set_title("(d) it is ACQUIRED, not inherited", loc="left", fontsize=10)
    _leg12b(ax, loc="upper left"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX[1, 1]                                                     # (e) the axis SIMBA launches along
    for _cl in KC:
        _d = pd.DataFrame(_W12C[_cl])
        _v = _d[_d.axis == "lin"].ang.dropna().to_numpy(float) if len(_d) else np.array([])
        if len(_v) >= P12B_NMIN:
            ax.plot(np.sort(_v), np.arange(1, len(_v) + 1) / len(_v), drawstyle="steps-post",
                    color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}, wander  ({len(_v)})")
        _u = np.array([_ang12c(_AX12C[(k, s)]["lin"], _AX12C[(k, s)]["n"]) for (k, s) in _AX12C
                       if CLS9.get(k) == _cl], float)
        _u = _u[np.isfinite(_u)]
        if len(_u) >= P12B_NMIN:
            ax.plot(np.sort(_u), np.arange(1, len(_u) + 1) / len(_u), drawstyle="steps-post",
                    color=KCLS[_cl]["color"], lw=1.4, ls="--", label=f"{KCLS[_cl]['short']}, vs the disc  ({len(_u)})")
    _th12c = np.linspace(0, 90, 181)
    ax.plot(_th12c, 1 - np.cos(np.radians(_th12c)), color="0.6", lw=1.2, ls=":", label="two random axes")
    ax.set_xlim(0, 90); ax.set_ylim(0, 1.02); ax.set_xticks([0, 15, 30, 45, 60, 75, 90])
    ax.set_xlabel(rf"angle [deg]:  $L_{{\rm in}}$ between stages (solid), $L_{{\rm in}}$ vs the disc (dashed)")
    ax.set_ylabel("cumulative fraction")
    ax.set_title(f"(e) the axis SIMBA launches along ({P12B_NLAUNCH} nearest)", loc="left", fontsize=10)
    _leg12b(ax, loc="lower right", fontsize=7); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX[1, 2]                                                     # (f) where R_esc actually comes from
    for _cl in KC:
        for _wk, _ls, _mk, _lab in (("fpol_all", "--", "o", "the reservoir"), ("fpol", "-", "s", "the wind")):
            _p, _l, _h = [], [], []
            for _s12c in P12B_STAGES:
                _v = np.array([v[_wk] for (k_, s_), v in W12B.items()
                               if s_ == _s12c and v["ok"] and CLS9.get(k_) == _cl], float)
                _v = _v[np.isfinite(_v)]
                if len(_v) < P12B_NMIN:
                    _p.append(np.nan); _l.append(np.nan); _h.append(np.nan); continue
                _bs = np.median(np.random.default_rng(29).choice(_v, (P12C_NBOOT, len(_v))), axis=1)
                _p.append(float(np.median(_v))); _l.append(float(np.percentile(_bs, 16)))
                _h.append(float(np.percentile(_bs, 84)))
                ROW12C.append(dict(block="polar_share", stage=_s12c, kcls=_cl, quantity=_wk, n_gal=len(_v),
                                   med=_p[-1], lo=_l[-1], hi=_h[-1]))
            _p, _l, _h = np.array(_p), np.array(_l), np.array(_h)
            ax.errorbar(_xs12c, _p, yerr=[_p - _l, _h - _p], fmt=_mk + _ls, color=KCLS[_cl]["color"],
                        lw=KCLS[_cl]["lw"], ms=5, mec="k", mew=0.4, capsize=2.5,
                        mfc=(KCLS[_cl]["color"] if _wk == "fpol" else "white"),
                        label=f"{KCLS[_cl]['short']}, {_lab}")
    ax.axhline(0.5, color="0.3", lw=1.2, ls=":")
    ax.annotate("isotropic", xy=(0.02, 0.5), xycoords=("axes fraction", "data"), xytext=(0, 4),
                textcoords="offset points", fontsize=8, color="0.3")
    ax.set_xticks(_xs12c); ax.set_xticklabels([_SSHORT12[s] for s in P12B_STAGES], fontsize=8.5)
    ax.set_ylabel(rf"polar share of the mass  ($\mu > {P12B_POL:g}$)")
    ax.set_title("(f) the wind is more isotropic than the medium", loc="left", fontsize=10)
    _leg12b(ax, loc="lower left", fontsize=7); ax.grid(alpha=0.2, lw=0.5)
    print("\n  (f) the polar share of the RESERVOIR against the polar share of the WIND: R_esc > 1 is the gap between "
          "them,\n      not a wind aimed at the pole - a planar reservoir under a near-isotropic wind.")
    fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the null of the polar excess, and the axis the code fixes",
                 y=0.995, fontsize=12, fontweight="bold")
    fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_outflow_null"); plt.show()
    pd.DataFrame(ROW12C).to_csv(os.path.join(OUT, "kappa_m25_outflow_null.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_m25_outflow_null.csv')}")
```

## Part 12d — the accretion history behind the split: is the black hole the upstream cause?

Parts 12b and 12c leave the mechanism open. What they establish is a **timing**: the two classes are indistinguishable at the
AGN ignition and part company afterwards — and the parting is *not* in the jet leg. Ignition → the X-ray gate opening takes
0.58 Gyr for the rotators against 0.51 Gyr for the rest ($p = 0.63$); what differs is what happens **after** the gate opens,
where the rest truncate as it opens ($t_{\rm SFT} - t_{\rm x1} = -0.04$ Gyr) and the rotators hold out ($+1.47$ Gyr,
$p = 0.024$).

The obvious upstream variable is the one neither part measures: **the accretion rate itself**. The hypothesis to test:

> a fast accretor moves quickly from the jet mode to the X-ray mode, quenches quickly and leaves its disc intact; a slow
> accretor sits in the jet mode long enough to stir the disc before the X-ray channel ever fires.

### What it reads

Part 9's chains keep only the state code and the BHAR, so `f_Edd`, $M_{\rm BH}$ and SIMBA's own gas gate
$f_{\rm gas} = M_{\rm gas}/M_\star$ are not in them. This cell therefore makes **one pass** over the histories file Parts 3
and 9 already read (`ism_prediction_histories_{box}.hdf5`), restricted to the Part 9 chains. No catalogues, no particle
files. The states come from Part 0's `agn_states()` **verbatim**, so the gates here are the gates Parts 3, 4, 9 and 12 date
their stages with: jets at $M_{\rm BH} > 10^{7.5}$ and $f_{\rm Edd} < 0.2$, full velocity at $f_{\rm Edd} \leq 0.02$, and the
X-ray channel on top of that at $f_{\rm gas} < 0.2$.

### What it measures, per galaxy, on the branch's own sampling

* **the accretion** — $\log f_{\rm Edd}$ at the ignition, and its median before the ignition / on the jet leg / after the
  gate; the peak before the truncation; the time spent *above* the jet threshold (the radiative mode) and the time in the
  jet state; $\log M_{\rm BH}$ at the ignition and its growth from the gate to the anchor;
* **the duty cycle**, which is what the hypothesis is really about — the fraction of the post-gate window with the X-ray
  channel actually firing, the same over the survival window $t_{\rm x1} \to t_{\rm SFT}$, the number of contiguous firing
  episodes, the longest firing-free gap, and the quiet time between the last firing and the anchor;
* **the thing to explain** — $\mathrm{delay} = t_{\rm SFT} - t_{\rm x1}$, the Gyr and a half the rotators survive an open
  channel.

### And the test is *not* the class contrast

A contrast over ~13 rotators is scored here with the same hand-rolled Benjamini–Hochberg as Part 12c, and reported honestly
— but the question the hypothesis actually asks is a **correlation over every galaxy at once**: does the accretion rate, or
the duty cycle, *predict* the delay? That is scored with Spearman on the pooled sample and within each class, and it needs
no class balance at all.

**What would falsify it**: $f_{\rm Edd}$ at the ignition the same in both classes *and* uncorrelated with the delay.
**What would confirm it**: the rotators accreting faster at the ignition, and the delay falling as $f_{\rm Edd}$ rises.

*A pure read* of one existing product plus Parts 9 and 12's globals → `kappa_m25_accretion{.png,.pdf}`,
`kappa_m25_accretion.csv` and `kappa_m25_accretion_galaxies.csv` (one row per galaxy).


```python
# ── Part 12d — the accretion history behind the split: is the black hole the upstream cause? ──
# Parts 12b/12c leave the mechanism open. What they establish is a TIMING: the two classes are indistinguishable at the AGN
# ignition and part company afterwards, and the parting is not in the jet leg (ignition -> the X-ray gate opening takes
# 0.58 Gyr for the rotators against 0.51 for the rest) but in what happens AFTER the gate opens — the rest truncate as it
# opens (t_SFT - t_x1 = -0.04 Gyr) while the rotators hold out (+1.47 Gyr). The obvious upstream variable is the one neither
# part measures: THE ACCRETION RATE ITSELF. This cell tests it, on the hypothesis that
#     a fast accretor moves quickly from the jet mode to the X-ray mode, quenches quickly and leaves its disc intact, while a
#     slow accretor sits in the jet mode long enough to stir the disc before the X-ray channel ever fires.
# THE THREE THINGS IT NEEDS, none of which is in Part 9's chains (CH9 keeps only the state code and the BHAR):
#   f_Edd itself, M_BH, and SIMBA's own gas gate f_gas = M_gas/M_* — so this cell makes ONE pass over the histories file
#   Part 3 and Part 9 already read (ism_prediction_histories_{box}.hdf5), restricted to the Part 9 chains. It reads no
#   catalogues and no particle files, and the AGN states come from Part 0's agn_states() VERBATIM, so the gates here are the
#   same gates Parts 3, 4, 9 and 12 date their stages with: jets at M_BH > 10^JET_LOGMBH and f_Edd < JET_FEDD, full velocity
#   at w_jet >= W_FULL (f_Edd <= 0.02), the X-ray channel on top of that at f_gas < XRAY_FGAS_MAX.
# WHAT IS MEASURED, per galaxy, on the branch's own sampling (dt = gradient(t), as Part 3 weights it):
#   (i)  THE ACCRETION: log f_Edd at the ignition, and its median before the ignition / on the jet leg / after the gate; the
#        peak f_Edd before the truncation; the time spent ABOVE the jet threshold (the radiative mode) and the time spent in
#        the jet state; log M_BH at the ignition and the growth Delta log M_BH from the gate to the anchor.
#   (ii) THE DUTY CYCLE, which is what the hypothesis is really about: the fraction of the post-gate window with the X-ray
#        channel actually firing (duty_post), the same on the SURVIVAL window t_x1 -> t_SFT (duty_surv), the number of
#        contiguous firing episodes, the longest firing-free gap, and the quiet time between the last firing and the anchor.
#   (iii) THE THING TO EXPLAIN: delay = t_SFT - t_x1, the 1.5 Gyr the rotators survive the open channel.
# AND THE TEST IS NOT THE CLASS CONTRAST. A contrast over ~13 rotators is scored here with the same hand-rolled
# Benjamini-Hochberg as Part 12c, and it is reported honestly, but the question the hypothesis actually asks is a
# CORRELATION on all the galaxies at once: does the accretion rate, or the duty cycle, PREDICT the delay? That is scored
# with Spearman on the pooled sample and within each class, which needs no class balance at all.
# WHAT WOULD FALSIFY THE HYPOTHESIS: f_Edd at the ignition the same in the two classes AND uncorrelated with the delay. What
# would confirm it: the rotators accreting faster at the ignition, and the delay falling as f_Edd rises.
# A PURE READ of one existing product plus Parts 9 and 12's globals; it writes one figure and one table.
P12D_FDR   = 0.05          # the Benjamini-Hochberg level the class-contrast grid is controlled at
P12D_NBOOT = 2000
P12D_TB    = np.arange(-2.0, 3.001, 0.25)   # the t - t_x1 clock the f_Edd tracks are drawn on [Gyr]
P12D_NMIN  = 3             # galaxies a drawn point needs
P12D_FEMIN = 1e-6          # the log floor of f_Edd: a branch step with no accretion is a real zero, not a missing value
# the scalars scored for the class contrast, in the order they are printed and drawn
P12D_QTY = [("lfedd_ign",  r"$\log f_{\rm Edd}$ at the ignition"),
            ("lfedd_jet",  r"$\log f_{\rm Edd}$ on the jet leg"),
            ("lfedd_post", r"$\log f_{\rm Edd}$ after the gate"),
            ("lfedd_pk",   r"peak $\log f_{\rm Edd}$ before the SFT"),
            ("t_quasar",   "time above the jet threshold [Gyr]"),
            ("t_jetstate", "time in the jet state [Gyr]"),
            ("duty_post",  "X-ray duty cycle after the gate"),
            ("duty_surv",  "X-ray duty cycle, gate -> SFT"),
            ("n_epi",      "firing episodes after the gate"),
            ("gap_max",    "longest firing-free gap [Gyr]"),
            ("lmbh_ign",   r"$\log M_{\rm BH}$ at the ignition"),
            ("dlmbh_post", r"$\Delta \log M_{\rm BH}$, gate -> anchor")]

if "KEEP9" not in globals() or "ST12" not in globals():
    print("Part 12d: needs Part 9 (the chains) and Part 12 (the critical points) — run them first.")
elif not os.path.exists(HIST_FILE.format(box=P9_BOX)):
    print(f"Part 12d: {HIST_FILE.format(box=P9_BOX)} is missing — paper_ism_prediction_boxes Part 6 writes it.")
else:
    # ── (A) ONE pass over the histories: f_Edd, M_BH, f_gas and the state masks on the retained chains ──
    A12D = {}
    with h5py.File(HIST_FILE.format(box=P9_BOX), "r") as _f12d:
        for _gn12d in _f12d:
            _a12d = int(_gn12d.replace("snap", ""))
            if not any(k[0] == _a12d for k in KEEP9):
                continue
            _g12d = _f12d[_gn12d]
            _t12d = _g12d["t_yr"][:] / 1e9; _o12d = np.argsort(_t12d); _t12d = _t12d[_o12d]
            if len(_t12d) < 3:
                continue
            _ms12d, _sfr12d, _mg12d, _mbh12d, _fe12d = (_g12d[c][:][_o12d]
                                                        for c in ("mstar", "sfr", "mgas", "mbh", "fedd"))
            _S12d, _D12d, _X12d = agn_states(_ms12d, _mg12d, _mbh12d, _fe12d)      # Part 0's gates, verbatim
            for _j12d, _gid12d in enumerate(np.asarray(_g12d["gal_ids"][:]).astype(np.int64)):
                _k12d = (_a12d, int(_gid12d))
                if _k12d not in KEEP9:
                    continue
                _ex12d = _X12d["exist"][:, _j12d]
                if _ex12d.sum() < 3:
                    continue
                _m = lambda a: np.nan_to_num(a[:, _j12d]) > 0.5
                A12D[_k12d] = dict(
                    t=_t12d, dt=np.gradient(_t12d), ex=_ex12d,
                    fedd=np.where(_ex12d, _fe12d[:, _j12d], np.nan),
                    mbh=np.where(_ex12d, _mbh12d[:, _j12d], np.nan),
                    fgas=np.where(_ex12d, _X12d["fgas"][:, _j12d], np.nan),
                    xr=_m(_S12d["xray"]) & _ex12d, jet=_m(_D12d["jet"]) & _ex12d,
                    rad=_m(_S12d["radiative"]) & _ex12d, chk=_m(_S12d["choked"]) & _ex12d,
                    dor=_m(_S12d["dormant"]) & _ex12d)
    print(f"Part 12d — the accretion history of {len(A12D)} of the {len(KEEP9)} retained {BOX_LABEL[P9_BOX]} chains "
          f"(SIMBA's gates: jets at log M_BH > {JET_LOGMBH:g} and f_Edd < {JET_FEDD:g}, full velocity at f_Edd <= "
          f"{JET_FEDD * 10 ** -W_FULL:.3f}, the X-ray channel on top of that at f_gas < {XRAY_FGAS_MAX:g})")

    # ── (B) the per-galaxy scalars ──
    def _at12d(d, tq):
        """The branch node nearest in time to tq — the value AT a stage, never an interpolation across a gap."""
        if not np.isfinite(tq) or not d["ex"].any():
            return -1
        _i = np.flatnonzero(d["ex"])
        return int(_i[np.argmin(np.abs(d["t"][_i] - tq))])

    def _w12d(d, m):
        """Time [Gyr] carried by a mask, on the branch's own sampling."""
        return float(np.sum(d["dt"][m & d["ex"]])) if (m & d["ex"]).any() else 0.0

    def _lf12d(d, m):
        """Median log f_Edd over a window, floored: a step with no accretion is a real zero."""
        v = d["fedd"][m & d["ex"]]
        v = v[np.isfinite(v)]
        return float(np.log10(np.median(np.clip(v, P12D_FEMIN, None)))) if len(v) else np.nan

    R12D = []
    for _k12d, _d12d in A12D.items():
        _t = _d12d["t"]; _ex = _d12d["ex"]
        _tig, _tx1 = ST12[_k12d]["agn_ign"], ST12[_k12d]["xray_on"]
        _tsf, _tan = ST12[_k12d]["sft"], ST12[_k12d]["anchor"]
        _pre = _ex & (_t < _tig) if np.isfinite(_tig) else np.zeros(len(_t), bool)
        _leg = _ex & (_t >= _tig) & (_t < _tx1) if np.isfinite(_tig) and np.isfinite(_tx1) else np.zeros(len(_t), bool)
        _pst = _ex & (_t >= _tx1) if np.isfinite(_tx1) else np.zeros(len(_t), bool)
        _srv = _pst & (_t < _tsf) if np.isfinite(_tsf) else np.zeros(len(_t), bool)
        _bsf = _ex & (_t <= _tsf) if np.isfinite(_tsf) else _ex
        _ii, _ix = _at12d(_d12d, _tig), _at12d(_d12d, _tx1)
        _fe = _d12d["fedd"]; _pk = np.nanmax(_fe[_bsf]) if _bsf.any() and np.isfinite(_fe[_bsf]).any() else np.nan
        # the firing episodes after the gate: contiguous runs of the X-ray mask, and the longest gap between them
        _xp = _d12d["xr"][_pst]; _tp = _t[_pst]
        _nep = int(np.sum(_xp[1:] & ~_xp[:-1]) + (1 if len(_xp) and _xp[0] else 0))
        if _xp.any():
            _on = _tp[_xp]
            _edges = np.concatenate([[_tp[0]], _on, [_tp[-1]]])
            _gap = float(np.max(np.diff(_edges))) if len(_edges) > 1 else 0.0
        else:
            _gap = float(_tp[-1] - _tp[0]) if len(_tp) > 1 else np.nan
        _lmb = lambda i: float(np.log10(_d12d["mbh"][i])) if (i >= 0 and np.isfinite(_d12d["mbh"][i])
                                                              and _d12d["mbh"][i] > 0) else np.nan
        _ia = _at12d(_d12d, _tan)
        R12D.append(dict(
            gal=_k12d, kcls=CLS9.get(_k12d),
            delay=(_tsf - _tx1) if np.isfinite(_tsf) and np.isfinite(_tx1) else np.nan,
            lead=(_tsf - _tig) if np.isfinite(_tsf) and np.isfinite(_tig) else np.nan,
            jetleg=(_tx1 - _tig) if np.isfinite(_tx1) and np.isfinite(_tig) else np.nan,
            lfedd_ign=(float(np.log10(max(_fe[_ii], P12D_FEMIN))) if _ii >= 0 and np.isfinite(_fe[_ii]) else np.nan),
            lfedd_pre=_lf12d(_d12d, _pre), lfedd_jet=_lf12d(_d12d, _leg), lfedd_post=_lf12d(_d12d, _pst),
            lfedd_pk=(float(np.log10(max(_pk, P12D_FEMIN))) if np.isfinite(_pk) else np.nan),
            t_quasar=_w12d(_d12d, _bsf & np.isfinite(_fe) & (np.nan_to_num(_fe) >= JET_FEDD)),
            t_jetstate=_w12d(_d12d, _bsf & _d12d["jet"]),
            duty_post=(_w12d(_d12d, _pst & _d12d["xr"]) / _w12d(_d12d, _pst)) if _w12d(_d12d, _pst) > 0 else np.nan,
            duty_surv=(_w12d(_d12d, _srv & _d12d["xr"]) / _w12d(_d12d, _srv)) if _w12d(_d12d, _srv) > 0 else np.nan,
            n_epi=float(_nep), gap_max=_gap,
            t_since_x=(float(_tan - _t[_pst & _d12d["xr"]][-1]) if (_pst & _d12d["xr"]).any()
                       and np.isfinite(_tan) else np.nan),
            lmbh_ign=_lmb(_ii), dlmbh_post=(_lmb(_ia) - _lmb(_ix)) if (_ia >= 0 and _ix >= 0) else np.nan,
            fgas_ign=(float(_d12d["fgas"][_ii]) if _ii >= 0 else np.nan),
            t_dor=_w12d(_d12d, _bsf & _d12d["dor"]), t_rad=_w12d(_d12d, _bsf & _d12d["rad"]),
            t_chk=_w12d(_d12d, _bsf & _d12d["chk"]), t_xr=_w12d(_d12d, _bsf & _d12d["xr"])))
    D12D = pd.DataFrame(R12D)
    ROW12D = []

    # ── (C) the class contrast, controlled at P12D_FDR — reported, but NOT the test ──
    print(f"\n(a) THE CLASS CONTRAST, Benjamini-Hochberg at q = {P12D_FDR:g}")
    _cells12d = []
    for _q, _lab in P12D_QTY:
        if _q not in D12D:
            continue
        _au, _p, _na, _nb = mw_auc(D12D[_q].to_numpy(float), D12D.kcls.eq(KC[0]).to_numpy(bool))
        if np.isfinite(_p):
            _cells12d.append(dict(q=_q, lab=_lab, auc=_au, p=_p, na=_na, nb=_nb,
                                  ma=float(np.nanmedian(D12D.loc[D12D.kcls.eq(KC[0]), _q].to_numpy(float))),
                                  mb=float(np.nanmedian(D12D.loc[~D12D.kcls.eq(KC[0]), _q].to_numpy(float)))))
    _cells12d.sort(key=lambda c: c["p"])
    _m12d = len(_cells12d)
    print(f"  {_m12d} of the {len(P12D_QTY)} scalars are scorable (a scalar the branch never defines is skipped)")
    for _r, _c in enumerate(_cells12d, 1):
        _thr = P12D_FDR * _r / _m12d
        _c["bh"] = bool(_c["p"] <= _thr)
        print(f"  {_c['lab']:<42} {KCLS[KC[0]]['short'][:9]:>9} {_c['ma']:8.3f}   rest {_c['mb']:8.3f}   "
              f"AUC {_c['auc']:.2f}  p {_c['p']:.4f}  N {_c['na']},{_c['nb']}   rank {_r:2d}/{_m12d}  "
              f"BH {_thr:.4f}   {'YES' if _c['bh'] else 'no'}")
        ROW12D.append(dict(block="contrast", quantity=_c["q"], kcls="rot-vs-iso", med=_c["ma"], lo=_c["mb"],
                           n_gal=_c["na"] + _c["nb"], auc=_c["auc"], p=_c["p"], bh=_c["bh"]))
    print(f"  -> {sum(c['bh'] for c in _cells12d)} of {_m12d} survive. With {int(D12D.kcls.eq(KC[0]).sum())} "
          f"{KCLS[KC[0]]['short']} galaxies, a single cell is suggestive at best — (b) is the test.")

    # ── (D) THE TEST: does the accretion, or the duty cycle, PREDICT the delay? No class balance needed ──
    print(f"\n(b) THE HYPOTHESIS, as a correlation on every galaxy at once: what predicts delay = t_SFT - t_x1?")
    print(f"  {'quantity':<42} {'all':>18}   {KCLS[KC[0]]['short'][:13]:>18}   {'the rest':>18}")
    for _q, _lab in P12D_QTY + [("lead", "ignition -> truncation [Gyr]"), ("jetleg", "the jet leg [Gyr]")]:
        if _q not in D12D:
            continue
        _out = []
        for _cl, _sub in (("all", D12D), (KC[0], D12D[D12D.kcls.eq(KC[0])]), ("rest", D12D[~D12D.kcls.eq(KC[0])])):
            _dd = _sub.dropna(subset=[_q, "delay"])
            if len(_dd) < 6:
                _out.append("      n too small"); continue
            _xa, _ya = _dd[_q].to_numpy(float), _dd["delay"].to_numpy(float)
            if np.nanstd(_xa) == 0 or np.nanstd(_ya) == 0:
                _out.append(f"    constant ({len(_dd):2d})"); continue
            _rr = spearmanr(_xa, _ya)
            _out.append(f"{_rr.statistic:+.2f} (p {_rr.pvalue:.3f}, {len(_dd):2d})")
            ROW12D.append(dict(block="predict_delay", quantity=_q, kcls=_cl, n_gal=len(_dd),
                               med=float(_rr.statistic), p=float(_rr.pvalue)))
        print(f"  {_lab:<42} " + "   ".join(f"{o:>18}" for o in _out))
    print("  (the hypothesis predicts a NEGATIVE rho for f_Edd — a faster accretor truncating sooner after the gate — and a "
          "positive one for the duty cycle)")

    # ── (E) FIG 5: the accretion, the duty cycle, and what they buy ──
    fig, _AX12D = plt.subplots(2, 3, figsize=(16.4, 7.6))
    ax = _AX12D[0, 0]                                                  # (a) f_Edd on the t - t_x1 clock
    for _cl in KC:
        _A = []
        for _k12d, _d12d in A12D.items():
            if CLS9.get(_k12d) != _cl or not np.isfinite(ST12[_k12d]["xray_on"]):
                continue
            _v = np.log10(np.clip(np.where(_d12d["ex"], _d12d["fedd"], np.nan), P12D_FEMIN, None))
            _A.append(to_grid(_d12d["t"] - ST12[_k12d]["xray_on"], _v, grid=P12D_TB))
        if len(_A) < P12D_NMIN:
            continue
        _c, _l, _h, _n = med_band(np.array(_A, float), nmin=P12D_NMIN, nboot=P12D_NBOOT)
        ax.fill_between(P12D_TB, _l, _h, color=KCLS[_cl]["color"], alpha=0.20, lw=0)
        ax.plot(P12D_TB, _c, color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_A)})")
        for _x, _y, _lo, _hi, _nn in zip(P12D_TB, _c, _l, _h, _n):
            ROW12D.append(dict(block="fedd_clock", kcls=_cl, x=float(_x), med=float(_y), lo=float(_lo),
                               hi=float(_hi), n_gal=int(_nn)))
    ax.axhline(np.log10(JET_FEDD), color="#7b3294", lw=1.0, ls="--")
    ax.annotate("the jet threshold", xy=(0.02, np.log10(JET_FEDD)), xycoords=("axes fraction", "data"),
                xytext=(0, 3), textcoords="offset points", fontsize=7.5, color="#7b3294")
    ax.axhline(np.log10(JET_FEDD) - W_FULL, color="#008837", lw=1.0, ls=":")
    ax.annotate("full velocity", xy=(0.02, np.log10(JET_FEDD) - W_FULL), xycoords=("axes fraction", "data"),
                xytext=(0, 3), textcoords="offset points", fontsize=7.5, color="#008837")
    ax.axvline(0.0, color="0.3", lw=1.2, ls="-")
    ax.set_xlabel(r"$t - t_{\rm x1}$ [Gyr]   (0 = the X-ray gate opens)"); ax.set_ylabel(r"$\log f_{\rm Edd}$")
    ax.set_title("(a) the accretion rate through the gate", loc="left", fontsize=10)
    _leg12b(ax, loc="lower left"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX12D[0, 1]                                                  # (b) the duty cycle
    for _cl in KC:
        _v = D12D[D12D.kcls.eq(_cl)].duty_post.dropna().to_numpy(float)
        if len(_v) < P12D_NMIN:
            continue
        ax.plot(np.sort(_v), np.arange(1, len(_v) + 1) / len(_v), drawstyle="steps-post",
                color=KCLS[_cl]["color"], lw=KCLS[_cl]["lw"], label=f"{KCLS[_cl]['short']}  ({len(_v)})")
        ROW12D.append(dict(block="duty", kcls=_cl, n_gal=len(_v), med=float(np.median(_v))))
    ax.set_xlim(0, 1); ax.set_ylim(0, 1.02)
    ax.set_xlabel("fraction of the post-gate window with the X-ray channel firing")
    ax.set_ylabel("cumulative fraction of the class")
    ax.set_title("(b) once open, does it fire continuously?", loc="left", fontsize=10)
    _leg12b(ax, loc="upper left"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX12D[0, 2]                                                  # (c) what buys the delay
    for _cl in KC:
        _d = D12D[D12D.kcls.eq(_cl)].dropna(subset=["duty_post", "delay"])
        ax.scatter(_d.duty_post, _d.delay, s=26, alpha=0.8, color=KCLS[_cl]["color"], edgecolor="k",
                   linewidth=0.3, label=f"{KCLS[_cl]['short']}  ({len(_d)})")
    _dd = D12D.dropna(subset=["duty_post", "delay"])
    if len(_dd) >= 8 and np.nanstd(_dd.duty_post) > 0 and np.nanstd(_dd.delay) > 0:
        _rr = spearmanr(_dd.duty_post.to_numpy(float), _dd.delay.to_numpy(float))
        ax.annotate(rf"$\rho = {_rr.statistic:+.2f}$  (p {_rr.pvalue:.3f})", xy=(0.03, 0.96),
                    xycoords="axes fraction", va="top", fontsize=9, color="0.25")
    ax.axhline(0.0, color="0.3", lw=1.2, ls="--")
    ax.annotate("truncation AT the gate", xy=(0.97, 0.0), xycoords=("axes fraction", "data"), xytext=(0, 4),
                textcoords="offset points", fontsize=7.5, color="0.3", ha="right")
    ax.set_xlabel("X-ray duty cycle after the gate"); ax.set_ylabel(r"$t_{\rm SFT} - t_{\rm x1}$ [Gyr]")
    ax.set_title("(c) how long the galaxy survives the open channel", loc="left", fontsize=10)
    _leg12b(ax, loc="upper right"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX12D[1, 0]                                                  # (d) the time budget of the AGN states
    _ST12D = [("t_dor", "dormant", "#bdbdbd"), ("t_rad", "radiative", "#2166ac"),
              ("t_chk", "jet, choked", "#e08214"), ("t_xr", "X-ray on", "#b2182b")]
    _w12, _xs = 0.34, np.arange(len(_ST12D))
    for _ci, _cl in enumerate(KC):
        _sub = D12D[D12D.kcls.eq(_cl)]
        _v = [float(np.nanmedian(_sub[c].to_numpy(float))) if len(_sub) else np.nan for c, _, _ in _ST12D]
        ax.bar(_xs + (_ci - 0.5) * _w12, _v, _w12, color=KCLS[_cl]["color"], alpha=0.85, edgecolor="k",
               linewidth=0.4, label=f"{KCLS[_cl]['short']}  ({len(_sub)})")
        for _c, _vv in zip([c for c, _, _ in _ST12D], _v):
            ROW12D.append(dict(block="state_time", kcls=_cl, quantity=_c, n_gal=len(_sub), med=_vv))
    ax.set_xticks(_xs); ax.set_xticklabels([l for _, l, _ in _ST12D], fontsize=8.5)
    ax.set_ylabel("median time before the SFT [Gyr]")
    ax.set_title("(d) where the pre-truncation time is spent", loc="left", fontsize=10)
    _leg12b(ax, loc="upper left"); ax.grid(alpha=0.2, lw=0.5, axis="y")
    ax = _AX12D[1, 1]                                                  # (e) the black hole itself
    for _cl in KC:
        _d = D12D[D12D.kcls.eq(_cl)].dropna(subset=["lmbh_ign", "lfedd_ign"])
        ax.scatter(_d.lmbh_ign, _d.lfedd_ign, s=26, alpha=0.8, color=KCLS[_cl]["color"], edgecolor="k",
                   linewidth=0.3, label=f"{KCLS[_cl]['short']}  ({len(_d)})")
    ax.axvline(JET_LOGMBH, color="#7b3294", lw=1.0, ls="--")
    ax.annotate("the jet mass gate", xy=(JET_LOGMBH, 0.02), xycoords=("data", "axes fraction"), xytext=(3, 0),
                textcoords="offset points", fontsize=7.5, color="#7b3294", rotation=90)
    ax.axhline(np.log10(JET_FEDD), color="0.4", lw=1.0, ls=":")
    ax.set_xlabel(r"$\log M_{\rm BH}$ at the ignition"); ax.set_ylabel(r"$\log f_{\rm Edd}$ at the ignition")
    ax.set_title("(e) the hole the two classes ignite with", loc="left", fontsize=10)
    _leg12b(ax, loc="lower left"); ax.grid(alpha=0.2, lw=0.5)
    ax = _AX12D[1, 2]                                                  # (f) the contrast, as a forest
    _cc = [c for c in _cells12d if np.isfinite(c["auc"])][::-1]
    if _cc:
        _y = np.arange(len(_cc))
        ax.barh(_y, [c["auc"] - 0.5 for c in _cc], left=0.5, height=0.62,
                color=["#b2182b" if c["bh"] else "0.72" for c in _cc], edgecolor="k", linewidth=0.4)
        for _yy, _c in zip(_y, _cc):
            ax.annotate(f"p {_c['p']:.3f}", xy=(1.0, _yy), xytext=(-3, 0), textcoords="offset points",
                        fontsize=7, va="center", ha="right", color="0.3")
        ax.set_yticks(_y); ax.set_yticklabels([c["lab"] for c in _cc], fontsize=7.5)
        ax.set_xlim(0, 1); ax.axvline(0.5, color="0.3", lw=1.2, ls="--")
    ax.set_xlabel(f"AUC ({KCLS[KC[0]]['short']} > the rest)")
    ax.set_title(f"(f) the contrast, red = survives BH at q = {P12D_FDR:g}", loc="left", fontsize=10)
    ax.grid(alpha=0.2, lw=0.5, axis="x")
    fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the accretion history behind the kinematic split",
                 y=0.995, fontsize=12, fontweight="bold")
    fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_accretion"); plt.show()
    pd.DataFrame(ROW12D).to_csv(os.path.join(OUT, "kappa_m25_accretion.csv"), index=False)
    D12D.assign(gal=[f"{a}_{g}" for a, g in D12D.gal]).to_csv(
        os.path.join(OUT, "kappa_m25_accretion_galaxies.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_m25_accretion.csv')}"
          f"\n         {os.path.join(OUT, 'kappa_m25_accretion_galaxies.csv')} (one row per galaxy)")
```

## Part 13 — the ejection and the heating: the CGM on the clock of the X-ray onset

**The hypothesis under test** (stated as a chain, so that each link can fail on its own):

| | claim | window |
|---|---|---|
| **H1** | *before* the jet / X-ray phase, the galaxies that end up **dust-poor** had already pushed more gas — and more dust — out of the ISM into the CGM, and **at higher velocity** | before |
| **H2** | *after* the X-ray channel fires, the AGN **heats** that CGM | after |
| **H3** | the heat **destroys** the grains out there — the CGM dust-to-gas ratio falls, rather than merely diluting | after |
| **H4** | the heating shuts the **cold inflow** off, so nothing comes back | after |
| **H5** | in the other case little was displaced, the cold gas keeps arriving, and $f_{\rm gas}$ stays **above SIMBA's own X-ray gate** — the channel is *choked* rather than firing | after |
| **H6** | what is left there is a **compact ISM**, in which dust can re-form | after |
| **H7** | what actually separates them is not *how much* left but whether it left **bound**: material pushed out below the local escape velocity is a **fountain** that comes back with its dust, material pushed out above it is gone | before |

**The clock is therefore $t - t_{x1}$**, the first step with the X-ray channel on (Part 12's `xray_on`, read off Part 9's state
code) — not $t_{\rm SFT}$. The hypothesis is a statement about before and after *that* event, so that event has to be the
origin. Every claim is scored in a window **before** the onset (`P13_PRE`) and one **after** it (`P13_POST`), separately: if
the difference H1 needs before the onset is only there afterwards, the causal order fails and the scorecard says so.

**What is measured**, from the same reduced particle files Part 10 reads, in a cache of its own (Part 10's is keyed on the
columns it already holds, and a new column there would mean deleting 7 GiB — so this part keeps its census beside it):

* **the CGM in phases** — cold ($T <$ `P13_TCOLD`), warm, hot ($T >$ `P13_THOT`) — the gas mass, the dust mass and the count
  of each, in four radial shells, with the mass-weighted **and the dust-weighted** mean temperature. The second one is the
  point of H3: in one m25 galaxy the CGM sits at $\langle\log T\rangle_M = 5.9$ while its surviving dust sits at $4.5$ — the
  grains are in the cold phase, and what the heat does is remove that phase.
* **the flux through a fixed shell**, split into inflow and outflow *and* into cold and total, for the gas and for the dust:
  the cold inflow of H4 is a measured number, not an inference. (It counts every particle in the shell, member or not — a
  grain does not know the membership list.)
* **the ISM compactness** — the gas and the dust inside 1 kpc, and $R_{50}$ of the member gas — so H6 and Part 12's hole are
  two views of one claim.
* **the escape velocity at every particle's radius**, for H7. It *cannot* be measured from the cutouts: they hold gas and
  stars only, and in a halo the dark matter is most of the mass — so the potential comes from the **caesar halo** the galaxy
  sits in ($M_{200c}$, $R_{200c}$ of `parent_halo_index`, an NFW profile, the concentration of Dutton & Macciò 2014), with
  the baryons the cutout *does* resolve added as $-\Phi_b(r) = G\,[\,M(<r)/r + \sum_{r_i > r} m_i/r_i\,]$ — the second
  term matters, since $G M(<r)/r$ alone goes to zero at the centre where the potential is deepest. `P13_VESC_TO` sets the
  convention: `"r200"` (the default) is the speed needed to *leave the halo*, `"inf"` the textbook escape to infinity.
* **and the apocentre of whatever is moving out** — because in these halos $v_{\rm esc}(10\ {\rm kpc}) \approx 900$–$1100$
  km s$^{-1}$ and almost nothing a quenching galaxy ejects is formally unbound, so the bound/unbound *binary* is nearly
  degenerate and the informative number is **how far out it gets before it turns around**: solve
  $-\Phi(r_{\rm apo}) = -\Phi(r) - v^2/2$ for a radial orbit in the same potential (angular momentum would only bring it
  back sooner, so this is the outer envelope), censored at $5\,R_{200c}$.

**The split is the outcome**, the dust label at the anchor (`P13_SPLIT = "pop"`), not Part 2's kinematic selection. That is
legitimate here and it is not what Part 2 does: this part is not *predicting* the label, it is asking what happened to the
galaxies that carry it. `"kcls"` and Part 12's `"hole"` are the other two settings, and the scorecard follows whichever is on.

**One thing is true by construction** and must not be read as a result: $f_{\rm gas}$ crosses SIMBA's gate at $t = 0$, because
that gate is part of the definition of $t_{x1}$. What is *not* by construction is whether a group climbs back **over** the
gate afterwards — that is H5, and it is the panel to read.

| figure | what it shows |
|---|---|
| **1 the ejection** | $M_{\rm dust}^{\rm CGM}/M_\star$, the share of the dust that is outside, the outflow speed of the outer ISM, and the cumulative dust that crossed 10 kpc |
| **2 the heating** | $\langle\log T\rangle_M$ of the CGM, its cold mass fraction, the temperature the surviving grains sit at, and the CGM dust-to-gas ratio against the ISM's own |
| **3 the return** | the cold inflow at the shell, $f_{\rm gas}$ against the gate, the share of each group *firing* vs *choked*, and how much of the ISM is in the central kpc |
| **1b the escape** | H7: $v_r/v_{\rm esc}$ of the outer ISM, the apocentre the ejected dust will reach, the dust returning through the shell against the dust leaving it, and the cumulative dust that left **bound** against the dust that left **unbound** |
| **4 the test** | the per-galaxy scalars — what it did before the onset, what its CGM did after — against the dust it actually ended with, with Spearman $\rho$ |

**The scorecard** is written in the vocabulary of the *outcome*: a claim that says `"dusty"` means *whichever group of the
split in force ends up dust-rich* — the dusty QGs under `"pop"`, the rotating class under `"kcls"`, the hole galaxies under
`"hole"` (Part 12 measured that those end ~1.8 dex dustier). It prints every claim with its two group medians, the AUC, the same AUC pooled within terciles of
$\log M_\star$ (the mass is the obvious confound: a more massive halo has a hotter CGM whatever the AGN did), the
Mann–Whitney $p$, and whether the direction is the one the hypothesis predicts.


```python
# ── Part 13 — THE EJECTION AND THE HEATING: the CGM on the clock of the X-ray onset, and the hypothesis it tests ──
# THE HYPOTHESIS (the user's, 2026-09-09), as a chain of statements each of which can be false on its own:
#   H1  BEFORE the jet / X-ray phase, the galaxies that end up dust-POOR had already pushed more gas — and more dust — out of
#       the ISM and into the CGM, and at higher velocity.
#   H2  AFTER the X-ray channel fires, the AGN heats that CGM.
#   H3  the heat DESTROYS the dust that is out there (the dust-to-gas ratio of the CGM falls; it does not merely dilute).
#   H4  the heating shuts the COLD INFLOW off, so nothing comes back.
#   H5  in the other case the ISM was compact, little of it was displaced, and the cold gas that keeps arriving holds
#       f_gas above SIMBA's own X-ray gate — the channel is CHOKED rather than firing.
#   H6  what is left is a compact ISM in which dust can re-form.
#   H7  THE VELOCITY THAT DECIDES IT (the question of 2026-09-09): what separates the two may not be HOW MUCH gas left the
#       ISM but whether it left BOUND. Material pushed out BELOW the local escape velocity is a fountain — it stays in the
#       halo, cools, and rains back with its dust; material pushed out ABOVE it is gone, and with it the reservoir the dust
#       would re-form from. The dustier galaxies should then be the ones whose ejecta sat below v_esc, and the two groups
#       can push the SAME mass out and still end up on opposite sides of this.
# ONE THING IS TRUE BY CONSTRUCTION and must not be read as a result: f_gas crosses SIMBA's gate at t = 0, because that gate
# is part of the definition of t_x1. What is NOT by construction is whether a group climbs back OVER the gate afterwards —
# that is H5, and it is the panel to read.
# THE CLOCK IS THEREFORE t - t_x1, the FIRST STEP WITH THE X-RAY CHANNEL ON (Part 12's `xray_on`, off Part 9's state code),
# not t_SFT: the hypothesis is a statement about before and after that event, so the event has to be the origin. Every claim
# is measured in a window BEFORE the onset (P13_PRE) and one AFTER it (P13_POST) and scored separately — if the difference
# the hypothesis needs before the onset is only there after it, the causal order fails and the scorecard says so.
# WHAT IS MEASURED, from the same reduced particle files Part 10 reads, in its own cache (Part 10's is keyed on the columns it
# already has; a new column there would mean deleting 7 GiB, so this part keeps its own census beside it):
#   the CGM in PHASES — cold (T < P13_TCOLD), warm, hot (T > P13_THOT) — with the gas mass, the dust mass and the count of
#   each, in four radial shells, with the mass-weighted AND the DUST-WEIGHTED mean temperature. The second one is the whole
#   point of H3: in one m25 galaxy the CGM sits at <log T> = 5.9 while its surviving dust sits at 4.5 — the grains are in the
#   cold phase, and what the heat does is remove that phase.
#   the FLUX through a fixed shell, split into inflow and outflow AND into cold and total, for the gas and for the dust: the
#   cold inflow of H4 is a directly measured number, not an inference.
#   the ISM COMPACTNESS (the gas and the dust inside 1 kpc, and R50 of the member gas) for H6, so this part and Part 12's hole
#   are two views of the same claim.
#   the ESCAPE VELOCITY at every particle's radius, for H7. It CANNOT be measured from the cutout: the reduced files hold gas
#   and stars only, and in a halo the dark matter is most of the mass, so the potential comes from the CAESAR halo the galaxy
#   sits in — M200c and R200c of `parent_halo_index`, an NFW profile, and the concentration of Dutton & Maccio (2014) — with
#   the baryons the cutout DOES resolve added as the enclosed mass they are (P13_VESC_BAR; it double counts the halo's own
#   baryon share at large radius and is worth having anyway inside ~10 kpc, where the analytic NFW knows nothing about the
#   condensed gas and stars). P13_VESC_TO decides the convention: "r200" = the speed needed to reach R200c and leave the halo
#   (the default — a fountain is about coming back, and the halo is what it comes back to), "inf" = the textbook escape to
#   infinity, which is larger. Then a particle is BOUND if its full speed in the galaxy rest frame is below v_esc(r), and an
#   OUTFLOW is unbound if v_r >= v_esc(r) — the flux through the shell is split on exactly that, so "the dust that left as a
#   fountain" and "the dust that left as a wind" are two separately integrated numbers.
#   ONE CAVEAT that the `central` column carries: for a SATELLITE the parent halo is the group's, its centre is not the
#   galaxy's, and v_esc(r) evaluated at the galaxy-centric radius is then the group's escape velocity — too high, and at the
#   wrong origin. The census records the flag so the scorecard can be repeated on centrals alone.
# THE SPLIT is the OUTCOME the hypothesis is about — the dust label at the anchor (P13_SPLIT = "pop") — and not Part 2's
# kinematic selection; "kcls" and Part 12's "hole" are the other two settings, and the scorecard prints the AUC of whichever
# is in force. Splitting on the outcome is legitimate here and it is not what Part 2 does: this part is not predicting the
# label, it is asking what happened to the galaxies that carry it.
P13_DIR    = os.path.join(os.getcwd(), "output", P9_BOX, "reduced_particles")
P13_PREFIX = {"cis25": "m25n512", "cis50": "m50n512", "cis50nox": "m50n512", "cis100": "m100n1024"}.get(P9_BOX, P9_BOX)
P13_CACHE  = os.path.join(OUT, f"kappa_{P9_BOX}_cgm_phases.fits")   # the census cache (delete to rebuild; appended to)
P13_CLOCK  = "sft"          # "x1" = t - t_x1, the X-ray onset (THE clock of this part) | "sft" = the truncation, for comparison
P13_SPLIT  = "kcls"         # "pop" = the dust label at the anchor (the outcome) | "kcls" = Part 2's selection | "hole" = Part 12's
P13_TB     = np.arange(-3.0, 4.001, 0.5)      # the clock bin EDGES [Gyr] (binned, like Part 10: a node is a snapshot with a file)
P13_PRE    = (-1.5, -0.25)  # the window "before the onset" every claim is scored in [Gyr]
P13_POST   = (0.5, 2.5)     # ... and the one "after" it
P13_RMAX   = 100.0         # the CGM aperture [kpc] — the files' own
P13_TCOLD  = 1e5           # [K] cold / warm boundary: the phase the dust and the accretion live in
P13_THOT   = 1e6           # [K] warm / hot boundary
P13_RSH, P13_DRSH = 10.0, 2.0     # the shell the fluxes are measured through [kpc]
P13_SHELLS = [(0.0, 10.0), (10.0, 30.0), (30.0, 60.0), (60.0, 100.0)]   # the radial decomposition of the CGM
P13_RIN    = 0.8           # [kpc] "compact": the ISM gas and dust inside this radius
P13_VFAST  = 100.0         # [km/s] the outward tail, as in Part 10   # the ABSOLUTE threshold, kept beside the relative one
P13_VESC     = True        # H7: measure every velocity against the LOCAL escape velocity of the halo the galaxy sits in
P13_VESC_TO  = "r200"      # "r200" = escape to R200c, i.e. leave the halo (the default) | "inf" = escape to infinity
P13_VESC_BAR = True        # count the gas + stars the cutout resolves in the potential too (see the note above)
P13_CONC     = None        # the NFW concentration c200c; None = Dutton & Maccio (2014), c(M200c, z)
P13_XB       = 1.0         # the bound / unbound boundary in units of v_esc (1 = the escape velocity itself)
P13_HCACHE   = os.path.join(OUT, f"kappa_{P9_BOX}_halo.csv")   # (snap, gx) -> M200c, R200c, central: read once, cached
G_KPC13      = 4.300917270e-6     # G in kpc (km/s)^2 / Msun
P13_HUBFLOW, P13_OM = True, 0.3
P13_CUMSTART = -1.0        # [Gyr] a galaxy enters the CUMULATIVE ejection panel only if its branch is sampled from at least
                            # this far before the onset — otherwise the integral of two galaxies starts at different times
P13_NMIN, P13_NBOOT, P13_MTERC = 3, 300, 3
P13_MAXNODE = None         # a cap on the files read (None = all) — set it to smoke-test the cell
KMS_KPC_PER_YR13 = 1.0227121650537077e-9

if "CH9" not in globals() or "BR9" not in globals():
    print("Part 13: needs Part 9 — run it first (the chains, the state code and the branch index all come from it).")
elif not os.path.isdir(P13_DIR):
    print(f"Part 13: {P13_DIR} is missing — build_reduced_particles_job.py writes it (cluster).")
elif "st" not in next(iter(CH9.values())):
    print("Part 13: the chains carry no AGN state code — re-run Part 9 (its cell now stores 'st' and 'mdot').")
else:
    import glob as _glob13, re as _re13, time as _time13

    # ── (A) THE ORIGIN OF THE CLOCK: the first X-ray step of every retained galaxy ──
    T13 = {}
    for _k13 in KEEP9:
        _d13 = CH9[_k13]
        _x13 = _d13["t"][np.nan_to_num(_d13["st"], nan=-1) >= 3]
        T13[_k13] = (float(_x13[0]) if len(_x13) else np.nan) if P13_CLOCK == "x1" else _tsft9.get(_k13, np.nan)
    _CLAB13 = r"$t - t_{x1}$  [Gyr]" if P13_CLOCK == "x1" else r"$t - t_{\rm SFT}$  [Gyr]"
    _CNAM13 = "the X-ray onset" if P13_CLOCK == "x1" else "the truncation"
    _dtx13 = np.array([T13[k] - _tsft9.get(k, np.nan) for k in KEEP9], float)
    print(f"Part 13 — the clock is {'t - t_x1' if P13_CLOCK == 'x1' else 't - t_SFT'}: "
          f"{int(np.isfinite(np.array(list(T13.values()), float)).sum())}/{len(KEEP9)} of the retained {P9_BOX} galaxies "
          "have one"
          + (f";  it sits {np.nanmedian(_dtx13):+.2f} Gyr from their truncation in the median "
             f"[{np.nanpercentile(_dtx13, 16):+.2f}, {np.nanpercentile(_dtx13, 84):+.2f}]" if P13_CLOCK == "x1" else ""))

    # ── (B) THE GROUPS: the outcome the hypothesis is about ──
    _S13 = QM[QM["box"] == P9_BOX]
    _pop13 = {(int(a_), int(g_)): str(p_) for a_, g_, p_ in zip(_S13["snap"], _S13["gal_id"], _S13["pop"])}
    if P13_SPLIT == "pop":
        G13 = {k: _pop13.get(k) for k in KEEP9}
        GST13 = {g: dict(POPS[g], short=("dusty QG" if g == "dusty" else "dust-poor QG")) for g in ("dusty", "pure")}
        GK13, _SPLAB13 = ["dusty", "pure"], f"the dust label at the anchor (log M_dust/M* {DUSTY_CUT:g})"
    elif P13_SPLIT == "hole" and "HOLE12" in globals():
        G13, GST13, GK13 = dict(HOLE12), HCLS, HC
        _SPLAB13 = "Part 12's central hole"
    else:
        G13, GST13, GK13 = dict(CLS9), KCLS, KC
        _SPLAB13 = "Part 2's kinematic selection"
    G13 = {k: v for k, v in G13.items() if v in GK13 and np.isfinite(T13.get(k, np.nan))}
    print(f"  the split: {_SPLAB13} — " + ", ".join(f"{GST13[g]['short']} {sum(1 for v in G13.values() if v == g)}"
                                                    for g in GK13))

    # ── (C) THE CENSUS: the CGM in phases, the fluxes, the compactness — one row per (snap, gx), cached ──
    _inv13 = {}
    for _p13 in _glob13.glob(os.path.join(P13_DIR, "snap_*", f"{P13_PREFIX}_snap*_gal*.h5")):
        _m13 = _re13.search(r"_snap(\d+)_gal(\d+)\.h5$", _p13)
        if _m13:
            _inv13[(int(_m13.group(1)), int(_m13.group(2)))] = _p13
    NOD13 = BR9[[(int(a_), int(g_)) in G13 for a_, g_ in zip(BR9["anchor_snap"], BR9["gal_id"])]].copy()
    NOD13 = NOD13[[(int(s_), int(x_)) in _inv13 for s_, x_ in zip(NOD13["snap"], NOD13["gx"])]].reset_index(drop=True)
    _uniq13 = sorted({(int(s_), int(x_)) for s_, x_ in zip(NOD13["snap"], NOD13["gx"])})
    print(f"  {len(_inv13)} particle files on disk; {len(NOD13)} are nodes of the {len(G13)} galaxies of the split "
          f"({len(_uniq13)} distinct files)")

    def _wq13(v, w, q):
        ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
        if not ok.any():
            return np.nan
        v, w = v[ok], w[ok]
        o = np.argsort(v); v, w = v[o], w[o]
        return float(np.interp(q, (np.cumsum(w) - 0.5 * w) / w.sum(), v))

    def _conc13(m200, z, hub):
        """The NFW concentration c200c of a halo of M200c [Msun] at redshift z — Dutton & Maccio (2014), their relation for
        c200c against M200c in units of 10^12 h^-1 Msun. P13_CONC overrides it with a fixed value."""
        if P13_CONC is not None:
            return float(P13_CONC)
        _z = max(float(z), 0.0)
        _a = 0.520 + (0.905 - 0.520) * np.exp(-0.617 * _z ** 1.21)
        _b = -0.101 + 0.026 * _z
        return float(10.0 ** (_a + _b * np.log10(max(float(m200), 1.0) * float(hub) / 1e12)))

    def _psib13(re, rp, mp):
        """-Phi(r) [(km/s)^2] of the baryons the cutout actually resolves: G [ M(<r)/r + sum over the particles OUTSIDE r of
        m_i/r_i ]. The second term is not optional — G M(<r)/r alone goes to zero at the centre, where the potential is in
        fact deepest, and a profile built from it is not monotone and cannot be inverted for an apocentre."""
        o = np.argsort(rp); r_s = np.clip(np.asarray(rp, float)[o], 1e-3, None); m_s = np.nan_to_num(np.asarray(mp, float)[o])
        if not len(r_s):
            return np.zeros(np.shape(re))
        _min = np.cumsum(m_s); _cmr = np.cumsum(m_s / r_s); _tot = float(_cmr[-1])
        re = np.clip(np.asarray(re, float), 1e-3, None)
        j = np.searchsorted(r_s, re, side="right") - 1
        _ok = j >= 0
        _jc = np.clip(j, 0, len(r_s) - 1)
        inside = np.where(_ok, _min[_jc] / re, 0.0)
        outside = np.where(_ok, _tot - _cmr[_jc], _tot)
        return G_KPC13 * (inside + outside)

    def _vesc13(r, m200, r200, c, psib=None, psib0=0.0, to=None):
        """The escape speed [km/s] at physical radius r [kpc] from an NFW halo (M200c, R200c, concentration c).

        -Phi(r) = G M200 / (R200 mu(c)) ln(1 + c x) / x  with x = r/R200 and mu(y) = ln(1+y) - y/(1+y); v_esc = sqrt(2|Phi|).
        With P13_VESC_TO = "r200" the reference is Phi(R200) rather than 0, so what is measured is the speed needed to LEAVE
        THE HALO, not to reach infinity. `psib` is -Phi of the resolved baryons at the same radii (from _psib13), `psib0`
        its value at R200c — the reference the "r200" convention needs."""
        to = P13_VESC_TO if to is None else to
        _mu = np.log1p(c) - c / (1.0 + c)
        x = np.clip(np.asarray(r, float) / r200, 1e-4, None)
        phi = np.log1p(c * x) / x
        if to == "r200":
            phi = phi - np.log1p(c)                       # ... only as far as R200c, where the halo ends
        v2 = 2.0 * G_KPC13 * m200 / (r200 * _mu) * phi
        if psib is not None:
            v2 = v2 + 2.0 * (np.asarray(psib, float) - (float(psib0) if to == "r200" else 0.0))
        return np.sqrt(np.clip(v2, 0.0, None))

    def _halo13(keys):
        """{(snap, gx): (M200c [Msun], R200c [physical kpc], central)} for the nodes asked for, one catalogue open per
        snapshot. The particle files carry NO dark matter, so this is the only place the potential can come from."""
        try:
            cdir, fmt = _catcfg3(P9_BOX)
        except Exception as _e:
            return {}, [f"the catalogue directory is unknown ({type(_e).__name__}: {_e})"]
        _by, HAL, miss = {}, {}, []
        for _s, _g in keys:
            _by.setdefault(int(_s), []).append(int(_g))
        for _sn in sorted(_by):
            try:
                with h5py.File(os.path.join(cdir or "", fmt.format(snap=int(_sn))), "r") as f:
                    _sf = float(f["simulation_attributes"].attrs["scale_factor"])
                    _ph = f["galaxy_data/parent_halo_index"][:].astype(int)
                    _cen = (f["galaxy_data/central"][:].astype(bool) if "galaxy_data/central" in f
                            else np.zeros(len(_ph), bool))
                    _m2 = f["halo_data/dicts/virial_quantities.m200c"][:].astype(float)
                    _r2 = f["halo_data/dicts/virial_quantities.r200c"][:].astype(float) * _sf   # kpccm -> physical kpc
            except (OSError, KeyError) as _e:
                miss.append(f"snap {_sn}: {type(_e).__name__}")
                continue
            for _gx in _by[_sn]:
                if 0 <= _gx < len(_ph) and 0 <= _ph[_gx] < len(_m2) and _m2[_ph[_gx]] > 0 and _r2[_ph[_gx]] > 0:
                    HAL[(_sn, _gx)] = (float(_m2[_ph[_gx]]), float(_r2[_ph[_gx]]), bool(_cen[_gx]))
        return HAL, miss

    def _cgm13(path, halo=None):
        """ONE reduced particle file -> the phase, radial, kinematic and flux census of its CGM, and the ISM it came from."""
        with h5py.File(path, "r") as h:
            A = dict(h.attrs); G, St = h["gas"], h["star"]
            gp = G["pos"][:].astype(float); gmem = G["member"][:].astype(bool)
            gm = G["m_gas"][:].astype(float); gd = G["m_dust"][:].astype(float)
            gh2 = G["m_H2"][:].astype(float) if "m_H2" in G else np.full(len(gm), np.nan)
            gT = G["temp"][:].astype(float) if "temp" in G else np.full(len(gm), np.nan)
            gz = G["zgas"][:].astype(float) if "zgas" in G else np.full(len(gm), np.nan)
            gv = G["vel"][:].astype(float) if "vel" in G else np.full((len(gm), 3), np.nan)
            sp = St["pos"][:].astype(float); sm = St["m_star"][:].astype(float)
            smem = St["member"][:].astype(bool)
            sv = St["vel"][:].astype(float) if "vel" in St else np.full((len(sm), 3), np.nan)
        z, hub = float(A["redshift"]), float(A["hub"])
        r, rs = np.linalg.norm(gp, axis=1), np.linalg.norm(sp, axis=1)
        _sok = smem & np.isfinite(sv).all(1) & (sm > 0)
        _gok = gmem & np.isfinite(gv).all(1) & (gm > 0)
        vsys = (np.average(sv[_sok], axis=0, weights=sm[_sok]) if _sok.sum() >= 10 else
                np.average(gv[_gok], axis=0, weights=gm[_gok]) if _gok.sum() >= 10 else np.full(3, np.nan))
        _hz13 = ((100.0 * hub * np.sqrt(P13_OM * (1 + z) ** 3 + 1 - P13_OM)) / 1e3) if P13_HUBFLOW else 0.0   # km/s per kpc
        _v3 = (gv - vsys) + _hz13 * gp                     # the velocity in the galaxy's rest frame, the Hubble flow in it
        vr = (_v3 * (gp / np.clip(r, 1e-9, None)[:, None])).sum(1)
        vtot = np.linalg.norm(_v3, axis=1)                 # the SPEED, which is what "bound" is about
        vesc = np.full(len(gm), np.nan)                    # H7: the escape velocity at each particle's radius
        rapo = np.full(len(gm), np.nan)                    # ... and the apocentre of whatever is moving out
        _m200 = _r200 = _conc = np.nan
        _cen13 = False
        if P13_VESC and halo is not None and np.isfinite(halo[0]) and halo[0] > 0 and halo[1] > 0:
            _m200, _r200, _cen13 = float(halo[0]), float(halo[1]), bool(halo[2])
            _conc = _conc13(_m200, z, hub)
            _rb = np.concatenate([r, rs]); _mb = np.concatenate([np.nan_to_num(gm), np.nan_to_num(sm)])
            _pb13 = (lambda _x: _psib13(_x, _rb, _mb)) if P13_VESC_BAR else (lambda _x: None)
            _pb0 = float(_psib13(np.array([_r200]), _rb, _mb)[0]) if P13_VESC_BAR else 0.0
            vesc = _vesc13(r, _m200, _r200, _conc, _pb13(r), _pb0)
            # HOW FAR DOES IT GET? In these halos v_esc(10 kpc) is ~1000 km/s and almost nothing a galaxy ejects is formally
            # unbound, so the binary is nearly degenerate and the informative number is the APOCENTRE: solve
            # -Phi(r_apo) = -Phi(r) - v^2/2 for a radial orbit in the same potential (angular momentum would only bring it
            # back sooner, so this is the outer envelope of where the material goes). Censored at 5 R200c, which is also
            # what "never comes back" is written as.
            _vinf = _vesc13(r, _m200, _r200, _conc, _pb13(r), to="inf")
            _rg13 = np.geomspace(0.05, 5.0 * _r200, 512)
            _psi13 = np.minimum.accumulate(0.5 * _vesc13(_rg13, _m200, _r200, _conc, _pb13(_rg13), to="inf") ** 2)
            _tgt13 = 0.5 * (_vinf ** 2 - np.nan_to_num(vtot, nan=np.inf) ** 2)         # -E per unit mass
            rapo = np.where(_tgt13 > _psi13[-1], np.interp(_tgt13, _psi13[::-1], _rg13[::-1]), 5.0 * _r200)
            rapo = np.where(np.isfinite(_tgt13), rapo, np.nan)
        lT = np.log10(np.where(gT > 0, gT, np.nan))
        ism, cgm = gmem, (~gmem) & (r <= P13_RMAX)
        out = gmem & (r >= np.nanmedian(r[gmem])) if gmem.sum() >= 4 else np.zeros(len(gm), bool)   # the outer half of the ISM
        row = dict(snap=int(A["snap"]), gx=int(A["gx"]), z=z, has_vel=bool(np.isfinite(vsys).all()),
                   m_star_mem=float(np.nansum(sm[smem])), n_cgm=int(cgm.sum()), n_ism=int(ism.sum()))
        if smem.sum() >= 5 and np.nansum(sm[smem]) > 0:
            _o = np.argsort(rs[smem])
            row["r50_star"] = float(np.interp(0.5, np.cumsum(sm[smem][_o]) / np.nansum(sm[smem]), rs[smem][_o]))
        else:
            row["r50_star"] = np.nan
        if gmem.sum() >= 5 and np.nansum(gm[gmem]) > 0:
            _o = np.argsort(r[gmem])
            row["r50_gasmem"] = float(np.interp(0.5, np.cumsum(gm[gmem][_o]) / np.nansum(gm[gmem]), r[gmem][_o]))
        else:
            row["r50_gasmem"] = np.nan
        for _nm, _mk in (("ism", ism), ("cgm", cgm)):
            row[f"mgas_{_nm}"] = float(np.nansum(gm[_mk])); row[f"mdust_{_nm}"] = float(np.nansum(gd[_mk]))
            row[f"mmet_{_nm}"] = float(np.nansum((gz * gm)[_mk])); row[f"mh2_{_nm}"] = float(np.nansum(gh2[_mk]))
            row[f"lT_{_nm}_mw"] = float(np.average(lT[_mk & np.isfinite(lT)], weights=gm[_mk & np.isfinite(lT)])) \
                if (_mk & np.isfinite(lT)).any() and np.nansum(gm[_mk & np.isfinite(lT)]) > 0 else np.nan
            _dm = _mk & np.isfinite(lT) & (gd > 0)
            row[f"lT_{_nm}_dw"] = float(np.average(lT[_dm], weights=gd[_dm])) if _dm.any() else np.nan
        _in1 = ism & (r < P13_RIN)                                     # H6: the compact ISM
        row["mgas_in1"], row["mdust_in1"] = float(np.nansum(gm[_in1])), float(np.nansum(gd[_in1]))
        PH13 = {"cold": gT < P13_TCOLD, "warm": (gT >= P13_TCOLD) & (gT < P13_THOT), "hot": gT >= P13_THOT}
        for _ph, _pm in PH13.items():                                  # H2 / H3: the phases of the CGM
            _m = cgm & _pm
            row[f"mgas_{_ph}"] = float(np.nansum(gm[_m])); row[f"mdust_{_ph}"] = float(np.nansum(gd[_m]))
            row[f"n_{_ph}"] = int(_m.sum())
        for _a, _b in P13_SHELLS:                                      # where in the CGM it happens
            _m = cgm & (r >= _a) & (r < _b)
            _t = _m & np.isfinite(lT)
            row[f"mgas_s{int(_b)}"] = float(np.nansum(gm[_m])); row[f"mdust_s{int(_b)}"] = float(np.nansum(gd[_m]))
            row[f"lT_s{int(_b)}"] = float(np.average(lT[_t], weights=gm[_t])) if _t.any() and np.nansum(gm[_t]) > 0 else np.nan
        _sh = np.abs(r - P13_RSH) < P13_DRSH                           # H1 / H4: what crosses the shell, and how cold it is
        for _wn, _wv in (("g", gm), ("d", gd)):
            for _pn, _pm in (("", np.ones(len(gm), bool)), ("c", PH13["cold"]), ("h", PH13["hot"])):
                _w = np.where(_sh & _pm & np.isfinite(vr) & np.isfinite(_wv), _wv, 0.0)
                row[f"mdot_out_{_wn}{_pn}"] = float(np.sum(_w * np.clip(vr, 0, None)) / (2 * P13_DRSH) * KMS_KPC_PER_YR13)
                row[f"mdot_in_{_wn}{_pn}"] = float(-np.sum(_w * np.clip(vr, None, 0)) / (2 * P13_DRSH) * KMS_KPC_PER_YR13)
        for _zn, _zm in (("cgm", cgm), ("out", out)):                  # H1: how fast it is moving
            for _wn, _wv in (("g", gm), ("d", gd)):
                row[f"q50_{_wn}_{_zn}"] = _wq13(vr[_zm], _wv[_zm], 0.50)
                row[f"q90_{_wn}_{_zn}"] = _wq13(vr[_zm], _wv[_zm], 0.90)
                _w = np.where(np.isfinite(vr[_zm]) & np.isfinite(_wv[_zm]), _wv[_zm], 0.0)
                row[f"ffast_{_wn}_{_zn}"] = float(_w[vr[_zm] > P13_VFAST].sum() / _w.sum()) if _w.sum() > 0 else np.nan
        # ── H7: every velocity in units of the LOCAL escape velocity — a fountain, or a wind ──
        row["m200c"], row["r200c"], row["conc"] = _m200, _r200, _conc
        row["central"] = bool(_cen13)
        row["vesc_rsh"] = (float(np.interp(float(P13_RSH), r[np.argsort(r)], vesc[np.argsort(r)]))
                           if np.isfinite(_m200) and np.isfinite(vesc).any() else np.nan)
        row["vesc_r50"] = (float(np.interp(max(row["r50_gasmem"], 0.1), r[np.argsort(r)], vesc[np.argsort(r)]))
                           if np.isfinite(_m200) and np.isfinite(row["r50_gasmem"]) and np.isfinite(vesc).any() else np.nan)
        _xr13 = np.where(np.isfinite(vesc) & (vesc > 0), vr / np.where(vesc > 0, vesc, np.nan), np.nan)   # v_r / v_esc
        _bnd13 = np.isfinite(vtot) & np.isfinite(vesc) & (vtot < P13_XB * vesc)      # bound: it is coming back
        for _zn, _zm in (("cgm", cgm), ("out", out)):
            for _wn, _wv in (("g", gm), ("d", gd)):
                row[f"xesc50_{_wn}_{_zn}"] = _wq13(_xr13[_zm], _wv[_zm], 0.50)
                row[f"xesc90_{_wn}_{_zn}"] = _wq13(_xr13[_zm], _wv[_zm], 0.90)
                _w = np.where(np.isfinite(_xr13[_zm]) & np.isfinite(_wv[_zm]), _wv[_zm], 0.0)
                row[f"fbnd_{_wn}_{_zn}"] = float(_w[_bnd13[_zm]].sum() / _w.sum()) if _w.sum() > 0 else np.nan
                _om13 = _zm & (vr > 0) & np.isfinite(_xr13)                 # of what is MOVING OUT, the unbound share
                _wo = np.where(np.isfinite(_wv), _wv, 0.0)
                row[f"funb_{_wn}_{_zn}"] = (float(_wo[_om13 & (_xr13 >= P13_XB)].sum() / _wo[_om13].sum())
                                            if _wo[_om13].sum() > 0 else np.nan)
                _ma13 = _om13 & np.isfinite(rapo)                           # of what is moving out, HOW FAR it will get
                row[f"rapo50_{_wn}_{_zn}"] = _wq13(rapo[_ma13], _wv[_ma13], 0.50)
                row[f"rapo90_{_wn}_{_zn}"] = _wq13(rapo[_ma13], _wv[_ma13], 0.90)
                row[f"xapo50_{_wn}_{_zn}"] = (row[f"rapo50_{_wn}_{_zn}"] / _r200) if np.isfinite(_r200) else np.nan
        for _wn, _wv in (("g", gm), ("d", gd)):        # the outward flux through the shell, split at the escape velocity
            for _bn, _bm in (("b", np.isfinite(vesc) & (vr < P13_XB * vesc)),
                             ("u", np.isfinite(vesc) & (vr >= P13_XB * vesc))):
                _w = np.where(_sh & _bm & np.isfinite(vr) & np.isfinite(_wv), _wv, 0.0)
                row[f"mdot_out_{_wn}{_bn}"] = float(np.sum(_w * np.clip(vr, 0, None)) / (2 * P13_DRSH) * KMS_KPC_PER_YR13)
        row["mgas_bnd"] = float(np.nansum(gm[cgm & _bnd13])); row["mdust_bnd"] = float(np.nansum(gd[cgm & _bnd13]))
        return row

    HAL13, _hmiss13 = {}, []
    if P13_VESC:
        if os.path.exists(P13_HCACHE):
            _h13 = pd.read_csv(P13_HCACHE)
            HAL13 = {(int(a_), int(b_)): (float(c_), float(d_), bool(e_)) for a_, b_, c_, d_, e_
                     in zip(_h13["snap"], _h13["gx"], _h13["m200c"], _h13["r200c"], _h13["central"])}
        _need13 = [k for k in _uniq13 if k not in HAL13]
        if _need13:
            _nh13, _hmiss13 = _halo13(_need13)
            HAL13.update(_nh13)
            pd.DataFrame([dict(snap=k[0], gx=k[1], m200c=v[0], r200c=v[1], central=v[2])
                          for k, v in sorted(HAL13.items())]).to_csv(P13_HCACHE, index=False)
        _nhal13 = [k for k in _uniq13 if k in HAL13]
        print(f"  the halo each node sits in, for the escape velocity: {len(_nhal13)}/{len(_uniq13)} found in the "
              f"catalogues ({sum(1 for k in _nhal13 if HAL13[k][2])} of them centrals) -> {P13_HCACHE}"
              + (f";  NOT read: {'; '.join(_hmiss13[:3])}" if _hmiss13 else ""))
        if not _nhal13:
            print("     no halo was found — the H7 columns will be NaN and its panels will not be drawn. The catalogues are "
                  "where M200c comes from; the reduced files hold no dark matter and cannot give the potential.")

    Z13 = pd.DataFrame()
    if os.path.exists(P13_CACHE):
        Z13 = Table.read(P13_CACHE).to_pandas()
        print(f"  the census, cached -> {P13_CACHE}  ({len(Z13)} files already measured)")
        if P13_VESC and HAL13 and "xesc90_g_out" not in Z13.columns:
            print("     that cache predates the escape-velocity columns (H7) — it is measured again from scratch. This is "
                  "the one extra pass; delete nothing, the new file replaces it when the pass ends.")
            Z13 = pd.DataFrame()
    _done13 = set(zip(Z13["snap"].astype(int), Z13["gx"].astype(int))) if len(Z13) else set()
    _todo13 = [k for k in _uniq13 if k not in _done13]
    if P13_MAXNODE is not None:
        _todo13 = _todo13[:int(P13_MAXNODE)]
    if _todo13:
        print(f"  measuring {len(_todo13)} files (~{sum(os.path.getsize(_inv13[k]) for k in _todo13) / 2 ** 30:.1f} GiB to "
              "read; the result is cached, and this is a SECOND pass over the same files Part 10 read — the two censuses "
              "keep separate caches so neither has to be rebuilt for the other)", flush=True)
        _new13, _t013, _bad13 = [], _time13.time(), []
        for _i13, _k13 in enumerate(_todo13):
            try:
                _new13.append(_cgm13(_inv13[_k13], HAL13.get(_k13)))
            except Exception as _e13:
                _bad13.append(f"{os.path.basename(_inv13[_k13])}: {type(_e13).__name__}: {_e13}")
            if (_i13 + 1) % 200 == 0:
                _el = _time13.time() - _t013
                print(f"     {_i13 + 1}/{len(_todo13)} files, {_el:.0f} s "
                      f"(~{_el / (_i13 + 1) * (len(_todo13) - _i13 - 1):.0f} s left)", flush=True)
        for _b13 in _bad13[:5]:
            print(f"     file not read, {_b13}")
        if _new13:
            Z13 = pd.concat([Z13, pd.DataFrame(_new13)], ignore_index=True)
            Table.from_pandas(Z13).write(P13_CACHE, overwrite=True)
            print(f"  the census -> {P13_CACHE}  ({len(Z13)} files, {_time13.time() - _t013:.0f} s)")
    if not len(Z13):
        print("  nothing measured — Part 13 stops here.")
    else:
        # ── (D) the nodes on the clock, with the chain's own f_gas and AGN state attached by SNAPSHOT ──
        D13 = NOD13.merge(Z13, on=["snap", "gx"], how="inner")
        D13["grp"] = [G13.get((int(a_), int(g_))) for a_, g_ in zip(D13["anchor_snap"], D13["gal_id"])]
        D13["t0"] = [T13.get((int(a_), int(g_)), np.nan) for a_, g_ in zip(D13["anchor_snap"], D13["gal_id"])]
        D13["dt"] = D13["t"].to_numpy(float) - D13["t0"].to_numpy(float)
        _stn, _fgn, _fdn = [], [], []
        for _a, _g, _s in zip(D13["anchor_snap"], D13["gal_id"], D13["snap"]):
            _d = CH9.get((int(_a), int(_g)))
            _j = np.where(_d["snaps"] == int(_s))[0] if _d is not None else np.array([], int)
            _stn.append(float(_d["st"][_j[0]]) if len(_j) else np.nan)
            _fgn.append(float(_d["lfgas"][_j[0]]) if len(_j) else np.nan)
            _fdn.append(float(_d["lfd"][_j[0]]) if len(_j) else np.nan)
        D13["st"], D13["lfgas_cat"], D13["lfd_cat"] = _stn, _fgn, _fdn
        with np.errstate(all="ignore"):
            _mst = D13["m_star_mem"].to_numpy(float)
            _dc, _di = D13["mdust_cgm"].to_numpy(float), D13["mdust_ism"].to_numpy(float)
            _gc, _gi = D13["mgas_cgm"].to_numpy(float), D13["mgas_ism"].to_numpy(float)
            _lg = lambda a, b: np.log10(np.where((a > 0) & (b > 0), a / np.where(b > 0, b, np.nan), np.nan))
            D13["lfd_cgm"] = _lg(_dc, _mst)                       # H1: the displaced dust per unit stellar mass
            D13["lfg_cgm"] = _lg(_gc, _mst)
            D13["f_dust_cgm"] = np.where(_dc + _di > 0, _dc / (_dc + _di), np.nan)
            D13["f_gas_cgm"] = np.where(_gc + _gi > 0, _gc / (_gc + _gi), np.nan)
            D13["ldg_cgm"] = _lg(_dc, _gc)                        # H3: destruction (falls) vs dilution (flat)
            D13["ldg_ism"] = _lg(_di, _gi)
            D13["dtm_cgm"] = np.where(_dc + D13["mmet_cgm"].to_numpy(float) > 0,
                                      _dc / (_dc + D13["mmet_cgm"].to_numpy(float)), np.nan)
            for _ph in ("cold", "warm", "hot"):                   # H2: the phase structure
                D13[f"fm_{_ph}"] = np.where(_gc > 0, D13[f"mgas_{_ph}"].to_numpy(float) / _gc, np.nan)
                D13[f"fd_{_ph}"] = np.where(_dc > 0, D13[f"mdust_{_ph}"].to_numpy(float) / _dc, np.nan)
            for _c in ("mdot_in_gc", "mdot_in_g", "mdot_out_g", "mdot_out_d", "mdot_in_dc", "mdot_in_d",
                       "mdot_out_gb", "mdot_out_gu", "mdot_out_db", "mdot_out_du"):
                if _c in D13.columns:
                    D13["l" + _c] = np.log10(np.where(D13[_c].to_numpy(float) > 0, D13[_c].to_numpy(float), np.nan))
            if "mdot_out_db" in D13.columns:                   # H7: of the dust leaving RIGHT NOW, the fountain share
                _fb13 = D13["mdot_out_db"].to_numpy(float); _fu13 = D13["mdot_out_du"].to_numpy(float)
                D13["fbnd_flux_d"] = np.where(_fb13 + _fu13 > 0,
                                              _fb13 / np.where(_fb13 + _fu13 > 0, _fb13 + _fu13, np.nan), np.nan)
                _gb13 = D13["mdot_out_gb"].to_numpy(float); _gu13 = D13["mdot_out_gu"].to_numpy(float)
                D13["fbnd_flux_g"] = np.where(_gb13 + _gu13 > 0,
                                              _gb13 / np.where(_gb13 + _gu13 > 0, _gb13 + _gu13, np.nan), np.nan)
                _di13 = D13["mdot_in_d"].to_numpy(float); _do13 = D13["mdot_out_d"].to_numpy(float)
                D13["lfret_d"] = np.log10(np.where((_di13 > 0) & (_do13 > 0),                # the fountain, in the act:
                                                   _di13 / np.where(_do13 > 0, _do13, np.nan), np.nan))
                for _c in ("rapo50_d_out", "rapo50_g_out", "rapo90_d_out"):                  # what returns / what leaves
                    if _c in D13.columns:
                        D13["l" + _c] = np.log10(np.where(D13[_c].to_numpy(float) > 0, D13[_c].to_numpy(float), np.nan))
            D13["f_in1"] = np.where(_gi > 0, D13["mgas_in1"].to_numpy(float) / _gi, np.nan)   # H6: the compact ISM
            D13["lsig_in1"] = np.log10(np.where(D13["mgas_in1"].to_numpy(float) > 0,
                                                D13["mgas_in1"].to_numpy(float) / (np.pi * P13_RIN ** 2 * 1e6), np.nan))
            D13["xray_on"] = np.where(np.isfinite(D13["st"].to_numpy(float)),
                                      (D13["st"].to_numpy(float) >= 2.5).astype(float), np.nan)
            D13["choked"] = np.where(np.isfinite(D13["st"].to_numpy(float)),
                                     ((D13["st"].to_numpy(float) > 1.5) & (D13["st"].to_numpy(float) < 2.5)).astype(float),
                                     np.nan)
        D13 = D13[np.isfinite(D13["dt"].to_numpy(float))].reset_index(drop=True)
        _XB13 = 0.5 * (P13_TB[1:] + P13_TB[:-1])
        D13["tbin"] = np.digitize(D13["dt"].to_numpy(float), P13_TB) - 1
        # the cumulative dust that crossed the shell outward, integrated along each branch. A galaxy only enters it if its
        # branch is sampled from P13_CUMSTART onwards, so that two galaxies' integrals start at the same point of the clock.
        _CUM13 = [(_c, _s) for _c, _s in (("mej_cum", "mdot_out_d"), ("mejb_cum", "mdot_out_db"),
                                          ("meju_cum", "mdot_out_du")) if _s in D13.columns]
        for _cn13, _ in _CUM13:
            D13[_cn13] = np.nan
        _ncum13 = 0
        for _k13, _s13 in D13.groupby(["anchor_snap", "gal_id"]):
            _o = np.argsort(_s13["dt"].to_numpy(float))
            _dt13 = _s13["dt"].to_numpy(float)[_o]
            if _dt13[0] > P13_CUMSTART:
                continue
            _tt = _dt13 * 1e9
            for _cn13, _sc13 in _CUM13:
                _md = _s13[_sc13].to_numpy(float)[_o]
                _cum = np.concatenate([[0.0], np.cumsum(0.5 * (np.nan_to_num(_md[1:]) + np.nan_to_num(_md[:-1]))
                                                        * np.diff(_tt))])
                D13.loc[_s13.index[_o], _cn13] = _cum
            _ncum13 += 1
        with np.errstate(all="ignore"):
            _ms13 = D13["m_star_mem"].to_numpy(float)               # this frame's own column: D13 was filtered above
            for _cn13, _ in _CUM13:
                D13["l" + _cn13] = np.log10(np.where((D13[_cn13].to_numpy(float) > 0) & (_ms13 > 0),
                                                     D13[_cn13].to_numpy(float) / _ms13, np.nan))
            if "mejb_cum" in D13.columns:            # H7: of ALL the dust that has left, the share that left bound
                _cb13 = D13["mejb_cum"].to_numpy(float); _cu13 = D13["meju_cum"].to_numpy(float)
                D13["fbnd_cum_d"] = np.where(_cb13 + _cu13 > 0,
                                             _cb13 / np.where(_cb13 + _cu13 > 0, _cb13 + _cu13, np.nan), np.nan)
        print(f"  the cumulative ejection panel: {_ncum13} of "
              f"{D13.groupby(['anchor_snap', 'gal_id']).ngroups} galaxies are sampled from {P13_CUMSTART:g} Gyr onwards "
              "and can carry a comparable integral")
        print(f"  the clock: {len(D13)} galaxy-nodes in [{P13_TB[0]:g}, {P13_TB[-1]:g}] Gyr of {_CNAM13}, "
              + ", ".join(f"{GST13[g]['short']} {int((D13['grp'] == g).sum())} nodes / "
                          f"{D13[D13['grp'] == g].groupby(['anchor_snap', 'gal_id']).ngroups} galaxies" for g in GK13))
        ROW13 = []

        def _track13(col, grp, agg="median"):
            cen = np.full(len(_XB13), np.nan); lo = cen.copy(); hi = cen.copy(); n = np.zeros(len(_XB13), int)
            sub = D13[(D13["grp"] == grp) & np.isfinite(D13[col].to_numpy(float))]
            rng = np.random.default_rng(13)
            for _b in range(len(_XB13)):
                _s = sub[sub["tbin"] == _b]
                if not len(_s):
                    continue
                _v = _s.groupby(["anchor_snap", "gal_id"])[col].median().to_numpy(float)
                _v = _v[np.isfinite(_v)]; n[_b] = len(_v)
                if len(_v) < P13_NMIN:
                    continue
                _f = np.median if agg == "median" else np.mean
                cen[_b] = _f(_v)
                _bs = _f(_v[rng.integers(0, len(_v), (P13_NBOOT, len(_v)))], axis=1)
                lo[_b], hi[_b] = np.percentile(_bs, 16), np.percentile(_bs, 84)
            return cen, lo, hi, n

        def _panel13(ax, col, ylab, title, block="", pct=False, leg=False, agg="median"):
            for _g in GK13:
                cen, lo, hi, n = _track13(col, _g, agg)
                _ok = np.isfinite(cen)
                if _ok.any():
                    ax.fill_between(_XB13[_ok], lo[_ok], hi[_ok], color=GST13[_g]["color"], alpha=0.20, lw=0)
                    ax.plot(_XB13[_ok], cen[_ok], "o-", color=GST13[_g]["color"], lw=GST13[_g]["lw"], ms=4, mec="k",
                            mew=0.4, label=f"{GST13[_g]['short']}  ({int(np.nanmax(n))} gal max)")
                for _x, _c, _l, _h, _n in zip(_XB13, cen, lo, hi, n):
                    ROW13.append(dict(block=block, quantity=col, grp=_g, x=float(_x), n_gal=int(_n),
                                      med=float(_c), lo=float(_l), hi=float(_h)))
            ax.axvline(0, color="k", lw=1.4, ls="--", alpha=0.9)
            ax.set_xlim(P13_TB[0], P13_TB[-1]); ax.set_ylabel(ylab); ax.set_title(title, loc="left", fontsize=10)
            ax.grid(alpha=0.22, lw=0.5)
            if pct:
                ax.set_ylim(-0.02, 1.02)
            if leg and ax.get_legend_handles_labels()[0]:
                ax.legend(loc="best", frameon=False, fontsize=8)

        def _mark13(ax):
            ax.annotate(_CNAM13, xy=(0, 0.0), xycoords=("data", "axes fraction"), xytext=(4, 6),
                        textcoords="offset points", fontsize=8, color="0.3")

        # ── FIG 1 — H1: what left the galaxy, and how fast, BEFORE the onset ──
        fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
        _panel13(axs[0, 0], "lfd_cgm", r"$\log\, M_{\rm dust}^{\rm CGM}/M_\star$",
                 "(a) the dust that is outside the galaxy", block="ejection", leg=True)
        _panel13(axs[0, 1], "f_dust_cgm", r"$M_{\rm dust}^{\rm CGM}/M_{\rm dust}^{\rm total}$",
                 "(b) the SHARE of the dust that is outside", block="ejection", pct=True)
        _panel13(axs[1, 0], "q90_g_out", r"$v_r$ (90th pct, mass-weighted)  [km s$^{-1}$]",
                 "(c) how fast the outer ISM is moving out", block="ejection")
        axs[1, 0].axhline(0, color="0.4", lw=0.9, ls=":")
        _panel13(axs[1, 1], "lmej_cum", r"$\log\, \int \dot M_{\rm dust}^{\rm out}\,dt \,/\, M_\star$",
                 rf"(d) the dust that crossed $r = {P13_RSH:g}$ kpc outward, cumulative", block="ejection")
        for ax in axs[1]:
            ax.set_xlabel(_CLAB13)
        _mark13(axs[0, 0])
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — H1: the ejection into the CGM, on the clock of {_CNAM13}",
                     y=0.99, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_cgm_ejection"); plt.show()

        # ── FIG 1b — H7: a fountain or a wind? the same velocities, in units of the LOCAL escape velocity ──
        if P13_VESC and "xesc90_g_out" in D13.columns and np.isfinite(D13["xesc90_g_out"].to_numpy(float)).any():
            fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
            _panel13(axs[0, 0], "xesc90_g_out", r"$v_r\,/\,v_{\rm esc}(r)$   (90th pct, mass-weighted)",
                     "(a) how fast the outer ISM leaves, per unit escape velocity", block="escape", leg=True)
            axs[0, 0].axhline(1.0, color="0.2", lw=1.3, ls=":"); axs[0, 0].axhline(0.0, color="0.6", lw=0.9, ls=":")
            axs[0, 0].annotate("above this it does not come back", xy=(0.985, 1.0), xycoords=("axes fraction", "data"),
                               xytext=(0, 3), textcoords="offset points", fontsize=7, color="0.25", ha="right", va="bottom")
            _panel13(axs[0, 1], "lrapo50_d_out", r"$\log\, r_{\rm apo}$  [kpc]  (mass-weighted median)",
                     "(b) how far out the dust leaving the ISM will get before it turns around", block="escape")
            axs[0, 1].axhline(np.log10(P13_RSH), color="0.45", lw=0.9, ls=":")
            axs[0, 1].annotate(rf"the shell, $r = {P13_RSH:g}$ kpc", xy=(0.985, np.log10(P13_RSH)),
                               xycoords=("axes fraction", "data"), xytext=(0, 3), textcoords="offset points",
                               fontsize=7, color="0.35", ha="right", va="bottom")
            _panel13(axs[1, 0], "lfret_d",
                     r"$\log\, \dot M_{\rm dust}^{\rm in} / \dot M_{\rm dust}^{\rm out}$",
                     rf"(c) at $r = {P13_RSH:g}$ kpc, the dust coming back against the dust leaving",
                     block="escape")
            axs[1, 0].axhline(0.0, color="0.2", lw=1.2, ls=":")
            axs[1, 0].annotate("above the line more dust returns than leaves", xy=(0.985, 0.0),
                               xycoords=("axes fraction", "data"), xytext=(0, 3), textcoords="offset points",
                               fontsize=7, color="0.25", ha="right", va="bottom")
            for _g in GK13:                       # (d) the two integrals side by side: what left bound, what left for good
                for _c13, _ls13, _nm13b in (("lmejb_cum", "-", "bound — a fountain"),
                                            ("lmeju_cum", "--", "unbound — a wind")):
                    if _c13 not in D13.columns:
                        continue
                    cen, lo, hi, n = _track13(_c13, _g)
                    _ok = np.isfinite(cen)
                    if _ok.any():
                        axs[1, 1].plot(_XB13[_ok], cen[_ok], ls=_ls13, lw=2.2, color=GST13[_g]["color"],
                                       label=f"{GST13[_g]['short']} — {_nm13b}")
                    for _x, _c, _l, _h, _n in zip(_XB13, cen, lo, hi, n):
                        ROW13.append(dict(block="escape", quantity=_c13, grp=_g, x=float(_x), n_gal=int(_n),
                                          med=float(_c), lo=float(_l), hi=float(_h)))
            axs[1, 1].axvline(0, color="k", lw=1.4, ls="--", alpha=0.9); axs[1, 1].grid(alpha=0.22, lw=0.5)
            axs[1, 1].set_xlim(P13_TB[0], P13_TB[-1])
            axs[1, 1].set_ylabel(r"$\log\, \int \dot M_{\rm dust}^{\rm out}\,dt \,/\, M_\star$")
            axs[1, 1].set_title(rf"(d) the dust that crossed $r = {P13_RSH:g}$ kpc, split at $v_{{\rm esc}}$",
                                loc="left", fontsize=10)
            if axs[1, 1].get_legend_handles_labels()[0]:
                axs[1, 1].legend(loc="best", frameon=False, fontsize=7.5)
            for ax in axs[1]:
                ax.set_xlabel(_CLAB13)
            _mark13(axs[0, 0])
            fig.suptitle(f"{BOX_LABEL[P9_BOX]} — H7: was what left the galaxy bound to it?", y=0.99, fontsize=12,
                         fontweight="bold")
            fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_cgm_escape"); plt.show()
            _vs13 = D13[np.abs(D13["dt"].to_numpy(float)) <= 0.5]
            print("  the scale the panel above is drawn against — v_esc at the shell in the half Gyr around "
                  f"{_CNAM13}: "
                  + ", ".join(f"{GST13[_g]['short']} "
                              f"{np.nanmedian(_vs13[_vs13['grp'] == _g]['vesc_rsh'].to_numpy(float)):.0f} km/s"
                              for _g in GK13)
                  + f"  (M200c = " + ", ".join(f"{np.log10(np.nanmedian(_vs13[_vs13['grp'] == _g]['m200c'].to_numpy(float))):.2f}"
                                               for _g in GK13) + " in the log)")

        # ── FIG 2 — H2 / H3: the heating, and what it does to the grains ──
        fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
        _panel13(axs[0, 0], "lT_cgm_mw", r"$\langle \log T \rangle_{M}$  [K]",
                 "(a) the temperature of the CGM", block="heating", leg=True)
        _panel13(axs[0, 1], "fm_cold", rf"$M_{{\rm gas}}(T < 10^{{{np.log10(P13_TCOLD):.0f}}}\,$K$)/M_{{\rm gas}}^{{\rm CGM}}$",
                 "(b) the cold phase of the CGM", block="heating", pct=True)
        _panel13(axs[1, 0], "lT_cgm_dw", r"$\langle \log T \rangle_{M_{\rm dust}}$  [K]",
                 "(c) the temperature the surviving grains sit at", block="heating")
        _panel13(axs[1, 1], "ldg_cgm", r"$\log\, M_{\rm dust}/M_{\rm gas}$  (CGM)",
                 "(d) the dust-to-gas ratio out there — destruction, or dilution", block="heating")
        for _g in GK13:                                             # the ISM's own D/G behind it, as the reference
            cen, lo, hi, n = _track13("ldg_ism", _g)
            _ok = np.isfinite(cen)
            if _ok.any():
                axs[1, 1].plot(_XB13[_ok], cen[_ok], ls="--", lw=1.4, color=GST13[_g]["color"], alpha=0.8)
            for _x, _c, _l, _h, _n in zip(_XB13, cen, lo, hi, n):
                ROW13.append(dict(block="heating", quantity="ldg_ism", grp=_g, x=float(_x), n_gal=int(_n),
                                  med=float(_c), lo=float(_l), hi=float(_h)))
        axs[1, 1].legend(handles=[Line2D([], [], color="0.3", lw=2.0, label="the CGM"),
                                  Line2D([], [], color="0.3", lw=1.4, ls="--", label="the ISM it came from")],
                         loc="best", frameon=False, fontsize=7.5)
        for ax in axs[1]:
            ax.set_xlabel(_CLAB13)
        _mark13(axs[0, 0])
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — H2 and H3: the heating of the CGM, and the grains in it",
                     y=0.99, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_cgm_heating"); plt.show()

        # ── FIG 3 — H4 / H5 / H6: the inflow, the choke, and what is left ──
        fig, axs = plt.subplots(2, 2, figsize=(11.6, 7.4), sharex=True)
        _panel13(axs[0, 0], "lmdot_in_gc", r"$\log\, \dot M_{\rm in}(T < 10^5\,$K$)$  [$M_\odot$ yr$^{-1}$]",
                 rf"(a) the COLD inflow through $r = {P13_RSH:g}$ kpc", block="return", leg=True)
        _panel13(axs[0, 1], "lfgas_cat", r"$\log\, M_{\rm gas}/M_\star$",
                 "(b) SIMBA's own X-ray gate", block="return")
        axs[0, 1].axhline(np.log10(XRAY_FGAS_MAX), color="0.25", lw=1.2, ls=":")
        axs[0, 1].set_ylim(top=max(axs[0, 1].get_ylim()[1], np.log10(XRAY_FGAS_MAX) + 0.25))   # keep the gate in frame
        axs[0, 1].annotate(rf"the gate: $f_{{\rm gas}} < {XRAY_FGAS_MAX:g}$ — below it the channel can fire."
                           + "\nCrossing it at 0 is by construction; climbing back over it is not.",
                           xy=(0.985, np.log10(XRAY_FGAS_MAX)), xycoords=("axes fraction", "data"), xytext=(0, -4),
                           textcoords="offset points", fontsize=7, color="0.25", ha="right", va="top")
        for _g in GK13:                                             # (c) the state itself: firing, or choked by the gas
            for _c13, _ls, _nm in (("xray_on", "-", "X-ray on"), ("choked", "--", "choked (jets, gas-rich)")):
                _sub = D13[(D13["grp"] == _g) & np.isfinite(D13[_c13].to_numpy(float))]
                _p, _n = np.full(len(_XB13), np.nan), np.zeros(len(_XB13), int)
                for _b in range(len(_XB13)):
                    _s = _sub[_sub["tbin"] == _b]
                    if not len(_s):
                        continue
                    _v = _s.groupby(["anchor_snap", "gal_id"])[_c13].max().to_numpy(float)
                    _n[_b] = len(_v)
                    if len(_v) >= P13_NMIN:
                        _p[_b] = float(np.mean(_v >= 0.5))
                    ROW13.append(dict(block="return", quantity=_c13, grp=_g, x=float(_XB13[_b]), n_gal=int(_n[_b]),
                                      med=float(_p[_b]), lo=np.nan, hi=np.nan))
                _ok = np.isfinite(_p)
                if _ok.any():
                    axs[1, 0].plot(_XB13[_ok], _p[_ok], ls=_ls, color=GST13[_g]["color"], lw=2.2,
                                   label=f"{GST13[_g]['short']} — {_nm}")
        axs[1, 0].set_ylim(-0.02, 1.02); axs[1, 0].axvline(0, color="k", lw=1.4, ls="--", alpha=0.9)
        axs[1, 0].set_xlim(P13_TB[0], P13_TB[-1]); axs[1, 0].grid(alpha=0.22, lw=0.5)
        axs[1, 0].set_ylabel("fraction of the group in the state")
        axs[1, 0].set_title("(c) firing, or choked by its own gas", loc="left", fontsize=10)
        axs[1, 0].legend(loc="best", frameon=False, fontsize=7.5)
        _panel13(axs[1, 1], "f_in1", rf"$M_{{\rm gas}}(r<{P13_RIN:g}\,$kpc$)/M_{{\rm gas}}^{{\rm ISM}}$",
                 "(d) how much of the ISM is in the central kpc", block="return", pct=True)
        for ax in axs[1]:
            ax.set_xlabel(_CLAB13)
        _mark13(axs[0, 0])
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — H4, H5 and H6: the cold return, the choke, and the ISM that is left",
                     y=0.99, fontsize=12, fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_m25_cgm_return"); plt.show()

        # ── (E) THE TEST: the per-galaxy scalars, before and after, against the outcome ──
        def _win13(col, w, how="median"):
            """One number per galaxy: `col` aggregated over the clock window w."""
            _s = D13[(D13["dt"] >= w[0]) & (D13["dt"] <= w[1]) & np.isfinite(D13[col].to_numpy(float))]
            if not len(_s):
                return {}
            _g = _s.groupby(["anchor_snap", "gal_id"])[col]
            _v = (_g.max() if how == "max" else _g.median())
            return {k: float(v) for k, v in _v.items()}

        P13_CLAIMS = [
            ("H1  the dust displaced into the CGM", "lfd_cgm", "pre", "pure"),
            ("H1  the SHARE of the dust displaced", "f_dust_cgm", "pre", "pure"),
            ("H1  how fast the outer ISM moves out", "q90_g_out", "pre", "pure"),
            ("H1  the dust that crossed the shell", "lmej_cum", "pre", "pure"),
            ("H2  the CGM temperature", "lT_cgm_mw", "post", "pure"),
            ("H2  the cold phase of the CGM", "fm_cold", "post", "dusty"),
            ("H3  the CGM dust-to-gas ratio", "ldg_cgm", "post", "dusty"),
            ("H3  the temperature the grains sit at", "lT_cgm_dw", "post", "pure"),
            ("H4  the cold inflow at the shell", "lmdot_in_gc", "post", "dusty"),
            ("H5  f_gas — SIMBA's X-ray gate", "lfgas_cat", "post", "dusty"),
            ("H6  the ISM inside 1 kpc", "f_in1", "post", "dusty"),
            ("H6  the size of the ISM", "r50_gasmem", "post", "pure"),
            ("H7  the ejection speed per v_esc", "xesc90_g_out", "pre", "pure"),
            ("H7  the unbound share of what leaves", "funb_d_cgm", "pre", "pure"),
            ("H7  the CGM that is still bound", "fbnd_g_cgm", "pre", "dusty"),
            ("H7  how far the ejecta get [kpc]", "rapo50_d_out", "pre", "pure"),
            ("H7  ... in units of R200c", "xapo50_d_out", "pre", "pure"),
            ("H7  the dust returning / leaving", "lfret_d", "post", "dusty"),
            ("H7  the dust that left BOUND", "lmejb_cum", "pre", "dusty"),
            ("H7  the dust that left UNBOUND", "lmeju_cum", "pre", "pure"),
            ("H7  of the dust that left, the bound share", "fbnd_cum_d", "pre", "dusty"),
        ]
        # The claims are written in the vocabulary of the OUTCOME: "dusty" means whichever group of the split IN FORCE ends
        # up dust-rich — GK13[0] in all three of them (the dusty QGs; Part 2's rotating class; Part 12's hole galaxies, which
        # end ~1.8 dex dustier). Without this mapping the column compared a group name to the string "dusty" and every row
        # read NO the moment the split was not "pop".
        _EXPK13 = {"dusty": GK13[0], "pure": GK13[1]}
        _lm13 = {}
        for _k13 in G13:
            _r = _S13[(_S13["snap"] == _k13[0]) & (_S13["gal_id"] == _k13[1])]
            _lm13[_k13] = float(_r["log_mstar"].iloc[0]) if len(_r) else np.nan
        _out13 = {k: float(_S13[(_S13["snap"] == k[0]) & (_S13["gal_id"] == k[1])]["lfd"].iloc[0])
                  for k in G13 if len(_S13[(_S13["snap"] == k[0]) & (_S13["gal_id"] == k[1])])}

        def _auc13(vals, mcell=True):
            """AUC of `vals` (a {key: value} dict) for GK13[0] against GK13[1], raw and pooled within log M* terciles."""
            _k = [k for k in vals if np.isfinite(vals[k])]
            if len(_k) < 6:
                return np.nan, np.nan, 0, 0
            x = np.array([vals[k] for k in _k], float); g = np.array([G13[k] == GK13[0] for k in _k], bool)
            a_raw = mw_auc(x, g)
            if not mcell:
                return a_raw[0], a_raw[1], int(g.sum()), int((~g).sum())
            m = np.array([_lm13.get(k, np.nan) for k in _k], float)
            _ok = np.isfinite(m)
            if _ok.sum() < 6:
                return a_raw[0], a_raw[1], int(g.sum()), int((~g).sum())
            _q = np.quantile(m[_ok], np.linspace(0, 1, P13_MTERC + 1)[1:-1])
            _c = np.digitize(m, _q)
            _num, _den = 0.0, 0.0
            for _ci in np.unique(_c[_ok]):
                _s = _ok & (_c == _ci)
                if (g & _s).sum() >= 2 and ((~g) & _s).sum() >= 2:
                    _w = (g & _s).sum() * ((~g) & _s).sum()
                    _num += _w * float(np.mean([(xa > xb) + 0.5 * (xa == xb) for xa in x[g & _s] for xb in x[(~g) & _s]]))
                    _den += _w
            return a_raw[0], (_num / _den if _den else np.nan), int(g.sum()), int((~g).sum())

        print(f"\nTHE SCORECARD — every claim scored in the window it is about; the split is {_SPLAB13}.")
        print(f"  before = {P13_PRE[0]:+g} to {P13_PRE[1]:+g} Gyr, after = {P13_POST[0]:+g} to {P13_POST[1]:+g} Gyr of "
              f"{_CNAM13}.  AUC = P({GST13[GK13[0]]['short']} > {GST13[GK13[1]]['short']}); [.] is the same AUC pooled "
              f"within {P13_MTERC} terciles of log M*.")
        print(f"  {'claim':44s} {'window':7s} {GST13[GK13[0]]['short'][:11]:>11s} {GST13[GK13[1]]['short'][:11]:>11s} "
              f"{'AUC':>6s} {'[M*]':>6s} {'p':>7s}  hypothesis")
        for _nm13, _col13, _w13, _exp13 in P13_CLAIMS:
            if _col13 not in D13.columns:
                continue
            _wd = P13_PRE if _w13 == "pre" else P13_POST
            _v13 = _win13(_col13, _wd)
            if not _v13:
                print(f"  {_nm13:44s} {_w13:7s}  — no galaxies in the window")
                continue
            _a, _am, _n1, _n0 = _auc13(_v13)
            _m1 = np.median([v for k, v in _v13.items() if G13[k] == GK13[0]] or [np.nan])
            _m0 = np.median([v for k, v in _v13.items() if G13[k] == GK13[1]] or [np.nan])
            _p = mw_auc(np.array(list(_v13.values()), float),
                        np.array([G13[k] == GK13[0] for k in _v13], bool))[1]
            _got = (GK13[0] if (np.isfinite(_am) and _am > 0.5) or (not np.isfinite(_am) and _a > 0.5) else GK13[1])
            _exk13 = _EXPK13[_exp13]
            _ok13 = ("yes" if _got == _exk13 else "NO ") if np.isfinite(_a) else "—  "
            print(f"  {_nm13:44s} {_w13:7s} {_m1:11.3g} {_m0:11.3g} {_a:6.2f} "
                  + (f"{_am:6.2f}" if np.isfinite(_am) else "     —") + f" {_p:7.3f}  {_ok13} "
                  + f"({GST13[_exk13]['short']} higher)")
            ROW13.append(dict(block="scorecard", quantity=_col13, grp=_w13, x=np.nan, n_gal=_n1 + _n0,
                              med=float(_m1), lo=float(_m0), hi=float(_am if np.isfinite(_am) else _a)))

        # ── FIG 4 — the chain itself: what happened before, against how the galaxy ended ──
        _TESTS13 = [("lfd_cgm", "pre", r"$\log M_{\rm dust}^{\rm CGM}/M_\star$ before the onset",
                     "(a) H1 — was the dust already out?"),
                    ("q90_g_out", "pre", r"$v_r$(90th pct) of the outer ISM before  [km s$^{-1}$]",
                     "(b) H1 — was it moving out faster?"),
                    ("lT_cgm_mw", "post", r"$\langle \log T\rangle_M$ of the CGM after the onset",
                     "(c) H2 — did its CGM end up hotter?"),
                    ("fbnd_cum_d", "pre", r"bound share of the dust that crossed $r = 10$ kpc outward",
                     "(d) H7 — did what left stay bound?")]
        _TESTS13 = [t for t in _TESTS13 if t[0] in D13.columns]
        fig, axs = plt.subplots(2, 2, figsize=(10.4, 7.8))
        for _ax, (_col13, _w13, _xlab, _ttl) in zip(axs.ravel(), _TESTS13):
            _v13 = _win13(_col13, P13_PRE if _w13 == "pre" else P13_POST)
            _kk = [k for k in _v13 if np.isfinite(_v13[k]) and np.isfinite(_out13.get(k, np.nan))]
            if len(_kk) < 4:
                _ax.set_title(_ttl + "  — too few", loc="left", fontsize=10); continue
            _x = np.array([_v13[k] for k in _kk], float); _y = np.array([_out13[k] for k in _kk], float)
            for _g in GK13:
                _m = np.array([G13[k] == _g for k in _kk], bool)
                _ax.plot(_x[_m], _y[_m], "o", color=GST13[_g]["color"], ms=6, mec="k", mew=0.4,
                         label=f"{GST13[_g]['short']}  ({int(_m.sum())})")
            _rho, _pv = spearmanr(_x, _y)
            _ax.annotate(rf"Spearman $\rho = {_rho:+.2f}$   $p = {_pv:.3f}$   $N = {len(_kk)}$",
                         xy=(0.03, 0.04), xycoords="axes fraction", fontsize=8.5, color="0.25")
            _ax.axhline(DUSTY_CUT, color="0.35", lw=1.0, ls=":")
            _ax.set_xlabel(_xlab); _ax.set_ylabel(r"$\log M_{\rm dust}/M_\star$ at the anchor")
            _ax.set_title(_ttl, loc="left", fontsize=10); _ax.grid(alpha=0.2, lw=0.5)
            for _xx, _yy, _kk2 in zip(_x, _y, _kk):
                ROW13.append(dict(block="test", quantity=_col13, grp=G13[_kk2], x=float(_xx), n_gal=1,
                                  med=float(_yy), lo=np.nan, hi=np.nan))
        for _ax in axs.ravel()[len(_TESTS13):]:
            _ax.axis("off")
        if axs.ravel()[0].get_legend_handles_labels()[0]:
            axs.ravel()[0].legend(loc="best", frameon=False, fontsize=8)
        fig.suptitle(f"{BOX_LABEL[P9_BOX]} — the test: what the galaxy did around {_CNAM13}, against the dust it ended with",
                     y=1.02, fontsize=12, fontweight="bold")
        fig.tight_layout(); paper_save(fig, "kappa_m25_cgm_test"); plt.show()

        pd.DataFrame(ROW13).to_csv(os.path.join(OUT, "kappa_m25_cgm.csv"), index=False)
        D13.to_csv(os.path.join(OUT, "kappa_m25_cgm_nodes.csv"), index=False)
        print(f"\ntables -> {os.path.join(OUT, 'kappa_m25_cgm.csv')} (the drawn points and the scorecard), "
              f"{os.path.join(OUT, 'kappa_m25_cgm_nodes.csv')} (every node)")
```

## Part 14 — the jet leg as the upstream cause: does a long pre-truncation jet mode make a compact, dusty, H₂-poor rotator?

**The hypothesis, in full.** A galaxy that spends a **long time in the jet mode before the truncation** keeps its rotating
gas disc and *suppresses* the X-ray channel by compactifying its ISM; the dust survives, and the high H₂ surface density
lets it grow back. A galaxy that instead truncates **at the opening of the X-ray gate**, with no long pre-SFT jet leg, has
its ISM stirred by whatever does the truncating: the dust is destroyed, the remnant is pressure-supported, it goes on
accreting gas — but the gate stays open and the dust cannot re-form.

It is a **chain**, and this part tests it link by link, always in the causal direction — split on the *cause*, look at the
*outcome*. That is the opposite of every other part of this notebook, which splits on the outcome class:

$$\text{the pre-SFT jet leg} \;\longrightarrow\; \text{a compact, rotating ISM} \;\longrightarrow\;
\text{the X-ray channel suppressed} \;\longrightarrow\; \text{the dust survives and grows.}$$

### Why the jet leg is the cause worth trying

Part 12d already found it in the m25 chains, without looking for it: the **time in the jet state before the SFT** is
$1.26$ Gyr for the rotating class against $0.44$ Gyr for the rest, and it is the single strongest predictor of the survival
delay $t_{\rm SFT} - t_{x1}$ ($\rho = +0.65$ pooled, $+0.61$ inside the rotating class). The other end of the same statement
is the **X-ray duty cycle after the gate**: $0.87$ for the rotators against $0.97$ — that part's top-ranked contrast — with
longer firing-free gaps. The channel is *open* in both classes; what differs is how continuously it fires, and how long the
galaxy lived in the jet mode before any of it began.

### What is measured

**The cause**, per galaxy, on the branch's own sampling and *before the truncation only*, with Part 0's `agn_states()`
verbatim:

| | |
|---|---|
| `t_jet_pre` | time in the **jet** state ($M_{\rm BH} > 10^{7.5}$, $f_{\rm Edd} < 0.2$) — Part 12d's `t_jetstate`, the default axis |
| `t_chk_pre` | time in the **choked** state — a jet with the X-ray gate shut. The jet leg *without* the channel, which is what the hypothesis is really about |
| `f_jet_pre`, `f_chk_pre` | the same as a **fraction** of the sampled pre-SFT branch — the version an old galaxy cannot win simply by having lived longer |

All four are scored and the cell prints which one carries the signal; `P14_CAUSE` switches the axis.

**The middle link (the suppression):** the X-ray duty cycle after the gate opens and the longest firing-free gap; the share
of the post-truncation branch with $f_{\rm gas}$ under SIMBA's gate; $f_{\rm gas}$ at the truncation and at the anchor; and
the post-SFT gas budget (anchor minus truncation, and the re-accretion from the post-SFT minimum) — the *"keeps accreting
gas"* half of the hypothesis.

**The outcome at the anchor**, from the catalogue rows (Part 4b's `_anc4b`, same logic): $\kappa_{\rm rot}$(gas), the
compactness $-\log R_{50}({\rm gas})/R_{50}(\star)$, $\log \Sigma_{\rm H_2}$, $\log \Sigma_{\rm gas}$, $\log R_{50}$(gas),
and the dust and H₂ fractions. The compound outcome the question names — **a compact rotator with high $f_{\rm dust}$ and
low $f_{\rm H_2}$** — is four booleans, each drawn on its own so a broken link can be read off the ladder instead of
disappearing inside the conjunction. `P14_CQTY` chooses what *compact* means (the extent ratio by default, or
$\Sigma_{\rm H_2}$, $\Sigma_{\rm gas}$, $R_{50}$(gas)); the cuts are the box's own medians unless set.

### The figures

* **FIG 1 `kappa_jetleg_outcome`** — *off by default* (`P14_FIG1`): the user found it uninformative, and its
  numbers are the printed blocks (a) and (b) above; Part 15 replaces its axis with the state at the jet activation.
  When switched on it draws (a) the **ladder**: $P(\text{outcome} \mid \text{the cause})$ in equal-count bins with
  Wilson 68 % bars, the four ingredients and the conjunction; (b) the plane the question names, compactness against the
  dust, every galaxy coloured by its jet leg; (c) dust against the H₂ it grew from; (d) the **middle link**, the X-ray duty
  cycle after the gate against the jet leg.
* **FIG 2 `kappa_jetleg_clocks`** — the **order of events**: the long and short jet-leg groups on the $t - t_{\rm SFT}$
  clock for the compactness, $\Sigma_{\rm H_2}$, $\kappa_{\rm rot}$, $f_{\rm gas}$ (with SIMBA's gate drawn),
  $f_{\rm dust}$ and $f_{\rm H_2}$, with each group's median jet start (▼) and first X-ray firing (|) marked on the axis.
  **This is where the hypothesis can die**: the compactification has to happen *during the jet leg*, i.e. left of zero. If
  the two groups separate only after the truncation, the jet leg is a label on the outcome and not its cause.

### The confound, and the control

The jet leg needs a big black hole, so it is partly *"how long ago $M_{\rm BH}$ crossed $10^{7.5}$"*, and both ends of the
chain grow with $M_\star$ and with the epoch. The group split is therefore **mass-controlled** by default — the cause is
ranked inside quartiles of $\log M_\star$ (`P14_MCTRL`) — and every score is printed twice, raw and **partial at fixed
$\log M_\star$ and $\log M_{\rm BH}$**. `log_mstar`, `log_mbh` and `t_SFT` are carried in the link table as control rows: if
they move with the groups, the split is not clean.

**The decisive control is `cis50nox`** — jets **on**, X-ray heating compiled **out**. If the chain is real, the
jet leg → compact rotating disc link must *survive* there (the jets are still there), while the jet leg → dust link must
*collapse* (nothing is destroying the dust, so there is nothing to be protected from). That double expectation is printed
as its own block, and the dust row is the one to read.

*A pure read* of `ism_prediction_histories_<box>.hdf5` restricted to Part 3's rows, plus Part 3's tracks and Part 0's `QM`
for the anchor geometry → `kappa_jetleg_outcome{.png,.pdf}`, `kappa_jetleg_clocks{.png,.pdf}`, `kappa_jetleg.csv` and
`kappa_jetleg_galaxies.csv` (one row per branch, every column of the chain).

```python
# ── Part 14 — THE JET LEG AS THE UPSTREAM CAUSE: does a long pre-SFT jet mode make a compact, dusty, H2-poor rotator? ──
# THE HYPOTHESIS (the user's, in full, and this cell exists to try to break it):
#   a galaxy that spends a LONG time in the jet mode BEFORE the truncation keeps its rotating gas disc and SUPPRESSES the
#   X-ray channel by compactifying its ISM; the dust survives, and the high H2 surface density lets it grow back. A galaxy
#   that instead truncates AT the opening of the X-ray gate, with no long pre-SFT jet leg, has its ISM stirred by whatever
#   does the truncating: the dust is destroyed, the remnant is pressure-supported, it goes on accreting gas — but the gate
#   stays open and the dust cannot re-form.
# It is a CHAIN, and the cell tests it link by link, always in the causal direction (split on the CAUSE, look at the
# OUTCOME) — the opposite of every other part of this notebook, which splits on the outcome class:
#   the pre-SFT jet leg  ->  a compact, rotating ISM  ->  the X-ray channel suppressed  ->  the dust survives and grows.
# WHY THE JET LEG IS THE CAUSE TO TRY. Part 12d already found it: the time in the jet state before the SFT is 1.26 Gyr for
# the rotating class against 0.44 Gyr for the rest, and it is the single strongest predictor of the survival delay
# t_SFT - t_x1 (rho = +0.65 pooled, +0.61 within the rotating class). The X-ray duty cycle after the gate is the other end
# of the same statement: 0.87 for the rotators against 0.97 (the top-ranked contrast of that part), with longer firing-free
# gaps. The channel is OPEN in both classes; what differs is how continuously it fires and how long the galaxy went on
# living in the jet mode before any of it started.
# THE CAUSE, per galaxy, on the branch's own sampling and before the truncation only (Part 0's agn_states() VERBATIM):
#   t_jet_pre  time in the JET state (M_BH > 10^7.5, f_Edd < 0.2)  — Part 12d's t_jetstate, THE axis by default
#   t_chk_pre  time in the CHOKED state (a jet with the X-ray gate shut: f_gas >= XRAY_FGAS_MAX or the jet under full
#              velocity) — the jet leg WITHOUT the channel, which is what the hypothesis is really about
#   f_jet_pre / f_chk_pre   the same as a FRACTION of the sampled pre-SFT branch — the version an old galaxy cannot win by
#              simply having lived longer; all four are scored, and the cell prints which one carries the signal.
# THE MIDDLE LINK (the suppression): the X-ray duty cycle after the gate and the longest firing-free gap; the share of the
#   post-truncation branch with f_gas under SIMBA's gate; f_gas at the truncation and at the anchor; and the post-SFT gas
#   budget (the anchor minus the truncation, and the re-accretion from the post-SFT minimum) — the "keeps accreting gas"
#   half of the hypothesis.
# THE OUTCOME at the anchor, from the CATALOGUE rows (Part 4b's `_anc4b`, same logic): kappa_rot(gas), the compactness
#   -log R50(gas)/R50(*), log Sigma_H2, log Sigma_gas, log R50(gas), and the dust and H2 fractions. The compound outcome the
#   question names — a COMPACT ROTATOR with HIGH f_dust and LOW f_H2 — is built from four booleans, each printed on its own
#   so that a failure can be read off the ladder instead of hidden in the conjunction.
# THE CONFOUND: the jet leg needs a big black hole, so it is partly "how long ago M_BH crossed 10^7.5", and both ends of the
#   chain grow with M* and with the epoch. The group split is therefore MASS-CONTROLLED by default (the cause is ranked
#   inside quartiles of log M*, P14_MCTRL) and every score is printed twice, raw and partial at fixed log M*, log M_BH and
#   t_SFT.
# THE DECISIVE CONTROL is cis50nox: jets ON, X-ray heating compiled OUT. If the chain is real, the jet leg -> compact disc
#   link must SURVIVE there (the jets are still there) while the jet leg -> dust link must COLLAPSE (nothing is destroying
#   the dust, so there is nothing to be protected from). That double expectation is printed as its own block.
# WHAT IT READS: ONE pass over ism_prediction_histories_<box>.hdf5 restricted to Part 3's rows (same sample, classes and
# clocks), plus Part 3's tracks for the clock figure and Part 0's QM for the anchor geometry. No catalogues, no particles.
P14_BOXES  = None           # None = the boxes Part 3 built, the channel-off box last
P14_CAUSE  = "t_jet_pre"    # THE axis: "t_jet_pre" (Part 12d's) | "f_jet_pre" | "t_chk_pre" | "f_chk_pre"
P14_CQTY   = "comp"         # what "compact" means: "comp" = -log R50(gas)/R50(*) | "lsig_h2" | "lsig_gas" | "lr_gas" (inverted)
P14_CCUT   = None           # the compactness cut (None = the box's own median of the sample)
P14_H2CUT  = None           # log 1.36 M_H2/M* below this = "H2-poor" (None = the box's own median)
P14_NB     = 5              # equal-count bins of the cause in the ladder panel
P14_NMIN   = 8              # galaxies a drawn bin needs (drop it to ~3 with more bins to sample the coarse boxes finer)
P14_TERC   = 3              # the group split for the clock figure: the top and bottom 1/P14_TERC of the cause
P14_MCTRL  = True           # rank the cause INSIDE quartiles of log M* before splitting (the mass-controlled groups)
P14_MQ     = 4              # how many log M* cells the control uses
P14_NBOOT  = 500
P14_FIG1   = False          # FIG 1 (the ladder / the planes / duty_post) — OFF: the user found it uninformative, the
                            # numbers behind it are the printed blocks (a) and (b), and FIG 2 is the figure this part keeps
P14_TRK    = ["comp", "lsig_h2", "kappa_gas", "lfgas", "lfd", "lfh2"]      # the clock figure's rows, in order
P14_PLANE2 = ("lfh2", "lfd")   # the second plane of FIG 1: (x, y) at the anchor
P14_GCOL   = {"long": "#e08214", "short": "#8073ac"}
P14_GLAB   = {"long": "LONG jet leg", "short": "SHORT jet leg"}
P14_CLAB   = {"t_jet_pre": "time in the jet state before the SFT [Gyr]",
              "t_chk_pre": "time in the choked jet state before the SFT [Gyr]",
              "f_jet_pre": "share of the pre-SFT branch in the jet state",
              "f_chk_pre": "share of the pre-SFT branch in the choked jet state"}
P14_OLAB   = {"rot": "a rotator", "compact": "compact", "dusty": rf"dusty", "h2poor": r"H$_2$-poor", "all4": "ALL FOUR"}
P14_QLAB   = {"comp": r"$-\log R_{50}({\rm gas})/R_{50}(\star)$", "lsig_h2": r"$\log \Sigma_{\rm H_2}$",
              "lsig_gas": r"$\log \Sigma_{\rm gas}$", "lr_gas": r"$\log R_{50}({\rm gas})$",
              "kappa_gas": r"$\kappa_{\rm rot}$(gas)", "lfd": r"$\log M_{\rm dust}/M_\star$",
              "lfh2": r"$\log 1.36\,M_{\rm H_2}/M_\star$", "lfgas": r"$\log M_{\rm gas}/M_\star$",
              "lms": r"$\log M_\star$", "lr_star": r"$\log R_{50}(\star)$", "lsig_star": r"$\log \Sigma_\star$"}

D14 = pd.DataFrame()
_BX14 = [b for b in (list(H) if P14_BOXES is None else P14_BOXES) if b in H]
_BX14 = [b for b in _BX14 if b != NOX_BOX] + [b for b in _BX14 if b == NOX_BOX]
if not _BX14:
    print("Part 14: Part 3 built no branch records — run it first.")
else:
    from scipy import stats as _st14

    def _pass14(box, want):
        """ONE pass over one box's histories, restricted to Part 3's rows: the pre-SFT AGN-state budget (the CAUSE), the
        post-gate firing pattern and the post-SFT gas budget (the MIDDLE LINK), on the branch's own sampling."""
        out = []
        with h5py.File(HIST_FILE.format(box=box), "r") as f:
            for gname in f:
                a = int(gname.replace("snap", ""))
                if not any(k[0] == a for k in want):
                    continue
                gh = f[gname]
                t = gh["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
                if len(t) < 3:
                    continue
                ms, mg, mbh, fedd = (gh[c][:][o] for c in ("mstar", "mgas", "mbh", "fedd"))
                S14, D14b, X14 = agn_states(ms, mg, mbh, fedd)              # Part 0's gates, verbatim
                M = {k: np.nan_to_num(v) > 0.5 for k, v in
                     (("xray", S14["xray"]), ("choked", S14["choked"]), ("rad", S14["radiative"]),
                      ("dor", S14["dormant"]), ("jet", D14b["jet"]), ("gate", D14b["gate"]))}
                dtw = np.gradient(t)
                for j, gid in enumerate(np.asarray(gh["gal_ids"][:]).astype(np.int64)):
                    key = (a, int(gid))
                    if key not in want:
                        continue
                    r, ex = want[key], X14["exist"][:, j]
                    if ex.sum() < 3:
                        continue
                    t_sft = float(r["t_sft"])
                    if not np.isfinite(t_sft):
                        continue
                    m = {k: v[:, j] & ex for k, v in M.items()}
                    pre, post = ex & (t <= t_sft), ex & (t > t_sft)
                    w = lambda k: float(np.sum(dtw[k]))                     # time [Gyr] carried by a mask
                    t_pre = w(pre)
                    t_x1 = float(t[m["xray"]][0]) if m["xray"].any() else np.nan
                    # the firing pattern AFTER the channel first opens: how continuously it actually fires
                    aft = ex & (t >= t_x1) if np.isfinite(t_x1) else np.zeros(len(t), bool)
                    _xa, _ta = m["xray"][aft], t[aft]
                    if _xa.any():
                        _edges = np.concatenate([[_ta[0]], _ta[_xa], [_ta[-1]]])
                        gap_max = float(np.max(np.diff(_edges))) if len(_edges) > 1 else 0.0
                    else:
                        gap_max = float(_ta[-1] - _ta[0]) if len(_ta) > 1 else np.nan
                    with np.errstate(all="ignore"):
                        lfg = np.log10(np.where(ex & (mg[:, j] > 0) & (ms[:, j] > 0), mg[:, j] / ms[:, j], np.nan))
                    _yp = lfg[post]; _yp = _yp[np.isfinite(_yp)]
                    _isf = lfg[pre]; _isf = _isf[np.isfinite(_isf)]
                    out.append(dict(
                        box=box, snap=a, gal_id=int(gid),
                        t_pre=t_pre, t_jet_pre=w(pre & m["jet"]), t_chk_pre=w(pre & m["choked"]),
                        t_xr_pre=w(pre & m["xray"]), t_rad_pre=w(pre & m["rad"]), t_dor_pre=w(pre & m["dor"]),
                        f_jet_pre=w(pre & m["jet"]) / t_pre if t_pre > 0 else np.nan,
                        f_chk_pre=w(pre & m["choked"]) / t_pre if t_pre > 0 else np.nan,
                        f_gate_pre=w(pre & m["gate"]) / t_pre if t_pre > 0 else np.nan,
                        duty_post=(w(aft & m["xray"]) / w(aft)) if w(aft) > 0 else np.nan,
                        gap_max=gap_max, n_epi_post=float(np.sum(_xa[1:] & ~_xa[:-1]) + (1 if len(_xa) and _xa[0] else 0)),
                        f_gate_post=(w(post & m["gate"]) / w(post)) if w(post) > 0 else np.nan,
                        fgas_sft=float(10 ** _isf[-1]) if len(_isf) else np.nan,
                        fgas_anchor=float(10 ** _yp[-1]) if len(_yp) else np.nan,
                        dlfgas_post=float(_yp[-1] - _isf[-1]) if (len(_yp) and len(_isf)) else np.nan,
                        regas=float(_yp[-1] - np.min(_yp)) if len(_yp) else np.nan))
        return pd.DataFrame(out)

    def _anc14(b):
        """The ANCHOR values of the geometry, straight from the catalogue rows of Part 0's sample and aligned to Part 3's
        frame — Part 4b's `_anc4b`, same logic, so the two parts cannot drift apart."""
        F = H[b]["F"]
        g = QM[QM["box"] == b].drop_duplicates(subset=["snap", "gal_id"]).copy()
        g = g.set_index([g["snap"].astype(int), g["gal_id"].astype(int)])
        key = list(zip(F["snap"].astype(int), F["gal_id"].astype(int)))
        take = lambda c: (g[c].reindex(key).to_numpy(float) if c in g.columns else np.full(len(key), np.nan))
        ms, mg, mh = take("mstar"), take("mgas"), take("mh2")
        rs, rg = take("r_half_star"), take("r_half_gas")
        with np.errstate(all="ignore"):
            return {"kappa_gas": take(KAPPA_COL),
                    "lsig_star": np.log10(np.where((rs > 0) & (ms > 0), 0.5 * ms / (np.pi * rs ** 2), np.nan)),
                    "lsig_gas": np.log10(np.where((rg > 0) & (mg > 0), 0.5 * mg / (np.pi * rg ** 2), np.nan)),
                    "lsig_h2": np.log10(np.where((rg > 0) & (mh > 0), 0.5 * HE * mh / (np.pi * rg ** 2), np.nan)),
                    "comp": -np.log10(np.where((rg > 0) & (rs > 0), rg / rs, np.nan)),
                    "lr_gas": np.log10(np.where(rg > 0, rg, np.nan)), "lr_star": np.log10(np.where(rs > 0, rs, np.nan))}

    _D14 = []
    for _b14 in _BX14:
        F14 = H[_b14]["F"]
        _w14 = {(int(_r["snap"]), int(_r["gal_id"])): _r for _, _r in F14.iterrows()}
        _d14 = _pass14(_b14, _w14)
        if not len(_d14):
            continue
        _keep = ["snap", "gal_id", "z", "age", "log_mstar", "log_mbh", "kcls", "rot", "pop", "lfd", "lfh2", "lfgas",
                 "lssfr", "kappa", "kz", "ngas", "central", "t_sft", "t_qt", "t_anchor", "t_x1", "t_gate", "t_big",
                 "t_jet", "e_x", "f_x", "f_x_post", "t_post", "lfd_net", "lfd_reb"]
        _base = F14[[c for c in _keep if c in F14.columns]].copy()
        for _k, _v in _anc14(_b14).items():                     # the anchor geometry, by position in Part 3's frame
            _base[_k] = _v
        _d14 = _base.merge(_d14, on=["snap", "gal_id"], how="inner")
        _d14["box"] = _b14
        _D14.append(_d14)
        print(f"  [{_b14}] {len(_d14)} branches: the pre-SFT state budget, the post-gate firing pattern and the anchor geometry"
              f"  ({int(np.isfinite(_d14['comp']).sum())} carry R50(gas))")
    D14 = pd.concat(_D14, ignore_index=True) if _D14 else pd.DataFrame()

if not len(D14):
    print("Part 14: no branch carried both an SFT and a histories chain — nothing to draw.")
else:
    _CQ, _CA = P14_CQTY, P14_CAUSE
    _CSIGN = -1.0 if _CQ == "lr_gas" else 1.0                   # a SMALL R50(gas) is compact; every other axis is +
    # ── the outcome booleans, cut box by box so a resolution cannot decide the answer by itself ──
    _cuts14 = {}
    for b in _BX14:
        g = D14[D14.box == b]
        _cc = P14_CCUT if P14_CCUT is not None else float(np.nanmedian(_CSIGN * g[_CQ].to_numpy(float)))
        _hc = P14_H2CUT if P14_H2CUT is not None else float(np.nanmedian(g["lfh2"].to_numpy(float)))
        _cuts14[b] = (_cc, _hc)
    D14["compact"] = [(_CSIGN * v >= _cuts14[b][0]) if np.isfinite(v) else False
                      for b, v in zip(D14.box, D14[_CQ].to_numpy(float))]
    D14["dusty"] = D14["lfd"].to_numpy(float) >= DUSTY_CUT
    D14["h2poor"] = [(v <= _cuts14[b][1]) if np.isfinite(v) else False
                     for b, v in zip(D14.box, D14["lfh2"].to_numpy(float))]
    D14["rot"] = D14["rot"].astype(bool)
    D14["all4"] = D14.rot & D14.compact & D14.dusty & D14.h2poor
    _OUT14 = ["rot", "compact", "dusty", "h2poor", "all4"]

    # ── the mass-controlled rank of the cause, and the LONG / SHORT groups ──
    def _mrank14(g, col):
        """The cause's rank in [0, 1] INSIDE cells of log M* (P14_MQ quantile cells), so a group split on it cannot be a
        mass split in disguise. With P14_MCTRL off it is the plain rank."""
        v, lm = g[col].to_numpy(float), g["log_mstar"].to_numpy(float)
        r = np.full(len(v), np.nan)
        if not P14_MCTRL or not np.isfinite(lm).any():
            ok = np.isfinite(v)
            if ok.sum() > 1:
                r[ok] = (_st14.rankdata(v[ok]) - 0.5) / ok.sum()
            return r
        e = np.unique(np.nanquantile(lm, np.linspace(0, 1, P14_MQ + 1)))
        cell = np.clip(np.digitize(lm, e[1:-1]), 0, max(len(e) - 2, 0))
        for c in np.unique(cell[np.isfinite(lm)]):
            ok = (cell == c) & np.isfinite(v) & np.isfinite(lm)
            if ok.sum() > 1:
                r[ok] = (_st14.rankdata(v[ok]) - 0.5) / ok.sum()
        return r

    D14["crank"] = np.concatenate([_mrank14(D14[D14.box == b], _CA) for b in _BX14]) if len(_BX14) else np.nan
    D14["grp"] = np.where(D14.crank >= 1 - 1.0 / P14_TERC, "long",
                          np.where(D14.crank <= 1.0 / P14_TERC, "short", "mid"))

    def _rho14(x, y):
        x, y = np.asarray(x, float), np.asarray(y, float)
        ok = np.isfinite(x) & np.isfinite(y)
        if ok.sum() < 6 or np.nanstd(x[ok]) == 0 or np.nanstd(y[ok]) == 0:
            return np.nan, np.nan, int(ok.sum())
        r = spearmanr(x[ok], y[ok])
        return float(r.statistic), float(r.pvalue), int(ok.sum())

    def _prho14(x, y, ctrl):
        """PARTIAL Spearman: both axes and the controls rank-transformed, the controls regressed out by least squares, the
        residuals correlated; p from the t approximation with k controls."""
        cols = [np.asarray(x, float), np.asarray(y, float)] + [np.asarray(c, float) for c in ctrl]
        ok = np.all([np.isfinite(c) for c in cols], axis=0)
        if ok.sum() < len(ctrl) + 8:
            return np.nan, np.nan, int(ok.sum())
        R = np.array([_st14.rankdata(c[ok]) for c in cols], float).T
        sd = R.std(0)
        if not (sd[:2] > 0).all():
            return np.nan, np.nan, int(ok.sum())
        R = R[:, np.r_[True, True, sd[2:] > 0]]         # a CONSTANT control carries no information and would only make
        sd = sd[np.r_[True, True, sd[2:] > 0]]          # the least squares singular — it is dropped, not fatal
        k = R.shape[1] - 2
        R = (R - R.mean(0)) / sd
        A = np.column_stack([np.ones(len(R)), R[:, 2:]])
        res = [R[:, i] - A @ np.linalg.lstsq(A, R[:, i], rcond=None)[0] for i in (0, 1)]
        if np.std(res[0]) == 0 or np.std(res[1]) == 0:
            return np.nan, np.nan, int(ok.sum())
        r, n = float(np.corrcoef(res[0], res[1])[0, 1]), int(ok.sum())
        dof = n - 2 - k
        return r, (float(2 * _st14.t.sf(abs(r) * np.sqrt(dof / max(1 - r * r, 1e-12)), dof)) if dof > 0 else np.nan), n

    def _eq14(x, nb=None):
        v = np.asarray(x, float); v = v[np.isfinite(v)]
        if len(v) < 2:
            return np.array([0.0, 1.0])
        e = np.unique(np.quantile(v, np.linspace(0, 1, (P14_NB if nb is None else nb) + 1)))
        return e if len(e) > 2 else np.array([float(v.min()), float(v.max()) + 1e-9])

    def _share14(x, m, edges, nmin=None):
        """The share of a BOOLEAN outcome in equal-count bins of the cause, with its Wilson 68 % interval."""
        nmin = P14_NMIN if nmin is None else nmin
        x, m = np.asarray(x, float), np.asarray(m, bool)
        ok = np.isfinite(x); x, m = x[ok], m[ok]
        out = []
        for i, (lo, hi) in enumerate(zip(edges[:-1], edges[1:])):
            s = (x >= lo) & ((x <= hi) if i == len(edges) - 2 else (x < hi))
            if int(s.sum()) < nmin:
                continue
            p, plo, phi = wilson(int(m[s].sum()), int(s.sum()))
            out.append((float(np.median(x[s])), p, plo, phi, int(s.sum())))
        return out

    ROW14 = []
    print("=" * 132)
    print(f"Part 14 — THE JET LEG AS THE CAUSE.  the axis: {P14_CLAB[_CA]}   |   'compact' = {P14_QLAB[_CQ]}"
          f"{' (inverted)' if _CSIGN < 0 else ''} above the box median")
    print(f"  the chain under test:  a long pre-SFT jet leg  ->  a compact rotating ISM  ->  the X-ray channel suppressed  "
          f"->  the dust survives and grows")
    print("=" * 132)

    # ══ (A) the cause itself, and the outcome base rates ═════════════════════════════════════════════════════════════
    for b in _BX14:
        g = D14[D14.box == b]
        _nox14 = b == NOX_BOX
        print(f"\n  {BOX_LABEL[b]}{'   [jets ON, X-ray heating OFF: the control]' if _nox14 else ''}   {len(g)} branches")
        print("    the pre-SFT time budget [Gyr, medians]:  " + "   ".join(
            f"{lab} {np.nanmedian(g[q].to_numpy(float)):.2f}" for q, lab in
            (("t_pre", "sampled"), ("t_dor_pre", "dormant"), ("t_rad_pre", "radiative"), ("t_chk_pre", "choked jet"),
             ("t_xr_pre", "X-ray"), ("t_jet_pre", "JET (choked + X-ray)"))))
        print("    the outcome at the anchor: " + "   ".join(
            f"{P14_OLAB[o]} {100 * g[o].mean():.0f} %" for o in _OUT14) +
            f"    [cuts: {_CQ} {_cuts14[b][0]:+.2f}, log f_H2 {_cuts14[b][1]:+.2f}]")
        for q in ("t_pre", "t_jet_pre", "t_chk_pre", "f_jet_pre", "f_chk_pre"):
            v = g[q].to_numpy(float); v = v[np.isfinite(v)]
            if len(v):
                ROW14.append(dict(box=b, block="census", quantity=q, med=float(np.median(v)),
                                  lo=float(np.percentile(v, 16)), hi=float(np.percentile(v, 84)), n_gal=len(v)))
        for o in _OUT14:
            ROW14.append(dict(box=b, block="baserate", quantity=o, med=float(g[o].mean()), n_gal=len(g)))

    # ══ (B) THE QUESTION, answered directly: is a long jet leg more likely to end as the compound outcome? ═══════════
    print("\n" + "-" * 132)
    print("(a) THE QUESTION — P(outcome | the cause), the top third of the cause against the bottom third"
          + ("  [the thirds are taken INSIDE quartiles of log M*]" if P14_MCTRL else ""))
    print(f"  {'box':10s} {'outcome':12s} {'P(long)':>16s} {'P(short)':>16s} {'ratio':>7s} {'AUC(cause)':>11s} "
          f"{'rho | M*, M_BH':>15s}")
    for b in _BX14:
        g = D14[D14.box == b]
        L, S = g[g.grp == "long"], g[g.grp == "short"]
        for o in _OUT14:
            if not len(L) or not len(S):
                continue
            pl, pll, plh = wilson(int(L[o].sum()), len(L))
            ps, psl, psh = wilson(int(S[o].sum()), len(S))
            au, p, na, nb_ = mw_auc(g[_CA].to_numpy(float), g[o].to_numpy(bool))
            ap, pp, _n = _prho14(g[_CA].to_numpy(float), g[o].to_numpy(float),
                                 [g.log_mstar.to_numpy(float), g.log_mbh.to_numpy(float)])
            print(f"  {b:10s} {P14_OLAB[o]:12s} {100 * pl:6.1f} [{100 * pll:4.1f},{100 * plh:5.1f}] "
                  f"{100 * ps:6.1f} [{100 * psl:4.1f},{100 * psh:5.1f}] {(pl / ps if ps > 0 else np.inf):7.2f} "
                  f"{au:6.2f} {('*' if p < 0.05 else ' ')}   {ap:+8.2f}{'*' if (np.isfinite(pp) and pp < 0.05) else ' '}")
            ROW14.append(dict(box=b, block="ladder", quantity=o, med=pl, lo=ps, hi=(pl / ps if ps > 0 else np.nan),
                              auc=au, p=p, n_gal=len(L) + len(S)))
    print("  (AUC(cause) = P(the cause is larger in the galaxies WITH the outcome); the last column is the partial rank")
    print("   correlation of the cause with the outcome at fixed log M* and log M_BH — the jet leg needs a big hole, and")
    print("   both ends of the chain grow with the mass, so that is the number the hypothesis has to survive.)")

    # ══ (C) the links, one at a time, as contrasts between the two groups ════════════════════════════════════════════
    _LINK = [("comp", "the ISM extent"), ("lsig_h2", "the ISM density"), ("lsig_gas", "the ISM density"),
             ("kappa_gas", "the rotation"), ("lfd", "the dust"), ("lfh2", "the H2"), ("lfgas", "the gas"),
             ("duty_post", "THE SUPPRESSION"), ("gap_max", "THE SUPPRESSION"), ("f_gate_post", "THE GATE"),
             ("fgas_anchor", "THE GATE"), ("dlfgas_post", "the accretion"), ("regas", "the accretion"),
             ("lfd_net", "the dust budget"), ("lfd_reb", "the dust budget"), ("log_mstar", "CONTROL"),
             ("log_mbh", "CONTROL"), ("t_sft", "CONTROL")]
    print("\n" + "-" * 132)
    print("(b) THE LINKS — the same galaxies, the long jet leg against the short one, quantity by quantity")
    print(f"  {'box':10s} {'link':17s} {'quantity':14s} {'LONG':>9s} {'SHORT':>9s} {'D':>8s} {'AUC':>6s} {'p':>9s}")
    for b in _BX14:
        g = D14[D14.box == b]
        L, S = g[g.grp == "long"], g[g.grp == "short"]
        for q, link in _LINK:
            if q not in g.columns:
                continue
            a_, b_ = L[q].to_numpy(float), S[q].to_numpy(float)
            if np.isfinite(a_).sum() < 4 or np.isfinite(b_).sum() < 4:
                continue
            au, p, _na, _nb = mw_auc(np.concatenate([a_, b_]),
                                     np.concatenate([np.ones(len(a_), bool), np.zeros(len(b_), bool)]))
            print(f"  {b:10s} {link:17s} {q:14s} {np.nanmedian(a_):9.2f} {np.nanmedian(b_):9.2f} "
                  f"{np.nanmedian(a_) - np.nanmedian(b_):+8.2f} {au:6.2f} {p:9.2e}")
            ROW14.append(dict(box=b, block="link", quantity=q, med=float(np.nanmedian(a_)), lo=float(np.nanmedian(b_)),
                              auc=au, p=p, n_gal=len(a_) + len(b_)))
    print("  (the hypothesis wants: comp / Sigma_H2 / kappa / f_dust UP, f_H2 DOWN, duty_post DOWN, gap_max UP in the LONG")
    print("   group, and the CONTROL rows flat — if log M* or log M_BH move with them the split is not clean.)")

    # ══ (D) the counterfactual: jets on, channel off ════════════════════════════════════════════════════════════════
    if NOX_BOX in _BX14:
        print("\n" + "-" * 132)
        print("(c) THE CONTROL — cis50nox has the jets and NOT the channel. If the chain is real, the jet leg must still")
        print("    build the compact rotating disc there, and must NOT buy any dust (there is nothing to be protected from).")
        print(f"  {'quantity':14s} " + " ".join(f"{b:>14s}" for b in _BX14))
        for q in (_CQ, "kappa_gas", "lsig_h2", "lfd", "lfh2", "duty_post"):
            _r = []
            for b in _BX14:
                g = D14[D14.box == b]
                r0, p0, n0 = _rho14(g[_CA].to_numpy(float), g[q].to_numpy(float))
                _r.append(f"{r0:+.2f}{'*' if (np.isfinite(p0) and p0 < 0.05) else ' '} ({n0:4d})" if np.isfinite(r0)
                          else f"{'-':>14s}")
                ROW14.append(dict(box=b, block="nox_rho", quantity=q, med=r0, p=p0, n_gal=n0))
            print(f"  {q:14s} " + " ".join(f"{v:>14s}" for v in _r))
        print("  (rho of the cause with each quantity at the anchor. The dust row is the decisive one: it should be strong in")
        print("   the fiducial boxes and NEAR ZERO in the channel-off box.)")

    def _leg14(ax, **kw):
        kw.setdefault("frameon", False); kw.setdefault("fontsize", 7.5)
        h, l = ax.get_legend_handles_labels()
        if h:
            ax.legend(h, l, **kw)

    # ══ (E) FIG 1 — the answer, box by box ══════════════════════════════════════════════════════════════════════════
    if P14_FIG1:
        fig, AX = plt.subplots(len(_BX14), 4, figsize=(18.6, 3.9 * len(_BX14)), squeeze=False)

        _OCOL = {"rot": KCLS["rot"]["color"], "compact": "#4d9221", "dusty": "#762a83", "h2poor": "#01665e",
                 "all4": "k"}
        _cv14 = D14[_CA].to_numpy(float)
        _vlo, _vhi = (np.nanpercentile(_cv14, 5), np.nanpercentile(_cv14, 95)) if np.isfinite(_cv14).any() else (0, 1)
        for i, b in enumerate(_BX14):
            g = D14[D14.box == b]
            _nox14 = b == NOX_BOX
            x = g[_CA].to_numpy(float)
            ax = AX[i, 0]                                            # (a) THE LADDER
            _e = _eq14(x)
            for o in _OUT14:
                pts = _share14(x, g[o].to_numpy(bool), _e)
                if not pts:
                    continue
                ax.errorbar([p[0] for p in pts], [p[1] for p in pts],
                            yerr=np.clip([[p[1] - p[2] for p in pts], [p[3] - p[1] for p in pts]], 0, None),
                            fmt="-o", ms=5 if o != "all4" else 6.5, lw=2.0 if o != "all4" else 3.0, color=_OCOL[o],
                            capsize=2.5, zorder=3 if o != "all4" else 5, label=P14_OLAB[o])
                for _p in pts:
                    ROW14.append(dict(box=b, block=f"share_{o}", quantity=_CA, x=_p[0], med=_p[1], lo=_p[2], hi=_p[3],
                                      n_gal=_p[4]))
            ax.set_ylim(-0.02, 1.02)
            ax.set_xlabel(P14_CLAB[_CA], fontsize=9.5)
            ax.set_ylabel("share of the bin", fontsize=9.5)
            ax.set_title(f"(a) {BOX_LABEL[b]}", loc="left", fontsize=9.5, color="#7f3b08" if _nox14 else "k")
            _leg14(ax, loc="best", ncol=2); ax.grid(alpha=0.2, lw=0.5)
            ax = AX[i, 1]                                            # (b) the plane the question names
            _s = ax.scatter(g[_CQ].to_numpy(float), g["lfd"].to_numpy(float), c=np.clip(x, _vlo, _vhi), s=22,
                            cmap="inferno_r", vmin=_vlo, vmax=_vhi, edgecolor="k", linewidth=0.2)
            _cb = fig.colorbar(_s, ax=ax, pad=0.02); _cb.set_label(P14_CLAB[_CA], fontsize=8); _cb.ax.tick_params(labelsize=7.5)
            ax.axhline(DUSTY_CUT, color="0.35", lw=1.2, ls="--")
            ax.axvline(_CSIGN * _cuts14[b][0], color="0.35", lw=1.2, ls="--")
            ax.set_xlabel(P14_QLAB[_CQ], fontsize=9.5); ax.set_ylabel(P14_QLAB["lfd"], fontsize=9.5)
            ax.set_title("(b) the corner the hypothesis predicts", loc="left", fontsize=9.5)
            ax.grid(alpha=0.2, lw=0.5)
            ax = AX[i, 2]                                            # (c) the second plane: high dust, low H2
            _qx, _qy = P14_PLANE2
            _s = ax.scatter(g[_qx].to_numpy(float), g[_qy].to_numpy(float), c=np.clip(x, _vlo, _vhi), s=22,
                            cmap="inferno_r", vmin=_vlo, vmax=_vhi, edgecolor="k", linewidth=0.2)
            _cb = fig.colorbar(_s, ax=ax, pad=0.02); _cb.set_label(P14_CLAB[_CA], fontsize=8); _cb.ax.tick_params(labelsize=7.5)
            if _qy == "lfd":
                ax.axhline(DUSTY_CUT, color="0.35", lw=1.2, ls="--")
            if _qx == "lfh2":
                ax.axvline(_cuts14[b][1], color="0.35", lw=1.2, ls="--")
            ax.set_xlabel(P14_QLAB.get(_qx, _qx), fontsize=9.5); ax.set_ylabel(P14_QLAB.get(_qy, _qy), fontsize=9.5)
            ax.set_title("(c) dust against the H$_2$ it grew from", loc="left", fontsize=9.5)
            ax.grid(alpha=0.2, lw=0.5)
            ax = AX[i, 3]                                            # (d) THE MIDDLE LINK: the suppression
            _xhi = float(np.nanpercentile(x, 99)) if np.isfinite(x).any() else 1.0
            _xd = np.clip(x, None, _xhi)                             # the long tail of the cause is drawn AT the edge
            for c in KC:
                m = (g.kcls == c).to_numpy(bool)
                if m.sum():
                    ax.scatter(_xd[m], g["duty_post"].to_numpy(float)[m], s=20, alpha=0.6, color=KCLS[c]["color"],
                               edgecolor="none", label=f"{KCLS[c]['short']}  ({int(m.sum())})")
            _e2 = _eq14(x, nb=max(P14_NB, 4))
            _pts = []
            for _i, (lo, hi) in enumerate(zip(_e2[:-1], _e2[1:])):
                s = np.isfinite(x) & (x >= lo) & ((x <= hi) if _i == len(_e2) - 2 else (x < hi))
                v = g["duty_post"].to_numpy(float)[s]; v = v[np.isfinite(v)]
                if len(v) >= P14_NMIN:
                    _pts.append((float(np.median(x[s])), float(np.median(v))))
            if _pts:
                ax.plot([p[0] for p in _pts], [p[1] for p in _pts], "-o", color="k", lw=2.6, ms=5.5, zorder=5,
                        label="the median")
            r0, p0, n0 = _rho14(x, g["duty_post"].to_numpy(float))
            if np.isfinite(r0):
                ax.annotate(rf"$\rho = {r0:+.2f}$  (p {p0:.3f}, N {n0})", xy=(0.97, 0.96), xycoords="axes fraction",
                            ha="right", va="top", fontsize=8.5, color="0.2")
            ax.set_ylim(-0.03, 1.03)
            if np.isfinite(_xhi):
                ax.set_xlim(np.nanmin(_xd) - 0.03 * (_xhi - np.nanmin(_xd) + 1e-9), _xhi + 0.05 * (_xhi - np.nanmin(_xd) + 1e-9))
            ax.set_xlabel(P14_CLAB[_CA], fontsize=9.5)
            ax.set_ylabel("X-ray duty cycle after the gate opens", fontsize=9.5)
            ax.set_title("(d) the middle link: is the channel suppressed?", loc="left", fontsize=9.5)
            _leg14(ax, loc="lower left"); ax.grid(alpha=0.2, lw=0.5)
        fig.suptitle("Does a long pre-truncation jet leg make a compact, dusty, H$_2$-poor rotator?", y=0.998, fontsize=12.5,
                     fontweight="bold")
        fig.tight_layout(rect=(0, 0, 1, 0.975)); paper_save(fig, "kappa_jetleg_outcome"); plt.show()

    # ══ (F) FIG 2 — the ORDER OF EVENTS: the two groups on the truncation clock ══════════════════════════════════════
    _TRK14 = [q for q in P14_TRK if any(q in H[b]["T"] for b in _BX14)]
    fig2, AX2 = plt.subplots(len(_TRK14), len(_BX14), figsize=(4.5 * len(_BX14), 2.5 * len(_TRK14)), squeeze=False)
    for jb, b in enumerate(_BX14):
        F, T = H[b]["F"], H[b]["T"]
        g = D14[D14.box == b]
        _gmap = dict(zip(zip(g.snap.astype(int), g.gal_id.astype(int)), g.grp))
        _grp = np.array([_gmap.get(k, "") for k in zip(F["snap"].astype(int), F["gal_id"].astype(int))])
        _gm = {k: (_grp == k) for k in ("long", "short")}
        for iq, q in enumerate(_TRK14):
            ax = AX2[iq, jb]
            if q not in T:
                ax.set_axis_off(); continue
            for k in ("short", "long"):
                A = T[q]["sft"][_gm[k]]
                if not len(A) or not np.isfinite(A).any():
                    continue
                cen, lo, hi, n_t = med_band(A, nboot=P14_NBOOT)
                ax.fill_between(TG, lo, hi, color=P14_GCOL[k], alpha=0.20, lw=0)
                ax.plot(TG, cen, color=P14_GCOL[k], lw=2.5, label=f"{P14_GLAB[k]}  ({int(_gm[k].sum())})")
                for _x, _c, _l, _h, _n in zip(TG, cen, lo, hi, n_t):
                    ROW14.append(dict(box=b, block=f"clock_{q}", quantity=k, x=float(_x), med=float(_c), lo=float(_l),
                                      hi=float(_h), n_gal=int(_n)))
                # where that group's jet leg and first firing sit on this clock
                _sub = g[g.grp == k]
                for _c14, _m14, _ls in ((_sub.t_jet.to_numpy(float) - _sub.t_sft.to_numpy(float), "v", "-"),
                                        (_sub.t_x1.to_numpy(float) - _sub.t_sft.to_numpy(float), "|", ":")):
                    _c14 = _c14[np.isfinite(_c14)]
                    if len(_c14) >= 3:
                        ax.plot([np.median(_c14)], [0.02 + (0.06 if k == "long" else 0.0)], _m14, ms=7,
                                color=P14_GCOL[k], transform=ax.get_xaxis_transform(), clip_on=False, zorder=6)
            ax.axvline(0.0, color="k", lw=1.2, ls="--")
            ax.grid(alpha=0.2, lw=0.5)
            if iq == 0:
                ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
                if jb == 0:
                    _leg14(ax, loc="best")
            if jb == 0:
                ax.set_ylabel(P14_QLAB.get(q, q), fontsize=9)
            if iq == len(_TRK14) - 1:
                ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]", fontsize=9.5)
            if q == "lfgas":
                ax.axhline(np.log10(XRAY_FGAS_MAX), color="#2ca02c", lw=1.1, ls=":")
                if jb == 0:
                    ax.annotate(rf"the X-ray gate $f_{{\rm gas}} < {XRAY_FGAS_MAX:g}$", xy=(0.02, np.log10(XRAY_FGAS_MAX)),
                                xycoords=("axes fraction", "data"), xytext=(0, 3), textcoords="offset points",
                                fontsize=7, color="#2ca02c")
    fig2.suptitle("The order of events: the two jet-leg groups on the truncation clock"
                  + ("   [split inside quartiles of $\\log M_\\star$]" if P14_MCTRL else "")
                  + "   —   $\\blacktriangledown$ the jet, $|$ the first X-ray firing", y=0.999, fontsize=12,
                  fontweight="bold")
    fig2.tight_layout(rect=(0, 0, 1, 0.978)); paper_save(fig2, "kappa_jetleg_clocks"); plt.show()

    # ══ (G) the epochs table: when each group is where ══════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(d) THE ORDER OF EVENTS — the group medians on the truncation clock (the claim needs the ISM to compactify")
    print("    BEFORE the truncation, i.e. during the jet leg, not after it)")
    print(f"  {'box':10s} {'quantity':12s} {'group':6s} " + " ".join(f"{f't{e:+.0f}':>8s}" for e in (-2, -1, 0, 1, 2))
          + f" {'anchor':>8s}")
    _iep = {e: int(np.argmin(np.abs(TG - e))) for e in (-2, -1, 0, 1, 2)}
    for b in _BX14:
        F, T = H[b]["F"], H[b]["T"]
        g = D14[D14.box == b]
        _gmap = dict(zip(zip(g.snap.astype(int), g.gal_id.astype(int)), g.grp))
        _grp = np.array([_gmap.get(k, "") for k in zip(F["snap"].astype(int), F["gal_id"].astype(int))])
        for q in _TRK14:
            if q not in T:
                continue
            for k in ("long", "short"):
                m = (_grp == k)
                if m.sum() < 3:
                    continue
                A = T[q]["sft"][m]
                _anc = g.loc[g.grp == k, q].to_numpy(float) if q in g.columns else np.array([np.nan])
                print(f"  {b:10s} {q:12s} {k:6s} " + " ".join(f"{np.nanmedian(A[:, _iep[e]]):8.2f}" for e in (-2, -1, 0, 1, 2))
                      + f" {np.nanmedian(_anc):8.2f}")
                ROW14.append(dict(box=b, block="epochs", quantity=q, kcls=k, n_gal=int(m.sum()),
                                  med=float(np.nanmedian(A[:, _iep[0]])), lo=float(np.nanmedian(A[:, _iep[-2]])),
                                  hi=float(np.nanmedian(_anc))))
    pd.DataFrame(ROW14).to_csv(os.path.join(OUT, "kappa_jetleg.csv"), index=False)
    D14.to_csv(os.path.join(OUT, "kappa_jetleg_galaxies.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_jetleg.csv')}"
          f"\n         {os.path.join(OUT, 'kappa_jetleg_galaxies.csv')} (one row per branch)")
```

## Part 15 — the state at the jet activation: how much cold ISM the X-ray channel finds when it fires

**What Part 14's clock figure showed, read as the physics.** In every box the **short** jet leg is activated *on a
spike*: the gas compactifies, $\Sigma_{\rm gas}$, $\Sigma_{\rm H_2}$ and the dust all peak together, and
$\kappa_{\rm rot}$(gas) falls at the same moment. In `cis50nox` the compactness barely moves, but the H₂ density still
peaks for the short leg and the rotation still drops. **After** the truncation the two groups run *parallel* in
everything except the dust — and in `cis50nox` the slopes are shallower but still parallel, $f_{\rm dust}$ decays
smoothly and far more slowly than in the X-ray runs, while $\kappa$ climbs back fast.

So the X-ray channel is the factor that matters: it **steepens** the disappearance of the cold ISM and it **stirs**
what is left. SIMBA's X-ray feedback (Davé et al. 2019) deposits half the energy in ISM gas as heat and half as a
**radial kick**, so the stirring is momentum delivered to whatever dense gas happens to be there when the channel
opens. A galaxy that has already lived a long time in the jet mode has lost part of its central H₂ by then: the same
channel finds less to push, moves fewer particles, and its dust survives. A galaxy that reaches the gate *on* the
compaction spike hands the channel a dense central ISM and is stirred.

### The division this part makes

Not a duration — **a state**: the H₂ surface density **at the jet activation**,

$$\text{DENSE at jet-on} \quad\text{vs}\quad \text{PRE-DEPLETED at jet-on}, \qquad
\Sigma_{\rm H_2}(t_{\rm jet}) \gtrless \text{the box median},$$

with `P15_EPOCH` able to read the same axis at the **X-ray onset** ($t_{x1}$, where the mechanism says the coupling
actually happens) or at the truncation, and `P15_CAUSE` able to swap the axis for $\Sigma_{\rm gas}$, $f_{\rm H_2}$,
sSFR, the compactness or $\kappa$. All of them are scored **side by side against the same outcome** in block (a), so
the choice of axis is made on the data and not here. Where Part 3's kinematic sidecar was never built the surface
density has no radius to stand on and the axis falls back to the mass fraction of the same phase, loudly
(`P15_FALLBACK`).

**The names are a claim, so the part measures them.** *Post-starburst* versus *normal quiescent* requires a burst:
`f_burst` is the stellar mass formed in the last `P15_BURST` Gyr before the truncation, **integrated from the branch's
own SFR** so that a merger cannot fake it; `b_ssfr` is the sSFR peak over that branch's own earlier baseline; and the
mass-weighted stellar age at the anchor is the third leg — a post-starburst branch has to be bursty **and** end young.

### The decay, fitted galaxy by galaxy

From the truncation to the anchor, each galaxy's dust, sSFR and H₂ are fitted with a **double exponential**

$$y(t) \;=\; A_1\,e^{-(t-t_{\rm SFT})/\tau_1} \;+\; A_2\,e^{-(t-t_{\rm SFT})/\tau_2}, \qquad \tau_1 \le \tau_2 ,$$

in $\log_{10}$ (the residual is in dex, so a two-decade decline is not owned by its first point). The single
exponential the model nests **is a straight line** in $\log_{10} y$, so the two are compared by BIC on the same
points: $\Delta\mathrm{BIC} > 0$ means the bend is worth its two extra parameters. Out come $\tau_1$ (the initial
e-folding), $\tau_2$ (the late one), $t_{\rm break}$ (where the slow term takes over), $f_{\rm slow}$ (its share at
the truncation) — and beside them the **model-free** slopes over the first and the last `P15_DWIN` Gyr, which is the
same statement without a model. **A point sitting on a floor is censored, not fitted**: $\log M_{\rm dust}/M_\star$ is
clipped at `P9_FLOOR` and the sSFR at `SSFR_FLOOR`, and a flat run of floored values would *be* the flattening the fit
is being asked to find; the time the branch first reached the floor is kept instead (`t_floor`).

The prediction under test is the user's: the two branches leave the truncation on **the same initial slope**, and then
the pre-depleted one **bends to something flatter** — a long $\tau_2$ and a $t_{\rm break}$ inside the sampled window
there — while the sSFR is shallower for it throughout. In AUC terms: `s_early` near 0.5, `s_late` not.

> **Superseded by Part 18.** Read back against the data this fit is *not identified*: two of its fits in five sit on one
> of the optimiser's bounds and three quarters give $f_{\rm slow} < 1\%$, so its $\tau_1$ / $\tau_2$ are the bounds and
> not measurements, and its median-track figure flattens partly because the fastest branches reach the floor and leave the
> sample. **Part 18 refits the same claim** with a broken exponential in $\log_{10} y$ profiled over the break, the
> floored points kept as left-censored, and the anchor and the snapshot spacing controlled — and answers the distribution
> question ("is $\tau$ bimodal?") there. Blocks (c), FIG 3 and FIG 4 here are kept for continuity and for the comparison
> Part 18 draws against them; the numbers to quote are Part 18's.

### The figures

### The anchor test, and where its outflow comes from

The outflow axis is **not** Part 13's fixed 10 kpc shell. It is measured through a shell at **each galaxy's own
$R_{50}(\star)$** — $|r - R_{50}^\star| < $ `P15_ORW` $R_{50}^\star$ — from the **anchor** reduced particle file, one
file per branch, cached in `OUT`: the member stars give both the radius and the rest frame, the Hubble flow inside the
cutout is added, and $\dot M_{\rm out} = \sum_{v_r>0} m\,v_r / \text{width}$ for the gas, the cold gas and the dust,
with the 90th-percentile outflow speed beside them. `P15_OQ` picks which of them is the axis (the default is
$\dot M_{\rm out}/M_\star$, so that "strong" is not just "massive"). This is the one block of Part 15 that reads
particles, and it exists only for the boxes that have reduced particle files — m25, and whatever snapshots of m100 are
on disk. Every score is printed twice: raw, and with the axis **ranked inside stellar-mass cells**.

### The open question, stated as it stands

A long jet leg should also have **heated the CGM**, preventing accretion and preventing the compaction in the first
place. What happens afterwards is genuinely undecided, and the part is built to show which way it falls: either the
jet mode *continues* in the stirred, compact branch and turns **preventive** (no new accretion, no new dust), or the
CGM of the other branch **cools** and lets its ISM stay dense. FIG 5 is the direct test in every box; FIG 7 is the
only place the CGM itself can be seen, and it exists for m25 alone.

### The controls, and what is switched off

The axis is ranked **inside cells of $\log M_\star$** before the split (`P15_MCTRL`), every score is repeated as a
partial rank correlation at fixed $\log M_\star$ and $\log M_{\rm BH}$, **and once more at fixed epoch** — Part 14's
jet leg carries a $+1.5$ to $+2.8$ Gyr difference in $t_{\rm SFT}$ by itself, i.e. it is half a statement about cosmic
time, and any state measured at $t_{\rm jet}$ inherits that risk. With `P15_ZCTRL` the groups are re-formed inside
cells of the epoch as well and the key contrasts printed twice. `log_mstar`, `log_mbh`, `t_sft` and the epoch itself
are carried as CONTROL rows of the contrast table. The **stirring test** — the change in $\kappa$ across the X-ray
onset against the $\Sigma_{\rm H_2}$ the onset found, with `cis50nox` as its null — is kept as code but is **off**
(`P15_STIRTEST`); switch it on to get block (e) and a third panel of FIG 5 back.

*A pure read* of `ism_prediction_histories_<box>.hdf5` restricted to Part 3's rows, plus Part 3's kinematic sidecar for
the radii and $\kappa$, plus one anchor particle file per branch for the outflow →
`kappa_jetstate_{clocks,dustage,decay,slopes,after,auc,cgm}{.png,.pdf}`, `kappa_jetstate.csv`,
`kappa_jetstate_galaxies.csv` (one row per branch, every column of the census) and
`kappa_jetstate_outflow_<box>.csv` (the cached shell fluxes).


```python
# ── Part 15 — THE STATE AT THE JET ACTIVATION: how much cold ISM the X-ray channel finds when it fires ──
# WHAT PART 14's CLOCK FIGURE ACTUALLY SHOWED (the user's reading of it, which this part turns into a division):
#   in EVERY box the SHORT jet leg is activated ON a spike — the gas compactifies, Sigma_gas and Sigma_H2 and the dust all
#   peak, and kappa_rot(gas) drops at the same time. In cis50nox the compactness barely moves, but the H2 density still
#   peaks for the short leg and the rotation still falls. AFTER the truncation the two groups run PARALLEL in everything
#   except the dust; in cis50nox the slopes are shallower but still parallel, f_dust falls smoothly and far more slowly
#   than in the X-ray runs, and kappa climbs back fast.
# THE READING: the X-ray channel is the factor. It steepens the disappearance of the cold ISM and it STIRS what is left.
#   SIMBA's X-ray feedback (Dave+19) puts half of the energy into ISM gas as heat and half as a RADIAL KICK, so the
#   stirring is momentum delivered to whatever dense gas is there when the channel opens. A galaxy that had already been
#   in the jet mode for a long time has lost part of its central H2 by then: the same channel finds less to push, moves
#   fewer particles, and the dust survives. A galaxy that reaches the gate ON the compaction spike hands the channel a
#   dense central ISM and is stirred.
# SO THE DIVISION IS NOT A DURATION BUT A STATE: high / low H2 surface density AT the jet activation (P15_EPOCH = "jet";
#   "x1" reads the same axis at the X-ray onset, which is where the mechanism says the coupling happens, and "sft" at the
#   truncation — all three are scored side by side so the choice is made on the data and not here).
# THE NAMES THE TWO BRANCHES ARE PROPOSED TO CARRY — post-starburst vs normal quiescent — are a claim, not a label, so the
#   part MEASURES them: the stellar mass formed in the last P15_BURST Gyr before the truncation (integrated from the
#   branch's own SFR, so a merger cannot fake it), the sSFR peak over that branch's own baseline, and the mass-weighted
#   stellar age at the anchor. A post-starburst branch has to be bursty AND end young.
# WHAT THIS PART DOES
#   (1) THE DIVISION: the axis at the epoch, the two groups, and whether it divides the outcome better than Part 14's
#       jet-leg duration — every candidate axis scored raw, at fixed log M* and log M_BH, and at fixed epoch.
#   (2) FIG 1 the clock, Part 14's figure with the new split and the jet activation as the origin: the compaction spike
#       is at zero by construction, so what is being read is what the two groups do AROUND it and after it.
#   (3) FIG 2 the f_dust - age plane AT THE ANCHOR: THE TRACKS ONLY, and the CRITICAL POINTS on them — the sSFR peak,
#       the jet activation, the truncation, the anchor. The age along the branch is integrated BACK from the
#       catalogue's own mass-weighted age at the anchor (see _agetrk15: exact when nothing forms, and biased young by
#       merged-in stars, which is why the pre-SFT leg is drawn faint).
#   (4) THE DECAY, FITTED: a DOUBLE EXPONENTIAL per galaxy from the truncation to the anchor, in the dust, the sSFR and
#       the H2 (_decay15). The claim being tested is the user's: the two branches leave the truncation on the SAME
#       initial slope and then the pre-depleted one BENDS to something flatter, while the sSFR is shallower there
#       throughout. tau1 / tau2 / t_break / f_slow per galaxy, the BIC against the single exponential the model nests,
#       and beside them the model-free slopes over the first and the last P15_DWIN Gyr — FIG 3 and FIG 4.
#   (5) THE AFTER, which is the open question: does the jet keep firing in the stirred branch and turn PREVENTIVE (no
#       accretion, no dust), while the other branch's CGM cools and keeps its ISM dense? The post-SFT state budget, the
#       gas re-accretion (FIG 5) and, where Part 13 has been run, the m25 CGM itself. The STIRRING TEST is kept as code
#       but is OFF (P15_STIRTEST).
#   (6) THE ANCHOR TEST (FIG 6): the two observables against the dust label they are supposed to predict — is a
#       ROTATING galaxy dusty too, and is one with a STRONG OUTFLOW RIGHT NOW not? The outflow is measured through a
#       shell at each galaxy's OWN R50(stars) — not the fixed 10 kpc shell of Part 13 — from the anchor particle file
#       (_oflow15), so this block reads particles where the rest of the part reads only catalogue chains.
# A PURE READ of ism_prediction_histories_<box>.hdf5 restricted to Part 3's rows, plus Part 3's kinematic sidecar for the
# radii and kappa — and, for the outflow axis alone, one reduced particle file per ANCHOR (cached in OUT). NOTE that in cis50nox the "X-ray" state is COUNTERFACTUAL: agn_states applies SIMBA's rule, but that
# run has the heating compiled out — the state says the channel WOULD have fired, and the figures use it as such.
P15_BOXES  = None          # None = the boxes Part 3 built, the channel-off box last
P15_EPOCH  = "jet"         # THE epoch the division is read at: "jet" = the jet activation | "x1" = the X-ray onset | "sft"
P15_CAUSE  = "lsig_h2"     # THE axis at that epoch: "lsig_h2" | "lsig_gas" | "lfh2" | "lssfr" | "comp" | "kappa_gas" | "lfd"
P15_MCTRL  = ["log_mstar"] # rank the axis INSIDE quantile cells of these columns before splitting ([] = the plain rank)
P15_MQ     = 4             # how many cells per control column
P15_ZCTRL  = True          # ALSO re-form the groups inside cells of the epoch itself, and print both scores side by side
P15_BASE   = (0.5, 2.0)    # the spike is measured against the median over [epoch - 2.0, epoch - 0.5] Gyr
P15_BURST  = 1.0           # the burst window before the SFT [Gyr]: the mass formed in it, and the sSFR peak over it
P15_JETDEF = "first"       # "first" = the first jet step | "sustained" = the first one at least half of whose next P15_SUST Gyr is jet
P15_SUST   = 0.5
P15_STIR   = 1.0           # THE STIRRING TEST window [Gyr]: the change across the X-ray onset, from t_x1 to t_x1 + this
P15_TRK    = ["lsig_h2", "comp", "kappa_gas", "lssfr", "lfgas", "lfd"]   # the rows of FIG 1
P15_PLANE  = ("age", "lfd")     # FIG 2's plane: (x, y), drawn as tracks and as the anchor cloud
P15_NBOOT  = 500
P15_NMIN   = 5             # a clock epoch is drawn only where this many galaxies contribute
P15_FALLBACK = True        # where the surface density is unmeasurable (no kinematic sidecar), divide on the mass fraction
P15_MINCOV = 0.30          # ... below this share of branches carrying the axis
P15_CGM    = True          # the m25 CGM block, if Part 13 has been run in this kernel
P15_Q      = 0.5           # the split: hi = the top P15_Q of the ranked axis, lo = the bottom (0.5 = a median split)
P15_GK     = ["hi", "lo"]
P15_GCOL   = {"hi": "#d6604d", "lo": "#4393c3"}
P15_GLAB   = {"hi": "DENSE at jet-on", "lo": "PRE-DEPLETED at jet-on"}
P15_GSH    = {"hi": "DENSE", "lo": "DEPLETED"}
P15_GLONG  = {"hi": "dense ISM when the jet turns on — the compaction branch",
              "lo": "thin ISM when the jet turns on — the pre-depleted branch"}
P15_ELAB   = {"jet": r"$t_{\rm jet}$", "x1": r"$t_{x1}$", "sft": r"$t_{\rm SFT}$"}
P15_ENAM   = {"jet": "the jet activation", "x1": "the X-ray onset", "sft": "the truncation"}
P15_QLAB   = {"lsig_h2": r"$\log \Sigma_{\rm H_2}$", "lsig_gas": r"$\log \Sigma_{\rm gas}$",
              "comp": r"$-\log R_{50}({\rm gas})/R_{50}(\star)$", "kappa_gas": r"$\kappa_{\rm rot}$(gas)",
              "lfd": r"$\log M_{\rm dust}/M_\star$", "lfh2": r"$\log 1.36\,M_{\rm H_2}/M_\star$",
              "lfgas": r"$\log M_{\rm gas}/M_\star$", "lssfr": r"$\log$ sSFR  [yr$^{-1}$]", "lms": r"$\log M_\star$",
              "lr_gas": r"$\log R_{50}({\rm gas})$  [kpc]", "age": r"mass-weighted stellar age  [Gyr]"}
P15_VKEYS  = ["lsig_h2", "lsig_gas", "comp", "kappa_gas", "lfd", "lfh2", "lfgas", "lssfr", "lms", "lr_gas", "age"]
P15_STIRTEST = False       # the stirring-test block (e) and its panel of FIG 4 — OFF: the user dropped it. The code is
                           # kept because it is the only falsifiable form of the kick mechanism (its null is cis50nox)
# ── THE DECAY: a double exponential per galaxy, from the truncation to the anchor ──────────────────────────────────
P15_DQ     = ["lfd", "lssfr", "lfh2"]   # the quantities fitted (log10; the fit is on the linear quantity behind them)
P15_DMAX   = 4.0           # the decay clock t - t_SFT the figures are drawn on [Gyr]
P15_DWIN   = 1.0           # the MODEL-FREE windows: the slope over the first / the last this many Gyr [dex/Gyr]
P15_DMIN   = 6             # the double exponential needs this many un-floored points ...
P15_DSPAN  = 1.0           # ... spanning at least this much time [Gyr]
P15_DFLOOR = 0.05          # a point within this many dex of a floor is CENSORED, not fitted (see _decay15)
P15_PKW    = 3.0           # the sSFR peak marked in the plane is the maximum over this window before the SFT [Gyr]
# ── THE ANCHOR TEST: is a rotator dusty, is a galaxy with a strong outflow not? ────────────────────────────────────
P15_AUCQ   = None          # the axes scored against the dust label at the anchor (None = kappa_rot(gas) and the outflow)
P15_ABINS  = 9             # the bins of the histograms, and of the dusty fraction drawn over them
P15_OUTFLOW = True         # measure the CURRENT outflow from the particle files (one anchor file per galaxy, cached)
P15_OQ     = "lsout"       # THE outflow axis: "lsout" = log Mdot_out/M* | "lmdot_out" | "lsout_c" (cold) | "v_out90"
P15_ORW    = 0.5           # the shell is |r - R50*| < this x R50(stars) — the user's radius, in place of a fixed 10 kpc
P15_OTCOLD = 1e5           # [K] the cold phase of the outflow
P15_OHUB, P15_OOM = True, 0.3      # the Hubble flow inside the cutout, as in Parts 10 and 13
P15_ODIR   = os.path.join(os.getcwd(), "output", "{box}", "reduced_particles")
P15_OPREFIX = {"cis25": "m25n512", "cis50": "m50n512", "cis50nox": "m50n512", "cis100": "m100n1024"}
P15_OCACHE = os.path.join(OUT, "kappa_jetstate_outflow_{box}.csv")   # (snap, gx) -> the fluxes; delete to re-measure

TDG    = np.arange(0.0, P15_DMAX + 1e-9, 0.25)      # THE DECAY CLOCK: t - t_SFT [Gyr], the grid the decay is drawn on
_TK15  = P15_VKEYS + [f"dec_{q}" for q in P15_DQ]   # the tracks carried: the epoch clock, and the decay clock
_GRID15 = {q: (TDG if q.startswith("dec_") else TG) for q in _TK15}
_DFL15 = {"lfd": P9_FLOOR, "lssfr": float(np.log10(SSFR_FLOOR))}   # the CENSORED values: a floor is not a measurement

D15, T15 = pd.DataFrame(), {}
_BX15 = [b for b in (list(H) if P15_BOXES is None else P15_BOXES) if b in H]
_BX15 = [b for b in _BX15 if b != NOX_BOX] + [b for b in _BX15 if b == NOX_BOX]
if not _BX15:
    print("Part 15: Part 3 built no branch records — run it first.")
else:
    from scipy import stats as _st15
    import glob as _glob15, re as _re15
    try:
        from scipy.optimize import least_squares as _lsq15
    except Exception:                                              # without it only the model-free slopes are measured
        _lsq15 = None

    def _at15(t, y, t0):
        """y at t0, linearly interpolated on the branch's own sampling and NEVER extrapolated past its ends."""
        ok = np.isfinite(t) & np.isfinite(y)
        if not np.isfinite(t0) or ok.sum() < 2:
            return np.nan
        tt, yy = t[ok], y[ok]
        return float(np.interp(t0, tt, yy)) if (tt[0] - 1e-9) <= t0 <= (tt[-1] + 1e-9) else np.nan

    def _win15(t, y, lo, hi):
        """the median of y over the window [lo, hi] of the branch's own time — NaN if the branch does not cover it."""
        ok = np.isfinite(t) & np.isfinite(y) & (t >= lo) & (t <= hi)
        return float(np.median(y[ok])) if ok.sum() else np.nan

    def _lslope15(tau, yl, lo, hi):
        """the least-squares slope of log10 y over the window [lo, hi] of the decay clock [dex/Gyr]; NaN under 3 points."""
        m = np.isfinite(tau) & np.isfinite(yl) & (tau >= lo - 1e-9) & (tau <= hi + 1e-9)
        if m.sum() < 3 or (np.nanmax(tau[m]) - np.nanmin(tau[m])) < 0.2:
            return np.nan
        return float(np.polyfit(tau[m], yl[m], 1)[0])

    def _decay15(tau, yl, floor=None):
        """ONE galaxy's decay from the truncation to the anchor: log10 y against tau = t - t_SFT.
        (i) MODEL-FREE — the slope of log10 y over the FIRST P15_DWIN Gyr and over the LAST P15_DWIN Gyr, in dex/Gyr.
            The claim being tested ("the same initial slope, and then the pre-depleted branch bends to something
            flatter") is exactly: s_early the same in the two branches, s_late not.
        (ii) THE DOUBLE EXPONENTIAL  y = A1 exp(-tau/tau1) + A2 exp(-tau/tau2),  tau1 <= tau2,  fitted to log10 y — the
            residual is in dex, which is what a two-decade decline needs if its first point is not to own the fit. The
            single exponential it nests IS a straight line in log10 y, so the two are compared by BIC on the same
            points: d BIC > 0 means the bend is worth its two extra parameters. tau1 is the initial e-folding, tau2 the
            late one, t_break the time the slow term takes over, f_slow its share of y at the truncation.
        POINTS ON A FLOOR ARE CENSORED, NOT FITTED: log M_dust/M* is clipped at P9_FLOOR and the sSFR at SSFR_FLOOR, and
        a flat run of floored values would BE the flattening this part is looking for. The time the branch first
        reaches the floor is kept instead (t_floor), which is the honest version of "it went to zero here"."""
        out = dict(n=0, span=np.nan, t_floor=np.nan, s_early=np.nan, s_late=np.nan, ds=np.nan, lA1=np.nan, tau1=np.nan,
                   lA2=np.nan, tau2=np.nan, tbrk=np.nan, fslow=np.nan, dbic=np.nan, rms=np.nan)
        tau, yl = np.asarray(tau, float), np.asarray(yl, float)
        ok = np.isfinite(tau) & np.isfinite(yl) & (tau >= -1e-9)
        if floor is not None:
            fl = ok & (yl <= floor + P15_DFLOOR)
            out["t_floor"] = float(np.min(tau[fl])) if fl.any() else np.nan
            ok = ok & ~fl
        if ok.sum() < 3:
            return out
        tau, yl = tau[ok], yl[ok]
        o = np.argsort(tau); tau, yl = tau[o], yl[o]
        out["n"], out["span"] = int(len(tau)), float(tau[-1] - tau[0])
        out["s_early"] = _lslope15(tau, yl, 0.0, P15_DWIN)
        out["s_late"] = _lslope15(tau, yl, tau[-1] - P15_DWIN, tau[-1])
        out["ds"] = out["s_late"] - out["s_early"] if np.isfinite(out["s_late"]) and np.isfinite(out["s_early"]) else np.nan
        if len(tau) < P15_DMIN or out["span"] < P15_DSPAN or _lsq15 is None:
            return out
        r1 = yl - np.polyval(np.polyfit(tau, yl, 1), tau)          # the single exponential, on the same points
        ssr1, n_ = float(np.sum(r1 ** 2)), len(tau)

        def _res15(p):
            A1, t1, A2, t2 = 10.0 ** p[0], 10.0 ** p[1], 10.0 ** p[2], 10.0 ** p[1] + 10.0 ** p[3]
            return np.log10(np.clip(A1 * np.exp(-tau / t1) + A2 * np.exp(-tau / t2), 1e-300, None)) - yl

        t1g = float(np.clip(0.4343 / abs(out["s_early"]), 0.05, 20.0)) if (np.isfinite(out["s_early"]) and out["s_early"] < -1e-6) else 1.0
        lo = [float(yl.min()) - 4.0, np.log10(0.05), float(yl.min()) - 8.0, -2.0]
        hi = [float(yl.max()) + 3.0, np.log10(20.0), float(yl.max()) + 3.0, np.log10(50.0)]
        p0 = np.clip([yl[0] + np.log10(0.9), np.log10(t1g), yl[0] - 1.0, np.log10(max(2.0 * t1g, 0.06))], lo, hi)
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                f = _lsq15(_res15, p0, bounds=(lo, hi), max_nfev=4000)
        except Exception:
            return out
        A1, t1, A2 = 10.0 ** f.x[0], 10.0 ** f.x[1], 10.0 ** f.x[2]
        t2, ssr2 = t1 + 10.0 ** f.x[3], float(np.sum(np.asarray(f.fun) ** 2))
        out.update(lA1=float(f.x[0]), tau1=float(t1), lA2=float(f.x[2]), tau2=float(t2),
                   rms=float(np.sqrt(ssr2 / n_)), fslow=float(A2 / (A1 + A2)) if (A1 + A2) > 0 else np.nan,
                   dbic=float(n_ * np.log(max(ssr1, 1e-30) / n_) + 2 * np.log(n_)
                              - n_ * np.log(max(ssr2, 1e-30) / n_) - 4 * np.log(n_)))
        if A1 > 0 and A2 > 0 and t2 > t1 * (1 + 1e-6):
            tb = float(np.log(A1 / A2) / (1.0 / t1 - 1.0 / t2))
            out["tbrk"] = tb if 0.0 <= tb <= tau[-1] else np.nan
        return out

    def _dcurve15(lA1, tau1, lA2, tau2, span, grid):
        """the fitted double exponential on the decay grid, NEVER beyond the range the galaxy was fitted over."""
        if not all(np.isfinite([lA1, tau1, lA2, tau2, span])):
            return np.full(len(grid), np.nan)
        with np.errstate(all="ignore"):
            y = np.log10(np.clip(10.0 ** lA1 * np.exp(-grid / tau1) + 10.0 ** lA2 * np.exp(-grid / tau2), 1e-300, None))
        return np.where(grid <= span + 1e-9, y, np.nan)

    KMS15 = 1.0227121650537077e-9              # (Msun km/s / kpc) -> Msun/yr, as in Parts 10 and 13

    def _wq15(v, w, q):
        """the weighted quantile of v, used for the speed of the outflowing gas."""
        ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
        if not ok.any():
            return np.nan
        v, w = v[ok], w[ok]
        o = np.argsort(v); v, w = v[o], w[o]
        return float(np.interp(q, (np.cumsum(w) - 0.5 * w) / w.sum(), v))

    def _oflow15(path):
        """ONE reduced particle file -> the flux through a shell at THAT galaxy's own R50(stars).
        The velocity is taken in the galaxy's rest frame (the mass-weighted velocity of its member stars, the Hubble
        flow inside the cutout added), the shell is |r - R50*| < P15_ORW R50* and, for the gas / the dust and for the
        whole / the cold phase,   Mdot_out = sum_{v_r > 0} m v_r / (the shell width).
        R50* is measured from the member stars of the file itself, so the radius is the galaxy's, not a fixed 10 kpc."""
        with h5py.File(path, "r") as h:
            A = dict(h.attrs); G, St = h["gas"], h["star"]
            gp = G["pos"][:].astype(float); gmem = G["member"][:].astype(bool)
            gm = G["m_gas"][:].astype(float); gd = G["m_dust"][:].astype(float)
            gT = G["temp"][:].astype(float) if "temp" in G else np.full(len(gm), np.nan)
            gv = G["vel"][:].astype(float) if "vel" in G else np.full((len(gm), 3), np.nan)
            sp = St["pos"][:].astype(float); sm = St["m_star"][:].astype(float)
            smem = St["member"][:].astype(bool)
            sv = St["vel"][:].astype(float) if "vel" in St else np.full((len(sm), 3), np.nan)
        z, hub = float(A["redshift"]), float(A["hub"])
        r, rs = np.linalg.norm(gp, axis=1), np.linalg.norm(sp, axis=1)
        row = dict(snap=int(A["snap"]), gx=int(A["gx"]), z=z, r50s=np.nan, n_sh=0, v_out90=np.nan, f_out=np.nan,
                   m_star_mem=float(np.nansum(sm[smem])) if smem.any() else np.nan)
        _sok = smem & (sm > 0)
        if _sok.sum() >= 5 and np.nansum(sm[_sok]) > 0:
            _o = np.argsort(rs[_sok])
            row["r50s"] = float(np.interp(0.5, np.cumsum(sm[_sok][_o]) / np.nansum(sm[_sok]), rs[_sok][_o]))
        _svok, _gvok = _sok & np.isfinite(sv).all(1), gmem & np.isfinite(gv).all(1) & (gm > 0)
        vsys = (np.average(sv[_svok], axis=0, weights=sm[_svok]) if _svok.sum() >= 10 else
                np.average(gv[_gvok], axis=0, weights=gm[_gvok]) if _gvok.sum() >= 10 else np.full(3, np.nan))
        if not (np.isfinite(row["r50s"]) and row["r50s"] > 0 and np.isfinite(vsys).all()):
            return row
        _hz = ((100.0 * hub * np.sqrt(P15_OOM * (1 + z) ** 3 + 1 - P15_OOM)) / 1e3) if P15_OHUB else 0.0
        _v3 = (gv - vsys) + _hz * gp
        vr = (_v3 * (gp / np.clip(r, 1e-9, None)[:, None])).sum(1)
        dR = P15_ORW * row["r50s"]
        sh = (np.abs(r - row["r50s"]) < dR) & np.isfinite(vr) & (gm > 0)
        row["n_sh"] = int(sh.sum())
        cold = np.isfinite(gT) & (gT < P15_OTCOLD)
        for _wn, _wv in (("g", gm), ("d", gd)):
            for _pn, _pm in (("", np.ones(len(gm), bool)), ("c", cold)):
                _w = np.where(sh & _pm & np.isfinite(_wv), _wv, 0.0)
                row[f"mdot_out_{_wn}{_pn}"] = float(np.sum(_w * np.clip(vr, 0, None)) / (2 * dR) * KMS15)
                row[f"mdot_in_{_wn}{_pn}"] = float(-np.sum(_w * np.clip(vr, None, 0)) / (2 * dR) * KMS15)
        _ow = sh & (vr > 0)
        if _ow.any() and np.nansum(gm[sh]) > 0:
            row["v_out90"] = _wq15(vr[_ow], gm[_ow], 0.90)
            row["f_out"] = float(np.nansum(gm[_ow]) / np.nansum(gm[sh]))
        return row

    def _outpass15(box, keys):
        """the anchor node of every branch of one box, measured once and cached -> (frame, a message)."""
        _pf = P15_OPREFIX.get(box); _dir = P15_ODIR.format(box=box)
        if _pf is None or not os.path.isdir(_dir):
            return pd.DataFrame(), f"[{box}] no reduced particle files ({_dir}) — the outflow axis is not measured here"
        _cache = P15_OCACHE.format(box=box)
        Z = pd.read_csv(_cache) if os.path.exists(_cache) else pd.DataFrame()
        _done = set(zip(Z["snap"].astype(int), Z["gx"].astype(int))) if len(Z) else set()
        _need = [k for k in keys if k not in _done]
        _inv, _miss = {}, 0
        for _s in sorted({int(s_) for s_, _ in _need}):
            for _p in _glob15.glob(os.path.join(_dir, f"snap_{_s:03d}", f"{_pf}_snap*_gal*.h5")):
                _m = _re15.search(r"_snap(\d+)_gal(\d+)\.h5$", _p)
                if _m:
                    _inv[(int(_m.group(1)), int(_m.group(2)))] = _p
        _todo = [k for k in _need if k in _inv]
        _miss = len(_need) - len(_todo)
        if _todo:
            print(f"  [{box}] measuring the outflow at R50(stars) in {len(_todo)} anchor files "
                  f"(~{sum(os.path.getsize(_inv[k]) for k in _todo) / 2 ** 30:.1f} GiB, cached afterwards)", flush=True)
            _new, _bad = [], []
            for _k in _todo:
                try:
                    _new.append(_oflow15(_inv[_k]))
                except Exception as _e:
                    _bad.append(f"{os.path.basename(_inv[_k])}: {type(_e).__name__}: {_e}")
            for _b in _bad[:3]:
                print(f"     file not read, {_b}")
            if _new:
                Z = pd.concat([Z, pd.DataFrame(_new)], ignore_index=True)
                Z.to_csv(_cache, index=False)
        if not len(Z):
            return pd.DataFrame(), f"[{box}] no anchor file could be measured"
        return Z, (f"[{box}] the outflow at R50(stars): {len(Z)} anchors cached -> {os.path.basename(_cache)}"
                   + (f";  {_miss} branches have no particle file at their anchor" if _miss else ""))

    def _agetrk15(t, ms, sfr, age_anc, ka):
        """The mass-weighted stellar age BACK along the branch, anchored on the catalogue's own age at the anchor.
        Between two steps the mass M_i already there ages by dt and the mass formed in between (the branch's own SFR,
        integrated, so that merged-in stars are not counted as newly born) enters with mean age dt/2:
            A_{i+1} = [ M_i (A_i + dt) + dM dt/2 ] / M_{i+1}   ->   A_i = [ A_{i+1} M_{i+1} - dM dt/2 ] / M_i - dt.
        It is EXACT where nothing forms (the age then simply ticks with the clock, which is the post-truncation leg this
        part draws) and it inherits every error of the SFH before that — mass brought in by mergers is old and is not
        credited, so the reconstructed age runs SHORT in the star-forming leg. That leg is drawn faint in FIG 2."""
        A = np.full(len(t), np.nan)
        if not (np.isfinite(age_anc) and 0 <= ka < len(t)):
            return A
        A[ka] = age_anc
        for i in range(ka - 1, -1, -1):
            dt = float(t[i + 1] - t[i])
            if not (np.isfinite(ms[i]) and ms[i] > 0 and np.isfinite(A[i + 1]) and dt > 0):
                break
            dm = float(np.clip(0.5 * (np.nan_to_num(sfr[i]) + np.nan_to_num(sfr[i + 1])) * dt * 1e9, 0.0, ms[i + 1]))
            a = (A[i + 1] * ms[i + 1] - dm * 0.5 * dt) / ms[i] - dt
            if not np.isfinite(a) or a <= 0:
                break
            A[i] = a
        return A

    def _kinalign15(gh, K, o):
        """Part 3's sidecar alignment: the (n_t, n_gal) block of each kinematic column, matched on the SNAPSHOT numbers
        and the galaxy ids themselves (never on the array order), in the histories' own time ordering `o`."""
        out = {}
        _sh = np.asarray(gh["snaps"][:]).astype(int)[o] if "snaps" in gh else None
        _gid = np.asarray(gh["gal_ids"][:]).astype(np.int64)
        if K is None or _sh is None:
            return out
        _rs = {int(v): i for i, v in enumerate(np.asarray(K["snaps"]).astype(int))}
        _ri = {int(v): j for j, v in enumerate(np.asarray(K["gal_ids"]).astype(np.int64))}
        _row = np.array([_rs.get(int(v), -1) for v in _sh]); _col = np.array([_ri.get(int(v), -1) for v in _gid])
        _mr, _mc = _row >= 0, _col >= 0
        for c in ("kappa_gas", "r_gas", "r_star"):
            if c in K and _mr.any() and _mc.any():
                B = np.full((len(_sh), len(_gid)), np.nan)
                B[np.ix_(_mr, _mc)] = np.asarray(K[c], float)[np.ix_(_row[_mr], _col[_mc])]
                out[c] = B
        return out

    def _pass15(box, want):
        """ONE pass over one box's histories, restricted to Part 3's rows: the STATE at each of the three epochs, the
        spike it sits on, the burst that preceded it, the post-SFT budget, the stirring across the X-ray onset — and the
        tracks of every quantity on the clock of P15_EPOCH."""
        rows, TR = [], {q: [] for q in _TK15}
        KIN = _kinread3(box) if "_kinread3" in globals() else {}
        with h5py.File(HIST_FILE.format(box=box), "r") as f:
            for gname in f:
                a = int(gname.replace("snap", ""))
                if not any(k[0] == a for k in want):
                    continue
                gh = f[gname]
                t = gh["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
                if len(t) < 3:
                    continue
                ms, sfr, mg, mbh, fedd = (gh[c][:][o] for c in ("mstar", "sfr", "mgas", "mbh", "fedd"))
                md = gh["mdust"][:][o] if "mdust" in gh else np.full_like(ms, np.nan)
                mh = gh["mh2"][:][o] if "mh2" in gh else np.full_like(ms, np.nan)
                S, D, X = agn_states(ms, mg, mbh, fedd)                    # Part 0's gates, verbatim
                M = {k: np.nan_to_num(v) > 0.5 for k, v in
                     (("xray", S["xray"]), ("choked", S["choked"]), ("rad", S["radiative"]), ("dor", S["dormant"]),
                      ("jet", D["jet"]), ("gate", D["gate"]))}
                kin = _kinalign15(gh, KIN.get(a), o)
                _nan2 = np.full_like(ms, np.nan)
                _kg, _rg, _rs2 = (kin.get(c, _nan2) for c in ("kappa_gas", "r_gas", "r_star"))
                dtw = np.gradient(t)
                for j, gid in enumerate(np.asarray(gh["gal_ids"][:]).astype(np.int64)):
                    key = (a, int(gid))
                    if key not in want:
                        continue
                    r, ex = want[key], X["exist"][:, j]
                    if ex.sum() < 3:
                        continue
                    m = {k: v[:, j] & ex for k, v in M.items()}
                    ka = int(np.flatnonzero(ex)[-1])
                    with np.errstate(all="ignore"):
                        den = np.where(ex, ms[:, j], np.nan)
                        v = {"lfd":   np.log10(np.where(ex, np.clip(md[:, j] / den, 10 ** P9_FLOOR, None), np.nan)),
                             "lfh2":  np.log10(np.where((mh[:, j] > 0) & ex, HE * mh[:, j] / den, np.nan)),
                             "lfgas": np.log10(np.where((mg[:, j] > 0) & ex, mg[:, j] / den, np.nan)),
                             "lssfr": np.log10(np.where(ex, np.clip(np.maximum(sfr[:, j], 0.0) / den, SSFR_FLOOR, None), np.nan)),
                             "lms":   np.log10(np.where(ex, ms[:, j], np.nan)),
                             "kappa_gas": np.where(ex, _kg[:, j], np.nan),
                             "comp":  -np.log10(np.where((_rg[:, j] > 0) & (_rs2[:, j] > 0) & ex, _rg[:, j] / _rs2[:, j], np.nan)),
                             "lr_gas": np.log10(np.where((_rg[:, j] > 0) & ex, _rg[:, j], np.nan)),
                             # the surface densities, Part 3's convention: half the mass inside the half-mass radius, the H2
                             # spread over the COLD-GAS radius because caesar carries no H2 radius
                             "lsig_gas": np.log10(np.where((_rg[:, j] > 0) & (mg[:, j] > 0) & ex,
                                                           0.5 * mg[:, j] / (np.pi * _rg[:, j] ** 2), np.nan)),
                             "lsig_h2": np.log10(np.where((_rg[:, j] > 0) & (mh[:, j] > 0) & ex,
                                                          0.5 * HE * mh[:, j] / (np.pi * _rg[:, j] ** 2), np.nan))}
                    v["age"] = _agetrk15(t, ms[:, j], sfr[:, j], float(r.get("age", np.nan)), ka)
                    # ── the three epochs ──
                    t_sft = float(r.get("t_sft", np.nan))
                    t_x1 = float(t[m["xray"]][0]) if m["xray"].any() else np.nan
                    _jt = t[m["jet"]]
                    if P15_JETDEF == "sustained" and m["jet"].any():
                        t_jet = np.nan
                        for _c in _jt:                                     # the first activation that stays on
                            _w = ex & (t >= _c) & (t <= _c + P15_SUST)
                            if _w.sum() and float(np.sum(dtw[_w & m["jet"]]) / max(np.sum(dtw[_w]), 1e-9)) >= 0.5:
                                t_jet = float(_c); break
                        if not np.isfinite(t_jet) and len(_jt):
                            t_jet = float(_jt[0])
                    else:
                        t_jet = float(_jt[0]) if len(_jt) else np.nan
                    EP = {"jet": t_jet, "x1": t_x1, "sft": t_sft}
                    t0 = EP.get(P15_EPOCH, np.nan)
                    rec = dict(box=box, snap=a, gal_id=int(gid), t_jet=t_jet, t_x1=t_x1, t_sft=t_sft,
                               t_anchor=float(t[ka]), t0=t0, t_span=float(t[ka] - t[ex][0]))
                    for _e, _te in EP.items():                             # the STATE at each epoch, and the SPIKE it sits on
                        for q in ("lsig_h2", "lsig_gas", "comp", "kappa_gas", "lfd", "lfh2", "lfgas", "lssfr", "lms"):
                            _val = _at15(t, v[q], _te)
                            rec[f"{q}_{_e}"] = _val
                            if _e == P15_EPOCH:
                                _b0 = _win15(t, v[q], _te - P15_BASE[1], _te - P15_BASE[0])
                                rec[f"d{q}"] = _val - _b0 if np.isfinite(_val) and np.isfinite(_b0) else np.nan
                    # ── the BURST: the mass formed in the last P15_BURST Gyr before the truncation, from the SFR itself ──
                    if np.isfinite(t_sft):
                        _bw = ex & (t >= t_sft - P15_BURST) & (t <= t_sft)
                        _mform = float(np.sum(np.nan_to_num(np.maximum(sfr[_bw, j], 0.0)) * dtw[_bw]) * 1e9) if _bw.any() else np.nan
                        _msft = _at15(t, np.where(ex, ms[:, j], np.nan), t_sft)
                        rec["f_burst"] = _mform / _msft if (np.isfinite(_mform) and np.isfinite(_msft) and _msft > 0) else np.nan
                        _pk = ex & (t >= t_sft - P15_BURST) & (t <= t_sft)
                        _bl = ex & (t >= t_sft - 3.0) & (t < t_sft - P15_BURST)
                        rec["ssfr_pk"] = float(np.nanmax(v["lssfr"][_pk])) if _pk.any() and np.isfinite(v["lssfr"][_pk]).any() else np.nan
                        rec["b_ssfr"] = (rec["ssfr_pk"] - float(np.nanmedian(v["lssfr"][_bl]))) if (_bl.any() and np.isfinite(rec["ssfr_pk"])) else np.nan
                        # WHERE THE STARBURST PEAKED — the critical point the plane marks, the maximum of the sSFR over
                        # the P15_PKW Gyr before the truncation (the peak of the last episode, not of the whole history)
                        _pw15 = ex & (t >= t_sft - P15_PKW) & (t <= t_sft) & np.isfinite(v["lssfr"])
                        _ip15 = int(np.flatnonzero(_pw15)[int(np.argmax(v["lssfr"][_pw15]))]) if _pw15.any() else -1
                        rec["t_pk"] = float(t[_ip15]) if _ip15 >= 0 else np.nan
                        rec["lssfr_pkw"] = float(v["lssfr"][_ip15]) if _ip15 >= 0 else np.nan
                    else:
                        rec.update(f_burst=np.nan, ssfr_pk=np.nan, b_ssfr=np.nan, t_pk=np.nan, lssfr_pkw=np.nan)
                    # ── THE STIRRING: what the X-ray onset does to the rotation and the extent in the P15_STIR Gyr after it ──
                    for q in ("kappa_gas", "comp", "lfd", "lsig_h2", "lfgas"):
                        _a0, _a1 = _at15(t, v[q], t_x1), _at15(t, v[q], t_x1 + P15_STIR)
                        rec[f"stir_{q}"] = _a1 - _a0 if (np.isfinite(_a0) and np.isfinite(_a1)) else np.nan
                    # ── the pre-epoch and post-SFT time budgets: the jet leg (Part 14's axis) and what fires afterwards ──
                    pre = ex & (t <= t_sft) if np.isfinite(t_sft) else np.zeros(len(t), bool)
                    post = ex & (t > t_sft) if np.isfinite(t_sft) else np.zeros(len(t), bool)
                    w = lambda k: float(np.sum(dtw[k]))
                    rec["t_pre"], rec["t_post"] = w(pre), w(post)
                    rec["t_jet_pre"] = w(pre & m["jet"])                   # Part 14's cause, on the same rows
                    rec["f_jet_pre"] = rec["t_jet_pre"] / rec["t_pre"] if rec["t_pre"] > 0 else np.nan
                    for _s, _k in (("jet", "jet"), ("chk", "choked"), ("xr", "xray"), ("rad", "rad"), ("dor", "dor")):
                        rec[f"f_{_s}_post"] = w(post & m[_k]) / rec["t_post"] if rec["t_post"] > 0 else np.nan
                    aft = ex & (t >= t_x1) if np.isfinite(t_x1) else np.zeros(len(t), bool)
                    _xa, _ta = m["xray"][aft], t[aft]
                    rec["duty_post"] = (w(aft & m["xray"]) / w(aft)) if w(aft) > 0 else np.nan
                    if _xa.any():
                        _edges = np.concatenate([[_ta[0]], _ta[_xa], [_ta[-1]]])
                        rec["gap_max"] = float(np.max(np.diff(_edges))) if len(_edges) > 1 else 0.0
                    else:
                        rec["gap_max"] = float(_ta[-1] - _ta[0]) if len(_ta) > 1 else np.nan
                    # ── the gas budget after the truncation: the "keeps accreting" half of the story ──
                    _yp = v["lfgas"][post]; _yp = _yp[np.isfinite(_yp)]
                    _pr = v["lfgas"][pre]; _pr = _pr[np.isfinite(_pr)]
                    rec["dlfgas_post"] = float(_yp[-1] - _pr[-1]) if (len(_yp) and len(_pr)) else np.nan
                    rec["regas"] = float(_yp[-1] - np.min(_yp)) if len(_yp) else np.nan
                    # ── the anchor, read off the branch itself (its last step IS Part 0's catalogue row) ──
                    for q in P15_VKEYS:
                        rec[f"{q}_anc"] = float(v[q][ka]) if np.isfinite(v[q][ka]) else np.nan
                        TR[q].append(to_grid(t - t0, v[q]) if np.isfinite(t0) else np.full(len(TG), np.nan))
                    rec["dlfgas_late"] = rec["lfgas_anc"] - _at15(t, v["lfgas"], float(t[ka]) - 1.0)
                    # ── THE DECAY from the truncation to the anchor: the fit per galaxy, and the track on its own clock
                    _dt15 = (t - t_sft) if np.isfinite(t_sft) else np.full(len(t), np.nan)
                    _sg15 = ex & np.isfinite(_dt15) & (_dt15 >= -1e-9)
                    for q in P15_DQ:
                        for _k15, _v15 in _decay15(_dt15[_sg15], v[q][_sg15], _DFL15.get(q)).items():
                            rec[f"{q}_{_k15}"] = _v15
                        _yq15 = v[q] if _DFL15.get(q) is None else np.where(v[q] <= _DFL15[q] + P15_DFLOOR, np.nan, v[q])
                        TR[f"dec_{q}"].append(to_grid(_dt15, _yq15, TDG))   # drawn = fitted: the floor is censored in both
                    for c in ("z", "age", "log_mstar", "log_mbh", "lfd", "lfh2", "lfgas", "lssfr", "kcls", "rot", "pop",
                              "central", "kappa", "ngas", "t_qt", "e_x", "f_x", "f_x_post", "lfd_net", "lfd_reb"):
                        if c in r:
                            rec[c] = r[c]
                    rows.append(rec)
        return pd.DataFrame(rows), {q: np.asarray(TR[q], float).reshape(len(rows), len(_GRID15[q])) for q in _TK15}

    _D15, _T15 = [], {q: [] for q in _TK15}
    for _b15 in _BX15:
        F15 = H[_b15]["F"]
        _w15 = {(int(_r["snap"]), int(_r["gal_id"])): _r for _, _r in F15.iterrows()}
        _d15, _t15 = _pass15(_b15, _w15)
        if not len(_d15):
            continue
        _D15.append(_d15)
        for q in _TK15:
            _T15[q].append(_t15[q])
        _cov = lambda c: int(np.isfinite(_d15[c].to_numpy(float)).sum())
        print(f"  [{_b15}] {len(_d15)} branches:  {_cov('t_jet')} reach the jet state, {_cov('t_x1')} fire the channel, "
              f"{_cov('t_sft')} have an SFT;  the axis {P15_CAUSE} at {P15_ENAM[P15_EPOCH]} is measured for "
              f"{_cov(P15_CAUSE + '_' + P15_EPOCH)} of them"
              + (f"  [{_cov('lsig_h2_' + P15_EPOCH)} carry R50(gas): the kinematic sidecar]" if "sig" in P15_CAUSE else ""))
    D15 = pd.concat(_D15, ignore_index=True) if _D15 else pd.DataFrame()
    T15 = {q: np.vstack(_T15[q]) for q in _TK15} if _D15 else {}

if not len(D15):
    print("Part 15: no branch carried both a jet activation and a histories chain — nothing to draw.")
else:
    _CA15 = f"{P15_CAUSE}_{P15_EPOCH}"
    # the surface densities need Part 3's kinematic sidecar (they are a mass over a radius): where it was never built the
    # axis is empty, and the part falls back on the MASS FRACTION of the same phase rather than dividing on nothing
    _CAFB = {"lsig_h2": "lfh2", "lsig_gas": "lfgas", "comp": "lfgas", "kappa_gas": "lfh2", "lr_gas": "lfgas"}
    if (P15_FALLBACK and np.isfinite(D15[_CA15].to_numpy(float)).mean() < P15_MINCOV
            and f"{_CAFB.get(P15_CAUSE, '')}_{P15_EPOCH}" in D15.columns):
        print(f"  !! {P15_CAUSE} at {P15_ENAM[P15_EPOCH]} is measured for only "
              f"{100 * np.isfinite(D15[_CA15].to_numpy(float)).mean():.0f} % of the branches — Part 3's kinematic sidecar "
              f"(P3_KIN_BUILD / P3_CATDIR) carries the radii it needs.\n"
              f"     THE AXIS FALLS BACK to {_CAFB[P15_CAUSE]}, the mass fraction of the same phase. Set P15_FALLBACK = "
              "False to stop instead, or build the sidecar and re-run Part 3.")
        P15_CAUSE = _CAFB[P15_CAUSE]
        _CA15 = f"{P15_CAUSE}_{P15_EPOCH}"
    _CLAB15 = f"{P15_QLAB.get(P15_CAUSE, P15_CAUSE)} at {P15_ELAB[P15_EPOCH]}"

    def _prho15(x, y, ctrl):
        """PARTIAL Spearman: both axes and the controls rank-transformed, the controls regressed out by least squares and
        the residuals correlated; p from the t approximation with k controls. A constant control is dropped, not fatal."""
        cols = [np.asarray(x, float), np.asarray(y, float)] + [np.asarray(c, float) for c in ctrl]
        ok = np.all([np.isfinite(c) for c in cols], axis=0)
        if ok.sum() < len(ctrl) + 8:
            return np.nan, np.nan, int(ok.sum())
        R = np.array([_st15.rankdata(c[ok]) for c in cols], float).T
        sd = R.std(0)
        if not (sd[:2] > 0).all():
            return np.nan, np.nan, int(ok.sum())
        R, sd = R[:, np.r_[True, True, sd[2:] > 0]], sd[np.r_[True, True, sd[2:] > 0]]
        k = R.shape[1] - 2
        R = (R - R.mean(0)) / sd
        A = np.column_stack([np.ones(len(R)), R[:, 2:]])
        res = [R[:, i] - A @ np.linalg.lstsq(A, R[:, i], rcond=None)[0] for i in (0, 1)]
        if np.std(res[0]) == 0 or np.std(res[1]) == 0:
            return np.nan, np.nan, int(ok.sum())
        r_, n_ = float(np.corrcoef(res[0], res[1])[0, 1]), int(ok.sum())
        dof = n_ - 2 - k
        return r_, (float(2 * _st15.t.sf(abs(r_) * np.sqrt(dof / max(1 - r_ * r_, 1e-12)), dof)) if dof > 0 else np.nan), n_

    def _rho15(x, y):
        x, y = np.asarray(x, float), np.asarray(y, float)
        ok = np.isfinite(x) & np.isfinite(y)
        if ok.sum() < 6 or np.nanstd(x[ok]) == 0 or np.nanstd(y[ok]) == 0:
            return np.nan, np.nan, int(ok.sum())
        r_ = spearmanr(x[ok], y[ok])
        return float(r_.statistic), float(r_.pvalue), int(ok.sum())

    def _crank15(g, col, ctrl):
        """The axis's rank in [0, 1] INSIDE the cartesian cells of `ctrl` (P15_MQ quantile bins each), so that a division
        on it cannot be a division on the controls in disguise. An empty `ctrl` gives the plain rank."""
        v = g[col].to_numpy(float) if col in g.columns else np.full(len(g), np.nan)
        r, cells, ok = np.full(len(v), np.nan), np.zeros(len(v), np.int64), np.isfinite(v)
        for c in ctrl:
            if c not in g.columns:
                continue
            x = g[c].to_numpy(float)
            e = np.unique(np.nanquantile(x, np.linspace(0, 1, P15_MQ + 1))) if np.isfinite(x).any() else np.array([0.0, 1.0])
            lab = np.where(np.isfinite(x), np.clip(np.digitize(x, e[1:-1]), 0, max(len(e) - 2, 0)), -1)
            cells = cells * (P15_MQ + 1) + (lab + 1)
            ok &= np.isfinite(x)
        for c in np.unique(cells[ok]):
            m = ok & (cells == c)
            if m.sum() > 1:
                r[m] = (_st15.rankdata(v[m]) - 0.5) / m.sum()
        return r

    _EPCOL = {"jet": "t_jet", "x1": "t_x1", "sft": "t_sft"}[P15_EPOCH]
    D15["crank"] = np.concatenate([_crank15(D15[D15.box == b], _CA15, P15_MCTRL) for b in _BX15])
    D15["grp"] = np.where(D15.crank >= 1 - P15_Q, "hi", np.where(D15.crank <= P15_Q, "lo", "mid"))
    D15["crank_z"] = np.concatenate([_crank15(D15[D15.box == b], _CA15, list(P15_MCTRL) + [_EPCOL]) for b in _BX15])
    D15["grp_z"] = np.where(D15.crank_z >= 1 - P15_Q, "hi", np.where(D15.crank_z <= P15_Q, "lo", "mid"))
    D15["dusty"] = D15["lfd_anc"].to_numpy(float) >= DUSTY_CUT

    # ── THE CURRENT OUTFLOW, through each galaxy's OWN R50(stars) — the user's radius, in place of the fixed 10 kpc
    # shell Part 13 uses. One reduced particle file per ANCHOR (not the whole ladder), measured once and cached.
    _OLAB15 = {"lsout": r"$\log\, \dot M_{\rm out}/M_\star$ at $R_{50}(\star)$  [yr$^{-1}$]",
               "lsout_c": r"$\log\, \dot M_{\rm out}(T<10^5\,$K$)/M_\star$ at $R_{50}(\star)$  [yr$^{-1}$]",
               "lsout_d": r"$\log\, \dot M_{\rm out}({\rm dust})/M_\star$ at $R_{50}(\star)$  [yr$^{-1}$]",
               "lmdot_out": r"$\log\, \dot M_{\rm out}$ at $R_{50}(\star)$  [$M_\odot$ yr$^{-1}$]",
               "v_out90": r"$v_r$(90th pct) of the outflow at $R_{50}(\star)$  [km s$^{-1}$]",
               "f_out": r"the outward mass share in the $R_{50}(\star)$ shell", "r50s": r"$R_{50}(\star)$  [kpc]"}
    if P15_OUTFLOW:
        _OM15 = []
        for b in _BX15:
            _gb15 = D15[D15.box == b]
            try:
                _Z15, _msg15 = _outpass15(b, sorted({(int(s_), int(i_)) for s_, i_ in
                                                     zip(_gb15["snap"], _gb15["gal_id"])}))
            except Exception as _e15:
                _Z15, _msg15 = pd.DataFrame(), f"[{b}] the outflow pass failed: {type(_e15).__name__}: {_e15}"
            print("  " + _msg15)
            if len(_Z15):
                _Z15 = _Z15.copy(); _Z15["box"] = b
                _OM15.append(_Z15)
        if _OM15:
            _ZZ15 = pd.concat(_OM15, ignore_index=True)
            _ix15 = {(str(b_), int(s_), int(x_)): i for i, (b_, s_, x_)
                     in enumerate(zip(_ZZ15["box"], _ZZ15["snap"], _ZZ15["gx"]))}
            _row15 = np.array([_ix15.get((str(b_), int(s_), int(i_)), -1)
                               for b_, s_, i_ in zip(D15["box"], D15["snap"], D15["gal_id"])])
            for _c15 in ("mdot_out_g", "mdot_out_gc", "mdot_out_d", "mdot_out_dc", "mdot_in_g", "mdot_in_gc",
                         "v_out90", "f_out", "r50s", "n_sh"):
                if _c15 in _ZZ15.columns:
                    _v15 = _ZZ15[_c15].to_numpy(float)[np.clip(_row15, 0, None)]
                    D15[_c15] = np.where(_row15 >= 0, _v15, np.nan)
            _ms15 = (10 ** D15["log_mstar"].to_numpy(float) if "log_mstar" in D15.columns
                     else 10 ** D15["lms_anc"].to_numpy(float))
            with np.errstate(all="ignore"):
                _lg15 = lambda a: np.log10(np.where(np.asarray(a, float) > 0, np.asarray(a, float), np.nan))
                if "mdot_out_g" in D15.columns:
                    D15["lmdot_out"] = _lg15(D15["mdot_out_g"])
                    D15["lsout"] = _lg15(D15["mdot_out_g"].to_numpy(float) / _ms15)
                if "mdot_out_gc" in D15.columns:
                    D15["lsout_c"] = _lg15(D15["mdot_out_gc"].to_numpy(float) / _ms15)
                if "mdot_out_d" in D15.columns:
                    D15["lsout_d"] = _lg15(D15["mdot_out_d"].to_numpy(float) / _ms15)
            print("  the outflow axis " + P15_OQ + " is measured for "
                  + ", ".join(f"{b}: {int(np.isfinite(D15[D15.box == b][P15_OQ].to_numpy(float)).sum())}/"
                              f"{int((D15.box == b).sum())}" for b in _BX15 if P15_OQ in D15.columns)
                  + f"   (the shell |r - R50*| < {P15_ORW:g} R50*, the velocity in the galaxy's rest frame)")

    ROW15 = []
    print("=" * 132)
    print(f"Part 15 — THE STATE AT {P15_ENAM[P15_EPOCH].upper()}.  the division: {P15_CAUSE} at {P15_ELAB[P15_EPOCH]}, "
          f"{'above / below the box median' if P15_Q == 0.5 else f'the top / bottom {100 * P15_Q:.0f} %'}"
          + (f", ranked inside cells of {' x '.join(P15_MCTRL)}" if P15_MCTRL else ", on the raw axis"))
    print("  the mechanism under test:  a dense central ISM when the channel opens  ->  the X-ray kick has something to "
          "push  ->  the ISM is stirred and the dust is destroyed")
    print("=" * 132)

    # ══ (A) IS THIS THE RIGHT AXIS? every candidate scored against the same outcome ══════════════════════════════════
    _CAND = [(f"lsig_h2_{e}", f"log Sigma_H2 at {P15_ENAM[e]}") for e in ("jet", "x1", "sft")] + \
            [(f"lfh2_{P15_EPOCH}", "log f_H2 at the epoch"), (f"lssfr_{P15_EPOCH}", "log sSFR at the epoch"),
             (f"comp_{P15_EPOCH}", "the compactness at the epoch"), (f"kappa_gas_{P15_EPOCH}", "kappa_rot at the epoch"),
             (f"d{P15_CAUSE}", "THE SPIKE the axis sits on"), ("f_burst", "the burst: M formed in the last Gyr"),
             ("b_ssfr", "the burst: the sSFR peak over baseline"), ("t_jet_pre", "PART 14's axis: the jet leg [Gyr]")]
    _CAND = ([(_CA15, f"THE AXIS: {P15_CAUSE} at {P15_ENAM[P15_EPOCH]}")] if _CA15 not in [q for q, _ in _CAND] else []) + _CAND
    print("\n" + "-" * 132)
    print("(a) THE AXIS — every candidate against the SAME outcome: the dust at the anchor (AUC of the axis between the")
    print(f"    dusty and the dust-poor QGs, log M_dust/M* {DUSTY_CUT:g}), then the rank correlation with the dust itself")
    print(f"  {'box':10s} {'candidate axis':38s} {'N':>5s} {'AUC(dusty)':>11s} {'rho(lfd)':>9s} "
          f"{'rho | M*,M_BH':>13s} {'rho | + epoch':>13s}")
    for b in _BX15:
        g = D15[D15.box == b]
        for q, lab in _CAND:
            if q not in g.columns or np.isfinite(g[q].to_numpy(float)).sum() < 8:
                continue
            au, p_, _n1, _n0 = mw_auc(g[q].to_numpy(float), g["dusty"].to_numpy(bool))
            r0, p0, n0 = _rho15(g[q].to_numpy(float), g["lfd_anc"].to_numpy(float))
            r1, p1, _ = _prho15(g[q].to_numpy(float), g["lfd_anc"].to_numpy(float),
                                [g.log_mstar.to_numpy(float), g.log_mbh.to_numpy(float)])
            r2, p2, _ = _prho15(g[q].to_numpy(float), g["lfd_anc"].to_numpy(float),
                                [g.log_mstar.to_numpy(float), g.log_mbh.to_numpy(float), g[_EPCOL].to_numpy(float)])
            _st = lambda r_, p_: f"{r_:+.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}"
            print(f"  {b:10s} {lab:38s} {n0:5d} {au:8.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}   "
                  f"{_st(r0, p0):>9s} {_st(r1, p1):>13s} {_st(r2, p2):>13s}")
            ROW15.append(dict(box=b, block="axis", quantity=q, auc=au, p=p_, med=r0, lo=r1, hi=r2, n_gal=n0))
    print("  (* = p < 0.05. The last column holds the epoch fixed as well: a candidate that survives it is a STATE, one")
    print("   that does not is the epoch — Part 14's jet leg carries a +1.5 to +2.8 Gyr difference in t_SFT by itself.")
    print(f"   The AUC is undefined wherever one side of the dust label is empty, which is the normal case in {NOX_BOX}:")
    print("   with the channel compiled out essentially every branch ends dusty, so there read the rho columns.)")

    # ══ (B) the two groups: what they are, and what the controls do ═════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print(f"(b) THE TWO BRANCHES — {P15_GLONG['hi']}  /  {P15_GLONG['lo']}")
    print(f"  {'box':10s} {'quantity':16s} {'DENSE':>9s} {'DEPLETED':>9s} {'D':>8s} {'AUC':>6s} {'p':>9s}   what it is")
    _LINK15 = [(f"{P15_CAUSE}_{P15_EPOCH}", "THE AXIS"), (f"d{P15_CAUSE}", "the spike it sits on"),
               (f"lssfr_{P15_EPOCH}", "the sSFR at the epoch — the same axis?"),
               ("f_burst", "THE BURST: M*(1 Gyr)/M*"), ("b_ssfr", "the sSFR peak over its own baseline"),
               ("age", "THE AGE at the anchor [Gyr]"), ("lfd_anc", "THE DUST at the anchor"),
               ("kappa_gas_anc", "the rotation at the anchor"), ("comp_anc", "the extent at the anchor"),
               ("lfh2_anc", "the H2 at the anchor"), ("lfgas_anc", "the gas at the anchor"),
               ("stir_kappa_gas", "THE STIRRING: d kappa across t_x1"), ("stir_comp", "d compactness across t_x1"),
               ("stir_lfd", "d dust across t_x1"), ("duty_post", "the X-ray duty after the gate"),
               ("gap_max", "the longest firing-free gap [Gyr]"), ("f_jet_post", "the jet AFTER the SFT"),
               ("f_dor_post", "dormant after the SFT"), ("dlfgas_post", "the gas budget after the SFT"),
               ("regas", "the re-accretion after the minimum"), ("t_jet_pre", "Part 14's jet leg [Gyr]"),
               (P15_OQ, "THE OUTFLOW at R50* now"), ("v_out90", "its speed [km/s]"),
               ("lfd_s_early", "THE DECAY: the dust slope, the first Gyr"), ("lfd_s_late", "... and the last one"),
               ("lfd_tau2", "the late e-folding of the dust [Gyr]"),
               ("log_mstar", "CONTROL"), ("log_mbh", "CONTROL"), ("t_sft", "CONTROL: the epoch"),
               (_EPCOL, "CONTROL: the epoch of the split")]
    for b in _BX15:
        g = D15[D15.box == b]
        L, S = g[g.grp == "hi"], g[g.grp == "lo"]
        print(f"  --- {BOX_LABEL[b]}   {len(L)} dense / {len(S)} pre-depleted"
              + ("   [jets ON, X-ray heating OFF: the counterfactual]" if b == NOX_BOX else ""))
        for q, lab in _LINK15:
            if q not in g.columns:
                continue
            a_, b_ = L[q].to_numpy(float), S[q].to_numpy(float)
            if np.isfinite(a_).sum() < 4 or np.isfinite(b_).sum() < 4:
                continue
            au, p_, _na, _nb = mw_auc(np.concatenate([a_, b_]),
                                      np.concatenate([np.ones(len(a_), bool), np.zeros(len(b_), bool)]))
            print(f"  {b:10s} {q:16s} {np.nanmedian(a_):9.2f} {np.nanmedian(b_):9.2f} "
                  f"{np.nanmedian(a_) - np.nanmedian(b_):+8.2f} {au:6.2f} {p_:9.2e}   {lab}")
            ROW15.append(dict(box=b, block="link", quantity=q, med=float(np.nanmedian(a_)), lo=float(np.nanmedian(b_)),
                              auc=au, p=p_, n_gal=len(a_) + len(b_)))
    print("  (the post-starburst claim needs f_burst and b_ssfr HIGH and the anchor age LOW in the dense branch; the")
    print("   stirring claim needs d kappa across t_x1 more NEGATIVE there, and that difference to vanish in cis50nox.)")
    if P15_ZCTRL:
        print("\n  the same contrasts with the groups re-formed inside cells of the epoch as well (t of the split):")
        print(f"  {'box':10s} " + " ".join(f"{q:>14s}" for q in ("lfd_anc", "age", "kappa_gas_anc", "stir_kappa_gas", "f_burst")))
        for b in _BX15:
            g = D15[D15.box == b]
            L, S = g[g.grp_z == "hi"], g[g.grp_z == "lo"]
            _o = []
            for q in ("lfd_anc", "age", "kappa_gas_anc", "stir_kappa_gas", "f_burst"):
                if q not in g.columns or not len(L) or not len(S):
                    _o.append(f"{'-':>14s}"); continue
                a_, b_ = L[q].to_numpy(float), S[q].to_numpy(float)
                au, p_, _na, _nb = mw_auc(np.concatenate([a_, b_]),
                                          np.concatenate([np.ones(len(a_), bool), np.zeros(len(b_), bool)]))
                _d = np.nanmedian(a_) - np.nanmedian(b_)
                _o.append(f"{_d:+7.2f} {au:5.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}")
                ROW15.append(dict(box=b, block="link_zctrl", quantity=q, med=float(np.nanmedian(a_)),
                                  lo=float(np.nanmedian(b_)), auc=au, p=p_, n_gal=len(a_) + len(b_)))
            print(f"  {b:10s} " + " ".join(f"{v:>14s}" for v in _o))
        print("  (D and AUC of dense minus pre-depleted. What survives here is not the epoch.)")

    def _leg15(ax, **kw):
        kw.setdefault("frameon", False); kw.setdefault("fontsize", 7.5)
        h, l = ax.get_legend_handles_labels()
        if h:
            ax.legend(h, l, **kw)

    _GM15 = {b: {k: ((D15.box == b) & (D15.grp == k)).to_numpy(bool) for k in P15_GK} for b in _BX15}

    # ══ (C) FIG 1 — THE CLOCK, with the jet activation as the origin ═════════════════════════════════════════════════
    _TRK15 = [q for q in P15_TRK if q in T15]
    fig, AX = plt.subplots(len(_TRK15), len(_BX15), figsize=(4.5 * len(_BX15), 2.5 * len(_TRK15)), squeeze=False)
    for jb, b in enumerate(_BX15):
        g = D15[D15.box == b]
        for iq, q in enumerate(_TRK15):
            ax = AX[iq, jb]
            for k in P15_GK:
                m = _GM15[b][k]
                if not m.sum():
                    continue
                A = T15[q][m]
                if not np.isfinite(A).any():
                    continue
                cen, lo, hi, n_t = med_band(A, nmin=P15_NMIN, nboot=P15_NBOOT)
                ax.fill_between(TG, lo, hi, color=P15_GCOL[k], alpha=0.20, lw=0)
                ax.plot(TG, cen, color=P15_GCOL[k], lw=2.5, label=f"{P15_GLAB[k]}  ({int(m.sum())})")
                for _x, _c, _l, _h, _n in zip(TG, cen, lo, hi, n_t):
                    ROW15.append(dict(box=b, block=f"clock_{q}", quantity=k, x=float(_x), med=float(_c), lo=float(_l),
                                      hi=float(_h), n_gal=int(_n)))
                _sub = g[g.grp == k]
                for _c15, _ls in ((_sub.t_sft.to_numpy(float) - _sub.t0.to_numpy(float), "--"),
                                  (_sub.t_x1.to_numpy(float) - _sub.t0.to_numpy(float), ":")):
                    _c15 = _c15[np.isfinite(_c15)]
                    if len(_c15) >= 3:
                        ax.axvline(float(np.median(_c15)), color=P15_GCOL[k], lw=1.2, ls=_ls, alpha=0.85, zorder=1)
            ax.axvline(0.0, color="k", lw=1.4)
            ax.grid(alpha=0.2, lw=0.5)
            if iq == 0:
                ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
                if jb == 0:
                    _leg15(ax, loc="best")
            if jb == 0:
                ax.set_ylabel(P15_QLAB.get(q, q), fontsize=9)
            if iq == len(_TRK15) - 1:
                ax.set_xlabel(rf"$t - {P15_ELAB[P15_EPOCH][1:-1]}$  [Gyr]", fontsize=9.5)
            if q == "lfgas":
                ax.axhline(np.log10(XRAY_FGAS_MAX), color="#2ca02c", lw=1.1, ls=":")
                if jb == 0:
                    ax.annotate(rf"the X-ray gate $f_{{\rm gas}} < {XRAY_FGAS_MAX:g}$", xy=(0.02, np.log10(XRAY_FGAS_MAX)),
                                xycoords=("axes fraction", "data"), xytext=(0, 3), textcoords="offset points",
                                fontsize=7, color="#2ca02c")
            if q == "lfd":
                ax.axhline(DUSTY_CUT, color="0.35", lw=1.0, ls="--")
    fig.suptitle(f"The two branches on the clock of {P15_ENAM[P15_EPOCH]}   —   split on {P15_QLAB.get(P15_CAUSE, P15_CAUSE)}"
                 f" at {P15_ELAB[P15_EPOCH]}" + (f", inside cells of {' x '.join(P15_MCTRL)}" if P15_MCTRL else "")
                 + "   ( -- the truncation,  ..  the X-ray onset, each group's own median )", y=0.999, fontsize=12,
                 fontweight="bold")
    fig.tight_layout(rect=(0, 0, 1, 0.978)); paper_save(fig, "kappa_jetstate_clocks"); plt.show()

    # ══ (D) FIG 2 — THE OBSERVABLE PLANE: f_dust against the age — THE TRACKS AND THEIR CRITICAL POINTS ═════════════
    # the anchor cloud is deliberately NOT drawn: what is being read here is the PATH each branch takes through the
    # plane and where its critical points sit on it (the sSFR peak, the jet activation, the truncation, the anchor).
    _PX, _PY = P15_PLANE
    _MK15 = [plt.Line2D([], [], ls="none", marker=_m, ms=_s, mfc="w", mec="0.25", mew=1.6, label=_l)
             for _m, _s, _l in (("^", 8, "the sSFR peak"), ("o", 7, P15_ENAM[P15_EPOCH]), ("s", 7, "the truncation"))]
    _MK15.append(plt.Line2D([], [], ls="none", marker="*", ms=13, color="0.25", label="the median anchor"))
    fig2, AX2 = plt.subplots(1, len(_BX15), figsize=(4.6 * len(_BX15), 4.3), squeeze=False)
    for jb, b in enumerate(_BX15):
        ax = AX2[0, jb]
        g = D15[D15.box == b]
        for k in P15_GK:
            m = _GM15[b][k]
            if not m.sum():
                continue
            _gx = g[g.grp == k]
            X_, Y_ = T15[_PX][m], T15[_PY][m]
            n_t = np.minimum(np.isfinite(X_).sum(0), np.isfinite(Y_).sum(0))
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", RuntimeWarning)
                xc, yc = np.nanmedian(X_, axis=0), np.nanmedian(Y_, axis=0)
            ok = (n_t >= P15_NMIN) & np.isfinite(xc) & np.isfinite(yc)
            if ok.sum() < 2:
                continue
            _xsft = np.nanmedian(_gx.t_sft.to_numpy(float) - _gx.t0.to_numpy(float))
            _pre = ok & (TG <= (_xsft if np.isfinite(_xsft) else 0.0))     # the star-forming leg: the age reconstruction
            _post = ok & (TG >= (_xsft if np.isfinite(_xsft) else 0.0))    # is only exact once nothing forms
            if _pre.sum() > 1:
                ax.plot(xc[_pre], yc[_pre], color=P15_GCOL[k], lw=1.4, ls=":", alpha=0.75, zorder=4)
            if _post.sum() > 1:
                ax.plot(xc[_post], yc[_post], color=P15_GCOL[k], lw=3.0, zorder=5,
                        label=f"{P15_GLAB[k]}  ({int(m.sum())})")
            _xpk = (float(np.nanmedian(_gx.t_pk.to_numpy(float) - _gx.t0.to_numpy(float)))
                    if "t_pk" in _gx.columns and np.isfinite(_gx.t_pk.to_numpy(float)).any() else np.nan)
            for _e15, _mk, _ms15 in ((_xpk, "^", 9), (0.0, "o", 8), (_xsft, "s", 8)):
                if np.isfinite(_e15) and TG[0] <= _e15 <= TG[-1]:
                    _i = int(np.argmin(np.abs(TG - _e15)))
                    if ok[_i]:
                        ax.plot([xc[_i]], [yc[_i]], _mk, ms=_ms15, mfc="w", mec=P15_GCOL[k], mew=2.0, zorder=6)
                        ROW15.append(dict(box=b, block="plane_mark", quantity=f"{k}_{_mk}", x=float(xc[_i]),
                                          med=float(yc[_i]), lo=float(_e15)))
            # the END of the track is where the clock grid runs out, NOT the anchor: the branches reach their anchors at
            # different times, so the star is the group's own median ANCHOR and the dashes are the leg in between
            _ia = int(np.flatnonzero(ok)[-1])
            _ax15 = float(np.nanmedian(_gx[_PX].to_numpy(float) if _PX in _gx.columns
                                       else _gx[f"{_PX}_anc"].to_numpy(float)))
            _ay15 = float(np.nanmedian(_gx[f"{_PY}_anc"].to_numpy(float)))
            if np.isfinite(_ax15) and np.isfinite(_ay15):
                ax.plot([xc[_ia], _ax15], [yc[_ia], _ay15], color=P15_GCOL[k], lw=1.4, ls=(0, (5, 3)), alpha=0.8, zorder=5)
                ax.plot([_ax15], [_ay15], "*", ms=18, color=P15_GCOL[k], mec="k", mew=0.6, zorder=7)
                ROW15.append(dict(box=b, block="plane_anchor", quantity=k, x=_ax15, med=_ay15, n_gal=int(m.sum())))
            else:
                ax.plot([xc[_ia]], [yc[_ia]], "*", ms=17, color=P15_GCOL[k], mec="k", mew=0.6, zorder=7)
            for _x15, _y15, _n15 in zip(xc[ok], yc[ok], n_t[ok]):
                ROW15.append(dict(box=b, block="plane", quantity=k, x=float(_x15), med=float(_y15), n_gal=int(_n15)))
        ax.axhline(DUSTY_CUT, color="0.35", lw=1.2, ls="--")
        _y015, _y115 = ax.get_ylim()                      # a clean band under the tracks, for the legend and the stars
        ax.set_ylim(_y015 - 0.28 * (_y115 - _y015), _y115)
        ax.set_xlabel(P15_QLAB.get(_PX, _PX), fontsize=9.5)
        ax.set_ylabel(P15_QLAB.get(_PY, _PY), fontsize=9.5)
        ax.set_title(BOX_LABEL[b], loc="left", fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.grid(alpha=0.2, lw=0.5)
        if jb == 0:
            _h15, _l15 = ax.get_legend_handles_labels()
            ax.legend(_h15 + _MK15, _l15 + [_x.get_label() for _x in _MK15], frameon=False, fontsize=7.5,
                      loc="lower left", ncol=2)
    fig2.suptitle("The path each branch takes through the observable plane, and where its critical points sit on it"
                  "\n(triangle = the sSFR peak, circle = " + P15_ENAM[P15_EPOCH] + ", square = the truncation, star = the"
                  " median anchor reached along the dashes; the dotted leg is still forming stars, where the"
                  " reconstructed age is a lower limit)", y=1.02, fontsize=11.5, fontweight="bold")
    fig2.tight_layout(); paper_save(fig2, "kappa_jetstate_dustage"); plt.show()

    # ══ (E) THE DECAY, FITTED — the same initial slope, and then a bend? ═════════════════════════════════════════════
    _DLAB15 = {"lfd": "the dust f_dust", "lssfr": "the sSFR", "lfh2": "the H2 f_H2", "lfgas": "the gas f_gas",
               "lsig_h2": "Sigma_H2", "kappa_gas": "kappa_rot(gas)"}
    _DQ15 = [q for q in P15_DQ if f"{q}_tau1" in D15.columns]
    print("\n" + "-" * 132)
    print(f"(c) THE DECAY FROM THE TRUNCATION TO THE ANCHOR — a double exponential per galaxy, y = A1 exp(-t/tau1) + "
          f"A2 exp(-t/tau2)")
    print(f"    fitted in log10 (tau1 = the initial e-folding, tau2 the late one, t_brk where the slow term takes over,")
    print(f"    f_slow its share at the truncation, BIC+ = the share of galaxies where the bend beats a single")
    print(f"    exponential), beside the MODEL-FREE slopes over the first and the last {P15_DWIN:g} Gyr [dex/Gyr]")
    print(f"  {'box':10s} {'quantity':16s} {'branch':9s} {'N':>4s} {'tau1':>7s} {'tau2':>7s} {'t_brk':>7s} "
          f"{'f_slow':>7s} {'BIC+':>6s} {'s_early':>8s} {'s_late':>8s} {'span':>6s}")
    for b in _BX15:
        for q in _DQ15:
            g = D15[D15.box == b]
            for k in P15_GK:
                s_ = g[g.grp == k]
                if not len(s_):
                    continue
                _f = lambda c: float(np.nanmedian(s_[f"{q}_{c}"].to_numpy(float))) if f"{q}_{c}" in s_.columns else np.nan
                _nf = int(np.isfinite(s_[f"{q}_tau1"].to_numpy(float)).sum())
                _db = s_[f"{q}_dbic"].to_numpy(float)
                _bp = float(np.mean(_db[np.isfinite(_db)] > 0)) if np.isfinite(_db).any() else np.nan
                print(f"  {b:10s} {_DLAB15.get(q, q):16s} {P15_GSH[k]:9s} {_nf:4d} {_f('tau1'):7.2f} {_f('tau2'):7.2f} "
                      f"{_f('tbrk'):7.2f} {_f('fslow'):7.2f} {100 * _bp:5.0f}% {_f('s_early'):+8.2f} "
                      f"{_f('s_late'):+8.2f} {_f('span'):6.2f}")
                for c in ("tau1", "tau2", "tbrk", "fslow", "s_early", "s_late", "ds", "span", "t_floor", "dbic"):
                    if f"{q}_{c}" in s_.columns:
                        ROW15.append(dict(box=b, block=f"decay_{q}", quantity=c, kcls=k, med=_f(c), n_gal=_nf))
            _o = []
            for c in ("s_early", "s_late", "ds", "tau1", "tau2"):
                a_, c_ = g[g.grp == "hi"][f"{q}_{c}"].to_numpy(float), g[g.grp == "lo"][f"{q}_{c}"].to_numpy(float)
                au, p_, _na, _nb = mw_auc(np.concatenate([a_, c_]),
                                          np.concatenate([np.ones(len(a_), bool), np.zeros(len(c_), bool)]))
                _o.append(f"{c} {au:.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}")
                ROW15.append(dict(box=b, block=f"decay_auc_{q}", quantity=c, auc=au, p=p_, n_gal=len(a_) + len(c_)))
            print(f"  {'':10s} {'':16s} {'AUC dense/depleted:':>19s} " + "  ".join(_o))
    print(f"  (the reading the fit is being asked for: s_early THE SAME in the two branches — AUC near 0.5 — and s_late")
    print( "   HIGHER (flatter) in the pre-depleted one, i.e. a long tau2 and a t_brk inside the sampled window there.")
    print(f"   A point within {P15_DFLOOR:g} dex of a floor is censored, so a branch that reaches zero dust ends its fit")
    print( "   there rather than being fitted a flat tail; t_floor in the CSV is when that happened.)")

    # ══ FIG 3 — the decay, drawn: the data, the fits, and the slopes they are made of ════════════════════════════════
    _WFIT15 = {}
    for q in _DQ15:
        _WFIT15[q] = np.array([_dcurve15(a_, b_, c_, d_, e_, TDG) for a_, b_, c_, d_, e_ in
                               zip(D15[f"{q}_lA1"], D15[f"{q}_tau1"], D15[f"{q}_lA2"], D15[f"{q}_tau2"],
                                   D15[f"{q}_span"])]) if len(D15) else np.zeros((0, len(TDG)))
    if _DQ15:
        fig3, AX3 = plt.subplots(len(_DQ15), len(_BX15), figsize=(4.5 * len(_BX15), 2.7 * len(_DQ15)), squeeze=False)
        for jb, b in enumerate(_BX15):
            for iq, q in enumerate(_DQ15):
                ax = AX3[iq, jb]
                for k in P15_GK:
                    m = _GM15[b][k]
                    if not m.sum():
                        continue
                    A = T15[f"dec_{q}"][m]
                    cen, lo, hi, n_t = med_band(A, nmin=P15_NMIN, nboot=P15_NBOOT)
                    _ok = n_t >= P15_NMIN
                    if _ok.any():
                        ax.fill_between(TDG[_ok], lo[_ok], hi[_ok], color=P15_GCOL[k], alpha=0.16, lw=0)
                        ax.plot(TDG[_ok], cen[_ok], "o", color=P15_GCOL[k], ms=3.5, alpha=0.85)
                        for _x, _c, _l, _h, _n in zip(TDG[_ok], cen[_ok], lo[_ok], hi[_ok], n_t[_ok]):
                            ROW15.append(dict(box=b, block=f"decayclock_{q}", quantity=k, x=float(_x), med=float(_c),
                                              lo=float(_l), hi=float(_h), n_gal=int(_n)))
                    W = _WFIT15[q][m]
                    _nw = np.isfinite(W).sum(0)
                    with warnings.catch_warnings():
                        warnings.simplefilter("ignore", RuntimeWarning)
                        _wc = np.nanmedian(W, axis=0)
                    _okw = (_nw >= P15_NMIN) & np.isfinite(_wc)
                    _t1 = float(np.nanmedian(D15[f"{q}_tau1"].to_numpy(float)[m]))
                    _t2 = float(np.nanmedian(D15[f"{q}_tau2"].to_numpy(float)[m]))
                    if _okw.sum() > 1:
                        ax.plot(TDG[_okw], _wc[_okw], color=P15_GCOL[k], lw=2.6, zorder=5,
                                label=f"{P15_GSH[k]} ({int(np.isfinite(D15[f'{q}_tau1'].to_numpy(float)[m]).sum())})"
                                      + (rf"  $\tau_1$={_t1:.2f}, $\tau_2$={_t2:.1f}" if np.isfinite(_t1) else ""))
                    _tb = float(np.nanmedian(D15[f"{q}_tbrk"].to_numpy(float)[m]))
                    if np.isfinite(_tb) and _okw.any():
                        ax.axvline(_tb, color=P15_GCOL[k], lw=1.2, ls="--", alpha=0.8)
                ax.grid(alpha=0.2, lw=0.5)
                if iq == 0:
                    ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
                if jb == 0:
                    ax.set_ylabel(P15_QLAB.get(q, q), fontsize=9)
                if iq == len(_DQ15) - 1:
                    ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]", fontsize=9.5)
                if q == "lfd":
                    ax.axhline(DUSTY_CUT, color="0.35", lw=1.0, ls=":")
                _leg15(ax, loc="best", fontsize=7)
        fig3.suptitle("The decay after the truncation: the points are the branches themselves, the lines the median "
                      "double exponential fitted to each galaxy\n(the dashed vertical is where the slow term takes over "
                      "— a bend inside the window is a break, no bend is a single exponential)", y=1.005, fontsize=11.5,
                      fontweight="bold")
        fig3.tight_layout(); paper_save(fig3, "kappa_jetstate_decay"); plt.show()

        # ── FIG 4 — the two slopes of every galaxy, which is the claim in one plane ──
        fig4, AX4 = plt.subplots(len(_DQ15), len(_BX15), figsize=(3.6 * len(_BX15), 3.3 * len(_DQ15)), squeeze=False)
        for jb, b in enumerate(_BX15):
            for iq, q in enumerate(_DQ15):
                ax = AX4[iq, jb]
                g = D15[D15.box == b]
                _lim = []
                for k in P15_GK:
                    s_ = g[g.grp == k]
                    x_, y_ = s_[f"{q}_s_early"].to_numpy(float), s_[f"{q}_s_late"].to_numpy(float)
                    if np.isfinite(x_).any():
                        ax.scatter(x_, y_, s=18, alpha=0.55, color=P15_GCOL[k], edgecolor="none",
                                   label=f"{P15_GSH[k]} ({int(np.isfinite(x_ + y_).sum())})")
                        ax.plot([np.nanmedian(x_)], [np.nanmedian(y_)], "*", ms=16, color=P15_GCOL[k], mec="k",
                                mew=0.6, zorder=6)
                        _lim += [np.nanpercentile(x_[np.isfinite(x_)], [2, 98]).tolist() if np.isfinite(x_).sum() > 2 else [],
                                 np.nanpercentile(y_[np.isfinite(y_)], [2, 98]).tolist() if np.isfinite(y_).sum() > 2 else []]
                _lim = [v for s in _lim for v in s]
                if _lim:
                    _l0, _l1 = float(min(_lim)), float(max(_lim))
                    _pd15 = 0.1 * max(_l1 - _l0, 0.1)
                    ax.plot([_l0 - _pd15, _l1 + _pd15], [_l0 - _pd15, _l1 + _pd15], color="0.5", lw=1.0, ls="--")
                    ax.set_xlim(_l0 - _pd15, _l1 + _pd15); ax.set_ylim(_l0 - _pd15, _l1 + _pd15)
                _ae, _al = [mw_auc(np.concatenate([g[g.grp == "hi"][f"{q}_{c}"].to_numpy(float),
                                                   g[g.grp == "lo"][f"{q}_{c}"].to_numpy(float)]),
                                   np.concatenate([np.ones((g.grp == "hi").sum(), bool),
                                                   np.zeros((g.grp == "lo").sum(), bool)]))[0]
                            for c in ("s_early", "s_late")]
                ax.set_title(f"{BOX_LABEL[b] if iq == 0 else ''}\nAUC  early {_ae:.2f}   late {_al:.2f}", loc="left",
                             fontsize=8.5, color="#7f3b08" if b == NOX_BOX else "k")
                ax.grid(alpha=0.2, lw=0.5)
                if jb == 0:
                    ax.set_ylabel(f"{_DLAB15.get(q, q)}\nthe LAST {P15_DWIN:g} Gyr  [dex/Gyr]", fontsize=8.5)
                if iq == len(_DQ15) - 1:
                    ax.set_xlabel(f"the FIRST {P15_DWIN:g} Gyr after the SFT  [dex/Gyr]", fontsize=9)
                _leg15(ax, loc="best", fontsize=7)
        fig4.suptitle("The initial slope against the late one, galaxy by galaxy — the claim is: the same x, a higher y "
                      "for the pre-depleted branch\n(the dashed line is no bend at all; above it the decay flattened, "
                      "below it, it steepened)", y=1.005, fontsize=11.5, fontweight="bold")
        fig4.tight_layout(); paper_save(fig4, "kappa_jetstate_slopes"); plt.show()

    # ══ (F) THE AFTER — preventive, or a CGM that cools? ═════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(d) WHAT HAPPENS AFTER THE TRUNCATION — the share of the post-SFT branch each AGN state holds, the gas that")
    print("    comes back, and whether the channel goes on firing (the PREVENTIVE case) or falls quiet")
    print(f"  {'box':10s} {'branch':10s} {'N':>4s} " + " ".join(f"{s_:>10s}" for s_ in ("dormant", "radiative", "choked", "X-ray"))
          + f" {'duty':>7s} {'gap':>7s} {'d f_gas':>8s} {'regas':>7s} {'d dust':>8s}")
    for b in _BX15:
        for k in P15_GK:
            g = D15[(D15.box == b) & (D15.grp == k)]
            if not len(g):
                continue
            _f = lambda c: np.nanmedian(g[c].to_numpy(float)) if c in g.columns else np.nan
            print(f"  {b:10s} {P15_GSH[k]:10s} {len(g):4d} "
                  + " ".join(f"{100 * _f(c):9.0f}%" for c in ("f_dor_post", "f_rad_post", "f_chk_post", "f_xr_post"))
                  + f" {_f('duty_post'):7.2f} {_f('gap_max'):7.2f} {_f('dlfgas_post'):+8.2f} {_f('regas'):+7.2f} "
                    f"{_f('lfd_net'):+8.2f}")
            for c in ("f_dor_post", "f_rad_post", "f_chk_post", "f_xr_post", "duty_post", "gap_max", "dlfgas_post",
                      "regas", "lfd_net", "dlfgas_late"):
                if c in g.columns:
                    ROW15.append(dict(box=b, block="after", quantity=c, kcls=k, med=float(_f(c)), n_gal=len(g)))
    print("  (PREVENTIVE would be: the stirred branch keeps a high jet / X-ray share after the SFT, does NOT re-accrete")
    print("   (d f_gas and regas near zero) and loses its dust; the other branch falls quiet and holds its ISM.)")

    if P15_STIRTEST:                                     # the block the user dropped; the code is kept, not run
        print("\n" + "-" * 132)
        print(f"(e) THE STIRRING TEST — the change across the X-ray onset (t_x1 to t_x1 + {P15_STIR:g} Gyr) against the ISM")
        print("    density the onset FOUND. The mechanism says: the denser the ISM at the onset, the more the kick stirs it.")
        print(f"  {'box':10s} {'rho(Sigma_H2(x1), d kappa)':>28s} {'rho(., d comp)':>16s} {'rho(., d dust)':>16s} {'N':>5s}")
        for b in _BX15:
            g = D15[D15.box == b]
            _o, _n = [], 0
            _SQ = "lsig_h2_x1" if np.isfinite(g["lsig_h2_x1"].to_numpy(float)).sum() >= 8 else "lfh2_x1"
            for q in ("stir_kappa_gas", "stir_comp", "stir_lfd"):
                r_, p_, n_ = _rho15(g[_SQ].to_numpy(float), g[q].to_numpy(float))
                _o.append(f"{r_:+.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}"); _n = max(_n, n_)
                ROW15.append(dict(box=b, block="stir", quantity=q, med=r_, p=p_, n_gal=n_))
            print(f"  {b:10s} {_o[0]:>28s} {_o[1]:>16s} {_o[2]:>16s} {_n:5d}"
                  + ("   [the channel is compiled OUT here: this row is the null]" if b == NOX_BOX else ""))
        print("  (the claim is a NEGATIVE first column in the X-ray runs — a denser ISM at the onset is stirred harder — and")
        print("   NO such relation in cis50nox, where the state fires in the bookkeeping but no energy is ever deposited.)")

    # ══ FIG 5 — the after, drawn ════════════════════════════════════════════════════════════════════════════════════
    _AFT15 = ["budget"] + (["stir"] if P15_STIRTEST else []) + ["gas"]
    fig5, AX5 = plt.subplots(len(_BX15), len(_AFT15), figsize=(5.0 * len(_AFT15), 3.6 * len(_BX15)), squeeze=False)
    _SC15 = globals().get("SCOL", {"dormant": "0.55", "radiative": "#9467bd", "choked": "#2ca02c", "xray": "#1f77b4"})
    for i, b in enumerate(_BX15):
        g = D15[D15.box == b]
        _pan = lambda nm: (AX5[i, _AFT15.index(nm)], "abc"[_AFT15.index(nm)])
        ax, _tg = _pan("budget")                                  # the post-SFT state budget
        _yy = np.arange(len(P15_GK))
        for _iy, k in enumerate(P15_GK):
            gg = g[g.grp == k]
            _left = 0.0
            for _st15n, _cc in (("f_dor_post", "dormant"), ("f_rad_post", "radiative"), ("f_chk_post", "choked"),
                                ("f_xr_post", "xray")):
                _v = float(np.nanmedian(gg[_st15n].to_numpy(float))) if len(gg) else np.nan
                if not np.isfinite(_v):
                    continue
                ax.barh(_iy, _v, left=_left, height=0.55, color=_SC15[_cc], edgecolor="w", lw=0.8,
                        label=_cc if _iy == 0 else None)
                if _v > 0.06:
                    ax.text(_left + _v / 2, _iy, f"{100 * _v:.0f}", ha="center", va="center", fontsize=8,
                            color="w" if _cc != "dormant" else "k")
                _left += _v
        ax.set_yticks(_yy); ax.set_yticklabels([P15_GLAB[k] for k in P15_GK], fontsize=8.5)
        ax.set_xlim(0, 1); ax.set_ylim(-0.55, len(P15_GK) - 1 + 0.95)     # headroom for the state legend
        ax.set_xlabel("the share of the post-truncation branch", fontsize=9.5)
        ax.set_title(f"({_tg}) {BOX_LABEL[b]}: what the BH does after the SFT", loc="left", fontsize=9.5,
                     color="#7f3b08" if b == NOX_BOX else "k")
        if i == 0:
            _leg15(ax, loc="upper center", ncol=4, bbox_to_anchor=(0.5, 1.0))
        if P15_STIRTEST:
            ax, _tg = _pan("stir")                                # THE STIRRING TEST
            _SQ = "lsig_h2_x1" if np.isfinite(g["lsig_h2_x1"].to_numpy(float)).sum() >= 3 * P15_NMIN else "lfh2_x1"
            _x15 = g[_SQ].to_numpy(float); _y15 = g["stir_kappa_gas"].to_numpy(float)
            for k in P15_GK:
                m = (g.grp == k).to_numpy(bool)
                if m.sum():
                    ax.scatter(_x15[m], _y15[m], s=20, alpha=0.55, color=P15_GCOL[k], edgecolor="none",
                               label=f"{P15_GLAB[k]}  ({int(m.sum())})")
            _fin = np.isfinite(_x15) & np.isfinite(_y15)
            if _fin.sum() >= 3 * P15_NMIN:
                _e15 = np.unique(np.quantile(_x15[_fin], np.linspace(0, 1, 6)))
                _pts = []
                for _ii, (_lo, _hi) in enumerate(zip(_e15[:-1], _e15[1:])):
                    _s15 = _fin & (_x15 >= _lo) & ((_x15 <= _hi) if _ii == len(_e15) - 2 else (_x15 < _hi))
                    if _s15.sum() >= P15_NMIN:
                        _pts.append((float(np.median(_x15[_s15])), float(np.median(_y15[_s15]))))
                if _pts:
                    ax.plot([p[0] for p in _pts], [p[1] for p in _pts], "-o", color="k", lw=2.4, ms=5, zorder=6,
                            label="the median")
            r_, p_, n_ = _rho15(_x15, _y15)
            ax.axhline(0.0, color="0.4", lw=1.0, ls="--")
            ax.set_xlabel((P15_QLAB.get(_SQ[:-3], _SQ) + " the X-ray onset finds"), fontsize=9.5)
            ax.set_ylabel(rf"$\Delta \kappa_{{\rm rot}}$(gas) over {P15_STIR:g} Gyr", fontsize=9.5)
            ax.set_title(f"({_tg}) stirred harder where the ISM is denser?"
                         + (rf"  $\rho = {r_:+.2f}$" if np.isfinite(r_) else "")
                         + ("  [X-ray OFF: the null]" if b == NOX_BOX else ""), loc="left", fontsize=9.0)
            _leg15(ax, loc="best"); ax.grid(alpha=0.2, lw=0.5)
        ax, _tg = _pan("gas")                                     # the budget plane: the gas that came back, the dust that went
        _qx, _qy = "dlfgas_post", ("lfd_net" if "lfd_net" in g.columns else "stir_lfd")
        for k in P15_GK:
            m = (g.grp == k).to_numpy(bool)
            if m.sum():
                ax.scatter(g[_qx].to_numpy(float)[m], g[_qy].to_numpy(float)[m], s=20, alpha=0.55,
                           color=P15_GCOL[k], edgecolor="none", label=f"{P15_GLAB[k]}  ({int(m.sum())})")
                ax.plot([np.nanmedian(g[_qx].to_numpy(float)[m])], [np.nanmedian(g[_qy].to_numpy(float)[m])], "*",
                        ms=16, color=P15_GCOL[k], mec="k", mew=0.6, zorder=6)
        r_, p_, n_ = _rho15(g[_qx].to_numpy(float), g[_qy].to_numpy(float))
        ax.set_xlabel(r"$\Delta \log f_{\rm gas}$, the truncation to the anchor", fontsize=9.5)
        ax.set_ylabel(r"$\Delta \log M_{\rm dust}/M_\star$ after it", fontsize=9.5)
        ax.set_title(f"({_tg}) no new gas, no new dust?" + (rf"  $\rho = {r_:+.2f}$" if np.isfinite(r_) else ""),
                     loc="left", fontsize=9.0)
        _leg15(ax, loc="best"); ax.grid(alpha=0.2, lw=0.5)
    fig5.suptitle("After the truncation: preventive, or a reservoir that comes back", y=0.999, fontsize=12,
                  fontweight="bold")
    fig5.tight_layout(rect=(0, 0, 1, 0.978)); paper_save(fig5, "kappa_jetstate_after"); plt.show()

    # ══ (G) THE ANCHOR TEST — is a ROTATOR dusty?  is a galaxy with a STRONG OUTFLOW not? ════════════════════════════
    _AQ15 = [(q, l) for q, l in ((("kappa_gas_anc", r"$\kappa_{\rm rot}$(gas) at the anchor"),
                                  (P15_OQ, _OLAB15.get(P15_OQ, P15_OQ))) if P15_AUCQ is None else
                                 [(q, P15_QLAB.get(q, _OLAB15.get(q, q))) for q in P15_AUCQ]) if q in D15.columns
             and np.isfinite(D15[q].to_numpy(float)).sum() >= 8]
    print("\n" + "-" * 132)
    print("(f) THE ANCHOR TEST — the two observables, scored against the dust label they are supposed to predict")
    print(f"    'a rotating galaxy is dusty too' and 'a galaxy blowing gas out through its own R50(stars) right now is")
    print(f"    NOT dusty': AUC = P(the axis is higher in a dusty QG), so > 0.5 means the axis goes WITH the dust")
    print(f"  {'box':10s} {'axis':28s} {'N':>5s} {'dusty':>6s} {'AUC':>6s} {'p':>9s} {'AUC | M*':>9s} "
          f"{'f_dusty low 1/3':>16s} {'f_dusty high 1/3':>17s}")
    for b in _BX15:
        g = D15[D15.box == b]
        for q, _ in _AQ15:
            x_, d_ = g[q].to_numpy(float), g["dusty"].to_numpy(bool)
            if np.isfinite(x_).sum() < 8:
                continue
            au, p_, _nd, _np15 = mw_auc(x_, d_)
            _cr = _crank15(g, q, ["log_mstar"])
            aum, pm, _, _ = mw_auc(_cr, d_)
            _fin = np.isfinite(x_)
            _t1, _t2 = np.nanquantile(x_[_fin], [1 / 3, 2 / 3]) if _fin.sum() >= 6 else (np.nan, np.nan)
            _lo15, _hi15 = _fin & (x_ <= _t1), _fin & (x_ >= _t2)
            _wl = wilson(int((d_ & _lo15).sum()), int(_lo15.sum()))
            _wh = wilson(int((d_ & _hi15).sum()), int(_hi15.sum()))
            print(f"  {b:10s} {q:28s} {int(_fin.sum()):5d} {int((d_ & _fin).sum()):6d} "
                  f"{au:6.2f}{'*' if (np.isfinite(p_) and p_ < 0.05) else ' '}{p_:9.1e} {aum:9.2f} "
                  f"{100 * _wl[0]:9.0f} %/{int(_lo15.sum()):3d} {100 * _wh[0]:12.0f} %/{int(_hi15.sum()):3d}")
            ROW15.append(dict(box=b, block="anchor_auc", quantity=q, auc=au, p=p_, med=aum, lo=_wl[0], hi=_wh[0],
                              n_gal=int(_fin.sum())))
        if "rot" in g.columns and g["rot"].notna().any():
            _r15 = g["rot"].to_numpy(bool) & np.isfinite(g["lfd_anc"].to_numpy(float))
            _n15 = (~g["rot"].to_numpy(bool)) & np.isfinite(g["lfd_anc"].to_numpy(float))
            _d15b = g["dusty"].to_numpy(bool)
            _a, _c = wilson(int((_d15b & _r15).sum()), int(_r15.sum())), wilson(int((_d15b & _n15).sum()), int(_n15.sum()))
            print(f"  {'':10s} {'Part 2 rotators are dusty':28s} {int(_r15.sum() + _n15.sum()):5d} {'':6s} {'':6s} "
                  f"{'':9s} {'':9s} {100 * _c[0]:9.0f} %/{int(_n15.sum()):3d} {100 * _a[0]:12.0f} %/{int(_r15.sum()):3d}"
                  "   (not-rotating / rotating)")
            ROW15.append(dict(box=b, block="anchor_auc", quantity="rot_dusty_frac", med=_a[0], lo=_c[0],
                              n_gal=int(_r15.sum() + _n15.sum())))
    print("  (the last two columns are the same statement without a model: the share of DUSTY QGs among the lowest and")
    print("   the highest third of the axis, with its Wilson 68 % interval in the CSV. 'AUC | M*' repeats the score on")
    print("   the axis RANKED INSIDE stellar-mass cells, because both axes carry the mass with them.)")

    # ══ FIG 6 — the same two tests as distributions, which is what an AUC is a summary of ════════════════════════════
    if _AQ15:
        fig6, AX6 = plt.subplots(len(_BX15), len(_AQ15), figsize=(5.2 * len(_AQ15), 3.1 * len(_BX15)), squeeze=False)
        for ib, b in enumerate(_BX15):
            g = D15[D15.box == b]
            for iq, (q, lab) in enumerate(_AQ15):
                ax = AX6[ib, iq]
                x_, d_ = g[q].to_numpy(float), g["dusty"].to_numpy(bool)
                fin = np.isfinite(x_)
                if fin.sum() < 6:
                    ax.text(0.5, 0.5, f"{q}\nnot measured in {BOX_LABEL[b]}", ha="center", va="center", fontsize=9,
                            color="0.45", transform=ax.transAxes)
                    ax.set_xticks([]); ax.set_yticks([])
                    continue
                _e = np.linspace(*np.nanpercentile(x_[fin], [1, 99]), P15_ABINS + 1)
                if not np.all(np.diff(_e) > 0):
                    _e = np.linspace(np.nanmin(x_[fin]) - 0.5, np.nanmax(x_[fin]) + 0.5, P15_ABINS + 1)
                for _lb, _m in (("dusty", d_ & fin), ("pure", (~d_) & fin)):
                    if _m.sum():
                        ax.hist(x_[_m], bins=_e, histtype="stepfilled", density=True, alpha=0.30,
                                color=POPS[_lb]["color"], lw=0)
                        ax.hist(x_[_m], bins=_e, histtype="step", density=True, color=POPS[_lb]["color"], lw=2.0,
                                label=f"{POPS[_lb]['label'].split('(')[0].strip()}  ({int(_m.sum())})")
                ax2 = ax.twinx()                                     # the same thing read forwards: P(dusty | the axis)
                _px, _py, _pe = [], [], []
                for _l15, _h15 in zip(_e[:-1], _e[1:]):
                    _m = fin & (x_ >= _l15) & (x_ < _h15 + (1e-9 if _h15 == _e[-1] else 0))
                    if _m.sum() >= 3:
                        _p15, _lo15, _hi15 = wilson(int((d_ & _m).sum()), int(_m.sum()))
                        _px.append(0.5 * (_l15 + _h15)); _py.append(_p15)
                        _pe.append((max(_p15 - _lo15, 0.0), max(_hi15 - _p15, 0.0)))   # the Wilson centre is shifted
                        ROW15.append(dict(box=b, block=f"anchor_frac_{q}", quantity="f_dusty", x=float(_px[-1]),
                                          med=_p15, lo=_lo15, hi=_hi15, n_gal=int(_m.sum())))
                if _px:
                    ax2.errorbar(_px, _py, yerr=np.array(_pe).T, fmt="-o", color="k", lw=1.8, ms=4.5, capsize=2.5,
                                 zorder=6, label="the dusty share")
                ax2.set_ylim(-0.03, 1.03); ax2.set_ylabel("dusty share", fontsize=8.5)
                ax2.axhline(float(np.mean(d_[fin])), color="0.45", lw=1.0, ls=":")
                if q.startswith("kappa"):
                    ax.axvline(1 / 3.0, color="#7f3b08", lw=1.2, ls="--")
                    if ib == 0:
                        ax.annotate("isotropic", xy=(1 / 3.0, 0.98), xycoords=("data", "axes fraction"), xytext=(3, -8),
                                    textcoords="offset points", fontsize=7, color="#7f3b08")
                au, p_, _, _ = mw_auc(x_, d_)
                ax.set_title(f"{BOX_LABEL[b]}   AUC = {au:.2f}" + ("*" if (np.isfinite(p_) and p_ < 0.05) else "")
                             + f"   (N = {int(fin.sum())})", loc="left", fontsize=9,
                             color="#7f3b08" if b == NOX_BOX else "k")
                ax.set_xlabel(lab, fontsize=9.5); ax.set_ylabel("density", fontsize=8.5)
                ax.grid(alpha=0.2, lw=0.5)
                if ib == 0 and iq == 0:
                    _leg15(ax, loc="best", fontsize=7)
        fig6.suptitle("The two anchor tests: does a rotating galaxy end up dusty, and does one that is blowing gas out "
                      "through its own $R_{50}(\\star)$ end up dust-poor?\n(the filled histograms are the two dust "
                      "classes, the black line is the dusty share bin by bin with its Wilson interval, the dotted line "
                      "the sample's own dusty share)", y=1.005, fontsize=11.5, fontweight="bold")
        fig6.tight_layout(); paper_save(fig6, "kappa_jetstate_auc"); plt.show()

    # ══ (H) the m25 CGM, where Part 13 has been run in this kernel ═══════════════════════════════════════════════════
    if P15_CGM and "D13" in globals() and isinstance(globals().get("D13"), pd.DataFrame) and len(globals()["D13"]):
        _P13B = globals().get("P9_BOX", "cis25")
        _g15 = {(int(s_), int(i_)): k_ for s_, i_, k_ in
                zip(D15[D15.box == _P13B]["snap"], D15[D15.box == _P13B]["gal_id"], D15[D15.box == _P13B]["grp"])}
        _Q13 = D13.copy()
        _Q13["grp15"] = [_g15.get((int(a_), int(g_))) for a_, g_ in zip(_Q13["anchor_snap"], _Q13["gal_id"])]
        _Q13 = _Q13[[v in P15_GK for v in _Q13["grp15"]]]
        _CGMQ = [("lT_cgm_mw", r"$\langle \log T \rangle_M$ of the CGM", "the CGM temperature"),
                 ("fm_cold", r"cold mass share of the CGM", "the cold phase"),
                 ("lmdot_in_gc", r"$\log \dot M_{\rm in}$(cold gas) at 10 kpc", "the cold inflow"),
                 ("lfd_cgm", r"$\log M_{\rm dust}^{\rm CGM}/M_\star$", "the displaced dust")]
        _CGMQ = [(q, l, s) for q, l, s in _CGMQ if q in _Q13.columns]
        if len(_Q13) and _CGMQ:
            _TB15 = globals().get("P13_TB", np.arange(-3.0, 4.001, 0.5))
            print("\n" + "-" * 132)
            print(f"(g) THE CGM of the {_P13B} branches (Part 13's census, re-split on Part 15's division): "
                  f"{len(_Q13)} nodes, " + ", ".join(f"{k} {int((_Q13.grp15 == k).sum())}" for k in P15_GK))
            print(f"  {'quantity':16s} " + " ".join(f"{w:>22s}" for w in ("before the clock 0", "after it")))
            fig7, AX7 = plt.subplots(1, len(_CGMQ), figsize=(4.4 * len(_CGMQ), 3.6), squeeze=False)
            for iq, (q, lab, short) in enumerate(_CGMQ):
                ax = AX7[0, iq]
                _txt = []
                for k in P15_GK:
                    s_ = _Q13[_Q13.grp15 == k]
                    _x, _y = s_["dt"].to_numpy(float), s_[q].to_numpy(float)
                    _px, _py, _pl, _ph = [], [], [], []
                    for _lo, _hi in zip(_TB15[:-1], _TB15[1:]):
                        _m = np.isfinite(_x) & np.isfinite(_y) & (_x >= _lo) & (_x < _hi)
                        if _m.sum() >= 3:
                            _v = _y[_m]
                            _px.append(0.5 * (_lo + _hi)); _py.append(float(np.median(_v)))
                            _rng15 = np.random.default_rng(11 + len(_px))
                            _b = np.array([np.median(_rng15.choice(_v, len(_v))) for _ in range(200)])
                            _pl.append(float(np.percentile(_b, 16))); _ph.append(float(np.percentile(_b, 84)))
                    if _px:
                        ax.fill_between(_px, _pl, _ph, color=P15_GCOL[k], alpha=0.20, lw=0)
                        ax.plot(_px, _py, "-o", color=P15_GCOL[k], lw=2.4, ms=4.5, label=P15_GLAB[k])
                        for _a15, _b15, _c15, _d15 in zip(_px, _py, _pl, _ph):
                            ROW15.append(dict(box=_P13B, block=f"cgm_{q}", quantity=k, x=_a15, med=_b15, lo=_c15, hi=_d15))
                    _pre = np.isfinite(_y) & (_x < 0); _post = np.isfinite(_y) & (_x >= 0)
                    _txt.append((float(np.median(_y[_pre])) if _pre.sum() else np.nan,
                                 float(np.median(_y[_post])) if _post.sum() else np.nan))
                print(f"  {q:16s} " + " ".join(f"{P15_GLAB[k][:5]} {_t[0]:+7.2f} / {_t[1]:+7.2f}   "
                                               for k, _t in zip(P15_GK, _txt)))
                ax.axvline(0.0, color="k", lw=1.2, ls="--"); ax.grid(alpha=0.2, lw=0.5)
                ax.set_xlabel(globals().get("_CLAB13", r"$t - t_0$  [Gyr]"), fontsize=9.5)
                ax.set_ylabel(lab, fontsize=9.5)
                ax.set_title(f"({'abcd'[iq]}) {short}", loc="left", fontsize=9.5)
                if iq == 0:
                    _leg15(ax, loc="best")
            fig7.suptitle(f"The CGM of the two branches ({_P13B}, Part 13's census on Part 15's division)", y=1.01,
                          fontsize=11.5, fontweight="bold")
            fig7.tight_layout(); paper_save(fig7, "kappa_jetstate_cgm"); plt.show()
            print("  (the open question this answers: does the CGM of the pre-depleted branch COOL — a rising cold share")
            print("   and a returning inflow — while the stirred branch stays hot and preventive?)")
    elif P15_CGM:
        print("\n  (the CGM block needs Part 13 in this kernel — run Parts 9-13 first for the m25 CGM of the two branches.)")

    pd.DataFrame(ROW15).to_csv(os.path.join(OUT, "kappa_jetstate.csv"), index=False)
    D15.to_csv(os.path.join(OUT, "kappa_jetstate_galaxies.csv"), index=False)
    print(f"\ntable -> {os.path.join(OUT, 'kappa_jetstate.csv')}"
          f"\n         {os.path.join(OUT, 'kappa_jetstate_galaxies.csv')} (one row per branch, every column of the census)")
```

## Part 16 — the order of events, and the CGM cycle the truncation starts

Part 15 ended on a division that does not divide: the ISM density when the channel opens is worth AUC 0.48 in m25, while
$\kappa_{\rm rot}$(gas) at the anchor is worth 0.68–0.74 in every box. So the question moves **upstream** — to *when* the
branches fire, and to the halo they fire into. Five claims, each falsifiable on its own:

| | claim | where |
|---|---|---|
| **C1** | a branch that reaches the **jet mode early** reaches the truncation early and fires the X-ray channel early: the three events are one chain, not three draws | (a) |
| **C2** | that early jet leaves a **higher $\kappa_{\rm rot}$(gas)** at the anchor, because what the jet removes is the **low-$j$ gas** — the central kpc empties, $R_{50}$(gas) grows, and the rotating part is what survives | (b) |
| **C3** | the CGM temperature has **three phases** with the truncation as the hinge: it **rises** before it (the galaxy is still forming stars — supernova heating), **falls** just after it (the ISM is displaced outward, cold gas is *added* to the CGM and the mass-weighted mean is diluted — so the cold phase must rise at the same time), and **rises again** once the X-ray channel is on, with the cold inflow and the CGM dust falling with it | (c) |
| **C4** | the two classes are **the same for ~1 Gyr** after the truncation and only then part company: the rotators' CGM then cools, keeps its dust and goes on accreting | (d) |
| **C5** | **is the dust just the star formation?** Every galaxy on the $f_{\rm dust}$–sSFR plane with the region it could have filled **in one Gyr at its own sSFR** shaded behind it. Above it, the dust is older than the window and has survived; inside it, nothing but the galaxy's own recent star formation is needed | (e) |

**What is read.** Nothing is measured from particles in this part. The chains and Part 3's kinematic sidecar give the three
event times and every track; the CGM comes from **Part 13's own per-node census** (`kappa_m25_cgm_nodes.csv`), re-clocked
here on this part's $t_{\rm SFT}$ and re-split on this part's classes so that `P13_CLOCK` and `P13_SPLIT` cannot change what
this part means; the plane of (e) is Part 0's catalogue cache — **every** galaxy in it above $\log M_\star >$ `MASS_MIN`, not
only the quenched sample. If Part 13 has not been run, (c) and (d) say so and are skipped; everything else still runs.

**"Early" is a rank, never a date.** $t_{\rm jet}$ is a cosmic time and each branch is watched up to a different anchor, so
the raw number confounds *fired early* with *was looked at early*. `_crank16` ranks it inside the cartesian cells of
`P16_CTRL` (the anchor time × the stellar mass, `P16_CELLQ` quantiles each): early **for its kind**. Both the raw and the
ranked correlation are printed side by side, so the difference between them is visible rather than assumed.

**The dust budget of (e)** is $f_{\rm dust}^{\rm new} = y_{\rm d}\,{\rm sSFR}\,\Delta t$ with $\Delta t =$ `P16_DT` and
$y_{\rm d}$ the mass of dust condensed per unit mass of stars formed. SIMBA condenses a fixed fraction of the metals leaving
the SNII and AGB ejecta ($\delta_{\rm SNII} = 0.15$, $\delta_{\rm AGB} = 0.2$; Davé+19, Li, Narayanan & Davé 2019), and with
a Chabrier IMF returning $y_Z \sim 0.01$–$0.03$ per unit mass formed over a Gyr that is $y_{\rm d} \sim 1.5$–$6\times10^{-3}$.
The **whole range is shaded**, and the claim being tested is an order-of-magnitude one, so nothing depends on picking a number
inside it. Two independent controls are drawn on the same plane: the **star-forming locus** (where galaxies that *are* forming
stars actually sit) and, per branch, the same excess computed from the mass its **own SFH** formed in that Gyr — no
constant-sSFR assumption. A galaxy with no star formation at all can make nothing, so its budget is the dust floor and it is
drawn in its own strip beside the plane rather than dropped off the edge.

| figure | what it shows |
|---|---|
| **1 the order** | every branch's three events on one timeline, sorted by $t_{\rm jet}$, and what the rank of the first one leaves behind: the length of the jet leg, $\kappa$ at the anchor, the dusty share, the X-ray it then takes |
| **2 the jet leg** | $\kappa_{\rm rot}$(gas), $R_{50}$(gas), $\Sigma_{\rm gas}$, $f_{\rm dust}$ and the X-ray occupancy on the clock $t - t_{\rm jet}$, early jet against late |
| **3 the low-$j$ removal** | the same leg in the m25 **particles**: the gas and the dust inside 1 kpc, $R_{50}$(gas) and the ISM mass — C2's mechanism, measured |
| **4 the CGM cycle** | the temperature, its cold phase, the inner-halo temperature, the cold inflow, the CGM dust and its dust-to-gas ratio on $t - t_{\rm SFT}$, with C3's three windows shaded and the X-ray occupancy behind |
| **5 the divergence** | class **minus** class, bin by bin, with a bootstrap over galaxies and $t_{\rm div}$ — the first epoch at or after the truncation where the difference holds for `P16_DIVSIG` bins running |
| **6 the dust budget** | the $f_{\rm dust}$–sSFR plane with the Gyr of dust production shaded, and the per-branch excess over what its own SFH could have made |

**The scorecard of (c)** is a set of per-galaxy slopes: one straight line through $\langle\log T\rangle_M$(CGM) in each of the
three windows of `P16_WIN`, correlated against what C3 says makes it — the sSFR of the same window, the ISM mass the galaxy
loses, the cold CGM mass it gains, the X-ray share of the window, and the cold inflow and CGM dust as they fall. The `d`
columns are **late minus early inside the window**, so a galaxy that *loses* ISM has `d < 0` and the predicted correlation
with a *cooling* CGM is **positive**.

```python
# ── Part 16 — THE ORDER OF EVENTS, AND THE CGM CYCLE THE TRUNCATION STARTS ──
# Part 15 ended on a division that does not divide: the ISM density when the channel opens is worth AUC 0.48 in m25, while
# kappa_rot(gas) at the anchor is worth 0.68-0.74 everywhere. So the question moves upstream, to WHEN the branches fire and
# to the halo they fire into. The claims tested here are stated first, each one falsifiable on its own:
#
#   C1  THE ORDER.  A branch that reaches the jet mode EARLY reaches the truncation early and fires the X-ray channel early
#       — the three events are one chain, not three independent draws. (a)
#   C2  THE ROTATION IS MADE BY THE JET.  That early jet leaves a HIGHER kappa_rot(gas) at the anchor, because what the jet
#       removes is the LOW-j gas: the central kpc empties, R50(gas) grows, and the surviving gas is the rotating part. The
#       counterfactual is the box with no X-ray channel — if the mechanism is the feedback, the relation must weaken there. (b)
#   C3  THE CGM TEMPERATURE HAS THREE PHASES, and the truncation is the hinge:  it RISES before the SFT (the galaxy is still
#       forming stars — supernova shock heating), FALLS just after it (the ISM is displaced outward: cold gas is ADDED to the
#       CGM and the mass-weighted temperature is diluted, so the cold phase must rise at the same time), and RISES again once
#       the X-ray channel is on, with the cold inflow and the CGM dust falling with it. (c)
#   C4  THE TWO CLASSES ARE THE SAME FOR ~1 Gyr AFTER THE TRUNCATION and only then part company: the cold inflow and the
#       displaced dust are indistinguishable at first, and afterwards the rotators' CGM cools, keeps its dust, and goes on
#       accreting. The measurement is a DIVERGENCE TIME with a bootstrap band, not an eyeballed curve. (d)
#   C5  IS THE DUST JUST THE STAR FORMATION?  Every galaxy on the f_dust - sSFR plane, with the region a galaxy could have
#       FILLED IN ONE Gyr at its own sSFR shaded behind it (the condensation yield of SIMBA's own dust model). A galaxy above
#       that region holds more dust than its recent star formation can have made — the dust is inherited and survived; one
#       inside it needs nothing but its own SFR. The expectation under test: the X-ray runs sit above, the no-X-ray box
#       inside. (e)
#
# WHAT IS READ.  Nothing is measured from particles here. The chains and Part 3's kinematic sidecar give the events and the
# tracks; the CGM comes from PART 13's census (kappa_<box>_cgm_nodes.csv, the per-node table it writes) re-clocked on this
# part's own t_SFT so that P13_CLOCK cannot change what this part means; the plane of (e) is Part 0's catalogue cache, every
# galaxy in it and not only the quenched sample. If Part 13 has not been run, (c) and (d) say so and are skipped.

P16_BOXES  = [b for b in (BOXES + [NOX_BOX]) if os.path.exists(HIST_FILE.format(box=b))]
P16_SPLIT  = "kcls"        # what (c) and (d) are split by: "kcls" = Part 2's selection | "pop" = the dust label
P16_CTRL   = ["t_anchor", "log_mstar"]   # what "early" is measured AGAINST: t_jet is ranked inside cells of these (an early jet
P16_CELLQ  = 3             # at z = 3 is not an early jet at z = 0.3, and the mass sets the whole clock) — quantiles per axis
P16_MINCELL = 6            # a cell smaller than this contributes no rank
P16_TRK    = ["kappa_gas", "lr_gas", "lsig_gas", "lfd", "lfgas", "lfh2", "lssfr", "lms", "xr"]   # the tracks of (b)
P16_JTRK   = ["kappa_gas", "lr_gas", "lsig_gas", "lfd", "xr"]   # ... and the rows drawn on the jet clock
P16_ORDER  = ["t_jet", "t_sft", "t_x1"]                    # the chain of C1, in the order it is claimed to happen

# ── (c) / (d): the CGM, from Part 13's per-node census, re-clocked here ──
P16_CGM    = True
P16_CGMBOX = P9_BOX if "P9_BOX" in globals() else "cis25"
P16_NODES  = os.path.join(OUT, "kappa_m25_cgm_nodes.csv")   # Part 13's own product (it names it after the m25 box it runs on)
P16_TB     = np.arange(-3.0, 4.001, 0.5)    # the bin EDGES of t - t_SFT the nodes are collected in [Gyr]
P16_NMIN   = 4             # galaxies (not nodes) a drawn bin needs
P16_NBOOT  = 400           # the bootstrap over GALAXIES behind every band and every divergence test
P16_WIN    = [("before", -1.5, 0.0), ("displaced", 0.0, 1.0), ("heated", 1.0, 3.0)]   # C3's three phases
P16_DIVQ   = ["lmdot_in_gc", "lT_cgm_mw", "fm_cold", "lfd_cgm", "ldgr_cgm", "lfd_ism"]  # what the divergence time is measured for
P16_DIVSIG = 2                # consecutive bins whose bootstrap 16-84 interval must exclude zero before the classes are parted

# ── (e) the dust budget: how much dust the last P16_DT Gyr of star formation can have made ──
P16_DT     = 1.0           # [Gyr] the window of the claim ("the dust generated in 1 Gyr by that sSFR")
# The yield is SIMBA's own condensation channel: a fraction of the metals leaving the SNII / AGB ejecta condenses into grains
# (delta_SNII = 0.15, delta_AGB = 0.2 in Dave+19 / Li, Narayanan & Dave 2019). With a Chabrier IMF the metal mass returned per
# unit mass of stars formed is y_Z ~ 0.01-0.03 over a Gyr, so the dust made per unit mass FORMED is y_d ~ 1.5e-3 to 6e-3. The
# band is drawn over that whole range and nothing below depends on picking one number inside it — the statement being tested is
# an ORDER OF MAGNITUDE one (is the dust of a quenched galaxy within reach of its own recent star formation at all?).
P16_YD     = (1.5e-3, 6.0e-3)
P16_YDLAB  = r"$y_{\rm d} = 1.5$–$6\times10^{-3}$  (condensation, $\delta_{\rm SNII} = 0.15$, $\delta_{\rm AGB} = 0.2$)"
P16_SFCUT  = -10.5         # log sSFR above this = still star forming, for the locus drawn through the main sequence
P16_PLANE  = ["lssfr", "lfd"]
P16_MAXPT  = 12000         # a cap on the points drawn per panel (the sample is drawn, the statistics use everything)

import scipy.stats as _st16
import time as _time16

_KLAB16 = {"kappa_gas": r"$\kappa_{\rm rot}$(gas)", "lr_gas": r"$\log R_{50}$(gas) [kpc]",
           "lsig_gas": r"$\log \Sigma_{\rm gas}$ [M$_\odot$ kpc$^{-2}$]", "lfd": r"$\log M_{\rm dust}/M_\star$",
           "lfgas": r"$\log M_{\rm gas}/M_\star$", "lfh2": r"$\log M_{\rm H_2}/M_\star$",
           "lssfr": r"$\log$ sSFR [yr$^{-1}$]", "lms": r"$\log M_\star$", "xr": "X-ray channel on"}
_ELAB16 = {"t_jet": "the jet activation", "t_sft": "the truncation", "t_x1": "the X-ray onset"}
_ESHORT16 = {"t_jet": "jet", "t_sft": "SFT", "t_x1": "X-ray"}
_EMK16 = {"t_jet": ("o", "#2ca02c"), "t_sft": ("s", "#d62728"), "t_x1": ("^", "#1f77b4")}


def _at16(t, y, t0):
    """One track's value at one epoch, linearly interpolated, never extrapolated."""
    ok = np.isfinite(t) & np.isfinite(y)
    t_, y_ = np.asarray(t)[ok], np.asarray(y)[ok]
    return float(np.interp(t0, t_, y_)) if (np.isfinite(t0) and ok.sum() >= 2 and t_[0] <= t0 <= t_[-1]) else np.nan


def _win16(t, y, lo, hi):
    """The median of a track inside a window of its own clock."""
    m = np.isfinite(t) & np.isfinite(y) & (t >= lo) & (t <= hi)
    return float(np.median(np.asarray(y)[m])) if m.any() else np.nan


def _slope16(t, y, lo, hi, nmin=3):
    """The straight-line slope of a track inside a window [per Gyr], and how many points carried it."""
    m = np.isfinite(t) & np.isfinite(y) & (t >= lo) & (t <= hi)
    if m.sum() < nmin or np.ptp(np.asarray(t)[m]) <= 0:
        return np.nan, int(m.sum())
    return float(np.polyfit(np.asarray(t)[m], np.asarray(y)[m], 1)[0]), int(m.sum())


def _push16(rows, block, grid, c, lo, hi, n, **kw):
    """Every drawn point, one row per epoch, so the figures' own numbers leave the notebook as a table."""
    for i in range(len(grid)):
        if np.isfinite(c[i]):
            rows.append(dict(block=block, t=float(grid[i]), val=float(c[i]),
                             lo=float(lo[i]) if np.isfinite(lo[i]) else np.nan,
                             hi=float(hi[i]) if np.isfinite(hi[i]) else np.nan, n=int(n[i]), **kw))


def _rho16(x, y):
    x, y = np.asarray(x, float), np.asarray(y, float)
    ok = np.isfinite(x) & np.isfinite(y)
    if ok.sum() < 6 or np.nanstd(x[ok]) == 0 or np.nanstd(y[ok]) == 0:
        return np.nan, np.nan, int(ok.sum())
    r_ = spearmanr(x[ok], y[ok])
    return float(r_.statistic), float(r_.pvalue), int(ok.sum())


def _prho16(x, y, ctrl):
    """PARTIAL Spearman (Part 15's, verbatim): everything rank-transformed, the controls regressed out, the residuals
    correlated; the p from the t approximation with k controls."""
    cols = [np.asarray(x, float), np.asarray(y, float)] + [np.asarray(c, float) for c in ctrl]
    ok = np.all([np.isfinite(c) for c in cols], axis=0)
    if ok.sum() < len(ctrl) + 8:
        return np.nan, np.nan, int(ok.sum())
    R = np.array([_st16.rankdata(c[ok]) for c in cols], float).T
    sd = R.std(0)
    if not (sd[:2] > 0).all():
        return np.nan, np.nan, int(ok.sum())
    R, sd = R[:, np.r_[True, True, sd[2:] > 0]], sd[np.r_[True, True, sd[2:] > 0]]
    k = R.shape[1] - 2
    R = (R - R.mean(0)) / sd
    A = np.column_stack([np.ones(len(R)), R[:, 2:]])
    res = [R[:, i] - A @ np.linalg.lstsq(A, R[:, i], rcond=None)[0] for i in (0, 1)]
    if np.std(res[0]) == 0 or np.std(res[1]) == 0:
        return np.nan, np.nan, int(ok.sum())
    r_, n_ = float(np.corrcoef(res[0], res[1])[0, 1]), int(ok.sum())
    dof = n_ - 2 - k
    return r_, (float(2 * _st16.t.sf(abs(r_) * np.sqrt(dof / max(1 - r_ * r_, 1e-12)), dof)) if dof > 0 else np.nan), n_


def _crank16(g, col, ctrl):
    """THE MEANING OF 'EARLY' — the rank of `col` in [0, 1] inside the cartesian cells of `ctrl`. t_jet is a cosmic time and
    every branch is watched up to a different anchor, so the raw number confounds 'fired early' with 'was looked at early';
    ranking inside cells of the anchor time and the stellar mass asks the question that was meant: early FOR ITS KIND."""
    v = g[col].to_numpy(float) if col in g.columns else np.full(len(g), np.nan)
    r, cells, ok = np.full(len(v), np.nan), np.zeros(len(v), np.int64), np.isfinite(v)
    for c in ctrl:
        if c not in g.columns:
            continue
        x = g[c].to_numpy(float)
        e = np.unique(np.nanquantile(x, np.linspace(0, 1, P16_CELLQ + 1))) if np.isfinite(x).any() else np.array([0.0, 1.0])
        lab = np.where(np.isfinite(x), np.clip(np.digitize(x, e[1:-1]), 0, max(len(e) - 2, 0)), -1)
        cells = cells * (P16_CELLQ + 1) + (lab + 1)
        ok &= np.isfinite(x)
    for c in np.unique(cells[ok]):
        m = ok & (cells == c)
        if m.sum() >= P16_MINCELL:
            r[m] = (_st16.rankdata(v[m]) - 0.5) / m.sum()
    return r


def _kinalign16(gh, K, o):
    """Part 3's sidecar, matched on the snapshot numbers and the galaxy ids themselves (never on the array order)."""
    out = {}
    _sh = np.asarray(gh["snaps"][:]).astype(int)[o] if "snaps" in gh else None
    _gid = np.asarray(gh["gal_ids"][:]).astype(np.int64)
    if K is None or _sh is None:
        return out
    _rs = {int(v): i for i, v in enumerate(np.asarray(K["snaps"]).astype(int))}
    _ri = {int(v): j for j, v in enumerate(np.asarray(K["gal_ids"]).astype(np.int64))}
    _row = np.array([_rs.get(int(v), -1) for v in _sh]); _col = np.array([_ri.get(int(v), -1) for v in _gid])
    _mr, _mc = _row >= 0, _col >= 0
    for c in ("kappa_gas", "r_gas", "r_star"):
        if c in K and _mr.any() and _mc.any():
            B = np.full((len(_sh), len(_gid)), np.nan)
            B[np.ix_(_mr, _mc)] = np.asarray(K[c], float)[np.ix_(_row[_mr], _col[_mc])]
            out[c] = B
    return out


def _pass16(box, want):
    """One box's chains -> the three event times of every branch, the state around each of them, the star formation the dust
    budget of (e) is measured against, and the tracks on BOTH clocks (t - t_jet for C2, t - t_SFT for everything else)."""
    rows, TR = [], {q: {"jet": [], "sft": []} for q in P16_TRK}
    KIN = _kinread3(box) if "_kinread3" in globals() else {}
    with h5py.File(HIST_FILE.format(box=box), "r") as f:
        for gname in f:
            a = int(gname.replace("snap", ""))
            if not any(k[0] == a for k in want):
                continue
            gh = f[gname]
            t = gh["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
            if len(t) < 3:
                continue
            ms, sfr, mg, mbh, fedd = (gh[c][:][o] for c in ("mstar", "sfr", "mgas", "mbh", "fedd"))
            md = gh["mdust"][:][o] if "mdust" in gh else np.full_like(ms, np.nan)
            mh = gh["mh2"][:][o] if "mh2" in gh else np.full_like(ms, np.nan)
            S, D, X = agn_states(ms, mg, mbh, fedd)                      # Part 0's gates, verbatim
            xray = np.nan_to_num(S["xray"]) > 0.5
            jet = np.nan_to_num(D["jet"]) > 0.5
            kin = _kinalign16(gh, KIN.get(a), o)
            _nan2 = np.full_like(ms, np.nan)
            _kg, _rg, _rs2 = (kin.get(c, _nan2) for c in ("kappa_gas", "r_gas", "r_star"))
            dtw = np.gradient(t)
            for j, gid in enumerate(np.asarray(gh["gal_ids"][:]).astype(np.int64)):
                key = (a, int(gid))
                if key not in want:
                    continue
                r, ex = want[key], X["exist"][:, j]
                if ex.sum() < 3:
                    continue
                ka = int(np.flatnonzero(ex)[-1]); t_anc = float(t[ka])
                xj, jj = xray[:, j] & ex, jet[:, j] & ex
                first = lambda m: float(t[m][0]) if m.any() else np.nan
                t_x1, t_jet, t_sft = first(xj), first(jj), float(r.get("t_sft", np.nan))
                with np.errstate(all="ignore"):
                    den = np.where(ex, ms[:, j], np.nan)
                    v = {"lfd":   np.log10(np.where(ex, np.clip(md[:, j] / den, 10 ** P9_FLOOR, None), np.nan)),
                         "lfgas": np.log10(np.where((mg[:, j] > 0) & ex, mg[:, j] / den, np.nan)),
                         "lfh2":  np.log10(np.where((mh[:, j] > 0) & ex, HE * mh[:, j] / den, np.nan)),
                         "lssfr": np.log10(np.where(ex, np.clip(np.maximum(sfr[:, j], 0.0) / den, SSFR_FLOOR, None), np.nan)),
                         "lms":   np.log10(np.where(ex, ms[:, j], np.nan)),
                         "xr":    np.where(ex, xj.astype(float), np.nan),
                         "kappa_gas": np.where(ex, _kg[:, j], np.nan),
                         "lr_gas": np.log10(np.where((_rg[:, j] > 0) & ex, _rg[:, j], np.nan)),
                         "lsig_gas": np.log10(np.where((_rg[:, j] > 0) & (mg[:, j] > 0) & ex,
                                                       0.5 * mg[:, j] / (np.pi * _rg[:, j] ** 2), np.nan))}
                rec = dict(box=box, snap=a, gal_id=int(gid), t_anchor=t_anc, t_jet=t_jet, t_sft=t_sft, t_x1=t_x1,
                           t_span=t_anc - float(t[ex][0]))
                rec["lag_sft"] = t_sft - t_jet                            # C1: the waits between the links of the chain
                rec["lag_x1"] = t_x1 - t_jet
                rec["lag_xs"] = t_x1 - t_sft
                rec["order_ok"] = bool(np.isfinite([t_jet, t_sft, t_x1]).all() and t_jet <= t_sft and t_jet <= t_x1)
                for _e, _te in (("jet", t_jet), ("sft", t_sft), ("x1", t_x1)):
                    for q in ("kappa_gas", "lr_gas", "lsig_gas", "lfd", "lfgas", "lfh2", "lssfr", "lms"):
                        rec[f"{q}_{_e}"] = _at16(t, v[q], _te)
                for q in P16_TRK:                                          # the anchor, read off the branch's own last step
                    rec[f"{q}_anc"] = float(v[q][ka]) if np.isfinite(v[q][ka]) else np.nan
                    TR[q]["jet"].append(to_grid(t - t_jet, v[q]) if np.isfinite(t_jet) else np.full(len(TG), np.nan))
                    TR[q]["sft"].append(to_grid(t - t_sft, v[q]) if np.isfinite(t_sft) else np.full(len(TG), np.nan))
                # C2: what the jet leg DOES to the rotation and to the extent, on the jet's own clock
                for q in ("kappa_gas", "lr_gas", "lsig_gas", "lfgas"):
                    _b16 = _win16(t - t_jet, v[q], -0.75, -0.05)
                    rec[f"d{q}_jet"] = rec[f"{q}_anc"] - _b16 if np.isfinite(rec[f"{q}_anc"]) else np.nan
                    rec[f"s{q}_jet"], _ = _slope16(t - t_jet, v[q], 0.0, 1.5)
                rec["f_jet_pre"] = (float(np.sum(dtw[jj & ex & (t <= t_sft)]) / np.sum(dtw[ex & (t <= t_sft)]))
                                    if np.isfinite(t_sft) and (ex & (t <= t_sft)).any() else np.nan)
                rec["e_x"] = float(np.sum(dtw[xj]))
                rec["f_x_post"] = (float(np.sum(dtw[xj & ex & (t >= t_sft)]) / np.sum(dtw[ex & (t >= t_sft)]))
                                   if np.isfinite(t_sft) and (ex & (t >= t_sft)).any() else np.nan)
                # (e) THE DUST BUDGET: the stellar mass the branch ACTUALLY formed in the last P16_DT Gyr — the sSFR of the
                # plane assumes that rate was constant over the window, and this is the same number without the assumption
                _bw = ex & (t >= t_anc - P16_DT) & (t <= t_anc)
                rec["dms_dt"] = float(np.sum(np.nan_to_num(np.maximum(sfr[_bw, j], 0.0)) * dtw[_bw]) * 1e9) if _bw.any() else np.nan
                _ps = ex & np.isfinite(t_sft) & (t >= t_sft)
                rec["dms_post"] = (float(np.sum(np.nan_to_num(np.maximum(sfr[_ps, j], 0.0)) * dtw[_ps]) * 1e9)
                                   if _ps.any() else np.nan)
                rec["m_star_anc"] = float(ms[ka, j]) if np.isfinite(ms[ka, j]) else np.nan
                rec["m_dust_anc"] = float(md[ka, j]) if np.isfinite(md[ka, j]) else np.nan
                for c in ("z", "age", "log_mstar", "log_mbh", "lfd", "lssfr", "kcls", "rot", "pop", "central", "kappa",
                          "ngas", "kz", "t_qt"):
                    if c in r:
                        rec[c] = r[c]
                rows.append(rec)
    F = pd.DataFrame(rows)
    T = {q: {k: np.asarray(vv, float).reshape(len(F), len(TG)) for k, vv in d.items()} for q, d in TR.items()}
    return F, T


print("=" * 132)
print("Part 16 — THE ORDER OF EVENTS, AND THE CGM CYCLE THE TRUNCATION STARTS.  'early' = the rank of t_jet inside cells of "
      + " x ".join(P16_CTRL))
print("=" * 132)

_D16, _T16 = [], {q: {"jet": [], "sft": []} for q in P16_TRK}
for _b16 in P16_BOXES:
    _sub16 = QM[QM["box"] == _b16]
    _w16 = {(int(s_), int(i_)): r_ for (s_, i_), r_ in zip(zip(_sub16["snap"], _sub16["gal_id"]), _sub16.to_dict("records"))}
    _t016 = _time16.time()
    _f16, _tr16 = _pass16(_b16, _w16)
    if not len(_f16):
        print(f"  [{_b16}] no chain of Part 0's sample is in the histories cache — skipped"); continue
    _D16.append(_f16)
    for q in P16_TRK:
        for k in ("jet", "sft"):
            _T16[q][k].append(_tr16[q][k])
    _cov16 = lambda c: int(np.isfinite(_f16[c].to_numpy(float)).sum())
    print(f"  [{_b16}] {len(_f16)} branches in {_time16.time() - _t016:.0f} s: {_cov16('t_jet')} reach the jet mode, "
          f"{_cov16('t_sft')} have a truncation, {_cov16('t_x1')} fire the X-ray channel; "
          f"{int(_f16['order_ok'].sum())} carry all three in the order jet -> SFT / X-ray"
          f";  {_cov16('kappa_gas_anc')} carry Part 3's kinematic sidecar")
D16 = pd.concat(_D16, ignore_index=True) if _D16 else pd.DataFrame()
T16 = {q: {k: np.vstack(v) for k, v in d.items()} for q, d in _T16.items()} if _D16 else {}

if not len(D16):
    print("Part 16: no branch carried a histories chain — nothing to draw.")
else:
    _BX16 = [b for b in P16_BOXES if (D16["box"] == b).any()]
    D16["dusty"] = D16["lfd_anc"].to_numpy(float) >= DUSTY_CUT
    D16["early"] = np.concatenate([_crank16(D16[D16.box == b], "t_jet", P16_CTRL) for b in _BX16])   # 0 = the earliest jet
    D16["egrp"] = np.where(D16["early"] <= 1.0 / 3, "early", np.where(D16["early"] >= 2.0 / 3, "late", "mid"))
    D16["cls"] = D16[P16_SPLIT].astype(str) if P16_SPLIT in D16.columns else "all"
    _EG16 = {"early": dict(color="#7b3294", lw=2.6, ls="-", label="jet EARLY for its epoch and mass (lowest third)"),
             "late":  dict(color="#008837", lw=2.4, ls="-", label="jet LATE (highest third)")}
    _CL16 = ({k: dict(color=v["color"], lw=v["lw"], ls=v.get("ls", "-"), label=v.get("label", k)) for k, v in KCLS.items()}
             if P16_SPLIT == "kcls" else {k: dict(color=v["color"], lw=v["lw"], ls=v.get("ls", "-"), label=v["label"])
                                          for k, v in POPS.items()})
    _CLN16 = ({k: KCLS[k].get("short", k) for k in KCLS} if P16_SPLIT == "kcls"
              else {"dusty": "dusty QG", "pure": "dust-poor QG"})

    # ── (a) THE ORDER: is an early jet an early truncation and an early X-ray onset? ──
    print("\n" + "-" * 132)
    print("(a) THE CHAIN — the three events of every branch, and whether they move together.  rho is Spearman; 'rank' is the\n"
          "    same correlation with t_jet replaced by its rank inside cells of " + " x ".join(P16_CTRL)
          + " (the epoch cannot then do the work),\n    and the two right-hand columns are what the early jet LEAVES at the anchor.")
    print(f"  {'box':10s} {'N':>5s} {'order':>6s} {'t_jet':>7s} {'lag SFT':>8s} {'lag Xray':>9s} | "
          f"{'rho t_sft':>9s} {'rank':>6s} | {'rho t_x1':>9s} {'rank':>6s} | {'rho kappa':>9s} {'rank':>6s} | "
          f"{'rho lfd':>8s} {'rank':>6s}")
    ROW16 = []
    for b in _BX16:
        g = D16[D16.box == b]
        _n = int(np.isfinite(g["t_jet"]).sum())
        _o = float(g["order_ok"].mean()) if len(g) else np.nan
        _md = lambda c: float(np.nanmedian(g[c].to_numpy(float))) if np.isfinite(g[c].to_numpy(float)).any() else np.nan
        _cells = []
        for y in ("t_sft", "t_x1", "kappa_gas_anc", "lfd_anc"):
            r1, p1, _ = _rho16(g["t_jet"], g[y])
            r2, p2, _ = _rho16(g["early"], g[y])
            _cells += [f"{r1:+.2f}" + ("*" if np.isfinite(p1) and p1 < 0.05 else " "),
                       f"{r2:+.2f}" + ("*" if np.isfinite(p2) and p2 < 0.05 else " ")]
            ROW16.append(dict(block="a", box=b, y=y, rho_tjet=r1, p_tjet=p1, rho_rank=r2, p_rank=p2, n=_n))
        print(f"  {b:10s} {_n:5d} {100 * _o:5.0f}% {_md('t_jet'):7.2f} {_md('lag_sft'):8.2f} {_md('lag_x1'):9.2f} | "
              + " | ".join(f"{_cells[2 * i]:>9s} {_cells[2 * i + 1]:>6s}" for i in range(4)))
    print("  (an early jet that reaches the truncation early gives rho > 0 in the two left blocks; C2 is the third: rho < 0\n"
          "   there means the EARLIER the jet, the HIGHER kappa_rot(gas) at the anchor. "
          + (NOX_BOX if NOX_BOX in _BX16 else "the no-X-ray box") + " is the counterfactual: if the\n"
          "   mechanism is the feedback, the third block must weaken or reverse there.)")

    # the same statement without a model: what each third of the axis ends up as
    print(f"\n  {'box':10s} {'group':7s} {'N':>5s} {'t_jet':>7s} {'lag SFT':>8s} {'kappa_anc':>10s} {'log fdust':>10s} "
          f"{'dusty':>8s} {'rotating':>9s} {'f_x post':>9s}")
    for b in _BX16:
        for e in ("early", "late"):
            g = D16[(D16.box == b) & (D16.egrp == e)]
            if not len(g):
                continue
            _k, _n = int(g["dusty"].sum()), len(g)
            _p, _lo, _hi = wilson(_k, _n)
            _r = float(np.mean(g["rot"].astype(float))) if "rot" in g.columns else np.nan
            print(f"  {b:10s} {e:7s} {_n:5d} {np.nanmedian(g['t_jet']):7.2f} {np.nanmedian(g['lag_sft']):8.2f} "
                  f"{np.nanmedian(g['kappa_gas_anc']):10.3f} {np.nanmedian(g['lfd_anc']):10.2f} "
                  f"{100 * _p:6.0f} % {100 * _r:8.0f} % {np.nanmedian(g['f_x_post']):9.2f}")
            ROW16.append(dict(block="a-group", box=b, grp=e, n=_n, t_jet=float(np.nanmedian(g["t_jet"])),
                              lag_sft=float(np.nanmedian(g["lag_sft"])), kappa=float(np.nanmedian(g["kappa_gas_anc"])),
                              lfd=float(np.nanmedian(g["lfd_anc"])), f_dusty=_p, f_dusty_lo=_lo, f_dusty_hi=_hi,
                              f_rot=_r, f_x_post=float(np.nanmedian(g["f_x_post"]))))

    # ── FIG 1 — the chain itself: every branch's three events, and what the rank of the first one leaves behind ──
    _BC16 = {b: c for b, c in zip(_BX16, ["#1b9e77", "#d95f02", "#7570b3", "#666666"] * 3)}
    _RB16 = np.linspace(0.0, 1.0, 6)                     # the rank bins of the trend panels
    _RM16 = 0.5 * (_RB16[:-1] + _RB16[1:])

    def _rtrend16(g, col, how="median"):
        """One box's trend against the early-rank: the median (or the share) in each rank bin, with a bootstrap band."""
        x, y = g["early"].to_numpy(float), g[col].to_numpy(float)
        c, lo, hi, n = (np.full(len(_RM16), np.nan) for _ in range(4))
        rng = np.random.default_rng(4)
        for i in range(len(_RM16)):
            m = np.isfinite(x) & np.isfinite(y) & (x >= _RB16[i]) & (x < _RB16[i + 1] + (1e-9 if i == len(_RM16) - 1 else 0))
            n[i] = m.sum()
            if m.sum() < 4:
                continue
            v = y[m]
            if how == "share":
                c[i], lo[i], hi[i] = wilson(float(np.nansum(v)), int(m.sum()))
            else:
                c[i] = np.nanmedian(v)
                bs = np.nanmedian(v[rng.integers(0, len(v), (min(P16_NBOOT, 300), len(v)))], axis=1)
                lo[i], hi[i] = np.nanpercentile(bs, 16), np.nanpercentile(bs, 84)
        return c, lo, hi, n

    _TRD16 = [("lag_sft", r"$t_{\rm SFT} - t_{\rm jet}$ [Gyr]", "how long the jet leg lasts", "median"),
              ("kappa_gas_anc", r"$\kappa_{\rm rot}$(gas) at the anchor", "C2: what the jet leg leaves rotating", "median"),
              ("dusty", "share of dusty QGs", "and whether it ends dusty", "share"),
              ("f_x_post", r"$f_{\rm X-ray}$ after the truncation", "how much X-ray it then takes", "median"),
              ("lfd_anc", r"$\log M_{\rm dust}/M_\star$ at the anchor", "the dust itself", "median")]
    _nb16 = len(_BX16); _nc16 = max(_nb16, 3)
    fig, axs = plt.subplots(2, _nc16, figsize=(3.7 * _nc16, 8.6), squeeze=False)
    for i in range(_nc16):
        ax = axs[0][i]
        if i >= _nb16:
            ax.axis("off"); continue
        b = _BX16[i]
        g = D16[(D16.box == b) & np.isfinite(D16.t_jet)].sort_values("t_jet")
        if not len(g):
            ax.axis("off"); continue
        _y16 = np.arange(len(g))
        _lo16 = np.nanmin(np.c_[g["t_jet"], g["t_sft"], g["t_x1"]], axis=1)
        _hi16 = np.nanmax(np.c_[g["t_jet"], g["t_sft"], g["t_x1"]], axis=1)
        ax.hlines(_y16, _lo16, _hi16, color="0.72", lw=0.4, alpha=0.7, rasterized=True)
        for _e in P16_ORDER:
            _mk, _c = _EMK16[_e]
            ax.scatter(g[_e].to_numpy(float), _y16, s=2.2 if len(g) > 800 else 8, marker=_mk, c=_c, lw=0,
                       label=_ESHORT16[_e], rasterized=True)
        ax.scatter(g["t_anchor"].to_numpy(float), _y16, s=1.4 if len(g) > 800 else 5, marker="|", c="0.35", lw=0.5,
                   label="anchor", rasterized=True)
        ax.set_xlabel("cosmic time [Gyr]"); ax.set_ylabel("branches, sorted by $t_{\\rm jet}$" if i == 0 else "")
        ax.set_title(f"{BOX_LABEL.get(b, b)}\n{len(g)} branches", fontsize=9.5)
        ax.set_ylim(-0.5, len(g) - 0.5); ax.grid(alpha=0.2, lw=0.5, axis="x")
        if i == 0:
            ax.legend(loc="lower right", fontsize=8, markerscale=2.6, framealpha=0.9)
    for i in range(_nc16):
        ax = axs[1][i]
        if i >= len(_TRD16):
            ax.axis("off"); continue
        col, ylab, ttl, how = _TRD16[i]
        for b in _BX16:
            g = D16[D16.box == b]
            if col not in g.columns or not np.isfinite(g[col].to_numpy(float)).any():
                continue
            c_, lo_, hi_, n_ = _rtrend16(g, col, how)
            ax.plot(_RM16, c_, "-o", ms=4, color=_BC16[b], lw=2.2 if b != NOX_BOX else 2.0,
                    ls="--" if b == NOX_BOX else "-", label=BOX_LABEL.get(b, b).split(" ")[0])
            ax.fill_between(_RM16, lo_, hi_, color=_BC16[b], alpha=0.16, lw=0)
            _push16(ROW16, "a-trend", _RM16, c_, lo_, hi_, n_, box=b, y=col)
        ax.set_xlabel("rank of $t_{\\rm jet}$   (0 = earliest for its epoch and mass)")
        ax.set_ylabel(ylab); ax.set_title(ttl, fontsize=9.5); ax.grid(alpha=0.22, lw=0.5)
        if i == 0:
            ax.legend(fontsize=8, loc="best")
    fig.suptitle("Part 16 (a) — THE ORDER OF EVENTS: the jet fires first, and how early it fires decides the wait to the "
                 "truncation, the rotation left at the anchor, and the dust", fontsize=11.5)
    fig.tight_layout(rect=(0, 0, 1, 0.955)); paper_save(fig, "kappa_timing_order"); plt.show()

    # ── (b) WHAT THE JET LEG DOES: the tracks on t - t_jet, and the rotation it leaves ──
    def _tband16(A, how="median", nt=None):
        """A (n_gal, n_t) stack -> the central track, its bootstrap 16-84 band OVER GALAXIES, and the per-epoch count."""
        A = np.asarray(A, float)
        nt = len(TG) if nt is None else int(nt)
        if A.ndim != 2 or not len(A):
            return (np.full(nt, np.nan),) * 3 + (np.zeros(nt, int),)
        n = np.isfinite(A).sum(axis=0)
        f = np.nanmean if how == "mean" else np.nanmedian
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", RuntimeWarning)
            c = f(A, axis=0)
            rng = np.random.default_rng(7)
            bs = np.array([f(A[rng.integers(0, len(A), len(A))], axis=0) for _ in range(min(P16_NBOOT, 300))])
            lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
        ok = n >= P16_NMIN
        return np.where(ok, c, np.nan), np.where(ok, lo, np.nan), np.where(ok, hi, np.nan), n

    print("\n" + "-" * 132)
    print("(b) THE JET LEG — C2: the jet removes the LOW-j gas, so after it fires the central kpc empties, R50(gas) GROWS,\n"
          "    Sigma_gas falls and what is left rotates.  d(...) is the anchor minus the 0.75 Gyr before the activation,\n"
          "    s(...) the slope over the 1.5 Gyr after it [per Gyr]; AUC = P(early jet > late jet).")
    print(f"  {'box':10s} {'quantity':16s} {'EARLY':>8s} {'LATE':>8s} {'D':>7s} {'AUC':>6s} {'p':>9s}   what it is")
    _B16Q = [("kappa_gas_jet", r"kappa at t_jet", "the rotation when the jet fires"),
             ("kappa_gas_anc", r"kappa at anchor", "... and at the anchor"),
             ("dkappa_gas_jet", r"d kappa", "THE CHANGE the jet leg leaves"),
             ("skappa_gas_jet", r"s kappa", "its slope in the first 1.5 Gyr"),
             ("dlr_gas_jet", r"d log R50(gas)", "the cold gas SPREADS if the centre is emptied"),
             ("dlsig_gas_jet", r"d log Sigma_gas", "... and thins"),
             ("lag_sft", r"the jet leg [Gyr]", "how long it runs before the truncation"),
             ("lfd_anc", r"log fdust anchor", "the outcome")]
    for b in _BX16:
        for col, lab, what in _B16Q:
            g = D16[D16.box == b]
            if col not in g.columns:
                continue
            e_, l_ = g[g.egrp == "early"][col].to_numpy(float), g[g.egrp == "late"][col].to_numpy(float)
            if np.isfinite(e_).sum() < 4 or np.isfinite(l_).sum() < 4:
                continue
            _au, _p, _, _ = mw_auc(g[col].to_numpy(float), (g.egrp == "early").to_numpy(bool))
            print(f"  {b:10s} {lab:16s} {np.nanmedian(e_):8.3f} {np.nanmedian(l_):8.3f} "
                  f"{np.nanmedian(e_) - np.nanmedian(l_):+7.3f} {_au:6.2f} {_p:9.1e}   {what}")
            ROW16.append(dict(block="b", box=b, y=col, early=float(np.nanmedian(e_)), late=float(np.nanmedian(l_)),
                              auc=_au, p=_p, n_early=int(np.isfinite(e_).sum()), n_late=int(np.isfinite(l_).sum())))
    print("  (C2 reads: d kappa > 0 and d log R50(gas) > 0 in both groups — the jet leg makes the rotation — and the EARLY\n"
          "   group ends higher in kappa. In " + (NOX_BOX if NOX_BOX in _BX16 else "a box with no X-ray channel")
          + " the jet still fires; what is missing is the X-ray kick that follows it.)")

    _nrj16 = len(P16_JTRK)
    fig, axs = plt.subplots(_nrj16, len(_BX16), figsize=(3.9 * len(_BX16), 2.35 * _nrj16), squeeze=False, sharex=True)
    for ci, b in enumerate(_BX16):
        _m16 = (D16["box"] == b).to_numpy(bool)
        for ri, q in enumerate(P16_JTRK):
            ax = axs[ri][ci]
            _how = "mean" if q == "xr" else "median"
            for e, st in _EG16.items():
                A = T16[q]["jet"][_m16 & (D16["egrp"] == e).to_numpy(bool)]
                c_, lo_, hi_, n_ = _tband16(A, _how)
                ax.plot(TG, c_, color=st["color"], lw=st["lw"], ls=st["ls"], label=st["label"] if (ri == 0 and ci == 0) else None)
                ax.fill_between(TG, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
            ax.axvline(0, color="0.35", lw=1.0, ls=":")
            if q == "kappa_gas":
                ax.axhline(KAPPA_ISO, color="0.5", lw=0.9, ls="--")
            ax.grid(alpha=0.2, lw=0.5)
            if ci == 0:
                ax.set_ylabel(_KLAB16.get(q, q), fontsize=9.5)
            if ri == 0:
                ax.set_title(BOX_LABEL.get(b, b), fontsize=9.5)
            if ri == _nrj16 - 1:
                ax.set_xlabel(r"$t - t_{\rm jet}$ [Gyr]")
            if ri == 0 and ci == 0:
                ax.legend(fontsize=8, loc="best")
    fig.suptitle("Part 16 (b) — THE JET LEG, on its own clock: the cold gas spreads and thins while the rotation rises — "
                 "what the jet takes is the low-$j$ gas", fontsize=11.5)
    fig.tight_layout(rect=(0, 0, 1, 0.965)); paper_save(fig, "kappa_timing_jetclock"); plt.show()

    # ── the CGM: PART 13's per-node census, re-clocked on this part's own t_SFT (so that P13_CLOCK cannot change what
    # this part means) and re-split on this part's own classes ──
    _MID16 = 0.5 * (P16_TB[:-1] + P16_TB[1:])
    N16 = pd.DataFrame()
    if P16_CGM and os.path.exists(P16_NODES):
        _n16 = pd.read_csv(P16_NODES)
        _k16 = D16[D16.box == P16_CGMBOX].set_index(["snap", "gal_id"])
        _cols16 = [c for c in ("t_sft", "t_jet", "t_x1", "t_anchor", "cls", "egrp", "dusty", "log_mstar", "lfd_anc",
                               "kappa_gas_anc") if c in _k16.columns]
        _n16 = _n16.merge(_k16[_cols16], left_on=["anchor_snap", "gal_id"], right_index=True, how="inner")
        if len(_n16):
            with np.errstate(all="ignore"):
                _n16["dt"] = _n16["t"].to_numpy(float) - _n16["t_sft"].to_numpy(float)
                _n16["dtj"] = _n16["t"].to_numpy(float) - _n16["t_jet"].to_numpy(float)
                _gi, _gc = _n16["mgas_ism"].to_numpy(float), _n16["mgas_cgm"].to_numpy(float)
                _di, _dc = _n16["mdust_ism"].to_numpy(float), _n16["mdust_cgm"].to_numpy(float)
                _ms16 = _n16["m_star_mem"].to_numpy(float)
                _n16["f_in1"] = _n16["mgas_in1"].to_numpy(float) / np.where(_gi > 0, _gi, np.nan)
                _n16["fd_in1"] = _n16["mdust_in1"].to_numpy(float) / np.where(_di > 0, _di, np.nan)
                _n16["lr50g"] = np.log10(np.where(_n16["r50_gasmem"].to_numpy(float) > 0, _n16["r50_gasmem"], np.nan))
                _n16["lmgas_ism"] = np.log10(np.where(_gi > 0, _gi, np.nan))
                _n16["lfd_ism"] = np.log10(np.where((_di > 0) & (_ms16 > 0), _di / _ms16, np.nan))
                _n16["lfd_cgm"] = np.log10(np.where((_dc > 0) & (_ms16 > 0), _dc / _ms16, np.nan))
                _n16["ldgr_cgm"] = np.log10(np.where((_dc > 0) & (_gc > 0), _dc / _gc, np.nan))
                _n16["ldgr_ism"] = np.log10(np.where((_di > 0) & (_gi > 0), _di / _gi, np.nan))
                _n16["fm_cold"] = _n16["mgas_cold"].to_numpy(float) / np.where(_gc > 0, _gc, np.nan)
                _n16["lm_cold"] = np.log10(np.clip(_n16["mgas_cold"].to_numpy(float), 1.0, None))
                _n16["lmgas_cgm"] = np.log10(np.where(_gc > 0, _gc, np.nan))
                _n16["lmdot_in_gc"] = np.log10(np.clip(_n16["mdot_in_gc"].to_numpy(float), 1e-3, None))
                _n16["lmdot_out_gc"] = np.log10(np.clip(_n16["mdot_out_gc"].to_numpy(float), 1e-3, None))
                _n16["lmdot_in_dc"] = np.log10(np.clip(_n16["mdot_in_dc"].to_numpy(float), 1e-8, None))
                _n16["f_out_dust"] = _n16["mdust_cgm"].to_numpy(float) / np.where((_dc + _di) > 0, _dc + _di, np.nan)
            _n16["_gkey"] = [f"{int(a_)}_{int(g_)}" for a_, g_ in zip(_n16["anchor_snap"], _n16["gal_id"])]
            N16 = _n16
        print(f"\n  the CGM census: {len(N16)} nodes of {N16['_gkey'].nunique() if len(N16) else 0} {P16_CGMBOX} branches "
              f"re-clocked on t - t_SFT (Part 13's {os.path.basename(P16_NODES)})"
              + (f";  classes: " + ", ".join(f"{_CLN16.get(k, k)} {v}" for k, v in
                                             N16.groupby('cls')['_gkey'].nunique().items()) if len(N16) else ""))
    elif P16_CGM:
        print(f"\n  Part 13 has not been run in this kernel — {P16_NODES} is missing, so (c) and (d) are skipped. It is the\n"
              "  part that measures the CGM from the reduced particle files; run it once and this part reads its table.")

    def _gmat16(d, col, tcol="dt"):
        """(n_gal, n_bin): the per-GALAXY median of `col` in each bin of the clock, so that a branch sampled at 40 snapshots
        does not outvote one sampled at 6."""
        if not len(d) or col not in d.columns:
            return np.zeros((0, len(_MID16))), np.array([])
        y, x, keys = d[col].to_numpy(float), d[tcol].to_numpy(float), d["_gkey"].to_numpy()
        b = np.digitize(x, P16_TB) - 1
        ok = np.isfinite(x) & np.isfinite(y) & (b >= 0) & (b < len(_MID16))
        if not ok.any():
            return np.zeros((0, len(_MID16))), np.array([])
        t = pd.DataFrame(dict(g=keys[ok], b=b[ok], y=y[ok])).groupby(["g", "b"])["y"].median()
        ug = np.array(sorted(set(keys[ok])))
        gi = {g_: i for i, g_ in enumerate(ug)}
        M = np.full((len(ug), len(_MID16)), np.nan)
        for (g_, b_), v_ in t.items():
            M[gi[g_], int(b_)] = v_
        return M, ug

    # ── FIG 3 — the same jet leg in the m25 PARTICLES: the central kpc, the size of the cold gas, and the dust in it ──
    if len(N16):
        _LOWJ16 = [("f_in1", r"$M_{\rm gas}(r < 1\,{\rm kpc})\,/\,M_{\rm gas}^{\rm ISM}$", "the low-$j$ gas: the central kpc"),
                   ("lr50g", r"$\log R_{50}$(gas) [kpc]", "the size of what is left"),
                   ("fd_in1", r"$M_{\rm dust}(r < 1\,{\rm kpc})\,/\,M_{\rm dust}^{\rm ISM}$", "the dust in the same centre"),
                   ("lmgas_ism", r"$\log M_{\rm gas}^{\rm ISM}$", "and the ISM as a whole")]
        fig, axs = plt.subplots(1, len(_LOWJ16), figsize=(3.6 * len(_LOWJ16), 3.5), squeeze=False)
        for i, (col, ylab, ttl) in enumerate(_LOWJ16):
            ax = axs[0][i]
            for e, st in _EG16.items():
                M, ug = _gmat16(N16[N16.egrp == e], col, "dtj")
                c_, lo_, hi_, n_ = _tband16(M, "median", len(_MID16))
                ax.plot(_MID16, c_, color=st["color"], lw=st["lw"], ls=st["ls"],
                        label=(st["label"] + f"  [{len(ug)}]") if i == 0 else None)
                ax.fill_between(_MID16, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
                _push16(ROW16, "b-lowj", _MID16, c_, lo_, hi_, n_, box=P16_CGMBOX, grp=e, y=col)
            ax.axvline(0, color="0.35", lw=1.0, ls=":")
            ax.set_xlabel(r"$t - t_{\rm jet}$ [Gyr]"); ax.set_ylabel(ylab, fontsize=9.5)
            ax.set_title(ttl, fontsize=9.5); ax.grid(alpha=0.22, lw=0.5)
            if i == 0:
                ax.legend(fontsize=7.5, loc="upper right", framealpha=0.9)
        fig.suptitle(f"Part 16 (b) — the jet leg in the {BOX_LABEL.get(P16_CGMBOX, P16_CGMBOX)} particles: the central kpc "
                     "empties while the cold gas spreads — the low-$j$ removal C2 rests on", fontsize=11.5)
        fig.tight_layout(rect=(0, 0, 1, 0.93)); paper_save(fig, "kappa_timing_lowj"); plt.show()

    # ── (c) THE CGM CYCLE: the three phases of the temperature, and what drives each of them ──
    if len(N16):
        _CQ16 = [("lT_cgm_mw", r"$\langle \log T \rangle_M$ of the CGM", "C3: the temperature, in three phases"),
                 ("fm_cold", r"$M_{\rm cold}/M_{\rm gas}$ (CGM)", "the cold phase — it must RISE as the ISM is displaced"),
                 ("lT_s10", r"$\langle \log T \rangle_M$, 0–10 kpc", "the inner halo, where the displaced ISM lands"),
                 ("lmdot_in_gc", r"$\log \dot M_{\rm in}$(cold) [M$_\odot$/yr]", "the cold inflow at the 10 kpc shell"),
                 ("lfd_cgm", r"$\log M_{\rm dust}^{\rm CGM}/M_\star$", "the dust that is outside"),
                 ("ldgr_cgm", r"$\log M_{\rm dust}/M_{\rm gas}$ (CGM)", "... and whether it survives out there")]
        _CLS16 = [c for c in (KC if P16_SPLIT == "kcls" else ["dusty", "pure"]) if (N16["cls"] == c).any()]
        fig, axs = plt.subplots(2, 3, figsize=(14.4, 7.6), squeeze=False)
        for i, (col, ylab, ttl) in enumerate(_CQ16):
            ax = axs[i // 3][i % 3]
            for _wi16, (_wn16, _lo16, _hi16) in enumerate(P16_WIN):   # C3's three windows, drawn behind everything
                ax.axvspan(_lo16, _hi16, color="0.82" if _wi16 % 2 else "0.93", alpha=0.5, lw=0, zorder=0)
                if i == 0:
                    ax.annotate(_wn16, (0.5 * (_lo16 + _hi16), 0.015), xycoords=("data", "axes fraction"), ha="center",
                                va="bottom", fontsize=7, color="0.3")
            for c in _CLS16:
                M, ug = _gmat16(N16[N16.cls == c], col, "dt")
                c_, lo_, hi_, n_ = _tband16(M, "median", len(_MID16))
                st = _CL16.get(c, dict(color="0.4", lw=2.0, ls="-"))
                ax.plot(_MID16, c_, color=st["color"], lw=st["lw"], ls=st["ls"],
                        label=(f"{_CLN16.get(c, c)}  [{len(ug)}]") if i == 0 else None)
                ax.fill_between(_MID16, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
                _push16(ROW16, "c", _MID16, c_, lo_, hi_, n_, box=P16_CGMBOX, grp=c, y=col)
            _X16, _ = _gmat16(N16, "xray_on", "dt") if "xray_on" in N16.columns else (np.zeros((0, len(_MID16))), None)
            if len(_X16):
                _ax2 = ax.twinx()
                _ax2.plot(_MID16, np.nanmean(_X16, axis=0), color="#1f77b4", lw=1.0, ls=":", alpha=0.8)
                _ax2.set_ylim(0, 1.05); _ax2.set_yticks([0, 1])
                _ax2.set_ylabel("X-ray on", fontsize=7.5, color="#1f77b4")
                _ax2.tick_params(axis="y", labelsize=7, colors="#1f77b4")
            ax.axvline(0, color="0.35", lw=1.0, ls=":")
            ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]"); ax.set_ylabel(ylab, fontsize=9.5)
            ax.set_title(ttl, fontsize=9.5); ax.grid(alpha=0.22, lw=0.5)
            if i == 0:
                ax.legend(fontsize=8, loc="best")
        fig.suptitle(f"Part 16 (c) — THE CGM CYCLE in {BOX_LABEL.get(P16_CGMBOX, P16_CGMBOX)}, on the truncation clock: "
                     "heated by its own star formation, cooled by the ISM it loses, heated again once the channel is on",
                     fontsize=11.5)
        fig.tight_layout(rect=(0, 0, 1, 0.955)); paper_save(fig, "kappa_cgm_cycle"); plt.show()

        # ── the drivers: one slope per galaxy in each window, against the thing C3 says makes it ──
        _G16 = []
        for _g16, _d16 in N16.groupby("_gkey"):
            _r16 = dict(gkey=_g16, cls=_d16["cls"].iloc[0], egrp=_d16["egrp"].iloc[0], dusty=bool(_d16["dusty"].iloc[0]),
                        n_node=len(_d16))
            _x16 = _d16["dt"].to_numpy(float)
            for _wn, _wlo, _whi in P16_WIN:
                _r16[f"sT_{_wn}"], _r16[f"nT_{_wn}"] = _slope16(_x16, _d16["lT_cgm_mw"].to_numpy(float), _wlo, _whi)
                _r16[f"sTin_{_wn}"], _ = _slope16(_x16, _d16["lT_s10"].to_numpy(float), _wlo, _whi)
                for _q16 in ("fm_cold", "lmgas_ism", "lm_cold", "lfd_cgm", "lmdot_in_gc", "ldgr_cgm"):
                    _a16 = _win16(_x16, _d16[_q16].to_numpy(float), _wlo, min(_wlo + 0.5, _whi))
                    _b16 = _win16(_x16, _d16[_q16].to_numpy(float), max(_whi - 0.5, _wlo), _whi)
                    _r16[f"d{_q16}_{_wn}"] = _b16 - _a16 if np.isfinite(_a16) and np.isfinite(_b16) else np.nan
                    _r16[f"{_q16}_{_wn}"] = _win16(_x16, _d16[_q16].to_numpy(float), _wlo, _whi)
            _G16.append(_r16)
        G16 = pd.DataFrame(_G16)
        # the branch's own star formation and X-ray occupancy in the same windows, from the chains (not the particles)
        _kk16 = D16[D16.box == P16_CGMBOX].copy()
        _kk16["_gkey"] = [f"{int(a_)}_{int(g_)}" for a_, g_ in zip(_kk16["snap"], _kk16["gal_id"])]
        _ix16 = {k_: i_ for i_, k_ in enumerate(_kk16["_gkey"])}
        _rows16 = np.array([_ix16.get(k_, -1) for k_ in G16["gkey"]])
        _sub16 = (D16["box"] == P16_CGMBOX).to_numpy(bool)
        for _wn, _wlo, _whi in P16_WIN:
            _mw16 = (TG >= _wlo) & (TG <= _whi)
            for _q16, _nm16 in (("lssfr", "lssfr"), ("xr", "f_xray"), ("lfgas", "lfgas"), ("lfd", "lfd")):
                _A16 = T16[_q16]["sft"][_sub16][:, _mw16]
                with warnings.catch_warnings():
                    warnings.simplefilter("ignore", RuntimeWarning)
                    _v16 = np.nanmean(_A16, axis=1) if _q16 == "xr" else np.nanmedian(_A16, axis=1)
                G16[f"{_nm16}_{_wn}"] = [(_v16[r_] if r_ >= 0 else np.nan) for r_ in _rows16]
        print("\n" + "-" * 132)
        print("(c) WHAT MOVES THE CGM TEMPERATURE — one slope per galaxy in each of C3's three windows [dex/Gyr], and the\n"
              "    Spearman of that slope against the thing the claim says makes it (the class is not controlled: it is\n"
              "    the same physics in both, and (d) is where they are compared).")
        print(f"  {'window':11s} {'N':>4s} {'slope T':>8s} {'inner':>7s} | {'against':41s} {'rho':>6s} {'p':>9s}")
        _DRV16 = {"before":    [("lssfr_before", "the sSFR of the window (the SN heating)"),
                                ("dlmgas_ism_before", "the ISM it is losing already")],
                  "displaced": [("dlmgas_ism_displaced", "the ISM MASS it loses (C3: it goes out)"),
                                ("dlm_cold_displaced", "the COLD CGM mass it gains"),
                                ("dfm_cold_displaced", "the cold SHARE of the CGM")],
                  "heated":    [("f_xray_heated", "the X-ray share of the window"),
                                ("dlmdot_in_gc_heated", "the cold inflow, as it shuts"),
                                ("dlfd_cgm_heated", "the CGM dust, as it goes")]}
        for _wn, _wlo, _whi in P16_WIN:
            _s16 = G16[f"sT_{_wn}"].to_numpy(float)
            for _c16, _lab16 in _DRV16.get(_wn, []):
                if _c16 not in G16.columns:
                    continue
                _r_, _p_, _n_ = _rho16(_s16, G16[_c16].to_numpy(float))
                print(f"  {_wn:11s} {_n_:4d} {np.nanmedian(_s16):8.3f} {np.nanmedian(G16[f'sTin_{_wn}']):7.3f} | "
                      f"{_lab16:41s} {_r_:+6.2f} {_p_:9.1e}")
                ROW16.append(dict(block="c-driver", box=P16_CGMBOX, window=_wn, x=_c16, rho=_r_, p=_p_, n=_n_,
                                  slope_T=float(np.nanmedian(_s16))))
        print("  (C3 predicts: a POSITIVE slope before the truncation that follows the sSFR; a NEGATIVE one just after it that\n"
              "   goes with the ISM leaving and the cold CGM growing — the dilution reading; and a POSITIVE one late that goes\n"
              "   with the X-ray share, with the cold inflow and the CGM dust falling as it does.)")

    # ── (d) THE DIVERGENCE TIME: how long the two classes stay the same after the truncation ──
    if len(N16) and len(_CLS16) >= 2:
        _A16n, _B16n = _CLS16[0], _CLS16[1]

        def _diff16(col):
            """The difference between the two classes bin by bin, with a bootstrap OVER GALAXIES inside each class, and the
            first epoch at or after the truncation where it holds its sign for P16_DIVSIG bins running."""
            MA, ua = _gmat16(N16[N16.cls == _A16n], col, "dt")
            MB, ub = _gmat16(N16[N16.cls == _B16n], col, "dt")
            d_, lo_, hi_ = (np.full(len(_MID16), np.nan) for _ in range(3))
            if not (len(MA) and len(MB)):
                return d_, lo_, hi_, np.nan, len(MA), len(MB)
            rng = np.random.default_rng(19)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", RuntimeWarning)
                nA, nB = np.isfinite(MA).sum(axis=0), np.isfinite(MB).sum(axis=0)
                d_ = np.nanmedian(MA, axis=0) - np.nanmedian(MB, axis=0)
                bs = np.array([np.nanmedian(MA[rng.integers(0, len(MA), len(MA))], axis=0)
                               - np.nanmedian(MB[rng.integers(0, len(MB), len(MB))], axis=0) for _ in range(P16_NBOOT)])
                lo_, hi_ = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
            _ok16 = (nA >= P16_NMIN) & (nB >= P16_NMIN)
            d_ = np.where(_ok16, d_, np.nan); lo_ = np.where(_ok16, lo_, np.nan); hi_ = np.where(_ok16, hi_, np.nan)
            _sig = np.isfinite(lo_) & np.isfinite(hi_) & ((lo_ > 0) | (hi_ < 0))
            t_div, _run = np.nan, 0
            for i in range(len(_MID16)):
                if _MID16[i] < 0:
                    continue
                _run = _run + 1 if _sig[i] else 0
                if _run >= P16_DIVSIG:
                    t_div = float(_MID16[i - P16_DIVSIG + 1]); break
            return d_, lo_, hi_, t_div, len(MA), len(MB)

        print("\n" + "-" * 132)
        print(f"(d) WHEN DO THEY PART? — the difference {_CLN16.get(_A16n, _A16n)} MINUS {_CLN16.get(_B16n, _B16n)}, bin by bin,\n"
              f"    with a bootstrap over galaxies; t_div is the first epoch at or after the truncation where the difference\n"
              f"    holds its sign at 68 % for {P16_DIVSIG} bins running (C4 says it should be near +1 Gyr, not at 0).")
        print(f"  {'quantity':16s} {'N':>9s} {'D before':>9s} {'D 0-1 Gyr':>10s} {'D 1-3 Gyr':>10s} {'t_div':>7s}   what it is")
        _DIVL16 = {"lmdot_in_gc": "the cold inflow", "lT_cgm_mw": "the CGM temperature", "fm_cold": "its cold share",
                   "lfd_cgm": "the dust outside", "ldgr_cgm": "the dust-to-gas out there", "lfd_ism": "the dust inside"}
        _DIV16 = {}
        for col in P16_DIVQ:
            if col not in N16.columns:
                continue
            d_, lo_, hi_, t_div, na, nb = _diff16(col)
            _wm = lambda a, b: float(np.nanmedian(d_[(_MID16 >= a) & (_MID16 <= b)])) if np.isfinite(d_).any() else np.nan
            _DIV16[col] = (d_, lo_, hi_, t_div)
            print(f"  {col:16s} {na:4d}/{nb:4d} {_wm(-3, -0.01):9.2f} {_wm(0, 1.0):10.2f} {_wm(1.0, 3.0):10.2f} "
                  f"{t_div:7.2f}   {_DIVL16.get(col, '')}")
            ROW16.append(dict(block="d", box=P16_CGMBOX, y=col, n_a=na, n_b=nb, t_div=t_div,
                              d_before=_wm(-3, -0.01), d_early=_wm(0, 1.0), d_late=_wm(1.0, 3.0)))
            _push16(ROW16, "d-track", _MID16, d_, lo_, hi_, np.full(len(_MID16), na + nb), box=P16_CGMBOX, y=col)
        _nd16 = len(_DIV16)
        if _nd16:
            _ncd16 = min(3, _nd16)
            _nrd16 = int(np.ceil(_nd16 / _ncd16))
            fig, axs = plt.subplots(_nrd16, _ncd16, figsize=(4.4 * _ncd16, 3.2 * _nrd16), squeeze=False)
            for i, (col, (d_, lo_, hi_, t_div)) in enumerate(_DIV16.items()):
                ax = axs[i // _ncd16][i % _ncd16]
                ax.axhline(0, color="0.3", lw=1.0)
                ax.axvline(0, color="0.35", lw=1.0, ls=":")
                ax.fill_between(_MID16, lo_, hi_, color="#4d4d9c", alpha=0.22, lw=0)
                ax.plot(_MID16, d_, "-o", ms=3.4, color="#2b2b6e", lw=2.0)
                if np.isfinite(t_div):
                    _dv16 = float(np.interp(t_div, _MID16, np.where(np.isfinite(d_), d_, 0.0)))
                    ax.axvline(t_div, color="#d95f02", lw=1.8, ls="--")
                    ax.annotate(rf"$t_{{\rm div}} = {t_div:+.2f}$ Gyr" + f"\n$\\Delta = {_dv16:+.2f}$ there",
                                (t_div, 0.94), xycoords=("data", "axes fraction"), ha="left", va="top", fontsize=8,
                                color="#d95f02", bbox=dict(fc="white", ec="none", alpha=0.75, pad=1.2))
                else:
                    ax.annotate("never parts", (0.02, 0.94), xycoords="axes fraction", ha="left", va="top",
                                fontsize=8.5, color="0.35")
                ax.set_title(_DIVL16.get(col, col), fontsize=9.5)
                ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]")
                ax.set_ylabel(rf"$\Delta$ {col}", fontsize=9)
                ax.grid(alpha=0.22, lw=0.5)
            for i in range(_nd16, _nrd16 * _ncd16):
                axs[i // _ncd16][i % _ncd16].axis("off")
            fig.suptitle(f"Part 16 (d) — {_CLN16.get(_A16n, _A16n)} MINUS {_CLN16.get(_B16n, _B16n)}: the two classes leave the "
                         "truncation together and part company about a Gyr later", fontsize=11.5)
            fig.tight_layout(rect=(0, 0, 1, 0.945)); paper_save(fig, "kappa_cgm_divergence"); plt.show()

    # ── (e) CAN THE DUST BE THE STAR FORMATION? the f_dust - sSFR plane, with the Gyr of dust production shaded ──
    print("\n" + "-" * 132)
    print(f"(e) THE DUST BUDGET — every galaxy of the catalogue cache with log M* > {MASS_MIN:g} on the f_dust - sSFR plane,\n"
          f"    against the dust its own star formation can have made in {P16_DT:g} Gyr:  f_dust^new = y_d x sSFR x dt, with\n"
          f"    y_d = {P16_YD[0]:.1e} to {P16_YD[1]:.1e} (SIMBA's condensation channel). ABOVE the band = the dust is older\n"
          "    than the window and has survived; INSIDE = the recent star formation is enough to explain it on its own.")
    _CATP = CAT[np.isfinite(CAT["mstar"].to_numpy(float)) & (CAT["mstar"].to_numpy(float) > 0)].copy()
    with np.errstate(all="ignore"):
        _msc = _CATP["mstar"].to_numpy(float); _sfc = np.maximum(_CATP["sfr"].to_numpy(float), 0.0)
        _CATP["lms"] = np.log10(_msc)
        _CATP["lssfr_p"] = np.where(_sfc > 0, np.log10(_sfc), np.nan)      # NaN = no star formation at all
        _fdc = _CATP["mdust"].to_numpy(float) #/ _msc
        _CATP["lfd_p"] = np.where(_fdc > 0, np.log10(np.clip(_fdc, 10 ** P9_FLOOR, None)), P9_FLOOR)
    _CATP = _CATP[_CATP["lms"].to_numpy(float) > MASS_MIN]
    _QK16 = set(zip(QM["box"], QM["snap"].astype(int), QM["gal_id"].astype(int)))
    _CATP["quenched"] = [(b_, int(s_), int(g_)) in _QK16 for b_, s_, g_ in
                         zip(_CATP["box"], _CATP["snap"], _CATP["gal_id"])]
    _yd16 = float(np.sqrt(P16_YD[0] * P16_YD[1]))                                  # the middle of the band, in the log
    _band16 = lambda x, y: np.log10(y) + np.asarray(x, float) + np.log10(P16_DT * 1e9)
    _XL16 = (-13.2, -8.2)
    print(f"  {'box':10s} {'sample':14s} {'N':>6s} {'above':>7s} {'inside':>7s} {'below':>7s} {'median excess':>14s} "
          f"{'... from the SFH':>17s}")
    for b in _BX16:
        _g16 = _CATP[_CATP["box"] == b]
        _d16 = D16[D16.box == b]
        for _nm16, _sel16, _dsel16 in (("all galaxies", np.ones(len(_g16), bool), None),
                                       ("star forming", _g16["lssfr_p"].to_numpy(float) > P16_SFCUT, None),
                                       ("quenched", _g16["quenched"].to_numpy(bool), np.ones(len(_d16), bool)),
                                       ("dusty QGs", _g16["quenched"].to_numpy(bool)
                                        & (_g16["lfd_p"].to_numpy(float) >= DUSTY_CUT),
                                        (_d16["dusty"].to_numpy(bool) if len(_d16) else None))):
            _s16 = _g16[_sel16]
            if not len(_s16):
                continue
            _x16, _y16 = _s16["lssfr_p"].to_numpy(float), _s16["lfd_p"].to_numpy(float)
            _hi16, _lo16 = _band16(_x16, P16_YD[1]), _band16(_x16, P16_YD[0])
            _hi16 = np.where(np.isfinite(_x16), _hi16, P9_FLOOR)                   # no star formation: it can make nothing
            _lo16 = np.where(np.isfinite(_x16), _lo16, P9_FLOOR)
            _ab = np.mean(_y16 > _hi16); _bl = np.mean(_y16 < _lo16)
            _nw16 = np.where(np.isfinite(_x16), np.maximum(_band16(_x16, _yd16), P9_FLOOR), P9_FLOOR)
            _ex16 = _y16 - _nw16
            _exs16 = np.nan
            if _dsel16 is not None and len(_d16):
                _dd16 = _d16[_dsel16]
                with np.errstate(all="ignore"):
                    _mk16 = (_dd16["dms_dt"].to_numpy(float) > 0) & (_dd16["m_star_anc"].to_numpy(float) > 0)
                    _new16 = np.where(_mk16, np.log10(np.clip(_yd16 * _dd16["dms_dt"].to_numpy(float)
                                                              / _dd16["m_star_anc"].to_numpy(float), 10 ** P9_FLOOR, None)),
                                      P9_FLOOR)
                    _exs16 = float(np.nanmedian(_dd16["lfd_anc"].to_numpy(float) - _new16))
            print(f"  {b:10s} {_nm16:14s} {len(_s16):6d} {100 * _ab:6.0f} % {100 * (1 - _ab - _bl):6.0f} % "
                  f"{100 * _bl:6.0f} % {np.nanmedian(_ex16):14.2f} "
                  + (f"{_exs16:17.2f}" if np.isfinite(_exs16) else f"{'-':>17s}"))
            ROW16.append(dict(block="e", box=b, sample=_nm16, n=len(_s16), f_above=float(_ab), f_below=float(_bl),
                              excess=float(np.nanmedian(_ex16)), excess_sfh=_exs16))
    print("  ('excess' is log M_dust/M* minus the middle of the band, in dex: 0 = exactly what the last Gyr of its own star\n"
          "   formation could have condensed. The last column repeats it with the mass the BRANCH actually formed in that Gyr\n"
          "   (its own SFH integrated), which needs no constant-sSFR assumption; a branch that formed nothing sits at the floor.)")

    _nc16e = len(_BX16)
    fig, axs = plt.subplots(2, _nc16e, figsize=(3.9 * _nc16e, 7.4), squeeze=False)
    _xg16 = np.linspace(_XL16[0], _XL16[1], 100)
    for i, b in enumerate(_BX16):
        ax = axs[0][i]
        _g16 = _CATP[_CATP["box"] == b]
        ax.fill_between(_xg16, _band16(_xg16, P16_YD[0]), _band16(_xg16, P16_YD[1]), color="#fdae61", alpha=0.45, lw=0,
                        zorder=0, label=f"made in {P16_DT:g} Gyr at that sSFR" if i == 0 else None)
        ax.plot(_xg16, _band16(_xg16, P16_YD[1]), color="#e6550d", lw=1.0, alpha=0.8)
        # nothing is dropped off the left edge: a galaxy with no star formation at all, or one below the axis, is drawn in
        # its own strip beside the plane (it can make NO dust in the window, so its budget is the floor)
        _xp16 = _g16["lssfr_p"].to_numpy(float)
        _off16 = ~np.isfinite(_xp16) | (_xp16 < _XL16[0])
        _rng16 = np.random.default_rng(2)
        _xp16 = np.where(_off16, _XL16[0] - 0.42 + 0.10 * _rng16.standard_normal(len(_xp16)), _xp16)
        _g16 = _g16.assign(_xp=_xp16)
        _q16 = _g16[_g16["quenched"].to_numpy(bool)]
        _s16 = _g16[~_g16["quenched"].to_numpy(bool)]
        _s16 = _s16.sample(P16_MAXPT, random_state=3) if len(_s16) > P16_MAXPT else _s16
        ax.scatter(_s16["_xp"], _s16["lfd_p"], s=3.0, c="0.72", lw=0, alpha=0.55, rasterized=True,
                   label="not quenched" if i == 0 else None)
        for _p16 in ("pure", "dusty"):
            _m16 = (_q16["lfd_p"].to_numpy(float) >= DUSTY_CUT) if _p16 == "dusty" else (_q16["lfd_p"].to_numpy(float) < DUSTY_CUT)
            ax.scatter(_q16["_xp"][_m16], _q16["lfd_p"][_m16], s=9, c=POPS[_p16]["color"], lw=0, alpha=0.75,
                       rasterized=True, label=(POPS[_p16]["label"].split("  (")[0] + ", quenched") if i == 0 else None)
        if _off16.any():
            ax.axvline(_XL16[0] - 0.2, color="0.5", lw=0.8, ls=":")
            ax.annotate(f"no SF, or\nbelow the axis\n{int(_off16.sum())}", (_XL16[0] - 0.42, 0.015),
                        xycoords=("data", "axes fraction"), ha="center", va="bottom", fontsize=6.5, color="0.35")
        _sfm16 = _g16["lssfr_p"].to_numpy(float) > P16_SFCUT                       # where the star-forming galaxies really sit
        if _sfm16.sum() > 20:
            _be16 = np.linspace(P16_SFCUT, min(_XL16[1], np.nanmax(_g16["lssfr_p"][_sfm16])), 6)
            _bm16 = 0.5 * (_be16[:-1] + _be16[1:])
            _md16 = [np.nanmedian(_g16["lfd_p"].to_numpy(float)[_sfm16 & (_g16["lssfr_p"].to_numpy(float) >= _be16[k])
                                                                & (_g16["lssfr_p"].to_numpy(float) < _be16[k + 1])])
                     for k in range(len(_bm16))]
            ax.plot(_bm16, _md16, "-k", lw=1.6, ls="--", label="the star-forming locus" if i == 0 else None)
        ax.axhline(DUSTY_CUT, color="0.35", lw=0.9, ls="-.")
        ax.set_xlim(_XL16[0] - 0.62, _XL16[1]); ax.set_ylim(P9_FLOOR - 0.2, -1.4)
        ax.set_xlabel(r"$\log$ sSFR [yr$^{-1}$]"); ax.set_ylabel(r"$\log M_{\rm dust}/M_\star$" if i == 0 else "")
        ax.set_title(BOX_LABEL.get(b, b), fontsize=9.5); ax.grid(alpha=0.2, lw=0.5)
        if i == 0:
            ax.legend(fontsize=7.5, loc="upper left", framealpha=0.9)
        ax = axs[1][i]
        _d16 = D16[D16.box == b]
        with np.errstate(all="ignore"):
            _mk16 = (_d16["dms_dt"].to_numpy(float) > 0) & (_d16["m_star_anc"].to_numpy(float) > 0)
            _new16 = np.where(_mk16, np.log10(np.clip(_yd16 * _d16["dms_dt"].to_numpy(float)
                                                      / _d16["m_star_anc"].to_numpy(float), 10 ** P9_FLOOR, None)), P9_FLOOR)
            _ex16 = _d16["lfd_anc"].to_numpy(float) - _new16                      # the floor bounds a branch that formed nothing
        _bins16 = np.linspace(-1.0, 5.0, 25)
        for _p16 in ("pure", "dusty"):
            _m16 = (_d16["dusty"].to_numpy(bool)) if _p16 == "dusty" else (~_d16["dusty"].to_numpy(bool))
            if _m16.sum():
                ax.hist(np.clip(_ex16[_m16], _bins16[0], _bins16[-1]), bins=_bins16, histtype="step", lw=2.0,
                        color=POPS[_p16]["color"], label=(POPS[_p16]["label"].split("  (")[0]) if i == 0 else None)
        ax.axvline(0, color="#e6550d", lw=2.0)
        ax.annotate("what its own\nlast Gyr can make", (0, 0.96), xycoords=("data", "axes fraction"), ha="center", va="top",
                    fontsize=7.5, color="#e6550d")
        ax.set_xlabel(r"$\log M_{\rm dust} - \log (y_{\rm d}\, \Delta M_\star)$   [dex above the budget]")
        ax.set_ylabel("quenched branches" if i == 0 else ""); ax.grid(alpha=0.22, lw=0.5)
        ax.set_title(f"the excess, from each branch's own SFH ({int(np.isfinite(_ex16).sum())})", fontsize=9)
        if i == 0:
            ax.legend(fontsize=7.5, loc="upper right")
    fig.suptitle("Part 16 (e) — is the dust just the star formation? " + P16_YDLAB, fontsize=11.5)
    fig.tight_layout(rect=(0, 0, 1, 0.955)); paper_save(fig, "kappa_dust_budget"); plt.show()

    # ── what leaves the notebook ──
    D16.to_csv(os.path.join(OUT, "kappa_timing_galaxies.csv"), index=False)
    pd.DataFrame(ROW16).to_csv(os.path.join(OUT, "kappa_timing.csv"), index=False)
    print(f"\ntables -> {os.path.join(OUT, 'kappa_timing.csv')} (every printed number and every drawn point),"
          f"\n         {os.path.join(OUT, 'kappa_timing_galaxies.csv')} (one row per branch: the three event times, the state "
          "at each of them, the jet-leg changes and the dust budget)")
    if "G16" in globals() and len(G16):
        G16.to_csv(os.path.join(OUT, "kappa_cgm_cycle.csv"), index=False)
        print(f"         {os.path.join(OUT, 'kappa_cgm_cycle.csv')} (one row per {P16_CGMBOX} branch: the three temperature "
              "slopes and what moves them)")
```

```python
# ─────────────────────────────────────────────────────────────────────
# Stellar dust budget in the M_dust – SFR plane
# ─────────────────────────────────────────────────────────────────────

# Generous stellar/SN dust yield; assume no destruction
YD_SN = 0.01#float(P16_YD[1])

# Reference production timescales [Gyr]
DT_LINES = [0.1, 1.0, 3.0, 6.0]


def stellar_dust_budget(sfr, dt_gyr):
    """
    Maximum dust mass produced by stars/SNe over dt_gyr,
    assuming constant SFR and no dust destruction.
    """
    return YD_SN * np.asarray(sfr) * dt_gyr * 1e9


# ── catalogue ────────────────────────────────────────────────────────

_CATP = CAT.copy()

mstar = _CATP["mstar"].to_numpy(float)
sfr   = _CATP["sfr"].to_numpy(float)
mdust = _CATP["mdust"].to_numpy(float)

valid = (
    np.isfinite(mstar) & (mstar > 0)
    & np.isfinite(mdust) & (mdust > 0)
)

_CATP = _CATP[valid].copy()

mstar = _CATP["mstar"].to_numpy(float)
sfr   = np.maximum(_CATP["sfr"].to_numpy(float), 0.0)
mdust = _CATP["mdust"].to_numpy(float)

_CATP["sfr_plot"] = sfr
_CATP["mdust_plot"] = mdust
_CATP["log_mstar"] = np.log10(mstar)
_CATP["log_fdust"] = np.log10(mdust / mstar)

_CATP = _CATP[_CATP["log_mstar"] > MASS_MIN].copy()


# ── quenched flag ────────────────────────────────────────────────────

_QK16 = set(
    zip(
        QM["box"],
        QM["snap"].astype(int),
        QM["gal_id"].astype(int),
    )
)

_CATP["quenched"] = [
    (b, int(s), int(g)) in _QK16
    for b, s, g in zip(
        _CATP["box"],
        _CATP["snap"],
        _CATP["gal_id"],
    )
]


# ── axis limits ──────────────────────────────────────────────────────

positive_sfr = _CATP.loc[_CATP["sfr_plot"] > 0, "sfr_plot"].to_numpy()
positive_md  = _CATP["mdust_plot"].to_numpy()

XLIM = (
    10**np.floor(np.log10(np.nanpercentile(positive_sfr, 0.5)) - 0.3),
    10**np.ceil (np.log10(np.nanpercentile(positive_sfr, 99.5)) + 0.3),
)

YLIM = (
    10**np.floor(np.log10(np.nanpercentile(positive_md, 0.5)) - 0.3),
    10**np.ceil (np.log10(np.nanpercentile(positive_md, 99.5)) + 0.3),
)

sfr_grid = np.logspace(
    np.log10(XLIM[0]),
    np.log10(XLIM[1]),
    300,
)


# ── figure ───────────────────────────────────────────────────────────

fig, axs = plt.subplots(
    1,
    len(_BX16),
    figsize=(4.0 * len(_BX16), 4.2),
    squeeze=False,
)

for i, box in enumerate(_BX16):

    ax = axs[0, i]
    g = _CATP[_CATP["box"] == box]


    # --------------------------------------------------------------
    # Region reachable by stellar production within 1 Gyr
    # --------------------------------------------------------------

    budget_1gyr = stellar_dust_budget(sfr_grid, 1.0)

    ax.fill_between(
        sfr_grid,
        YLIM[0],
        budget_1gyr,
        alpha=0.20,
        label=(
            r"stellar production within $1$ Gyr"
            if i == 0 else None
        ),
    )


    # --------------------------------------------------------------
    # Other characteristic production timescales
    # --------------------------------------------------------------

    for dt in DT_LINES:

        ax.plot(
            sfr_grid,
            stellar_dust_budget(sfr_grid, dt),
            lw=2.0 if dt == 1.0 else 1.0,
            ls="-" if dt == 1.0 else "--",
            alpha=1.0 if dt == 1.0 else 0.7,
            label=rf"${dt:g}\,$Gyr" if i == 0 else None,
        )


    # --------------------------------------------------------------
    # Non-quenched galaxies
    # --------------------------------------------------------------

    sf = g[
        (~g["quenched"])
        & (g["sfr_plot"] > 0)
    ]

    if len(sf) > P16_MAXPT:
        sf = sf.sample(P16_MAXPT, random_state=3)

    ax.scatter(
        sf["sfr_plot"],
        sf["mdust_plot"],
        s=4,
        c="0.72",
        lw=0,
        alpha=0.5,
        rasterized=True,
        label="not quenched" if i == 0 else None,
    )


    # --------------------------------------------------------------
    # Quenched galaxies
    # --------------------------------------------------------------

    q = g[
        g["quenched"]
        & (g["sfr_plot"] > 0)
    ]

    for pop in ("pure", "dusty"):

        if pop == "dusty":
            mask = q["log_fdust"] >= DUSTY_CUT
        else:
            mask = q["log_fdust"] < DUSTY_CUT

        ax.scatter(
            q.loc[mask, "sfr_plot"],
            q.loc[mask, "mdust_plot"],
            s=12,
            c=POPS[pop]["color"],
            lw=0,
            alpha=0.8,
            rasterized=True,
            label=(
                POPS[pop]["label"].split("  (")[0] + ", quenched"
                if i == 0 else None
            ),
        )


    # --------------------------------------------------------------
    # Cosmetics
    # --------------------------------------------------------------

    ax.set_xscale("log")
    ax.set_yscale("log")

    ax.set_xlim(XLIM)
    ax.set_ylim(YLIM)

    ax.set_xlabel(
        r"${\rm SFR}\ [{\rm M_\odot\,yr^{-1}}]$"
    )

    if i == 0:
        ax.set_ylabel(
            r"$M_{\rm dust}\ [{\rm M_\odot}]$"
        )

    ax.set_title(
        BOX_LABEL.get(box, box),
        fontsize=10,
    )

    ax.grid(False)

    if i == 0:
        ax.legend(
            fontsize=7.5,
            frameon=False,
        )


fig.suptitle(
    "Can stellar dust production explain the observed dust?",
    fontsize=11,
)

fig.tight_layout()

paper_save(
    fig,
    "kappa_dust_budget",
)

plt.show()
```

## Part 17 — the regrowth time, particle by particle: what the grains can do where they are

Part 16 measured **when** the CGM dust appears and goes. This part measures **whether it can**. SIMBA does not make or destroy
dust with a state: it does it grain by grain, at a rate set by the density, the temperature and the metallicity of the gas the
grain sits in — and every one of those is in the reduced particle files, which have carried `nh` and `zgas` since 2026-09-02
for exactly this.

**The growth time** (accretion of gas-phase metals onto grains; Hirashita 2000 / Asano+13, the form Li, Narayanan & Davé 2019
implement in SIMBA):

$$\tau_{\rm grow} \;=\; \tau_{\rm ref}\,\frac{a}{0.1\,\mu{\rm m}}\left(\frac{n_{\rm H}}{n_{\rm ref}}\right)^{-1}
\left(\frac{T}{T_{\rm ref}}\right)^{-1/2}\left(\frac{Z}{Z_\odot}\right)^{-1}$$

with `P17_TAUG` $= 25.4$ Myr at $n_{\rm ref} = 100$ cm$^{-3}$, $T_{\rm ref} = 20$ K and solar metallicity. A single grain's
growth is limited by the metals **still in the gas**, so the rate carried here is $\dot m_{\rm d} = m_{\rm d} f_Z/\tau_{\rm
grow}$ with $f_Z = Z m_{\rm gas}/(m_{\rm dust} + Z m_{\rm gas})$ — the same bounded dust-to-metal form Part 9 draws. Above
`P17_TGROW` the sticking of a metal atom onto a grain collapses, so the number is **formal** there: every growth column is the
cold gas of its zone only, and the hot-CGM growth panel is deliberately empty.

**The sputtering time** (thermal sputtering; Tsai & Mathews 1995, SIMBA's own, and the same law the old notebook's sputtering
part used):

$$\tau_{\rm sp} \;=\; 0.17\ {\rm Gyr}\ \frac{a}{0.1\,\mu{\rm m}}\ \frac{10^{-27}}{\rho}\left[\left(\frac{T_0}{T}\right)^{\omega} + 1\right],
\qquad T_0 = 2\times10^6\ {\rm K},\ \omega = 2.5,\ \rho = n_{\rm H} m_p / X_{\rm H}$$

$\tau_{\rm sp}$ is the timescale of the **radius**; the mass goes as $a^3$, so the mass e-folding is $\tau_{\rm sp}/3$ and that
is what the budget uses. Both timescales are **linear in the grain radius**, so `P17_AGR` shifts the whole distribution and
changes no comparison.

**One thing must not be read as physics.** For star-forming gas the temperature is the effective-equation-of-state temperature,
not a physical one, so its $\tau_{\rm sp}$ means nothing at all. The star-forming and the non-star-forming ISM are therefore
kept apart in every row of every table, exactly as the old notebook's sputtering part did.

| | what is asked | where |
|---|---|---|
| **(a)** | **the two clocks of a grain**: $\tau_{\rm grow}$ and $\tau_{\rm sp}$ of the dust that is *actually there*, dust-mass-weighted, in the ISM and in the CGM, on the truncation clock — the user's question, "the regrowth timescale in single particles in the ISM and the CGM", is this figure | FIG 1 |
| **(b)** | **is the CGM dust made out there, or brought?** the in-situ growth summed over the CGM particles, the sputtering against it, and the **measured** change of the CGM dust mass between consecutive nodes of the same branch. If growth falls orders of magnitude short of the measured change, the dust is transported, not made — and Part 13's flux through the shell says which way it went | FIG 2 |
| **(c)** | **the outcome**: does a galaxy end dusty because its dust sits where growth wins? every timescale and every environment it is made of, scored against the dust at the anchor, raw and with the axis ranked inside terciles of $\log M_\star$ | FIG 3 |

**What it reads.** The **same** reduced particle files Parts 10 and 13 read, keyed by **Part 13's node table** so that the
branch walk is not repeated — which means Part 13 must have been run for the m25 box first, and Part 16 must have left its
per-branch table (in the kernel, or as `kappa_timing_galaxies.csv`). It keeps its **own** cache, `kappa_<box>_dust_timescales.fits`,
one row per (snapshot, galaxy), appended to; delete it to re-measure. Where `nh` or `zgas` is missing from a file every
timescale is NaN and the cell prints the one-line backfill (`build_reduced_particles_job` appends only the missing dataset — no
geometry redo).

**The number to read first** is `time to make it` in (b): $\log M_{\rm dust}/\dot M_{\rm grow}$, how long the zone would need
to build the dust it holds at the rate it is growing at *now*. Above a Hubble time (1.14 in the log) it cannot have made it,
and the dust it holds came from somewhere else.

```python
# ── Part 17 — THE REGROWTH TIME, PARTICLE BY PARTICLE: where dust can grow, where it is sputtered, and whether the CGM
# makes its own dust or is refilled with somebody else's ──
# Part 16 measured WHEN the CGM dust appears and goes; this part measures WHETHER IT CAN. SIMBA does not destroy or make dust
# with a state: it does it grain by grain, at a rate set by the density, the temperature and the metallicity of the gas the
# grain sits in, and both rates are computable for every particle in the reduced files (they carry nh and zgas since
# 2026-09-02 for exactly this).
#
#   THE GROWTH TIME (accretion of gas-phase metals onto grains; Hirashita 2000 / Asano+13, the form Li, Narayanan & Dave
#   2019 implement in SIMBA):
#         tau_grow = P17_TAUG (a/0.1um) (n_H/n_ref)^-1 (T/T_ref)^-1/2 (Z/Zsun)^-1
#   ... and the growth of a single grain is limited by the metals still IN THE GAS, so the rate carried here is
#         dm_d/dt = m_d f_Z / tau_grow      with   f_Z = Z m_gas / (m_dust + Z m_gas)
#   the same bounded dust-to-metal form Part 9 draws. Above P17_TGROW the sticking of a metal atom onto a grain collapses,
#   so the number is FORMAL there and the growth statistics are reported for the cold gas only.
#
#   THE SPUTTERING TIME (thermal sputtering of the grain; Tsai & Mathews 1995, SIMBA's own):
#         tau_sp = P17_TAUS (a/0.1um) (1e-27/rho) [ (T_0/T)^omega + 1 ]      rho = n_H m_p / X_H
#   tau_sp is the timescale of the RADIUS; the mass goes as a^3, so the mass e-folding is tau_sp/3 and that is what the
#   budget uses. For STAR-FORMING gas the temperature is the effective-equation-of-state temperature and is not a physical
#   temperature at all, so its tau_sp means nothing — the star-forming and the non-star-forming ISM are kept apart
#   throughout, exactly as the old notebook's sputtering part did.
#
#   WHAT IS ASKED
#   (a) THE TWO CLOCKS: tau_grow and tau_sp of the dust that is actually there, in the ISM and in the CGM, on t - t_SFT.
#       The user's question — "the regrowth timescale in single particles in the ISM and the CGM" — is this figure.
#   (b) IS THE CGM DUST MADE OUT THERE, OR BROUGHT? The in-situ growth rate summed over the CGM particles, the sputtering
#       rate against it, and the MEASURED change of the CGM dust mass between consecutive nodes of the same branch. If the
#       growth term is orders of magnitude short of the measured change, the CGM dust is transported, not made — and Part
#       13's own flux through the shell says which way it went.
#   (c) THE OUTCOME: does a galaxy end dusty because its dust sits where growth wins? Per branch, the timescales in the
#       window after the truncation against the dust at the anchor, with the class and the mass beside them.
#
# It reads the SAME reduced particle files Parts 10 and 13 read, keyed by PART 13's node table (so the branch walk is not
# repeated here), and keeps its own cache: one row per (snap, gx), appended to, delete to rebuild.

P17_BOX    = P16_CGMBOX if "P16_CGMBOX" in globals() else "cis25"
P17_NODES  = P16_NODES if "P16_NODES" in globals() else os.path.join(OUT, "kappa_m25_cgm_nodes.csv")
P17_DIR    = os.path.join(os.getcwd(), "output", P17_BOX, "reduced_particles")
P17_PREFIX = {"cis25": "m25n512", "cis50": "m50n512", "cis50nox": "m50n512", "cis100": "m100n1024"}.get(P17_BOX, P17_BOX)
P17_CACHE  = os.path.join(OUT, f"kappa_{P17_BOX}_dust_timescales.fits")
P17_RMAX   = 100.0         # the CGM aperture [kpc] — the files' own
P17_TCOLD, P17_THOT = 1e5, 1e6                  # the phase boundaries, Part 13's
P17_TB     = P16_TB if "P16_TB" in globals() else np.arange(-3.0, 4.001, 0.5)
P17_SPLIT  = P16_SPLIT if "P16_SPLIT" in globals() else "kcls"
P17_NMIN, P17_NBOOT = 4, 300
P17_MAXNODE = None         # a cap on the files read (None = all) — set it to smoke-test the cell
P17_WIN    = (0.0, 2.5)    # the window after the truncation block (c) scores

# ── the grain physics, in one place ──
MP_G17, XH17 = 1.67262192e-24, 0.76             # the proton mass [g] and the hydrogen mass fraction behind rho = n_H m_p / X_H
P17_ZSUN   = 0.0134                             # Asplund+09
P17_TAUG   = 0.0254                             # [Gyr] the accretion time AT the reference point
P17_NREF, P17_TREF = 100.0, 20.0                # ... which is n_H = 100 cm^-3 and T = 20 K, at solar metallicity
P17_TAUS, P17_T0, P17_OMEGA = 0.17, 2.0e6, 2.5  # Tsai & Mathews (1995): the coefficient [Gyr], T_0 [K], the exponent
P17_AGR    = 1.0                                # the grain radius in units of 0.1 um — BOTH timescales are linear in it
P17_TGROW  = 1.0e5                              # [K] above this tau_grow is formal (the sticking coefficient collapses)
P17_FAST   = 1.0                                # [Gyr] "fast": the share of the dust whose timescale is shorter than this
P17_ZONES  = [("ism_nsf", "the ISM, not star forming"), ("ism_sf", "the ISM, star forming"),
              ("cgm", "the CGM"), ("cgm_cold", "the cold CGM"), ("cgm_hot", "the hot CGM")]

import glob as _glob17
import re as _re17
import time as _time17


def _taug17(nh, T, zg):
    """The grain-growth (metal accretion) time [Gyr] of a 0.1 um grain in gas of density n_H [cm^-3], temperature T [K] and
    metal mass fraction zg. Linear in the grain radius, so P17_AGR only shifts the whole distribution."""
    with np.errstate(all="ignore"):
        nh = np.asarray(nh, float); T = np.asarray(T, float); z = np.asarray(zg, float) / P17_ZSUN
        ok = (nh > 0) & (T > 0) & (z > 0)
        return np.where(ok, P17_AGR * P17_TAUG * (P17_NREF / np.where(nh > 0, nh, np.nan))
                        * np.sqrt(P17_TREF / np.where(T > 0, T, np.nan)) / np.where(z > 0, z, np.nan), np.nan)


def _taus17(nh, T):
    """The thermal-sputtering time of the grain RADIUS [Gyr] (Tsai & Mathews 1995). The MASS goes as a^3, so the mass
    e-folding is this over 3 — which is what the budget of (b) uses."""
    with np.errstate(all="ignore"):
        nh = np.asarray(nh, float); T = np.asarray(T, float)
        rho = nh * MP_G17 / XH17
        ok = (rho > 0) & (T > 0)
        return np.where(ok, P17_AGR * P17_TAUS * (1e-27 / np.where(rho > 0, rho, np.nan))
                        * ((P17_T0 / np.where(T > 0, T, np.nan)) ** P17_OMEGA + 1.0), np.nan)


def _wq17(v, w, q):
    """The weighted quantile of v with weights w (Part 13's, verbatim)."""
    v, w = np.asarray(v, float), np.asarray(w, float)
    ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
    if not ok.any():
        return np.nan
    v, w = v[ok], w[ok]
    o = np.argsort(v); v, w = v[o], w[o]
    return float(np.interp(q, (np.cumsum(w) - 0.5 * w) / w.sum(), v))


def _dust17(path):
    """ONE reduced particle file -> the growth and sputtering census of its dust, zone by zone."""
    with h5py.File(path, "r") as h:
        A = dict(h.attrs); G = h["gas"]
        r = np.linalg.norm(G["pos"][:].astype(float), axis=1)
        mem = G["member"][:].astype(bool)
        mg = G["m_gas"][:].astype(float); md = G["m_dust"][:].astype(float)
        T = G["temp"][:].astype(float) if "temp" in G else np.full(len(mg), np.nan)
        nh = G["nh"][:].astype(float) if "nh" in G else np.full(len(mg), np.nan)
        zg = G["zgas"][:].astype(float) if "zgas" in G else np.full(len(mg), np.nan)
        sfr = G["sfr"][:].astype(float) if "sfr" in G else np.zeros(len(mg))
    tg, ts = _taug17(nh, T, zg), _taus17(nh, T)
    with np.errstate(all="ignore"):
        mz = np.where(np.isfinite(zg), zg * mg, np.nan)                 # the metals still in the GAS phase
        fz = np.where((md + mz) > 0, mz / (md + mz), np.nan)            # ... as a share of all the metals: the room to grow
        grow = np.where(np.isfinite(tg) & (tg > 0), md * np.nan_to_num(fz) / tg, 0.0)          # [Msun / Gyr]
        sput = np.where(np.isfinite(ts) & (ts > 0), 3.0 * md / ts, 0.0)                        # the MASS rate
    cgm = (~mem) & (r <= P17_RMAX)
    ZM17 = {"ism_nsf": mem & (sfr <= 0), "ism_sf": mem & (sfr > 0), "cgm": cgm,
            "cgm_cold": cgm & (T < P17_TCOLD), "cgm_hot": cgm & (T >= P17_THOT)}
    row = dict(snap=int(A["snap"]), gx=int(A["gx"]), z=float(A["redshift"]), n_gas=int(len(mg)),
               has_nh=bool(np.isfinite(nh).any()), has_z=bool(np.isfinite(zg).any()))
    for zn, zm in ZM17.items():
        _md, _mg = md[zm], mg[zm]
        row[f"mdust_{zn}"] = float(np.nansum(_md)); row[f"mgas_{zn}"] = float(np.nansum(_mg))
        row[f"mmet_{zn}"] = float(np.nansum(mz[zm])); row[f"n_{zn}"] = int(zm.sum())
        _grow_ok = zm & (T < P17_TGROW)                                 # where growth is physical at all
        row[f"ltg_{zn}"] = np.log10(_wq17(tg[_grow_ok], md[_grow_ok], 0.5)) if _grow_ok.any() else np.nan
        row[f"lts_{zn}"] = np.log10(_wq17(ts[zm], _md, 0.5)) if zm.any() else np.nan
        row[f"lnh_{zn}"] = np.log10(_wq17(nh[zm], _md, 0.5)) if zm.any() else np.nan
        row[f"lT_{zn}"] = np.log10(_wq17(T[zm], _md, 0.5)) if zm.any() else np.nan
        row[f"fz_{zn}"] = _wq17(fz[zm], _md, 0.5) if zm.any() else np.nan
        _w = np.where(np.isfinite(_md), _md, 0.0)
        row[f"fgrow_{zn}"] = (float(_w[(tg[zm] < P17_FAST) & (T[zm] < P17_TGROW)].sum() / _w.sum())
                              if _w.sum() > 0 else np.nan)              # the share of the DUST that can regrow in P17_FAST
        row[f"fsput_{zn}"] = float(_w[ts[zm] < P17_FAST].sum() / _w.sum()) if _w.sum() > 0 else np.nan
        row[f"grow_{zn}"] = float(np.nansum(grow[_grow_ok]))            # [Msun/Gyr] the in-situ rates themselves
        row[f"sput_{zn}"] = float(np.nansum(sput[zm]))
    return row


print("=" * 132)
print(f"Part 17 — THE REGROWTH TIME, PARTICLE BY PARTICLE, in {BOX_LABEL.get(P17_BOX, P17_BOX)}:  "
      f"tau_grow = {P17_TAUG * 1e3:.1f} Myr (n_H/{P17_NREF:g})^-1 (T/{P17_TREF:g})^-1/2 (Z/Zsun)^-1,  "
      f"tau_sp = {P17_TAUS:g} Gyr (1e-27/rho) [(T_0/T)^{P17_OMEGA:g} + 1]")
print("=" * 132)

Z17 = pd.DataFrame()
if not os.path.exists(P17_NODES):
    print(f"  Part 13's node table is missing ({P17_NODES}) — it is what maps a particle file to a branch and an epoch.\n"
          "  Run Part 13 once (it builds the CGM census from the same files) and this part reads its table. Skipped.")
elif not os.path.isdir(P17_DIR):
    print(f"  no reduced particle files ({P17_DIR}) — this part measures the grains in them and cannot run without. Skipped.")
else:
    _N17 = pd.read_csv(P17_NODES)
    _S17 = D16 if "D16" in globals() and len(D16) else pd.read_csv(need(os.path.join(OUT, "kappa_timing_galaxies.csv"),
                                                                        "Part 16"))
    _S17 = _S17[_S17["box"] == P17_BOX]
    _c17 = [c for c in ("t_sft", "t_jet", "t_x1", "t_anchor", "cls", "dusty", "log_mstar", "lfd_anc", "kappa_gas_anc",
                        "egrp") if c in _S17.columns]
    if "cls" not in _c17 and P17_SPLIT in _S17.columns:
        _S17 = _S17.assign(cls=_S17[P17_SPLIT].astype(str)); _c17.append("cls")
    _N17 = _N17.merge(_S17.set_index(["snap", "gal_id"])[_c17], left_on=["anchor_snap", "gal_id"], right_index=True,
                      how="inner")
    _inv17 = {}
    for _p17 in _glob17.glob(os.path.join(P17_DIR, "snap_*", f"{P17_PREFIX}_snap*_gal*.h5")):
        _m17 = _re17.search(r"_snap(\d+)_gal(\d+)\.h5$", _p17)
        if _m17:
            _inv17[(int(_m17.group(1)), int(_m17.group(2)))] = _p17
    _N17 = _N17[[(int(s_), int(x_)) in _inv17 for s_, x_ in zip(_N17["snap"], _N17["gx"])]].reset_index(drop=True)
    _u17 = sorted({(int(s_), int(x_)) for s_, x_ in zip(_N17["snap"], _N17["gx"])})
    print(f"  {len(_inv17)} particle files on disk; {len(_N17)} are nodes of {_N17['gal_id'].nunique()} branches of Part 16's "
          f"table ({len(_u17)} distinct files)")
    C17 = pd.DataFrame()
    if os.path.exists(P17_CACHE):
        C17 = Table.read(P17_CACHE).to_pandas()
        print(f"  the census, cached -> {P17_CACHE}  ({len(C17)} files already measured)")
    _done17 = set(zip(C17["snap"].astype(int), C17["gx"].astype(int))) if len(C17) else set()
    _todo17 = [k for k in _u17 if k not in _done17]
    if P17_MAXNODE is not None:
        _todo17 = _todo17[:int(P17_MAXNODE)]
    if _todo17:
        print(f"  measuring {len(_todo17)} files (~{sum(os.path.getsize(_inv17[k]) for k in _todo17) / 2 ** 30:.1f} GiB to "
              "read; cached afterwards — this is a THIRD pass over the same files Parts 10 and 13 read, and it keeps its "
              "own cache so neither of theirs has to be rebuilt)", flush=True)
        _new17, _t017, _bad17 = [], _time17.time(), []
        for _i17, _k17 in enumerate(_todo17):
            try:
                _new17.append(_dust17(_inv17[_k17]))
            except Exception as _e17:
                _bad17.append(f"{os.path.basename(_inv17[_k17])}: {type(_e17).__name__}: {_e17}")
            if (_i17 + 1) % 200 == 0:
                _el17 = _time17.time() - _t017
                print(f"     {_i17 + 1}/{len(_todo17)} files, {_el17:.0f} s "
                      f"(~{_el17 / (_i17 + 1) * (len(_todo17) - _i17 - 1):.0f} s left)", flush=True)
        for _b17 in _bad17[:5]:
            print(f"     file not read, {_b17}")
        if _new17:
            C17 = pd.concat([C17, pd.DataFrame(_new17)], ignore_index=True)
            Table.from_pandas(C17).write(P17_CACHE, overwrite=True)
            print(f"  the census -> {P17_CACHE}  ({len(C17)} files, {_time17.time() - _t017:.0f} s)")
    if len(C17):
        _nnh17 = float(np.mean(C17["has_nh"].to_numpy(bool))) if "has_nh" in C17.columns else 0.0
        _nz17 = float(np.mean(C17["has_z"].to_numpy(bool))) if "has_z" in C17.columns else 0.0
        print(f"  the fields the grain physics needs: nh in {100 * _nnh17:.0f} % of the files, zgas in {100 * _nz17:.0f} %"
              + ("" if min(_nnh17, _nz17) > 0.99 else
                 "\n     !! where they are missing every timescale is NaN. They are backfilled by re-submitting the SAME "
                 "plan\n        (build_reduced_particles_job adds only the missing dataset — no geometry redo):\n"
                 "        sbatch --array=0-15 submit_reduced_particles.sh <the plan this box was built with>"))
        Z17 = _N17.merge(C17, on=["snap", "gx"], how="inner", suffixes=("", "_c17"))
        Z17["dt"] = Z17["t"].to_numpy(float) - Z17["t_sft"].to_numpy(float)
        Z17["_gkey"] = [f"{int(a_)}_{int(g_)}" for a_, g_ in zip(Z17["anchor_snap"], Z17["gal_id"])]
        with np.errstate(all="ignore"):
            for _zn17, _ in P17_ZONES:
                Z17[f"lmdust_{_zn17}"] = np.log10(np.clip(Z17[f"mdust_{_zn17}"].to_numpy(float), 1.0, None))
                Z17[f"lgrow_{_zn17}"] = np.log10(np.clip(Z17[f"grow_{_zn17}"].to_numpy(float), 1e-4, None))
                Z17[f"lsput_{_zn17}"] = np.log10(np.clip(Z17[f"sput_{_zn17}"].to_numpy(float), 1e-4, None))
                _md17 = Z17[f"mdust_{_zn17}"].to_numpy(float); _gr17 = Z17[f"grow_{_zn17}"].to_numpy(float)
                Z17[f"tmake_{_zn17}"] = np.log10(np.where((_gr17 > 0) & (_md17 > 0), _md17 / _gr17, np.nan))
        print(f"  {len(Z17)} nodes of {Z17['_gkey'].nunique()} branches carry both the census and the clock")

    if len(Z17):
        _MID17 = 0.5 * (P17_TB[:-1] + P17_TB[1:])
        _CL17 = ({k: dict(color=v["color"], lw=v["lw"], ls=v.get("ls", "-"), short=v.get("short", k)) for k, v in KCLS.items()}
                 if P17_SPLIT == "kcls" else
                 {k: dict(color=v["color"], lw=v["lw"], ls=v.get("ls", "-"),
                          short=("dusty QG" if k == "dusty" else "dust-poor QG")) for k, v in POPS.items()})
        _CK17 = [c for c in (KC if P17_SPLIT == "kcls" else ["dusty", "pure"]) if (Z17["cls"] == c).any()]

        def _gmat17(d, col):
            """(n_gal, n_bin) of the per-GALAXY median of `col` in each bin of t - t_SFT."""
            if not len(d) or col not in d.columns:
                return np.zeros((0, len(_MID17))), np.array([])
            y, x, keys = d[col].to_numpy(float), d["dt"].to_numpy(float), d["_gkey"].to_numpy()
            b = np.digitize(x, P17_TB) - 1
            ok = np.isfinite(x) & np.isfinite(y) & (b >= 0) & (b < len(_MID17))
            if not ok.any():
                return np.zeros((0, len(_MID17))), np.array([])
            t_ = pd.DataFrame(dict(g=keys[ok], b=b[ok], y=y[ok])).groupby(["g", "b"])["y"].median()
            ug = np.array(sorted(set(keys[ok])))
            gi = {g_: i for i, g_ in enumerate(ug)}
            M = np.full((len(ug), len(_MID17)), np.nan)
            for (g_, b_), v_ in t_.items():
                M[gi[g_], int(b_)] = v_
            return M, ug

        def _tband17(A):
            A = np.asarray(A, float)
            if A.ndim != 2 or not len(A):
                return (np.full(len(_MID17), np.nan),) * 3 + (np.zeros(len(_MID17), int),)
            n = np.isfinite(A).sum(axis=0)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", RuntimeWarning)
                c = np.nanmedian(A, axis=0)
                rng = np.random.default_rng(23)
                bs = np.array([np.nanmedian(A[rng.integers(0, len(A), len(A))], axis=0) for _ in range(P17_NBOOT)])
                lo, hi = np.nanpercentile(bs, 16, axis=0), np.nanpercentile(bs, 84, axis=0)
            ok = n >= P17_NMIN
            return np.where(ok, c, np.nan), np.where(ok, lo, np.nan), np.where(ok, hi, np.nan), n

        ROW17 = []
        print("\n" + "-" * 132)
        print("(a) THE TWO CLOCKS OF A GRAIN — the dust-mass-weighted median of each timescale in each zone, over the window\n"
              f"    {P17_WIN[0]:+.1f} to {P17_WIN[1]:+.1f} Gyr of the truncation [log Gyr], with the share of the dust whose "
              f"timescale is under {P17_FAST:g} Gyr beside it.")
        print(f"  {'zone':26s} {'N':>5s} {'log tau_grow':>12s} {'can regrow':>11s} | {'log tau_sp':>11s} {'sputtered':>10s} | "
              f"{'log n_H':>8s} {'log T':>7s} {'f_Z in gas':>10s}")
        _w17 = (Z17["dt"].to_numpy(float) >= P17_WIN[0]) & (Z17["dt"].to_numpy(float) <= P17_WIN[1])
        for zn, zlab in P17_ZONES:
            _d17 = Z17[_w17 & (Z17[f"n_{zn}"].to_numpy(float) > 0)]
            if not len(_d17):
                continue
            _m = lambda c: float(np.nanmedian(_d17[c].to_numpy(float))) if c in _d17.columns else np.nan
            print(f"  {zlab:26s} {len(_d17):5d} {_m('ltg_' + zn):12.2f} {100 * _m('fgrow_' + zn):9.0f} % | "
                  f"{_m('lts_' + zn):11.2f} {100 * _m('fsput_' + zn):8.0f} % | {_m('lnh_' + zn):8.2f} {_m('lT_' + zn):7.2f} "
                  f"{_m('fz_' + zn):10.2f}")
            ROW17.append(dict(block="a", zone=zn, n=len(_d17), ltg=_m("ltg_" + zn), lts=_m("lts_" + zn),
                              fgrow=_m("fgrow_" + zn), fsput=_m("fsput_" + zn), lnh=_m("lnh_" + zn), lT=_m("lT_" + zn),
                              fz=_m("fz_" + zn)))
        print(f"  (a timescale under 0 in the log is under a Gyr. 'f_Z in gas' is the share of the metals that are NOT yet in\n"
              f"   grains — the room the growth term still has. Above {P17_TGROW:.0e} K the growth number is formal and is not\n"
              "   counted: the growth columns are the COLD gas of that zone only, the sputtering columns are all of it.)")

        _ZD17 = [z for z in ("ism_nsf", "ism_sf", "cgm_cold", "cgm_hot") if (Z17[f"n_{z}"].to_numpy(float) > 0).any()]
        _RW17 = [("ltg_{}", r"$\log \tau_{\rm grow}$ [Gyr]", "the regrowth time of the dust that is there"),
                 ("lts_{}", r"$\log \tau_{\rm sp}$ [Gyr]", "the sputtering time of the same grains"),
                 ("lmdust_{}", r"$\log M_{\rm dust}$ [M$_\odot$]", "the dust in the zone")]
        fig, axs = plt.subplots(len(_RW17), len(_ZD17), figsize=(3.4 * len(_ZD17), 2.5 * len(_RW17)), squeeze=False,
                                sharex=True)
        for ci, zn in enumerate(_ZD17):
            for ri, (cf, ylab, ttl) in enumerate(_RW17):
                ax = axs[ri][ci]
                for c in _CK17:
                    M, ug = _gmat17(Z17[Z17.cls == c], cf.format(zn))
                    c_, lo_, hi_, n_ = _tband17(M)
                    st = _CL17.get(c, dict(color="0.4", lw=2.0, ls="-"))
                    ax.plot(_MID17, c_, color=st["color"], lw=st["lw"], ls=st["ls"],
                            label=(f"{st['short']}  [{len(ug)}]") if (ri == 0 and ci == 0) else None)
                    ax.fill_between(_MID17, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
                    for _q17 in range(len(_MID17)):
                        if np.isfinite(c_[_q17]):
                            ROW17.append(dict(block="a-track", zone=zn, grp=c, y=cf.format(zn), t=float(_MID17[_q17]),
                                              val=float(c_[_q17]), lo=float(lo_[_q17]), hi=float(hi_[_q17]),
                                              n=int(n_[_q17])))
                if cf.startswith("lt"):
                    ax.axhline(0.0, color="0.45", lw=0.9, ls="--")          # one Gyr
                    ax.axhline(np.log10(13.8), color="0.65", lw=0.9, ls=":")  # a Hubble time
                if not np.isfinite(Z17[cf.format(zn)].to_numpy(float)).any():
                    ax.annotate(f"not defined here\n($T > ${P17_TGROW:.0e} K)" if cf.startswith("ltg") else "not measured",
                                (0.5, 0.5), xycoords="axes fraction", ha="center", va="center", fontsize=8, color="0.45")
                ax.axvline(0, color="0.35", lw=1.0, ls=":")
                ax.grid(alpha=0.22, lw=0.5)
                if ci == 0:
                    ax.set_ylabel(ylab, fontsize=9.5)
                if ri == 0:
                    ax.set_title(dict(P17_ZONES).get(zn, zn), fontsize=9.5)
                if ri == len(_RW17) - 1:
                    ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]")
                if ri == 0 and ci == 0:
                    ax.legend(fontsize=8, loc="best")
        fig.suptitle(f"Part 17 (a) — the two clocks of a grain in {BOX_LABEL.get(P17_BOX, P17_BOX)}: how long the dust that is "
                     "there needs to regrow, and how long it has left  (dashed = 1 Gyr, dotted = a Hubble time)", fontsize=11.5)
        fig.tight_layout(rect=(0, 0, 1, 0.955)); paper_save(fig, "kappa_dust_timescales"); plt.show()

        # ── (b) IS THE CGM DUST MADE OUT THERE, OR BROUGHT? the in-situ rates against the measured change ──
        for _zn17 in ("cgm", "cgm_cold", "ism_nsf", "ism_sf"):
            Z17[f"dmdt_{_zn17}"] = np.nan
        for _g17, _d17 in Z17.groupby("_gkey"):
            _o17 = np.argsort(_d17["t"].to_numpy(float))
            _idx17 = _d17.index.to_numpy()[_o17]
            _tt17 = _d17["t"].to_numpy(float)[_o17]
            for _zn17 in ("cgm", "cgm_cold", "ism_nsf", "ism_sf"):
                _mm17 = _d17[f"mdust_{_zn17}"].to_numpy(float)[_o17]
                if len(_tt17) >= 2:
                    _r17 = np.full(len(_tt17), np.nan)
                    _r17[1:] = np.diff(_mm17) / np.clip(np.diff(_tt17), 1e-6, None)
                    Z17.loc[_idx17, f"dmdt_{_zn17}"] = _r17
        with np.errstate(all="ignore"):
            for _zn17 in ("cgm", "cgm_cold", "ism_nsf", "ism_sf"):
                Z17[f"ldmdt_{_zn17}"] = np.log10(np.clip(np.abs(Z17[f"dmdt_{_zn17}"].to_numpy(float)), 1e-4, None))
        print("\n" + "-" * 132)
        print("(b) MADE OUT THERE, OR BROUGHT? — the in-situ rates summed over the particles of a zone against the change of\n"
              f"    its dust mass actually measured between consecutive nodes of the same branch, in {P17_WIN[0]:+.1f} to "
              f"{P17_WIN[1]:+.1f} Gyr of the truncation [log M_sun/Gyr].")
        print(f"  {'zone':26s} {'N':>5s} {'growth':>8s} {'sputtering':>11s} {'|measured|':>11s} {'grow/|meas|':>12s} "
              f"{'log M_dust':>11s} {'time to make it':>16s}")
        for zn, zlab in P17_ZONES:
            if f"ldmdt_{zn}" not in Z17.columns:
                continue
            _d17 = Z17[_w17 & np.isfinite(Z17[f"dmdt_{zn}"].to_numpy(float))]
            if not len(_d17):
                continue
            _m = lambda c: float(np.nanmedian(_d17[c].to_numpy(float)))
            _rt17 = float(np.nanmedian(np.log10(np.clip(_d17[f"grow_{zn}"].to_numpy(float), 1e-6, None)
                                                / np.clip(np.abs(_d17[f"dmdt_{zn}"].to_numpy(float)), 1e-6, None))))
            print(f"  {zlab:26s} {len(_d17):5d} {_m('lgrow_' + zn):8.2f} {_m('lsput_' + zn):11.2f} "
                  f"{_m('ldmdt_' + zn):11.2f} {_rt17:12.2f} {_m('lmdust_' + zn):11.2f} {_m('tmake_' + zn):16.2f}")
            ROW17.append(dict(block="b", zone=zn, n=len(_d17), lgrow=_m("lgrow_" + zn), lsput=_m("lsput_" + zn),
                              ldmdt=_m("ldmdt_" + zn), lratio=_rt17, lmdust=_m("lmdust_" + zn), tmake=_m("tmake_" + zn)))
        print("  ('time to make it' = log M_dust / growth [Gyr]: how long the zone would need to build the dust it holds at\n"
              "   the rate it is growing at right now. Above a Hubble time (1.14) it cannot have made it — the dust was\n"
              "   brought there. 'grow/|meas|' near or above 0 means the in-situ term is big enough to explain the change;\n"
              "   far below 0 means transport does it, and Part 13's flux through the shell says which way.)")

        _BQ17 = [("cgm_cold", "the cold CGM"), ("ism_nsf", "the ISM, not star forming")]
        fig, axs = plt.subplots(2, len(_BQ17) + 1, figsize=(4.0 * (len(_BQ17) + 1), 6.6), squeeze=False)
        for ci, (zn, zlab) in enumerate(_BQ17):
            ax = axs[0][ci]
            for _cf17, _lab17, _co17 in ((f"lgrow_{zn}", "in-situ growth", "#1a9850"),
                                         (f"lsput_{zn}", "sputtering", "#d73027"),
                                         (f"ldmdt_{zn}", "|measured change|", "#2b2b6e")):
                M, ug = _gmat17(Z17, _cf17)
                c_, lo_, hi_, n_ = _tband17(M)
                ax.plot(_MID17, c_, color=_co17, lw=2.2, label=_lab17 if ci == 0 else None)
                ax.fill_between(_MID17, lo_, hi_, color=_co17, alpha=0.16, lw=0)
            ax.axvline(0, color="0.35", lw=1.0, ls=":")
            ax.set_title(zlab, fontsize=9.5); ax.grid(alpha=0.22, lw=0.5)
            ax.set_ylabel(r"$\log$ rate [M$_\odot$ Gyr$^{-1}$]" if ci == 0 else "", fontsize=9.5)
            ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]")
            if ci == 0:
                ax.legend(fontsize=8, loc="best")
            ax = axs[1][ci]
            for c in _CK17:
                M, ug = _gmat17(Z17[Z17.cls == c], f"tmake_{zn}")
                c_, lo_, hi_, n_ = _tband17(M)
                st = _CL17.get(c, dict(color="0.4", lw=2.0, ls="-"))
                ax.plot(_MID17, c_, color=st["color"], lw=st["lw"], ls=st["ls"], label=st["short"] if ci == 0 else None)
                ax.fill_between(_MID17, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
            ax.axhline(np.log10(13.8), color="0.5", lw=1.0, ls=":")
            ax.annotate("a Hubble time", (0.02, np.log10(13.8)), xycoords=("axes fraction", "data"), va="bottom",
                        fontsize=7.5, color="0.35")
            ax.axvline(0, color="0.35", lw=1.0, ls=":")
            ax.set_title("the time it would need to make its own dust", fontsize=9)
            ax.set_ylabel(r"$\log\,M_{\rm dust}/\dot M_{\rm grow}$ [Gyr]" if ci == 0 else "", fontsize=9.5)
            ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]"); ax.grid(alpha=0.22, lw=0.5)
            if ci == 0:
                ax.legend(fontsize=8, loc="best")
        ax = axs[0][len(_BQ17)]
        _s17 = Z17[_w17 & np.isfinite(Z17["dmdt_cgm"].to_numpy(float))]
        _x17 = np.log10(np.clip(_s17["grow_cgm_cold"].to_numpy(float), 1e-3, None))
        _y17 = np.log10(np.clip(np.abs(_s17["dmdt_cgm"].to_numpy(float)), 1e-3, None))
        for c in _CK17:
            _m17 = (_s17["cls"] == c).to_numpy(bool)
            st = _CL17.get(c, dict(color="0.4"))
            ax.scatter(_x17[_m17], _y17[_m17], s=7, c=st["color"], lw=0, alpha=0.6, rasterized=True, label=st["short"])
        if np.isfinite(np.r_[_x17, _y17]).any():
            _lim17 = [np.nanmin(np.r_[_x17, _y17]) - 0.2, np.nanmax(np.r_[_x17, _y17]) + 0.2]
            ax.plot(_lim17, _lim17, "-k", lw=1.2)
            ax.set_xlim(*_lim17); ax.set_ylim(*_lim17)
        else:
            ax.annotate("no node carries both a rate and a measured change", (0.5, 0.5), xycoords="axes fraction",
                        ha="center", va="center", fontsize=8, color="0.4")
        ax.set_xlabel(r"$\log \dot M_{\rm grow}$, cold CGM [M$_\odot$/Gyr]")
        ax.set_ylabel(r"$\log |\Delta M_{\rm dust}^{\rm CGM} / \Delta t|$")
        ax.set_title("what it can make against what it did", fontsize=9.5); ax.grid(alpha=0.22, lw=0.5)
        ax.legend(fontsize=8, loc="best")
        ax = axs[1][len(_BQ17)]
        for c in _CK17:
            M, ug = _gmat17(Z17[Z17.cls == c], "fgrow_cgm_cold")
            c_, lo_, hi_, n_ = _tband17(M)
            st = _CL17.get(c, dict(color="0.4", lw=2.0, ls="-"))
            ax.plot(_MID17, c_, color=st["color"], lw=st["lw"], ls=st["ls"], label=st["short"])
            ax.fill_between(_MID17, lo_, hi_, color=st["color"], alpha=0.17, lw=0)
        ax.axvline(0, color="0.35", lw=1.0, ls=":")
        ax.set_title(rf"the share of the cold-CGM dust that can regrow in {P17_FAST:g} Gyr", fontsize=9)
        ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]"); ax.set_ylabel("mass fraction", fontsize=9.5)
        ax.grid(alpha=0.22, lw=0.5); ax.legend(fontsize=8, loc="best")
        fig.suptitle("Part 17 (b) — the dust budget of the halo: what the grains out there can make for themselves, against "
                     "what the CGM's dust mass actually does", fontsize=11.5)
        fig.tight_layout(rect=(0, 0, 1, 0.955)); paper_save(fig, "kappa_dust_budget_cgm"); plt.show()

        # ── (c) THE OUTCOME: does a galaxy end dusty because its dust sits where growth wins? ──
        _G17rows = []
        for _g17, _d17 in Z17[_w17].groupby("_gkey"):
            _r17 = dict(gkey=_g17, cls=_d17["cls"].iloc[0], n_node=len(_d17),
                        dusty=bool(_d17["dusty"].iloc[0]) if "dusty" in _d17.columns else False,
                        log_mstar=float(_d17["log_mstar"].iloc[0]) if "log_mstar" in _d17.columns else np.nan,
                        lfd_anc=float(_d17["lfd_anc"].iloc[0]) if "lfd_anc" in _d17.columns else np.nan)
            for zn, _ in P17_ZONES:
                for c in ("ltg", "lts", "fgrow", "fsput", "lnh", "lT", "fz", "tmake", "lmdust"):
                    _cn = f"{c}_{zn}"
                    _r17[_cn] = (float(np.nanmedian(_d17[_cn].to_numpy(float)))
                                 if _cn in _d17.columns and np.isfinite(_d17[_cn].to_numpy(float)).any() else np.nan)
            _G17rows.append(_r17)
        G17 = pd.DataFrame(_G17rows)

        def _aucm17(v, m1, mass, nq=3):
            """The AUC, and the same AUC with the axis ranked inside terciles of log M* (both axes carry the mass)."""
            v = np.asarray(v, float); m1 = np.asarray(m1, bool); mass = np.asarray(mass, float)
            a_, p_, n1, n0 = mw_auc(v, m1)
            r = np.full(len(v), np.nan)
            if np.isfinite(mass).sum() > 3 * nq:
                e = np.unique(np.nanquantile(mass, np.linspace(0, 1, nq + 1)))
                lab = np.where(np.isfinite(mass), np.clip(np.digitize(mass, e[1:-1]), 0, max(len(e) - 2, 0)), -1)
                for c in np.unique(lab[lab >= 0]):
                    m = (lab == c) & np.isfinite(v)
                    if m.sum() > 1:
                        r[m] = (np.argsort(np.argsort(v[m])) + 0.5) / m.sum()
            am_, _, _, _ = mw_auc(r, m1)
            return a_, p_, am_, n1, n0

        print("\n" + "-" * 132)
        print("(c) THE OUTCOME — every timescale and every environment it is made of, scored against the dust the galaxy ends\n"
              f"    with (median over {P17_WIN[0]:+.1f} to {P17_WIN[1]:+.1f} Gyr of the truncation). AUC = P(the axis is higher "
              "in a dusty QG);\n    'AUC | M*' is the same with the axis ranked inside terciles of log M*.")
        print(f"  {'zone':26s} {'quantity':30s} {'N':>4s} {'dusty':>6s} {'dust-poor':>10s} {'AUC':>6s} {'AUC | M*':>9s} "
              f"{'p':>9s}")
        _QC17 = [("ltg", "log tau_grow of the dust"), ("lts", "log tau_sp of the dust"),
                 ("fgrow", f"share regrowing in {P17_FAST:g} Gyr"), ("fsput", f"share sputtered in {P17_FAST:g} Gyr"),
                 ("lnh", "log n_H the dust sits in"), ("lT", "log T the dust sits in"),
                 ("fz", "the metals still in the gas"), ("tmake", "log time to make its own dust")]
        _dm17 = G17["dusty"].to_numpy(bool)
        for zn, zlab in P17_ZONES:
            for c, clab in _QC17:
                _cn = f"{c}_{zn}"
                if _cn not in G17.columns or np.isfinite(G17[_cn].to_numpy(float)).sum() < 8:
                    continue
                _v17 = G17[_cn].to_numpy(float)
                if np.nanstd(_v17) == 0:
                    continue                                  # a column that never varies scores nothing
                _a17, _p17, _am17, _n1, _n0 = _aucm17(_v17, _dm17, G17["log_mstar"].to_numpy(float))
                if not np.isfinite(_a17):
                    continue
                print(f"  {zlab:26s} {clab:30s} {int(np.isfinite(_v17).sum()):4d} {np.nanmedian(_v17[_dm17]):6.2f} "
                      f"{np.nanmedian(_v17[~_dm17]):10.2f} {_a17:6.2f} {_am17:9.2f} {_p17:9.1e}")
                ROW17.append(dict(block="c", zone=zn, y=_cn, n=int(np.isfinite(_v17).sum()),
                                  dusty=float(np.nanmedian(_v17[_dm17])), pure=float(np.nanmedian(_v17[~_dm17])),
                                  auc=_a17, auc_m=_am17, p=_p17))

        _PL17 = [("ltg_ism_nsf", r"$\log \tau_{\rm grow}$, the non-SF ISM"), ("lnh_ism_nsf", r"$\log n_{\rm H}$ of that dust"),
                 ("ltg_cgm_cold", r"$\log \tau_{\rm grow}$, the cold CGM")]
        _PL17 = [(c, l) for c, l in _PL17 if c in G17.columns and np.isfinite(G17[c].to_numpy(float)).sum() >= 8]
        if _PL17:
            fig, axs = plt.subplots(1, len(_PL17), figsize=(4.1 * len(_PL17), 3.6), squeeze=False)
            for i, (col, xlab) in enumerate(_PL17):
                ax = axs[0][i]
                for _p16n, _m17 in (("dusty", _dm17), ("pure", ~_dm17)):
                    ax.scatter(G17[col].to_numpy(float)[_m17], G17["lfd_anc"].to_numpy(float)[_m17], s=22,
                               c=POPS[_p16n]["color"], lw=0, alpha=0.75,
                               label=(POPS[_p16n]["label"].split("  (")[0]) if i == 0 else None)
                _r17, _pv17, _n17 = ((lambda a: (a.statistic, a.pvalue, int(np.isfinite(G17[col]).sum())))(
                    spearmanr(G17[col].to_numpy(float)[np.isfinite(G17[col].to_numpy(float))],
                              G17["lfd_anc"].to_numpy(float)[np.isfinite(G17[col].to_numpy(float))]))
                    if np.isfinite(G17[col].to_numpy(float)).sum() > 6 else (np.nan, np.nan, 0))
                ax.axhline(DUSTY_CUT, color="0.35", lw=0.9, ls="-.")
                ax.set_xlabel(xlab); ax.set_ylabel(r"$\log M_{\rm dust}/M_\star$ at the anchor" if i == 0 else "")
                ax.set_title(rf"$\rho = {_r17:+.2f}$  ($p = {_pv17:.1g}$, $N = {_n17}$)", fontsize=9.5)
                ax.grid(alpha=0.22, lw=0.5)
                if i == 0:
                    ax.legend(fontsize=8, loc="best")
            fig.suptitle("Part 17 (c) — the outcome against the environment the grains sit in: is a dusty QG dusty because its "
                         "dust can regrow?", fontsize=11.5)
            fig.tight_layout(rect=(0, 0, 1, 0.93)); paper_save(fig, "kappa_dust_outcome"); plt.show()

        G17.to_csv(os.path.join(OUT, f"kappa_{P17_BOX}_dust_timescales_galaxies.csv"), index=False)
        pd.DataFrame(ROW17).to_csv(os.path.join(OUT, "kappa_dust_timescales.csv"), index=False)
        print(f"\ntables -> {os.path.join(OUT, 'kappa_dust_timescales.csv')} (every printed number and drawn point),"
              f"\n         {os.path.join(OUT, f'kappa_{P17_BOX}_dust_timescales_galaxies.csv')} (one row per branch),"
              f"\n         {P17_CACHE} (the per-file census — delete it to re-measure)")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# PART 17 — WHERE IS THE GAS CAPABLE OF GROWING DUST AFTER QUENCHING?
#
# Scientific question:
#
#   Independently of whether dust is already present, what fraction of the
#   available gas / metal reservoir lies in conditions where grain growth is
#   fast enough to operate before the final (anchor) epoch?
#
# For every galaxy and every branch snapshot we compute tau_grow from:
#
#       n_H, T, Z
#
# for ALL eligible gas particles, not only dust-bearing ones.
#
# For each particle we compare:
#
#       tau_grow < t_available(snapshot)
#
# where:
#
#       t_available(snapshot) = t_anchor - t_snapshot
#
# i.e. the time remaining for grain growth before the final selected epoch.
#
# We then ask, separately in the non-SF ISM and CGM:
#
#   f_Z,growable =
#       metal mass in particles with tau_grow < t_available
#       ---------------------------------------------------
#                     total metal mass
#
# This is the MAIN diagnostic.
#
# We also compute:
#
#   f_gas,growable =
#       gas mass in particles with tau_grow < t_available
#       -------------------------------------------------
#                         total gas mass
#
# as a robustness check.
#
# Finally, each galaxy is represented by the time-integrated "opportunity"
# over its post-quenching branch:
#
#       <f_growable>_time
#
# calculated as a time-weighted mean over 0 < t-t_SFT < P17_WIN[1].
#
# This avoids conditioning on where dust already happens to survive.
# ─────────────────────────────────────────────────────────────────────────────

import os
import glob
import re
import time

import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy.table import Table
from scipy.stats import spearmanr


# ── configuration ────────────────────────────────────────────────────────────

P17_BOX = (
    P16_CGMBOX
    if "P16_CGMBOX" in globals()
    else "cis25"
)

P17_NODES = (
    P16_NODES
    if "P16_NODES" in globals()
    else os.path.join(
        OUT,
        "kappa_m25_cgm_nodes.csv",
    )
)

P17_DIR = os.path.join(
    os.getcwd(),
    "output",
    P17_BOX,
    "reduced_particles",
)

P17_PREFIX = {
    "cis25": "m25n512",
    "cis50": "m50n512",
    "cis50nox": "m50n512",
    "cis100": "m100n1024",
}.get(P17_BOX, P17_BOX)

P17_CACHE = os.path.join(
    OUT,
    f"kappa_{P17_BOX}_growable_gas.fits",
)

# CGM radial aperture.
P17_RMAX = 100.0  # kpc

# Post-quenching interval included in the galaxy-level summary.
P17_WIN = (0.0, 2.5)  # Gyr

# Growth prescription should not be interpreted in hot gas.
P17_TGROW_MAX = 1e5  # K

# Minimum number of eligible particles required to quote a snapshot fraction.
P17_NMIN = 5

# Optional debugging limiter.
P17_MAXNODE = None


# ── grain-growth prescription ────────────────────────────────────────────────

P17_ZSUN = 0.0134

P17_TAUG = 0.0254     # Gyr
P17_NREF = 100.0      # cm^-3
P17_TREF = 20.0       # K
P17_AGR  = 1.0        # a / 0.1 micron


def _taug17(nh, T, zg):
    """
    Grain-growth timescale [Gyr].

        tau_grow = TAUG
                   (a/0.1um)
                   (nH/NREF)^-1
                   (T/TREF)^-1/2
                   (Z/Zsun)^-1

    This depends only on the physical gas state and does NOT require dust
    to already be present.
    """
    nh = np.asarray(nh, float)
    T  = np.asarray(T, float)
    zg = np.asarray(zg, float)

    out = np.full_like(nh, np.nan, dtype=float)

    with np.errstate(
        divide="ignore",
        invalid="ignore",
        over="ignore",
    ):
        zrel = zg / P17_ZSUN

        ok = (
            np.isfinite(nh)
            & np.isfinite(T)
            & np.isfinite(zrel)
            & (nh > 0)
            & (T > 0)
            & (zrel > 0)
        )

        out[ok] = (
            P17_AGR
            * P17_TAUG
            * (P17_NREF / nh[ok])
            * np.sqrt(P17_TREF / T[ok])
            / zrel[ok]
        )

    return out


# ── utility functions ────────────────────────────────────────────────────────

def _weighted_fraction(weight, valid, good):
    """
    Weighted fraction of a valid reservoir satisfying 'good'.

    Returns NaN if fewer than P17_NMIN valid particles or if the total
    positive weight is zero.
    """
    valid = np.asarray(valid, bool)
    good  = np.asarray(good, bool)
    weight = np.asarray(weight, float)

    valid = (
        valid
        & np.isfinite(weight)
        & (weight > 0)
    )

    if valid.sum() < P17_NMIN:
        return np.nan

    denom = np.sum(weight[valid])

    if denom <= 0:
        return np.nan

    return float(
        np.sum(weight[valid & good])
        / denom
    )


def _number_fraction(valid, good):
    """
    Simple particle-number fraction, kept only as an auxiliary diagnostic.
    """
    valid = np.asarray(valid, bool)
    good  = np.asarray(good, bool)

    n = int(valid.sum())

    if n < P17_NMIN:
        return np.nan

    return float(
        np.sum(valid & good)
        / n
    )


def _weighted_quantile(v, w, q=0.5):
    """
    Weighted quantile.
    """
    v = np.asarray(v, float)
    w = np.asarray(w, float)

    ok = (
        np.isfinite(v)
        & np.isfinite(w)
        & (w > 0)
    )

    if not np.any(ok):
        return np.nan

    v = v[ok]
    w = w[ok]

    order = np.argsort(v)
    v = v[order]
    w = w[order]

    cdf = (
        np.cumsum(w)
        - 0.5 * w
    ) / np.sum(w)

    return float(
        np.interp(q, cdf, v)
    )


def _time_weighted_mean(t, y):
    """
    Time-weighted mean of y(t) using trapezoidal integration:

        <y> = ∫ y dt / ∫ dt

    This gives each time interval weight according to its duration instead of
    simply giving every snapshot equal weight.
    """
    t = np.asarray(t, float)
    y = np.asarray(y, float)

    ok = (
        np.isfinite(t)
        & np.isfinite(y)
    )

    if ok.sum() < 2:
        return np.nan

    t = t[ok]
    y = y[ok]

    order = np.argsort(t)
    t = t[order]
    y = y[order]

    # Remove duplicate times.
    keep = np.r_[
        True,
        np.diff(t) > 0,
    ]

    t = t[keep]
    y = y[keep]

    if len(t) < 2:
        return np.nan

    dt_total = t[-1] - t[0]

    if dt_total <= 0:
        return np.nan

    return float(
        np.trapz(y, t)
        / dt_total
    )


# ─────────────────────────────────────────────────────────────────────────────
# ONE PARTICLE FILE
# ─────────────────────────────────────────────────────────────────────────────

def _growable17(path):
    """
    Measure the physical capacity for grain growth in one branch node.

    IMPORTANT:
    - no Mdust > 0 requirement is imposed;
    - the denominator is all eligible gas / metals in the zone;
    - star-forming ISM is excluded because its effective-EOS temperature
      should not be interpreted as a physical thermodynamic temperature.

    The criterion tau_grow < t_available cannot yet be applied here because
    t_available is branch-dependent and comes from the node table.  Therefore
    this cache stores the particle quantities needed to reconstruct several
    growth-time thresholds later.

    To avoid saving every particle, we store weighted cumulative information
    on a fixed tau_grow grid.
    """

    with h5py.File(path, "r") as h:

        A = dict(h.attrs)
        G = h["gas"]

        pos = G["pos"][:].astype(float)
        r = np.linalg.norm(
            pos,
            axis=1,
        )

        mem = G["member"][:].astype(bool)

        mg = G["m_gas"][:].astype(float)

        T = (
            G["temp"][:].astype(float)
            if "temp" in G
            else np.full(len(mg), np.nan)
        )

        nh = (
            G["nh"][:].astype(float)
            if "nh" in G
            else np.full(len(mg), np.nan)
        )

        zg = (
            G["zgas"][:].astype(float)
            if "zgas" in G
            else np.full(len(mg), np.nan)
        )

        sfr = (
            G["sfr"][:].astype(float)
            if "sfr" in G
            else np.zeros(len(mg))
        )

    tg = _taug17(
        nh,
        T,
        zg,
    )

    # Gas-phase metal mass.
    mZ = zg * mg

    zones = {
        "ism": (
            mem
            & (sfr <= 0)
        ),
        "cgm": (
            (~mem)
            & (r <= P17_RMAX)
        ),
    }

    row = {
        "snap": int(A["snap"]),
        "gx": int(A["gx"]),
        "z": float(A["redshift"]),
    }

    # Fixed thresholds [Gyr].
    #
    # t_available will generally fall in this range.  Storing a reasonably
    # fine CDF grid keeps the particle cache compact while allowing us to
    # evaluate f(tau_grow < t_available) later for each branch node.
    tau_grid = np.logspace(
        -4,
        1.0,
        151,
    )

    for zn, zone in zones.items():

        # Eligible gas for grain growth:
        #
        #   - physically meaningful nH, T, Z
        #   - T < 1e5 K
        #
        # Critically: NO requirement on current dust mass.
        eligible = (
            zone
            & np.isfinite(tg)
            & (tg > 0)
            & np.isfinite(mg)
            & (mg > 0)
            & np.isfinite(mZ)
            & (mZ > 0)
            & np.isfinite(T)
            & (T < P17_TGROW_MAX)
        )

        row[f"n_eligible_{zn}"] = int(
            eligible.sum()
        )

        row[f"mgas_eligible_{zn}"] = float(
            np.nansum(
                mg[eligible]
            )
        )

        row[f"mZ_eligible_{zn}"] = float(
            np.nansum(
                mZ[eligible]
            )
        )

        row[f"tg_med_Z_{zn}"] = (
            _weighted_quantile(
                tg[eligible],
                mZ[eligible],
                0.5,
            )
            if eligible.any()
            else np.nan
        )

        row[f"tg_med_gas_{zn}"] = (
            _weighted_quantile(
                tg[eligible],
                mg[eligible],
                0.5,
            )
            if eligible.any()
            else np.nan
        )

        # Weighted CDF evaluated on the fixed tau grid.
        #
        # These columns allow us to interpolate the fraction below any
        # galaxy-specific t_available later.
        if eligible.sum() >= P17_NMIN:

            tge = tg[eligible]
            mge = mg[eligible]
            mZe = mZ[eligible]

            gas_tot = np.sum(mge)
            z_tot = np.sum(mZe)
            n_tot = len(tge)

            for j, th in enumerate(tau_grid):

                good = tge < th

                row[f"cdfZ_{zn}_{j:03d}"] = (
                    float(
                        np.sum(mZe[good])
                        / z_tot
                    )
                    if z_tot > 0
                    else np.nan
                )

                row[f"cdfG_{zn}_{j:03d}"] = (
                    float(
                        np.sum(mge[good])
                        / gas_tot
                    )
                    if gas_tot > 0
                    else np.nan
                )

                row[f"cdfN_{zn}_{j:03d}"] = (
                    float(
                        np.sum(good)
                        / n_tot
                    )
                    if n_tot > 0
                    else np.nan
                )

        else:

            for j in range(len(tau_grid)):
                row[f"cdfZ_{zn}_{j:03d}"] = np.nan
                row[f"cdfG_{zn}_{j:03d}"] = np.nan
                row[f"cdfN_{zn}_{j:03d}"] = np.nan

    return row


# Same grid used by the cache.
P17_TAU_GRID = np.logspace(
    -4,
    1.0,
    151,
)


def _interp_cdf17(row, zone, prefix, t_available):
    """
    Interpolate cached weighted CDF at a galaxy-specific t_available.

    prefix:
        cdfZ = metal-mass weighted
        cdfG = gas-mass weighted
        cdfN = particle-number weighted
    """
    if not np.isfinite(t_available) or t_available <= 0:
        return np.nan

    vals = np.array([
        row.get(
            f"{prefix}_{zone}_{j:03d}",
            np.nan,
        )
        for j in range(
            len(P17_TAU_GRID)
        )
    ], dtype=float)

    good = np.isfinite(vals)

    if good.sum() < 2:
        return np.nan

    # CDF should be monotonic, but numerical noise / missing values can
    # occasionally spoil this.  Enforce monotonicity conservatively.
    x = P17_TAU_GRID[good]
    y = np.maximum.accumulate(
        vals[good]
    )

    if t_available <= x[0]:
        return float(y[0])

    if t_available >= x[-1]:
        return float(y[-1])

    return float(
        np.interp(
            np.log10(t_available),
            np.log10(x),
            y,
        )
    )


# ─────────────────────────────────────────────────────────────────────────────
# LOAD BRANCH TABLE + PARTICLE CACHE
# ─────────────────────────────────────────────────────────────────────────────

print("=" * 105)
print("Part 17 — GROWABLE GAS / METAL FRACTION AFTER QUENCHING")
print("=" * 105)

Z17 = pd.DataFrame()
G17 = pd.DataFrame()

if not os.path.exists(P17_NODES):

    print(
        f"Missing node table: {P17_NODES}"
    )

elif not os.path.isdir(P17_DIR):

    print(
        f"Missing particle directory: {P17_DIR}"
    )

else:

    N17 = pd.read_csv(
        P17_NODES
    )

    S17 = (
        D16.copy()
        if "D16" in globals()
        and len(D16)
        else pd.read_csv(
            need(
                os.path.join(
                    OUT,
                    "kappa_timing_galaxies.csv",
                ),
                "Part 16",
            )
        )
    )

    S17 = S17[
        S17["box"] == P17_BOX
    ].copy()

    meta_cols = [
        c for c in (
            "t_sft",
            "t_anchor",
            "cls",
            "dusty",
            "log_mstar",
            "lfd_anc",
            "kappa_gas_anc",
            "egrp",
        )
        if c in S17.columns
    ]

    if "cls" not in meta_cols:

        split_col = (
            P16_SPLIT
            if "P16_SPLIT" in globals()
            else "kcls"
        )

        if split_col in S17.columns:

            S17["cls"] = (
                S17[split_col]
                .astype(str)
            )

            meta_cols.append(
                "cls"
            )

    N17 = N17.merge(
        S17.set_index(
            ["snap", "gal_id"]
        )[meta_cols],
        left_on=[
            "anchor_snap",
            "gal_id",
        ],
        right_index=True,
        how="inner",
    )

    # ── inventory particle files ─────────────────────────────────────────────

    inv17 = {}

    pattern = os.path.join(
        P17_DIR,
        "snap_*",
        f"{P17_PREFIX}_snap*_gal*.h5",
    )

    for p in glob.glob(pattern):

        m = re.search(
            r"_snap(\d+)_gal(\d+)\.h5$",
            p,
        )

        if m:

            inv17[
                (
                    int(m.group(1)),
                    int(m.group(2)),
                )
            ] = p

    have = np.array([
        (
            int(s),
            int(gx),
        )
        in inv17
        for s, gx in zip(
            N17["snap"],
            N17["gx"],
        )
    ])

    N17 = (
        N17[have]
        .reset_index(drop=True)
    )

    nodes = sorted({
        (
            int(s),
            int(gx),
        )
        for s, gx in zip(
            N17["snap"],
            N17["gx"],
        )
    })

    print(
        f"{len(inv17)} particle files on disk; "
        f"{len(nodes)} unique branch nodes; "
        f"{N17['gal_id'].nunique()} galaxies."
    )

    # ── cache ────────────────────────────────────────────────────────────────

    C17 = pd.DataFrame()

    if os.path.exists(
        P17_CACHE
    ):

        C17 = (
            Table.read(
                P17_CACHE
            )
            .to_pandas()
        )

        print(
            f"Loaded cache: "
            f"{P17_CACHE} "
            f"({len(C17)} nodes)"
        )

    done = (
        set(
            zip(
                C17["snap"].astype(int),
                C17["gx"].astype(int),
            )
        )
        if len(C17)
        else set()
    )

    todo = [
        key
        for key in nodes
        if key not in done
    ]

    if P17_MAXNODE is not None:

        todo = todo[
            :int(P17_MAXNODE)
        ]

    if todo:

        print(
            f"Measuring {len(todo)} "
            f"new particle files...",
            flush=True,
        )

        rows = []
        failures = []

        t0 = time.time()

        for i, key in enumerate(
            todo
        ):

            try:

                rows.append(
                    _growable17(
                        inv17[key]
                    )
                )

            except Exception as exc:

                failures.append(
                    f"{os.path.basename(inv17[key])}: "
                    f"{type(exc).__name__}: "
                    f"{exc}"
                )

            if (
                (i + 1) % 200
                == 0
            ):

                print(
                    f"  {i+1}/"
                    f"{len(todo)}",
                    flush=True,
                )

        if failures:

            print(
                f"{len(failures)} files failed."
            )

            for msg in failures[:5]:
                print(
                    "  ",
                    msg,
                )

        if rows:

            C17 = pd.concat(
                [
                    C17,
                    pd.DataFrame(rows),
                ],
                ignore_index=True,
            )

            Table.from_pandas(
                C17
            ).write(
                P17_CACHE,
                overwrite=True,
            )

            print(
                f"Updated cache in "
                f"{time.time()-t0:.0f} s."
            )

    # ── attach cache to branch nodes ─────────────────────────────────────────

    if len(C17):

        Z17 = N17.merge(
            C17,
            on=[
                "snap",
                "gx",
            ],
            how="inner",
            suffixes=(
                "",
                "_p17",
            ),
        )

        Z17["dt"] = (
            Z17["t"].to_numpy(float)
            -
            Z17["t_sft"].to_numpy(float)
        )

        Z17["t_available"] = (
            Z17["t_anchor"].to_numpy(float)
            -
            Z17["t"].to_numpy(float)
        )

        Z17["_gkey"] = [
            f"{int(a)}_{int(g)}"
            for a, g in zip(
                Z17["anchor_snap"],
                Z17["gal_id"],
            )
        ]

        # Evaluate the CDF at the actual time remaining before the anchor.
        for zn in (
            "ism",
            "cgm",
        ):

            Z17[
                f"fgrow_Z_{zn}"
            ] = [
                _interp_cdf17(
                    row,
                    zn,
                    "cdfZ",
                    tav,
                )
                for (_, row), tav in zip(
                    Z17.iterrows(),
                    Z17["t_available"],
                )
            ]

            Z17[
                f"fgrow_gas_{zn}"
            ] = [
                _interp_cdf17(
                    row,
                    zn,
                    "cdfG",
                    tav,
                )
                for (_, row), tav in zip(
                    Z17.iterrows(),
                    Z17["t_available"],
                )
            ]

            Z17[
                f"fgrow_N_{zn}"
            ] = [
                _interp_cdf17(
                    row,
                    zn,
                    "cdfN",
                    tav,
                )
                for (_, row), tav in zip(
                    Z17.iterrows(),
                    Z17["t_available"],
                )
            ]

        print(
            f"{len(Z17)} branch nodes retained."
        )


# ─────────────────────────────────────────────────────────────────────────────
# ONE NUMBER PER GALAXY
#
# Main quantity:
#
#       time-weighted mean of f_Z,growable
#
# over the chosen post-quenching interval.
#
# This asks whether the galaxy spends a substantial fraction of its post-SFT
# evolution with a large metal reservoir in conditions where dust could grow
# quickly enough before the anchor.
# ─────────────────────────────────────────────────────────────────────────────

if len(Z17):

    rows = []

    for gkey, d in Z17.groupby(
        "_gkey"
    ):

        d = (
            d.sort_values("t")
            .copy()
        )

        w = (
            (d["dt"] >= P17_WIN[0])
            &
            (d["dt"] <= P17_WIN[1])
            &
            (d["t_available"] > 0)
        )

        dw = d[w].copy()

        row = {
            "gkey": gkey,

            "gal_id": int(
                d["gal_id"].iloc[0]
            ),

            "anchor_snap": int(
                d["anchor_snap"].iloc[0]
            ),

            "n_snap": int(
                len(dw)
            ),

            "lfd_anc": (
                float(
                    d["lfd_anc"].iloc[0]
                )
                if "lfd_anc" in d
                else np.nan
            ),

            "log_mstar": (
                float(
                    d["log_mstar"].iloc[0]
                )
                if "log_mstar" in d
                else np.nan
            ),

            "dusty": (
                bool(
                    d["dusty"].iloc[0]
                )
                if "dusty" in d
                else False
            ),

            "cls": (
                d["cls"].iloc[0]
                if "cls" in d
                else ""
            ),

            "t_since_sft": (
                float(
                    d["t_anchor"].iloc[0]
                    -
                    d["t_sft"].iloc[0]
                )
                if (
                    "t_anchor" in d
                    and "t_sft" in d
                )
                else np.nan
            ),
        }

        for zn in (
            "ism",
            "cgm",
        ):

            # MAIN: metal-mass-weighted growable fraction.
            row[
                f"fgrow_Z_{zn}"
            ] = _time_weighted_mean(
                dw["t"].to_numpy(float),
                dw[
                    f"fgrow_Z_{zn}"
                ].to_numpy(float),
            )

            # Robustness: gas-mass-weighted fraction.
            row[
                f"fgrow_gas_{zn}"
            ] = _time_weighted_mean(
                dw["t"].to_numpy(float),
                dw[
                    f"fgrow_gas_{zn}"
                ].to_numpy(float),
            )

            # Auxiliary: number fraction.
            row[
                f"fgrow_N_{zn}"
            ] = _time_weighted_mean(
                dw["t"].to_numpy(float),
                dw[
                    f"fgrow_N_{zn}"
                ].to_numpy(float),
            )

            # Median tau_grow itself, weighted by metals at each snapshot
            # then summarized over time.
            tg = (
                dw[
                    f"tg_med_Z_{zn}"
                ].to_numpy(float)
                if f"tg_med_Z_{zn}" in dw
                else np.array([])
            )

            tg = tg[
                np.isfinite(tg)
                & (tg > 0)
            ]

            row[
                f"tg_med_Z_{zn}"
            ] = (
                float(
                    np.nanmedian(tg)
                )
                if len(tg)
                else np.nan
            )

            nz = (
                dw[
                    f"n_eligible_{zn}"
                ].to_numpy(float)
                if f"n_eligible_{zn}" in dw
                else np.array([])
            )

            nz = nz[
                np.isfinite(nz)
            ]

            row[
                f"n_eligible_{zn}_med"
            ] = (
                float(
                    np.nanmedian(nz)
                )
                if len(nz)
                else np.nan
            )

        rows.append(
            row
        )

    G17 = pd.DataFrame(
        rows
    )


# ─────────────────────────────────────────────────────────────────────────────
# MAIN FIGURE
#
# Fraction of METALS residing in gas where tau_grow is shorter than the time
# remaining until the anchor, versus final dust fraction.
#
# One point = one galaxy.
# ─────────────────────────────────────────────────────────────────────────────

if len(G17):

    fig, axs = plt.subplots(
        1,
        2,
        figsize=(
            10.4,
            4.5,
        ),
        sharex=True,
        sharey=True,
    )

    for ax, zn, title in zip(
        axs,
        (
            "ism",
            "cgm",
        ),
        (
            "non-SF ISM",
            f"CGM (< {P17_RMAX:g} kpc)",
        ),
    ):

        x = G17[
            "lfd_anc"
        ].to_numpy(float)

        y = G17[
            f"fgrow_Z_{zn}"
        ].to_numpy(float)

        m = (
            np.isfinite(x)
            &
            np.isfinite(y)
        )

        ax.scatter(
            x[m],
            y[m],
            s=42,
            alpha=0.80,
            edgecolors="none",
            rasterized=True,
        )

        ax.set_title(
            title
        )

        ax.set_xlabel(
            r"final "
            r"$\log(M_{\rm dust}/M_\star)$"
        )

        ax.set_ylim(
            -0.04,
            1.04,
        )

        if m.sum() >= 6:

            rr = spearmanr(
                x[m],
                y[m],
            )

            ax.text(
                0.04,
                0.95,
                rf"$\rho={rr.statistic:+.2f}$"
                + "\n"
                + rf"$p={rr.pvalue:.2g}$",
                transform=ax.transAxes,
                ha="left",
                va="top",
            )

    axs[0].set_ylabel(
        r"metal-mass fraction with "
        r"$\tau_{\rm grow}<t_{\rm available}$"
    )

    fig.suptitle(
        "How much of the metal reservoir can grow dust before the final epoch?"
    )

    fig.tight_layout()

    paper_save(
        fig,
        "kappa_growable_metals_vs_final_fdust",
    )

    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# ROBUSTNESS FIGURE — GAS-MASS WEIGHTING
# ─────────────────────────────────────────────────────────────────────────────

if len(G17):

    fig, axs = plt.subplots(
        1,
        2,
        figsize=(
            10.4,
            4.5,
        ),
        sharex=True,
        sharey=True,
    )

    for ax, zn, title in zip(
        axs,
        (
            "ism",
            "cgm",
        ),
        (
            "non-SF ISM",
            f"CGM (< {P17_RMAX:g} kpc)",
        ),
    ):

        x = G17[
            "lfd_anc"
        ].to_numpy(float)

        y = G17[
            f"fgrow_gas_{zn}"
        ].to_numpy(float)

        m = (
            np.isfinite(x)
            &
            np.isfinite(y)
        )

        ax.scatter(
            x[m],
            y[m],
            s=42,
            alpha=0.80,
            edgecolors="none",
            rasterized=True,
        )

        ax.set_title(
            title
        )

        ax.set_xlabel(
            r"final "
            r"$\log(M_{\rm dust}/M_\star)$"
        )

        ax.set_ylim(
            -0.04,
            1.04,
        )

        if m.sum() >= 6:

            rr = spearmanr(
                x[m],
                y[m],
            )

            ax.text(
                0.04,
                0.95,
                rf"$\rho={rr.statistic:+.2f}$"
                + "\n"
                + rf"$p={rr.pvalue:.2g}$",
                transform=ax.transAxes,
                ha="left",
                va="top",
            )

    axs[0].set_ylabel(
        r"gas-mass fraction with "
        r"$\tau_{\rm grow}<t_{\rm available}$"
    )

    fig.suptitle(
        "How much gas occupies conditions capable of grain growth?"
    )

    fig.tight_layout()

    paper_save(
        fig,
        "kappa_growable_gas_vs_final_fdust",
    )

    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# DIAGNOSTIC FIGURE — TYPICAL GROWTH CLOCK
#
# This is useful for checking whether the fractions saturate because basically
# every eligible particle has a tiny tau_grow.
# ─────────────────────────────────────────────────────────────────────────────

if len(G17):

    fig, axs = plt.subplots(
        1,
        2,
        figsize=(
            10.4,
            4.5,
        ),
        sharex=True,
        sharey=True,
    )

    for ax, zn, title in zip(
        axs,
        (
            "ism",
            "cgm",
        ),
        (
            "non-SF ISM",
            f"CGM (< {P17_RMAX:g} kpc)",
        ),
    ):

        x = G17[
            "lfd_anc"
        ].to_numpy(float)

        tg = G17[
            f"tg_med_Z_{zn}"
        ].to_numpy(float)

        with np.errstate(
            divide="ignore",
            invalid="ignore",
        ):

            y = np.log10(
                tg
            )

        m = (
            np.isfinite(x)
            &
            np.isfinite(y)
        )

        ax.scatter(
            x[m],
            y[m],
            s=42,
            alpha=0.80,
            edgecolors="none",
            rasterized=True,
        )

        ax.axhline(
            0.0,
            lw=1.0,
            ls="--",
            color="0.35",
        )

        ax.set_title(
            title
        )

        ax.set_xlabel(
            r"final "
            r"$\log(M_{\rm dust}/M_\star)$"
        )

    axs[0].set_ylabel(
        r"$\log_{10}$ metal-weighted median "
        r"$\tau_{\rm grow}$ [Gyr]"
    )

    fig.suptitle(
        "Typical grain-growth timescale of eligible gas"
    )

    fig.tight_layout()

    paper_save(
        fig,
        "kappa_growth_timescale_vs_final_fdust",
    )

    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# PRINT CORRELATIONS
# ─────────────────────────────────────────────────────────────────────────────

if len(G17):

    print(
        "\n"
        + "-"
        * 105
    )

    print(
        "CORRELATION WITH FINAL DUST FRACTION"
    )

    print(
        "-"
        * 105
    )

    for col, label in [

        (
            "fgrow_Z_ism",
            "growable metal fraction — ISM",
        ),

        (
            "fgrow_Z_cgm",
            "growable metal fraction — CGM",
        ),

        (
            "fgrow_gas_ism",
            "growable gas fraction — ISM",
        ),

        (
            "fgrow_gas_cgm",
            "growable gas fraction — CGM",
        ),
    ]:

        x = G17[
            col
        ].to_numpy(float)

        y = G17[
            "lfd_anc"
        ].to_numpy(float)

        m = (
            np.isfinite(x)
            &
            np.isfinite(y)
        )

        if m.sum() >= 6:

            rr = spearmanr(
                x[m],
                y[m],
            )

            print(
                f"{label:34s}: "
                f"rho={rr.statistic:+.3f}, "
                f"p={rr.pvalue:.3g}, "
                f"N={m.sum()}"
            )


# ─────────────────────────────────────────────────────────────────────────────
# SAVE OUTPUTS
# ─────────────────────────────────────────────────────────────────────────────

if len(Z17):

    out_nodes = os.path.join(
        OUT,
        f"kappa_{P17_BOX}_growable_gas_nodes.csv",
    )

    Z17.to_csv(
        out_nodes,
        index=False,
    )

    print(
        "\nSaved:"
    )

    print(
        " ",
        out_nodes,
    )


if len(G17):

    out_gal = os.path.join(
        OUT,
        f"kappa_{P17_BOX}_growable_gas_galaxies.csv",
    )

    G17.to_csv(
        out_gal,
        index=False,
    )

    print(
        " ",
        out_gal,
    )

    print(
        " ",
        P17_CACHE,
    )
```

## Part 18 — the decay after the truncation, timed: the fits themselves, and how τ is distributed

Part 15 block (c) fitted a **double exponential** to every branch and printed the median $\tau_1$ / $\tau_2$ per group.
Read back against the data, those numbers do not survive inspection, and this part is the revisit.

### What is wrong with the double exponential

$$y(t) \;=\; A_1 e^{-t/\tau_1} + A_2 e^{-t/\tau_2}$$

has **four free parameters**, and the window it is fitted over holds ~10 points across ~1.2 Gyr, because it **drops
every point sitting on the dust floor** and most branches reach zero dust inside a Gyr. The audit block (b) prints the consequence on the
same branches: **two fits in five sit on one of the fitter's bounds** ($\tau_1$ at 0.05 Gyr, or $\tau_2$ at 50–70 Gyr),
and **three quarters give $f_{\rm slow} < 1\%$** — the slow term carries essentially none of the amplitude, so $\tau_2$ is
not measured at all, it is whatever the optimiser was left holding. Part 15's median-track figure flattens for a second
reason: the branches that fall fastest reach the floor and *leave the sample*, so the late track is drawn by the survivors.

### What replaces it

Everything here is a **straight line in $\log_{10} y$**, which is what an exponential is, so at a fixed break the fit is
linear and there is no local minimum to fall into:

$$\log_{10} y \;=\; c - s_1\,(t-t_{\rm SFT}) \quad (t \le t_{\rm brk}), \qquad
\log_{10} y \;=\; c - s_1 t_{\rm brk} - s_2\,(t - t_{\rm brk}) \quad (t > t_{\rm brk}),$$

$$\tau_1 = \frac{\log_{10} e}{s_1}, \qquad \tau_2 = \frac{\log_{10} e}{s_2}, \qquad
\tau_0 = \frac{\log_{10} e}{s_0}\ \text{(the single exponential the break is scored against, by BIC)} .$$

Three things make the numbers identifiable where Part 15's were not:

| | |
|---|---|
| **the window** | **every snapshot from the truncation to the anchor** (`P18_WIN = "all"`). Nothing is dropped and nothing is trimmed: whatever a branch does after its minimum is part of the record and part of the fit |
| **the floor** | $\log M_{\rm dust}/M_\star = $ `P9_FLOOR` is SIMBA saying the galaxy has **no dust left**, so it is fitted as the value it is (`P18_CENSOR = False`). A branch that reaches zero therefore comes out with a **flat second leg and no $\tau_2$ at all** — the honest form of *there is nothing left to lose*, and it is counted rather than hidden |
| **the profile** | $t_{\rm brk}$ is scanned over the observed times and chosen by the likelihood, with at least `P18_NSIDE` points and `P18_DTMIN` Gyr on each side; each leg's error comes from the fit's own covariance at that break, with $\sigma$ taken from the points **off** the floor (a run of identical floored values carries no scatter and would shrink every error bar to zero) |

Beside them, **model-free**: $t_{\rm 1dex}$, the time to lose the first dex, by interpolation and with no model at all.

The two narrower treatments are one knob away — `P18_WIN = "to_min"` stops at the minimum, `P18_CENSOR = True` turns the
floored points back into upper limits — and with `P18_ALT` the narrow one is fitted **alongside**, so block (b) prints
exactly what the choice changes: $\tau_1$ by nothing (0.138 against 0.150 Gyr in m100 — the steep leg is the same object
either way), $\tau_2$ by a factor two, because the flat floor tail exists only in the wide window.

### The shape classes, declared before any timescale is quoted

`steep→shallow` (the claim under test, **both legs still falling**) · `rebound` (it fell and then **stopped** falling:
the second leg is flat or rising — for the dust either the floor, i.e. nothing left, or real regrowth; `reb` and `n_fl`
in the CSV say which) · `shallow→steep` (a branch that holds its dust and *then* loses it) · `single` (no break
preferred) · `short` (the window holds one slope but not two) · `flat` (it never fell `P18_DROPMIN` dex) · `nofit`.
Block (a) prints the mix per box **and split by the dust label**, which is where the result is.

### The two confounders that manufacture false modes

* **The snapshot spacing** runs from 0.07 Gyr at $z=3$ to 0.18 Gyr at $z=0$. A $\tau$ under one step means the e-folding is
  shorter than the sampling: the *slope* is still fitted from the several points that cross the fall — `e(s)/s`, the fit's
  own relative error, says how well — but nothing *inside* one e-folding is resolved, and a mode piling up there is a
  limit, not a measurement. Every $\tau$ axis carries that line, and the share below it is printed.
* **The anchor.** Each box mixes ten anchors, $z = 3 \to 0$, and the median $\log \tau_0$ alone moves 0.3 dex along the
  ladder (FIG 3, bottom). A distribution pooled over anchors is therefore a *mixture* before any physics enters, so every
  test is run **pooled, again after centring each anchor on its own median, again per anchor, and again on the
  well-measured branches alone** (`e/s` < `P18_EMAX`).

**The test is Silverman's critical bandwidth**, calibrated by the smoothed bootstrap: the smallest bandwidth whose kernel
estimate still shows more than one mode, and how often a resample of the data smoothed at *that* bandwidth does the same.
$p < 0.05$ rejects unimodality. The 2-component Gaussian mixture drawn beside it is **descriptive only** — a mixture beats
a single Gaussian on skewness alone, which these distributions have in quantity — so what is printed with it is the Ashman
$D$ (a real gap needs $D > 2$) and whether the fitted mixture is *itself* two-humped.

### The figures

* **FIG 1 `kappa_decay_gallery`** — the fits themselves: a ladder of `P18_NGAL` branches from the fastest decay to the
  slowest, with every point (circles off the floor, filled ▽ on it), the broken fit, the single one, and Part 15's double
  exponential on the same branch. **Nothing below this figure is worth reading until this one passes the eye.**
* **FIG 2 `kappa_decay_classes`** — `P18_NCLS` examples of each shape class, so the classification can be checked by hand.
* **FIG 3 `kappa_decay_audit`** — is the number a measurement: $\tau_0$ in snapshot steps, $\tau_0$ against Part 15's
  $\tau_1$ / $\tau_2$ with its bounds drawn as dotted lines, and the anchor ladder.
* **FIG 4 `kappa_decay_taudist`** — **the question**: the distributions of $\tau_0$, $\tau_1$, $\tau_2$ and $t_{\rm 1dex}$
  per box, each with the anchor-centred version, the well-measured subset, the dust label, the mixture, and the Silverman
  $p$ in the title.
* **FIG 5 `kappa_decay_plane`** — the two legs of every branch in one plane. Above the diagonal the decay flattened, below
  it the branch held its dust and then lost it; the faint points are the branches where the break is not preferred, and
  they scatter along the diagonal as they should. The **triangles on the top edge** are the branches that stopped falling
  and have no second timescale at all — they belong at $\tau_2 \to \infty$, not off the figure. Below it, when the break
  falls.
* **FIG 6 `kappa_decay_dustlink`** — is the slow decay the dusty one? The dust's own $\tau$ against the dust it ends with
  is **circular** (a slower decay from the same start ends higher) and is drawn as such; the honest version is the same
  question asked of the **sSFR** and the **H₂** decay of the same galaxies, which the dust label does not enter.

### What it answers, and where the answer differs from the expectation

The expectation this part was opened with was: a steep first leg distributed normally, and a **bimodal** second leg, the
dustier galaxies decaying more slowly. What the fits give is (block G prints it from the data):

1. **the shape is right** — steep-then-shallow is the single most common class (26 % of the m100 branches, 30 % of the
   dust-poor ones) — **but it is not what a dusty QG does**. Only 2 % of the dusty QGs are still falling on both legs; a
   third of them fall and then **stop**, and for those the arrest is real **regrowth**: a median 2.4 dex recovered from a
   minimum that was the zero-dust floor for half of them. The dust-poor branches that stop are stopping *at* the floor;
2. $\log \tau$ is **right-skewed and not Gaussian** anywhere, and **Silverman's test rejects one mode nowhere** — nor
   does the model-free $t_{\rm 1dex}$. One mode with a long slow tail, the dusty QGs living inside that tail, and nothing
   a mixture would be entitled to name;
3. the dusty QGs *are* slower **in the dust** (AUC 0.81 in m100), but that test is circular, and the non-circular
   versions split by resolution: in the coarse boxes they lose their sSFR and their H₂ *faster* (0.45, 0.41), while in
   **m25 — the box that resolves the gas — they are slower in everything** (0.66, 0.59).

*A pure read* of `ism_prediction_histories_<box>.hdf5` restricted to Part 3's rows; Part 15 is optional and, where it has
been run, its double-exponential parameters and its branch split are carried along for the comparison →
`kappa_decay_{gallery,classes,audit,taudist,plane,dustlink}{.png,.pdf}`, `kappa_decay.csv` and
`kappa_decay_galaxies.csv` (one row per branch, every quantity's window, slopes, break, class and errors).

```python
# ── Part 18 — THE DECAY TIMESCALES AFTER THE TRUNCATION: the fits themselves, and how tau is distributed ──
# WHY THIS PART EXISTS. Part 15 block (c) fitted a DOUBLE EXPONENTIAL  y = A1 exp(-t/tau1) + A2 exp(-t/tau2)  to every
# branch and printed the median tau1 / tau2 per group. Read back, those numbers do not survive inspection:
#   - the median tau1 is 0.12-0.27 Gyr, at or BELOW the snapshot spacing (0.07 Gyr at z = 3, 0.18 Gyr at z = 0);
#   - TWO IN FIVE of its fits sit on one of the fitter's bounds (tau1 at 0.05 Gyr, or tau2 at 50-70 Gyr), and f_slow —
#     the weight the slow term carries at the truncation — is under 1 % for THREE QUARTERS of them, i.e. for most
#     branches the "second exponential" is not in the fit at all and tau2 is whatever the optimiser was left holding;
#   - the window it was fitted over is short: it DROPPED every point sitting on the dust floor, and what is left spans
#     1.2-1.5 Gyr with ~10 points, which cannot support four free parameters.
# So Part 15's tau1 / tau2 are mostly the fitter's bounds, not measurements — and its median-track figure flattens partly
# because the fast branches leave the sample when they reach the floor (a survivorship effect, not a slow leg).
# WHAT THIS PART DOES INSTEAD — the same claim, in a form that is identifiable:
#   (1) THE WINDOW: EVERY SNAPSHOT AFTER THE TRUNCATION, nothing dropped and nothing censored (P18_WIN = "all", the
#       user's rule). log M_dust/M* = P9_FLOOR is SIMBA saying the galaxy has no dust left, so it is fitted as the value
#       it is, and a branch that reaches zero simply comes out with a FLAT second leg and no tau_2 — the honest form of
#       "there is nothing left to lose". The time of the minimum, the regrowth after it and the number of floored steps
#       are carried per branch (t_min, reb, n_fl); P18_WIN / P18_CENSOR restore the two narrower treatments, and P18_ALT
#       fits the narrow one alongside so that block (b) can print exactly what the choice changes.
#   (2) THE MODEL. A BROKEN EXPONENTIAL in log10:  log y = c - s1 t  up to t_brk, then slope s2 — the same "steep then
#       shallow" shape, but with every parameter fixed by the data: the break is profiled over the observed times and the
#       two slopes are linear at fixed break. The error bars come from the scatter of the points that are NOT on a
#       floor, since a run of identical floored values carries none and left in would shrink every error bar to zero.
#       tau_1 = log10(e)/s1 and tau_2 = log10(e)/s2 are the e-folding times of the two legs; tau_0 is the single
#       exponential the break is scored against (BIC), i.e. THE characteristic timescale when there is only one.
#   (3) THE MODEL-FREE CHECK. t_1dex, the time to lose the first dex of dust, computed by interpolation and used to
#       verify that the fitted tau_0 is not an artefact of the model.
#   (4) THE FITS, DRAWN. FIG 1 a ladder of P18_NGAL galaxies from the fastest decay to the slowest, FIG 2 five examples
#       of each shape class, both with every point, the ones on the floor marked, the fits, and — where Part 15
#       has been run — its double exponential on the same branch. This is the part the eye has to pass before any tau
#       distribution is worth reading.
#   (5) THE DISTRIBUTIONS, which is the question: is log tau unimodal? FIG 4 per box, with TWO confounders handled
#       explicitly, because both manufacture false modes:
#         - THE SNAPSHOT SPACING. tau below ~P18_RESN steps is not resolved; the axis carries that line and the share
#           below it is printed.
#         - THE ANCHOR. Each box mixes ten anchors, z = 3 to 0, whose spacing and whose available post-SFT time differ by
#           a factor 2.5; the median log tau_0 alone moves 0.3 dex along the ladder. So every distribution is tested BOTH
#           pooled and after centring each anchor on its own median, and the per-anchor tests are printed beside them.
#       The test is SILVERMAN'S critical-bandwidth test (calibrated by a smoothed bootstrap) — the GMM decomposition is
#       drawn beside it but is NOT the test: a 2-component mixture beats a single Gaussian on skewness alone, which is
#       why the Ashman D and whether the fitted mixture is itself two-humped are printed with it.
#   (6) THE LINK TO THE DUST (FIG 5, FIG 6): the tau_1-tau_2 plane galaxy by galaxy, and tau against the anchor dust —
#       inside anchor cells, and in the two quantities where the test is NOT circular (the sSFR and the H2 decay against
#       the dust label; the dust's own tau against its own end point is close to a tautology and is marked as such).
# A PURE READ of ism_prediction_histories_<box>.hdf5 restricted to Part 3's rows. Part 15 is optional: where D15 exists
# its branch split and its double-exponential parameters are carried along for the comparison.
P18_BOXES   = None         # None = the boxes Part 3 built, the channel-off box last
P18_DQ      = ["lfd", "lssfr", "lfh2"]      # the quantities fitted (log10), the dust first
P18_QGAL    = "lfd"        # THE quantity the galleries and the distribution figures are drawn for
# ── the decay window ──
P18_WIN     = "all"        # THE WINDOW: "all" = EVERY snapshot from the truncation to the anchor, nothing dropped and
                           #   nothing censored (the user's rule, and the default) | "to_min" = stop at the first
                           #   arrival at the branch's post-SFT minimum, which is where the regrowth begins
P18_CENSOR  = False        # treat a point on a floor as a LEFT-CENSORED upper limit instead of a measurement.  With
                           #   P18_WIN = "all" this is off: log M_dust/M* = P9_FLOOR is SIMBA saying zero dust, and it
                           #   is fitted as the value it is — so a branch that reaches zero has a FLAT second leg and
                           #   no tau_2, which is the honest reading of "there is nothing left to lose"
P18_ENDTOL  = 0.0015         # for P18_WIN = "to_min": how close to the minimum counts as having arrived
P18_FLOORTOL = 0.00001        # a point within this many dex of a floor COUNTS as being on it (marked in every figure)
P18_DROPMIN = 0.6          # a branch that never falls this far is "flat": it has no decay to time
P18_ALT     = True         # also fit the OTHER treatment (window to the minimum, floors censored) for the gallery
                           #   quantity and print what it changes — the audit of the choice above
# ── the fit ──
P18_NMIN, P18_SPANMIN = 1, 0.3      # the single exponential needs this many points over this much time [Gyr]
P18_NBRK, P18_SPANBRK = 3, 0.5      # ... the broken one this many over this much
P18_NSIDE, P18_DTMIN  = 3, 0.3      # each leg needs this many points and this much time [Gyr]
P18_EMIT    = 25           # the censored-EM iterations, where P18_CENSOR asks for them
# ── the galleries ──
P18_NGAL    = 24           # FIG 1: how many galaxies
P18_GMODE   = "ladder"     # "ladder" = evenly along the sorted tau | "random" | "worst" = the largest residuals
P18_GBOX    = None         # None = every box; a box name draws the ladder from that box alone
P18_NCLS    = 5            # FIG 2: examples per shape class
P18_SEED    = 7
# ── the distributions ──
P18_TAUAX   = (-2.0, 1.6)  # the log10 tau axis [Gyr], and the range every histogram is clipped into
P18_TBIN    = 0.1
P18_RESN    = 1.0          # tau under this many median snapshot steps: one e-folding inside one snapshot step
P18_EMAX    = 0.35         # a tau is "well measured" when the fit's own relative error on its slope is under this
P18_SILV    = 400          # the Silverman bootstrap; 0 skips the test
P18_GMMBOOT = 0            # the GMM parametric bootstrap (slow, and NOT the multimodality test) — 0 = descriptive only
P18_MINDIST = 10           # a distribution is tested only above this many galaxies
P18_NANCH   = 10           # ... and an anchor is tested on its own only above this many
P18_QLAB    = {"lfd": r"$\log M_{\rm dust}/M_\star$", "lssfr": r"$\log$ sSFR", "lfh2": r"$\log 1.36 M_{\rm H_2}/M_\star$"}
P18_QSH     = {"lfd": "the dust", "lssfr": "the sSFR", "lfh2": "the H2"}
P18_CLS     = ["steep-shallow", "rebound", "shallow-steep", "single", "short", "flat", "nofit"]
P18_CCOL    = {"steep-shallow": "#d6604d", "rebound": "#e08214", "shallow-steep": "#4393c3", "single": "#5aae61",
               "short": "#b8860b", "flat": "#762a83", "nofit": "0.6"}
P18_CLAB    = {"steep-shallow": "steep then shallow — both legs falling",
               "rebound": "it fell and then STOPPED falling: the second leg is flat or rising (for the dust, usually the "
                          "floor — there is nothing left — or real regrowth)",
               "shallow-steep": "shallow then steep (a delayed collapse)",
               "single": "one exponential (no break preferred)", "short": "too short for a break — one slope only",
               "flat": "it never fell — flat, or rising, from the truncation on", "nofit": "not measurable"}

L10E18 = float(np.log10(np.e))          # dex/Gyr -> e-foldings: tau = L10E18 / s
_FL18  = {"lfd": P9_FLOOR, "lssfr": float(np.log10(SSFR_FLOOR))}     # the CENSORED values; the H2 has no floor
D18, R18, ROW18 = pd.DataFrame(), {}, []
_BX18 = [b for b in (list(H) if P18_BOXES is None else P18_BOXES) if b in H]
_BX18 = [b for b in _BX18 if b != NOX_BOX] + [b for b in _BX18 if b == NOX_BOX]
if not _BX18:
    print("Part 18: Part 3 built no branch records — run it first.")
else:
    import time as _time18
    from scipy import stats as _st18

    # ══ THE MACHINERY ═══════════════════════════════════════════════════════════════════════════════════════════════
    def _fitlin18(X, y, cens, floor, iters=None):
        """Least squares with LEFT-CENSORED rows (y <= floor), by the censored EM: the M-step is the ordinary lstsq, the
        E-step replaces a censored point by its conditional mean under the current fit. Returns (p, sigma, SSR over the
        OBSERVED points, the log-likelihood of observed + censored)."""
        n, k = len(y), X.shape[1]
        obs = ~cens
        if obs.sum() < k + 1:
            return None
        p = np.linalg.lstsq(X[obs], y[obs], rcond=None)[0]
        sig = max(float(np.sqrt(np.sum((y[obs] - X[obs] @ p) ** 2) / max(obs.sum() - k, 1))), 1e-3)
        if cens.any():
            yf = y.copy()
            for _ in range(int(P18_EMIT if iters is None else iters)):
                mu = X @ p
                a = (floor - mu[cens]) / sig
                lam = np.exp(_st18.norm.logpdf(a) - np.clip(_st18.norm.logcdf(a), -700, 0))    # phi/Phi, stable
                yf[cens] = mu[cens] - sig * lam
                v = sig ** 2 * np.clip(1.0 - a * lam - lam ** 2, 1e-6, None)
                pn = np.linalg.lstsq(X, yf, rcond=None)[0]
                sn = float(np.sqrt(max((np.sum((yf - X @ pn) ** 2) + np.sum(v)) / n, 1e-8)))
                done = np.max(np.abs(pn - p)) < 1e-7 and abs(sn - sig) < 1e-7
                p, sig = pn, sn
                if done:
                    break
        mu = X @ p
        a = (floor - mu[cens]) / sig
        ll = float(np.sum(_st18.norm.logpdf(y[obs], mu[obs], sig)) + np.sum(np.clip(_st18.norm.logcdf(a), -700, 0)))
        return p, sig, float(np.sum((y[obs] - mu[obs]) ** 2)), ll

    def _seg18(t, y, cens, nofl=None):
        """ONE galaxy's decay window: the single exponential, and the broken one profiled over the observed break times.
        Everything is a straight line in log10 y, so at fixed t_brk the fit is linear and there is no local minimum to
        fall into; the break is chosen by the likelihood, and scored against the single exponential by BIC.
        THE ERROR BARS come from the scatter of the points that are NOT on a floor: a run of identical floored values
        carries no scatter at all, and left in it would shrink every error bar towards zero."""
        out = {k: np.nan for k in ("c", "s0", "e_s0", "sig0", "rms0", "s1", "e_s1", "s2", "e_s2", "tb", "sig", "rms",
                                   "dbic", "ll", "ll0")}
        nofl = np.zeros(len(t), bool) if nofl is None else np.asarray(nofl, bool)
        out["n"], out["n_obs"], out["span"] = int(len(t)), int((~cens).sum()), float(t[-1] - t[0]) if len(t) else np.nan
        if len(t) < P18_NMIN or out["span"] < P18_SPANMIN or out["n_obs"] < 3:
            return out
        fl = float(np.min(y[cens])) if cens.any() else -np.inf
        n = len(t)
        X0 = np.vstack([np.ones(n), t]).T
        r0 = _fitlin18(X0, y, cens, fl)
        if r0 is None:
            return out
        p0, sig0, ssr0, ll0 = r0

        def _sige(p, X, sig):
            """the scatter the error bars are taken from: the off-floor points where there are enough of them."""
            m = (~cens) & (~nofl)
            if m.sum() >= X.shape[1] + 2:
                return max(float(np.sqrt(np.sum((y[m] - (X @ p)[m]) ** 2) / (m.sum() - X.shape[1]))), 1e-3)
            return sig

        C0 = _sige(p0, X0, sig0) ** 2 * np.linalg.pinv(X0.T @ X0)
        out.update(c=float(p0[0]), s0=float(-p0[1]), sig0=float(sig0), ll0=float(ll0),
                   e_s0=float(np.sqrt(max(C0[1, 1], 0.0))), rms0=float(np.sqrt(ssr0 / max(out["n_obs"], 1))))
        if n < P18_NBRK or out["span"] < P18_SPANBRK:
            return out
        best = None
        for tb in t[P18_NSIDE - 1:n - P18_NSIDE]:
            if (tb - t[0]) < P18_DTMIN or (t[-1] - tb) < P18_DTMIN:
                continue
            X = np.vstack([np.ones(n), t, np.maximum(t - tb, 0.0)]).T
            r = _fitlin18(X, y, cens, fl)
            if r is not None and (best is None or r[3] > best[0][3]):
                best = (r, tb, X)
        if best is None:
            return out
        (p, sig, ssr, ll), tb, X = best
        C = _sige(p, X, sig) ** 2 * np.linalg.pinv(X.T @ X)
        out.update(c=float(p[0]), s1=float(-p[1]), s2=float(-(p[1] + p[2])), tb=float(tb), sig=float(sig), ll=float(ll),
                   e_s1=float(np.sqrt(max(C[1, 1], 0.0))),
                   e_s2=float(np.sqrt(max(C[1, 1] + C[2, 2] + 2 * C[1, 2], 0.0))),
                   rms=float(np.sqrt(ssr / max(out["n_obs"], 1))),
                   dbic=float((3 * np.log(n) - 2 * ll0) - (5 * np.log(n) - 2 * ll)))
        return out

    def _tau18(s):
        """a slope in dex/Gyr -> the e-folding time [Gyr]; a flat or rising leg has no decay time."""
        return (L10E18 / s) if (np.isfinite(s) and s > 1e-6) else np.nan

    def _decay18(t, y, floor, win=None, censor=None):
        """ONE branch from the truncation on: the window, the model-free numbers, the fits and the shape class.
        With P18_WIN = "all" (the default) the window is EVERY snapshot after the truncation, a point on the floor
        included as the value it is — SIMBA's zero dust — so the second leg of a branch that reaches zero comes out
        FLAT and it has no tau_2 at all.  "to_min" and P18_CENSOR restore the two narrower treatments."""
        win = P18_WIN if win is None else win
        censor = P18_CENSOR if censor is None else censor
        out = dict(cls="nofit", n=0, n_obs=0, span=np.nan, drop=np.nan, reb=np.nan, t_end=np.nan, t_min=np.nan,
                   t_floor=np.nan, n_fl=0, y0=np.nan, ymin=np.nan, y_end=np.nan, t1dex=np.nan, tau0=np.nan,
                   tau1=np.nan, tau2=np.nan, dt=np.nan)
        t, y = np.asarray(t, float), np.asarray(y, float)
        ok = np.isfinite(t) & np.isfinite(y) & (t >= -1e-9)
        t, y = t[ok], y[ok]
        o = np.argsort(t); t, y = t[o], y[o]
        if len(t) < 3:
            return out, None
        out["dt"] = float(np.median(np.diff(t)))
        onfl = (y <= floor + P18_FLOORTOL) if floor is not None else np.zeros(len(t), bool)
        ymin = float(np.min(y))
        i_min = int(np.flatnonzero(y <= ymin + P18_ENDTOL)[0])           # where the fall ends and the regrowth starts
        keep = np.ones(len(t), bool) if win == "all" else np.concatenate([np.ones(i_min + 1, bool),
                                                                          np.zeros(len(t) - i_min - 1, bool)])
        out.update(y0=float(y[0]), ymin=ymin, y_end=float(y[-1]), drop=float(y[0] - ymin), t_min=float(t[i_min]),
                   t_end=float(t[keep][-1]), n_fl=int(onfl.sum()),
                   reb=float(np.max(y[i_min:]) - ymin) if i_min < len(t) - 1 else 0.0,
                   t_floor=float(t[onfl][0]) if onfl.any() else np.nan)
        tw, yw, cw = t[keep], y[keep], (onfl[keep] & bool(censor))
        blw = np.flatnonzero(yw <= out["y0"] - 1.0)                      # model-free: the time to lose the first dex
        if len(blw):
            i = int(blw[0])
            out["t1dex"] = float(np.interp(out["y0"] - 1.0, [yw[i], yw[i - 1]], [tw[i], tw[i - 1]])) if i > 0 else float(tw[0])
        s = _seg18(tw, yw, cw, nofl=onfl[keep])
        out.update(s)
        out["tau0"], out["tau1"], out["tau2"] = _tau18(s["s0"]), _tau18(s["s1"]), _tau18(s["s2"])
        if not np.isfinite(s["s0"]):
            out["cls"] = "nofit"
        elif out["drop"] < P18_DROPMIN:
            out["cls"] = "flat"
        elif not np.isfinite(s["s1"]):
            out["cls"] = "short"
        elif s["dbic"] <= 0:
            out["cls"] = "single"
        elif s["s1"] > 1e-6 and s["s2"] <= 1e-6:
            out["cls"] = "rebound"                       # it fell, then stopped falling — the floor, or real regrowth
        else:
            out["cls"] = "steep-shallow" if s["s2"] < s["s1"] else "shallow-steep"
        return out, dict(t=t, y=y, cens=(onfl & bool(censor)), onfl=onfl, keep=keep)

    # ══ (A) THE PASS: one read of the histories, restricted to Part 3's rows ═════════════════════════════════════════
    def _pass18(box):
        F = H[box]["F"]
        want = {(int(r["snap"]), int(r["gal_id"])): r for _, r in F.iterrows()}
        rows, ser = [], {}
        with h5py.File(HIST_FILE.format(box=box), "r") as f:
            for gname in f:
                a = int(gname.replace("snap", ""))
                if not any(k[0] == a for k in want):
                    continue
                gh = f[gname]
                t = gh["t_yr"][:] / 1e9; o = np.argsort(t); t = t[o]
                ms, sfr = gh["mstar"][:][o], gh["sfr"][:][o]
                md = gh["mdust"][:][o] if "mdust" in gh else np.full_like(ms, np.nan)
                mh = gh["mh2"][:][o] if "mh2" in gh else np.full_like(ms, np.nan)
                for j, gid in enumerate(gh["gal_ids"][:]):
                    key = (a, int(gid))
                    if key not in want:
                        continue
                    r = want[key]
                    t_sft = float(r.get("t_qt", np.nan))
                    ex = np.isfinite(ms[:, j]) & (ms[:, j] > 0)
                    if not np.isfinite(t_sft) or ex.sum() < 3:
                        continue
                    den = np.where(ex, ms[:, j], np.nan)
                    with np.errstate(all="ignore"):
                        v = {"lfd": np.log10(np.where(ex, np.clip(md[:, j] / den, 10 ** P9_FLOOR, None), np.nan)),
                             "lssfr": np.log10(np.where(ex, np.clip(np.maximum(sfr[:, j], 0.0) / den, SSFR_FLOOR, None), np.nan)),
                             "lfh2": np.log10(np.where((mh[:, j] > 0) & ex, HE * mh[:, j] / den, np.nan))}
                    dt15 = t - t_sft
                    m15 = ex & (dt15 >= -1e-9)
                    if m15.sum() < 3:
                        continue
                    rec = dict(box=box, snap=a, gal_id=int(gid), t_sft=t_sft, t_post=float(np.max(dt15[m15])))
                    for c in ("z", "pop", "lfd", "age", "log_mstar", "log_mbh", "kcls", "rot", "central", "t_anchor",
                              "kappa", "ngas", "t_x1", "e_x", "f_x_post", "t_qt"):
                        if c in r:
                            rec[c] = r[c]
                    for q in P18_DQ:
                        d_, s_ = _decay18(dt15[m15], v[q][m15], _FL18.get(q))
                        for k_, w_ in d_.items():
                            rec[f"{q}_{k_}"] = w_
                        if q == P18_QGAL and s_ is not None:
                            ser[(box, a, int(gid))] = s_
                        if q == P18_QGAL and P18_ALT:      # the other treatment, for the audit of the choice
                            a_, _ = _decay18(dt15[m15], v[q][m15], _FL18.get(q), win="to_min", censor=True)
                            for k_, w_ in a_.items():
                                rec[f"alt_{k_}"] = w_
                    rows.append(rec)
        return pd.DataFrame(rows), ser

    _D18 = []
    for _b18 in _BX18:
        _t018 = _time18.time()
        _d18, _s18 = _pass18(_b18)
        if not len(_d18):
            print(f"  [{_b18}] no branch with a truncation — skipped"); continue
        _D18.append(_d18); R18.update(_s18)
        _q = P18_QGAL
        _cc = _d18[f"{_q}_cls"].value_counts()
        print(f"  [{_b18}] {len(_d18)} branches ({_time18.time() - _t018:.0f} s):  median window "
              f"{np.nanmedian(_d18[f'{_q}_t_end']):.2f} Gyr over {np.nanmedian(_d18[f'{_q}_n']):.0f} steps, snapshot "
              f"spacing {np.nanmedian(_d18[f'{_q}_dt']):.3f} Gyr")
        print(f"      {P18_QSH.get(_q, _q)}: " + ", ".join(f"{_cc.get(c, 0)} {c}" for c in P18_CLS if _cc.get(c, 0)))
    D18 = pd.concat(_D18, ignore_index=True) if _D18 else pd.DataFrame()

if not len(D18):
    print("Part 18: nothing to fit.")
else:
    # Part 15's own fit, carried along for the comparison (its window and its parameterisation, untouched)
    _P15C = [c for c in (f"{P18_QGAL}_{k}" for k in ("tau1", "tau2", "lA1", "lA2", "fslow", "dbic", "span")) ]
    _HAS15 = "D15" in dir() and len(D15) and all(c in D15.columns for c in _P15C)
    if _HAS15:
        _m15 = D15[["box", "snap", "gal_id"] + _P15C + (["grp"] if "grp" in D15.columns else [])].copy()
        _m15 = _m15.rename(columns={c: f"p15_{c.split('_', 1)[1]}" for c in _P15C})
        D18 = D18.merge(_m15, on=["box", "snap", "gal_id"], how="left")
    print(f"  Part 15's double exponential for the same branches: {'carried along' if _HAS15 else 'not in this kernel'}")

    def _leg18(ax, **kw):
        kw.setdefault("frameon", False); kw.setdefault("fontsize", 7)
        h, l = ax.get_legend_handles_labels()
        if h:
            ax.legend(h, l, **kw)

    def _segc18(c, s1, s2, tb, g):
        return c - s1 * np.minimum(g, tb) - s2 * np.maximum(g - tb, 0.0)

    def _2expc18(lA1, tau1, lA2, tau2, g):
        with np.errstate(all="ignore"):
            return np.log10(np.clip(10.0 ** lA1 * np.exp(-g / tau1) + 10.0 ** lA2 * np.exp(-g / tau2), 1e-300, None))

    _BB18 = [b for b in _BX18 if (D18.box == b).any()]
    print("=" * 132)
    print(f"Part 18 — THE DECAY AFTER THE TRUNCATION, TIMED.  a broken exponential per galaxy over "
          + ("EVERY SNAPSHOT from the truncation to the anchor," if P18_WIN == "all"
             else "[t_SFT, the first arrival at the branch minimum],"))
    print(f"  a point on a floor {'CENSORED as an upper limit' if P18_CENSOR else 'fitted as the value it is (zero dust)'};"
          f"  tau = log10(e)/slope,  every distribution tested against the snapshot spacing and against the anchor.")
    print("=" * 132)

    # ══ (B) WHAT SHAPE THE DECAYS ARE ═══════════════════════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(a) THE SHAPE OF THE DECAY — every branch classified before any timescale is quoted.  'steep then shallow' is")
    print("    the claim under test; 'shallow then steep' is its opposite, a branch that holds its dust and THEN loses it")
    print(f"  {'box':10s} {'quantity':8s} {'N':>5s} " + " ".join(f"{c[:11]:>12s}" for c in P18_CLS)
          + f" {'window':>7s} {'drop':>6s} {'regrow':>7s}")
    for b in _BB18:
        for q in P18_DQ:
            g = D18[D18.box == b]
            cc = g[f"{q}_cls"].value_counts()
            print(f"  {b:10s} {P18_QSH.get(q, q):8s} {len(g):5d} "
                  + " ".join(f"{100 * cc.get(c, 0) / max(len(g), 1):11.0f}%" for c in P18_CLS)
                  + f" {np.nanmedian(g[f'{q}_t_end']):7.2f} {np.nanmedian(g[f'{q}_drop']):6.2f} "
                    f"{np.nanmedian(g[f'{q}_reb']):7.2f}")
            for c in P18_CLS:
                ROW18.append(dict(box=b, block=f"shape_{q}", quantity=c, n_gal=int(cc.get(c, 0)),
                                  med=float(cc.get(c, 0) / max(len(g), 1))))
    if "pop" in D18.columns:
        print(f"\n  the same for {P18_QSH.get(P18_QGAL, P18_QGAL)} alone, split by the dust the galaxy ENDS with — "
              f"which shape does a dusty QG have?")
        print(f"  {'box':10s} {'label':7s} {'N':>5s} " + " ".join(f"{c[:11]:>12s}" for c in P18_CLS))
        for b_ in _BB18:
            for p_ in ("dusty", "pure"):
                g_ = D18[(D18.box == b_) & (D18["pop"] == p_)]
                if len(g_) < 8:
                    continue
                cc_ = g_[f"{P18_QGAL}_cls"].value_counts()
                print(f"  {b_:10s} {p_:7s} {len(g_):5d} "
                      + " ".join(f"{100 * cc_.get(c, 0) / len(g_):11.0f}%" for c in P18_CLS))
                for c in P18_CLS:
                    ROW18.append(dict(box=b_, block=f"shape_{P18_QGAL}_pop", quantity=c, kcls=p_,
                                      n_gal=int(cc_.get(c, 0)), med=float(cc_.get(c, 0) / len(g_))))
    print(f"  (the window is "
          + ("[t_SFT -> the anchor]" if P18_WIN == "all" else "[t_SFT -> the first arrival at the branch minimum]")
          + f" in Gyr, the drop and the regrowth in dex.  'rebound' = it fell and then STOPPED falling — for the dust")
    print( "   that is usually the floor, i.e. nothing left to lose, and sometimes real regrowth (the CSV's reb says")
    print(f"   which).  'flat' = it never fell {P18_DROPMIN:g} dex,")
    print( "   'short' = the window holds one slope but not two, 'nofit' = under "
          f"{P18_NMIN} points or {P18_SPANMIN:g} Gyr after the truncation — the anchor came too soon to time anything.)")

    # ══ (C) FIG 1 — THE FITS THEMSELVES, a ladder from the fastest decay to the slowest ══════════════════════════════
    _Q, _FLQ = P18_QGAL, _FL18.get(P18_QGAL)
    _GD18 = D18 if P18_GBOX is None else D18[D18.box == P18_GBOX]
    _fit18 = _GD18[np.isfinite(_GD18[f"{_Q}_s0"].to_numpy(float))].copy()
    _rng18 = np.random.default_rng(P18_SEED)

    def _panel18(ax, row, note=""):
        """ONE galaxy: its points, the ones on the floor, the window, and the three fits."""
        key = (row["box"], int(row["snap"]), int(row["gal_id"]))
        s = R18.get(key)
        if s is None:
            ax.set_axis_off(); return
        t, y, cn, kp = s["t"], s["y"], s["cens"], s["keep"]
        fl = s.get("onfl", cn)
        ax.plot(t[kp & ~fl], y[kp & ~fl], "o", color="k", ms=4.2, zorder=4)
        ax.plot(t[kp & fl & ~cn], y[kp & fl & ~cn], "v", color="0.4", ms=5.0, zorder=4)     # on the floor, and fitted
        ax.plot(t[kp & cn], y[kp & cn], "v", mfc="none", mec="0.45", ms=5.5, mew=1.0, zorder=4)   # ... or censored
        if (~kp).any():
            ax.plot(t[~kp], y[~kp], "o", color="0.78", ms=3.4, zorder=2)          # outside the window (P18_WIN)
        gmax = float(row[f"{_Q}_t_end"])
        gg = np.linspace(0.0, max(gmax, 1e-3), 200)
        _c, _s0 = float(row[f"{_Q}_c"]), float(row[f"{_Q}_s0"])
        if np.isfinite(row[f"{_Q}_s1"]):
            _t1, _t2 = row[f"{_Q}_tau1"], row[f"{_Q}_tau2"]
            ax.plot(gg, _segc18(_c, row[f"{_Q}_s1"], row[f"{_Q}_s2"], row[f"{_Q}_tb"], gg), "-", color="#d6604d", lw=2.2,
                    zorder=5, label=(rf"$\tau_1$={_t1:.2f}" if np.isfinite(_t1) else r"$\tau_1$: flat/up")
                                    + (rf"  $\tau_2$={_t2:.2f}" if np.isfinite(_t2) else r"  $\tau_2$: flat/up")
                                    + f"\n$t_b$={row[f'{_Q}_tb']:.2f}  rms={row[f'{_Q}_rms']:.2f}  "
                                      f"$\\Delta$BIC={row[f'{_Q}_dbic']:+.0f}")
            ax.axvline(row[f"{_Q}_tb"], color="#d6604d", lw=1.0, ls="--", alpha=0.55)
        _lab0 = rf"$\tau_0$={row[f'{_Q}_tau0']:.2f}" if np.isfinite(row[f"{_Q}_tau0"]) else r"$\tau_0$: flat/up"
        ax.plot(gg, _c - _s0 * gg, ":", color="#2166ac", lw=1.6, zorder=3, label=_lab0 + f"  rms={row[f'{_Q}_rms0']:.2f}")
        if _HAS15 and np.isfinite(row.get("p15_tau1", np.nan)):
            g2 = np.linspace(0.0, max(float(row.get("p15_span", gmax)), 1e-3), 200)
            ax.plot(g2, _2expc18(row["p15_lA1"], row["p15_tau1"], row["p15_lA2"], row["p15_tau2"], g2), "--",
                    color="#7f3b08", lw=1.4, zorder=3,
                    label=f"P15 2exp {row['p15_tau1']:.2f} / {row['p15_tau2']:.1f}\n$f_{{slow}}$={row['p15_fslow']:.3f}")
        if _FLQ is not None:
            ax.axhline(_FLQ, color="0.75", lw=1.0, ls=":")
        ax.set_title(f"{row['box']} s{int(row['snap'])} g{int(row['gal_id'])}"
                     + (f" [{row['pop']}]" if "pop" in row else "")
                     + f"  {row[f'{_Q}_cls']}" + (f"  {note}" if note else ""),
                     fontsize=7.5, color=P18_CCOL.get(row[f"{_Q}_cls"], "k"))
        ax.grid(alpha=0.2, lw=0.5)
        _leg18(ax, loc="best", fontsize=6.2)
        _yv = y[np.isfinite(y)]
        if len(_yv):
            ax.set_ylim(np.min(_yv) - 0.35, np.max(_yv) + 0.45)

    if len(_fit18):
        _key18 = np.where(np.isfinite(_fit18[f"{_Q}_tau0"]), _fit18[f"{_Q}_tau0"], 1e3)
        if P18_GMODE == "random":
            _idx = _rng18.choice(len(_fit18), size=min(P18_NGAL, len(_fit18)), replace=False)
        elif P18_GMODE == "worst":
            _idx = np.argsort(-np.nan_to_num(_fit18[f"{_Q}_rms0"].to_numpy(float)))[:P18_NGAL]
        else:
            _o = np.argsort(_key18)
            _idx = _o[np.unique(np.linspace(0, len(_o) - 1, min(P18_NGAL, len(_o))).astype(int))]
        _nc = 6
        _nr = int(np.ceil(len(_idx) / _nc))
        fig1, AX1 = plt.subplots(_nr, _nc, figsize=(3.7 * _nc, 2.5 * _nr), squeeze=False)
        for ax, i in zip(AX1.ravel(), _idx):
            _panel18(ax, _fit18.iloc[int(i)])
        for ax in AX1.ravel()[len(_idx):]:
            ax.set_axis_off()
        for ax in AX1[-1]:
            ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]", fontsize=9)
        for ax in AX1[:, 0]:
            ax.set_ylabel(P18_QLAB.get(_Q, _Q), fontsize=9)
        fig1.suptitle(f"The decay of {P18_QSH.get(_Q, _Q)} after the truncation, galaxy by galaxy — "
                      + {"ladder": f"a ladder of {len(_idx)} branches from the FASTEST decay to the slowest",
                         "random": f"{len(_idx)} branches drawn at random",
                         "worst": f"the {len(_idx)} branches the single exponential fits worst"}[P18_GMODE]
                      + "\nevery snapshot after the truncation is fitted: circles off the floor, filled triangles ON it"
                      + (", open triangles censored" if P18_CENSOR else "")
                      + (", grey excluded by P18_WIN" if P18_WIN != "all" else "")
                      + ";  red = the broken fit, blue dotted = the single one"
                      + (", brown dashed = Part 15's double exponential" if _HAS15 else ""),
                      y=1.004, fontsize=12, fontweight="bold")
        fig1.tight_layout(); paper_save(fig1, "kappa_decay_gallery"); plt.show()

        # ── FIG 2 — the same, one row per shape class: what each class actually looks like ──
        _CLD = [c for c in P18_CLS if c != "nofit" and (_GD18[f"{_Q}_cls"] == c).sum() >= 1]
        if _CLD:
            fig2, AX2 = plt.subplots(len(_CLD), P18_NCLS, figsize=(3.5 * P18_NCLS, 2.4 * len(_CLD)), squeeze=False)
            for ir, c in enumerate(_CLD):
                _sc = _GD18[_GD18[f"{_Q}_cls"] == c]
                _pick = _rng18.choice(len(_sc), size=min(P18_NCLS, len(_sc)), replace=False)
                for ax, i in zip(AX2[ir], _pick):
                    _panel18(ax, _sc.iloc[int(i)])
                for ax in AX2[ir][len(_pick):]:
                    ax.set_axis_off()
                AX2[ir, 0].set_ylabel(f"{c}\n({int(len(_sc))})", fontsize=8.5, color=P18_CCOL.get(c, "k"),
                                      fontweight="bold")
            for ax in AX2[-1]:
                ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]", fontsize=9)
            fig2.suptitle(f"What each shape class is — {P18_NCLS} branches drawn from each, same symbols as above",
                          y=1.004, fontsize=12, fontweight="bold")
            fig2.tight_layout(); paper_save(fig2, "kappa_decay_classes"); plt.show()

    # ══ (D) THE AUDIT — is the number a measurement? ═════════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(b) IS THE TIMESCALE MEASURED OR MANUFACTURED — the fit against the snapshot spacing, the window and the")
    print("    residual, and Part 15's double exponential on the same branches")
    print(f"  {'box':10s} {'N fit':>6s} {'dt':>6s} {'tau0':>7s} {'tau0/dt':>8s} {'<%.0fdt' % P18_RESN:>6s} "
          f"{'rms0':>6s} {'rms':>6s} {'BIC+':>6s} {'e(s0)/s0':>9s} {'e(s1)/s1':>9s} {'e(s2)/s2':>9s}"
          + ("  |  P15: {:>6s} {:>7s} {:>6s} {:>6s}".format("tau1", "tau2", "rail", "fs<1%") if _HAS15 else ""))
    for b in _BB18:
        g = D18[D18.box == b]
        _t0v = g[f"{_Q}_tau0"].to_numpy(float); _dtv = g[f"{_Q}_dt"].to_numpy(float)
        _nf = int(np.isfinite(g[f"{_Q}_s0"]).sum())
        _e0 = np.abs(g[f"{_Q}_e_s0"].to_numpy(float) / g[f"{_Q}_s0"].to_numpy(float))
        _e1 = np.abs(g[f"{_Q}_e_s1"].to_numpy(float) / g[f"{_Q}_s1"].to_numpy(float))
        _e2 = np.abs(g[f"{_Q}_e_s2"].to_numpy(float) / g[f"{_Q}_s2"].to_numpy(float))
        _db = g[f"{_Q}_dbic"].to_numpy(float)
        _fin = np.isfinite(_t0v) & np.isfinite(_dtv)
        _row = (f"  {b:10s} {_nf:6d} {np.nanmedian(_dtv):6.3f} {np.nanmedian(_t0v):7.3f} "
                f"{np.nanmedian(_t0v / _dtv):8.2f} {100 * np.mean(_t0v[_fin] < P18_RESN * _dtv[_fin]) if _fin.any() else np.nan:5.0f}% "
                f"{np.nanmedian(g[f'{_Q}_rms0']):6.2f} {np.nanmedian(g[f'{_Q}_rms']):6.2f} "
                f"{100 * np.nanmean(_db[np.isfinite(_db)] > 0) if np.isfinite(_db).any() else np.nan:5.0f}% "
                f"{np.nanmedian(_e0):9.2f} {np.nanmedian(_e1):9.2f} {np.nanmedian(_e2):9.2f}")
        if _HAS15:
            _p1, _p2 = g["p15_tau1"].to_numpy(float), g["p15_tau2"].to_numpy(float)
            _f15 = np.isfinite(_p1) & np.isfinite(_p2)
            _fs = g["p15_fslow"].to_numpy(float)
            _rail = np.mean((_p1[_f15] <= 0.0501) | (_p2[_f15] >= 40.0)) if _f15.any() else np.nan
            _row += (f"  |  {np.nanmedian(_p1):6.2f} {np.nanmedian(_p2):7.2f} {100 * _rail:5.0f}% "
                     f"{100 * np.mean(_fs[np.isfinite(_fs)] < 0.01) if np.isfinite(_fs).any() else np.nan:5.0f}%")
        print(_row)
        ROW18.append(dict(box=b, block=f"audit_{_Q}", quantity="tau0_over_dt", med=float(np.nanmedian(_t0v / _dtv)),
                          lo=float(np.mean(_t0v[_fin] < P18_RESN * _dtv[_fin])) if _fin.any() else np.nan,
                          hi=float(np.nanmedian(_e0)), n_gal=_nf))
    if P18_ALT and "alt_tau0" in D18.columns:
        print(f"\n  THE TREATMENT OF THE FLOOR, audited — the same branches fitted the other way (window stopped at the")
        print(f"  branch minimum, the floored points censored as upper limits) against the one in force above")
        print(f"  {'box':10s} {'N fit':>6s} {'N alt':>6s} {'window':>7s} {'alt':>6s} {'tau0':>7s} {'alt':>6s} "
              f"{'tau1':>7s} {'alt':>6s} {'tau2':>7s} {'alt':>6s} {'no tau2':>8s} {'alt':>6s}")
        for b in _BB18:
            g = D18[D18.box == b]
            _n2 = lambda c: float(np.nanmedian(g[c].to_numpy(float)))
            _fl = lambda c: (lambda v: float(np.mean(np.isfinite(v) & (v <= 1e-6))) if len(v) else np.nan)(
                g[c].to_numpy(float))
            print(f"  {b:10s} {int(np.isfinite(g[f'{_Q}_s0']).sum()):6d} {int(np.isfinite(g['alt_s0']).sum()):6d} "
                  f"{_n2(f'{_Q}_t_end'):7.2f} {_n2('alt_t_end'):6.2f} {_n2(f'{_Q}_tau0'):7.3f} {_n2('alt_tau0'):6.3f} "
                  f"{_n2(f'{_Q}_tau1'):7.3f} {_n2('alt_tau1'):6.3f} {_n2(f'{_Q}_tau2'):7.3f} {_n2('alt_tau2'):6.3f} "
                  f"{100 * _fl(f'{_Q}_s2'):7.0f}% {100 * _fl('alt_s2'):5.0f}%")
            ROW18.append(dict(box=b, block=f"treatment_{_Q}", quantity="tau0", med=_n2(f"{_Q}_tau0"),
                              lo=_n2("alt_tau0"), hi=_n2(f"{_Q}_t_end"), n_gal=int(np.isfinite(g[f"{_Q}_s0"]).sum())))
        print( "  ('no tau2' is the share whose second leg is flat or rising, i.e. has no decay time.  Taking every")
        print( "   snapshot makes the window longer and the branches that reached zero dust end on a FLAT leg — that is")
        print( "   the difference between the two columns, and it is the honest statement that nothing is left to lose.)")
    print(f"  (tau0/dt is one e-folding in snapshot steps.  Under {P18_RESN:g} the e-folding is shorter than the spacing: the")
    print( "   SLOPE is still fitted from the several points that sample the fall — e(s)/s, the fit's own relative error on")
    print( "   it, says how well — but nothing inside one e-folding is resolved, so a mode piling up there is a limit and")
    print( "   not a measurement.  rms0 / rms are the residuals of the single and the broken fit in dex, BIC+ the share")
    print( "   where the break is worth its two parameters." + ("" if _HAS15 else ")"))
    if _HAS15:
        print( "   P15 'rail' = the share of Part 15's fits sitting on a bound (tau1 <= 0.05 or tau2 >= 40 Gyr), 'fs<1%'")
        print( "   the share whose slow term carries under 1 % of the amplitude — both mean it was never identified.)")

    # ── FIG 3 — the audit, drawn: the resolution, Part 15's fit against this one, and the anchor trend ──
    fig3, AX3 = plt.subplots(3, len(_BB18), figsize=(4.0 * len(_BB18), 8.4), squeeze=False)
    for jb, b in enumerate(_BB18):
        g = D18[D18.box == b]
        ax = AX3[0, jb]
        _r = (g[f"{_Q}_tau0"] / g[f"{_Q}_dt"]).replace([np.inf, -np.inf], np.nan).dropna().to_numpy()
        if len(_r):
            ax.hist(np.clip(np.log10(_r[_r > 0]), -1.0, 2.0), bins=np.arange(-1.0, 2.01, 0.1), color="0.75",
                    edgecolor="w")
            ax.axvline(np.log10(P18_RESN), color="#d6604d", lw=1.8, ls="--",
                       label=f"{P18_RESN:g} snapshot steps\n{100 * np.mean(_r < P18_RESN):.0f} % below")
        ax.set_xlabel(r"$\log\,\tau_0 / \Delta t_{\rm snap}$", fontsize=9); _leg18(ax, loc="upper right")
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        if jb == 0:
            ax.set_ylabel("THE RESOLUTION\nbranches", fontsize=9)
        ax = AX3[1, jb]
        if _HAS15:
            _x = np.log10(g[f"{_Q}_tau0"].to_numpy(float))
            for _c15, _cc, _lb in (("p15_tau1", "#4393c3", r"P15 $\tau_1$"), ("p15_tau2", "#7f3b08", r"P15 $\tau_2$")):
                _y = np.log10(np.clip(g[_c15].to_numpy(float), 1e-3, None))
                _m = np.isfinite(_x) & np.isfinite(_y)
                if _m.sum():
                    ax.scatter(_x[_m], _y[_m], s=8, alpha=0.35, color=_cc, edgecolor="none", label=f"{_lb} ({_m.sum()})")
            ax.axhline(np.log10(0.05), color="#4393c3", lw=1.0, ls=":")
            ax.axhline(np.log10(50.0), color="#7f3b08", lw=1.0, ls=":")
            _lo, _hi = -1.6, 1.9
            ax.plot([_lo, _hi], [_lo, _hi], color="0.5", lw=1.0, ls="--")
            _xf = _x[np.isfinite(_x)]
            _x0, _x1 = (np.percentile(_xf, [0.5, 99.5]) if len(_xf) > 10 else (_lo, 0.9))
            ax.set_xlim(_x0 - 0.15, _x1 + 0.15); ax.set_ylim(_lo, _hi)
        else:
            ax.text(0.5, 0.5, "Part 15 not run in this kernel", ha="center", va="center", fontsize=8, color="0.5",
                    transform=ax.transAxes)
        ax.set_xlabel(r"$\log\,\tau_0$  this part  [Gyr]", fontsize=9); _leg18(ax, loc="lower right")
        if jb == 0:
            ax.set_ylabel("PART 15's FIT\n" + r"$\log\,\tau$  [Gyr]   (dotted = its bounds)", fontsize=9)
        ax = AX3[2, jb]
        for _q18, _cc, _lb in ((f"{_Q}_tau0", "k", r"$\tau_0$"), (f"{_Q}_t1dex", "#5aae61", r"$t_{\rm 1dex}$"),
                               (f"{_Q}_dt", "#d6604d", r"$\Delta t_{\rm snap}$")):
            _u, _mm = [], []
            for _sn, _ga in g.groupby("snap"):
                _yv = np.log10(_ga[_q18].to_numpy(float))
                _yv = _yv[np.isfinite(_yv)]
                if len(_yv) < 5 or "z" not in _ga.columns:
                    continue
                _u.append(float(np.nanmedian(_ga["z"].to_numpy(float)))); _mm.append(float(np.median(_yv)))
                ROW18.append(dict(box=b, block=f"anchor_{_Q}", quantity=_q18, x=_u[-1], med=_mm[-1],
                                  n_gal=int(len(_yv))))
            if len(_u) > 1:
                _o = np.argsort(_u)
                ax.plot(np.array(_u)[_o], np.array(_mm)[_o], "o-", color=_cc, ms=4, lw=1.6, label=_lb)
        ax.set_xlabel("anchor redshift", fontsize=9); _leg18(ax, loc="best")
        if jb == 0:
            ax.set_ylabel("THE ANCHOR\nmedian $\\log$ [Gyr]", fontsize=9)
        for _a in AX3[:, jb]:
            _a.grid(alpha=0.2, lw=0.5)
    fig3.suptitle("Is the timescale a measurement? — the decay time against the snapshot spacing (top), against Part 15's "
                  "double exponential (middle),\nand along the anchor ladder (bottom: the spacing itself moves by a "
                  "factor 2.5, so any distribution pooled over anchors is a mixture)",
                  y=1.005, fontsize=12, fontweight="bold")
    fig3.tight_layout(); paper_save(fig3, "kappa_decay_audit"); plt.show()

    # ══ (E) THE DISTRIBUTIONS — is log tau unimodal? ═════════════════════════════════════════════════════════════════
    def _gmm18(x, k=2, n_init=8, iters=400, seed=0):
        """a 1-D Gaussian mixture by EM, the best of n_init starts -> (w, mu, sd, loglike). Descriptive, not a test."""
        x = np.asarray(x, float); x = x[np.isfinite(x)]
        n = len(x)
        if n < 4 * k:
            return None
        rng = np.random.default_rng(seed); best = None
        sd0 = max(float(np.std(x)), 1e-6)
        for _ in range(n_init):
            mu = rng.choice(x, size=k, replace=False) + 0.05 * sd0 * rng.standard_normal(k)
            sd, w, ll_old = np.full(k, sd0), np.full(k, 1.0 / k), -np.inf
            for _it in range(iters):
                P = w * np.exp(-0.5 * ((x[:, None] - mu) / sd) ** 2) / (sd * np.sqrt(2 * np.pi))
                s = np.where(P.sum(1) > 0, P.sum(1), 1e-300)
                ll = float(np.sum(np.log(s)))
                R = P / s[:, None]
                nk = R.sum(0) + 1e-12
                w, mu = nk / n, (R * x[:, None]).sum(0) / nk
                sd = np.maximum(np.sqrt(np.maximum((R * (x[:, None] - mu) ** 2).sum(0) / nk, 1e-6)), 0.02 * sd0)
                if abs(ll - ll_old) < 1e-8 * max(1.0, abs(ll)):
                    break
                ll_old = ll
            if best is None or ll > best[-1]:
                o = np.argsort(mu); best = (w[o], mu[o], sd[o], ll)
        return best

    def _nmod18(x, h, lo=None, hi=None, m=512):
        """the number of modes of the Gaussian KDE of x at bandwidth h — binned, so the cost does not grow with n."""
        lo = (float(np.min(x)) - 4 * h) if lo is None else lo
        hi = (float(np.max(x)) + 4 * h) if hi is None else hi
        cnt, edges = np.histogram(x, bins=m, range=(lo, hi))
        dx = edges[1] - edges[0]
        k = np.arange(-int(np.ceil(4 * h / dx)), int(np.ceil(4 * h / dx)) + 1) * dx
        d = np.convolve(cnt.astype(float), np.exp(-0.5 * (k / h) ** 2), mode="same")
        return int(np.sum((d[1:-1] > d[:-2]) & (d[1:-1] >= d[2:])))

    def _silv18(x, k=1, nboot=None, seed=2):
        """SILVERMAN'S CRITICAL-BANDWIDTH TEST of 'at most k modes': the smallest bandwidth whose KDE still shows k modes,
        calibrated by the smoothed bootstrap AT that bandwidth. A small p means the data have more than k modes."""
        nboot = P18_SILV if nboot is None else nboot
        x = np.asarray(x, float); x = x[np.isfinite(x)]
        n = len(x)
        if n < P18_MINDIST or nboot <= 0:
            return dict(n=n, h=np.nan, p=np.nan)
        lo, hi = 1e-3 * np.std(x), 10 * np.std(x)
        for _ in range(40):
            m = 0.5 * (lo + hi)
            if _nmod18(x, m) > k:
                lo = m
            else:
                hi = m
        h = 0.5 * (lo + hi)
        rng = np.random.default_rng(seed); sd = float(np.std(x, ddof=1)); cnt = 0
        _l, _h = float(np.min(x)) - 6 * h, float(np.max(x)) + 6 * h
        for _ in range(int(nboot)):
            i = rng.integers(0, n, n)
            y = x[i] + h * rng.standard_normal(n)
            y = np.mean(x[i]) + (y - np.mean(y)) / np.sqrt(1 + h ** 2 / sd ** 2)      # the variance-corrected version
            cnt += int(_nmod18(y, h, _l, _h) > k)
        return dict(n=n, h=float(h), p=float((1 + cnt) / (1 + int(nboot))))

    def _shape18(x):
        """the descriptive summary of one distribution: the moments, the mixture, and whether the mixture is two-humped."""
        x = np.asarray(x, float); x = x[np.isfinite(x)]
        out = dict(n=len(x), med=np.nan, sd=np.nan, skew=np.nan, w2=np.nan, mu1=np.nan, mu2=np.nan, sd1=np.nan,
                   sd2=np.nan, ash=np.nan, modes=np.nan, dbic=np.nan)
        if len(x) < P18_MINDIST:
            return out
        out.update(med=float(np.median(x)), sd=float(np.std(x, ddof=1)), skew=float(_st18.skew(x)))
        g = _gmm18(x, 2)
        if g is None:
            return out
        w, mu, sd, ll2 = g
        m0, s0 = float(np.mean(x)), float(np.std(x, ddof=1))
        ll1 = float(np.sum(_st18.norm.logpdf(x, m0, s0)))
        gg = np.linspace(x.min() - 1, x.max() + 1, 2000)
        dd = w[0] * _st18.norm.pdf(gg, mu[0], sd[0]) + w[1] * _st18.norm.pdf(gg, mu[1], sd[1])
        out.update(w2=float(w[1]), mu1=float(mu[0]), mu2=float(mu[1]), sd1=float(sd[0]), sd2=float(sd[1]),
                   ash=float(np.sqrt(2) * abs(mu[1] - mu[0]) / np.sqrt(sd[0] ** 2 + sd[1] ** 2)),
                   modes=int(np.sum((dd[1:-1] > dd[:-2]) & (dd[1:-1] >= dd[2:]))),
                   dbic=float((2 * np.log(len(x)) - 2 * ll1) - (5 * np.log(len(x)) - 2 * ll2)))
        return out

    def _lt18(g, q):
        """log10 of a timescale column, with the infinities (a flat or rising leg) dropped and counted separately."""
        v = g[q].to_numpy(float)
        return np.log10(v[np.isfinite(v) & (v > 0)])

    def _centre18(g, q):
        """the same, with each anchor centred on its OWN median: the anchor ladder cannot make a mode here."""
        v = np.log10(g[q].to_numpy(float).clip(min=1e-9))
        v[~np.isfinite(g[q].to_numpy(float))] = np.nan
        s = g["snap"].to_numpy()
        out = np.full(len(v), np.nan)
        for u in np.unique(s):
            m = (s == u) & np.isfinite(v)
            if m.sum() >= 5:
                out[m] = v[m] - np.median(v[m])
        return out[np.isfinite(out)]

    _TQ18 = [(f"{_Q}_tau0", r"$\tau_0$  (one exponential)"), (f"{_Q}_tau1", r"$\tau_1$  (the first leg)"),
             (f"{_Q}_tau2", r"$\tau_2$  (the second leg)"), (f"{_Q}_t1dex", r"$t_{\rm 1dex}$  (model-free)")]
    _ESL18 = {f"{_Q}_tau0": f"{_Q}_e_s0", f"{_Q}_tau1": f"{_Q}_e_s1", f"{_Q}_tau2": f"{_Q}_e_s2"}
    _SSL18 = {f"{_Q}_tau0": f"{_Q}_s0", f"{_Q}_tau1": f"{_Q}_s1", f"{_Q}_tau2": f"{_Q}_s2"}

    def _good18(g, q):
        """the branches whose slope is measured to better than P18_EMAX in relative terms — the precision cut."""
        if q not in _ESL18:
            return np.ones(len(g), bool)
        with np.errstate(all="ignore"):
            return np.abs(g[_ESL18[q]].to_numpy(float) / g[_SSL18[q]].to_numpy(float)) <= P18_EMAX

    def _noneg18(g, q):
        """how many branches have that leg FITTED but flat or rising, i.e. no decay time at all."""
        if q not in _SSL18:
            return 0
        v = g[_SSL18[q]].to_numpy(float)
        return int(np.sum(np.isfinite(v) & (v <= 1e-6)))
    print("\n" + "-" * 132)
    print(f"(c) HOW THE TIMESCALES ARE DISTRIBUTED — {P18_QSH.get(_Q, _Q)}, log10 tau [Gyr].  The test is Silverman's")
    print(f"    critical bandwidth ({P18_SILV} bootstraps): p < 0.05 means MORE THAN ONE MODE.  It is run pooled, and again")
    print( "    after centring each anchor on its own median, because the anchor ladder alone spreads the medians by 0.3 dex.")
    print( "    The mixture beside it is DESCRIPTIVE: a 2-component fit beats one Gaussian on skewness alone, so what")
    print( "    matters is the Ashman D (a real gap needs D > 2) and whether the fitted mixture is itself two-humped.")
    print(f"  {'box':10s} {'tau':10s} {'N':>5s} {'med':>6s} {'sd':>5s} {'skew':>6s} {'<res':>5s} {'e/s':>5s} | "
          f"{'Silv p':>7s} {'centred':>8s} {'precise':>8s} {'per-anchor':>16s} | {'w2':>5s} {'mu1':>6s} {'mu2':>6s} "
          f"{'AshD':>5s} {'humps':>6s}")
    for b in _BB18:
        g = D18[D18.box == b]
        for q, lab in _TQ18:
            x = _lt18(g, q)
            if len(x) < P18_MINDIST:
                print(f"  {b:10s} {q.split('_', 1)[1]:10s} {len(x):5d}   — too few"); continue
            xc = _centre18(g, q)
            sv, svc, sh = _silv18(x), _silv18(xc), _shape18(x)
            _pa = []
            for _s18, _g18 in g.groupby("snap"):
                _xa = _lt18(_g18, q)
                if len(_xa) >= P18_NANCH:
                    _pa.append(_silv18(_xa)["p"])
            _tv, _dv = g[q].to_numpy(float), g[f"{_Q}_dt"].to_numpy(float)
            _fin = np.isfinite(_tv) & np.isfinite(_dv)
            _res = float(np.mean(_tv[_fin] < P18_RESN * _dv[_fin])) if _fin.any() else np.nan
            _gd = _lt18(g[_good18(g, q)], q)
            _er = np.abs(g[_ESL18[q]].to_numpy(float) / g[_SSL18[q]].to_numpy(float)) if q in _ESL18 else np.array([np.nan])
            _svg = _silv18(_gd) if len(_gd) >= P18_MINDIST else dict(p=np.nan)
            print(f"  {b:10s} {q.split('_', 1)[1]:10s} {len(x):5d} {sh['med']:+6.2f} {sh['sd']:5.2f} {sh['skew']:+6.2f} "
                  f"{100 * _res:4.0f}% {np.nanmedian(_er):5.2f} | {sv['p']:7.3f} {svc['p']:8.3f} "
                  f"{_svg['p']:8.3f} "
                  + (f"{np.min(_pa):.3f}-{np.max(_pa):.3f} ({len(_pa)})" if _pa else "—").rjust(16)
                  + f" | {sh['w2']:5.2f} {sh['mu1']:+6.2f} {sh['mu2']:+6.2f} {sh['ash']:5.2f} {sh['modes']:6.0f}")
            ROW18.append(dict(box=b, block=f"dist_{q}", quantity="silverman_precise", p=_svg["p"], n_gal=len(_gd)))
            ROW18.append(dict(box=b, block=f"dist_{q}", quantity="silverman", p=sv["p"], med=sh["med"], n_gal=len(x)))
            ROW18.append(dict(box=b, block=f"dist_{q}", quantity="silverman_centred", p=svc["p"], n_gal=len(xc)))
            ROW18.append(dict(box=b, block=f"dist_{q}", quantity="mixture", med=sh["mu1"], hi=sh["mu2"], lo=sh["w2"],
                              auc=sh["ash"], n_gal=int(sh["modes"]) if np.isfinite(sh["modes"]) else 0))
    print(f"  ('<res' is the share whose e-folding is shorter than {P18_RESN:g} snapshot step, 'e/s' the median relative error")
    print(f"   the fit puts on the slope, and 'precise' the same Silverman test run on the branches with e/s < {P18_EMAX:g}")
    print( "   alone — if a mode is real it has to survive there too.  For tau_2 read that column with care: a leg that is")
    print( "   nearly flat has a large RELATIVE error by construction, so the cut removes the slow tail it is testing.)")

    # ── FIG 4 — the distributions themselves ──
    _BINS18 = np.arange(P18_TAUAX[0], P18_TAUAX[1] + 1e-9, P18_TBIN)
    fig4, AX4 = plt.subplots(len(_TQ18), len(_BB18), figsize=(4.2 * len(_BB18), 2.9 * len(_TQ18)), squeeze=False)
    for jb, b in enumerate(_BB18):
        g = D18[D18.box == b]
        for iq, (q, lab) in enumerate(_TQ18):
            ax = AX4[iq, jb]
            x = _lt18(g, q)
            if len(x) < 8:
                ax.set_axis_off(); continue
            xc = _centre18(g, q)
            _n_inf = _noneg18(g, q)
            ax.hist(np.clip(x, _BINS18[0], _BINS18[-1]), bins=_BINS18, color="0.78", edgecolor="w", lw=0.4,
                    label=f"all ({len(x)})")
            if len(xc) >= P18_MINDIST:
                ax.hist(np.clip(xc + np.median(x), _BINS18[0], _BINS18[-1]), bins=_BINS18, histtype="step",
                        color="#1a1a1a", lw=1.4, ls="--", label="anchor-centred")
            _xg = _lt18(g[_good18(g, q)], q)
            if len(_xg) >= 8 and len(_xg) < len(x):
                ax.hist(np.clip(_xg, _BINS18[0], _BINS18[-1]), bins=_BINS18, histtype="step", color="#5aae61",
                        lw=1.6, label=f"e/s < {P18_EMAX:g} ({len(_xg)})")
            for _p18 in ("dusty", "pure"):
                if "pop" not in g.columns:
                    break
                _xp = _lt18(g[g["pop"] == _p18], q)
                if len(_xp) >= 8:
                    ax.hist(np.clip(_xp, _BINS18[0], _BINS18[-1]), bins=_BINS18, histtype="step",
                            color=POPS[_p18]["color"], lw=1.8, label=f"{_p18} ({len(_xp)})")
            _sv = _silv18(x)
            gg = np.linspace(_BINS18[0], _BINS18[-1], 500)
            _h0 = 0.9 * min(np.std(x, ddof=1), (np.percentile(x, 75) - np.percentile(x, 25)) / 1.349) * len(x) ** -0.2
            if _h0 > 0:
                _d = np.exp(-0.5 * ((gg[:, None] - x[None, :]) / _h0) ** 2).sum(1)
                ax.plot(gg, _d * len(x) * P18_TBIN / max(np.sum(_d) * (gg[1] - gg[0]), 1e-12), color="k", lw=1.6)
            if np.isfinite(_sv["h"]) and np.isfinite(_sv["p"]) and _sv["p"] < 0.05:
                _d = np.exp(-0.5 * ((gg[:, None] - x[None, :]) / _sv["h"]) ** 2).sum(1)
                ax.plot(gg, _d * len(x) * P18_TBIN / max(np.sum(_d) * (gg[1] - gg[0]), 1e-12), color="#7f3b08", lw=1.0,
                        ls="--", label=f"one mode only at h = {_sv['h']:.2f}")
            _sh = _shape18(x)
            if np.isfinite(_sh["w2"]) and min(_sh["sd1"], _sh["sd2"]) > 0.03:
                for _w, _m, _s in ((1 - _sh["w2"], _sh["mu1"], _sh["sd1"]), (_sh["w2"], _sh["mu2"], _sh["sd2"])):
                    ax.plot(gg, _w * _st18.norm.pdf(gg, _m, _s) * len(x) * P18_TBIN, color="#762a83", lw=1.0, ls=":")
            _rl = np.log10(P18_RESN * np.nanmedian(g[f"{_Q}_dt"].to_numpy(float)))
            ax.axvspan(_BINS18[0], _rl, color="#d6604d", alpha=0.045, lw=0)
            ax.axvline(_rl, color="#d6604d", lw=1.2, ls="--")
            if _n_inf > 0:
                ax.annotate(f"+{_n_inf} flat or rising:\nno decay time", xy=(0.02, 0.70), xycoords="axes fraction",
                            fontsize=6.5, color="0.35")
            ax.set_title((f"{BOX_LABEL[b]}\n" if iq == 0 else "") + f"{lab}   Silverman p = {_sv['p']:.3f}"
                         + (f",  centred {_silv18(xc)['p']:.3f}" if len(xc) >= P18_MINDIST else ""),
                         fontsize=8.5, color="#7f3b08" if b == NOX_BOX else "k")
            ax.set_xlim(_BINS18[0], _BINS18[-1]); ax.grid(alpha=0.2, lw=0.5)
            if iq == len(_TQ18) - 1:
                ax.set_xlabel(r"$\log_{10}\,\tau$  [Gyr]", fontsize=9.5)
            if jb == 0:
                ax.set_ylabel("branches", fontsize=9)
            _leg18(ax, loc="upper right", fontsize=6.2)
    fig4.suptitle(f"How the decay timescales of {P18_QSH.get(_Q, _Q)} are distributed — grey = all, black dashed = the "
                  "same after each anchor is centred on its own median, green = the well-measured branches,\nred/blue = "
                  "the anchor dust label;  black = the KDE, brown dashed = the KDE at the bandwidth where the extra mode "
                  "dies, violet dotted = the 2-component mixture (descriptive)\nthe shaded band on the left is where one "
                  f"e-folding is shorter than {P18_RESN:g} snapshot step",
                  y=1.005, fontsize=12, fontweight="bold")
    fig4.tight_layout(); paper_save(fig4, "kappa_decay_taudist"); plt.show()

    # ── FIG 5 — the two legs together: the plane where the shape lives, and where the break falls ──
    fig5, AX5 = plt.subplots(2, len(_BB18), figsize=(4.0 * len(_BB18), 7.4), squeeze=False)
    for jb, b in enumerate(_BB18):
        g = D18[D18.box == b]
        ax = AX5[0, jb]
        _x = np.log10(g[f"{_Q}_tau1"].to_numpy(float)); _y = np.log10(g[f"{_Q}_tau2"].to_numpy(float))
        _br = np.isfinite(g[f"{_Q}_dbic"].to_numpy(float)) & (g[f"{_Q}_dbic"].to_numpy(float) > 0)
        for _p18, _mk in (("pure", "o"), ("dusty", "o")):
            if "pop" not in g.columns:
                break
            _m = (g["pop"].to_numpy() == _p18) & np.isfinite(_x) & np.isfinite(_y)
            if _m.sum():
                ax.scatter(_x[_m & ~_br], _y[_m & ~_br], s=9, alpha=0.25, color=POPS[_p18]["color"], edgecolor="none")
                ax.scatter(_x[_m & _br], _y[_m & _br], s=20, alpha=0.7, color=POPS[_p18]["color"], edgecolor="none",
                           label=f"{_p18}, break preferred ({int((_m & _br).sum())})")
                ax.plot([np.nanmedian(_x[_m])], [np.nanmedian(_y[_m])], "*", ms=15, color=POPS[_p18]["color"], mec="k",
                        mew=0.6, zorder=6)
            # the branches that STOPPED falling have no tau_2 at all: they belong at the top of the axis, not off it
            _mr = ((g["pop"].to_numpy() == _p18) & np.isfinite(_x) & _br
                   & np.isfinite(g[f"{_Q}_s2"].to_numpy(float)) & (g[f"{_Q}_s2"].to_numpy(float) <= 1e-6))
            if _mr.sum():
                ax.plot(_x[_mr], np.full(int(_mr.sum()), P18_TAUAX[1] - 0.06), "^", ms=5, alpha=0.7,
                        color=POPS[_p18]["color"], mec="none",
                        label=f"{_p18}, second leg flat or rising ({int(_mr.sum())})")
        _lo, _hi = P18_TAUAX
        ax.plot([_lo, _hi], [_lo, _hi], color="0.5", lw=1.1, ls="--")
        ax.annotate("steep then shallow", xy=(0.04, 0.94), xycoords="axes fraction", fontsize=7.5, color="#d6604d",
                    fontweight="bold")
        ax.annotate("shallow then steep", xy=(0.96, 0.06), xycoords="axes fraction", ha="right", fontsize=7.5,
                    color="#4393c3", fontweight="bold")
        ax.axhline(P18_TAUAX[1] - 0.06, color="0.7", lw=0.8, ls=":")
        ax.set_xlim(_lo, _hi); ax.set_ylim(_lo, _hi + 0.02); ax.grid(alpha=0.2, lw=0.5)
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.set_xlabel(r"$\log\,\tau_1$  the first leg  [Gyr]", fontsize=9)
        if jb == 0:
            ax.set_ylabel(r"$\log\,\tau_2$  the second leg  [Gyr]", fontsize=9)
        _leg18(ax, loc="lower left", fontsize=6.5)
        ax = AX5[1, jb]
        for c in ("steep-shallow", "rebound", "shallow-steep", "single"):
            _tb = g[g[f"{_Q}_cls"] == c][f"{_Q}_tb"].to_numpy(float)
            _tb = _tb[np.isfinite(_tb)]
            if len(_tb) >= 5:
                ax.hist(np.clip(_tb, 0, 4), bins=np.arange(0, 4.01, 0.2), histtype="step", lw=1.8,
                        color=P18_CCOL[c], label=f"{c} ({len(_tb)}, med {np.median(_tb):.2f})")
        ax.set_xlabel(r"$t_{\rm brk} - t_{\rm SFT}$  [Gyr]", fontsize=9); ax.grid(alpha=0.2, lw=0.5)
        if jb == 0:
            ax.set_ylabel("branches", fontsize=9)
        _leg18(ax, loc="upper right", fontsize=6.5)
    fig5.suptitle("The two legs of every branch (top) and when the break falls (bottom) — above the diagonal the decay "
                  "flattens, below it the branch held its dust and then lost it\n(faint points: the break is not "
                  "preferred over a single exponential, and they scatter along the diagonal as they should;  the "
                  "triangles on the top edge STOPPED falling and have no second timescale at all)",
                  y=1.005, fontsize=12, fontweight="bold")
    fig5.tight_layout(); paper_save(fig5, "kappa_decay_plane"); plt.show()

    # ══ (F) IS THE SLOW DECAY THE DUSTY ONE? ════════════════════════════════════════════════════════════════════════
    def _resid18(g, col, by="snap"):
        """a column with each anchor centred on its own median — the plain way to ask a question at fixed epoch."""
        v = g[col].to_numpy(float); s = g[by].to_numpy()
        out = np.full(len(v), np.nan)
        for u in np.unique(s):
            m = (s == u) & np.isfinite(v)
            if m.sum() >= 5:
                out[m] = v[m] - np.median(v[m])
        return out

    print("\n" + "-" * 132)
    print("(d) IS THE SLOW DECAY THE DUSTY ONE — the timescale against the dust label, raw and inside anchor cells.")
    print("    CIRCULARITY: the dust's OWN tau against the dust it ends with is close to a tautology (a slower decay from")
    print("    the same start ends higher), so it is marked (*) and the honest tests are the OTHER two quantities — the")
    print("    sSFR and the H2 decay against the dust label — plus tau at fixed starting dust.")
    print(f"  {'box':10s} {'decay of':9s} {'tau':6s} {'N dusty':>8s} {'N pure':>7s} {'med dusty':>10s} {'med pure':>9s} "
          f"{'AUC':>6s} {'p':>9s} {'AUC | anchor':>13s} {'rho(tau, dust)':>15s}")
    for b in _BB18:
        g = D18[D18.box == b]
        if "pop" not in g.columns:
            continue
        for q in P18_DQ:
            for tk in ("tau0", "tau2"):
                col = f"{q}_{tk}"
                if col not in g.columns:
                    continue
                lv = np.log10(g[col].to_numpy(float))
                lv[~np.isfinite(g[col].to_numpy(float))] = np.nan
                m1 = (g["pop"].to_numpy() == "dusty")
                au, p_, na, nb = mw_auc(lv, m1)
                if na < 8 or nb < 8:
                    continue
                _gg = g.copy(); _gg["_l"] = lv
                auz = mw_auc(_resid18(_gg, "_l"), m1)[0]
                _rr = np.nan
                if "lfd" in g.columns:
                    _ok = np.isfinite(lv) & np.isfinite(g["lfd"].to_numpy(float))
                    if _ok.sum() > 10:
                        _rr = float(spearmanr(lv[_ok], g["lfd"].to_numpy(float)[_ok]).statistic)
                _star = "*" if q == "lfd" else " "
                print(f"  {b:10s} {P18_QSH.get(q, q):9s} {tk:6s}{_star}{na:7d} {nb:7d} "
                      f"{np.nanmedian(lv[m1]):+10.2f} {np.nanmedian(lv[~m1]):+9.2f} {au:6.3f} {p_:9.1e} "
                      f"{auz:13.3f} {_rr:15.2f}")
                ROW18.append(dict(box=b, block=f"dustlink_{q}", quantity=tk, auc=au, p=p_, med=auz, hi=_rr,
                                  n_gal=na + nb))
    print( "  (AUC = P(tau larger in a dusty QG); 0.5 is no link, above 0.5 means the dusty ones decay SLOWER.")
    print( "   'AUC | anchor' repeats it after centring each anchor on its own median — the anchor ladder alone can")
    print( "   produce the whole effect, since the dusty fraction runs from ~60 % at z = 3 to ~10 % at z = 0.)")

    # ── FIG 6 — the same, drawn ──
    _DQ6 = [q for q in P18_DQ if f"{q}_tau0" in D18.columns]
    fig6, AX6 = plt.subplots(2, len(_BB18), figsize=(4.0 * len(_BB18), 7.0), squeeze=False)
    for jb, b in enumerate(_BB18):
        g = D18[D18.box == b]
        ax = AX6[0, jb]
        _x = g["lfd"].to_numpy(float) if "lfd" in g.columns else np.full(len(g), np.nan)
        _y = np.log10(g[f"{_Q}_tau0"].to_numpy(float))
        _y[~np.isfinite(g[f"{_Q}_tau0"].to_numpy(float))] = np.nan
        _m = np.isfinite(_x) & np.isfinite(_y)
        if _m.sum() > 5:
            ax.scatter(_x[_m], _y[_m], s=10, alpha=0.3, color="0.45", edgecolor="none")
            _bn = np.quantile(_x[_m], np.linspace(0, 1, 7))
            _bn = np.unique(np.round(_bn, 3))
            _cx, _cy, _lo, _hi = [], [], [], []
            for _i in range(len(_bn) - 1):
                _s = _m & (_x >= _bn[_i]) & (_x <= _bn[_i + 1])
                if _s.sum() >= P18_MINDIST // 4:
                    _cx.append(float(np.median(_x[_s]))); _cy.append(float(np.median(_y[_s])))
                    _b = np.array([np.median(np.random.default_rng(k).choice(_y[_s], _s.sum())) for k in range(120)])
                    _lo.append(float(np.percentile(_b, 16))); _hi.append(float(np.percentile(_b, 84)))
                    ROW18.append(dict(box=b, block=f"tau_vs_dust_{_Q}", quantity="tau0", x=_cx[-1], med=_cy[-1],
                                      lo=_lo[-1], hi=_hi[-1], n_gal=int(_s.sum())))
            if _cx:
                ax.fill_between(_cx, _lo, _hi, color="#d6604d", alpha=0.25, lw=0)
                ax.plot(_cx, _cy, "o-", color="#d6604d", lw=2.2, ms=5)
            ax.axvline(DUSTY_CUT, color="0.35", lw=1.0, ls=":")
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.set_xlabel(r"$\log M_{\rm dust}/M_\star$ at the anchor", fontsize=9); ax.grid(alpha=0.2, lw=0.5)
        if jb == 0:
            ax.set_ylabel("THE CIRCULAR ONE\n" + r"$\log\,\tau_0$ of the dust  [Gyr]", fontsize=9)
        ax = AX6[1, jb]
        _any = False
        for q, _cc in zip([q for q in _DQ6 if q != "lfd"], ("#5aae61", "#762a83")):
            if "pop" not in g.columns:
                break
            for _p18, _ls in (("dusty", "-"), ("pure", "--")):
                _v = _lt18(g[g["pop"] == _p18], f"{q}_tau0")
                if len(_v) < 8:
                    continue
                _any = True
                ax.hist(np.clip(_v, P18_TAUAX[0], P18_TAUAX[1]), bins=np.arange(P18_TAUAX[0], P18_TAUAX[1] + 1e-9, 0.15),
                        histtype="step", lw=1.8, ls=_ls, color=_cc, density=True,
                        label=f"{P18_QSH.get(q, q)} — {_p18} ({len(_v)})")
            _au = mw_auc(np.log10(g[f"{q}_tau0"].to_numpy(float)), (g["pop"].to_numpy() == "dusty"))[0]
            if np.isfinite(_au):
                ax.annotate(f"{P18_QSH.get(q, q)}: AUC {_au:.2f}", xy=(0.03, 0.95 - 0.08 * _DQ6.index(q)),
                            xycoords="axes fraction", fontsize=7.5, color=_cc, fontweight="bold")
        if not _any:
            ax.set_axis_off()
        else:
            ax.set_xlabel(r"$\log\,\tau_0$  [Gyr]", fontsize=9); ax.grid(alpha=0.2, lw=0.5)
            if jb == 0:
                ax.set_ylabel("THE NON-CIRCULAR ONE\nsolid = dusty, dashed = dust-poor\ndensity", fontsize=9)
            _leg18(ax, loc="upper right", fontsize=6.2)
    fig6.suptitle("Is the slow decay the dusty one? — the dust's own timescale against the dust it ends with (top, "
                  "circular by construction and shown for completeness),\nand the same question asked of the sSFR and "
                  "the H2 decay, which the dust label does not enter (bottom)", y=1.005, fontsize=12, fontweight="bold")
    fig6.tight_layout(); paper_save(fig6, "kappa_decay_dustlink"); plt.show()

    # ══ (G) THE READING — the questions this part was opened with, answered off the numbers above ═══════════════════
    _R18 = pd.DataFrame(ROW18)
    print("\n" + "=" * 132)
    print("WHAT THE FITS SAY")
    _b0 = _BB18[0]
    _g0 = D18[D18.box == _b0]
    _cc0 = _g0[f"{_Q}_cls"].value_counts()
    _ss = 100 * _cc0.get("steep-shallow", 0) / max(len(_g0), 1)
    _sh = 100 * _cc0.get("shallow-steep", 0) / max(len(_g0), 1)
    print(f"  1. THE SHAPE.  Over every snapshot after the truncation, 'a steep decay followed by a shallower one' is the")
    print(f"     shape of {_ss:.0f} % of the {_b0} branches and the most common one; {_sh:.0f} % are its opposite (a plateau and then a")
    print( "     collapse) and the rest need only one exponential or end too soon to tell.")
    if "pop" in _g0.columns:
        _du = _g0[_g0["pop"] == "dusty"][f"{_Q}_cls"].value_counts()
        _pu = _g0[_g0["pop"] == "pure"][f"{_Q}_cls"].value_counts()
        _nd, _np_ = max((_g0["pop"] == "dusty").sum(), 1), max((_g0["pop"] == "pure").sum(), 1)
        _dss, _pss = 100 * _du.get("steep-shallow", 0) / _nd, 100 * _pu.get("steep-shallow", 0) / _np_
        _drb, _prb = 100 * _du.get("rebound", 0) / _nd, 100 * _pu.get("rebound", 0) / _np_
        print(f"     SPLIT BY THE DUST LABEL IT DOES DIVIDE: steep-then-shallow — both legs still falling — is "
              f"{_pss:.0f} % of the dust-poor")
        print(f"     and {_dss:.0f} % of the dusty; the branches that fall and then STOP falling are {_drb:.0f} % of the dusty against"
              f" {_prb:.0f} %.")
        _rb = _g0[_g0[f"{_Q}_cls"] == "rebound"]
        if len(_rb) and "pop" in _rb.columns:
            _o = []
            for _p18 in ("dusty", "pure"):
                _r18 = _rb[_rb["pop"] == _p18]
                if len(_r18) >= 5:
                    _o.append(f"{_p18} {np.nanmedian(_r18[f'{_Q}_reb']):.2f} dex back from the minimum, "
                              f"{100 * np.mean(_r18[f'{_Q}_n_fl'].to_numpy(float) > 0):.0f} % of them off the floor")
            if _o:
                print( "     And that arrest is not the same thing in the two:")
                for _l18 in _o:
                    print(f"       {_l18}")
                print( "     For the dusty QGs it is REGROWTH — a third of them reach (near) zero dust and then rebuild it.")
    _dbv = D18[f"{_Q}_dbic"].to_numpy(float)
    print(f"  2. ARE THE FITS RIGHT.  The broken exponential is preferred over one straight line for "
          f"{100 * np.mean(_dbv[np.isfinite(_dbv)] > 0) if np.isfinite(_dbv).any() else np.nan:.0f} % of the branches")
    print(f"     that can be tested, and it cuts the residual from {np.nanmedian(D18[f'{_Q}_rms0']):.2f} to "
          f"{np.nanmedian(D18[f'{_Q}_rms']):.2f} dex; the slopes carry a "
          f"{100 * np.nanmedian(np.abs(D18[f'{_Q}_e_s0'] / D18[f'{_Q}_s0'])):.0f} % error.")
    if _HAS15:
        _q1, _q2 = D18["p15_tau1"].to_numpy(float), D18["p15_tau2"].to_numpy(float)
        _qf = np.isfinite(_q1) & np.isfinite(_q2)
        _qs = D18["p15_fslow"].to_numpy(float); _qs = _qs[np.isfinite(_qs)]
        print(f"     Part 15's double exponential is NOT identified on the same data: "
              f"{100 * np.mean((_q1[_qf] <= 0.0501) | (_q2[_qf] >= 40)):.0f} % of its {int(_qf.sum())} fits sit on a bound and")
        print(f"     {100 * np.mean(_qs < 0.01):.0f} % give the slow term under 1 % of the amplitude — its tau1 / tau2"
              f" should not be quoted.")
    _sv18 = _R18[_R18.block == f"dist_{_Q}_tau0"] if "block" in _R18.columns else pd.DataFrame()

    def _pv(qn, bx):
        _r = _sv18[(_sv18.quantity == qn) & (_sv18.box == bx)] if len(_sv18) else []
        return float(_r["p"].iloc[0]) if len(_r) else np.nan

    _pp = {bx: _pv("silverman", bx) for bx in _BB18}
    _pg = {bx: _pv("silverman_precise", bx) for bx in _BB18}
    _rej = [bx for bx in _BB18 if np.isfinite(_pp[bx]) and _pp[bx] < 0.05]
    print( "  3. IS THERE A BIMODALITY.  log tau is right-skewed and not Gaussian in every box, with a long tail towards")
    print( "     slow decays.  Silverman's test on tau_0 — " + ", ".join(
        f"{bx} p = {_pp[bx]:.3f}" for bx in _BB18 if np.isfinite(_pp[bx])) + " —")
    if not _rej:
        print( "     does not reject one mode ANYWHERE, and neither does the model-free t_1dex.  ONE mode with a slow tail,")
        print( "     the dusty QGs living inside that tail: no second population, and nothing that a mixture would name.")
    else:
        print(f"     rejects one mode in {', '.join(_rej)}; on the well-measured branches alone it gives " + ", ".join(
            f"{bx} {_pg[bx]:.3f}" for bx in _BB18 if np.isfinite(_pg[bx])) + ".")
        print( "     Where the rejection does not survive that cut it is the imprecise tail talking, not the data.")
    _dl = _R18[(_R18.block == f"dustlink_{_Q}") & (_R18.quantity == "tau0")] if "block" in _R18.columns else []
    if len(_dl):
        print( "  4. THE DUST.  The dusty QGs decay more slowly IN THE DUST — AUC (and at fixed anchor) "
              + ", ".join(f"{r.box} {r.auc:.2f} ({r.med:.2f})" for r in _dl.itertuples() if np.isfinite(r.auc)) + " —")
        print( "     but that test is CIRCULAR: a slower decay from the same start ends with more dust.  The non-circular")
        print( "     versions are the same galaxies' OTHER decays, which the dust label does not enter:")
        _nc = _R18[(_R18.block.astype(str).str.startswith("dustlink_")) & (_R18.block != f"dustlink_{_Q}")
                   & (_R18.quantity == "tau0")]
        for bx in _BB18:
            _r = _nc[_nc.box == bx]
            if not len(_r):
                continue
            _lab = ", ".join(f"{str(r.block).split('_', 1)[1]} {r.auc:.2f}" for r in _r.itertuples() if np.isfinite(r.auc))
            _dir = ("SLOWER in those too" if all(r.auc > 0.5 for r in _r.itertuples() if np.isfinite(r.auc))
                    else "FASTER in those" if all(r.auc < 0.5 for r in _r.itertuples() if np.isfinite(r.auc)) else "mixed")
            print(f"       {bx:10s} {_lab}   ->  the dusty ones are {_dir}")
        print( "     So in the coarse boxes what is slow about a dusty QG is the DUST ALONE; in m25, the box that resolves")
        print( "     the gas, it is slow in everything.")
    print("=" * 132)

    # ══ (H) THE PRODUCTS ════════════════════════════════════════════════════════════════════════════════════════════
    pd.DataFrame(ROW18).to_csv(os.path.join(OUT, "kappa_decay.csv"), index=False)
    D18.to_csv(os.path.join(OUT, "kappa_decay_galaxies.csv"), index=False)
    print("\ntable  ->", os.path.join(OUT, "kappa_decay.csv"), " (every row of every block above)")
    print("table  ->", os.path.join(OUT, "kappa_decay_galaxies.csv"),
          f" ({len(D18)} branches x {len(P18_DQ)} quantities: the window, the two slopes, the break and the classes)")
```

```python
# ── Part 19 — HOW LONG THE DUST STAYS OBSERVABLE: the visibility time after the truncation ──
# THE QUESTION (the user's): Part 18 timed the decay; this part asks the thing an observer actually measures — HOW LONG
# a quenched galaxy stays above the dusty threshold after its truncation, i.e. WHEN the falling exponential crosses
# log M_dust/M* = DUSTY_CUT.  It is the same fit read as a clock instead of a rate, and it is the natural candidate for
# the division Part 18 looked for and did not find in log tau: tau is a rate, and a rate divides nothing on its own
# because a galaxy that starts with ten times the dust survives ten times longer at the same tau.  THE LIFETIME IS THE
# RATE AND THE HEAD START TOGETHER,  t_vis ~ (y_0 - thr) / s,  and it is what the label is made of.
# THE CIRCULARITY, STATED FIRST, BECAUSE IT IS TOTAL.  The dust label IS "above the threshold at the anchor", so the
# measured visibility time and the label are two readings of one number: a dusty QG is exactly a branch whose crossing
# has NOT happened yet by the time we look.  Read naively, "t_vis divides dusty from dust-poor" is not a result, it is
# the definition — and worse, t_vis is not even measurable for the dusty ones: their crossing lies in the future of the
# last snapshot.  So this part is built on the two things that break the circle:
#   (1) THE CENSORING IS THE POINT, NOT A NUISANCE.  A branch still above the threshold at the anchor is a RIGHT-CENSORED
#       lifetime (t_vis > t_post), not a missing one.  KAPLAN-MEIER uses both kinds and estimates the survival curve
#       S(t) = P(still observably dusty t Gyr after the truncation) without ever needing the label, so the median
#       visibility time and the never-crossing plateau are population statements, not restatements of the cut.
#   (2) THE COUNTERFACTUAL LIFETIME.  t_ini = (y_0 - thr) / s_1 is what the branch's OWN first leg implies, out of the
#       state it had at the truncation and the rate it was falling at then — measured entirely before the anchor and
#       defined for the dusty and the dust-poor alike.  Whether THAT already separates the label is a real question,
#       and the difference t_vis - t_ini is Part 18's arrest measured in the currency of the observation: the extra
#       Gyr of visibility the branches that stop falling (and regrow) buy themselves.
# WHAT IS DRAWN AND TESTED:
#   (a) the census: never visible / crossed / still visible / crossed and came back, per box and per class;
#   (b) FIG 1 the survival curves — by kinematic class (the SELECTION, which is not the label), by anchor epoch, and up
#       the whole THRESHOLD LADDER, because "observable" is a depth: the answer must be quoted with the cut it used;
#   (c) FIG 2 THE TWO CLOCKS: the visibility time against the time since the truncation at which each galaxy is caught.
#       The label is the diagonal.  From S(t) and the distribution of that delay the dusty FRACTION per anchor is
#       PREDICTED (leave-one-anchor-out) and compared with the observed one — the demography test, and the honest way
#       to ask whether the z-trend of the dusty fraction is a lifetime or a look-back-time effect;
#   (d) FIG 3 is it a divider — the counterfactual lifetime's distribution (Silverman, as in Part 18) and the plane it
#       is made in: the head start against the initial rate, with the iso-lifetime diagonals drawn on it;
#   (e) FIG 4 the arrest: t_vis against t_ini, and how much visible time the branches that stop falling gain.
# A PURE RE-READ of Part 18's kernel objects: D18 (the fits) and R18 (the stored branch series of P18_QGAL).  No file
# is opened, nothing is re-fitted, and the numbers therefore refer to Part 18's window and floor treatment exactly.
P19_THR    = None       # THE VISIBILITY THRESHOLD in log M_dust/M*.  None = DUSTY_CUT, the label's own cut
P19_LADDER = None       # the thresholds the curves are repeated at.  None = THR + 0.5, THR, THR - 0.5, THR - 1.0
P19_KCLS   = True       # split the survival curves by the kinematic class as well (Part 2's selection)
P19_ZBINS  = 3          # the anchor-epoch split: this many equal-count bins in z
P19_HZ     = None       # the RMST horizon [Gyr]: None = the box's own median follow-up, so both classes cover it
P19_NBOOT  = 200        # the bootstrap on the RMST difference
P19_MINN   = 15         # a curve is drawn / a group compared only above this many branches
P19_MINDIST = None      # a distribution is tested only above this many (None = P18_MINDIST)
P19_SILV   = None       # the Silverman bootstrap (None = P18_SILV; 0 skips the test)
P19_TAX    = (-1.3, 0.9)   # the log10 t axis [Gyr] of FIG 3
P19_SMAX   = None       # the survival curves' time axis [Gyr]; None = the 95th percentile of the follow-up
P19_SEED   = 19

_OK19 = ("D18" in dir()) and len(D18) and len(R18)
if not _OK19:
    print("Part 19: needs Part 18 in the kernel — D18 (the fits) and R18 (the stored branches).  Run Part 18 first.")
else:
    from scipy import stats as _st19

    _Q19 = P18_QGAL                      # the quantity Part 18 kept the series of — the dust
    THR19 = float(DUSTY_CUT if P19_THR is None else P19_THR)
    _LAD19 = [THR19 + 0.5, THR19, THR19 - 0.5, THR19 - 1.0] if P19_LADDER is None else list(P19_LADDER)
    LAD19 = sorted({round(float(t_), 3) for t_ in _LAD19 if t_ > P9_FLOOR + 0.1}, reverse=True)
    _MND19 = P18_MINDIST if P19_MINDIST is None else P19_MINDIST
    _SLV19 = P18_SILV if P19_SILV is None else P19_SILV
    ROW19, V19 = [], pd.DataFrame()

    # ══ THE MACHINERY ═══════════════════════════════════════════════════════════════════════════════════════════════
    def _cross19(t, y, thr):
        """ONE branch against ONE threshold.  y is log10 of the quantity, t is Gyr from the truncation, so a straight
        line between two points IS the exponential and the crossing is exact under linear interpolation.
        Returns the FIRST down-crossing (the lifetime), whether it happened inside the window (the event, as opposed to
        a right-censored branch still above the threshold at the anchor), the TOTAL time spent above it (which counts a
        return), the crossings both ways, and the state at each end."""
        t, y = np.asarray(t, float), np.asarray(y, float)
        m = np.isfinite(t) & np.isfinite(y)
        t, y = t[m], y[m]
        out = dict(y0=np.nan, head=np.nan, vis0=False, vis_end=False, t_vis=np.nan, ev=False, t_above=np.nan,
                   n_down=0, n_up=0, t_ret=np.nan, t0=np.nan, t_post=np.nan)
        if len(t) < 2:
            return out
        ab = y >= thr
        out.update(y0=float(y[0]), head=float(y[0] - thr), vis0=bool(ab[0]), vis_end=bool(ab[-1]),
                   t0=float(t[0]), t_post=float(t[-1]))
        dn = np.flatnonzero(ab[:-1] & ~ab[1:])          # a down-crossing lies inside [i, i+1]
        up = np.flatnonzero((~ab[:-1]) & ab[1:])
        out["n_down"], out["n_up"] = int(len(dn)), int(len(up))

        def _tc(i):
            """the crossing time inside the interval [i, i+1], linear in log10 y = exponential in y."""
            dy = y[i] - y[i + 1]
            f = (y[i] - thr) / dy if abs(dy) > 1e-12 else 0.0
            return float(t[i] + np.clip(f, 0.0, 1.0) * (t[i + 1] - t[i]))

        if not ab[0]:                                    # it was already below the threshold at the first step after
            out.update(t_vis=0.0, ev=True)               # the truncation: no visible life to time (counted apart)
        elif len(dn):
            out.update(t_vis=_tc(int(dn[0])), ev=True)
        else:
            out.update(t_vis=float(t[-1]), ev=False)     # RIGHT-CENSORED: still above the threshold when we look
        if len(up) and ab[-1] and len(dn):
            out["t_ret"] = _tc(int(up[-1]))              # it came back: the last return into visibility
        a = 0.0                                          # the total time above, with the partial intervals counted
        for i in range(len(t) - 1):
            dt = t[i + 1] - t[i]
            if ab[i] and ab[i + 1]:
                a += dt
            elif ab[i] or ab[i + 1]:                 # a partial interval: the same interpolation both ways
                dy = y[i] - y[i + 1]
                f = (y[i] - thr) / dy if abs(dy) > 1e-12 else 0.0
                f = float(np.clip(f, 0.0, 1.0))
                a += dt * (f if ab[i] else (1.0 - f))
        out["t_above"] = float(a)
        return out

    def _km19(t, ev):
        """KAPLAN-MEIER for right-censored lifetimes: S(t) = P(still above the threshold at t), with Greenwood's
        standard error.  A censored branch (still visible at the anchor) contributes its time at risk and no event —
        which is exactly what a dusty QG is, and why the label never has to be used here."""
        t, ev = np.asarray(t, float), np.asarray(ev, bool)
        m = np.isfinite(t) & (t >= 0)
        t, ev = t[m], ev[m]
        out = dict(t=np.array([0.0]), S=np.array([1.0]), se=np.array([0.0]), n=int(len(t)), nev=int(ev.sum()),
                   tmax=float(np.max(t)) if len(t) else np.nan, nrisk=np.array([float(len(t))]))
        if not len(t) or not ev.any():
            return out
        T, S, SE, NR = [0.0], [1.0], [0.0], [float(len(t))]
        s, v = 1.0, 0.0
        for u in np.unique(t[ev]):
            n_r = float(np.sum(t >= u - 1e-9))
            d = float(np.sum(ev & (np.abs(t - u) < 1e-9)))
            if n_r <= 0 or d <= 0:
                continue
            s *= 1.0 - d / n_r
            v += d / max(n_r * (n_r - d), 1e-12)
            T.append(float(u)); S.append(float(s)); SE.append(float(s * np.sqrt(v))); NR.append(n_r)
        out.update(t=np.asarray(T), S=np.asarray(S), se=np.asarray(SE), nrisk=np.asarray(NR))
        return out

    def _kmq19(km, q=0.5, off=0.0):
        """the q-quantile of the survival curve: the first time S drops to or below 1 - q.  `off` shifts S by that many
        standard errors first, which is how the 68 % interval on the median is read off the band.  NaN = never."""
        S = km["S"] + off * km["se"]
        i = np.flatnonzero(S <= 1.0 - q + 1e-12)
        return float(km["t"][i[0]]) if len(i) else np.nan

    def _rmst19(km, h):
        """the restricted mean survival time to the horizon h: the mean visibility time, counting anything beyond h as
        h.  It is the honest 'how long on average' when the follow-up differs between the groups compared."""
        T, S = km["t"], km["S"]
        e = np.append(T[1:], np.inf)
        return float(np.sum(S * np.clip(np.minimum(e, h) - np.minimum(T, h), 0.0, None)))

    def _logrank19(t, ev, g):
        """the two-group LOG-RANK test: are the two survival curves the same?  (O - E)^2 / V on group 1, chi2 with 1 dof."""
        t, ev, g = np.asarray(t, float), np.asarray(ev, bool), np.asarray(g, bool)
        m = np.isfinite(t)
        t, ev, g = t[m], ev[m], g[m]
        O1 = E1 = V = 0.0
        for u in np.unique(t[ev]):
            at = t >= u - 1e-9
            n, n1 = float(at.sum()), float((at & g).sum())
            d = float(np.sum(ev & (np.abs(t - u) < 1e-9)))
            d1 = float(np.sum(ev & (np.abs(t - u) < 1e-9) & g))
            if n < 2 or d <= 0:
                continue
            O1 += d1; E1 += d * n1 / n
            V += d * (n - d) * n1 * (n - n1) / (n * n * (n - 1.0))
        if V <= 0:
            return np.nan, np.nan
        chi = (O1 - E1) ** 2 / V
        return float(chi), float(_st19.chi2.sf(chi, 1))

    def _sat19(km, t):
        """S evaluated at arbitrary times: the step function, held flat beyond the last event (and flagged by the
        caller, since an extrapolated S is an assumption and not an estimate)."""
        i = np.searchsorted(km["t"], np.asarray(t, float), side="right") - 1
        return km["S"][np.clip(i, 0, len(km["S"]) - 1)]

    # ══ (A) THE PASS: every stored branch against every threshold of the ladder ══════════════════════════════════════
    _rows19, TH19 = [], {}
    _c19 = [c for c in ("z", "pop", "lfd", "age", "log_mstar", "log_mbh", "kcls", "rot", "central", "t_anchor",
                        "kappa", "ngas", "t_x1", "e_x", "f_x_post", "t_qt", "t_sft", "t_post") if c in D18.columns]
    _f19 = [f"{_Q19}_{k}" for k in ("cls", "s0", "s1", "s2", "tau0", "tau1", "tau2", "t1dex", "drop", "reb", "t_min",
                                    "ymin", "y_end", "n_fl", "dt", "tb", "e_s0", "e_s1", "n")]
    _f19 = [c for c in _f19 if c in D18.columns]
    for _, _r19 in D18.iterrows():
        _k19 = (_r19["box"], int(_r19["snap"]), int(_r19["gal_id"]))
        _s19 = R18.get(_k19)
        if _s19 is None:
            continue
        _t, _y = _s19["t"], _s19["y"]
        _rec = dict(box=_r19["box"], snap=int(_r19["snap"]), gal_id=int(_r19["gal_id"]))
        _rec.update({c: _r19[c] for c in _c19})
        _rec.update({c: _r19[c] for c in _f19})
        for _th in LAD19:
            _cv = _cross19(_t, _y, _th)
            _sfx = "" if abs(_th - THR19) < 1e-9 else f"_{_th:+.2f}".replace("+", "p").replace("-", "m").replace(".", "")
            for _k, _v in _cv.items():
                _rec[f"{_k}{_sfx}"] = _v
        # THE COUNTERFACTUAL LIFETIME: the state at the truncation over the rate the branch was falling at then.
        _s1 = float(_r19.get(f"{_Q19}_s1", np.nan)); _s0 = float(_r19.get(f"{_Q19}_s0", np.nan))
        _si = _s1 if np.isfinite(_s1) else _s0
        _rec["s_ini"], _rec["ini_leg"] = _si, (1 if np.isfinite(_s1) else 0)
        _hd = _rec.get("head", np.nan)
        with np.errstate(all="ignore"):
            _rec["t_ini"] = float(_hd / _si) if (np.isfinite(_hd) and np.isfinite(_si) and _si > 1e-6) else np.inf
            _rec["t_ini0"] = float(_hd / _s0) if (np.isfinite(_hd) and np.isfinite(_s0) and _s0 > 1e-6) else np.inf
        if not np.isfinite(_hd) or _hd <= 0:
            _rec["t_ini"] = _rec["t_ini0"] = 0.0            # already below the threshold: no visible life to predict
        _rows19.append(_rec)
    V19 = pd.DataFrame(_rows19)
    if "lfd_pre" not in V19.columns and "H" in dir():        # the pre-truncation dust, where Part 3 has it: it says
        try:                                                # whether a branch that is never visible ever WAS dusty
            _pre = pd.concat([H[b]["F"][["box", "snap", "gal_id", "lfd_pre"]] for b in H
                              if "lfd_pre" in H[b]["F"].columns], ignore_index=True)
            V19 = V19.merge(_pre, on=["box", "snap", "gal_id"], how="left")
        except Exception as _e19:
            print(f"  (the pre-truncation dust could not be joined: {_e19})")
    _BB19 = [b for b in (list(H) if "H" in dir() else sorted(V19.box.unique())) if (V19.box == b).any()]
    _BB19 = [b for b in _BB19 if b != NOX_BOX] + [b for b in _BB19 if b == NOX_BOX]
    _tp19 = V19["t_post"].to_numpy(float)[V19.box.to_numpy() != NOX_BOX]
    if not len(_tp19):
        _tp19 = V19["t_post"].to_numpy(float)
    _SMAX19 = float(np.nanpercentile(_tp19, 95)) if P19_SMAX is None else float(P19_SMAX)

    print("=" * 132)
    print(f"Part 19 — HOW LONG THE DUST STAYS OBSERVABLE.  the visibility threshold is log M_dust/M* = {THR19:+.2f}"
          f"  (the ladder: {', '.join(f'{t_:+.2f}' for t_ in LAD19)}),")
    print(f"  the clock is t - t_SFT, and a branch still above the threshold at its anchor is RIGHT-CENSORED, not "
          f"dropped — which is what a dusty QG is.")
    print("=" * 132)

    # ══ (B) THE CENSUS ══════════════════════════════════════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(a) THE CENSUS OF VISIBILITY — what each branch does with the threshold, and the Kaplan-Meier lifetime that")
    print("    uses the crossed and the still-visible branches together (so the label is never needed to compute it)")
    print(f"  {'box':10s} {'N':>5s} {'never':>6s} {'crossed':>8s} {'still vis':>10s} {'came back':>10s} "
          f"{'follow-up':>10s} {'t_vis KM':>9s} {'68 %':>13s} {'S(end)':>7s} {'RMST':>6s} {'t_above':>8s} {'label ok':>9s}")
    for b in _BB19:
        g = V19[V19.box == b]
        n = len(g)
        v0 = g["vis0"].to_numpy(bool)
        km = _km19(g["t_vis"].to_numpy(float)[v0], g["ev"].to_numpy(bool)[v0])
        hz = float(np.nanmedian(g["t_post"])) if P19_HZ is None else float(P19_HZ)
        _lab = np.nan
        if "pop" in g.columns:
            _lab = float(np.mean((g["vis_end"].to_numpy(bool)) == (g["pop"].to_numpy() == "dusty")))
        print(f"  {b:10s} {n:5d} {100 * np.mean(~v0):5.0f}% {100 * np.mean(v0 & g['ev'].to_numpy(bool)):7.0f}% "
              f"{100 * np.mean(v0 & ~g['ev'].to_numpy(bool)):9.0f}% {100 * np.mean(g['n_up'].to_numpy(float) > 0):9.0f}% "
              f"{np.nanmedian(g['t_post']):10.2f} {_kmq19(km):9.2f} "
              f"{_kmq19(km, off=-1):5.2f}-{_kmq19(km, off=+1):<7.2f} {km['S'][-1]:7.2f} {_rmst19(km, hz):6.2f} "
              f"{np.nanmedian(g['t_above']):8.2f} {100 * _lab:8.0f}%")
        ROW19.append(dict(box=b, block="census", quantity="km_median", thr=THR19, n_gal=int(v0.sum()),
                          med=_kmq19(km), lo=_kmq19(km, off=-1), hi=_kmq19(km, off=+1), x=float(km["S"][-1])))
        ROW19.append(dict(box=b, block="census", quantity="rmst", thr=THR19, n_gal=int(v0.sum()), med=_rmst19(km, hz),
                          x=hz))
        ROW19.append(dict(box=b, block="census", quantity="shares", thr=THR19, n_gal=n, med=float(np.mean(~v0)),
                          lo=float(np.mean(v0 & g["ev"].to_numpy(bool))), hi=float(np.mean(v0 & ~g["ev"].to_numpy(bool)))))
    print(f"  ('never' = already below {THR19:+.2f} at the first snapshot after the truncation — it never had an")
    print( "   observable dust phase to lose and is excluded from every survival curve (it has no time at risk).")
    print( "   'crossed' = the crossing is in the window, 'still vis' = right-censored at the anchor, 'came back' = it")
    print( "   crossed and returned above the threshold at least once.  t_vis KM is the median of the survival curve")
    print( "   with its 68 % band, S(end) the plateau — the share that has NOT crossed by the end of the follow-up —")
    print(f"   RMST the mean visibility time restricted to the box's own median follow-up, and t_above the total time")
    print( "   spent above the threshold (which counts a return).  'label ok' checks that 'above at the last step' is")
    print( "   the anchor dust label: it must be ~100 % when the threshold IS DUSTY_CUT, and it is the audit of that.)")

    if "pop" in V19.columns:
        print(f"\n  the same split by the dust label — WHICH IS THE CIRCLE ITSELF, printed so that it cannot be mistaken")
        print(f"  for a result: a dusty QG is BY DEFINITION a branch whose crossing has not happened yet, so its t_vis")
        print(f"  is censored at the anchor and its 'median' below is a lower limit, nothing more.")
        print(f"  {'box':10s} {'label':7s} {'N':>5s} {'never':>6s} {'crossed':>8s} {'censored':>9s} {'med t_vis':>10s} "
              f"{'med t_ini':>10s} {'med head':>9s} {'med s_ini':>10s}")
        for b in _BB19:
            for p_ in ("dusty", "pure"):
                g = V19[(V19.box == b) & (V19["pop"] == p_)]
                if len(g) < 8:
                    continue
                v0 = g["vis0"].to_numpy(bool); ev = g["ev"].to_numpy(bool)
                _ti = g["t_ini"].to_numpy(float)
                print(f"  {b:10s} {p_:7s} {len(g):5d} {100 * np.mean(~v0):5.0f}% {100 * np.mean(v0 & ev):7.0f}% "
                      f"{100 * np.mean(v0 & ~ev):8.0f}% {np.nanmedian(g['t_vis']):10.2f} "
                      f"{np.nanmedian(_ti[np.isfinite(_ti)]):10.2f} {np.nanmedian(g['head']):9.2f} "
                      f"{np.nanmedian(g['s_ini']):10.2f}")
                ROW19.append(dict(box=b, block="census_pop", quantity=p_, thr=THR19, n_gal=len(g),
                                  med=float(np.nanmedian(g["t_vis"])), lo=float(np.nanmedian(g["head"])),
                                  hi=float(np.nanmedian(_ti[np.isfinite(_ti)]))))
        print( "  (t_ini = head / s_ini, the lifetime the branch's OWN first leg implies out of the dust it had at the")
        print( "   truncation and the rate it was then losing it at — measured before the anchor, defined for both")
        print( "   labels, and infinite where the first leg is flat or rising (those are excluded from its median).)")

    # ══ (C) FIG 1 — THE SURVIVAL CURVES ═════════════════════════════════════════════════════════════════════════════
    def _leg19(ax, **kw):
        kw.setdefault("frameon", False); kw.setdefault("fontsize", 7.5)
        h, l = ax.get_legend_handles_labels()
        if h:
            ax.legend(h, l, **kw)

    def _drawkm19(ax, km, color, label, ls="-", band=True):
        if km["n"] < 1:
            return
        t = np.append(km["t"], max(_SMAX19, km["t"][-1]))
        S = np.append(km["S"], km["S"][-1])
        ax.step(t, S, where="post", color=color, lw=2.0, ls=ls, label=label, zorder=4)
        if band:
            lo = np.clip(np.append(km["S"] - km["se"], km["S"][-1] - km["se"][-1]), 0, 1)
            hi = np.clip(np.append(km["S"] + km["se"], km["S"][-1] + km["se"][-1]), 0, 1)
            ax.fill_between(t, lo, hi, step="post", color=color, alpha=0.18, lw=0, zorder=2)

    print("\n" + "-" * 132)
    print("(b) THE SURVIVAL CURVES — S(t) = P(still above the threshold t Gyr after the truncation), and what splits it")
    print(f"  {'box':10s} {'split':22s} {'group':16s} {'N':>5s} {'events':>7s} {'t_vis KM':>9s} {'RMST':>6s} "
          f"{'diff':>6s} {'68 %':>13s} {'log-rank p':>11s}")
    _KMB19 = {}
    for b in _BB19:
        g = V19[(V19.box == b) & V19["vis0"].to_numpy(bool)]
        if len(g) < P19_MINN:
            continue
        hz = float(np.nanmedian(V19[V19.box == b]["t_post"])) if P19_HZ is None else float(P19_HZ)
        _KMB19[b] = dict(all=_km19(g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool)), hz=hz)
        for split, col, groups in (("the kinematic class", "kcls", KC if "KC" in dir() else ["rot", "iso"]),
                                   ("the dust label (circ.)", "pop", ["dusty", "pure"])):
            if col not in g.columns:
                continue
            _k, _n = {}, {}
            for gr in groups:
                m = (g[col].to_numpy() == gr)
                if m.sum() >= P19_MINN:
                    _k[gr], _n[gr] = _km19(g["t_vis"].to_numpy(float)[m], g["ev"].to_numpy(bool)[m]), int(m.sum())
            if len(_k) < 2:
                continue
            g1 = groups[0]
            chi, p_ = _logrank19(g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool), (g[col].to_numpy() == g1))
            _rm = {gr: _rmst19(_k[gr], hz) for gr in _k}
            _d = _rm[groups[0]] - _rm[groups[1]]
            _bs = []
            _rng19 = np.random.default_rng(P19_SEED)
            _tv, _ev, _gc = g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool), g[col].to_numpy()
            for _ib in range(int(P19_NBOOT)):
                i = _rng19.integers(0, len(g), len(g))
                try:
                    _bs.append(_rmst19(_km19(_tv[i][_gc[i] == groups[0]], _ev[i][_gc[i] == groups[0]]), hz)
                               - _rmst19(_km19(_tv[i][_gc[i] == groups[1]], _ev[i][_gc[i] == groups[1]]), hz))
                except Exception:
                    pass
            _lo, _hi = (np.percentile(_bs, [16, 84]) if len(_bs) > 20 else (np.nan, np.nan))
            for j, gr in enumerate(_k):
                print(f"  {b if j == 0 else '':10s} {split if j == 0 else '':22s} {gr:16s} {_n[gr]:5d} "
                      f"{_k[gr]['nev']:7d} {_kmq19(_k[gr]):9.2f} {_rm[gr]:6.2f} "
                      + (f"{_d:+6.2f} {_lo:+6.2f},{_hi:+6.2f} {p_:11.1e}" if j == 0 else ""))
                ROW19.append(dict(box=b, block=f"survival_{col}", quantity=gr, thr=THR19, n_gal=_n[gr],
                                  med=_kmq19(_k[gr]), x=_rm[gr], p=p_))
            _KMB19[b][col] = _k
        _kl = {}
        for th in LAD19:
            _sfx = "" if abs(th - THR19) < 1e-9 else f"_{th:+.2f}".replace("+", "p").replace("-", "m").replace(".", "")
            gg = V19[(V19.box == b) & V19[f"vis0{_sfx}"].to_numpy(bool)]
            if len(gg) < P19_MINN:
                continue
            _kl[th] = _km19(gg[f"t_vis{_sfx}"].to_numpy(float), gg[f"ev{_sfx}"].to_numpy(bool))
            ROW19.append(dict(box=b, block="ladder", quantity="km_median", thr=th, n_gal=len(gg),
                              med=_kmq19(_kl[th]), x=float(_kl[th]["S"][-1])))
        _KMB19[b]["ladder"] = _kl
    print( "  (the median is the first time S drops to 0.5 and is NaN where the curve never gets there — the honest")
    print( "   form of 'more than half are still visible when the follow-up ends'.  RMST is the mean visibility time")
    print( "   restricted to the box's median follow-up, its difference bootstrapped over galaxies; the log-rank test")
    print( "   compares the whole curves.  The dust-label rows are the circle drawn on purpose: they compare a group")
    print( "   defined by not having crossed with a group defined by having crossed.)")

    _nrow = 2 + int("kcls" in V19.columns and P19_KCLS)
    fig1, AX1 = plt.subplots(_nrow, len(_BB19), figsize=(4.0 * len(_BB19), 3.1 * _nrow), squeeze=False, sharey=True)
    for jb, b in enumerate(_BB19):
        K = _KMB19.get(b)
        ax = AX1[0, jb]
        if K:
            _drawkm19(ax, K["all"], "0.25", f"every branch ({K['all']['n']}, {K['all']['nev']} crossings)")
            for th, c_ in zip(K["ladder"], plt.cm.viridis(np.linspace(0.1, 0.85, max(len(K["ladder"]), 1)))):
                if abs(th - THR19) < 1e-9:
                    continue
                _drawkm19(ax, K["ladder"][th], c_, rf"$\log M_{{\rm dust}}/M_\star = {th:+.1f}$", ls="--", band=False)
            _m = _kmq19(K["all"])
            if np.isfinite(_m):
                ax.plot([_m, _m], [0, 0.5], color="0.25", lw=1.0, ls=":")
                ax.annotate(f"{_m:.2f} Gyr", xy=(_m, 0.5), xytext=(3, 3), textcoords="offset points", fontsize=8,
                            color="0.25", fontweight="bold")
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.axhline(0.5, color="0.8", lw=0.8, ls=":")
        ax.set_xlim(0, _SMAX19); ax.set_ylim(0, 1.02); ax.grid(alpha=0.2, lw=0.5)
        if jb == 0:
            ax.set_ylabel("THE THRESHOLD LADDER\n" + r"$S(t)$ = still above the cut", fontsize=9)
        _leg19(ax, loc="lower left", fontsize=6.5)
        r_ = 1
        if "kcls" in V19.columns and P19_KCLS:
            ax = AX1[r_, jb]
            for c_ in (KC if "KC" in dir() else ["rot", "iso"]):
                if K and "kcls" in K and c_ in K["kcls"]:
                    _drawkm19(ax, K["kcls"][c_], KCLS[c_]["color"], f"{KCLS[c_]['short']} ({K['kcls'][c_]['n']})")
            if K and "kcls" in K and len(K["kcls"]) == 2:
                g = V19[(V19.box == b) & V19["vis0"].to_numpy(bool)]
                _c, _p = _logrank19(g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool),
                                    (g["kcls"].to_numpy() == (KC[0] if "KC" in dir() else "rot")))
                ax.annotate(f"log-rank p = {_p:.1e}", xy=(0.97, 0.95), xycoords="axes fraction", ha="right",
                            fontsize=7.5, fontweight="bold")
            ax.axhline(0.5, color="0.8", lw=0.8, ls=":")
            ax.set_xlim(0, _SMAX19); ax.set_ylim(0, 1.02); ax.grid(alpha=0.2, lw=0.5)
            if jb == 0:
                ax.set_ylabel("THE SELECTION\n" + r"$S(t)$", fontsize=9)
            _leg19(ax, loc="lower left")
            r_ += 1
        ax = AX1[r_, jb]
        g = V19[(V19.box == b) & V19["vis0"].to_numpy(bool)]
        if len(g) >= P19_MINN and "z" in g.columns:
            _z = g["z"].to_numpy(float)
            _ed = np.unique(np.nanquantile(_z, np.linspace(0, 1, P19_ZBINS + 1)))
            for i_ in range(len(_ed) - 1):
                m = (_z >= _ed[i_]) & (_z <= _ed[i_ + 1] if i_ == len(_ed) - 2 else _z < _ed[i_ + 1])
                if m.sum() < P19_MINN:
                    continue
                km = _km19(g["t_vis"].to_numpy(float)[m], g["ev"].to_numpy(bool)[m])
                _drawkm19(ax, km, plt.cm.plasma(0.15 + 0.7 * i_ / max(len(_ed) - 2, 1)),
                          f"z = {_ed[i_]:.1f}-{_ed[i_ + 1]:.1f} ({int(m.sum())})", band=False)
                ROW19.append(dict(box=b, block="survival_z", quantity=f"{_ed[i_]:.2f}-{_ed[i_ + 1]:.2f}", thr=THR19,
                                  n_gal=int(m.sum()), med=_kmq19(km), x=float(np.median(_z[m]))))
        ax.axhline(0.5, color="0.8", lw=0.8, ls=":")
        ax.set_xlim(0, _SMAX19); ax.set_ylim(0, 1.02); ax.grid(alpha=0.2, lw=0.5)
        ax.set_xlabel(r"$t - t_{\rm SFT}$  [Gyr]", fontsize=9)
        if jb == 0:
            ax.set_ylabel("THE EPOCH\n" + r"$S(t)$", fontsize=9)
        _leg19(ax, loc="lower left")
    fig1.suptitle("How long the dust stays observable after the truncation — the Kaplan-Meier survival curve of the "
                  f"visibility time at $\\log M_{{\\rm dust}}/M_\\star = {THR19:+.1f}$,\nwith the branches still above the "
                  "threshold at their anchor entering as right-censored (which is what a dusty QG is)",
                  y=1.005, fontsize=12, fontweight="bold")
    fig1.tight_layout(); paper_save(fig1, "kappa_visibility_survival"); plt.show()

    # ══ (D) THE TWO CLOCKS AND THE DEMOGRAPHY ═══════════════════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(c) THE TWO CLOCKS — a galaxy is seen as a dusty QG when its visibility time outlasts the time since its own")
    print("    truncation.  So the dusty FRACTION is predictable: P(dusty) = P(vis at t=0) x S(t_post), with S taken")
    print("    from the OTHER anchors of the same box (leave-one-anchor-out), never from the anchor being predicted")
    print(f"  {'box':10s} {'anchor z':>9s} {'N':>5s} {'med delay':>10s} {'dusty obs':>10s} {'predicted':>10s} "
          f"{'68 %':>13s} {'never vis':>10s} {'extrap':>7s}")
    for b in _BB19:
        g = V19[V19.box == b]
        if "z" not in g.columns or len(g) < P19_MINN:
            continue
        for a_ in sorted(g["snap"].unique()):
            ga = g[g["snap"] == a_]
            go = g[g["snap"] != a_]
            if len(ga) < P19_MINN or len(go) < P19_MINN:
                continue
            km = _km19(go["t_vis"].to_numpy(float)[go["vis0"].to_numpy(bool)],
                       go["ev"].to_numpy(bool)[go["vis0"].to_numpy(bool)])
            _d = ga["t_post"].to_numpy(float)
            _pr = np.mean(ga["vis0"].to_numpy(bool) * _sat19(km, _d))
            _bs = []
            _rng19 = np.random.default_rng(P19_SEED + int(a_))
            for _ib in range(int(P19_NBOOT)):
                i = _rng19.integers(0, len(go), len(go))
                _k = _km19(go["t_vis"].to_numpy(float)[i][go["vis0"].to_numpy(bool)[i]],
                           go["ev"].to_numpy(bool)[i][go["vis0"].to_numpy(bool)[i]])
                j = _rng19.integers(0, len(ga), len(ga))
                _bs.append(float(np.mean(ga["vis0"].to_numpy(bool)[j] * _sat19(_k, _d[j]))))
            _lo, _hi = (np.percentile(_bs, [16, 84]) if len(_bs) > 20 else (np.nan, np.nan))
            _ob = float(np.mean(ga["pop"].to_numpy() == "dusty")) if "pop" in ga.columns else np.nan
            _ex = float(np.mean(_d > km["t"][-1]))
            print(f"  {b:10s} {np.median(ga['z']):9.2f} {len(ga):5d} {np.median(_d):10.2f} {100 * _ob:9.0f}% "
                  f"{100 * _pr:9.0f}% {100 * _lo:6.0f}-{100 * _hi:<3.0f}% "
                  f"{100 * np.mean(~ga['vis0'].to_numpy(bool)):9.0f}% {100 * _ex:6.0f}%")
            ROW19.append(dict(box=b, block="demography", quantity=f"snap{int(a_)}", thr=THR19, n_gal=len(ga),
                              x=float(np.median(ga["z"])), med=_pr, lo=_lo, hi=_hi, auc=_ob))
    print( "  ('med delay' is the median t_anchor - t_SFT, i.e. how long ago these galaxies truncated; 'extrap' the")
    print( "   share of them whose delay is longer than the longest crossing seen at the other anchors, where S has to")
    print( "   be held flat and the prediction is an assumption.  A predicted fraction that tracks the observed one")
    print( "   means the epoch trend of the dusty fraction is a LOOK-BACK effect — the same lifetime, caught earlier.)")

    fig2, AX2 = plt.subplots(2, len(_BB19), figsize=(4.0 * len(_BB19), 7.2), squeeze=False)
    for jb, b in enumerate(_BB19):
        g = V19[V19.box == b]
        ax = AX2[0, jb]
        _d = g["t_post"].to_numpy(float); _v = g["t_vis"].to_numpy(float); _e = g["ev"].to_numpy(bool)
        _v0 = g["vis0"].to_numpy(bool)
        _cl = (np.where(g["pop"].to_numpy() == "dusty", POPS["dusty"]["color"], POPS["pure"]["color"])
               if "pop" in g.columns else np.full(len(g), "0.4"))
        m = _v0 & _e
        ax.scatter(_d[m], _v[m], s=9, c=_cl[m], alpha=0.45, edgecolor="none", zorder=3)
        m = _v0 & ~_e
        ax.scatter(_d[m], _v[m], s=22, c=_cl[m], alpha=0.6, marker="^", edgecolor="none", zorder=4)
        m = ~_v0
        ax.scatter(_d[m], np.full(m.sum(), 0.0), s=9, c=_cl[m], alpha=0.35, marker="x", zorder=2)
        _lim = (0, max(_SMAX19, 1e-3))
        ax.plot(_lim, _lim, "-", color="0.3", lw=1.2)
        ax.annotate("above the line: still visible\n(= a dusty QG)", xy=(0.04, 0.93), xycoords="axes fraction",
                    fontsize=7.5, va="top", fontweight="bold")
        ax.set_xlim(*_lim); ax.set_ylim(-0.05 * _lim[1], _lim[1] * 1.05); ax.grid(alpha=0.2, lw=0.5)
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.set_xlabel(r"$t_{\rm anchor} - t_{\rm SFT}$  — when we look  [Gyr]", fontsize=9)
        if jb == 0:
            ax.set_ylabel("THE TWO CLOCKS\n" + r"$t_{\rm vis}$  ($\blacktriangle$ = censored)  [Gyr]", fontsize=9)
        ax = AX2[1, jb]
        _R = pd.DataFrame([r for r in ROW19 if r.get("block") == "demography" and r.get("box") == b])
        if len(_R):
            _R = _R.sort_values("x")
            ax.fill_between(_R["x"], 100 * _R["lo"], 100 * _R["hi"], color="#4393c3", alpha=0.25, lw=0)
            ax.plot(_R["x"], 100 * _R["med"], "s--", color="#2166ac", lw=2.0, ms=5,
                    label="predicted from $S(t)$ of the other anchors")
            ax.plot(_R["x"], 100 * _R["auc"], "o-", color=POPS["dusty"]["color"], lw=2.2, ms=5.5, label="observed")
        ax.set_xlabel("anchor redshift", fontsize=9); ax.grid(alpha=0.2, lw=0.5)
        if jb == 0:
            ax.set_ylabel("THE DEMOGRAPHY\ndusty QGs  [%]", fontsize=9)
        _leg19(ax, loc="upper left")
    fig2.suptitle("The label is two clocks — how long the dust stays above the cut, against how long ago the galaxy "
                  "truncated (top); and the dusty fraction\nof each anchor predicted from the survival curve of the "
                  "OTHER anchors plus those delays (bottom)", y=1.005, fontsize=12, fontweight="bold")
    fig2.tight_layout(); paper_save(fig2, "kappa_visibility_clocks"); plt.show()

    # ══ (E) IS IT A DIVIDER ═════════════════════════════════════════════════════════════════════════════════════════
    def _l19(v):
        v = np.asarray(v, float)
        out = np.full(len(v), np.nan)
        m = np.isfinite(v) & (v > 0)
        out[m] = np.log10(v[m])
        return out

    print("\n" + "-" * 132)
    print("(d) IS THE LIFETIME THE DIVIDER — the distribution of the COUNTERFACTUAL lifetime t_ini (defined for every")
    print("    branch, measured before the anchor) and of its two factors, tested for two modes exactly as Part 18")
    print("    tested log tau, and scored against the dust label")
    print(f"  {'box':10s} {'quantity':14s} {'N':>5s} {'med':>6s} {'sd':>5s} {'skew':>6s} | {'Silv p':>7s} {'centred':>8s} "
          f"{'AshD':>5s} {'humps':>6s} | {'AUC':>6s} {'p':>9s} {'AUC | anchor':>13s}")
    _QD19 = [("t_ini", r"$t_{\rm ini}$ (counterfactual)"), ("t_vis", r"$t_{\rm vis}$ (measured*)"),
             ("head", "the head start [dex]"), ("s_ini", "the initial rate")]
    for b in _BB19:
        g = V19[V19.box == b]
        if len(g) < _MND19:
            continue
        for q, _lab in _QD19:
            x = g[q].to_numpy(float) if q == "head" else _l19(g[q].to_numpy(float))
            x = np.where(np.isfinite(x), x, np.nan)
            if np.isfinite(x).sum() < _MND19:
                continue
            sh = _shape18(x[np.isfinite(x)]) if "_shape18" in dir() else dict(med=np.nan, sd=np.nan, skew=np.nan,
                                                                             ash=np.nan, modes=np.nan)
            sv = _silv18(x[np.isfinite(x)], nboot=_SLV19) if "_silv18" in dir() else dict(p=np.nan)
            _gg = g.copy(); _gg["_x"] = x
            xc = _resid18(_gg, "_x") if "_resid18" in dir() else np.full(len(g), np.nan)
            svc = _silv18(xc[np.isfinite(xc)], nboot=_SLV19) if ("_silv18" in dir() and np.isfinite(xc).sum() >= _MND19) \
                else dict(p=np.nan)
            au = p_ = auz = np.nan
            if "pop" in g.columns:
                m1 = (g["pop"].to_numpy() == "dusty")
                au, p_, na, nb = mw_auc(x, m1)
                auz = mw_auc(xc, m1)[0]
            _st = "*" if q in ("t_vis",) else " "
            print(f"  {b:10s} {q:13s}{_st}{int(np.isfinite(x).sum()):5d} {sh['med']:6.2f} {sh['sd']:5.2f} "
                  f"{sh['skew']:+6.2f} | {sv['p']:7.3f} {svc['p']:8.3f} {sh['ash']:5.2f} {sh['modes']:6.0f} | "
                  f"{au:6.3f} {p_:9.1e} {auz:13.3f}")
            ROW19.append(dict(box=b, block="divider", quantity=q, thr=THR19, n_gal=int(np.isfinite(x).sum()),
                              med=sh["med"], p=sv["p"], auc=au, x=auz, lo=sh["ash"]))
    print( "  (every quantity is log10 except the head start, which is already in dex.  (*) t_vis is CIRCULAR — it is")
    print( "   censored precisely for the dusty ones, so its AUC is a restatement of the label and is printed only to")
    print( "   show how total that is.  t_ini, the head start and the rate are not: all three are fixed by the branch")
    print( "   before the anchor.  'centred' repeats Silverman after removing each anchor's own median, since the")
    print( "   anchor ladder alone spreads them.  AshD > 2 and humps = 2 would be needed for a real gap.)")

    fig3, AX3 = plt.subplots(2, len(_BB19), figsize=(4.0 * len(_BB19), 7.2), squeeze=False)
    for jb, b in enumerate(_BB19):
        g = V19[V19.box == b]
        ax = AX3[0, jb]
        _bins = np.arange(P19_TAX[0], P19_TAX[1] + 1e-9, 0.1)
        _x = _l19(g["t_ini"].to_numpy(float))
        if np.isfinite(_x).sum() >= _MND19:
            ax.hist(np.clip(_x[np.isfinite(_x)], *P19_TAX), bins=_bins, color="0.75", lw=0, label="every branch")
            if "pop" in g.columns:
                for p_ in ("dusty", "pure"):
                    m = np.isfinite(_x) & (g["pop"].to_numpy() == p_)
                    if m.sum() >= 8:
                        ax.hist(np.clip(_x[m], *P19_TAX), bins=_bins, histtype="step", lw=1.9,
                                color=POPS[p_]["color"], label=f"{p_} ({int(m.sum())})")
            _sv = _silv18(_x[np.isfinite(_x)], nboot=_SLV19) if "_silv18" in dir() else dict(p=np.nan)
            ax.set_title(f"{BOX_LABEL[b]}\nSilverman p = {_sv['p']:.3f}", fontsize=9.5,
                         color="#7f3b08" if b == NOX_BOX else "k")
        _md = float(np.nanmedian(V19[V19.box == b]["t_post"]))
        ax.axvline(np.log10(max(_md, 1e-3)), color="0.35", lw=1.2, ls="--")
        ax.annotate("the median follow-up", xy=(np.log10(max(_md, 1e-3)), 0.97), xycoords=("data", "axes fraction"),
                    rotation=90, ha="right", va="top", fontsize=7, color="0.35")
        ax.set_xlim(*P19_TAX); ax.grid(alpha=0.2, lw=0.5)
        ax.set_xlabel(r"$\log_{10}\,t_{\rm ini}$  [Gyr]", fontsize=9)
        if jb == 0:
            ax.set_ylabel("THE COUNTERFACTUAL LIFETIME\nbranches", fontsize=9)
        _leg19(ax, loc="upper right")
        ax = AX3[1, jb]
        _sx, _hy = _l19(g["s_ini"].to_numpy(float)), g["head"].to_numpy(float)
        m = np.isfinite(_sx) & np.isfinite(_hy)
        if m.sum() > 5:
            _c = (np.where(g["pop"].to_numpy() == "dusty", POPS["dusty"]["color"], POPS["pure"]["color"])
                  if "pop" in g.columns else np.full(len(g), "0.4"))
            ax.scatter(_sx[m], _hy[m], s=10, c=_c[m], alpha=0.45, edgecolor="none", zorder=3)
            _gx = np.linspace(np.nanpercentile(_sx[m], 1), np.nanpercentile(_sx[m], 99), 50)
            for _tt in (0.25, 0.5, 1.0, 2.0, 4.0):
                ax.plot(_gx, _tt * 10 ** _gx, "-", color="0.45", lw=0.9, alpha=0.8, zorder=2)
                ax.annotate(f"{_tt:g} Gyr", xy=(_gx[-1], _tt * 10 ** _gx[-1]), fontsize=6.5, color="0.45",
                            ha="right", va="bottom")
            ax.set_ylim(0, np.nanpercentile(_hy[m], 99.5))
        ax.grid(alpha=0.2, lw=0.5)
        ax.set_xlabel(r"$\log_{10}\,s_1$  — the initial rate  [dex Gyr$^{-1}$]", fontsize=9)
        if jb == 0:
            ax.set_ylabel("THE PLANE THE LIFETIME LIVES IN\n" + r"$y_0 - $ threshold  [dex]", fontsize=9)
    fig3.suptitle("Is the visibility time the divider? — the lifetime each branch's own first leg implies (top), and "
                  "the two factors it is made of\nwith the iso-lifetime lines drawn (bottom): a long-lived galaxy is "
                  "one that starts high, falls slowly, or both", y=1.005, fontsize=12, fontweight="bold")
    fig3.tight_layout(); paper_save(fig3, "kappa_visibility_split"); plt.show()

    # ══ (F) THE ARREST, IN THE CURRENCY OF THE OBSERVATION ══════════════════════════════════════════════════════════
    print("\n" + "-" * 132)
    print("(e) WHAT THE ARREST BUYS — the measured visibility time against the one the first leg implied.  A branch is")
    print("    still above the threshold at the anchor for one of TWO reasons, and they are physically different: its")
    print("    INITIAL RATE alone was slow enough to outlast the window (t_ini > t_post), or it should have crossed at")
    print("    that rate and did not — the decay ARRESTED.  The two shares below add up to the label")
    print(f"  {'box':10s} {'label':7s} {'N':>5s} {'med t_ini':>10s} {'med t_vis':>10s} {'slow start':>11s} "
          f"{'arrested':>9s} {'fell early':>11s} {'gain | crossed':>15s} {'gain | limit':>13s} {'rho':>6s}")
    for b in _BB19:
        for p_ in (("dusty", "pure") if "pop" in V19.columns else ("all",)):
            g = V19[(V19.box == b) & ((V19["pop"] == p_) if p_ != "all" else True)]
            g = g[g["vis0"].to_numpy(bool)]
            if len(g) < 8:
                continue
            ti, tv, ev, tp = (g["t_ini"].to_numpy(float), g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool),
                              g["t_post"].to_numpy(float))
            _fin = np.isfinite(ti)
            _slow = (~_fin) | (ti > tp)                    # the initial rate alone keeps it above the cut past the anchor
            _arr = _fin & (ti <= tp) & ~ev                 # it should have crossed inside the window, and it did not
            _early = _fin & (ti > tp) & ev                 # ... or it crossed although its first leg said it would not
            _gc = np.nanmedian((tv - ti)[_fin & ev]) if (_fin & ev).sum() >= 3 else np.nan
            _gl = np.nanmedian((tp - ti)[_arr]) if _arr.sum() >= 3 else np.nan   # only where the arrest is what saved it
            _ok = _fin & np.isfinite(tv) & ev
            _rr = float(spearmanr(ti[_ok], tv[_ok]).statistic) if _ok.sum() > 10 else np.nan
            print(f"  {b:10s} {p_:7s} {len(g):5d} {np.nanmedian(ti[_fin]):10.2f} {np.nanmedian(tv):10.2f} "
                  f"{100 * np.mean(_slow):10.0f}% {100 * np.mean(_arr):8.0f}% {100 * np.mean(_early):10.0f}% "
                  f"{_gc:+15.2f} {_gl:+13.2f} {_rr:6.2f}")
            ROW19.append(dict(box=b, block="arrest", quantity=p_, thr=THR19, n_gal=len(g), med=_gc,
                              lo=float(np.mean(_slow)), hi=float(np.mean(_arr)), x=_rr, p=float(np.mean(_early))))
    print( "  ('slow start' includes the branches whose first leg is flat or rising, which have no finite t_ini at all.")
    print( "   'gain | crossed' is t_vis - t_ini where both are measured; 'gain | limit' is t_post - t_ini over the")
    print( "   ARRESTED branches alone, a LOWER limit on what the arrest bought them.  rho is over the branches that DID cross,")
    print( "   where both numbers are measurements — it says how much of the lifetime the initial state and rate fix.)")

    # ── the same question as a budget: what is the visibility time MADE of?  log t_ini = log(head) - log(rate) ──
    print(f"\n  THE BUDGET OF THE LIFETIME.  t_ini = head / rate exactly, so var(log t) = var(log head) + var(log rate)")
    print(f"  - 2 cov: the two terms say whether an observably dusty QG is one that STARTED with more dust or one that")
    print(f"  LOSES it more slowly.  (A negative covariance term means the two work together.)")
    print(f"  {'box':10s} {'N':>5s} {'sd log t':>9s} {'var head':>9s} {'var rate':>9s} {'-2 cov':>8s} "
          f"{'rho(head, rate)':>16s} {'rho(t, head)':>13s} {'rho(t, rate)':>13s}")
    for b in _BB19:
        g = V19[V19.box == b]
        _h, _s = _l19(np.clip(g["head"].to_numpy(float), 1e-3, None)), _l19(g["s_ini"].to_numpy(float))
        _t = _h - _s
        m = np.isfinite(_h) & np.isfinite(_s)
        if m.sum() < _MND19:
            continue
        vh, vs = float(np.var(_h[m])), float(np.var(_s[m]))
        cv = -2.0 * float(np.cov(_h[m], _s[m])[0, 1])
        tot = vh + vs + cv
        print(f"  {b:10s} {int(m.sum()):5d} {np.sqrt(max(tot, 0)):9.2f} {100 * vh / tot:8.0f}% {100 * vs / tot:8.0f}% "
              f"{100 * cv / tot:7.0f}% {float(spearmanr(_h[m], _s[m]).statistic):16.2f} "
              f"{float(spearmanr(_t[m], _h[m]).statistic):13.2f} {float(spearmanr(_t[m], _s[m]).statistic):13.2f}")
        ROW19.append(dict(box=b, block="budget", quantity="var_shares", thr=THR19, n_gal=int(m.sum()),
                          med=float(np.sqrt(max(tot, 0))), lo=vh / tot, hi=vs / tot, x=cv / tot))
    print( "  (the head start is in dex and the rate in dex/Gyr, both taken in log10 so that the identity is exact.)")

    fig4, AX4 = plt.subplots(1, len(_BB19), figsize=(4.0 * len(_BB19), 3.8), squeeze=False)
    for jb, b in enumerate(_BB19):
        g = V19[(V19.box == b) & V19["vis0"].to_numpy(bool)]
        ax = AX4[0, jb]
        ti, tv, ev = g["t_ini"].to_numpy(float), g["t_vis"].to_numpy(float), g["ev"].to_numpy(bool)
        _c = (np.where(g["pop"].to_numpy() == "dusty", POPS["dusty"]["color"], POPS["pure"]["color"])
              if "pop" in g.columns else np.full(len(g), "0.4"))
        m = np.isfinite(ti) & np.isfinite(tv) & ev
        ax.scatter(np.clip(ti[m], 1e-2, None), tv[m], s=10, c=_c[m], alpha=0.45, edgecolor="none", zorder=3)
        m = np.isfinite(ti) & np.isfinite(tv) & ~ev
        ax.scatter(np.clip(ti[m], 1e-2, None), tv[m], s=24, c=_c[m], alpha=0.6, marker="^", edgecolor="none", zorder=4)
        _lim = (1e-2, max(_SMAX19, 1e-2))
        ax.plot(_lim, _lim, "-", color="0.3", lw=1.2)
        ax.set_xscale("log"); ax.set_xlim(*_lim); ax.set_ylim(0, _lim[1] * 1.05); ax.grid(alpha=0.2, lw=0.5)
        ax.set_title(BOX_LABEL[b], fontsize=10, color="#7f3b08" if b == NOX_BOX else "k")
        ax.set_xlabel(r"$t_{\rm ini}$ — what the first leg implied  [Gyr]", fontsize=9)
        if jb == 0:
            ax.set_ylabel(r"$t_{\rm vis}$ — what happened  [Gyr]" + "\n($\\blacktriangle$ = still visible: a limit)",
                          fontsize=9)
    fig4.suptitle("What the arrest buys — the visibility time against the one the branch's own first leg implied; "
                  "above the diagonal the decay stopped\nand the galaxy stayed observable longer than its initial "
                  "rate allowed", y=1.02, fontsize=12, fontweight="bold")
    fig4.tight_layout(); paper_save(fig4, "kappa_visibility_arrest"); plt.show()

    # ══ (G) THE READING ═════════════════════════════════════════════════════════════════════════════════════════════
    _R19 = pd.DataFrame(ROW19)

    def _g19(bx, blk, q, col="med"):
        m = (_R19.box == bx) & (_R19.block == blk) & (_R19.quantity == q)
        return float(_R19[m][col].iloc[0]) if m.any() and col in _R19.columns else np.nan

    _b019 = _BB19[0]
    print("\n" + "=" * 132)
    print("WHAT THE VISIBILITY TIME SAYS")
    print(f"  1. HOW LONG.  Above log M_dust/M* = {THR19:+.1f}, the median quenched branch stays observably dusty for "
          + ", ".join(f"{_g19(b, 'census', 'km_median'):.2f} Gyr ({b})" for b in _BB19 if np.isfinite(_g19(b, "census", "km_median")))
          + ".")
    _pl = {b: _g19(b, "census", "km_median", "x") for b in _BB19}
    print( "     The curve does not go to zero: a plateau of "
          + ", ".join(f"{100 * _pl[b]:.0f} % ({b})" for b in _BB19 if np.isfinite(_pl[b]))
          + " has not crossed when the follow-up ends,")
    print( "     and THAT plateau is the dusty QGs — not a slower decay, an arrested one.")
    _lad = _R19[(_R19.block == "ladder") & (_R19.box == _b019)].sort_values("thr", ascending=False)
    if len(_lad):
        print(f"  2. AT WHICH DEPTH.  'observable' is a threshold, and the lifetime moves with it — in {_b019}: "
              + ", ".join(f"{r.med:.2f} Gyr at {r.thr:+.1f}" for r in _lad.itertuples() if np.isfinite(r.med)) + ".")
    _dv = _R19[(_R19.block == "divider") & (_R19.quantity == "t_ini")]
    if len(_dv):
        print( "  3. DOES IT DIVIDE.  The counterfactual lifetime — what the branch's own first leg implies, before the")
        print( "     anchor is reached — scores AUC "
              + ", ".join(f"{r.auc:.2f} ({r.box})" for r in _dv.itertuples() if np.isfinite(r.auc))
              + " against the dust label,")
        print( "     and Silverman's test on it gives p = "
              + ", ".join(f"{r.p:.3f} ({r.box})" for r in _dv.itertuples() if np.isfinite(r.p))
              + ".  A p above 0.05 means ONE mode: the")
        print( "     lifetime separates the two labels on average without the population itself being two populations —")
        print( "     the same answer Part 18 got for log tau, now for the quantity the label is actually made of.")
    _ar = _R19[(_R19.block == "arrest") & (_R19.quantity == "dusty")]
    if len(_ar):
        print( "  4. WHY A DUSTY QG IS STILL VISIBLE — two reasons, and they are not the same physics.  Of the dusty")
        print( "     branches, "
              + ", ".join(f"{100 * r.lo:.0f} % ({r.box})" for r in _ar.itertuples() if np.isfinite(r.lo))
              + " outlast the window on their INITIAL RATE alone, while ")
        print( "     "
              + ", ".join(f"{100 * r.hi:.0f} % ({r.box})" for r in _ar.itertuples() if np.isfinite(r.hi))
              + " should have crossed the threshold inside the window at the rate they")
        print( "     started with and did not: for those the observability is bought AFTER the truncation, by the arrest")
        print( "     Part 18 timed, and not inherited from the dust they had at it.")
    _bg = _R19[(_R19.block == "budget")]
    if len(_bg):
        print( "     THE BUDGET.  Of the variance of the log lifetime, the head start carries "
              + ", ".join(f"{100 * r.lo:.0f} % ({r.box})" for r in _bg.itertuples() if np.isfinite(r.lo))
              + " and the initial rate")
        print( "     " + ", ".join(f"{100 * r.hi:.0f} %" for r in _bg.itertuples() if np.isfinite(r.hi))
              + " — which of the two an observably dusty QG owes its dust to.")
    _dm = _R19[(_R19.block == "demography") & (_R19.box == _b019)]
    if len(_dm) > 2:
        _rr = float(spearmanr(_dm["med"], _dm["auc"]).statistic)
        _bi = float(np.nanmedian(_dm["med"] - _dm["auc"]))
        print(f"  5. THE EPOCH TREND IS A LOOK-BACK EFFECT (rho = {_rr:+.2f} between the predicted and the observed dusty")
        print(f"     fraction across the {len(_dm)} anchors of {_b019}, median offset {100 * _bi:+.0f} points).  The survival curve of")
        print( "     the other anchors plus the delay since each galaxy's own truncation reproduces the fraction seen:")
        print( "     high-z anchors are dusty-rich because they catch their quenched galaxies sooner, not because the")
        print( "     dust there lives longer.  Where the offset is large the lifetime itself differs — that is the")
        print( "     residual worth reading.")
    print("=" * 132)

    # ══ (H) THE PRODUCTS ════════════════════════════════════════════════════════════════════════════════════════════
    pd.DataFrame(ROW19).to_csv(os.path.join(OUT, "kappa_visibility.csv"), index=False)
    V19.to_csv(os.path.join(OUT, "kappa_visibility_galaxies.csv"), index=False)
    print("\ntable  -> " + os.path.join(OUT, "kappa_visibility.csv") + "  (every row of every block above)")
    print("table  -> " + os.path.join(OUT, "kappa_visibility_galaxies.csv")
          + f"  ({len(V19)} branches: the crossing at every threshold of the ladder, the head start, the initial rate,")
    print("          the counterfactual lifetime and the censoring flag)")
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# Part 18 — DUST EVOLUTION AFTER SFT
#
# Goal:
#   Fit the post-SFT evolution of fdust = Mdust/Mstar galaxy by galaxy.
#
#   The evolution is NOT assumed to be a decay.
#   Each component of the double exponential may:
#
#       k > 0   -> decay
#       k = 0   -> flat
#       k < 0   -> growth
#
#
# Model:
#
#   fdust(t) / fdust(t0) =
#
#       (1-f2) exp(-k1 t) + f2 exp(-k2 t)
#
#
# Because the curve is normalized at t=0, the model has only 3 free parameters:
#
#       k1, k2, f2
#
# so a fit is attempted whenever >= 3 snapshots are available.
#
#
# OUTPUT
# ------
# D18 contains one row per galaxy with:
#
#   k_fast, k_slow        signed rates [Gyr^-1]
#   tau_fast, tau_slow    |1/k| [Gyr]
#   fslow                 weight of the slower-rate component
#
#   k_eff0                instantaneous effective rate at SFT
#   tau_eff0              |1/k_eff0|
#
#   delta_lfd             total change in log fdust over observed interval
#   lfd_final             final log fdust
#   evolution             decay / flat / growth according to net fitted change
#
#
# FIGURES
# -------
# 1. Gallery of galaxy-by-galaxy fits
# 2. Final fdust against effective signed evolution rate
# 3. Final fdust against characteristic timescale, separating
#    declining and growing galaxies
#
# ═══════════════════════════════════════════════════════════════════════════════


import numpy as np
import pandas as pd
import h5py
import matplotlib.pyplot as plt

from scipy.optimize import least_squares
from scipy.stats import spearmanr


# =============================================================================
# SETTINGS
# =============================================================================

P18_BOXES = None

# fit every galaxy with at least this many post-SFT snapshots
P18_NMIN = 3

# allowed signed exponential rates [Gyr^-1]
#
# |k| = 20 Gyr^-1 corresponds to tau = 0.05 Gyr
P18_KMAX = 20.0

# below this the rate is effectively flat
P18_KZERO = 0.02

# number of different starting points for the optimizer
P18_NSTART = 20

# fit gallery
P18_NSHOW = 24
P18_SEED = 7


# =============================================================================
# MODEL
# =============================================================================

def _double_exp18(t, k1, k2, f2):
    """
    Normalized two-component exponential evolution.

    k > 0 : decay
    k = 0 : constant
    k < 0 : growth

    By construction model(0) = 1.
    """

    return (
        (1.0 - f2) * np.exp(-k1 * t)
        + f2 * np.exp(-k2 * t)
    )


# =============================================================================
# FIT ONE GALAXY
# =============================================================================

def _fit_double_exp18(t, fd):
    """
    Fit one post-SFT fdust history.

    Parameters
    ----------
    t : array
        Cosmic time [Gyr].

    fd : array
        Mdust / Mstar.

    Returns
    -------
    dictionary or None
    """

    t = np.asarray(t, dtype=float)
    fd = np.asarray(fd, dtype=float)

    # -------------------------------------------------------------------------
    # valid snapshots
    # -------------------------------------------------------------------------

    good = (
        np.isfinite(t)
        & np.isfinite(fd)
        & (fd > 0)
    )

    t = t[good]
    fd = fd[good]

    if len(t) < P18_NMIN:
        return None

    # chronological order
    o = np.argsort(t)

    t = t[o]
    fd = fd[o]

    # -------------------------------------------------------------------------
    # time since first available post-SFT snapshot
    # -------------------------------------------------------------------------

    x = t - t[0]

    span = float(x[-1])

    if span <= 0:
        return None

    # -------------------------------------------------------------------------
    # normalize by the first post-SFT fdust
    # -------------------------------------------------------------------------

    fd0 = float(fd[0])

    y = fd / fd0

    if not np.all(np.isfinite(y)):
        return None

    # -------------------------------------------------------------------------
    # floor in normalized units
    #
    # P9_FLOOR is the log10 fdust floor already used elsewhere in the analysis.
    # -------------------------------------------------------------------------

    floor_abs = 10.0 ** P9_FLOOR

    floor_norm = max(
        floor_abs / fd0,
        1e-15,
    )

    logy = np.log10(
        np.clip(y, floor_norm, None)
    )

    # -------------------------------------------------------------------------
    # residual in log(fdust)
    #
    # fitting in log space prevents the first high-fdust points from completely
    # dominating the fit.
    # -------------------------------------------------------------------------

    def residual(p):

        k1, k2, f2 = p

        ym = _double_exp18(
            x,
            k1,
            k2,
            f2,
        )

        logym = np.log10(
            np.clip(ym, 1e-15, None)
        )

        return logym - logy

    # -------------------------------------------------------------------------
    # rough net rate from first to last point
    #
    # y(end) = exp(-k_net * span)
    #
    # therefore:
    #
    # k_net = -ln(y_end)/span
    #
    # This is ONLY used as an optimizer starting point.
    # -------------------------------------------------------------------------

    yend = max(float(y[-1]), 1e-15)

    k_net_guess = -np.log(yend) / span

    k_net_guess = np.clip(
        k_net_guess,
        -0.8 * P18_KMAX,
        +0.8 * P18_KMAX,
    )

    # -------------------------------------------------------------------------
    # multiple starting points
    #
    # Two-exponential fits are degenerate under component swapping and can have
    # local minima. Try several starting combinations and keep the lowest-cost
    # solution.
    # -------------------------------------------------------------------------

    rng = np.random.default_rng(
        int(
            abs(
                hash(
                    (
                        round(float(t[0]), 5),
                        round(float(fd0), 12),
                        len(t),
                    )
                )
            )
            % (2**32)
        )
    )

    starts = []

    # sensible deterministic starts
    starts.extend(
        [
            [k_net_guess, 0.5 * k_net_guess, 0.5],
            [2.0 * k_net_guess, 0.5 * k_net_guess, 0.3],
            [0.5 * k_net_guess, 2.0 * k_net_guess, 0.7],

            [1.0, 0.1, 0.5],
            [3.0, 0.3, 0.3],
            [0.3, 3.0, 0.7],

            [-1.0, -0.1, 0.5],
            [-3.0, -0.3, 0.3],
            [-0.3, -3.0, 0.7],

            [1.0, -0.5, 0.5],
            [-0.5, 1.0, 0.5],
        ]
    )

    # random starts
    while len(starts) < P18_NSTART:

        starts.append(
            [
                rng.uniform(
                    -0.8 * P18_KMAX,
                    +0.8 * P18_KMAX,
                ),

                rng.uniform(
                    -0.8 * P18_KMAX,
                    +0.8 * P18_KMAX,
                ),

                rng.uniform(
                    0.05,
                    0.95,
                ),
            ]
        )

    bounds = (
        [
            -P18_KMAX,
            -P18_KMAX,
            0.0,
        ],
        [
            +P18_KMAX,
            +P18_KMAX,
            1.0,
        ],
    )

    best = None

    for p0 in starts:

        p0 = np.asarray(
            p0,
            dtype=float,
        )

        p0[:2] = np.clip(
            p0[:2],
            -0.999 * P18_KMAX,
            +0.999 * P18_KMAX,
        )

        p0[2] = np.clip(
            p0[2],
            1e-4,
            1.0 - 1e-4,
        )

        try:

            fit = least_squares(
                residual,
                p0,
                bounds=bounds,
                max_nfev=10000,
            )

        except Exception:

            continue

        if not fit.success:
            continue

        if (
            best is None
            or fit.cost < best.cost
        ):
            best = fit

    if best is None:
        return None

    # =========================================================================
    # FITTED PARAMETERS
    # =========================================================================

    k1, k2, f2 = best.x

    # -------------------------------------------------------------------------
    # order by rate
    #
    # Larger k means faster decline.
    #
    # Therefore:
    #
    #     k_fast >= k_slow
    #
    # Note that k_slow can be negative, meaning the second component grows.
    # -------------------------------------------------------------------------

    if k1 >= k2:

        k_fast = float(k1)
        k_slow = float(k2)

        fslow = float(f2)

    else:

        k_fast = float(k2)
        k_slow = float(k1)

        fslow = float(1.0 - f2)

    ffast = 1.0 - fslow

    # -------------------------------------------------------------------------
    # corresponding absolute timescales
    #
    # Keep the sign separately in k.
    # -------------------------------------------------------------------------

    tau_fast = (
        1.0 / abs(k_fast)
        if abs(k_fast) > P18_KZERO
        else np.inf
    )

    tau_slow = (
        1.0 / abs(k_slow)
        if abs(k_slow) > P18_KZERO
        else np.inf
    )

    # -------------------------------------------------------------------------
    # effective rate at t=0
    #
    # Since y(0)=1,
    #
    #       -dy/dt |0 = (1-fslow) k_fast + fslow k_slow
    #
    # so this is the instantaneous signed evolution rate at SFT.
    #
    # positive -> initially declining
    # negative -> initially growing
    # -------------------------------------------------------------------------

    k_eff0 = (
        ffast * k_fast
        + fslow * k_slow
    )

    tau_eff0 = (
        1.0 / abs(k_eff0)
        if abs(k_eff0) > P18_KZERO
        else np.inf
    )

    # -------------------------------------------------------------------------
    # fitted model
    # -------------------------------------------------------------------------

    model = _double_exp18(
        x,
        k_fast,
        k_slow,
        fslow,
    )

    logmodel = np.log10(
        np.clip(
            model,
            1e-15,
            None,
        )
    )

    rms = float(
        np.sqrt(
            np.mean(
                (logmodel - logy) ** 2
            )
        )
    )

    # -------------------------------------------------------------------------
    # net fitted change over the actual observed interval
    #
    # This is used only to say whether the WHOLE fitted track ends above or
    # below where it started.
    # -------------------------------------------------------------------------

    yfit_final = float(model[-1])

    delta_lfd_fit = float(
        np.log10(
            max(
                yfit_final,
                1e-15,
            )
        )
    )

    # observed change
    delta_lfd = float(
        np.log10(
            max(
                y[-1],
                1e-15,
            )
        )
    )

    # -------------------------------------------------------------------------
    # simple label based on total fitted change
    #
    # ±0.1 dex is treated as approximately flat.
    # -------------------------------------------------------------------------

    if delta_lfd_fit < -0.1:

        evolution = "decay"

    elif delta_lfd_fit > +0.1:

        evolution = "growth"

    else:

        evolution = "flat"

    # -------------------------------------------------------------------------
    # component labels
    # -------------------------------------------------------------------------

    def _component_type(k):

        if k > P18_KZERO:
            return "decay"

        if k < -P18_KZERO:
            return "growth"

        return "flat"

    type_fast = _component_type(
        k_fast
    )

    type_slow = _component_type(
        k_slow
    )

    return dict(

        # sampling
        n=int(len(x)),
        span=float(span),

        # actual data
        t=x,
        fd=fd,
        fd0=fd0,
        y=y,

        # fitted components
        k_fast=k_fast,
        k_slow=k_slow,

        tau_fast=tau_fast,
        tau_slow=tau_slow,

        ffast=ffast,
        fslow=fslow,

        type_fast=type_fast,
        type_slow=type_slow,

        # effective initial evolution
        k_eff0=float(k_eff0),
        tau_eff0=float(tau_eff0),

        # observed/fitted overall evolution
        delta_lfd=float(delta_lfd),
        delta_lfd_fit=float(delta_lfd_fit),

        evolution=evolution,

        rms=rms,
    )


# =============================================================================
# READ THE HISTORIES
# =============================================================================

BOXES18 = (
    list(H)
    if P18_BOXES is None
    else [
        b
        for b in P18_BOXES
        if b in H
    ]
)

# put the channel-off box last
BOXES18 = (
    [
        b
        for b in BOXES18
        if b != NOX_BOX
    ]
    +
    [
        b
        for b in BOXES18
        if b == NOX_BOX
    ]
)


rows18 = []
R18 = {}


for box in BOXES18:

    print(
        f"Reading {box}..."
    )

    F = H[box]["F"]

    # galaxies belonging to the Part-3 sample
    want = {
        (
            int(r["snap"]),
            int(r["gal_id"]),
        ): r

        for _, r in F.iterrows()
    }

    with h5py.File(
        HIST_FILE.format(
            box=box
        ),
        "r",
    ) as f:

        for gname in f:

            snap = int(
                gname.replace(
                    "snap",
                    "",
                )
            )

            # galaxies required at this anchor
            wanted_here = {
                key: row
                for key, row in want.items()
                if key[0] == snap
            }

            if not wanted_here:
                continue

            gh = f[gname]

            # --------------------------------------------------------------
            # times
            # --------------------------------------------------------------

            t = (
                gh["t_yr"][:]
                / 1e9
            )

            order = np.argsort(t)

            t = t[order]

            # --------------------------------------------------------------
            # masses
            # --------------------------------------------------------------

            ms = gh["mstar"][:][order]

            md = (
                gh["mdust"][:][order]
                if "mdust" in gh
                else np.full_like(
                    ms,
                    np.nan,
                )
            )

            # --------------------------------------------------------------
            # loop through galaxies
            # --------------------------------------------------------------

            for j, gid0 in enumerate(
                gh["gal_ids"][:]
            ):

                gid = int(gid0)

                key = (
                    snap,
                    gid,
                )

                if key not in wanted_here:
                    continue

                r = wanted_here[key]

                t_sft = float(
                    r.get(
                        "t_qt",
                        np.nan,
                    )
                )

                if not np.isfinite(t_sft):
                    continue

                # ----------------------------------------------------------
                # fdust = Mdust/Mstar
                # ----------------------------------------------------------

                valid = (
                    np.isfinite(
                        ms[:, j]
                    )
                    &
                    np.isfinite(
                        md[:, j]
                    )
                    &
                    (ms[:, j] > 0)
                )

                with np.errstate(
                    divide="ignore",
                    invalid="ignore",
                ):

                    fd = np.where(
                        valid,

                        np.clip(
                            md[:, j]
                            / ms[:, j],

                            10.0 ** P9_FLOOR,
                            None,
                        ),

                        np.nan,
                    )

                # ----------------------------------------------------------
                # ALL AVAILABLE SNAPSHOTS FROM SFT ONWARD
                # ----------------------------------------------------------

                post = (
                    valid
                    &
                    np.isfinite(fd)
                    &
                    (
                        t
                        >= t_sft - 1e-9
                    )
                )

                if (
                    np.sum(post)
                    < P18_NMIN
                ):
                    continue

                tt = t[post]
                ff = fd[post]

                # ----------------------------------------------------------
                # fit
                # ----------------------------------------------------------

                fit = _fit_double_exp18(
                    tt,
                    ff,
                )

                if fit is None:
                    continue

                # ----------------------------------------------------------
                # final dust content
                # ----------------------------------------------------------

                fd_final = float(
                    ff[-1]
                )

                lfd_final = float(
                    np.log10(
                        fd_final
                    )
                )

                rec = dict(

                    box=box,
                    snap=snap,
                    gal_id=gid,

                    t_sft=t_sft,

                    n=fit["n"],
                    span=fit["span"],

                    k_fast=fit["k_fast"],
                    k_slow=fit["k_slow"],

                    tau_fast=fit["tau_fast"],
                    tau_slow=fit["tau_slow"],

                    ffast=fit["ffast"],
                    fslow=fit["fslow"],

                    type_fast=fit["type_fast"],
                    type_slow=fit["type_slow"],

                    k_eff0=fit["k_eff0"],
                    tau_eff0=fit["tau_eff0"],

                    delta_lfd=fit["delta_lfd"],
                    delta_lfd_fit=fit["delta_lfd_fit"],

                    evolution=fit["evolution"],

                    rms=fit["rms"],

                    fd0=fit["fd0"],
                    fd_final=fd_final,
                    lfd_final=lfd_final,
                )

                # ----------------------------------------------------------
                # carry Part-3 properties
                # ----------------------------------------------------------

                for c in (
                    "z",
                    "pop",
                    "lfd",
                    "age",
                    "log_mstar",
                    "log_mbh",
                    "kcls",
                    "rot",
                    "central",
                    "t_anchor",
                    "kappa",
                    "ngas",
                    "t_qt",
                ):

                    if c in r:
                        rec[c] = r[c]

                rows18.append(rec)

                R18[
                    (
                        box,
                        snap,
                        gid,
                    )
                ] = fit


D18 = pd.DataFrame(
    rows18
)


# =============================================================================
# SUMMARY
# =============================================================================

print(
    "\n"
    + "=" * 100
)

print(
    "DOUBLE-EXPONENTIAL POST-SFT DUST EVOLUTION"
)

print(
    "=" * 100
)


if len(D18) == 0:

    print(
        "No galaxy could be fitted."
    )

else:

    print(
        f"\nTotal fitted galaxies: "
        f"{len(D18)}"
    )

    print(
        "\nEvolution is classified from "
        "the TOTAL fitted change over the observed post-SFT interval."
    )

    for box in BOXES18:

        g = D18[
            D18.box == box
        ]

        if len(g) == 0:
            continue

        ndec = np.sum(
            g.evolution == "decay"
        )

        nflat = np.sum(
            g.evolution == "flat"
        )

        ngrow = np.sum(
            g.evolution == "growth"
        )

        print(
            f"\n{box}"
        )

        print(
            f"    N = {len(g)}"
        )

        print(
            f"    decay  = "
            f"{ndec:4d} "
            f"({100*ndec/len(g):5.1f}%)"
        )

        print(
            f"    flat   = "
            f"{nflat:4d} "
            f"({100*nflat/len(g):5.1f}%)"
        )

        print(
            f"    growth = "
            f"{ngrow:4d} "
            f"({100*ngrow/len(g):5.1f}%)"
        )

        print(
            f"    median RMS = "
            f"{np.nanmedian(g.rms):.3f} dex"
        )

        print(
            f"    median snapshots = "
            f"{np.nanmedian(g.n):.0f}"
        )

        print(
            f"    median span = "
            f"{np.nanmedian(g.span):.2f} Gyr"
        )


# =============================================================================
# FIGURE 1
#
# FIT GALLERY
#
# Show tracks spanning the full range of net evolution:
# strong decay -> flat -> strong growth.
# =============================================================================

if len(D18):

    tmp = D18[
        np.isfinite(
            D18.delta_lfd_fit
        )
    ].sort_values(
        "delta_lfd_fit"
    )

    nshow = min(
        P18_NSHOW,
        len(tmp),
    )

    idx = np.unique(
        np.linspace(
            0,
            len(tmp) - 1,
            nshow,
        ).astype(int)
    )

    show = tmp.iloc[idx]

    ncol = 6

    nrow = int(
        np.ceil(
            len(show)
            / ncol
        )
    )

    fig, axes = plt.subplots(
        nrow,
        ncol,

        figsize=(
            3.5 * ncol,
            2.8 * nrow,
        ),

        squeeze=False,
    )

    for ax, (_, row) in zip(
        axes.ravel(),
        show.iterrows(),
    ):

        key = (
            row["box"],
            int(row["snap"]),
            int(row["gal_id"]),
        )

        fit = R18[key]

        t = fit["t"]
        y = fit["y"]

        # data
        ax.plot(
            t,
            np.log10(y),
            "o",
            color="k",
            ms=4.5,
            zorder=5,
        )

        # fitted curve
        xx = np.linspace(
            0,
            max(
                t[-1],
                1e-3,
            ),
            400,
        )

        yy = _double_exp18(
            xx,
            fit["k_fast"],
            fit["k_slow"],
            fit["fslow"],
        )

        ax.plot(
            xx,
            np.log10(
                np.clip(
                    yy,
                    1e-15,
                    None,
                )
            ),
            lw=2.0,
            zorder=4,
        )

        # zero-change level
        ax.axhline(
            0,
            color="0.75",
            lw=0.8,
            ls=":",
        )

        # --------------------------------------------------------------
        # useful labels
        # --------------------------------------------------------------

        def rate_label(
            k,
            tau,
        ):

            if k > P18_KZERO:

                return (
                    rf"$\tau={tau:.2f}$"
                    "\n"
                    "decay"
                )

            elif k < -P18_KZERO:

                return (
                    rf"$\tau={tau:.2f}$"
                    "\n"
                    "growth"
                )

            else:

                return "flat"

        lab1 = rate_label(
            row["k_fast"],
            row["tau_fast"],
        )

        lab2 = rate_label(
            row["k_slow"],
            row["tau_slow"],
        )

        ax.set_title(
            f"{row['box']}  "
            f"s{int(row['snap'])} "
            f"g{int(row['gal_id'])}\n"
            f"{row['evolution']}   "
            f"$f_2$={row['fslow']:.2f}\n"
            f"1: {lab1}   "
            f"2: {lab2}",
            fontsize=7.5,
        )

        ax.set_xlabel(
            r"$t-t_{\rm first,post-SFT}$ "
            r"[Gyr]",
            fontsize=8,
        )

        ax.set_ylabel(
            r"$\log_{10}"
            r"[f_{\rm dust}/f_{\rm dust,0}]$",
            fontsize=8,
        )

        ax.tick_params(
            labelsize=7
        )


    for ax in axes.ravel()[
        len(show):
    ]:

        ax.set_axis_off()


    fig.suptitle(
        "Post-SFT dust evolution: "
        "double-exponential fits\n"
        "from strongest decline to strongest growth",
        fontsize=13,
        fontweight="bold",
    )

    fig.tight_layout()

    paper_save(
        fig,
        "fdust_double_exponential_gallery",
    )

    plt.show()


# =============================================================================
# FIGURE 2
#
# MAIN RESULT:
#
# FINAL fdust AGAINST THE SIGNED EFFECTIVE RATE AT SFT
#
#       k_eff0 > 0 : initially declining
#       k_eff0 = 0 : stable
#       k_eff0 < 0 : initially growing
#
# This remains continuous through zero, unlike tau.
# =============================================================================

if len(D18):

    boxes_plot = [
        b
        for b in BOXES18
        if np.any(
            D18.box == b
        )
    ]

    fig, axes = plt.subplots(
        1,
        len(boxes_plot),

        figsize=(
            4.5 * len(boxes_plot),
            4.2,
        ),

        squeeze=False,
    )

    axes = axes[0]


    for ax, box in zip(
        axes,
        boxes_plot,
    ):

        g = D18[
            D18.box == box
        ].copy()

        ok = (
            np.isfinite(
                g.k_eff0
            )
            &
            np.isfinite(
                g.lfd_final
            )
        )

        g = g[ok]

        x = g.k_eff0.to_numpy(
            float
        )

        y = g.lfd_final.to_numpy(
            float
        )

        ax.scatter(
            x,
            y,
            s=18,
            alpha=0.55,
            edgecolor="none",
        )

        # zero = neither initial growth nor decline
        ax.axvline(
            0,
            color="0.5",
            lw=1,
            ls="--",
        )

        # correlation
        if len(g) >= 3:

            rho, p = spearmanr(
                x,
                y,
            )

            ax.text(
                0.05,
                0.95,

                rf"$N={len(g)}$"
                "\n"
                rf"$\rho={rho:.2f}$"
                "\n"
                rf"$p={p:.2g}$",

                transform=ax.transAxes,

                ha="left",
                va="top",

                fontsize=9,
            )

        ax.set_xlabel(
            r"signed effective rate at SFT "
            r"$k_{\rm eff,0}$ "
            r"[Gyr$^{-1}$]"
        )

        ax.set_title(
            BOX_LABEL.get(
                box,
                box,
            )
        )


    axes[0].set_ylabel(
        r"final "
        r"$\log_{10}(M_{\rm dust}/M_\star)$"
    )


    fig.suptitle(
        "Does post-SFT dust evolution determine how much dust remains?\n"
        r"$k<0$: growth   $\leftarrow$   "
        r"$k=0$   $\rightarrow$   "
        r"$k>0$: decay",
        fontsize=13,
        fontweight="bold",
    )

    fig.tight_layout()

    paper_save(
        fig,
        "fdust_signed_rate_vs_final_fdust",
    )

    plt.show()


# =============================================================================
# FIGURE 3
#
# CHARACTERISTIC TIMESCALE
#
# Here growth and decay are separated because |1/k| loses the sign.
#
# For declining galaxies:
#      longer tau_eff0 = slower initial decay
#
# For growing galaxies:
#      longer tau_eff0 = slower initial growth
# =============================================================================

if len(D18):

    boxes_plot = [
        b
        for b in BOXES18
        if np.any(
            D18.box == b
        )
    ]

    fig, axes = plt.subplots(
        1,
        len(boxes_plot),

        figsize=(
            4.5 * len(boxes_plot),
            4.2,
        ),

        squeeze=False,
    )

    axes = axes[0]


    for ax, box in zip(
        axes,
        boxes_plot,
    ):

        g = D18[
            D18.box == box
        ].copy()

        # --------------------------------------------------------------
        # initially declining
        # --------------------------------------------------------------

        gd = g[
            (
                g.k_eff0
                > P18_KZERO
            )
            &
            np.isfinite(
                g.tau_eff0
            )
            &
            np.isfinite(
                g.lfd_final
            )
        ]

        if len(gd):

            ax.scatter(
                np.log10(
                    gd.tau_eff0
                ),
                gd.lfd_final,

                s=20,
                alpha=0.55,

                label="initial decay",
                edgecolor="none",
            )

        # --------------------------------------------------------------
        # initially growing
        # --------------------------------------------------------------

        gg = g[
            (
                g.k_eff0
                < -P18_KZERO
            )
            &
            np.isfinite(
                g.tau_eff0
            )
            &
            np.isfinite(
                g.lfd_final
            )
        ]

        if len(gg):

            ax.scatter(
                np.log10(
                    gg.tau_eff0
                ),
                gg.lfd_final,

                marker="^",

                s=24,
                alpha=0.55,

                label="initial growth",
                edgecolor="none",
            )

        # --------------------------------------------------------------
        # correlation for DECAYING galaxies only
        #
        # This directly tests the original idea:
        #
        # slower decay -> more dust left?
        # --------------------------------------------------------------

        if len(gd) >= 3:

            rho, p = spearmanr(
                np.log10(
                    gd.tau_eff0
                ),
                gd.lfd_final,
            )

            text = (
                "decays only\n"
                rf"$N={len(gd)}$"
                "\n"
                rf"$\rho={rho:.2f}$"
                "\n"
                rf"$p={p:.2g}$"
            )

            ax.text(
                0.05,
                0.95,
                text,

                transform=ax.transAxes,

                ha="left",
                va="top",

                fontsize=9,
            )

        ax.set_xlabel(
            r"$\log_{10}"
            r"|\tau_{\rm eff,0}|$ "
            r"[Gyr]"
        )

        ax.set_title(
            BOX_LABEL.get(
                box,
                box,
            )
        )

        ax.legend(
            frameon=False,
            fontsize=8,
        )


    axes[0].set_ylabel(
        r"final "
        r"$\log_{10}(M_{\rm dust}/M_\star)$"
    )


    fig.suptitle(
        "Characteristic dust-evolution timescale\n"
        "for decaying galaxies, larger "
        r"$\tau$ means slower dust loss",
        fontsize=13,
        fontweight="bold",
    )

    fig.tight_layout()

    paper_save(
        fig,
        "fdust_timescale_vs_final_fdust",
    )

    plt.show()


# =============================================================================
# FIGURE 4
#
# THE TWO EXPONENTIAL COMPONENTS THEMSELVES
#
# Useful mainly as a diagnostic: are galaxies dominated by
#
#   decay + decay
#   decay + growth
#   growth + growth
#
# and how much weight does the slower component carry?
# =============================================================================

if len(D18):

    boxes_plot = [
        b
        for b in BOXES18
        if np.any(
            D18.box == b
        )
    ]

    fig, axes = plt.subplots(
        1,
        len(boxes_plot),

        figsize=(
            4.5 * len(boxes_plot),
            4.2,
        ),

        squeeze=False,
    )

    axes = axes[0]


    for ax, box in zip(
        axes,
        boxes_plot,
    ):

        g = D18[
            D18.box == box
        ]

        sc = ax.scatter(
            g.k_fast,
            g.k_slow,

            c=g.fslow,

            s=20,
            alpha=0.65,

            edgecolor="none",
        )

        ax.axhline(
            0,
            color="0.6",
            lw=0.8,
            ls="--",
        )

        ax.axvline(
            0,
            color="0.6",
            lw=0.8,
            ls="--",
        )

        lo = min(
            np.nanmin(
                g.k_fast
            ),
            np.nanmin(
                g.k_slow
            ),
            0,
        )

        hi = max(
            np.nanmax(
                g.k_fast
            ),
            np.nanmax(
                g.k_slow
            ),
            0,
        )

        ax.plot(
            [lo, hi],
            [lo, hi],
            ":",
            color="0.6",
            lw=1,
        )

        ax.set_xlabel(
            r"$k_{\rm fast}$ "
            r"[Gyr$^{-1}$]"
        )

        ax.set_ylabel(
            r"$k_{\rm slow}$ "
            r"[Gyr$^{-1}$]"
        )

        ax.set_title(
            BOX_LABEL.get(
                box,
                box,
            )
        )


    cbar = fig.colorbar(
        sc,
        ax=axes.tolist(),
        pad=0.02,
    )

    cbar.set_label(
        r"slow-component weight "
        r"$f_{\rm slow}$"
    )


    fig.suptitle(
        "The two fitted dust-evolution components\n"
        r"$k>0$: decay;   "
        r"$k<0$: growth",
        fontsize=13,
        fontweight="bold",
    )

    fig.subplots_adjust(
        top=0.80,
        bottom=0.15,
        wspace=0.30,
    )

    paper_save(
        fig,
        "fdust_double_exponential_components",
    )

    plt.show()


# =============================================================================
# SAVE TABLE
# =============================================================================

if len(D18):

    outcols = [
        c
        for c in D18.columns
        if c not in (
            "t",
            "fd",
            "y",
        )
    ]

    D18[
        outcols
    ].to_csv(
        "fdust_double_exponential_fits.csv",
        index=False,
    )

    print(
        "\nSaved:"
        "\n  fdust_double_exponential_fits.csv"
    )
```

### The selection

* `s_iso(N) = 0.385/sqrt(N)` is the **sampling** standard deviation of the Sales+12 $\kappa_{\rm rot}$ estimator under an
  isotropic velocity field: $\mathrm{Var}(v_\phi^2/v^2) = 4/45$ for a uniform direction and $N_{\rm eff} = 0.6\,N$ under the
  $v^2$ weighting. Both constants are knobs (`KAPPA_VARF`, `KAPPA_WEFF`). It is a *floor* on the uncertainty: it says nothing
  about clumping, about the finite gravitational softening, or about how well 76 particles trace a real velocity field.
* The observed spread of $\kappa$ is 5–9 times $s_{\rm iso}$, so the axis is dominated by real structure and the threshold
  only protects the poorly sampled end. Moving the cut between 1 and 3 $\sigma$ changes the selected fraction by a few per
  cent of the sample (Part 2d).
* **"Rotating" is a statement about the measurement, not a morphological class.** $\kappa_{\rm rot} = 1/3$ is isotropic and
  $\approx 0.5$ is rotation-supported; the cut lies between them. Part 2 prints the share above 0.5 outright for comparison.
* The classes are defined **at the anchor**. Every track in Parts 3–6 is therefore the history of the galaxies that *ended*
  in that class, not the path of a fixed population.
* **The rule has a second axis, `ROT_MASS = 10.5`** (the old Part 32's pair), ANDed into the kinematic cut in one place —
  `rot_rule()` in Part 0 — so Part 2's sample and Parts 6–8's larger K-S anchor set are cut identically. It is worth +0.02
  (m100) and +0.05 (m50) of out-of-fold AUC and roughly triples the purity of the selected group; in **m25 — the box that
  resolves the gas — it is worth nothing** ($\mathrm{AUC} = 0.53$, out of fold 0.47), and there it costs half the
  completeness for no purity. Part 35 of the old notebook says why: the physical axis there is $\Sigma_{\rm gas}$, and
  $M_\star$ is the coarse boxes' unresolved stand-in for it. Part 2 prints that trade box by box, and `ROT_MASS = None`
  recovers the purely kinematic rule everywhere. With the axis on, the second class is *everything else* — a galaxy is in it
  for either reason — which is what "the rest" means in every legend.
* **The AGN dose is not a division anywhere in this notebook.** $E_x$, $f_x$ and $f_x^{\rm post}$ are computed in Parts 3 and
  6 and written to the CSVs as per-galaxy numbers, but nothing is ever split by them: the earlier version's four-way
  dose $\times$ class grid, and the panel that scored the dose as an alternative divider, are both gone. What the X-ray
  channel does is Part 4's subject (it dates the dust destruction), and the channel-off box is the control there.

### The branch index, the particle zone census, the profiles and the CGM phases (Parts 9–13, m25 only)

Parts 9–13 are the 25 Mpc box only: it is the box that resolves the gas (Part 2 shows the mass axis buys nothing there), and
the only one with particle cutouts along the branch. They add four caches, all written beside the figures:

* **the branch index** (`P9_IDX`) — the most-massive-progenitor walk, $(\text{anchor snap}, \text{gal id}) \rightarrow
  g_x(\text{snap})$, from caesar's `tree_data/progen_galaxy_star` where the catalogue has it and from the `progen_links`
  sidecar otherwise (the m25 case). It is what lets Part 10 find the particle file of a *progenitor*, and what makes the
  de-duplication **exact**: two anchors are the same galaxy when one lies on the other's branch, not when a fingerprint
  matches (which is how Part 6 does it, and which cannot de-duplicate a branch that starts after its reference times).
  75 anchor rows → 44 galaxies, 11 rotating and 33 the rest.
* **the particle zone census** (`P10_CACHE`) — one row per reduced particle file on a retained branch: the dust, gas and H₂
  masses of the three zones, the mass-weighted radial kinematics of each under both weights, the half-mass radii and the
  shell fluxes. The first run reads several GiB; the second is instant. Delete the file to rebuild it, and note that adding a
  **column** to `_census10` needs the file deleted too — the cache is keyed on `(snap, gx)` and only *missing rows* are filled.
* **the radial profiles** (`P12_CACHE`, an `.npz`) — the binned mass and count profiles of gas, H₂, dust and stars on both
  radial grids (physical kpc, and $R/R_{50}^\star$), for the member gas and for every particle in the slab, one row per file.
  Only *masses and counts* are stored, so the surface densities, the apertures and the hole metric are all recomputed at
  draw time and every threshold can be changed without re-reading anything. The bins and the geometry are stored with it and
  checked on load: change `P12_RB`, `P12_ZMAX`, `P12_FRAME` or `P12_MEMBER` and the cell says so and rebuilds. By default it
  holds only the four stage nodes of each galaxy (~4 files each); `P12_ALLNODES = True` adds the rest, incrementally.
* **the CGM phase census** (`P13_CACHE`) — one row per file: the CGM decomposed into cold / warm / hot and into four radial
  shells, with the mass- and dust-weighted temperatures, the inflow and outflow through a fixed shell split by phase, the
  kinematics of the CGM and of the outer ISM, and the ISM's compactness. It is a **second pass over the same files Part 10
  reads**: the two censuses keep separate caches deliberately, so that adding this one does not force a 7 GiB rebuild of
  Part 10's (whose keying fills missing rows only). Same cost either way — one full pass — and nothing already paid for is
  lost. **A census written before the escape-velocity columns existed is detected and measured again from scratch** — that
  is one extra pass, once.
* **the halo table** (`P13_HCACHE`) — $M_{200c}$, $R_{200c}$ and the `central` flag of the parent halo of every node, one
  catalogue open per snapshot, cached as a CSV. It is what the escape velocity of H7 is built on, and the reason the part
  needs the catalogues at all.

### The kinematic sidecar

The boxes notebook's histories carry masses only, so the $\kappa_{\rm rot}$(gas) and $R_{50}$ tracks of Parts 3–4 come from a
**cached sidecar** that walks the same main branch (caesar `tree_data/progen_galaxy_star`, most massive progenitor, the
histories' own snapshot ladder) and reads the kinematic columns out of each catalogue — one file open per snapshot, ~100 per
box. `P3_KIN_FILE` points at the file `agn_states_dusty_vs_pure` Part 36 already builds, so an existing build is reused and
only the missing boxes are walked; delete a file to rebuild it, or set `P3_KIN_BUILD = False` to draw only what is cached.
It is matched to the histories **on the snapshot numbers and the galaxy ids**, never on array order, so a partial build leaves
NaNs rather than a misaligned branch. `P3_CATDIR` overrides the config's `catalog_dir` (the share's s50 catalogues moved, and
the fallback ladder in `_catcfg3` is what finds them).

### The AGN states

The four states are evaluated from the catalogue columns along the main branch, with SIMBA's own criteria: a jet needs
$M_{\rm BH} > 10^{7.5}\,M_\odot$ and $f_{\rm Edd} < 0.2$, its velocity ramps as $w_{\rm jet} = \mathrm{clip}(\log_{10}(0.2/f_{\rm Edd}), 0, 1)$,
and the X-ray channel additionally needs a full-velocity jet ($w_{\rm jet} \geq 0.999$, i.e. $f_{\rm Edd} \leq 0.02$) **and**
$f_{\rm gas} < $ `XRAY_FGAS_MAX` (0.2, Davé+19; the environment variable overrides it). $f_{\rm gas} = M_{\rm gas}/M_\star$ is
SIMBA's own gate on the *total* gas — it is a feedback criterion, not an ISM diagnostic, and it appears nowhere else in the
notebook except inside the metal budget of the dust-to-metal ratio.

**A dead black hole scores $w_{\rm jet} = 1$ for free**, since $w_{\rm jet}$ saturates for any $f_{\rm Edd} \leq 0.02$. Part 3
therefore records the X-ray *state* and not a coupling integral; if a coupling integral is ever added here, gate it on
$f_{\rm Edd}$ as Part 2a of the old notebook does.

### Caveats that decide how the results should be quoted

1. **The firing does not demonstrate causation of the truncation.** Part 4 draws all three clocks precisely so that this is
   visible: the channel's gas gate opens at the truncation, and the channel fires when the gate opens. What the channel
   demonstrably does is destroy the dust — the counterfactual box is what shows it.
2. **`cis50nox` has no X-ray channel at all**, so its `t_x1` is a *would-be* firing computed from the same criteria. It is a
   different simulation, not the same galaxies with one process removed, and its quenched population is small (178 chains in
   Part 0's sample) and almost entirely "rotating" (95 %) — which is itself a result, not a bug.
3. **The rebound depends on the floor.** A galaxy that loses all its dust and has some again later has its rebound measured
   from `P9_FLOOR`. The share of such galaxies, and the numbers with them excluded, are printed next to every rebound.
4. **The dust label and the dust ratios share a numerator.** Part 2's AUC is a genuine prediction (the axis is kinematic), but
   any statement of the form "the dusty ones have more dust" in Parts 6–7 is a tautology; the substance is in the panels where
   the *denominators* — the gas, the H$_2$, the metallicity — do not separate.
5. **The observed ages are not like for like** (CIGALE onset vs mass-weighted); Part 5's age band is hatched and labelled for
   that reason, and no scorecard uses it.
6. **`AGE_MIN` is a real selection**, not a neutral tidy-up: it removes the young, recently truncated galaxies. Part 1 says
   what that costs.
7. **The m25 numbers are tens of galaxies.** `P5_NMIN = 3` is deliberately below the notebook's `NMIN`; the counts are printed
   with every point rather than hidden, and a bootstrap band on 4 objects is a bootstrap band on 4 objects. With `ROT_MASS` on
   the selected class is the smaller of the two everywhere, and in m25 it is ~19 of 74 chains.
8. **$\Sigma_{\rm H_2}$ in Part 4b is not a resolved measurement.** caesar carries no H$_2$ half-mass radius, so it is the
   H$_2$ mass spread over the *cold-gas* half-mass radius. It is an ISM density in the same sense $\Sigma_\star$ is a stellar
   one, and the resolved statement lives in Parts 5 (B) and 7 (A), which measure from the particles in fixed apertures.
9. **The ISM/CGM split is caesar's member list**, not a radius. A grain that leaves the member list counts as CGM even if it
   is still at 2 kpc, and hot ambient gas inside the disc counts as CGM too (`P10_CGM_RMIN` raises the inner edge if that is
   not wanted). This is the operational meaning of "not the galaxy", and it is the same list every catalogue $M_{\rm dust}$
   in this notebook is a sum over — which is exactly why Part 10's budget is the right control on Parts 3–9.
10. **The Part 10 clock is binned, not interpolated.** A node exists only where a snapshot of the branch happens to have a
   particle file, so the sampling is set by whichever plan built the cutouts (in m25, a median of ~28 nodes per galaxy over
   the branch). Each galaxy contributes its own median inside each bin before the class median is taken, so a galaxy with
   many snapshots in a bin does not outvote one with a single snapshot — but the bins are *not* equally well sampled, and the
   per-bin galaxy count is drawn and written to the CSV for that reason.
11. **The rest frame of the velocities is the member stars.** Where a galaxy has fewer than 10 of them with velocities the
   member gas is used instead, and where neither exists the kinematic columns are NaN. Files built before
   `build_reduced_particles_job.py` gained its `vel` field (2026-08-28) carry no velocities at all; the cell counts them and
   says so, and re-submitting that plan's sbatch backfills them without a geometry redo.
12. **Part 11's sample is the intersection** of Part 0's m25 rows with the K-S products' own (differently selected) anchor
   set, so it is smaller than Parts 9–10's — 28 galaxies against 44. It is *not* a subsample chosen on anything measured here.
13. **The disc plane of Part 12 is the gas's own.** The profiles are measured face-on in the frame of the member gas's
   angular momentum, which is the right plane for a rotating disc and an arbitrary one for a gas distribution that is not
   rotating — for the second, a face-on profile is simply a projection of something roughly spherical, and the hole metric
   still means what it says. The fallbacks (the stellar angular momentum, then the stellar minor axis of the file's own
   `evecs`) are counted and printed, as is the misalignment between the gas plane and the stars: it is large, which is the
   premise of this notebook and not a failure of the frame.
14. **The hole metric is an aperture against an annulus.** $\Sigma(<R_{\rm in})$ is a cumulative mean and $\Sigma_{\rm peak}$
   a single ring, so a smoothly declining profile scores *negative* rather than zero — the threshold at $+0.3$ dex sits well
   clear of that regime. The centre is floored at one particle so an empty one is finite, and a galaxy is classified only
   where `n_exp` (the count the peak would have put inside the aperture) reaches `P12_NEXP`; the galaxies that fail it are
   counted, not silently dropped, and both counts are printed for the two groups.
15. **Part 12's stages are rebuilt from the histories**, not read from the m25 notebook's Part 2d windows table. `agn_ign`
   uses the same rule (the first crossing of half the pre-quenching BHAR peak) but on the boxes notebook's snapshot ladder,
   so it can differ from the `agn_ign` of Parts 5 and 7 by a snapshot; `xray_on` is this notebook's own `agn_states` and has
   no counterpart there. The advantage is that all four stages, the clocks and the classes come off one chain.
16. **In Part 13 the gate crossing at $t = 0$ is a definition, not a result.** $t_{x1}$ is the first step at which SIMBA's
   X-ray criterion is met, and one of its conditions is $f_{\rm gas} <$ `XRAY_FGAS_MAX`; the panel therefore *must* show
   both groups below the gate at the origin. The falsifiable statement is whether a group climbs back over it afterwards.
   The same warning applies to the AGN-state occupancy panel: being in the X-ray state at $t = 0^+$ is what the clock means.
17. **Part 13 splits on the outcome.** The groups are the dust label at the anchor, not a selection, so its figures are not
   predictions and must never be quoted as an AUC of the kind Part 2 reports. What they are is an anatomy of what happened
   to galaxies that ended dusty against galaxies that ended dust-poor — and the mass is the obvious confound in every one of
   them (a heavier halo has a hotter CGM whatever the AGN did), which is why the scorecard prints the AUC pooled within
   $\log M_\star$ terciles beside the raw one, and why the bracketed number is the one to quote.
18. **The escape velocity is the halo's, and for a satellite it is the wrong halo's.** `parent_halo_index` points at the
   group halo, whose centre is not the galaxy's, so $v_{\rm esc}(r)$ evaluated at a galaxy-centric radius is then too high
   and taken about the wrong origin. The census records `central` per node so the test can be repeated on centrals alone.
   Two further approximations are deliberate: the NFW concentration comes from a *mean* relation (Dutton & Macciò 2014) and
   not from the halo's own profile, and adding the resolved baryons double counts the halo's own baryon share at large
   radius — worth it inside ~10 kpc, where the analytic NFW knows nothing about the condensed gas and stars. Set
   `P13_VESC_BAR = False` to see how much of the answer rests on that.
19. **The apocentre is a radial-orbit upper bound.** It solves $-\Phi(r_{\rm apo}) = -\Phi(r) - v^2/2$ using the *full*
   speed, i.e. as if the orbit had no angular momentum; a real orbit with the same energy turns around sooner. It is
   therefore an outer envelope of where material goes, comparable between galaxies but not a prediction of any one orbit,
   and it is censored at $5\,R_{200c}$, which is also how "it never comes back" is written into the column.
20. **The shell flux counts every particle in the shell**, member or not, and the "cold inflow" is the inward part of it
   below `P13_TCOLD`. It is a flux across a fixed sphere at one instant, not an accretion rate onto the ISM: a particle can
   cross the shell and turn around. The cumulative version integrates instantaneous fluxes sampled at the snapshots that
   have a cutout (~0.15 Gyr apart), so it is indicative, and only the galaxies sampled from `P13_CUMSTART` onwards enter it.
21. **$\Sigma_\star$ shares its numerator with the rule** once `ROT_MASS` is on, so its separation at and after the truncation
   is half construction. Part 4b prints $\log M_\star$, $R_{50}(\star)$ and $R_{50}$(gas) as controls under the table for
   exactly this reason, and the statement that survives is the one made *before* the truncation, where the same galaxies are
   not yet apart.

22. **Part 14 splits on the CAUSE, and that is the whole point of it.** Every other part of this notebook groups galaxies
   by what they ended up as; Part 14 groups them by how long they lived in the jet mode *before* the truncation and then
   asks what they became. Three things follow. (i) The jet leg is not a clean instrument: it needs
   $M_{\rm BH} > 10^{7.5}$, so it is partly "how long ago the hole got big", and it grows with $M_\star$ and with the
   epoch — which is why the groups are split inside quartiles of $\log M_\star$ and every score is repeated partial at
   fixed $\log M_\star$ and $\log M_{\rm BH}$. (ii) `rot` already carries a mass cut when `ROT_MASS` is on, so the
   *rotator* rung of the ladder shares an axis with the control — read it beside the compactness rung, which does not.
   (iii) The claim is an ORDER, not a correlation: FIG 2 is where it can die, because the compactification has to appear
   left of zero, *during* the jet leg. If the two groups separate only after the truncation, the jet leg is a label on the
   outcome and not its cause. The channel-off box decides the last link: there the jet leg must still build the disc and
   must buy no dust at all.

23. **Part 16 asks the question upstream, and its axis is a RANK.** Part 15's division of the ISM at the jet activation
   is worth nothing in m25; what does divide the outcome is the rotation at the anchor, so Part 16 asks what makes *that*.
   Its axis is $t_{\rm jet}$ — but a cosmic time is not a property of a galaxy: each branch is watched up to a different
   anchor, and "fired early" and "was looked at early" are the same number unless the epoch is held. `_crank16` therefore
   ranks $t_{\rm jet}$ inside cells of the anchor time and the stellar mass, and the raw and the ranked correlation are
   printed side by side so the difference is visible. Two more things to hold on to. (i) The CGM half of the part
   **re-clocks Part 13's nodes on its own $t_{\rm SFT}$ and re-splits them on its own classes**, so `P13_CLOCK` and
   `P13_SPLIT` can be anything without changing what (c) and (d) mean — but the census itself is Part 13's, so that part
   has to have been run for the m25 box first. (ii) $t_{\rm div}$ in (d) is the first epoch where the difference holds
   for `P16_DIVSIG` bins running, **whatever its sign**: the value of the difference *at* $t_{\rm div}$ is printed and
   annotated beside it, because an early, small, transient excursion is not the parting C4 is about.

24. **Part 17 computes a rate, not a measurement, and the difference matters.** tau_grow and tau_sp are LAWS evaluated at
   each particle's own density, temperature and metallicity — they say what the grain *would* do in the gas it is in, not
   what SIMBA's solver did to it that step. Three things follow. (i) For STAR-FORMING gas the temperature is the effective
   equation of state, not a temperature, so its tau_sp is meaningless; the star-forming and the non-star-forming ISM are
   never merged. (ii) Above `P17_TGROW` the sticking coefficient collapses and the growth law is formal: every growth
   column is the COLD gas of its zone, and the hot-CGM growth panel is empty on purpose. (iii) Both laws are LINEAR in
   the grain radius, so `P17_AGR` slides the whole distribution and changes no comparison — which is why the part is
   written around ratios and shares (the dust that can regrow within `P17_FAST`, the time a zone would need to make its
   own dust) rather than around any single number.

25. **A decay timescale is not a measurement until three things are checked, and Part 18 checks them.** (i) *The
   parameterisation.* Part 15's double exponential has four free parameters on a ~10-point, ~1.2 Gyr window: two fits in
   five sit on one of its bounds and three quarters give the slow term under 1 % of the amplitude, so **its tau1 / tau2
   should not be quoted** — Part 18's broken exponential in log10, profiled over the break, is the identifiable version of
   the same claim. (ii) *The sampling.* One e-folding of the dust is 0.26 Gyr against a snapshot spacing of 0.07 Gyr at
   $z=3$ and 0.18 Gyr at $z=0$: the slope is still fitted from the points that cross the fall, but a mode piling up below
   one step would be an upper limit. (iii) *The anchor.* Ten anchors from $z=3$ to $z=0$ move the median $\log \tau_0$ by
   0.3 dex on their own, so a distribution pooled over them is already a mixture — every test in Part 18 is run pooled,
   centred on each anchor, per anchor and on the well-measured branches alone, and none of them finds a second mode.
   (iv) *The window.* Part 18 fits **every snapshot after the truncation** with the floored points taken at face value
   (`P18_WIN = "all"`, `P18_CENSOR = False`): `P9_FLOOR` is SIMBA saying there is no dust left, so a branch that reaches
   zero comes out with a flat second leg and **no tau_2 at all**, which is counted (the `rebound` class) rather than
   fitted away. `P18_ALT` fits the narrow treatment beside it: tau_1 is unchanged, tau_2 doubles.
26. **The dust's own timescale against the dust it ends with is circular** — a slower decay from the same starting point
   ends higher by construction. Part 18 draws it because it is the natural question, marks it `(*)` everywhere, and puts
   the non-circular versions beside it: the same galaxies' sSFR and H2 decay times against the same dust label.
