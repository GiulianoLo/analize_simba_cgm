# Paper figures under the X-ray classification — the m25 quenched sample and the boxes, one ordered notebook

**Goal.** One notebook of **pure reads** that rebuilds the paper figures of `paper_figures_quenched_m25.ipynb` (A) and of
`paper_ism_prediction_boxes.ipynb` (B) under a single, official division of the quenched galaxies by SIMBA's X-ray channel. Nothing is
measured here — no snapshot, no particle file, no `simbanator` import — only the cached products of the two analysis notebooks
(`output/cis25/`, `output/box_resolution/ism_prediction/`), `ks_tracks_lib.py` and `obs_data/almac11/`.

**The channel.** B Part 9 integrates SIMBA's gated X-ray channel after quenching from the catalogue histories:
$E_x = \int_{\rm SFT}^{\rm anchor} w_{\rm jet}\,[f_{\rm gas} < 0.2]\,{\rm d}t$, $T_{\rm gas\text{-}poor} = \int_{\rm SFT}^{\rm anchor} [f_{\rm gas} < 0.2]\,{\rm d}t$
and their ratio $r_x = E_x / T_{\rm gas\text{-}poor}$ = the mean jet weight over the time the galaxy spent gas-poor (the only time the X-ray channel
can fire); B Part 6a records $t_{\rm onset}$ = the first snapshot of the history where the gated coupling $x_{\rm coup} = w_{\rm jet}\,[f_{\rm gas} < 0.2] \geq 0.5$.

**The classification** (`XSCHEME = "onset"`): the **timing of the first X-ray episode** against the quench clock of the same history (B Part 9 panels d–f),
for every galaxy with a quench event (SFT and QT found):

* **`early`** — first X-ray episode **by QT**: the channel fired before star formation stopped (B's `lead`) or during the quenching (B's `concurrent`) — the
  reference class (`XREF`), the bulk of the quenched population and the driest;
* **`late`** — first episode after QT: quenched before the channel fired (short time since SFT, all fast quenchers);
* **`never`** — no episode by the anchor: never exposed (in the boxes almost all satellites without a jet-mode BH);
* galaxies without a quench event stay `no_event` (unclassified).

`XSCHEME = "onset_sft"` splits the same episode at **SFT** instead of QT — `before` (episode before SFT: gas-poor with a jet-mode BH while still
star-forming, B's `lead`) / `after` (in [SFT, QT] or after QT; the reference) / `never`; `XSCHEME = "onset4"` splits `early` back into `lead` / `concurrent`
(the four groups of B Part 9); the previous division — the jet weight while gas-poor cut at
`R_X_CUT` (`xlow` / `xhigh`) — is `XSCHEME = "rx"`. The four-way onset label and the rx label are always written to the class table (`onset_class`, `rx_class`). Every section reads `XCLASSES` / `XREF` / `XCOL` / `XNAME` / `XSHORT` / `XABBR` of Part 0, so the switch changes every figure; the two-sample tests
contrast each class with `XREF`. The baseline every scheme is compared with is the quenching-timescale dichotomy: **fast** quenchers have
$\log_{10}(\tau_q / t_{\rm QT}) < -1.25$ ($\tau_q = t_{\rm QT} - t_{\rm SFT}$, $t_{\rm QT}$ the cosmic time at QT), **slow** the rest.

* **Part 0** — the official class table for every box (`xray_classes.csv`), the counts and the per-class dust medians, the crosstabs with fast / slow, with the old
  weak / strong AGN-coupling classes and with the other scheme.
* **Section 1 — the feedback tracks** (A Parts 1b + 1c merged): the ladder of the critical points and the stacked $w_{\rm jet}$ (ungated), $x_{\rm coup}$ (gated) and
  $f_{\rm gas}$ tracks on the $t - t_{\rm SFT}$ clock, fast / slow beside the X-ray classes, with one Mann–Whitney strip per class against the reference class.
* **Section 2 — the $A_V$ test on the KS plane** (A Part 3): the $\log M_\star > 10.25$ quenched galaxies binned by their anchor $A_V$, followed SFT → QT → track end,
  one column per class (a class too small for the $A_V$ terciles is drawn unsplit).
* **Section 3 — the radial profiles** (A Part 4d): $A_V$, dust / gas / H$_2$ per stellar mass, age and $\kappa_{\rm rot}$ across the four annuli, per class + SF controls.
* **Section 4 — the critical-point sequences** (A Part 5 figure B): $\kappa_{\rm rot}$, sSFR / age / elapsed time, dust / H$_2$ fractions of the core and of the outskirt
  at ignition → jet on → SFT → QT → end, the classes side by side (`kinematics_sequence`, `stellar_sequence`, `ism_sequence`, $\log M_\star > 10.25$).
* **Section 5 — dust-to-gas against the exposure, the time since SFT and the gas-poor time** (B Part 9, rows 1–3): $\log M_{\rm dust}/M_{\rm gas}$ per box, the classes on
  every row, read from the exposure table; plus **the exposure as a diagnostic inside each class** — the rank correlation of $E_x$, $f_x$, $r_x$, $T_{\rm gas\text{-}poor}$,
  $t_{\rm anchor} - t_{\rm SFT}$ and $t_{\rm onset} - t_{\rm SFT}$ with the dust content, per box and class (`paper_xray_exposure_diagnostic.csv`).; plus **the dusty tail per class** — the unit-area distributions of $M_{\rm dust}/M_\star$ and $M_{\rm dust}/M_{\rm gas}$
  per class with the class shares of the whole box against the shares above a threshold ($-4$ / $-2.5$; hypergeometric P), `paper_xray_dust_distributions` + `_dust_tail_composition.csv`.
* **Section 5b — the ISM grid** (B Part 2 style): $M_{\rm dust}/M_\star$ and $M_{\rm gas}/M_\star$ against age, $\Sigma_{\rm e}$ and sSFR, one track per class (+ the
  small classes as points) with the observed galaxies coloured by $A_V$, per box and across the boxes.
* **Section 6 — the fiducial KS plane**: the same galaxies as Section 2 without the $A_V$ split — all classes / one column per class with the individual tracks, the class median
  track and the ALMA-C11 points at full strength, plus the below / on / above Bigiel+08 counts of the track ends against the observed ones.

**Inputs** (all cached; the notebook that writes each one in brackets): `output/cis25/tables/powderday_quenched_selection.fits` [m25 Part 3],
`agn_classifier_{windows,tracks}.fits` [m25 Part 2d], `annulus_av_allincl.fits`, `annulus_ism_truth.fits`, `aperture_truth.fits`, `annulus_kinematics.fits`,
`attenuation_vs_ism.fits` [m25 Parts 7–8]; `output/cis25/ks_tracks_pt/{ks_tracks,ks_galaxies}.fits` [KS Parts 4a / 5b], `output/cis25/ks_tracks/ks_stage_kinematics.fits`
[KS Part 3b]; `output/box_resolution/ism_prediction/{ism_prediction_catalogue.fits, ism_prediction_agn_classes.csv, ism_prediction_xray_exposure.csv}` [B Parts 1 / 6a / 9];
`obs_data/almac11/{ks_table,almac11_gas_dust}.csv`.

**Outputs** → `output/cis25/plots/paper_xray/`: `xray_classes.csv`; `paper_xray_feedback_tracks`; `paper_xray_ks_av_bins_R50_H2` + `paper_xray_ks_av_bins.csv`;
`paper_xray_radial_profiles_{mgt10p25,mlt10p25}` + `paper_xray_radial_profiles.csv`; `paper_xray_{kinematics,stellar,ism}_sequence_mgt10p25` + `paper_xray_kappa_sequence.csv`;
`paper_xray_dg_vs_exposure` + `.csv` + `_classes.csv` + `paper_xray_exposure_diagnostic.csv`; `paper_xray_ism_grid_{cis100,cis50,cis25,boxes}` + `.csv` + `_points.csv`;
`paper_xray_ks_fiducial_R50_H2` + `paper_xray_ks_fiducial.csv` (PNG + PDF).

Palette (B Part 9's onset colours): `early` red (`onset4`: `lead` light orange, `concurrent` red), `late` blue, `never` black; `onset_sft`: `before` orange, `after` purple; the `rx` scheme keeps `xlow` burnt orange / `xhigh` steel blue;
fast purple, slow ochre; SF controls grey — none of them the old weak / strong colours.

```python
# ── Part 0 — configuration + the official two-class X-ray classification (pure reads of the cached products of the analysis notebooks) ──
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Rectangle
from matplotlib.patheffects import withStroke
from astropy.table import Table
from scipy.stats import mannwhitneyu, ks_2samp, spearmanr
import ks_tracks_lib as kl                    # repo-root module: RELATIONS / relation_y / ecdf (the KS plane)

SIM_NAME = "cis25"
OUT      = os.path.join(os.getcwd(), "output", SIM_NAME)
TABLEDIR = os.path.join(OUT, "tables")
BOXDIR   = os.path.join(os.getcwd(), "output", "box_resolution", "ism_prediction")   # paper_ism_prediction_boxes.ipynb products
OBS_CSV  = os.path.join(os.getcwd(), "obs_data", "almac11", "ks_table.csv")
OBS_GD_CSV = os.path.join(os.getcwd(), "obs_data", "almac11", "almac11_gas_dust.csv")   # ALMA-C11 dust / CO detections + fiducial CIGALE age, M*, M_dust (Section 4 band)

# ── inputs: the m25 particle sample (A's tables), the KS products of the pre-SFT-threshold run, the boxes' catalogue products ──
SELECTION_FITS  = os.path.join(TABLEDIR, "powderday_quenched_selection.fits")   # m25 Part 3: pop, log_mstar, t_sft / t_qt [yr], tau_q_over_tH (= tau_q / t_QT), agn_class_pt, w_pre
P2D_FITS        = os.path.join(TABLEDIR, "agn_classifier_windows.fits")         # m25 Part 2d: one row per event galaxy (critical times on the t - t_SFT clock)
P2D_TRACKS_FITS = os.path.join(TABLEDIR, "agn_classifier_tracks.fits")          # m25 Part 2d: the tracks behind the stacks (galaxy x history snapshot)
ANNULUS_AV_FITS = os.path.join(TABLEDIR, "annulus_av_allincl.fits")             # m25 Part 8b: A_V per (aperture / annulus, sightline)
ATTEN_FITS      = os.path.join(TABLEDIR, "attenuation_vs_ism.fits")             # m25 Part 7a: A_V per aperture, fiducial sightline (fallback)
ANNULUS_ISM_FITS    = os.path.join(TABLEDIR, "annulus_ism_truth.fits")          # m25 Part 8a: dust / gas / H2 per (aperture / annulus, sightline)
APERTURE_TRUTH_FITS = os.path.join(TABLEDIR, "aperture_truth.fits")             # m25 Part 7e: M*, mass-weighted stellar age per (aperture, sightline)
ANNULUS_KIN_FITS    = os.path.join(TABLEDIR, "annulus_kinematics.fits")         # m25 Part 8j0: kappa_rot of gas / H2 / stars per spherical rung / shell
KS_RULE_TAG = "pt"                                                              # the KS notebook run whose products are read (KS_AGN_CLASSIFIER = pre_threshold)
KSDIR       = os.path.join(OUT, "ks_tracks")                                    # KS caches (rule-free)
KSOUT       = KSDIR + f"_{KS_RULE_TAG}"                                         # KS derived products of that run
TRACKS_FITS = os.path.join(KSOUT, "ks_tracks.fits")                             # KS Part 4a: epochs x apertures on the KS plane
GALS_FITS   = os.path.join(KSOUT, "ks_galaxies.fits")                           # KS Part 5b: one row per Q galaxy (region, clock, stage snapshots)
KIN_STAGE_FITS = os.path.join(KSDIR, "ks_stage_kinematics.fits")               # KS Part 3b: kappa_rot of gas / H2 / stars + stellar age per spherical zone at every epoch
CACHE_FITS  = os.path.join(BOXDIR, "ism_prediction_catalogue.fits")             # B Part 1: every galaxy with log M* > 9.5 of every (box, anchor)
CLASS_CSV   = os.path.join(BOXDIR, "ism_prediction_agn_classes.csv")            # B Part 6a: t_sft / t_qt / tau_q [Gyr], w_pre, agn_onset, the old classes
EXP_CSV     = os.path.join(BOXDIR, "ism_prediction_xray_exposure.csv")          # B Part 9: E_x, T_gaspoor, r_x, ... per galaxy with a quench event
PAPERDIR    = os.path.join(OUT, "plots", "paper_xray")
os.makedirs(PAPERDIR, exist_ok=True)

# ── the boxes (B Part 0): labels / colours; the m25 Part 1 selection evaluated on the catalogue columns (Section 5) ──
BOXES = {"cis100": dict(label="m100n1024 (100 Mpc/h)", color="#D55E00"), "cis50": dict(label="m50n512 (50 Mpc/h)", color="#009E73"),
         "cis25": dict(label="m25n512 (25 Mpc/h)", color="#0072B2"),
         "cis50nox": dict(label="m50n512 no X-ray feedback (s50nox)", color="#CC79A7"), "cis50noagn": dict(label="m50n512 no AGN feedback (s50noagn)", color="#E69F00")}
BOX_ORDER = ["cis100", "cis50", "cis25"]
BOX_SHORT = {"cis100": "m100", "cis50": "m50", "cis25": "m25", "cis50nox": "s50nox", "cis50noagn": "s50noagn"}
MASS_FLOOR, PASSIVE_FACTOR, NGAS_MIN, NSTAR_MIN, DUST_TO_H2_MIN = 10.0, 0.2, 21, 20, 1e-4   # the m25 cuts (log M* > 10, sSFR < 0.2 / t_H, >= 21 gas, >= 20 stars, M_dust >= 1e-4 M_H2)
MASS_MIN       = 10.25    # the mass cut of every per-class figure (mgt10p25)
CENTRALS_ONLY  = False
HE             = 1.36     # SIMBA H2 is hydrogen-only
SFR_COL, MSTAR_COL = "sfr", "mstar"
R_PROJ_OVER_3D = 0.75
SSFR_FLOOR     = 1e-13
JET_LOGMBH, JET_FEDD, XRAY_FGAS_MAX = 7.5, 0.2, 0.2   # SIMBA jet mode: log M_BH > 7.5 and f_Edd < 0.2; the X-ray gate f_gas < 0.2
KEY = ["box", "snap", "gal_id"]

# ── the classification: XSCHEME picks the official division of the quenched galaxies (every galaxy with a quench event = SFT and QT found) ──
XSCHEME  = "onset"        # "onset": the timing of the first X-ray episode (B Part 6a: t_onset = the first snapshot with w_jet [f_gas < 0.2] >= 0.5) -> by QT (before SFT or in [SFT, QT]) / after QT / never
                          # "onset_sft": the same episode split at SFT instead of QT -> before SFT / after SFT / never; "onset4": the by-QT class split into before SFT / in [SFT, QT];
                          # "rx": the jet weight while gas-poor r_x = E_x / T_gaspoor cut at R_X_CUT
                          # (the previous official division); the raw four-way onset label and the rx label are always kept as the columns onset_class / rx_class
R_X_CUT  = 0.97           # rx scheme: xlow below this (or undefined = never gas-poor since SFT), xhigh at or above; 0.97 = B Part 9's cis25 tercile edge (r_x piles at 1), 0.5 = the earlier fixed cut
FAST_CUT = -1.25          # fast quencher: log10(tau_q / t_QT) below this (the baseline dichotomy every scheme is compared with)
XSCHEMES = {
    "onset": dict(title="timing of the first X-ray episode", classes=["early", "late", "never"], ref="early",
                  onset_map={"lead": "early", "concurrent": "early", "late": "late", "never": "never"},        # B 6a agn_onset -> class
                  col={"early": "#d7191c", "late": "#2c7bb6", "never": "0.05", "no_event": "0.85"},
                  name={"early": "first X-ray episode by QT (before SFT or in [SFT, QT])", "late": "first X-ray episode after QT",
                        "never": "never exposed (no X-ray episode by the anchor)", "no_event": "no quench event"},
                  short={"early": "X-ray by QT", "late": "X-ray after QT", "never": "never exposed", "no_event": "no event"},
                  abbr={"early": "early", "late": "late", "never": "never"}),
    "onset_sft": dict(title="timing of the first X-ray episode against SFT", classes=["before", "after", "never"], ref="after",
                  onset_map={"lead": "before", "concurrent": "after", "late": "after", "never": "never"},   # the same episode, the clock split at SFT instead of QT
                  col={"before": "#e66101", "after": "#5e3c99", "never": "0.05", "no_event": "0.85"},
                  name={"before": "first X-ray episode before SFT (gas-poor with a jet-mode BH while still star-forming)",
                        "after": "first X-ray episode after SFT (in [SFT, QT] or after QT)",
                        "never": "never exposed (no X-ray episode by the anchor)", "no_event": "no quench event"},
                  short={"before": "X-ray before SFT", "after": "X-ray after SFT", "never": "never exposed", "no_event": "no event"},
                  abbr={"before": "before", "after": "after", "never": "never"}),
    "onset4": dict(title="timing of the first X-ray episode", classes=["lead", "concurrent", "late", "never"], ref="concurrent",
                  onset_map={"lead": "lead", "concurrent": "concurrent", "late": "late", "never": "never"},
                  col={"lead": "#fdae61", "concurrent": "#d7191c", "late": "#2c7bb6", "never": "0.05", "no_event": "0.85"},
                  name={"lead": "first X-ray episode before SFT", "concurrent": "first X-ray episode in [SFT, QT]", "late": "first X-ray episode after QT",
                        "never": "never exposed (no X-ray episode by the anchor)", "no_event": "no quench event"},
                  short={"lead": "X-ray before SFT", "concurrent": "X-ray in [SFT, QT]", "late": "X-ray after QT", "never": "never exposed", "no_event": "no event"},
                  abbr={"lead": "lead", "concurrent": "conc.", "late": "late", "never": "never"}),
    "rx":    dict(title="jet weight while gas-poor", classes=["xlow", "xhigh"], ref="xhigh",
                  col={"xlow": "#d95f0e", "xhigh": "#31688e", "no_event": "0.85"},
                  name={"xlow": r"jet off while gas-poor ($r_x < %g$)" % R_X_CUT, "xhigh": r"jet on while gas-poor ($r_x \geq %g$)" % R_X_CUT, "no_event": "no quench event"},
                  short={"xlow": r"low $r_x$", "xhigh": r"high $r_x$", "no_event": "no event"}, abbr={"xlow": "low", "xhigh": "high"}),
}
_XS = XSCHEMES[XSCHEME]
XTITLE, XCLASSES, XREF = _XS["title"], list(_XS["classes"]), _XS["ref"]   # XREF: the class every other class is contrasted with (Mann-Whitney) in the sections
XCOL, XNAME, XSHORT, XABBR = _XS["col"], _XS["name"], _XS["short"], _XS["abbr"]   # colour, long name, short name (legends / columns), abbreviation (annotations)
XOTHERS = [c for c in XCLASSES if c != XREF]
FS_CLASSES = ["fast", "slow"]
FS_COL  = {"fast": "#7b3294", "slow": "#a6761d"}
FS_NAME = {"fast": r"fast ($\log \tau_q/t_{\rm QT} < %g$)" % FAST_CUT, "slow": r"slow ($\log \tau_q/t_{\rm QT} \geq %g$)" % FAST_CUT}
C_SF = "0.5"              # the SF-control colour of every figure
P9_FLOOR = -6.5           # log M_dust/M* and log M_dust/M_gas of the zero-dust galaxies (drawn / binned at the floor, never dropped)

# ── style ──
PAPER_DPI = 300
INLINE_DPI = 110          # the notebook's inline rendering (the saved files use PAPER_DPI); the figure sizes below are generous on purpose — Jupyter scales wide figures to the cell
plt.rcParams.update({"font.size": 11, "axes.titlesize": 11, "axes.labelsize": 11, "legend.fontsize": 9, "xtick.labelsize": 9.5,
                     "ytick.labelsize": 9.5, "axes.titleweight": "bold", "pdf.fonttype": 42, "figure.dpi": INLINE_DPI})


def _s(col):
    """FITS string column -> stripped str array."""
    return np.char.strip(np.asarray(col).astype(str))


def need(path, made_by):
    if not os.path.exists(path):
        raise FileNotFoundError(f"{path} is missing: run {made_by} first")
    return path


def wrap_caption(text, width=150):
    """Greedy line wrap at `width` characters that never splits inside a $...$ mathtext span (existing newlines kept): long one-line captions and
    suptitles widen the tight bounding box that the inline display and paper_save use, and the panels shrink with it."""
    import re
    out = []
    for para in text.split("\n"):
        words, cur = [], ""
        for t in re.findall(r"\$[^$]*\$|[^\s$]+|\s+", para):
            if t.isspace():
                if cur:
                    words.append(cur); cur = ""
            else:
                cur += t
        if cur:
            words.append(cur)
        line = ""
        for w in words:
            if line and len(line) + 1 + len(w) > width:
                out.append(line); line = w
            else:
                line = (line + " " + w) if line else w
        out.append(line)
    return "\n".join(out)


def paper_save(fig, stem):
    for ext in ("png", "pdf"):
        fig.savefig(os.path.join(PAPERDIR, f"paper_xray_{stem}.{ext}"), dpi=PAPER_DPI, bbox_inches="tight")
    print("figure ->", os.path.join(PAPERDIR, f"paper_xray_{stem}.{{png,pdf}}"))


# ── the m25 selection table: pop, mass, the old class, tau_q / t_QT ──
SEL = Table.read(need(SELECTION_FITS, "powderday_flux_quenched_m25.ipynb Parts 2-3"))
_sk = [(int(a), int(b)) for a, b in zip(SEL["snap"], SEL["gal_id"])]
POP_OF = dict(zip(_sk, _s(SEL["pop"])))
LM_OF  = dict(zip(_sk, np.asarray(SEL["log_mstar"], float)))
OLD_OF = dict(zip(_sk, _s(SEL["agn_class_pt"])))                       # the previous classification (pre-SFT threshold on w_pre; weak / intermediate / strong / no_event)
TQH_OF = dict(zip(_sk, np.asarray(SEL["tau_q_over_tH"], float)))       # tau_q / t_QT of the m25 pipeline (checked against the catalogue histories below)

# ── the boxes' products: the exposure table (every galaxy of every classified anchor), the classes table (SFT / QT / tau_q / w_pre / old class), the catalogue cache ──
EXP = pd.read_csv(need(EXP_CSV, "paper_ism_prediction_boxes.ipynb Part 9"))
CLS = pd.read_csv(need(CLASS_CSV, "paper_ism_prediction_boxes.ipynb Part 6a"))
CAT = Table.read(need(CACHE_FITS, "paper_ism_prediction_boxes.ipynb Part 1")).to_pandas()
CAT["box"] = [v.decode() if isinstance(v, bytes) else str(v) for v in CAT["box"]]
CAT["box"] = CAT["box"].str.strip()
for _c in ("snap", "gal_id"):
    CAT[_c] = CAT[_c].astype(int)
with np.errstate(divide="ignore", invalid="ignore"):
    CAT["log_mstar"] = np.log10(np.where(CAT["mstar"] > 0, CAT["mstar"], np.nan))
XT = EXP.merge(CLS[KEY + ["t_sft", "t_qt", "tau_q", "w_pre", "t_onset", "agn_onset", "agn_class3", "agn_class"]], on=KEY, how="left") \
        .merge(CAT[KEY + ["log_mstar", "mstar", "mdust", "mgas", "central"]], on=KEY, how="left")
_ev = XT["t_sft"].notna() & XT["t_qt"].notna()
_rx = XT["r_x"].to_numpy(float)
XT["event"] = _ev
# the two divisions: rx_class (the fixed cut on r_x) and onset_class (B Part 6a's agn_onset: the first snapshot with the gated coupling >= 0.5 against SFT / QT;
# "no_AGN" = a history without a BH at all = never exposed); xray_class = the official one (XSCHEME); galaxies without a quench event are no_event in both
XT["rx_class"]    = np.where(~_ev, "no_event", np.where((_rx < R_X_CUT) | ~np.isfinite(_rx), "xlow", "xhigh"))
_onset = XT["agn_onset"].fillna("no_event").astype(str).str.strip().replace({"no_AGN": "never"})
XT["onset_class"] = np.where(~_ev, "no_event", _onset)
_bad = sorted(set(XT.loc[_ev, "onset_class"]) - set(XSCHEMES["onset4"]["classes"]))
assert not _bad, f"event galaxies with an agn_onset outside the onset classes: {_bad} (B Part 6a changed its labels?)"
XT["xray_class"] = XT["rx_class"] if XSCHEME == "rx" else np.where(_ev, XT["onset_class"].map(_XS["onset_map"]).fillna("no_event"), "no_event")
with np.errstate(divide="ignore", invalid="ignore"):
    XT["tau_q_over_tqt"] = XT["tau_q"] / XT["t_qt"]
    _lnt = np.log10(XT["tau_q_over_tqt"].to_numpy(float))
    _fd, _md, _mg = XT["mdust"].to_numpy(float) / XT["mstar"].to_numpy(float), XT["mdust"].to_numpy(float), XT["mgas"].to_numpy(float)
    XT["lfd"] = np.where(_fd > 0, np.log10(np.where(_fd > 0, _fd, 1.0)), P9_FLOOR)
    XT["ldg"] = np.where((_md > 0) & (_mg > 0), np.log10(np.where((_md > 0) & (_mg > 0), _md / np.where(_mg > 0, _mg, 1.0), 1.0)), np.where(_mg > 0, P9_FLOOR, np.nan))
XT["dt_onset_sft"] = XT["t_onset"] - XT["t_sft"]                          # the first X-ray episode on the t - t_SFT clock [Gyr] (NaN: never)
XT["dt_onset_qt"]  = XT["t_onset"] - XT["t_qt"]
XT["fast_slow"] = np.where(~_ev, "no_event", np.where(_lnt < FAST_CUT, "fast", "slow"))
XT["rx_undefined"] = _ev & ~np.isfinite(_rx)                            # a quench event but never gas-poor since SFT (T_gaspoor = 0, E_x = 0)
# the m25 particle sample seen through the same table: every Q galaxy of the selection has a cis25 row; its tau_q / t_QT and old class must agree
_m25 = XT[XT["box"] == "cis25"].copy()
_m25["key"] = list(zip(_m25["snap"].astype(int), _m25["gal_id"].astype(int)))
_qkeys = [k for k in _sk if POP_OF[k] == "Q"]
_missing = [k for k in _qkeys if k not in set(_m25["key"])]
assert not _missing, f"{len(_missing)} m25 Q galaxies have no row in {EXP_CSV}: re-run paper_ism_prediction_boxes Parts 6a + 9 with cis25 in H_BOXES"
_m25 = _m25.set_index("key").loc[_qkeys]
_dt = np.abs(np.log10(_m25["tau_q_over_tqt"].to_numpy(float)) - np.log10(np.array([TQH_OF[k] for k in _qkeys])))
_old = np.array([OLD_OF[k] for k in _qkeys])
print(f"m25 sample: {len(_qkeys)} Q galaxies, all in the exposure table; log(tau_q / t_QT) catalogue vs m25 pipeline: max |diff| {np.nanmax(_dt):.1e} dex "
      f"({int(np.isfinite(_dt).sum())} event galaxies); old 3-class label agreement {100 * np.mean(_m25['agn_class3'].to_numpy().astype(str) == _old):.0f} %")
assert np.nanmax(_dt) < 1e-3, "tau_q / t_QT of the catalogue histories disagrees with the m25 selection table"
XT["agn_class_m25"] = [OLD_OF.get(k, "") if b == "cis25" else "" for b, k in zip(XT["box"], zip(XT["snap"].astype(int), XT["gal_id"].astype(int)))]

# ── the official table ──
XT_COLS = KEY + ["log_mstar", "event", "r_x", "e_x", "t_gaspoor", "f_x", "dt_sft", "t_sft", "t_qt", "tau_q", "tau_q_over_tqt", "w_pre", "t_onset", "dt_onset_sft", "dt_onset_qt", "agn_onset",
                 "fast_slow", "xray_class", "onset_class", "rx_class", "rx_undefined", "agn_class", "agn_class3", "agn_class_m25", "lfd", "ldg", "central"]
XT = XT[XT_COLS].sort_values(KEY).reset_index(drop=True)
XT.to_csv(os.path.join(PAPERDIR, "xray_classes.csv"), index=False)
print(f"classes -> {os.path.join(PAPERDIR, 'xray_classes.csv')} ({len(XT)} rows: every galaxy of every classified anchor of {sorted(XT['box'].unique())})")
print(f"\nscheme {XSCHEME!r} = {XTITLE}: classes {XCLASSES} (every other class is contrasted with {XREF!r}); no quench event -> no_event; "
      f"rx_class (r_x < {R_X_CUT:g} or undefined -> xlow, else xhigh) and the four-way onset_class are both in the table; fast = log10(tau_q / t_QT) < {FAST_CUT:g}")
_OTHERS = ["onset_class"] if XSCHEME == "rx" else ["rx_class", "onset_class"]
print(f"\n{'box':8s} {'sample':14s} {'N':>5s} " + " ".join(f"{c[:10]:>10s}" for c in XCLASSES) + f" {'no_evt':>6s} {'undef':>5s} {'fast':>5s} {'slow':>5s}")
for box in BOX_ORDER:
    for sname, m in ((f"log M* > {MASS_MIN:g}", XT["log_mstar"] > MASS_MIN), (f"log M* <= {MASS_MIN:g}", XT["log_mstar"] <= MASS_MIN)):
        g = XT[(XT["box"] == box) & m]
        if not len(g):
            continue
        print(f"{box:8s} {sname:14s} {len(g):5d} " + " ".join(f"{int((g['xray_class'] == c).sum()):10d}" for c in XCLASSES)
              + f" {int((g['xray_class'] == 'no_event').sum()):6d} {int(g['rx_undefined'].sum()):5d} {int((g['fast_slow'] == 'fast').sum()):5d} {int((g['fast_slow'] == 'slow').sum()):5d}")
print(f"(undef: r_x undefined = a quench event but never gas-poor since SFT)")
print(f"\nmedians per class, log M* > {MASS_MIN:g} event galaxies: log M_dust/M_gas | log M_dust/M* | satellite fraction (zero dust at the floor); Mann-Whitney p against {XREF!r} on log D/G")
print(f"{'box':8s} " + " ".join(f"{c[:10]:>22s}" for c in XCLASSES) + "   " + " ".join(f"{'p ' + c[:8]:>10s}" for c in XOTHERS))
_gm = XT[(XT["log_mstar"] > MASS_MIN) & XT["event"]]
for box in BOX_ORDER:
    g = _gm[_gm["box"] == box]
    if not len(g):
        continue
    cells, ps = [], []
    for c in XCLASSES:
        h = g[g["xray_class"] == c]
        cells.append(f"{h['ldg'].median():6.2f} {h['lfd'].median():6.2f} {100 * (1 - h['central'].mean()):3.0f}% ({len(h):4d})" if len(h) else f"{'—':>22s}")
    ref = g.loc[g["xray_class"] == XREF, "ldg"].dropna()
    for c in XOTHERS:
        a = g.loc[g["xray_class"] == c, "ldg"].dropna()
        ps.append(f"{(mannwhitneyu(a, ref).pvalue if min(len(a), len(ref)) >= 3 else np.nan):10.1g}")
    print(f"{box:8s} " + " ".join(cells) + "   " + " ".join(ps))
for box in BOX_ORDER:
    g = _gm[_gm["box"] == box]
    if not len(g):
        continue
    print(f"\n{box}, log M* > {MASS_MIN:g}, event galaxies: the class x fast / slow, x the old AGN-coupling class (two-class scheme of B), x {' / '.join(_OTHERS)}:")
    print(pd.crosstab(g["xray_class"], g["fast_slow"]).to_string())
    print(pd.crosstab(g["xray_class"], g["agn_class"]).to_string())
    for _o in _OTHERS:
        print(pd.crosstab(g["xray_class"], g[_o]).to_string())
    a, b = (g.loc[g["fast_slow"] == c, "ldg"].dropna() for c in FS_CLASSES)
    print(f"  median log M_dust/M_gas fast / slow: {a.median():.2f} / {b.median():.2f} (Mann-Whitney p = {(mannwhitneyu(a, b).pvalue if min(len(a), len(b)) >= 3 else np.nan):.1g}); "
          f"log M_dust/M*: {g.loc[g['fast_slow'] == 'fast', 'lfd'].median():.2f} / {g.loc[g['fast_slow'] == 'slow', 'lfd'].median():.2f}")

# ── the m25 lookups used by every m25 section ──
_x25 = XT[XT["box"] == "cis25"]
_k25 = list(zip(_x25["snap"].astype(int), _x25["gal_id"].astype(int)))
XC_OF = dict(zip(_k25, _x25["xray_class"].astype(str)))
FS_OF = dict(zip(_k25, _x25["fast_slow"].astype(str)))
RX_OF = dict(zip(_k25, _x25["r_x"].to_numpy(float)))


def cls_of(snap, gid):
    """The class of an m25 galaxy: 'star_forming' for the SF controls of the selection, else its X-ray class ('no_event' without a quench event)."""
    k = (int(snap), int(gid))
    if POP_OF.get(k) == "SF":
        return "star_forming"
    return XC_OF.get(k, "no_event")


_nq = sum(1 for k in _sk if POP_OF[k] == "Q")
print(f"\nm25 lookups: {_nq} Q + {len(_sk) - _nq} SF galaxies of the selection; Q classes " + str({c: sum(1 for k in _sk if POP_OF[k] == 'Q' and XC_OF.get(k, 'no_event') == c) for c in XCLASSES + ['no_event']})
      + f"; figures -> {PAPERDIR}")
for _p, _by in ((P2D_FITS, "m25 Part 2d"), (P2D_TRACKS_FITS, "m25 Part 2d"), (ANNULUS_AV_FITS, "m25 Part 8b"), (ATTEN_FITS, "m25 Part 7a"), (ANNULUS_ISM_FITS, "m25 Part 8a"),
                (APERTURE_TRUTH_FITS, "m25 Part 7e"), (ANNULUS_KIN_FITS, "m25 Part 8j0"), (TRACKS_FITS, "KS Part 4a"), (GALS_FITS, "KS Part 5b"), (KIN_STAGE_FITS, "KS Part 3b"),
                (OBS_CSV, "git pull"), (OBS_GD_CSV, "git pull")):
    print(f"  {'ok     ' if os.path.exists(_p) else 'MISSING'} {os.path.relpath(_p, os.getcwd()) if _p.startswith(os.getcwd()) else _p}   [{_by}]")
```

## Section 1 — the feedback tracks: fast / slow quenchers against the X-ray classes

A Parts 1b and 1c merged into one figure. The rows are the sequence of the critical points (ladder: median, 16–84 %, 5–95 % of $t_k - t_{\rm SFT}$
per class) and the stacked tracks of the ungated jet weight $w_{\rm jet}$, of the gated coupling $x_{\rm coup} = w_{\rm jet}\,[f_{\rm gas} < 0.2]$
(with the fraction of the class that is gas-poor, dashed) and of the gas fraction $f_{\rm gas} = M_{\rm gas}/M_\star$ on the $t - t_{\rm SFT}$
clock; the columns are the two schemes — **fast / slow quenchers** ($\log \tau_q / t_{\rm QT}$ at $-1.25$) and the **X-ray classes** of Part 0.
The strips under each track panel mark the clock bins where a class differs from the column's reference class (slow; `XREF`), one strip per class,
bottom-up in class order (Mann–Whitney, black $p < 0.001$, dark grey $< 0.01$, light grey $< 0.05$).

The data are the m25 Part 2d tables (`agn_classifier_{windows,tracks}.fits`: every galaxy with a quench event, all masses; `P1_MASS_MIN`
restricts them). What to look for: the fast / slow split separates the *timing* of quenching but not the AGN state — both classes reach the
jet mode before SFT and open the gate at the same time. Under the onset classes the ladder *is* the definition: the `early` galaxies reach the
jet mode and the gate by QT (most of them around SFT; `onset4` separates the ones that were gas-poor before SFT), the `late` ones reach the jet mode
only after QT, and the `never` galaxies never build a jet-mode BH ($x_{\rm coup}$ stays at zero) while keeping a higher $f_{\rm gas}$ through and after QT.

```python
# ── Section 1 — the feedback tracks on the t - t_SFT clock: fast / slow quenchers beside the X-ray classes of Part 0 (A Parts 1a-1c) ──
G = Table.read(need(P2D_FITS, "powderday_flux_quenched_m25.ipynb Part 2d"))
_TT = Table.read(need(P2D_TRACKS_FITS, "powderday_flux_quenched_m25.ipynb Part 2d (2026-08-28+: it writes the tracks table)"))
G1 = {k: (_s(G[k]) if G[k].dtype.kind in "SU" else np.asarray(G[k], float)) for k in G.colnames}
N1 = len(G)
_gi = {(int(a), int(b)): i for i, (a, b) in enumerate(zip(G["snap"], G["gal_id"]))}
_row = np.array([_gi[(int(a), int(b))] for a, b in zip(_TT["snap"], _TT["gal_id"])], int)
_ord = np.lexsort((np.asarray(_TT["dt_sft"], float), _row))
_row = _row[_ord]
_cnt = np.bincount(_row, minlength=N1)
assert (_cnt > 0).all(), "every galaxy of the windows table needs a track (the two tables come from the same Part 2d run)"
_st0 = np.r_[0, np.cumsum(_cnt)[:-1]]
TRK = {k: [np.asarray(_TT[c], float)[_ord][s:s + n] for s, n in zip(_st0, _cnt)]
       for k, c in (("t", "dt_sft"), ("w", "w"), ("x", "x"), ("fg", "fg"), ("lfe", "lfe"), ("lmb", "lmb"), ("lmd", "lmd"))}

# rule parameters travel in the table headers; the SIMBA jet-mode / gate constants are those of the m25 notebook Part 0
PRE_WIN  = float(G.meta.get("PREWIN", 1.0))                 # pre-SFT window [Gyr] (the old rule's window, drawn for reference)
JET_ON   = float(G.meta.get("JETON", 0.5))                  # jet onset: first snapshot with the ungated w_jet >= this
IGN_FLOOR  = 1e-3                                           # BHAR floor of the ignition rule (m25 P2D_IGN_FLOOR)
FGAS_FLOOR = float(_TT.meta.get("FGASFLR", 1e-3))
CLOCK_BINS = np.arange(-2.5, 2.51, 0.25)
CTR, XL = 0.5 * (CLOCK_BINS[:-1] + CLOCK_BINS[1:]), (CLOCK_BINS[0], CLOCK_BINS[-1])
BAND_MIN = 8                                                # galaxies per clock bin for the 16-84 % band (A: 15; the small classes have ~10-20 galaxies)
CRIT = [("t_agn", r"$t_{\rm AGN}$ (ignition)", "*"), ("t_mbh", r"$M_{\rm BH} > 10^{%.1f}$" % JET_LOGMBH, "D"),
        ("t_jet", r"jet onset ($w_{\rm jet} \geq %g$)" % JET_ON, "^"), ("t_sft", "SFT", "s"),
        ("t_gate", r"$f_{\rm gas} < %g$ gate" % XRAY_FGAS_MAX, "v"), ("t_qt", "QT", "o"), ("t_anchor", "anchor", "P")]
CK = [k for k, _, _ in CRIT]
SHORT_CRIT = {"t_agn": "ignition", "t_mbh": r"$M_{\rm BH}$ threshold", "t_jet": "jet onset", "t_gate": "gate", "t_qt": "QT"}
LFE_SAT = np.log10(JET_FEDD) - 1.0                          # w_jet = 1 (f_Edd one dex below the jet threshold)
ALL = np.ones(N1, bool)
TAU_MED = float(np.nanmedian(G1["tau_q"]))
P1_MASS_MIN = None                                          # None: every event galaxy (all masses); e.g. MASS_MIN restricts the figure to log M* > 10.25
MASK1 = ALL if P1_MASS_MIN is None else (G1["log_mstar"] > P1_MASS_MIN)
# the two schemes: (title, {(snap, gal_id): label}, classes, colours, long names, short names, the reference class the others are contrasted with)
P1_SCHEMES = [("fast / slow quenchers", FS_OF, FS_CLASSES, FS_COL, FS_NAME, {c: c for c in FS_CLASSES}, "slow"),
              (XTITLE, XC_OF, XCLASSES, XCOL, XNAME, XSHORT, XREF)]
_gk1 = [(int(a), int(b)) for a, b in zip(G["snap"], G["gal_id"])]
LAB = {name: np.array([of.get(k, "no_event") for k in _gk1], dtype=object) for name, of, *_ in P1_SCHEMES}
print(f"Section 1: {N1} event galaxies; tracks: {len(_TT)} rows; median tau_q = {TAU_MED:.2f} Gyr" + (f"; drawn: log M* > {P1_MASS_MIN:g} ({int(MASK1.sum())})" if P1_MASS_MIN else "; all masses drawn"))
for name, _, classes, *_ in P1_SCHEMES:
    print(f"  {name:>34s}: " + ", ".join(f"{c} {int((MASK1 & (LAB[name] == c)).sum())}" for c in classes + ["no_event"]))


# ── track stacks on the t - t_SFT clock ──
def binvals(key, idx, lo, hi):
    """per galaxy in `idx`: mean of track `key` over lo <= t - t_SFT < hi (NaN without finite samples)."""
    out = np.full(len(idx), np.nan)
    for n, i in enumerate(idx):
        m = (TRK["t"][i] >= lo) & (TRK["t"][i] < hi) & np.isfinite(TRK[key][i])
        if m.any():
            out[n] = np.mean(TRK[key][i][m])
    return out


def stack(key, mask):
    out = np.full((len(CTR), 4), np.nan)
    idx = np.where(mask)[0]
    for j, (lo, hi) in enumerate(zip(CLOCK_BINS[:-1], CLOCK_BINS[1:])):
        v = binvals(key, idx, lo, hi)
        v = v[np.isfinite(v)]
        if len(v) >= 5:
            out[j] = (np.median(v), np.percentile(v, 16), np.percentile(v, 84), len(v))
    return out


def track(ax, key, mask, color, label=None, lw=2.2, band=True, ls="-", alpha=1.0):
    st = stack(key, mask)
    ok = st[:, 3] >= BAND_MIN
    if band:
        ax.fill_between(CTR, np.where(ok, st[:, 1], np.nan), np.where(ok, st[:, 2], np.nan), color=color, alpha=0.10, lw=0)
    ax.plot(CTR, st[:, 0], ls, color=color, lw=lw, label=label, alpha=alpha)
    return st


def gate_fraction(mask):
    """per clock bin: the fraction of the galaxies of `mask` whose bin-mean f_gas is below the gate (NaN with < 5 galaxies)."""
    idx = np.where(mask)[0]
    return np.array([np.mean(v[np.isfinite(v)] < XRAY_FGAS_MAX) if np.isfinite(v).sum() >= 5 else np.nan
                     for v in (binvals("fg", idx, lo, hi) for lo, hi in zip(CLOCK_BINS[:-1], CLOCK_BINS[1:]))])


def crit_stats(mask):
    """per critical point: median / 16 / 84 of t - t_SFT, the fraction of the group where it is defined, the fraction before SFT."""
    out = {}
    for k in CK:
        v = G1[k][mask]
        f = np.isfinite(v)
        out[k] = ((np.nanmedian(v), np.nanpercentile(v, 16), np.nanpercentile(v, 84), f.mean(), (v[f] < 0).mean() if f.any() else np.nan)
                  if f.sum() >= 3 else (np.nan,) * 5)
    return out


def crit_timing(ax, groups, xlim):
    """horizontal ladder of the critical points: median (marker), 16-84 (thick), 5-95 (thin) of t_k - t_SFT per group."""
    ng = len(groups)
    off = np.linspace(-0.3, 0.3, ng) if ng > 1 else [0.0]
    xhi = xlim[1] - 0.95                                     # text column on the right
    for gi, (gname, mask, color) in enumerate(groups):
        for ki, (k, lab, mk) in enumerate(CRIT):
            v = G1[k][mask]
            v = v[np.isfinite(v)]
            if len(v) < 3:
                continue
            y = ki + off[gi]
            p5, p16, p50, p84, p95 = np.percentile(v, [5, 16, 50, 84, 95])
            ax.plot([max(p5, xlim[0]), min(p95, xhi)], [y, y], "-", color=color, lw=0.9, alpha=0.7)
            ax.plot([max(p16, xlim[0]), min(p84, xhi)], [y, y], "-", color=color, lw=3.2, alpha=0.9, solid_capstyle="butt")
            ax.plot(np.clip(p50, xlim[0], xhi), y, mk, ms=8 if mk != "*" else 12, color=color, mec="k", mew=0.6, zorder=5,
                    label=gname if ki == 0 and ng > 1 else None)
            ax.text(xlim[1] - 0.03, y, f"{100 * np.isfinite(G1[k][mask]).mean():.0f}% | {100 * (v < 0).mean():.0f}%",
                    fontsize=7.5, color=color, ha="right", va="center")
    ax.set_yticks(range(len(CRIT))); ax.set_yticklabels([lab for _, lab, _ in CRIT]); ax.set_ylim(len(CRIT) - 0.5, -1.0)
    ax.axvline(0, color="k", lw=1); ax.axvline(xhi, color="0.7", lw=0.6); ax.set_xlim(*xlim); ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]")
    ax.text(xlim[1] - 0.03, -0.7, "defined |\nbefore SFT", fontsize=7.5, ha="right", va="center", color="0.3")
    ax.grid(axis="y", lw=0.4, alpha=0.4)
    if ng > 1:
        ax.legend(frameon=False, loc="lower left", fontsize=9)


def sig_strip(ax, key, lab, a, b, y0, y1):
    """bottom strip: class a vs class b Mann-Whitney per clock bin (black p < 0.001, dark grey < 0.01, light grey < 0.05)."""
    a_idx, b_idx = np.where(lab == a)[0], np.where(lab == b)[0]
    for lo, hi in zip(CLOCK_BINS[:-1], CLOCK_BINS[1:]):
        va, vb = binvals(key, a_idx, lo, hi), binvals(key, b_idx, lo, hi)
        va, vb = va[np.isfinite(va)], vb[np.isfinite(vb)]
        if len(va) < 5 or len(vb) < 5:
            continue
        p = mannwhitneyu(va, vb).pvalue
        col = "k" if p < 1e-3 else ("0.55" if p < 0.01 else ("0.85" if p < 0.05 else "none"))
        if col != "none":
            ax.fill_between([lo, hi], y0, y1, color=col, lw=0, transform=ax.get_xaxis_transform())


# ── the contrasts of each dichotomy: timing of the critical points, the jet lead, the window means of w_jet and f_gas ──
def _wmed(key, mask, lo, hi):
    v = binvals(key, np.where(mask)[0], lo, hi)
    return v[np.isfinite(v)]


print(f"\n{'scheme / class':>34s} {'N':>4s} {'t_jet':>6s} {'t_gate':>6s} {'t_qt':>6s} {'jet lead':>8s} {'w[-1,0)':>8s} {'w[0,.5)':>8s} {'x[0,.5)':>8s} {'x[.5,2)':>8s} {'fg[-.5,0)':>9s} {'fg[.5,2)':>8s}")
S1_ROWS = []
for name, _, classes, _, _, short, ref in P1_SCHEMES:
    lab = LAB[name]
    vals = {}
    for c in classes:
        m = MASK1 & (lab == c)
        cs = crit_stats(m)
        vals[c] = dict(w_pre=_wmed("w", m, -1.0, 0.0), w_post=_wmed("w", m, 0.0, 0.5), x_post=_wmed("x", m, 0.0, 0.5), x_late=_wmed("x", m, 0.5, 2.0),
                       fg_pre=_wmed("fg", m, -0.5, 0.0), fg_late=_wmed("fg", m, 0.5, 2.0))
        row = dict(scheme=name, cls=c, n=int(m.sum()), t_jet=cs["t_jet"][0], t_gate=cs["t_gate"][0], t_qt=cs["t_qt"][0], jet_lead=float(np.mean(G1["lead_jet"][m] > 0)) if m.any() else np.nan,
                   **{k: float(np.median(v)) if len(v) else np.nan for k, v in vals[c].items()})
        S1_ROWS.append(row)
        print(f"{name + ' / ' + c:>34s} {row['n']:4d} {row['t_jet']:+6.2f} {row['t_gate']:+6.2f} {row['t_qt']:+6.2f} {100 * row['jet_lead']:7.0f}% {row['w_pre']:8.2f} {row['w_post']:8.2f} "
              f"{row['x_post']:8.2f} {row['x_late']:8.2f} {row['fg_pre']:9.3f} {row['fg_late']:8.3f}")
    for a in [c for c in classes if c != ref]:
        _p = lambda k: (mannwhitneyu(vals[a][k], vals[ref][k]).pvalue if min(len(vals[a][k]), len(vals[ref][k])) >= 5 else np.nan)
        print(f"{'p ' + a + ' vs ' + ref:>34s} {'':>4s} " + " " * 22 + " ".join(f"{_p(k):8.1g}" for k in ("w_pre", "w_post", "x_post", "x_late")) + f" {_p('fg_pre'):9.1g} {_p('fg_late'):8.1g}")
print("(t_k: median t - t_SFT [Gyr] of the critical point; jet lead: jet mode before SFT; w / x / fg: median over the class of the per-galaxy mean in the clock window [Gyr]; p: Mann-Whitney against the reference class)")
pd.DataFrame(S1_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_feedback_tracks.csv"), index=False)

# ── the figure: one column per dichotomy; rows = the ladder, w_jet, x_coup (+ the gas-poor fraction), f_gas ──
P1_ROWS = [("w", r"median $w_{\rm jet}$ (ungated)"), ("x", r"median $x_{\rm coup} = w_{\rm jet}\,[f_{\rm gas}<%g]$ (gated)" % XRAY_FGAS_MAX), ("fg", r"median $f_{\rm gas} = M_{\rm gas}/M_\star$")]
X1 = (XL[0], 2.0)
fig, axes = plt.subplots(1 + len(P1_ROWS), len(P1_SCHEMES), figsize=(8.5 * len(P1_SCHEMES), 4.0 * (1 + len(P1_ROWS))))
for j, (name, _, classes, col, longname, short, ref) in enumerate(P1_SCHEMES):
    lab = LAB[name]
    others = [c for c in classes if c != ref]
    labm = np.where(MASK1, lab, "masked").astype(object)
    ax = axes[0, j]
    crit_timing(ax, [(f"{short[c]} ({int((MASK1 & (lab == c)).sum())})", MASK1 & (lab == c), col[c]) for c in classes], XL)
    ax.set_title(f"({'ab'[j]}) {name}: the sequence of feedback events", loc="left")
    for i, (key, ylab) in enumerate(P1_ROWS):
        ax = axes[1 + i, j]
        for c in classes:
            m = MASK1 & (lab == c)
            track(ax, key, m, col[c], label=f"{longname[c]} ({int(m.sum())})")
            if key == "x":
                ax.plot(CTR, gate_fraction(m), "--", color=col[c], lw=1.3, alpha=0.8, label=f"{short[c]}: fraction gas-poor ($f_{{\\rm gas}} < {XRAY_FGAS_MAX:g}$)")
        _sh = 0.03                                                     # strip height (axes fraction); one strip per class (bottom-up in class order), against the reference class
        for k_, c in enumerate(others):
            sig_strip(ax, key, labm, c, ref, _sh * k_, _sh * (k_ + 1))
            if len(others) > 1:
                ax.text(X1[1] - 0.02, _sh * (k_ + 0.5), short[c], transform=ax.get_xaxis_transform(), fontsize=6.0, color=col[c], ha="right", va="center", zorder=7)
        ax.axvline(0, color="k", lw=1); ax.axvline(TAU_MED, color="k", lw=0.8, ls="--"); ax.set_xlim(*X1)
        if key == "fg":
            ax.set_yscale("log"); ax.set_ylim(3e-3, 20); ax.axhline(XRAY_FGAS_MAX, color="k", lw=0.8, ls="-.")
            ax.text(X1[1] - 0.03, XRAY_FGAS_MAX, r"$f_{\rm gas}$ gate", fontsize=9, ha="right", va="bottom")
            ax.text(TAU_MED + 0.03, 0.07, "median QT", transform=ax.get_xaxis_transform(), ha="left", va="bottom", fontsize=9, color="0.35")
        else:
            ax.set_ylim(-0.03 - (1.06 * _sh * len(others)) / (1 - _sh * len(others)), 1.03)   # the strips sit below the w = 0 baseline
        ax.set_ylabel(ylab)
        ax.set_xlabel(r"$t - t_{\rm SFT}$ [Gyr]")
        ax.set_title(f"({'cdefgh'[2 * i + j]})", loc="left")
        ax.legend(frameon=False, loc="center left" if key != "x" else "upper left", fontsize=8.5)
fig.tight_layout()
fig.text(0.5, -0.01, wrap_caption("ladders: marker = median, thick = 16-84%, thin = 5-95% of t - t_SFT; right column: defined | before SFT. Bottom strips: the named class differs from the column's reference class "
         f"({', '.join(s[6] for s in P1_SCHEMES)}; Mann-Whitney p < 0.001 black, < 0.01 dark grey, < 0.05 light grey); shaded: 16-84% of the class (>= {BAND_MIN} galaxies per bin); x_coup dashed: the fraction of the class below the gate",
         width=int(14 * fig.get_figwidth())), fontsize=9.5, color="0.3", ha="center", va="top")
paper_save(fig, "feedback_tracks"); plt.show()
```

## Section 2 — the $A_V$ test on the KS plane, one column per X-ray class

A Part 3 under the new classes. The quenched galaxies with $\log M_\star > 10.25$ at the anchor (the mass sample of every per-class
figure here; A used $10.5 < \log M_\star < 11.2$, so the $A_V$ tercile edges are recomputed on this sample), binned by their **anchor
$A_V$** (core $r < 3.2$ kpc, dust_on / dust_off RT, median over the four sightlines of `annulus_av_allincl.fits`), followed on the
Kennicutt–Schmidt plane ($r < R_{50}({\rm H_2})$, $0.5 M / \pi R_{50}^2$, SFR over 100 Myr) from the quench start (SFT) through the
quench end (QT) to the track end — columns = all classes, then one column per class of Part 0; last row the $A_V$ ECDF of each column.

A class column with fewer than `P3_SPLIT_MIN` = 3 × `NMIN_BIN` galaxies is **not split by $A_V$**: its individual tracks and one median
track (in the class colour) are drawn instead — the `late` and `never` classes have ~12 and ~4 galaxies above the mass cut. The galaxies enter
through the KS products (`ks_galaxies.fits`: a track needs the KS notebook's own quench event, `has_clock`), so the per-class counts can be
smaller than Part 0's; the crosstab printed by cell 2a quantifies it. The printed $A_V$ tests contrast each class with `XREF`.

```python
# ── Section 2a — the KS products (Parts 4a / 5b of ks_tracks_quenched_m25) + the plane / presentation helpers of A Part 2a; the X-ray class on GALS ──
TRACKS = Table.read(need(TRACKS_FITS, "ks_tracks_quenched_m25.ipynb Part 4"))
GALS   = Table.read(need(GALS_FITS, "ks_tracks_quenched_m25.ipynb Part 5b (2026-08-28+: it writes ks_galaxies.fits)"))
OBS    = pd.read_csv(need(OBS_CSV, "git pull"))
OBS_S  = OBS[OBS["co_det"].astype(bool) & np.isfinite(OBS["logSigmaH2"])].reset_index(drop=True)


def _keys(tab):
    if "anchor_snap" in tab.colnames:
        return list(zip(np.asarray(tab["anchor_snap"], int), np.asarray(tab["gal_id"], int)))
    return [tuple(int(v) for v in g.split("_")) for g in _s(tab["gkey"])]


# ── conventions of the KS notebook (Parts 0 / 4b / 4c / 5b / 5c) ──
FIDUCIAL_AP = "R50_H2"
HE_FACTOR   = kl.HE_FACTOR
STAGES_DRAW = ["sft", "qt", "end"]
END_STAGES  = ("anchor", "end")
STAGE_MARKER = {"sft": "^", "qt": "s", "end": "o", "anchor": "o"}
STAGE_LABEL  = {"sft": "SFT (quench start)", "qt": "QT (quench end)", "end": "track end"}
END_DEF = r"track end = the anchor, or the last snapshot with $M_{\rm H_2}/M_\star > 10^{-4}$ when the anchor holds no H$_2$"
RELATIONS_DRAW, RELATIONS_FAINT = ("B08",), ("K98",)
KS_REF  = RELATIONS_DRAW[0]
KS_BAND = float(kl.RELATIONS[KS_REF]["sig"])
AP_TITLE = {"R50_H2": r"$r<R_{50}({\rm H_2})$"}
XLIM, YLIM = (-0.5, 3.6), (-4.6, 1.1)
C_OBS, C_OBS_EDGE = "#e7298a", "#3b0f2a"
NMIN_BIN = 4                                       # uncensored galaxies needed for a stage median inside one bin
Z_ALL    = (-1.0, 99.0, r"all anchors, $0.3\leq z\leq2$")
NMIN_ECDF = 4


# ── the KS plane ──
def draw_relations(ax, xs=None, tdep_lines=(0.1, 1.0, 10.0)):
    """The reference relation(s) with the published scatter band, the faint ones thin, + constant-t_dep lines. Returns legend items."""
    xs = np.linspace(XLIM[0] - 0.5, XLIM[1] + 0.5, 80) if xs is None else xs
    items = []
    for t, ls in zip(tdep_lines, (":", "-", "--")):
        ax.plot(xs, xs + 6 - np.log10(t * 1e9), color="0.65", ls=ls, lw=0.9, zorder=1)
        ax.text(XLIM[1] - 0.05, XLIM[1] - 0.05 + 6 - np.log10(t * 1e9) - 0.25, rf"$t_{{\rm dep}}$={t:g} Gyr",
                fontsize=7.5, color="0.45", ha="right", rotation=42, rotation_mode="anchor")
    for key in RELATIONS_DRAW:
        rel = kl.RELATIONS[key]
        y = rel["A"] + rel["N"] * xs
        ax.fill_between(xs, y - rel["sig"], y + rel["sig"], color=rel["color"], alpha=0.10, lw=0, zorder=0)
        ax.plot(xs, y, color=rel["color"], lw=1.4, ls=rel["ls"], zorder=2)
        items.append((Line2D([], [], color=rel["color"], lw=1.4, ls=rel["ls"]), rf"{rel['label']} $\pm$ {rel['scat_label']}"))
    for key in RELATIONS_FAINT:
        rel = kl.RELATIONS[key]
        ax.plot(xs, rel["A"] + rel["N"] * xs, color=rel["color"], lw=0.9, ls=rel["ls"], alpha=0.55, zorder=1)
        items.append((Line2D([], [], color=rel["color"], lw=0.9, ls=rel["ls"], alpha=0.55), rf"{rel['label']} (for reference)"))
    return items


def overlay_obs(ax, alpha=1.0):
    """ALMA-C11 points: filled = size measured (asymmetric errors), open + slope-1 arrow = unresolved (both Sigma lower limits)."""
    for _, r in OBS_S.iterrows():
        x, y = float(r["logSigmaH2"]), float(r["logSigmaSFR"])
        if bool(r["sigma_is_ll"]):
            ax.annotate("", xy=(x + 0.3, y + 0.3), xytext=(x, y),
                        arrowprops=dict(arrowstyle="-|>", color=C_OBS_EDGE, lw=1.2, mutation_scale=10, alpha=alpha), zorder=8)
            ax.scatter(x, y, s=110, marker="o", facecolors="white", edgecolors=C_OBS_EDGE, linewidths=1.4, alpha=alpha, zorder=9)
            ax.scatter(x, y, s=40, marker="o", c=C_OBS, edgecolors="none", alpha=alpha, zorder=10)
        else:
            elo = r["logSigmaSFR_elo"] if np.isfinite(r["logSigmaSFR_elo"]) else 0.0
            ehi = r["logSigmaSFR_ehi"] if np.isfinite(r["logSigmaSFR_ehi"]) else 0.0
            elo = min(elo, y - YLIM[0])
            xe = r["logSigmaH2_err"] if np.isfinite(r["logSigmaH2_err"]) else 0.0
            ax.errorbar(x, y, xerr=xe, yerr=[[elo], [ehi]], fmt="none", ecolor=C_OBS_EDGE, elinewidth=1.0, alpha=alpha, zorder=8)
            ax.scatter(x, y, s=110, marker="o", c=C_OBS, edgecolors=C_OBS_EDGE, linewidths=1.4, alpha=alpha, zorder=9)
    return [(Line2D([], [], marker="o", ls="", ms=9, mfc=C_OBS, mec=C_OBS_EDGE), rf"ALMA-C11 QGs, $z\approx0.4$ (N={len(OBS_S)}; size measured)"),
            (Line2D([], [], marker="o", ls="", ms=9, mfc="white", mec=C_OBS_EDGE), r"ALMA-C11 unresolved: $\Sigma$ lower limits")]


def ks_axes(ax, ap_label=FIDUCIAL_AP, sfr_note="SFR over 100 Myr (stars)"):
    conv = r"$0.5M/\pi R_{50}^2$" if ap_label.startswith("R50") else r"$M(<r)/\pi r^2$"
    ax.set_xlabel(r"$\log(\Sigma_{\rm H_2}/M_\odot\,{\rm pc}^{-2})$  [" + AP_TITLE[ap_label] + ", " + conv + rf", $\times{HE_FACTOR:g}$ He]")
    ax.set_ylabel(r"$\log(\Sigma_{\rm SFR}/M_\odot\,{\rm yr}^{-1}\,{\rm kpc}^{-2})$  [" + sfr_note + "]")
    ax.set_xlim(XLIM); ax.set_ylim(YLIM)
    ax.tick_params(direction="in", top=True, right=True); ax.grid(False)


def _sel_ap(ap_label, pop="Q"):
    return TRACKS[(_s(TRACKS["ap_label"]) == ap_label) & (_s(TRACKS["pop"]) == pop)]


def stage_medians(T, xcol="logSigmaH2", ycol="logSigmaSFR", stages=STAGES_DRAW, nmin=NMIN_BIN):
    """Per-stage median of x and y over the uncensored rows (keys x, y when >= nmin) and over ALL finite rows with the censored SFRs
    at their one-particle floor (x_all, y_all: an upper limit on the stage median)."""
    stg, ul = _s(T["stage"]), np.asarray(T["is_ul"], bool)
    x, y = np.asarray(T[xcol], float), np.asarray(T[ycol], float)
    out = []
    for st in stages:
        m_all = (stg == st) & np.isfinite(x) & np.isfinite(y)
        m = m_all & ~ul
        d = dict(stage=st, n=int(m.sum()), n_ul=int(((stg == st) & np.isfinite(x) & ul).sum()), n_all=int(m_all.sum()))
        if m_all.sum() >= nmin:
            d.update(x_all=np.median(x[m_all]), y_all=np.median(y[m_all]))
        if m.sum() >= nmin:
            d.update(x=np.median(x[m]), x16=np.percentile(x[m], 16), x84=np.percentile(x[m], 84),
                     y=np.median(y[m]), y16=np.percentile(y[m], 16), y84=np.percentile(y[m], 84))
        out.append(d)
    return out


def draw_median_track(ax, T, color, ycol="logSigmaSFR", lw=2.2, ls="-", zorder=6, ms=1.0):
    """Per-stage medians of T joined in stage order, stage symbols as markers; a stage with < NMIN_BIN uncensored galaxies but
    >= NMIN_BIN in total is drawn at its all-rows median: open marker + down arrow = an upper limit on the stage median."""
    meds = stage_medians(T, ycol=ycol)
    pts = [(m["x"], m["y"], m["stage"], False) if "x" in m else (m["x_all"], m["y_all"], m["stage"], True)
           for m in meds if ("x" in m) or ("x_all" in m)]
    if len(pts) >= 2:
        ax.plot([p[0] for p in pts], [p[1] for p in pts], ls=ls, color=color, lw=lw, alpha=0.9, zorder=zorder)
    for x, y, st, is_ul in pts:
        s = (110 if st in END_STAGES else 70) * ms
        if is_ul:
            ax.scatter(x, y, s=s, marker=STAGE_MARKER[st], facecolors="white", edgecolors=color, linewidths=1.4, zorder=zorder + 1)
            ax.annotate("", xy=(x, y - 0.35), xytext=(x, y), zorder=zorder + 1,
                        arrowprops=dict(arrowstyle="-|>", color=color, lw=1.2, mutation_scale=9, shrinkA=4, shrinkB=0))
        else:
            ax.scatter(x, y, s=s, marker=STAGE_MARKER[st], c=[color], edgecolors="k" if st in END_STAGES else "none", linewidths=0.9, zorder=zorder + 1)
    return meds


def ks_region(x, y, ul, ref=None, band=None):
    """Region of the points (x, y) w.r.t. relation `ref` (default KS_REF) with half-width `band` (default that relation's published scatter):
    'below' / 'within' / 'above'; 'undef' when x or y is NaN, or when the point is censored (`ul`, y = the one-particle floor) and its floor
    reaches the band. Returns (region str array, Delta = y - relation(x)). [ks_tracks_quenched_m25 Part 4c]"""
    ref = KS_REF if ref is None else ref
    band = float(kl.RELATIONS[ref]["sig"]) if band is None else float(band)
    x, y, ul = np.asarray(x, float), np.asarray(y, float), np.asarray(ul, bool)
    with np.errstate(invalid="ignore"):
        off = y - kl.relation_y(ref, x)
    fin = np.isfinite(off)
    reg = np.full(len(x), "undef", dtype=object)
    reg[fin & ~ul & (off < -band)] = "below"
    reg[fin & ~ul & (np.abs(off) <= band)] = "within"
    reg[fin & ~ul & (off > band)] = "above"
    reg[fin & ul & (off < -band)] = "below"                     # a censored point whose floor is already below the band
    return reg.astype(str), off


# ── presentation grids: no titles, no grid, column names above the first row, thin frame around the first column ──
def pres_grid(nrows, names, height=3.1, width=4.4):
    fig, axs = plt.subplots(nrows, len(names), figsize=(width * len(names), height * nrows), squeeze=False)
    for ax in axs.ravel():
        ax.grid(False)
        ax.tick_params(direction="in", top=True, right=True)
    for j, name in enumerate(names):
        axs[0, j].text(0.0, 1.02, name, transform=axs[0, j].transAxes, ha="left", va="bottom", fontsize=9.5, color="0.25")
    return fig, axs


def pres_finish(fig, axs, stem, pad=0.008, frame=True):
    """Common limits along every row (columns comparable), thin frame around the first column; paper_save."""
    for row in axs:
        row_ = [ax for ax in row if ax.has_data()]
        if len(row_) <= 1:
            continue
        if all(getattr(a, "_auto_y", False) for a in row_):
            logy = row_[0].get_yscale() == "log"
            ys = np.concatenate([np.asarray(l.get_ydata(), float) for a in row_ for l in a.get_lines()
                                 if len(l.get_ydata()) > 1 and l.get_transform() == a.transData])
            ys = ys[np.isfinite(ys) & ((ys > 0) if logy else True)]
            if logy:
                ys = np.log10(ys)
            if ys.size:
                lo, hi = ys.min(), ys.max()
                lo, hi = lo - 0.15 * (hi - lo), hi + 0.15 * (hi - lo)
                for a in row_:
                    a.set_ylim(*((10 ** lo, 10 ** hi) if logy else (lo, hi)))
        else:
            lims = np.array([a.get_ylim() for a in row_])
            lo, hi = (lims.min(), lims.max()) if lims[0, 0] <= lims[0, 1] else (lims.max(), lims.min())
            for a in row_:
                a.set_ylim(lo, hi)
        lims = np.array([a.get_xlim() for a in row_])
        for a in row_:
            a.set_xlim(lims.min(), lims.max())
    plt.tight_layout(w_pad=1.2, h_pad=1.0)
    if frame:
        fig.canvas.draw()
        inv = fig.transFigure.inverted()
        bbs = [ax.get_tightbbox(fig.canvas.get_renderer()).transformed(inv) for ax in axs[:, 0]]
        x0, x1 = min(b.x0 for b in bbs) - pad, max(b.x1 for b in bbs) + pad
        y0, y1 = min(b.y0 for b in bbs) - pad, max(b.y1 for b in bbs) + pad * 1.6
        fig.add_artist(Rectangle((x0, y0), x1 - x0, y1 - y0, transform=fig.transFigure, fill=False, ec="0.4", lw=0.9, zorder=20))
    paper_save(fig, stem)
    plt.show()


def _legend(ax, loc="best", **kw):
    kw.setdefault("handlelength", 1.8)
    ax.legend(frameon=False, fontsize=8, loc=loc, **kw)


# ── the galaxies: the X-ray class of every Q galaxy of the KS products (the products' own agn_class is kept only for the crosstab) ──
GK = _s(GALS["gkey"])
AZ = np.asarray(GALS["anchor_z"], float)
_cl  = np.array([XC_OF.get(k, "no_event") for k in _keys(GALS)])            # the class of every GALS row (this is what every section below reads)
_old = _s(GALS["agn_class"])                                                  # the KS run's own (old) AGN-coupling labels
_hc  = np.asarray(GALS["has_clock"], bool)                                    # the KS notebook found the quench event (SFT / QT of the particle histories)
_lmg = np.asarray(GALS["log_mstar_anchor"], float)
print(f"TRACKS: {len(TRACKS)} rows; GALS: {len(GALS)} Q galaxies ({int(_hc.sum())} with a KS quench clock); observed: {len(OBS_S)} ALMA-C11 sources with a surface density")
print(f"X-ray class of the GALS rows x the KS run's old class (all masses):")
print(pd.crosstab(pd.Series(_cl, name="xray_class"), pd.Series(_old, name="old class")).to_string())
print(f"X-ray class x KS quench clock (has_clock), log M* > {MASS_MIN:g}:")
print(pd.crosstab(pd.Series(_cl[_lmg > MASS_MIN], name="xray_class"), pd.Series(_hc[_lmg > MASS_MIN], name="has_clock")).to_string())
_nx = {c: int(((_cl == c) & ~_hc).sum()) for c in XCLASSES if ((_cl == c) & ~_hc).sum()}
if _nx:
    print(f"[note] classified galaxies without a KS quench clock (the KS notebook's particle-history event detector), which drop out of Sections 2, 4 and 6: {_nx}")
```

```python
# ── Section 2b — the mass-limited quenched galaxies binned by their anchor A_V on the KS plane, one column per X-ray class (A Part 3) ──
MASS_RANGE  = (MASS_MIN, np.inf)   # log M*/Msun at the anchor (selection-table catalogue mass): the mgt10p25 sample of every per-class figure
AV_APERTURE = "ap3kpc"          # the core (r < 3.2 kpc; the 'CORE' facet of m25 Part 8); any ap* / ann* label of annulus_av_allincl.fits
AV_EDGES    = None              # None: terciles of the mass-limited sample (the same edges in every column) | fixed edges, e.g. (0.1, 0.5) [mag]
AV_COLORS   = ["#2c7bb6", "#fdae61", "#d7191c", "#7b3294"]
P3_CLASSES  = list(XCLASSES)    # one column each (no_event galaxies have no SFT/QT -> no track)
P3_SPLIT_MIN = 3 * NMIN_BIN     # a class column with fewer galaxies is drawn unsplit: its individual tracks + one median track over all A_V
P3_ROWS     = [Z_ALL]           # one KS row per entry
SHOW_INDIVIDUAL = True          # thin lines: every galaxy's own track under its bin's median track
OBS_ALPHA   = 0.35              # the ALMA-C11 points (z ~ 0.4), faded
_AP_NAME = {"ap1kpc": r"$r<1$ kpc", "ap3kpc": r"$r<3.2$ kpc", "ap10kpc": r"$r<10$ kpc", "ap32kpc": r"$r<32$ kpc", "ap100kpc": r"$r<100$ kpc",
            "ann3kpc": r"$1<r<3.2$ kpc", "ann10kpc": r"$3.2<r<10$ kpc", "ann32kpc": r"$10<r<32$ kpc", "ann100kpc": r"$32<r<100$ kpc"}


def anchor_av():
    """{(snap, gal_id): A_V at the anchor} in AV_APERTURE: the median over the sightlines of annulus_av_allincl (m25 Part 8b), else the
    fiducial-sightline column of attenuation_vs_ism (m25 Part 7a)."""
    if os.path.exists(ANNULUS_AV_FITS):
        A = Table.read(ANNULUS_AV_FITS)
        A = A[_s(A["aperture"]) == AV_APERTURE]
        if not len(A):
            raise KeyError(f"{ANNULUS_AV_FITS} has no aperture {AV_APERTURE!r}")
        df = pd.DataFrame(dict(snap=np.asarray(A["snap"], int), gal_id=np.asarray(A["gal_id"], int), av=np.asarray(A["A_V"], float)))
        g = df.groupby(["snap", "gal_id"])["av"].median()
        return {(int(s), int(i)): float(v) for (s, i), v in g.items()}, f"median over {len(np.unique(_s(A['incl'])))} sightlines"
    A = Table.read(need(ATTEN_FITS, "powderday_flux_quenched_m25.ipynb Part 7a (or Part 8b for annulus_av_allincl.fits)"))
    col = f"A_V_{AV_APERTURE}"
    if col not in A.colnames:
        raise KeyError(f"{ATTEN_FITS} has no column {col!r}")
    return {(int(s), int(i)): float(v) for s, i, v in zip(A["snap"], A["gal_id"], np.asarray(A[col], float))}, "fiducial sightline"


AV_OF, AV_NOTE = anchor_av()
AV_LABEL = r"$A_V$ at the anchor [mag]"
AV_DEF = rf"$A_V$: dust_on / dust_off RT in the projected aperture {_AP_NAME.get(AV_APERTURE, AV_APERTURE)} at the anchor, {AV_NOTE}"
GALS["av_anchor"] = np.array([AV_OF.get(k, np.nan) for k in _keys(GALS)])
_lm, _av = np.asarray(GALS["log_mstar_anchor"], float), np.asarray(GALS["av_anchor"], float)
P3_MASS   = (_lm > MASS_RANGE[0]) & (_lm < MASS_RANGE[1])
P3_SAMPLE = P3_MASS & np.isfinite(_av) & np.isin(_cl, P3_CLASSES) & _hc
AV_EDGE = np.quantile(_av[P3_SAMPLE], [1 / 3, 2 / 3]) if AV_EDGES is None else np.asarray(AV_EDGES, float)
AV_BIN = np.where(P3_SAMPLE, np.searchsorted(AV_EDGE, np.where(np.isfinite(_av), _av, -1.0), side="right"), -1)
NB_AV = len(AV_EDGE) + 1
AV_BIN_LABEL = [rf"$A_V \leq {AV_EDGE[0]:.2f}$"] + [rf"${AV_EDGE[k - 1]:.2f} < A_V \leq {AV_EDGE[k]:.2f}$" for k in range(1, NB_AV - 1)] + [rf"$A_V > {AV_EDGE[-1]:.2f}$"]
B_OF = dict(zip(GK, AV_BIN))
print(f"A_V source: {AV_NOTE} in {AV_APERTURE}; Q galaxies with an A_V: {int(np.isfinite(_av).sum())}/{len(GALS)}")
print(f"mass-limited sample log M* > {MASS_RANGE[0]}: {int(P3_MASS.sum())} Q galaxies, of which {int(P3_SAMPLE.sum())} with a KS quench event, "
      f"an A_V and a class in {P3_CLASSES} (dropped: {dict(zip(*np.unique(_cl[P3_MASS & ~P3_SAMPLE], return_counts=True)))})")
print(f"A_V bins ({'terciles of the sample' if AV_EDGES is None else 'fixed edges'}): edges {np.round(AV_EDGE, 3)} mag")
print(f"{'class':>13s} " + " ".join(f"{'bin ' + str(k):>10s}" for k in range(NB_AV)) + f" {'N':>4s} {'med A_V':>8s} {'med logM*':>10s}")
for c in ["all"] + P3_CLASSES:
    m = P3_SAMPLE & ((_cl == c) if c != "all" else True)
    print(f"{c:>13s} " + " ".join(f"{int((m & (AV_BIN == k)).sum()):10d}" for k in range(NB_AV)) + f" {int(m.sum()):4d} {np.median(_av[m]):8.3f} {np.median(_lm[m]):10.2f}")
print("per bin (all classes): median log M* = " + ", ".join(f"{np.median(_lm[P3_SAMPLE & (AV_BIN == k)]):.2f}" for k in range(NB_AV))
      + "; median anchor z = " + ", ".join(f"{np.median(AZ[P3_SAMPLE & (AV_BIN == k)]):.2f}" for k in range(NB_AV)))
vb = _av[P3_SAMPLE & (_cl == XREF)]
for c in XOTHERS:
    va = _av[P3_SAMPLE & (_cl == c)]
    if len(va) >= NMIN_ECDF and len(vb) >= NMIN_ECDF:
        print(f"A_V distributions, {c} vs {XREF}: KS p = {ks_2samp(va, vb).pvalue:.2g}, Mann-Whitney p = {mannwhitneyu(va, vb).pvalue:.2g} "
              f"(medians {np.median(va):.3f} / {np.median(vb):.3f} mag, N = {len(va)} / {len(vb)})")

# the SF controls of the same mass range (anchor rows of the fiducial aperture)
_SF = _sel_ap(FIDUCIAL_AP, "SF")
_SF = _SF[(np.asarray(_SF["log_mstar_anchor"], float) > MASS_RANGE[0]) & (np.asarray(_SF["log_mstar_anchor"], float) < MASS_RANGE[1])]
_TQ = _sel_ap(FIDUCIAL_AP, "Q")
_TQ = _TQ[np.isin(_s(_TQ["stage"]), STAGES_DRAW)]
_tq_gk, _tq_st = _s(_TQ["gkey"]), _s(_TQ["stage"])
_stage_rank = {st: i for i, st in enumerate(STAGES_DRAW)}


def draw_tracks(ax, gks, color):
    """The individual tracks (thin) of the galaxies `gks` in stage order; returns their rows of _TQ."""
    T = _TQ[np.isin(_tq_gk, gks)]
    if SHOW_INDIVIDUAL:
        for g in gks:
            r = T[_s(T["gkey"]) == g]
            r = r[np.argsort([_stage_rank[s] for s in _s(r["stage"])])]
            x, y = np.asarray(r["logSigmaH2"], float), np.asarray(r["logSigmaSFR"], float)
            ok = np.isfinite(x) & np.isfinite(y)
            if ok.sum() >= 2:
                ax.plot(x[ok], y[ok], "-", color=color, lw=0.7, alpha=0.35, zorder=3)
    return T


def av_panel(ax, zlo, zhi, cls, obs_alpha=OBS_ALPHA):
    """KS panel of the mass-limited galaxies with anchor z in (zlo, zhi] and class `cls` (None = all P3_CLASSES): the SF cloud, then per
    A_V bin the individual tracks (thin) and the bin median track (thick); a class column below P3_SPLIT_MIN galaxies is drawn unsplit
    (bin -1, class colour). Returns [(bin, n_gal, meds), ...]."""
    m = P3_SAMPLE & (AZ > zlo) & (AZ <= zhi) & ((_cl == cls) if cls else np.ones(len(GALS), bool))
    gk_in = set(GK[m])
    sfz = np.asarray(_SF["anchor_z"], float)
    S = _SF[(sfz > zlo) & (sfz <= zhi)]
    ax.scatter(np.asarray(S["logSigmaH2"], float), np.asarray(S["logSigmaSFR"], float), s=7, c="0.6", alpha=0.3, lw=0, zorder=2)
    split = cls is None or len(gk_in) >= P3_SPLIT_MIN
    out = []
    for k in (range(NB_AV) if split else [-1]):
        gks = sorted(g for g in gk_in if (not split) or B_OF[g] == k)
        colk = AV_COLORS[k] if split else XCOL[cls]
        T = draw_tracks(ax, gks, colk)
        meds = draw_median_track(ax, T, colk) if len(gks) else []
        out.append((k, len(gks), meds))
    draw_relations(ax)
    overlay_obs(ax, alpha=obs_alpha)
    ks_axes(ax)
    ax.legend([Line2D([], [], color=AV_COLORS[k] if k >= 0 else XCOL[cls], lw=2.2, marker="o", mec="k", ms=6) for k, _, _ in out],
              [(f"{AV_BIN_LABEL[k]}  (N={n})" if k >= 0 else f"all $A_V$, unsplit  (N={n})") for k, n, _ in out], loc="lower right", frameon=False, fontsize=8, handlelength=1.8)
    return out


COL_NAMES = [None] + P3_CLASSES
COL_LABEL = {None: "all classes", **XSHORT}
fig, axs = pres_grid(len(P3_ROWS) + 1, [""] * len(COL_NAMES), height=5.2, width=5.4)
_rows_csv = []
for i, (zlo, zhi, zname) in enumerate(P3_ROWS):
    for j, cls in enumerate(COL_NAMES):
        out = av_panel(axs[i, j], zlo, zhi, cls)
        n_col = sum(n for _, n, _ in out)
        axs[i, j].text(0.0, 1.02, f"{COL_LABEL[cls]} (N={n_col}), {zname}", transform=axs[i, j].transAxes, ha="left", va="bottom", fontsize=9.5, color="0.25")
        for k, n, meds in out:
            for md_ in meds:
                _rows_csv.append(dict(z_row=zname.replace("$", ""), column=COL_LABEL[cls].replace("$", ""), av_bin=k, av_bin_label=AV_BIN_LABEL[k].replace("$", "") if k >= 0 else "all", n_gal=n,
                                      stage=md_["stage"], n_uncensored=md_["n"], n_censored=md_["n_ul"],
                                      logSigmaH2=md_.get("x", np.nan), logSigmaSFR=md_.get("y", np.nan),
                                      logSigmaH2_all=md_.get("x_all", np.nan), logSigmaSFR_all_upper=md_.get("y_all", np.nan)))
# last row: the A_V distribution of every column's galaxies (all anchors), SF controls of the same mass range dashed, the bin edges dotted
_sf_av = np.array([AV_OF.get(k, np.nan) for k in zip(np.asarray(_SF["anchor_snap"], int), np.asarray(_SF["gal_id"], int))])
for j, cls in enumerate(COL_NAMES):
    ax = axs[-1, j]
    m = P3_SAMPLE & ((_cl == cls) if cls else np.ones(len(GALS), bool))
    x, F = kl.ecdf(_av[m])
    ax.step(np.r_[x[0], x], np.r_[0, F], where="post", color=XCOL.get(cls, "k"), lw=2.0, zorder=4, label=f"{COL_LABEL[cls]} (N={int(m.sum())})")
    ax.plot([np.median(_av[m])], [0.5], marker="|", ms=12, mew=2.0, color=XCOL.get(cls, "k"), zorder=5)
    sv = _sf_av[np.isfinite(_sf_av)]
    if len(sv) >= NMIN_ECDF:
        xs_, Fs_ = kl.ecdf(sv)
        ax.step(np.r_[xs_[0], xs_], np.r_[0, Fs_], where="post", color=C_SF, lw=1.2, ls="--", zorder=3, label=f"SF controls, same mass range (N={len(sv)})")
    xlo = max(1e-3, 0.5 * np.nanmin(_av[P3_SAMPLE & (_av > 0)]))
    xhi = 1.5 * max(np.nanmax(_av[P3_SAMPLE]), np.nanmax(sv) if len(sv) else 0.0)
    for e in AV_EDGE:
        ax.axvline(e, color="0.4", lw=0.8, ls=":")
    for k in range(NB_AV):
        ax.axvspan(AV_EDGE[k - 1] if k else xlo, AV_EDGE[k] if k < NB_AV - 1 else xhi, color=AV_COLORS[k], alpha=0.08, lw=0, zorder=0)
    ax.set_xscale("log"); ax.set_xlim(xlo, xhi)
    ax.set_ylim(0, 1.02); ax.set_xlabel(AV_LABEL); ax.set_ylabel("cumulative fraction")
    _legend(ax, loc="lower right")
_items = [(Line2D([], [], marker=STAGE_MARKER[st], ls="", ms=8, mfc="0.4", mec="k" if st in END_STAGES else "none"), STAGE_LABEL[st]) for st in STAGES_DRAW]
_items += [(Line2D([], [], color="0.4", lw=2.2), "bin median track (stages in time order)"),
           (Line2D([], [], color="0.4", lw=0.7, alpha=0.5), "individual galaxy tracks (SFR = 0 epochs at the one-particle floor)"),
           (Line2D([], [], marker="o", ls="", ms=8, mfc="white", mec="0.4", mew=1.4), rf"open + $\downarrow$: < {NMIN_BIN} uncensored galaxies, median with SFR = 0 at the floor (upper limit)"),
           (Line2D([], [], marker="o", ls="", ms=5, mfc="0.6", mec="none"), f"SF controls, log $M_\\star$ > {MASS_RANGE[0]:g}, at their anchor")]
_rel = [(Line2D([], [], color=kl.RELATIONS[k]["color"], lw=1.4 if k in RELATIONS_DRAW else 0.9, ls=kl.RELATIONS[k]["ls"], alpha=1 if k in RELATIONS_DRAW else 0.55),
         rf"{kl.RELATIONS[k]['label']} $\pm$ {kl.RELATIONS[k]['scat_label']}" if k in RELATIONS_DRAW else rf"{kl.RELATIONS[k]['label']} (for reference)")
        for k in RELATIONS_DRAW + RELATIONS_FAINT]
_items += _rel + [(Line2D([], [], marker="o", ls="", ms=9, mfc=C_OBS, mec=C_OBS_EDGE), rf"ALMA-C11 QGs, $z\approx0.4$ (N={len(OBS_S)})")]
_caption = (f"{AV_DEF}; bins: the same edges in every column ({'terciles of the mass-limited sample' if AV_EDGES is None else 'fixed edges'}); "
            f"a class column with < {P3_SPLIT_MIN} galaxies is not split (one median track in the class colour); {END_DEF}")
fig.legend([h for h, _ in _items], [l for _, l in _items], loc="upper center", ncol=3, frameon=False, fontsize=8.5, bbox_to_anchor=(0.5, 0.0),
           title=wrap_caption(_caption, width=int(15 * fig.get_figwidth())), title_fontsize=8.5)
pres_finish(fig, axs, f"ks_av_bins_{FIDUCIAL_AP}")
AV_BINS = pd.DataFrame(_rows_csv)
AV_BINS.to_csv(os.path.join(PAPERDIR, "paper_xray_ks_av_bins.csv"), index=False)
print(f"bin medians -> {os.path.join(PAPERDIR, 'paper_xray_ks_av_bins.csv')} ({len(AV_BINS)} rows)")
print(AV_BINS.pivot_table(index=["column", "av_bin"], columns="stage", values="logSigmaSFR", aggfunc="first").reindex(columns=STAGES_DRAW).round(2).to_string())
```

## Section 3 — the radial profiles per class (A Part 4d)

Is a class more attenuated in its core, richer in dust or in gas there, younger, or just more rotating? One panel per quantity, the four
disjoint projected annuli (0–1, 1–3.2, 3.2–10, 10–32 kpc) inside → out along $x$, one line per class = the median over its galaxies
(each galaxy = the median of its four sightlines) with a galaxy-bootstrap 16–84 % band; the mass-matched SF controls in grey. One figure
per mass sample ($\log M_\star \gtrless 10.25$). Inputs: `annulus_ism_truth` (dust / gas / H$_2$), `annulus_av_allincl` (RT $A_V$),
`aperture_truth` ($M_\star$, age; $\geq 20$ stars), `annulus_kinematics` ($\kappa_{\rm rot}$ in the spherical rung / shell). A zero dust /
gas / H$_2$ mass is dropped, never floored (`n_dropped` in the CSV).

```python
# ── Section 3 — figure F of A (Part 4d): the radial trend per X-ray class (+ SF controls): A_V, dust / gas / H2 per stellar mass, stellar age and rotation support of each annulus ──
ANN_NAME    = {"ap1kpc": "0–1 kpc", "ann3kpc": "1–3.2 kpc", "ann10kpc": "3.2–10 kpc", "ann32kpc": "10–32 kpc", "ann100kpc": "32–100 kpc",
               "ap3kpc": "0–3.2 kpc", "ap10kpc": "0–10 kpc", "ap32kpc": "0–32 kpc", "ap100kpc": "0–100 kpc"}
CORE_LAB, OUT_LAB = "ap3kpc", "ann10kpc"      # core = the 0–3.2 kpc disc (m25 CORE9), outskirt = the 3.2–10 kpc annulus (OUT9); Section 4's zones
NSTAR_AP_MIN = 20                             # m25 7e sampling floor: M* and age need this many stars
KEY4        = ["snap", "gal_id", "incl", "aperture"]


def _frame(path, made_by, cols):
    t = Table.read(need(path, made_by))
    d = {}
    for c in cols:
        v = t[c]
        v = v.filled(np.nan) if hasattr(v, "filled") else v
        v = np.asarray(v)
        d[c] = _s(v) if c in ("incl", "aperture") else v.astype(v.dtype.type)   # native byte order (FITS is big-endian; pandas refuses it)
    return pd.DataFrame(d)


F_ANNULI   = ["ap1kpc", "ann3kpc", "ann10kpc", "ann32kpc"]          # the disjoint annuli (m25 T7 vocabulary; add "ann100kpc" for the CGM shell)
F_CLASSES  = ["star_forming"] + list(XCLASSES)                       # lines: SF controls first (grey), then the X-ray classes
F_MASS_SAMPLES = [(r"$\log M_\star > 10.25$", (MASS_MIN, np.inf), "mgt10p25"), (r"$\log M_\star < 10.25$", (-np.inf, MASS_MIN), "mlt10p25")]
F_KAPPA_LINE = 0.3                                                   # dashed: rotation- / pressure-supported (Section 4's KAPPA_PRESSURE)
F_NBOOT    = 500
F_NMIN     = 4                                                       # galaxies with a finite value for a point
# (column, axis label, log axis, CSV name): the per-galaxy value of an annulus; the ratios use the annulus' own stellar mass (m25 7e)
F_QUANT = [("A_V", r"$A_V$ [mag]", True, "A_V"), ("fdust", r"$M_{\rm dust}/M_\star$", True, "Mdust_over_Mstar"),
           ("fgas", r"$M_{\rm gas}/M_\star$", True, "Mgas_over_Mstar"), ("fH2", r"$M_{\rm H_2}/M_\star$", True, "MH2_over_Mstar"),
           ("age", "stellar age [Gyr]", False, "age_Gyr"), ("kappa_gas", r"$\kappa_{\rm rot}^{\rm gas}$", False, "kappa_gas"),
           ("kappa_H2", r"$\kappa_{\rm rot}^{\rm H_2}$", False, "kappa_H2"), ("kappa_star", r"$\kappa_{\rm rot}^{\star}$", False, "kappa_star")]
F_LAYOUT = (2, 4)

_ism = _frame(ANNULUS_ISM_FITS, "m25 Part 8a", KEY4 + ["ngas", "M_gas", "M_dust", "M_H2"])
_avt = _frame(ANNULUS_AV_FITS, "m25 Part 8b", KEY4 + ["A_V"])
_apt = _frame(APERTURE_TRUTH_FITS, "m25 Part 7e", KEY4 + ["nstar_ap", "mstar", "age_m_star_myr"])
_kin4 = _frame(ANNULUS_KIN_FITS, "m25 Part 8j0", ["snap", "gal_id", "aperture", "kappa_gas", "kappa_H2", "kappa_star"])
RF = _ism.merge(_avt, on=KEY4, how="inner").merge(_apt, on=KEY4, how="inner").merge(_kin4, on=["snap", "gal_id", "aperture"], how="left")
RF = RF[RF["aperture"].isin(F_ANNULI)].reset_index(drop=True)
RF["cls"] = [cls_of(s, g) for s, g in zip(RF["snap"], RF["gal_id"])]
RF["gkey"] = [f"{int(s)}_{int(g)}" for s, g in zip(RF["snap"], RF["gal_id"])]
RF["log_mstar"] = [LM_OF.get((int(s), int(g)), np.nan) for s, g in zip(RF["snap"], RF["gal_id"])]
_msF = np.asarray(RF["mstar"], float); _ageF = np.asarray(RF["age_m_star_myr"], float) / 1e3
_low = np.asarray(RF["nstar_ap"], int) < NSTAR_AP_MIN                # m25 7e floor: M* and the age need NSTAR_AP_MIN stars
_msF[_low], _ageF[_low] = np.nan, np.nan
with np.errstate(divide="ignore", invalid="ignore"):
    for c, num in (("fdust", "M_dust"), ("fgas", "M_gas"), ("fH2", "M_H2")):
        v = np.asarray(RF[num], float)
        RF[c] = np.where(np.isfinite(v) & (v > 0) & np.isfinite(_msF) & (_msF > 0), v / _msF, np.nan)   # zero mass -> NaN, never a floor
    RF["A_V"] = np.where(np.isfinite(RF["A_V"]) & (RF["A_V"] > 0), RF["A_V"], np.nan)
RF["age"] = _ageF
RF = RF[RF["cls"].isin(F_CLASSES)].reset_index(drop=True)
F_NAME = dict(XSHORT, star_forming="SF controls")
F_COLOR = dict(XCOL, star_forming=C_SF)


def _fboot(vals, keys, n=F_NBOOT, seed=0):
    """Per-galaxy medians over the sightline rows, then the bootstrap of their median -> (median, lo16, hi84, n_gal); NaN below F_NMIN."""
    v, k = np.asarray(vals, float), np.asarray(keys)
    ok = np.isfinite(v)
    v, k = v[ok], k[ok]
    uk, inv = np.unique(k, return_inverse=True)
    if len(uk) < F_NMIN:
        return np.nan, np.nan, np.nan, len(uk)
    pg = np.array([np.median(v[inv == i]) for i in range(len(uk))])
    bs = np.median(pg[np.random.default_rng(seed).integers(0, len(pg), (n, len(pg)))], axis=1)
    return float(np.median(pg)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), len(uk)


F_ROWS = []
_xF = np.arange(len(F_ANNULI))
for sname, mrange, stag in F_MASS_SAMPLES:
    S = RF[(RF["log_mstar"] > mrange[0]) & (RF["log_mstar"] < mrange[1])]
    nsel = {c: len(set(S["gkey"][S["cls"] == c])) for c in F_CLASSES}
    print(f"\n{sname.replace('$', '').replace(chr(92) + 'log', 'log').replace(chr(92) + 'star', '*')}: " + ", ".join(f"{F_NAME[c]} {nsel[c]}" for c in F_CLASSES))
    fig, axs = pres_grid(F_LAYOUT[0], [""] * F_LAYOUT[1], height=3.4, width=4.6)
    for q, (qv, qlab, islog, qname) in enumerate(F_QUANT):
        ax = axs[q // F_LAYOUT[1], q % F_LAYOUT[1]]
        for ci, c in enumerate(F_CLASSES):
            Sc = S[S["cls"] == c]
            pts = [_fboot(Sc[qv][Sc["aperture"] == lab], Sc["gkey"][Sc["aperture"] == lab]) for lab in F_ANNULI]
            for lab, (med, lo, hi, n) in zip(F_ANNULI, pts):
                nz = len(set(Sc["gkey"][(Sc["aperture"] == lab)])) - n
                F_ROWS.append(dict(sample=stag, quantity=qname, cls=c, annulus=lab, label=ANN_NAME[lab], n_gal=n, n_dropped=nz,
                                   median=med, lo16=lo, hi84=hi))
            y = np.array([p[0] for p in pts]); lo = np.array([p[1] for p in pts]); hi = np.array([p[2] for p in pts])
            ok = np.isfinite(y)
            if ok.sum() == 0:
                continue
            off = (ci - (len(F_CLASSES) - 1) / 2) * 0.06
            ax.fill_between(_xF[ok] + off, lo[ok], hi[ok], color=F_COLOR[c], alpha=0.18 if c != "star_forming" else 0.12, lw=0, zorder=2)
            ax.plot(_xF[ok] + off, y[ok], "-", color=F_COLOR[c], lw=1.8 if c != "star_forming" else 1.4, marker="o", ms=4.5, mec="k", mew=0.4, zorder=5,
                    label=f"{F_NAME[c]} (N={nsel[c]})" if q == 0 else None)
        if islog:
            ax.set_yscale("log")
        if qv.startswith("kappa"):
            ax.axhline(F_KAPPA_LINE, color="0.3", lw=1.0, ls="--", zorder=1)
            ax.set_ylim(0.0, 1.0)
        ax.set_xlim(-0.5, len(F_ANNULI) - 0.5)
        ax.set_xticks(_xF)
        ax.set_xticklabels([ANN_NAME[l].replace(" kpc", "") for l in F_ANNULI] if q // F_LAYOUT[1] == F_LAYOUT[0] - 1 else [])
        ax.set_ylabel(qlab)
        if q // F_LAYOUT[1] == F_LAYOUT[0] - 1:
            ax.set_xlabel("annulus [kpc]")
        ax.tick_params(direction="in", top=True, right=True); ax.grid(False)
    axs[0, 0].legend(loc="best", frameon=False, fontsize=8, handlelength=1.8)
    fig.text(0.0, 1.0, f"{sname}: the radial trend of each quantity at the anchor, per X-ray class and for the SF controls", fontsize=9.5, color="0.25", ha="left", va="bottom")
    fig.legend([Line2D([], [], color="0.4", lw=1.8, marker="o", mec="k", mew=0.4, ms=4.5), Rectangle((0, 0), 1, 1, fc="0.4", alpha=0.2, lw=0), Line2D([], [], color="0.3", lw=1.0, ls="--")],
               [f"class median over galaxies (each galaxy = the median of its 4 sightlines; >= {F_NMIN} galaxies)", "galaxy-bootstrap 16–84 %",
                rf"$\kappa_{{\rm rot}} = {F_KAPPA_LINE:g}$: rotation- / pressure-supported"],
               loc="upper center", ncol=3, frameon=False, fontsize=8.5, bbox_to_anchor=(0.5, 0.0),
               title=wrap_caption(f"projected annuli of the m25 ladder (dust_on / dust_off RT A_V; 8a dust / gas / H2; 7e M* and age, >= {NSTAR_AP_MIN} stars); a zero dust / gas / H2 mass is dropped "
                                  f"(n_dropped in the CSV); kappa in the spherical rung / shell (8j0, >= 10 weighted particles, sightline-independent); "
                                  + "; ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES), width=int(15 * fig.get_figwidth())), title_fontsize=8.5)
    plt.tight_layout(w_pad=1.0, h_pad=0.8)
    paper_save(fig, f"radial_profiles_{stag}")
    plt.show()

FR = pd.DataFrame(F_ROWS)
FR.to_csv(os.path.join(PAPERDIR, "paper_xray_radial_profiles.csv"), index=False)
print(f"\ntable -> {os.path.join(PAPERDIR, 'paper_xray_radial_profiles.csv')} ({len(FR)} rows)")
print("innermost annulus (0–1 kpc) per class: median [boot 16, 84] (N galaxies with a finite value):")
for stag in [s[2] for s in F_MASS_SAMPLES]:
    print(f"  {stag}:")
    for qv, _, islog, qname in F_QUANT:
        fmt = (lambda v: f"{v:.3g}") if islog else (lambda v: f"{v:.2f}")
        cells = []
        for c in F_CLASSES:
            r = FR[(FR["sample"] == stag) & (FR["quantity"] == qname) & (FR["cls"] == c) & (FR["annulus"] == F_ANNULI[0])].iloc[0]
            cells.append(f"{F_NAME[c].replace('$', '')[:8]:>8s} " + (f"{fmt(r['median'])} [{fmt(r['lo16'])},{fmt(r['hi84'])}] ({int(r['n_gal'])})" if np.isfinite(r["median"]) else f"— ({int(r['n_gal'])})"))
        print(f"    {qname:>16s}  " + " | ".join(cells))
```

## Section 4 — the critical-point sequences per class (A Part 5, figure B)

The three sequence figures of A Part 5 for the $\log M_\star > 10.25$ sample, the X-ray classes side by side in every panel on a
categorical axis — AGN ignition → jet-mode onset → SFT → QT → track end (the canonical order; the stages present in the KS products
are drawn):

* `kinematics_sequence` — $\kappa_{\rm rot}$ of the H$_2$ and of the stars in the **core** ($r < 3.2$ kpc sphere) and in the **outskirt**
  (3.2–10 kpc shell) of `ks_stage_kinematics.fits` (KS Part 3b; a $\kappa$ needs $\geq 10$ weighted particles — the H$_2$ rows are survivors);
* `stellar_sequence` — sSFR (SFR over 100 Myr of the KS tracks' projected apertures) and the mass-weighted stellar age of the same zones, plus
  the elapsed time of every interval over the cosmic time at its end (paired per galaxy);
* `ism_sequence` — $M_{\rm dust}/M_\star$ and $M_{\rm H_2}/M_\star$ of the projected core disc and of the 3.2–10 kpc annulus, with the range
  of the ALMA-C11 detections shaded.

Group median at every point with a galaxy-bootstrap 16–84 % interval ($\geq$ `NMIN_BIN` = 4 galaxies with a measurable value, so a point of a
small class can be blank). Because the small classes have ~4–12 galaxies, the figures have a single column (all $A_V$); `P5_AV_COLUMNS = True`

```python
# ── Section 4 — the critical-point sequences (A Part 5 figure B): kinematics / stellar / ISM of the core and of the outskirt, the X-ray classes side by side ──
P5_MASS_SAMPLES = [(r"$\log M_\star > 10.25$", (MASS_MIN, np.inf), "mgt10p25")]
P5_ZONES        = [CORE_LAB, OUT_LAB]      # core (0–3.2 kpc sphere), outskirt (3.2–10 kpc shell): the KS Part 3b zones (spherical)
P5_AV_COLUMNS   = False                    # True: one more column per anchor-A_V tercile beside the all-A_V column (the small classes then have ~3 galaxies per column)
P5_AV_EDGES     = None                     # None: terciles of the mass sample | fixed edges, e.g. (0.1, 0.5) [mag]
KAPPA_PRESSURE  = 0.3                      # dashed line: below it the component is pressure- rather than rotation-supported
P5_ZONE_NAME    = {CORE_LAB: "core, $r<3.2$ kpc", OUT_LAB: "outskirt, $3.2<r<10$ kpc", "ap10kpc": "$r<10$ kpc"}
P5_ZONE_SHORT   = {CORE_LAB: "core", OUT_LAB: "outskirt", "ap10kpc": "$r<10$ kpc"}
# the critical points in the canonical event order; agn_ign / jet_on need the KS notebook of 2026-08-28 evening+ (without them the figures use SFT / QT / end alone)
P5_STAGES       = ["agn_ign", "jet_on"] + list(STAGES_DRAW)
P5_STAGE_MARKER = dict(STAGE_MARKER, agn_ign="*", jet_on="D")
P5_STAGE_LABEL  = dict(STAGE_LABEL, agn_ign="AGN ignition (BHAR crosses half its pre-QT peak)", jet_on=r"jet-mode onset (ungated $w_{\rm jet} \geq 0.5$)")
P5_STAGE_SHORT  = {"agn_ign": "ignition", "jet_on": "jet on", "sft": "SFT", "qt": "QT", "end": "end"}
P5_NBOOT        = 500                      # galaxy bootstrap of every group median (16–84 % interval)
P5_FONT         = dict(label=13, tick=11.5, legend=10.5, header=12.5, tag=11.5, note=10.5)
P5_ZONE_COLOR   = {CORE_LAB: "#0f4c81", OUT_LAB: "#8c6d31"}                 # frames + tags: core deep blue, outskirt bronze (never used for lines)
P5_ZONE_TAG     = {CORE_LAB: "CORE\n$r<3.2$ kpc", OUT_LAB: "OUTSKIRT\n$3.2<r<10$ kpc"}
# (file stem, [(column, axis label, kind)], title, elapsed-time row): kind = kappa (0–1) | kappa_zoom (limits from the data) | lin | log
P5_SEQ_FIGS     = [("kinematics_sequence", [("kappa_H2", r"$\kappa_{\rm rot}^{\rm H_2}$", "kappa"), ("kappa_star", r"$\kappa_{\rm rot}^{\star}$", "kappa_zoom")],
                                            "rotation support of the H$_2$ and of the stars", False),
                   ("stellar_sequence",    [("ssfr", r"sSFR [yr$^{-1}$]", "log"), ("age_mw_gyr", "stellar age [Gyr]", "lin")], "sSFR, stellar age and elapsed time", True),
                   ("ism_sequence",        [("fdust", r"$M_{\rm dust}/M_\star$", "log"), ("fH2", r"$M_{\rm H_2}/M_\star$", "log")], "dust and H$_2$ fractions", False)]
P5_SSFR_COL     = "sfr100"                 # sSFR rows: SFR of the KS tracks' apertures averaged over 100 Myr (sfr100 | sfr25 | sfr_inst) over their M*
P5_SSFR_FLOOR   = 1e-13                    # [yr^-1] a zero SFR is floored here (kept in the medians)
P5_SSFR_LINE    = 1e-11                    # [yr^-1] dashed reference on the sSFR rows
P5_DT_NORM      = "end"                    # elapsed-time row: dt / cosmic time at the "end" | "start" of the interval | "anchor" | None (Gyr)
P5_OBS_DET      = "det_dust"               # age / M_dust band = the range covered by the ALMA-C11 sources with this flag (det_dust | det_co); None: no band
P5_OBS_DET_H2   = "det_co"                 # M_H2 / M* band = the CO-detected sources (MH2_fid: alpha_CO = 4.36 incl. He -> divided by P5_OBS_HE, SIMBA H2 is H-only)
P5_OBS_HE       = 1.36
P5_OBS_COLS     = ("logage", "logMstar", "logMdust")   # fiducial-run columns of almac11_gas_dust.csv (CIGALE age [log yr], log M*, log M_dust)
P5_OBS_COLOR    = C_OBS

KIN = Table.read(need(KIN_STAGE_FITS, "ks_tracks_quenched_m25.ipynb Part 3b (2026-08-28+)"))
NKIN_MIN = int(KIN.meta.get("NKIN_MIN", 10))
_kin = pd.DataFrame({c: (_s(KIN[c]) if c == "zone" else np.asarray(KIN[c]).astype(np.asarray(KIN[c]).dtype.type))
                     for c in ("snap", "gx", "zone", "kappa_gas", "kappa_H2", "kappa_star", "n_gas", "n_H2", "n_star", "m_H2", "m_star",
                               "age_mw_gyr", "has_vel")})
_kin = _kin[_kin["zone"].isin(P5_ZONES)].reset_index(drop=True)
# the epoch keys of the critical points (aperture-independent: one (snap, gx) per galaxy and stage) and their cosmic times, from TRACKS
_stgT, _popT = _s(TRACKS["stage"]), _s(TRACKS["pop"])
P5_HAVE = [st for st in P5_STAGES if st in set(_stgT[_popT == "Q"])]
_absent = [st for st in P5_STAGES if st not in P5_HAVE]
if _absent:
    print(f"[note] stages {_absent} are not in {os.path.relpath(TRACKS_FITS, os.getcwd())}: the KS notebook of 2026-08-28 evening+ adds them "
          "(Part 1 rebuild -> Part 2 BUILD_PLAN=True + sbatch -> Parts 3, 3b, 4); drawing the critical points " + " -> ".join(P5_HAVE))
_e = TRACKS[(_popT == "Q") & np.isin(_stgT, P5_HAVE)]
_ep = pd.DataFrame(dict(gkey=_s(_e["gkey"]), stage=_s(_e["stage"]), snap=np.asarray(_e["snap"], int), gx=np.asarray(_e["gx"], int),
                        t_gyr=np.asarray(_e["t_stage_gyr"], float))).drop_duplicates(["gkey", "stage"])
K5 = _ep.merge(_kin, on=["snap", "gx"], how="inner")
# M_dust / M*, M_H2 / M* and sSFR of the zones at every critical point from the KS tracks' projected apertures (m_dust, m_H2, sfr, m_star,
# n_star of ap3kpc and ap10kpc): core = the r < 3.2 kpc disc, outskirt = the 3.2–10 kpc annulus (ap10kpc - ap3kpc); NaN without
# NSTAR_AP_MIN stars; a zero dust / H2 mass -> NaN (never a floor), a zero SFR -> P5_SSFR_FLOOR (kept: quenching is the point)
_apT = _s(TRACKS["ap_label"])
_td = TRACKS[(_popT == "Q") & np.isin(_stgT, P5_HAVE) & np.isin(_apT, ["ap3kpc", "ap10kpc"])]
_td = pd.DataFrame(dict(gkey=_s(_td["gkey"]), stage=_s(_td["stage"]), ap=_s(_td["ap_label"]), m_dust=np.asarray(_td["m_dust"], float),
                        m_H2=np.asarray(_td["m_H2"], float), sfr=np.asarray(_td[P5_SSFR_COL], float), m_star=np.asarray(_td["m_star"], float),
                        n_star=np.asarray(_td["n_star"], float))).drop_duplicates(["gkey", "stage", "ap"])
_pv = _td.pivot_table(index=["gkey", "stage"], columns="ap", values=["m_dust", "m_H2", "sfr", "m_star", "n_star"], aggfunc="first")
_fd_rows = []
for (gk, st), r in _pv.iterrows():
    for zone, sel in ((CORE_LAB, lambda c: r[(c, "ap3kpc")]), (OUT_LAB, lambda c: r[(c, "ap10kpc")] - r[(c, "ap3kpc")])):
        ms, ns = sel("m_star"), sel("n_star")
        ok = np.isfinite(ms) and ms > 0 and ns >= NSTAR_AP_MIN
        d = dict(gkey=gk, stage=st, zone=zone)
        for c, num in (("fdust", "m_dust"), ("fH2", "m_H2")):
            v = sel(num)
            d[c] = (v / ms) if (ok and np.isfinite(v) and v > 0) else np.nan
        v = sel("sfr")
        d["ssfr"] = max(v / ms, P5_SSFR_FLOOR) if (ok and np.isfinite(v) and v >= 0) else np.nan
        _fd_rows.append(d)
K5 = K5.merge(pd.DataFrame(_fd_rows), on=["gkey", "stage", "zone"], how="left")
for c in ("fdust", "fH2", "ssfr"):
    if c not in K5:
        K5[c] = np.nan
# the parameter space covered by the observed sample: ALMA-C11 detections (fiducial CIGALE run), min–max of the age and of M_dust / M*
P5_OBS_BAND, P5_OBS_N, P5_OBS_N_H2 = {}, 0, 0
if P5_OBS_DET:
    OBS_GD = pd.read_csv(need(OBS_GD_CSV, "git pull"))
    _det = OBS_GD[OBS_GD[P5_OBS_DET].astype(bool)]
    _oage = 10 ** _det[P5_OBS_COLS[0]].to_numpy(float) / 1e9
    _ofd = 10 ** (_det[P5_OBS_COLS[2]].to_numpy(float) - _det[P5_OBS_COLS[1]].to_numpy(float))
    _oss = _det["SFR"].to_numpy(float) / 10 ** _det[P5_OBS_COLS[1]].to_numpy(float)      # fiducial CIGALE SFR over the fiducial M*
    P5_OBS_BAND = {"age_mw_gyr": (float(np.nanmin(_oage)), float(np.nanmax(_oage))), "fdust": (float(np.nanmin(_ofd)), float(np.nanmax(_ofd))),
                   "ssfr": (float(np.nanmin(_oss)), float(np.nanmax(_oss)))}
    P5_OBS_N = int(len(_det))
    print(f"observed band: {P5_OBS_N} ALMA-C11 sources with {P5_OBS_DET}: CIGALE age {P5_OBS_BAND['age_mw_gyr'][0]:.2f}–{P5_OBS_BAND['age_mw_gyr'][1]:.2f} Gyr, "
          f"M_dust/M* {P5_OBS_BAND['fdust'][0]:.2e}–{P5_OBS_BAND['fdust'][1]:.2e}, sSFR {P5_OBS_BAND['ssfr'][0]:.2e}–{P5_OBS_BAND['ssfr'][1]:.2e} /yr ({', '.join(_det['id'].astype(str))})")
    _co = OBS_GD[OBS_GD[P5_OBS_DET_H2].astype(bool)] if P5_OBS_DET_H2 else OBS_GD.iloc[:0]
    P5_OBS_N_H2 = int(len(_co))
    if P5_OBS_N_H2:
        _ofh = _co["MH2_fid"].to_numpy(float) / P5_OBS_HE / 10 ** _co[P5_OBS_COLS[1]].to_numpy(float)
        P5_OBS_BAND["fH2"] = (float(np.nanmin(_ofh)), float(np.nanmax(_ofh)))
        print(f"               {P5_OBS_N_H2} sources with {P5_OBS_DET_H2}: M_H2/M* {P5_OBS_BAND['fH2'][0]:.2e}–{P5_OBS_BAND['fH2'][1]:.2e} "
              f"(MH2_fid / {P5_OBS_HE:g} for He, over the fiducial M*; {', '.join(_co['id'].astype(str))})")
_nvel = int((~K5["has_vel"].astype(bool)).sum()) // max(len(P5_ZONES), 1)
_nz = max(len(P5_ZONES), 1)
print(f"stage kinematics: {len(KIN)} rows ({len(_kin) // _nz} files in the drawn zones); Q epochs matched: {len(K5) // _nz} of {len(_ep)} "
      + "(" + ", ".join(f"{st} {int((K5['stage'] == st).sum()) // _nz}/{int((_ep['stage'] == st).sum())}" for st in P5_HAVE) + ")"
      + (f"; [WARN] {_nvel} epochs come from files WITHOUT velocities (kappa NaN): re-run the Part 2 sbatch of the KS notebook + its Part 3b" if _nvel else ""))
_gal_cls = dict(zip(GK, _cl))
_gal_lm = dict(zip(GK, _lm))
_gal_tanc = dict(zip(GK, np.asarray(GALS["t_anchor_gyr"], float)))
_rank_canon = {st: i for i, st in enumerate(P5_STAGES)}


def p5_sample(mrange):
    """Galaxies (gkeys) of a mass sample with a KS quench event and an X-ray class (+ an anchor A_V when the A_V columns are drawn) + their A_V bin edges."""
    m = (_lm > mrange[0]) & (_lm < mrange[1]) & np.isin(_cl, XCLASSES) & _hc
    if P5_AV_COLUMNS:
        m &= np.isfinite(_av)
    ok = m & np.isfinite(_av)
    edges = np.quantile(_av[ok], [1 / 3, 2 / 3]) if (P5_AV_EDGES is None and ok.sum() >= 3) else np.asarray(P5_AV_EDGES if P5_AV_EDGES is not None else [0.0, 0.0], float)
    b = np.where(ok, np.searchsorted(edges, np.where(np.isfinite(_av), _av, -1.0), side="right"), -1)
    return m, edges, dict(zip(GK, b))


def _bin_labels(edges):
    nb = len(edges) + 1
    return [rf"$A_V \leq {edges[0]:.2f}$"] + [rf"${edges[k - 1]:.2f} < A_V \leq {edges[k]:.2f}$" for k in range(1, nb - 1)] + [rf"$A_V > {edges[-1]:.2f}$"]


def _bmed(v, n=P5_NBOOT, seed=0):
    """(median, 16, 84 of the bootstrap of the median, N) over the finite values; NaNs with fewer than NMIN_BIN."""
    v = np.asarray(v, float)
    v = v[np.isfinite(v)]
    if len(v) < NMIN_BIN:
        return np.nan, np.nan, np.nan, len(v)
    bs = np.median(v[np.random.default_rng(seed).integers(0, len(v), (n, len(v)))], axis=1)
    return float(np.median(v)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), len(v)


# ── the sequence figures: categorical x (critical points | intervals), groups side by side with a bootstrap interval ──
def _groups_and_columns(labels, nb):
    """(lines, columns) of the sequence figures: lines = the X-ray classes; columns = all A_V (+ the anchor-A_V bins when P5_AV_COLUMNS)."""
    lines = [(("cls", c), XCOL[c], XSHORT[c]) for c in XCLASSES]
    columns = [(None, "0.2", r"all $A_V$")] + ([(("av", k), AV_COLORS[k], labels[k]) for k in range(nb)] if P5_AV_COLUMNS else [])
    return lines, columns


def _members(gks, sel):
    """Galaxies of `gks` in the selection sel = ('cls', class) | ('av', bin) | None."""
    if sel is None:
        return list(gks)
    kind, val = sel
    return [g for g in gks if (_gal_cls[g] == val if kind == "cls" else _bin_of[g] == val)]


def seq_panel(ax, xs, groups, values, ylabel=None, zero=None, ylim=None, band=None):
    """values(group_key, i) -> array of per-galaxy values at x index i. Draws median + bootstrap 16–84 per group (offset in x);
    `band` = (lo, hi) shaded across the panel (the observed range). Returns {(group_key, i): (med, lo, hi, n)}."""
    res = {}
    ng = len(groups)
    for gi, (key, color, name) in enumerate(groups):
        off = (gi - (ng - 1) / 2) * 0.18
        xx, yy, lo, hi = [], [], [], []
        for i, x in enumerate(xs):
            med, l16, h84, n = _bmed(values(key, i))
            res[(key, i)] = (med, l16, h84, n)
            if np.isfinite(med):
                xx.append(i + off); yy.append(med); lo.append(med - l16); hi.append(h84 - med)
        if len(xx):
            ax.errorbar(xx, yy, yerr=np.array([lo, hi]), color=color, lw=2.2, elinewidth=1.4, capsize=3, marker="o", ms=6.5, mec="k", mew=0.6, zorder=5, label=name)
    if band is not None and np.all(np.isfinite(band)):
        ax.axhspan(band[0], band[1], color=P5_OBS_COLOR, alpha=0.13, lw=0, zorder=0)
        ax.axhline(band[0], color=P5_OBS_COLOR, lw=0.8, ls=":", alpha=0.7, zorder=1); ax.axhline(band[1], color=P5_OBS_COLOR, lw=0.8, ls=":", alpha=0.7, zorder=1)
    if zero is not None:
        ax.axhline(zero, color="0.3", lw=1.0, ls="--", zorder=1)
    ax.set_xlim(-0.6, len(xs) - 0.4)
    ax.set_xticks(range(len(xs)))
    ax.set_xticklabels(xs)
    if ylim is not None:
        ax.set_ylim(*ylim)
    if ylabel:
        ax.set_ylabel(ylabel, fontsize=P5_FONT["label"])
    ax.tick_params(direction="in", top=True, right=True, labelsize=P5_FONT["tick"]); ax.tick_params(axis="x", labelsize=P5_FONT["tick"] - 1); ax.grid(False)
    return res


def _zone_frame(ax, zone, tag=False):
    """The zone of a row: spines + tick marks in the zone colour; `tag` adds the rotated zone box left of the y label (first column)."""
    if zone not in P5_ZONE_COLOR:
        return
    c = P5_ZONE_COLOR[zone]
    for s in ax.spines.values():
        s.set_edgecolor(c); s.set_linewidth(1.7)
    ax.tick_params(color=c)
    if tag:
        ax.text(-0.36, 0.5, P5_ZONE_TAG[zone], transform=ax.transAxes, rotation=90, ha="center", va="center", color="white",
                fontsize=P5_FONT["tag"], fontweight="bold", bbox=dict(boxstyle="round,pad=0.35", fc=c, ec="none"), zorder=30)


def _seq_header(ax, cname_, lines, gk_col):
    """Column header: the column's name and, per line group, its number of galaxies (the class legend itself is figure-level)."""
    ax.text(0.0, 1.02, f"{cname_}  (N = {len(gk_col)})\n" + " $\\cdot$ ".join(f"{nm} {len(_members(gk_col, key))}" for key, _, nm in lines),
            transform=ax.transAxes, ha="left", va="bottom", fontsize=P5_FONT["header"] - 1.5, color="0.2", linespacing=1.15)


def _seq_class_legend(fig, lines):
    """The line groups (the classes) once per figure, top right, on the title line."""
    fig.legend([Line2D([], [], color=c, lw=2.2, marker="o", mec="k", mew=0.6, ms=6.5) for _, c, _ in lines], [nm for _, _, nm in lines],
               loc="lower right", bbox_to_anchor=(1.0, 1.035), ncol=min(len(lines), 2), frameon=False, fontsize=P5_FONT["legend"] + 1, handlelength=1.8, columnspacing=1.4)


def _pack(clauses, width=150, sep="; "):
    """Caption lines from clauses (never split inside a clause, so mathtext stays balanced), greedily packed to `width` characters."""
    lines, cur = [], ""
    for c in clauses:
        if cur and len(cur) + len(sep) + len(c) > width:
            lines.append(cur); cur = c
        else:
            cur = (cur + sep + c) if cur else c
    return "\n".join(lines + ([cur] if cur else []))


def _row_limits(res_list, kind, band=None):
    """Common y limits of one quantity over all its panels from the drawn medians +- bootstrap (and the observed band)."""
    vals = np.array([v for res in res_list for (m, l, h, n) in res.values() for v in (l, h) if np.isfinite(v)], float)
    if band is not None and np.all(np.isfinite(band)) and kind == "log":
        vals = np.append(vals, band)
    if kind == "kappa":
        return (-0.02, 1.02)
    if vals.size == 0:
        return None
    if kind == "log":
        vals = vals[vals > 0]
        return (vals.min() / 1.8, vals.max() * 1.8) if vals.size else None
    lo, hi = vals.min(), vals.max()
    if kind == "lin" and band is not None and np.all(np.isfinite(band)):
        d = np.array([v for res in res_list for (m, l, h, n) in res.values() for v in (l, h) if np.isfinite(v)], float)
        if d.size:                                         # the data decide the range; the band's lower edge always shows, its upper edge only when close
            lo, top = d.min(), max(d.max(), band[0] * 1.15)
            hi = band[1] if band[1] <= 1.4 * top else top
    pad = 0.08 * (hi - lo if hi > lo else 1.0)
    if kind == "kappa_zoom":
        return (max(-0.02, lo - pad), min(1.02, hi + pad))
    return (max(0.0, lo - pad) if lo >= 0 else lo - pad, hi + pad)


def _stage_vals(R, gks, stage, col):
    r = R[(R["gkey"].isin(gks)) & (R["stage"] == stage)]
    return r[col].to_numpy(float)


def _interval_vals(R, gks, a, b, col, norm=None):
    """Paired per-galaxy change col(b) - col(a) over the interval a -> b (both epochs measured); `norm` divides it by the cosmic
    time at the "end" (b) | "start" (a) of the interval | the "anchor" of the galaxy."""
    ra = R[(R["gkey"].isin(gks)) & (R["stage"] == a)].set_index("gkey")
    rb = R[(R["gkey"].isin(gks)) & (R["stage"] == b)].set_index("gkey")
    j = ra.index.intersection(rb.index)
    if not len(j):
        return np.zeros(0)
    d = rb.loc[j, col].to_numpy(float) - ra.loc[j, col].to_numpy(float)
    if norm == "end":
        d = d / rb.loc[j, "t_gyr"].to_numpy(float)
    elif norm == "start":
        d = d / ra.loc[j, "t_gyr"].to_numpy(float)
    elif norm == "anchor":
        d = d / np.array([_gal_tanc.get(g, np.nan) for g in j], float)
    return d


SEQ_ROWS, INT_ROWS = [], []
P5_INTERVALS = [(P5_HAVE[i], P5_HAVE[i + 1]) for i in range(len(P5_HAVE) - 1)]
for sname, mrange, stag in P5_MASS_SAMPLES:
    m_sample, edges, _bin_of = p5_sample(mrange)
    nb = len(edges) + 1
    labels = _bin_labels(edges)
    gk_all = list(GK[m_sample])
    print(f"\n{sname.replace('$', '').replace(chr(92) + 'log', 'log').replace(chr(92) + 'star', '*')}: {int(m_sample.sum())} Q galaxies "
          f"(KS quench event, class in {XCLASSES}" + (", anchor A_V" if P5_AV_COLUMNS else "") + "); per class: "
          + ", ".join(f"{c} {sum(1 for g in gk_all if _gal_cls[g] == c)}" for c in XCLASSES)
          + (f"; A_V edges {np.round(edges, 3)} mag; per class x bin: " + "; ".join(f"{c} " + "/".join(str(sum(1 for g in gk_all if _gal_cls[g] == c and _bin_of[g] == k)) for k in range(nb)) for c in XCLASSES) if P5_AV_COLUMNS else ""))
    lines, columns = _groups_and_columns(labels, nb)
    xs = [P5_STAGE_SHORT[st] for st in P5_HAVE]
    xsC = [f"{P5_STAGE_SHORT[a]}\n$\\rightarrow$ {P5_STAGE_SHORT[b]}" for a, b in P5_INTERVALS]
    _dt_lab = r"$\Delta t\,/\,t_{\rm cosmic}$" if P5_DT_NORM else r"$\Delta t$ [Gyr]"
    _dt_def = {"end": "the cosmic time at the end of the interval", "start": "the cosmic time at the start of the interval", "anchor": "the cosmic time of the anchor", None: None}[P5_DT_NORM]
    _grammar = ("lines = X-ray class" + (", columns = all $A_V$ | anchor-$A_V$ bins" if P5_AV_COLUMNS else "")
                + " (" + " $\\rightarrow$ ".join(P5_STAGE_SHORT[st] for st in P5_HAVE) + " = the canonical order of the critical points)")
    for stem, quants, ftitle, dt_row in P5_SEQ_FIGS:
        B_ROWS = [(zone, col, lab, kind) for col, lab, kind in quants for zone in P5_ZONES]
        nrow = len(B_ROWS) + (1 if (dt_row and P5_INTERVALS) else 0)
        fig, axs = pres_grid(nrow, [""] * len(columns), height=3.1, width=10.0 if len(columns) == 1 else 5.6)
        res_by_q = {col: [] for col, _, _ in quants}
        for j, (ckey, _, cname_) in enumerate(columns):
            gk_col = _members(gk_all, ckey)
            for i, (zone, col, ylab, kind) in enumerate(B_ROWS):
                ax = axs[i, j]
                R = K5[K5["zone"] == zone]
                band = P5_OBS_BAND.get(col)
                res = seq_panel(ax, xs, lines, lambda key, k, R=R, gk_col=gk_col, col=col: _stage_vals(R, _members(gk_col, key), P5_HAVE[k], col),
                                ylabel=ylab if j == 0 else None, zero=KAPPA_PRESSURE if kind.startswith("kappa") else (P5_SSFR_LINE if col == "ssfr" else None), band=band)
                if kind == "log":
                    ax.set_yscale("log")
                res_by_q[col].append(res)
                _zone_frame(ax, zone, tag=(j == 0))
                for (key, k), (med, lo, hi, n) in res.items():
                    SEQ_ROWS.append(dict(sample=stag, figure=stem, zone=zone, quantity=col, stage=P5_HAVE[k], column=cname_.replace("$", ""), line=str(key[1]),
                                         n_line=len(_members(gk_col, key)), n=n, median=med, lo16=lo, hi84=hi))
                if i == 0:
                    _seq_header(ax, cname_, lines, gk_col)
                if j > 0:
                    ax.tick_params(labelleft=False)
                if i < nrow - 1:
                    ax.tick_params(labelbottom=False)
            if dt_row and P5_INTERVALS:
                ax = axs[len(B_ROWS), j]
                R = K5[K5["zone"] == P5_ZONES[0]]
                res = seq_panel(ax, xsC, lines, lambda key, k, R=R, gk_col=gk_col: _interval_vals(R, _members(gk_col, key), *P5_INTERVALS[k], "t_gyr", norm=P5_DT_NORM),
                                ylabel=_dt_lab if j == 0 else None, zero=0.0)
                ax.tick_params(axis="x", labelsize=P5_FONT["tick"] - 1)
                for (key, k), (med, lo, hi, n) in res.items():
                    INT_ROWS.append(dict(sample=stag, zone="", quantity=f"dt_norm_{P5_DT_NORM}" if P5_DT_NORM else "t_gyr", interval=f"{P5_INTERVALS[k][0]}->{P5_INTERVALS[k][1]}",
                                         column=cname_.replace("$", ""), line=str(key[1]), n_line=len(_members(gk_col, key)), n_paired=n, median=med, lo16=lo, hi84=hi))
                if j > 0:
                    ax.tick_params(labelleft=False)
        for col, _, kind in quants:                       # one y range per quantity over both zones and every column
            lim = _row_limits(res_by_q[col], kind, P5_OBS_BAND.get(col))
            if lim is not None:
                for i, (zone, c_, _, _) in enumerate(B_ROWS):
                    if c_ == col:
                        for ax in axs[i]:
                            ax.set_ylim(*lim)
        fig.text(0.0, 1.0, f"{sname} — {ftitle} at the critical points", fontsize=P5_FONT["header"], color="0.2", ha="left", va="bottom")
        _seq_class_legend(fig, lines)
        _items = [(Line2D([], [], color="0.4", lw=2.2, marker="o", mec="k", mew=0.6, ms=6.5),
                   f"group median at each critical point, galaxy-bootstrap 16–84 % (>= {NMIN_BIN} galaxies with a measurable value" + ("; H$_2$ rows: survivors)" if "kappa_H2" in [q[0] for q in quants] else ")"))]
        if any(q[2].startswith("kappa") for q in quants):
            _items.append((Line2D([], [], color="0.3", lw=1.0, ls="--"), rf"$\kappa_{{\rm rot}} = {KAPPA_PRESSURE:g}$: rotation- / pressure-supported"))
        if any(q[0] in P5_OBS_BAND for q in quants):
            _qs = [q[0] for q in quants]
            _parts = [t for q, t in (("ssfr", f"sSFR {P5_OBS_BAND['ssfr'][0]:.1e}–{P5_OBS_BAND['ssfr'][1]:.1e} yr$^{{-1}}$"),
                                     ("age_mw_gyr", f"age {P5_OBS_BAND['age_mw_gyr'][0]:.1f}–{P5_OBS_BAND['age_mw_gyr'][1]:.1f} Gyr"),
                                     ("fdust", f"$M_{{\\rm dust}}/M_\\star$ {P5_OBS_BAND['fdust'][0]:.1e}–{P5_OBS_BAND['fdust'][1]:.1e}")) if q in _qs and q in P5_OBS_BAND]
            _btxt = "range of the ALMA-C11 detections (whole galaxy, fiducial CIGALE run): " + (", ".join(_parts) + f" of the {P5_OBS_N} {P5_OBS_DET.replace('det_', '')}-detected" if _parts else "")
            if "fH2" in P5_OBS_BAND and "fH2" in _qs:
                _btxt += ("; " if _parts else "") + f"$M_{{\\rm H_2}}/M_\\star$ {P5_OBS_BAND['fH2'][0]:.1e}–{P5_OBS_BAND['fH2'][1]:.1e} of the {P5_OBS_N_H2} CO-detected (H-only, $\\div${P5_OBS_HE:g} He)"
            _items.append((Rectangle((0, 0), 1, 1, fc=P5_OBS_COLOR, alpha=0.2, ec=P5_OBS_COLOR, ls=":"), _btxt))
        if dt_row and P5_INTERVALS:
            _items.append((Line2D([], [], color="0.3", lw=1.0, ls="--"), "last row: the elapsed time of each interval, paired per galaxy" + (f", divided by {_dt_def}" if _dt_def else "")
                           + "; negative = the canonical order was inverted (e.g. jet mode only after SFT)"))
        _clauses = [_grammar, "; ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES), "zones spherical about the galaxy centre", END_DEF]
        if P5_AV_COLUMNS:
            _clauses.append(AV_DEF)
        if any(q[2].startswith("kappa") for q in quants):
            _clauses.append(f"a kappa needs >= {NKIN_MIN} weighted particles")
        if {"fdust", "fH2", "ssfr"} & {q[0] for q in quants}:
            _clauses.append(f"$M_{{\\rm dust}}/M_\\star$, $M_{{\\rm H_2}}/M_\\star$ and sSFR from the KS tracks' projected apertures (core: $r<3.2$ kpc disc; outskirt: 3.2–10 kpc annulus), "
                            f">= {NSTAR_AP_MIN} stars, a zero mass dropped")
        if "ssfr" in [q[0] for q in quants]:
            _clauses.append(f"sSFR = SFR({P5_SSFR_COL.replace('sfr', '')} Myr) / $M_\\star$ of the aperture, a zero SFR floored at {P5_SSFR_FLOOR:g} yr$^{{-1}}$ "
                            f"(a group median on the floor = most of its galaxies form no stars there); dashed: {P5_SSFR_LINE:g} yr$^{{-1}}$")
        fig.legend([h for h, _ in _items], [l for _, l in _items], loc="upper center", ncol=1, frameon=False, fontsize=P5_FONT["note"], bbox_to_anchor=(0.5, 0.0),
                   title=_pack(_clauses, width=int(9.5 * fig.get_figwidth())), title_fontsize=P5_FONT["note"])
        pres_finish(fig, axs, f"{stem}_{stag}", frame=False)

pd.DataFrame(SEQ_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_kappa_sequence.csv"), index=False)
pd.DataFrame(INT_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_kappa_intervals.csv"), index=False)
print(f"\ntables -> {os.path.join(PAPERDIR, 'paper_xray_kappa_{sequence,intervals}.csv')} ({len(SEQ_ROWS)} / {len(INT_ROWS)} rows)")
_sq = pd.DataFrame(SEQ_ROWS)
print("\nthe classes at each critical point (median [boot 16, 84] (N), all A_V):")
for stem, quants, _, _ in P5_SEQ_FIGS:
    for col, ylab, kind in quants:
        for zone in P5_ZONES:
            fmt = (lambda v: f"{v:.2e}") if kind == "log" else (lambda v: f"{v:.2f}")
            cells = []
            for st in P5_HAVE:
                parts = []
                for c in XCLASSES:
                    r = _sq[(_sq["figure"] == stem) & (_sq["quantity"] == col) & (_sq["zone"] == zone) & (_sq["stage"] == st) & (_sq["column"] == "all A_V") & (_sq["line"] == c)]
                    if len(r) and np.isfinite(r["median"].iloc[0]):
                        parts.append(f"{c} {fmt(r['median'].iloc[0])} ({int(r['n'].iloc[0])})")
                    else:
                        parts.append(f"{c} — ({int(r['n'].iloc[0]) if len(r) else 0})")
                cells.append(f"{P5_STAGE_SHORT[st]}: " + " | ".join(parts))
            print(f"  {col:>11s} {P5_ZONE_SHORT[zone]:>8s}  " + "  ||  ".join(cells))
if P5_OBS_BAND:
    print(f"\ncore at each critical point vs the ALMA-C11 {P5_OBS_DET} band (fraction of the class inside the observed range):")
    for sname, mrange, stag in P5_MASS_SAMPLES:
        m_sample, edges, _bin_of = p5_sample(mrange)
        gk_all = list(GK[m_sample])
        R = K5[K5["zone"] == CORE_LAB]
        for col, nm in (("ssfr", "sSFR /yr"), ("age_mw_gyr", "age [Gyr]"), ("fdust", "Mdust/M*"), ("fH2", "MH2/M*")):
            if col not in P5_OBS_BAND:
                continue
            lo, hi = P5_OBS_BAND[col]
            cells = []
            for st in P5_HAVE:
                cells.append(f"{P5_STAGE_SHORT[st]}: " + " | ".join(
                    (lambda v: f"{XABBR[c]} {np.median(v):.3g} ({np.mean((v >= lo) & (v <= hi)):.0%} in band, N={len(v)})" if len(v) else f"{XABBR[c]} —")(
                        (lambda a: a[np.isfinite(a)])(_stage_vals(R, _members(gk_all, ("cls", c)), st, col))) for c in XCLASSES))
            print(f"  {stag} core {nm:>9s} [band {lo:.3g}–{hi:.3g}]  " + "  ||  ".join(cells))
```

## Section 5 — dust-to-gas against the X-ray exposure, its clock and its gate, per box (B Part 9, rows 1–3), and the exposure as a diagnostic inside each class

The first three rows of B's Part 9 figure under the classes of Part 0, read from the exposure table (no histories re-integrated here): for every
quenched galaxy of the m25 rule with $\log M_\star > 10.25$ (cis100 / cis50 / cis25 catalogue rows, B's `QM`) the dust-to-gas ratio
$\log M_{\rm dust}/M_{\rm gas}$ (zero dust at the floor, never dropped) against

* **(a–c)** the exposure $E_x = \int_{\rm SFT}^{\rm anchor} w_{\rm jet}\,[f_{\rm gas} < 0.2]\,{\rm d}t$ (fixed bins, the $E_x = 0$ pile its own bin);
* **(d–f)** the time since SFT, $t_{\rm anchor} - t_{\rm SFT}$ — the classes at fixed elapsed time (B's own row for the onset groups);
* **(g–i)** the gas-poor time $T_{\rm gas\text{-}poor} = \int_{\rm SFT}^{\rm anchor} [f_{\rm gas} < 0.2]\,{\rm d}t$ — the gate held fixed (B split this row in
  $r_x$ terciles; here the classes).

Every galaxy as a point, the binned medians of every class (equal-count bins on rows 2–3; a class below `P9_MIN` = 6 galaxies is one median
marker), the all-classified line dotted, and the median dust-to-gas ratio of the 50 Mpc runs without the X-ray channel (s50nox, s50noagn) as the
zero-exposure reference. Rows 2–3 are the test that matters: is a class dustier *at the same time since SFT and the same gas-poor time*? The printed
tables give every class's medians per box and the class contrasts (against `XREF`) inside terciles of the two clocks.

**The exposure as a diagnostic.** The onset class fixes *when* the channel first fired; what is left for the *amount* of exposure to explain? The table
`paper_xray_exposure_diagnostic.csv` gives, per box and class, the Spearman rank correlation of $E_x$, $f_x = E_x / \Delta t_{\rm SFT}$, $r_x$,
$T_{\rm gas\text{-}poor}$, $\Delta t_{\rm SFT}$ and $t_{\rm onset} - t_{\rm SFT}$ with $\log M_{\rm dust}/M_{\rm gas}$ and with $\log M_{\rm dust}/M_\star$.
Where $E_x$ is saturated (`early`: a jet-mode BH throughout, $E_x \approx \Delta t_{\rm SFT}$) only its normalised forms carry a signal; in the
`late` class $E_x$ is a clock (the anchor sits shortly after the first episode) and correlates strongly with the dust.

**The dusty tail per class (`paper_xray_dust_distributions`).** The medians of every class can sit below the observed threshold while the few galaxies above it belong
disproportionately to one class — and that class need not be the one that dominates the box. The figure draws $\log M_{\rm dust}/M_\star$ and
$\log M_{\rm dust}/M_{\rm gas}$ per class normalised to unit area (`P5_BIN` = 0.25 dex bins, the zero-dust galaxies as bars at the floor, a class below
`P5_HMIN` = 12 galaxies as individual marks), the tail threshold of `P5_QTY` dashed ($-4$ for $M_{\rm dust}/M_\star$, $-2.5$ for the dust-to-gas ratio), and in every
panel the class shares of the whole box against the shares among the galaxies above the threshold, with the hypergeometric probability of the tail count under random
draws from the box (the null: the tail is a random subset of the box). The printed table (`paper_xray_dust_tail_composition.csv`) scans the thresholds of `P5_SCAN`
and gives, per class, its share of the tail against its share of the box and the fraction of the class above each threshold.


```python
# ── Section 5 — dust-to-gas against the X-ray exposure per box, the classes of Part 0 (B Part 9 rows 1-3 from the exposure table) + the exposure as a diagnostic inside each class; B Part 1's Q / QM from the catalogue cache ──
def m25_cuts(df):
    """The m25 Part 1 cuts on the catalogue columns at the anchor -> {name: mask}."""
    ms, sfr = df["mstar"].to_numpy(float), df["sfr"].to_numpy(float)
    with np.errstate(divide="ignore", invalid="ignore"):
        ssfr = np.where(ms > 0, sfr / ms, np.nan)
        lm = np.log10(np.where(ms > 0, ms, np.nan))
    cuts = {"massive": lm > MASS_FLOOR,
            "passive": ssfr < PASSIVE_FACTOR / (df["t_H_gyr"].to_numpy(float) * 1e9),
            f"gas>={NGAS_MIN}": df["ngas"].to_numpy(float) >= NGAS_MIN,
            f"star>={NSTAR_MIN}": df["nstar"].to_numpy(float) >= NSTAR_MIN,
            "dust/H2": df["mdust"].to_numpy(float) >= DUST_TO_H2_MIN * df["mh2"].to_numpy(float)}
    if CENTRALS_ONLY:
        cuts["central"] = df["central"].to_numpy(float) > 0
    return cuts


def derive(df):
    """The axes of the figure from the catalogue columns (one row per galaxy)."""
    G = df.copy()
    ms = G[MSTAR_COL].to_numpy(float)
    ms = np.where(ms > 0, ms, np.nan)
    with np.errstate(divide="ignore", invalid="ignore"):
        G["log_mstar"] = np.log10(G["mstar"].to_numpy(float))                     # the selection's M* (whole galaxy)
        G["fdust"] = G["mdust"].to_numpy(float) / ms                              # zero dust / H2 -> dropped on the log axes (never floored)
        G["fh2"] = HE * G["mh2"].to_numpy(float) / ms
        G["fgas"] = G["mgas"].to_numpy(float) / ms                               # whole-galaxy gas over stars
        G["ssfr"] = np.maximum(G[SFR_COL].to_numpy(float), 0.0) / ms              # zero SFR kept: drawn at the floor
        G["re"] = R_PROJ_OVER_3D * G["r_half_star"].to_numpy(float)              # projected half-mass radius [kpc]
        G["sig_e"] = np.where(G["re"] > 0, 0.5 * ms / (np.pi * G["re"] ** 2), np.nan)
        mbh = G["mbh"].to_numpy(float)
        G["jet"] = (mbh > 0) & (np.log10(np.where(mbh > 0, mbh, 1.0)) > JET_LOGMBH) & (G["fedd"].to_numpy(float) < JET_FEDD)
        G["jet"] &= np.isfinite(G["fedd"].to_numpy(float))
        G["log_mbh"] = np.log10(np.clip(mbh, 1.0, None))
        _fd, _mg, _md = G["fdust"].to_numpy(float), G["mgas"].to_numpy(float), G["mdust"].to_numpy(float)
        G["lfd"] = np.where(_fd > 0, np.log10(np.where(_fd > 0, _fd, 1.0)), P9_FLOOR)                            # zero dust at the floor (the removal's end point)
        G["ldg"] = np.where((_md > 0) & (_mg > 0), np.log10(np.where((_md > 0) & (_mg > 0), _md / np.where(_mg > 0, _mg, 1.0), 1.0)), np.where(_mg > 0, P9_FLOOR, np.nan))
    G["gkey"] = [f"{b}_{int(s)}_{int(g)}" for b, s, g in zip(G["box"], G["snap"], G["gal_id"])]
    return G


CUTS = m25_cuts(CAT)
_allq = np.logical_and.reduce(list(CUTS.values()))
Q = derive(CAT[_allq])
QM = Q[Q["log_mstar"] > MASS_MIN].reset_index(drop=True)
_XC = ["event", "r_x", "e_x", "t_gaspoor", "f_x", "dt_sft", "tau_q_over_tqt", "w_pre", "agn_onset", "dt_onset_sft", "dt_onset_qt", "xray_class", "onset_class", "rx_class", "fast_slow", "agn_class"]
QM = QM.merge(XT[KEY + _XC], on=KEY, how="left")
QM["xray_class"] = QM["xray_class"].fillna("no_hist")                       # boxes without histories (the 50 Mpc variants) carry no class
print(f"the m25 rule on the catalogue cache: Q = {len(Q)} quenched galaxies; QM (log M* > {MASS_MIN:g}) = {len(QM)}: "
      + ", ".join(f"{b} {int((QM['box'] == b).sum())}" for b in sorted(QM["box"].unique())))
print("QM X-ray classes per box: " + "; ".join(f"{b}: " + ", ".join(f"{c} {int(((QM['box'] == b) & (QM['xray_class'] == c)).sum())}" for c in XCLASSES + ["no_event"]) for b in BOX_ORDER))

P9_MIN    = 6                       # galaxies per E_x bin for a binned median
P9_SMALL  = 150                     # a class with at most this many galaxies in a box is drawn emphasised (larger, darker points)
P9_EX_EDGES = [0.0, 0.0, 0.25, 0.5, 1.0, 2.0, 6.0]   # E_x bins [Gyr]: the first is E_x = 0 exactly (never exposed), then (0, 0.25], (0.25, 0.5], (0.5, 1], (1, 2], (2, 6]
P9_DG_REF = ["cis50nox", "cis50noagn"]              # the runs without the X-ray channel: their median dust-to-gas ratio drawn as the "zero exposure" reference
P9_BOXES  = [b for b in BOX_ORDER if (QM["xray_class"].isin(XCLASSES) & (QM["box"] == b)).sum() >= 2 * P9_MIN]
P8_EDGES  = (-5.0, -3.5)            # dust sides of B Part 8: no-dust below, dust-rich at or above (log M_dust/M*)
NBOOT     = 500
FONT      = dict(label=12.0, tick=10.0, legend=9.0, note=8.3, tag=10.5)
SEL_TXT   = (f"log $M_\\star$ > {MASS_MIN:g}, passive (sSFR < {PASSIVE_FACTOR:g} / $t_H$), >= {NGAS_MIN} gas and >= {NSTAR_MIN} star particles, "
             f"$M_{{\\rm dust}} \\geq 10^{{{np.log10(DUST_TO_H2_MIN):.0f}}} M_{{\\rm H_2}}$")
_dg_ref = {}
for v in P9_DG_REF:
    h = QM[(QM["box"] == v) & np.isfinite(QM["ldg"])]
    if len(h) >= P9_MIN:
        _dg_ref[v] = (float(h["ldg"].median()), float(h["lfd"].median()), len(h))
        print(f"  {v}: median log M_dust/M_gas {_dg_ref[v][0]:.2f}, log M_dust/M* {_dg_ref[v][1]:.2f} (N = {len(h)}) — the run without the X-ray channel, the zero-exposure reference")

# the classes per box: the dust content, the disc, the BH, the exposure quantities
print(f"\n{'box':7s} {'class':6s} {'N':>5s} {'log D/G':>8s} {'log f_d':>8s} {'no-dust':>7s} {'dust-rich':>9s} {'kappa_g':>7s} {'age':>5s} {'log MBH':>7s} {'jet':>5s} {'sat':>5s} {'E_x':>5s} {'T_gp':>5s} {'r_x':>5s} {'dt_SFT':>6s} {'w_pre':>6s} {'fast':>5s}")
S5_ROWS = []
for box in P9_BOXES:
    g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
    for c in XCLASSES + ["all"]:
        h = g if c == "all" else g[g["xray_class"] == c]
        if len(h) < 3:
            continue
        row = dict(box=box, cls=c, n=len(h), ldg=h["ldg"].median(), lfd=h["lfd"].median(), nodust=float((h["lfd"] < P8_EDGES[0]).mean()), dustrich=float((h["lfd"] >= P8_EDGES[1]).mean()),
                   kappa_gas=h["kappa_gas"].median(), age=h["age"].median(), log_mbh=h["log_mbh"].median(), jet=float(h["jet"].mean()), sat=float(1 - h["central"].mean()),
                   e_x=h["e_x"].median(), t_gaspoor=h["t_gaspoor"].median(), r_x=h["r_x"].median(), dt_sft=h["dt_sft"].median(), w_pre=h["w_pre"].median(), fast=float((h["fast_slow"] == "fast").mean()))
        S5_ROWS.append(row)
        print(f"{box:7s} {c:6s} {row['n']:5d} {row['ldg']:8.2f} {row['lfd']:8.2f} {100 * row['nodust']:6.0f}% {100 * row['dustrich']:8.0f}% {row['kappa_gas']:7.2f} {row['age']:5.2f} {row['log_mbh']:7.2f} "
              f"{100 * row['jet']:4.0f}% {100 * row['sat']:4.0f}% {row['e_x']:5.2f} {row['t_gaspoor']:5.2f} {row['r_x']:5.2f} {row['dt_sft']:6.2f} {row['w_pre']:6.2f} {100 * row['fast']:4.0f}%")
    for c in XOTHERS:
        _mw = lambda col: (mannwhitneyu(g.loc[g["xray_class"] == c, col].dropna(), g.loc[g["xray_class"] == XREF, col].dropna()).pvalue
                           if min((g["xray_class"] == c).sum(), (g["xray_class"] == XREF).sum()) >= 3 else np.nan)
        print(f"{'':7s} Mann-Whitney {c} vs {XREF}: log D/G p = {_mw('ldg'):.1g}; log f_d p = {_mw('lfd'):.1g}; kappa_gas p = {_mw('kappa_gas'):.1g}")
pd.DataFrame(S5_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_dg_vs_exposure_classes.csv"), index=False)
print(f"(no-dust: log M_dust/M* < {P8_EDGES[0]:g}; dust-rich: >= {P8_EDGES[1]:g}; T_gp = T_gaspoor [Gyr]; zero dust at the floor {P9_FLOOR:g})")

# ── the exposure as a diagnostic inside each class: the rank correlation of every exposure quantity with the dust content, class by class (the class fixes the
#    timing / the state of the channel; what is left for the amount of exposure to explain?) ──
D5_PRED = [("e_x", "E_x [Gyr]"), ("f_x", "f_x = E_x/dt_SFT"), ("r_x", "r_x = E_x/T_gaspoor"), ("t_gaspoor", "T_gaspoor [Gyr]"), ("dt_sft", "dt_SFT [Gyr]"), ("dt_onset_sft", "t_onset - t_SFT")]
D5_ROWS = []
print(f"\nthe exposure as a diagnostic inside each class (log M* > {MASS_MIN:g}): Spearman rho with log M_dust/M_gas (p) [rho with log M_dust/M*], per box and class; — below {2 * P9_MIN} galaxies")
for box in P9_BOXES:
    g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
    print(f"  {box:7s} {'class':11s} {'N':>5s} " + " ".join(f"{lab:>24s}" for _, lab in D5_PRED))
    for c in XCLASSES + ["all"]:
        h = g if c == "all" else g[g["xray_class"] == c]
        cells = []
        for col, lab in D5_PRED:
            x, y1, y2 = h[col].to_numpy(float), h["ldg"].to_numpy(float), h["lfd"].to_numpy(float)
            ok = np.isfinite(x) & np.isfinite(y1) & np.isfinite(y2)
            if ok.sum() >= 2 * P9_MIN and np.std(x[ok]) > 0:
                r1, r2 = spearmanr(x[ok], y1[ok]), spearmanr(x[ok], y2[ok])
                cells.append(f"{r1.statistic:+.2f} ({r1.pvalue:.0e}) [{r2.statistic:+.2f}]")
                D5_ROWS.append(dict(box=box, cls=c, predictor=col, n=int(ok.sum()), rho_ldg=float(r1.statistic), p_ldg=float(r1.pvalue), rho_lfd=float(r2.statistic), p_lfd=float(r2.pvalue)))
            else:
                cells.append("—")
        print(f"  {'':7s} {c:11s} {len(h):5d} " + " ".join(f"{s:>24s}" for s in cells))
pd.DataFrame(D5_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_exposure_diagnostic.csv"), index=False)
print(f"  -> {os.path.join(PAPERDIR, 'paper_xray_exposure_diagnostic.csv')}")


def bin_median(x, y, edges, nmin=P9_MIN, n=NBOOT, seed=0):
    """Median of y in the fixed x bins `edges` (the first bin = x == edges[0] exactly when edges[0] == edges[1]) with a bootstrap 16–84 % band -> DataFrame (x, y, lo, hi, n)."""
    x, y = np.asarray(x, float), np.asarray(y, float); ok = np.isfinite(x) & np.isfinite(y); x, y = x[ok], y[ok]
    rng, out = np.random.default_rng(seed), []
    for i in range(len(edges) - 1):
        m = (x == edges[0]) if (i == 0 and edges[0] == edges[1]) else ((x > edges[i]) & (x <= edges[i + 1]))
        if m.sum() < nmin:
            continue
        yy = y[m]; bs = np.median(yy[rng.integers(0, len(yy), (n, len(yy)))], axis=1)
        out.append((float(np.median(x[m])), float(np.median(yy)), float(np.percentile(bs, 16)), float(np.percentile(bs, 84)), int(m.sum())))
    return pd.DataFrame(out, columns=["x", "y", "lo", "hi", "n"])


# ── the figure: the first three rows of B's Part 9 figure with the two classes on every row — per box a column; (a-c) log D/G against the exposure E_x (fixed bins, the E_x = 0
#    pile its own bin); (d-f) against the time since SFT (the classes at fixed elapsed time); (g-i) against the gas-poor time T_gaspoor (the gate held fixed) ──
P9_ROWS = [("e_x", "ldg_vs_ex", r"X-ray exposure $E_x = \int_{\rm SFT}^{\rm anchor} w_{\rm jet}\,[f_{\rm gas} < 0.2]\,{\rm d}t$  [Gyr]", "dust-to-gas against the exposure", "fixed", 6.0),
           ("dt_sft", "ldg_vs_dtsft", r"time since SFT, $t_{\rm anchor} - t_{\rm SFT}$  [Gyr]", "the classes at fixed time since SFT", 6, 8.0),
           ("t_gaspoor", "ldg_vs_tgaspoor", r"gas-poor time $T_{\rm gas-poor} = \int_{\rm SFT}^{\rm anchor} [f_{\rm gas} < 0.2]\,{\rm d}t$  [Gyr]", "the classes at fixed gas-poor time", 5, 6.0)]
P9_TAGS = "abcdefghijkl"


def dg_panel(ax, g, xcol, edges, xlabel, title, rows, box, panel, xmax_cap, jitter_zero=False, ylabel=True):
    """One panel: per class the galaxies (points) and the binned medians (band = bootstrap 16-84 %), the all-classified dotted line, the no-X-ray references, rho(x, D/G)."""
    x = g[xcol].to_numpy(float)
    xj = x + (np.where(x <= 0, np.random.default_rng(1).uniform(-0.04, 0.04, len(g)), 0.0) if jitter_zero else 0.0)
    for c in XCLASSES:
        m = (g["xray_class"] == c).to_numpy()
        small = m.sum() <= P9_SMALL
        ax.scatter(xj[m], g.loc[m, "ldg"], s=11 if small else 9, color=XCOL[c], alpha=0.55 if small else 0.3, lw=0, zorder=3 if small else 2)
        tr = bin_median(x[m], g.loc[m, "ldg"].to_numpy(float), edges)
        if len(tr):
            ax.fill_between(tr["x"], tr["lo"], tr["hi"], color=XCOL[c], alpha=0.16, lw=0, zorder=3)
            ax.plot(tr["x"], tr["y"], color=XCOL[c], lw=2.6, marker="o", ms=6, mec="k", mew=0.5, zorder=4, path_effects=[withStroke(linewidth=4.2, foreground="white")],
                    label=f"{XSHORT[c]} (N = {int(m.sum())})")
            for _, r in tr.iterrows():
                rows.append(dict(box=box, panel=panel, group=c, x=r["x"], y=r["y"], lo=r["lo"], hi=r["hi"], n=r["n"]))
        elif m.sum():
            ax.plot([np.nanmedian(x[m])], [np.nanmedian(g.loc[m, "ldg"])], color=XCOL[c], ls="", marker="o", ms=9, mec="k", mew=0.6, zorder=6,
                    label=f"{XSHORT[c]} (N = {int(m.sum())}, one median)")
            rows.append(dict(box=box, panel=panel, group=c, x=float(np.nanmedian(x[m])), y=float(np.nanmedian(g.loc[m, "ldg"])), lo=np.nan, hi=np.nan, n=int(m.sum())))
    tr = bin_median(x, g["ldg"].to_numpy(float), edges)
    if len(tr):
        ax.plot(tr["x"], tr["y"], color="0.15", lw=2.0, ls=(0, (1.2, 1.4)), marker="s", ms=5, zorder=5, label=f"all classified (N = {len(g)})")
        for _, r in tr.iterrows():
            rows.append(dict(box=box, panel=panel, group="all", x=r["x"], y=r["y"], lo=r["lo"], hi=r["hi"], n=r["n"]))
    for v, (dg, fd, nv) in _dg_ref.items():
        ax.axhline(dg, color=BOXES[v]["color"], lw=1.6, ls=(0, (4.0, 1.6)), zorder=1, label=f"{BOX_SHORT.get(v, v)} (no X-ray channel) median, N = {nv}")
    ok = np.isfinite(x) & np.isfinite(g["ldg"].to_numpy(float))
    _rho = spearmanr(x[ok], g.loc[ok, "ldg"]) if ok.sum() >= 5 else None
    _rhc = {c: spearmanr(x[ok & (g["xray_class"] == c).to_numpy()], g.loc[ok & (g["xray_class"] == c).to_numpy(), "ldg"]) if (ok & (g["xray_class"] == c).to_numpy()).sum() >= 5 else None for c in XCLASSES}
    _xn = {"e_x": "$E_x$", "dt_sft": r"$\Delta t_{\rm SFT}$", "t_gaspoor": r"$T_{\rm gas-poor}$"}.get(xcol, xcol)
    if _rho is not None:
        ax.text(0.98, 0.97, rf"$\rho$({_xn}, D/G) = {_rho.statistic:+.2f} (p = {_rho.pvalue:.1g})" + "".join(f"; {XABBR[c]} {_rhc[c].statistic:+.2f}" for c in XCLASSES if _rhc[c] is not None),
                transform=ax.transAxes, ha="right", va="top", fontsize=FONT["note"] + 0.3, color="0.25", zorder=6)
    ax.axhline(P9_FLOOR + 0.25, color="0.6", lw=0.8, ls=":", zorder=1)
    ax.set_xlim(-0.15, min(xmax_cap, max(0.6, float(np.nanmax(x)) * 1.08))); ax.set_ylim(*_YL)
    ax.text(ax.get_xlim()[1] * 0.99, P9_FLOOR + 0.25, "zero dust (floor)", fontsize=FONT["note"] - 0.5, color="0.45", ha="right", va="bottom", zorder=7)
    ax.set_ylabel(r"$\log\,M_{\rm dust}\,/\,M_{\rm gas}$" if ylabel else ""); ax.set_xlabel(xlabel)
    ax.tick_params(direction="in", top=True, right=True, labelsize=FONT["tick"]); ax.grid(False)
    ax.set_title(title, loc="left", fontsize=FONT["tag"] - (0.8 if len(title) > 52 else 0.0))
    ax.legend(loc="lower left", frameon=False, fontsize=FONT["legend"] - 1.5)


if len(P9_BOXES):
    _nb = len(P9_BOXES)
    fig, axes = plt.subplots(len(P9_ROWS), _nb, figsize=(6.4 * _nb + 0.8, 5.2 * len(P9_ROWS)), squeeze=False, gridspec_kw=dict(hspace=0.34, wspace=0.20))
    _YL = (P9_FLOOR - 0.3, -0.2)
    _rows5 = []
    for i, box in enumerate(P9_BOXES):
        g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
        for r_, (xcol, panel, xlabel, ttl, binning, cap) in enumerate(P9_ROWS):
            if binning == "fixed":
                edges = P9_EX_EDGES
            else:
                xv = g[xcol].to_numpy(float); xv = xv[np.isfinite(xv)]
                edges = list(np.quantile(xv, np.linspace(0, 1, binning + 1))) if len(xv) >= P9_MIN * binning else [float(xv.min()), float(xv.max())] if len(xv) else [0.0, 1.0]
                edges[0] = edges[0] - 1e-9                                  # the lowest galaxy falls inside the first bin (bins are (lo, hi])
            dg_panel(axes[r_][i], g, xcol, edges, xlabel, f"({P9_TAGS[r_ * _nb + i]}) {BOXES[box]['label'].split(' (')[0]}: {ttl}", _rows5, box, panel, cap,
                     jitter_zero=(xcol == "e_x"), ylabel=(i == 0))
    fig.suptitle(wrap_caption(f"the X-ray channel after quenching: $E_x$ = the jet weight while $f_{{\\rm gas}} < {XRAY_FGAS_MAX:g}$, integrated from SFT to the anchor (catalogue histories); "
                              f"quenched galaxies, {SEL_TXT}; classes = {XTITLE}: " + ", ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES), width=int(13 * fig.get_figwidth())),
                 fontsize=FONT["note"] + 1, y=0.99)
    fig.subplots_adjust(top=0.94, bottom=0.05)
    paper_save(fig, "dg_vs_exposure")
    plt.show()
    pd.DataFrame(_rows5).to_csv(os.path.join(PAPERDIR, "paper_xray_dg_vs_exposure.csv"), index=False)
    print(f"  tables -> {os.path.join(PAPERDIR, 'paper_xray_dg_vs_exposure{.csv,_classes.csv}')}")
    # the classes at fixed control: inside terciles of the time since SFT and of the gas-poor time, the median D/G of each class and the Mann-Whitney between them
    print(f"\nthe classes at fixed elapsed time / gas-poor time (terciles of the box's classified galaxies): median log D/G per class (N), Mann-Whitney p against {XREF}")
    for box in P9_BOXES:
        g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
        for ctrl, clab in (("dt_sft", "dt_SFT"), ("t_gaspoor", "T_gaspoor")):
            xv = g[ctrl].to_numpy(float); ok = np.isfinite(xv)
            if ok.sum() < 3 * P9_MIN:
                continue
            q = np.quantile(xv[ok], [1 / 3, 2 / 3]); terc = np.where(~ok, -1, np.where(xv <= q[0], 0, np.where(xv <= q[1], 1, 2)))
            for k in range(3):
                h = g[terc == k]; ref = h.loc[h["xray_class"] == XREF, "ldg"].dropna()
                parts = []
                for c in XCLASSES:
                    a = h.loc[h["xray_class"] == c, "ldg"].dropna()
                    p = mannwhitneyu(a, ref).pvalue if (c != XREF and min(len(a), len(ref)) >= 3) else np.nan
                    parts.append(f"{XABBR[c]} {a.median() if len(a) else np.nan:5.2f} ({len(a):3d})" + (f" p={p:7.1g}" if c != XREF else "          "))
                print(f"  {box:7s} {clab:9s} T{k + 1} [{xv[ok & (terc == k)].min():.2f}, {xv[ok & (terc == k)].max():.2f}]: " + " | ".join(parts))
```

```python
# ── Section 5b — the figure-H grid of B (Parts 2-3): M_dust/M* and M_gas/M* against age, Sigma_e and sSFR, the X-ray classes per box + the observed points ──
from matplotlib.colors import LinearSegmentedColormap, Normalize
from matplotlib.gridspec import GridSpec
from matplotlib.ticker import FuncFormatter
GRID = [[("age", "fdust"), ("sigma_e", "fdust"), ("ssfr", "fdust")], [("age", "fh2"), ("sigma_e", "fh2"), ("ssfr", "fh2")]]
OBS_STRUCT_CSV = os.path.join(os.getcwd(), "obs_data", "almac11", "age_sersic_sigma.csv")       # ALMA-C11 R_e (COSMOS-Web / ACS), Sersic n
LIT_CSV        = os.path.join(os.getcwd(), "obs_data", "literature", "av_re_literature.csv")    # LEGA-C rows behind the Spilker+18 points
SPILKER_CSV    = os.path.join(os.getcwd(), "obs_data", "literature", "spilker18_legac.csv")     # Spilker+18 LEGA-C passive galaxies (M_H2)
SIM_GAS_ROW = "fgas"          # the simulation quantity on the H2 row: "fgas" = whole-galaxy M_gas/M* (B's choice) | "fh2" = 1.36 x the catalogue H2 over M*
OBS_COLOUR, OBS_CLIM = "av", (0.0, 1.4)
OBS_CTRL  = True              # the six ALMA-C11 controls as small circles
RE_KIND   = "maj"             # observed R_e: "maj" semi-major (van der Wel+14 convention) | "circ" circularised
CMAP      = ("YlGnBu", 0.2, 1.0)
NBIN_X, NMIN_X = 6, 30        # tracks: at most NBIN_X equal-count x bins of >= NMIN_X galaxies (fewer than 2 x NMIN_X galaxies: 4 bins of >= 5)
G5_POINTS_MAX = 150           # a class with at most this many galaxies is also drawn as individual points (the small classes)
G5_BOX_MARKER = {"cis100": "o", "cis50": "s", "cis25": "D"}
NOX_BOX = "cis50nox"          # the run without the X-ray channel: its whole quenched sample joins every grid as its own track (no classes: no histories yet)
UMEHATA   = dict(name="ADF22-QG1", ref="Umehata+25", z=3.0922, logM=11.11, re_kpc=1.01, av=0.5, logage=8.79, logMH2=10.26, logMd_ul=8.02, ssfr_ul=1.8e-12)
# axes: col = plot-space column (log10 for log axes), raw = the linear column, floor = drawn there below (log axes; without one non-positive values are dropped)
AXES = {"av":      dict(col="lav", raw="av", log=True, lim=(2e-3, 4.0), label=r"$A_V$ [mag]", short="A_V"),
        "sigma_e": dict(col="lsig", raw="sig_e", log=True, lim=(1.5e8, 5e10), label=r"$\Sigma_{\rm e} = M_\star\,/\,2\pi R_{\rm e}^2$  [M$_\odot$ kpc$^{-2}$]", short="Sigma_e",
                        nm=r"$\Sigma_{\rm e}$", obs_note="semi-major $R_{\\rm e}$ (obs) / " + f"{R_PROJ_OVER_3D:g} x the 3-D stellar half-mass radius (cat.)"),
        "age":     dict(col="age", raw="age", log=False, lim=(0.3, 8.8), label="stellar age  [Gyr]", short="age",
                        obs_note="SED fit (obs) / mass-weighted, whole galaxy (cat.)"),
        "ssfr":    dict(col="lssfr", raw="ssfr", log=True, floor=SSFR_FLOOR, lim=(6e-14, 4e-10), cens="ssfr_censor", label=r"sSFR  [yr$^{-1}$]", short="sSFR",
                        obs_note=f"SED fit (obs) / `{SFR_COL}` over $M_\\star$ (cat.); < 1e-13 drawn at 1e-13"),
        "re":      dict(col="lre", raw="re_kpc", log=True, lim=(0.4, 40.0), label=r"$R_{\rm e}$  [kpc]", short="R_e"),
        "fdust":   dict(col="lfdust", raw="fdust", log=True, lim=(2e-6, 6e-3), cens="fdust_censor", label=r"$M_{\rm dust}\,/\,M_\star$", short="M_dust/M*", nm=r"$M_{\rm dust}/M_\star$",
                        obs_note="CIGALE $M_{\\rm dust}$ / $M_\\star$ (FAST++: ADF22-QG1) / whole-galaxy dust over stars (cat.)"),
        "fh2":     dict(col="lfh2", raw="fh2", log=True, lim=(2e-4, 6e-1), cens="fh2_censor", label=r"$M_{\rm H_2}\,/\,M_\star$", short="M_H2/M*", nm=r"$M_{\rm H_2}/M_\star$",
                        obs_note=r"$\alpha_{\rm CO}$ masses incl. He (obs) / 1.36 x the whole-galaxy H$_2$ over stars (cat.)")}
if SIM_GAS_ROW == "fgas":     # the H2 row carries the whole-galaxy gas fraction for the simulation and the alpha_CO H2 for the observations (`lfh2` stays the catalogue H2)
    AXES["fh2"].update(col="lgasrow", label=r"$M_{\rm H_2}/M_\star$ (obs)   $M_{\rm gas}/M_\star$ (sim)", short="M_gas/M* (sim; obs M_H2/M*)", nm=r"$M_{\rm gas}/M_\star$ (sim)",
                       obs_note=r"the whole-galaxy gas (every member gas particle, He incl.) over stars (cat.) against the $\alpha_{\rm CO}$ H$_2$ masses incl. He (obs);"
                                "\n" r"   not nested with the catalogue H$_2$ (summed over the halo gas assigned to the galaxy): median $M_{\rm gas}/1.36\,M_{\rm H_2}$ = +0.08 dex (cis100)")
SIM_GAS_COL = {"fh2": "fh2", "fgas": "fgas"}[SIM_GAS_ROW]
_cm0 = plt.get_cmap(CMAP[0])
CM = LinearSegmentedColormap.from_list(CMAP[0] + "_xray", _cm0(np.linspace(CMAP[1], CMAP[2], 256)))
NORM = Normalize(*OBS_CLIM)
_cof = lambda v: CM(NORM(float(v))) if np.isfinite(v) else "white"


def _sigma_e(mstar, re_kpc):
    mstar, re_kpc = np.asarray(mstar, float), np.asarray(re_kpc, float)
    with np.errstate(divide="ignore", invalid="ignore"):
        return np.where(np.isfinite(mstar) & np.isfinite(re_kpc) & (re_kpc > 0), 0.5 * mstar / (np.pi * re_kpc ** 2), np.nan)


def _fs(spec, v):
    """Raw values -> plot space (log10 on log axes; a floor if the axis has one, else non-positive values are dropped)."""
    v = np.asarray(v, float)
    if not spec["log"]:
        return v
    fl = spec.get("floor")
    with np.errstate(divide="ignore", invalid="ignore"):
        if fl:
            return np.where(np.isfinite(v), np.log10(np.maximum(v, fl)), np.nan)
        return np.where(np.isfinite(v) & (v > 0), np.log10(np.where(v > 0, v, 1.0)), np.nan)


def _raw(spec, v):
    return 10 ** np.asarray(v, float) if spec["log"] else np.asarray(v, float)


def _add_fit_cols(df, cols):
    for key, raw in cols.items():
        df[AXES[key]["col"]] = _fs(AXES[key], df[raw])
    return df


for _G in (Q, QM):
    _add_fit_cols(_G, {"sigma_e": "sig_e", "age": "age", "ssfr": "ssfr", "re": "re", "fdust": "fdust", "fh2": SIM_GAS_COL})
    _G["lfh2"] = _fs(AXES["fh2"], _G["fh2"])                                       # the catalogue H2 under its own name, whatever the row shows

# ── the observed points (as figure H) ──
OBS = []
_gd = pd.read_csv(need(OBS_GD_CSV, "git pull")).set_index("id")
_st = pd.read_csv(need(OBS_STRUCT_CSV, "git pull")).set_index("id")
_tf = lambda v: str(v).strip().lower() in ("true", "1")
for sid, r in _gd.iterrows():
    if sid not in _st.index:
        print(f"  ALMA-C11 {sid}: no structure row -> skipped"); continue
    s = _st.loc[sid]
    re_ = float(s["re_kpc"]) if RE_KIND == "maj" else float(s["re_cosmosweb_kpc"] if np.isfinite(s["re_cosmosweb_kpc"]) else s["re_acs_kpc"])
    ctrl = str(sid).startswith("ctrl")
    if ctrl and not OBS_CTRL:
        continue
    M = 10 ** float(r["logMstar"])
    OBS.append(dict(sample="ALMA-C11 controls" if ctrl else "ALMA-C11", id=sid, z=float(r["z"]), av=float(r["AV_" + str(r["fid_run"]).strip()]), logM=float(r["logMstar"]),
                    re_kpc=re_, sig_e=float(_sigma_e(M, re_)), age=10 ** float(r["logage"]) / 1e9, ssfr=float(r["SFR"]) / M, ssfr_censor="det",
                    fdust=10 ** float(r["log_fdust"]), fdust_censor="ul" if _tf(r["dust_ul"]) else "det",
                    fh2=10 ** float(r["logMH2"]) / M, fh2_censor="ul" if _tf(r["co_ul"]) else "det",
                    marker="o", size=42 if ctrl else 120, edge="0.35" if ctrl else C_OBS_EDGE))
_lit = pd.read_csv(need(LIT_CSV, "git pull"))
_lit["re_use"] = _lit["re_5000_kpc"] if RE_KIND == "maj" else _lit["re_5000_kpc"] * np.sqrt(_lit["q"].clip(lower=0.05))
_lit["sig_e"] = _sigma_e(10 ** _lit["logM"].to_numpy(float), _lit["re_use"].to_numpy(float))
_lit["ssfr"] = np.where(_lit["logssfr"] > -50, 10 ** _lit["logssfr"].to_numpy(float), 0.0)
_lit["age"] = 10 ** _lit["logage_fit"].to_numpy(float)
_legac = _lit[_lit["sample"] == "LEGA-C"].drop_duplicates("id").set_index("id")
for _, r in pd.read_csv(need(SPILKER_CSV, "git pull")).iterrows():
    lid = int(str(r["galaxy"]).replace("LEGA-C", "").strip())
    if lid not in _legac.index:
        print(f"  Spilker+18 {r['galaxy']}: not in the LEGA-C literature rows -> skipped"); continue
    L = _legac.loc[lid]
    OBS.append(dict(sample="Spilker+18", id=f"LEGA-C {lid}", z=float(L["z"]), av=float(L["av"]), logM=float(L["logM"]), re_kpc=float(L["re_use"]), sig_e=float(L["sig_e"]),
                    age=float(L["age"]), ssfr=float(L["ssfr"]), ssfr_censor="det", fdust=np.nan, fdust_censor="none",
                    fh2=10 ** float(r["logMH2_h"]) / 10 ** float(L["logM"]), fh2_censor="det", marker="D", size=64, edge="0.2"))
U = UMEHATA
OBS.append(dict(sample=f"{U['name']} ({U['ref']})", id=U["name"], z=U["z"], av=U["av"], logM=U["logM"], re_kpc=U["re_kpc"], sig_e=float(_sigma_e(10 ** U["logM"], U["re_kpc"])),
                age=10 ** U["logage"] / 1e9, ssfr=U["ssfr_ul"], ssfr_censor="ul", fdust=10 ** (U["logMd_ul"] - U["logM"]), fdust_censor="ul",
                fh2=10 ** (U["logMH2"] - U["logM"]), fh2_censor="det", marker="*", size=300, edge="k"))
OBS = pd.DataFrame(OBS)
_add_fit_cols(OBS, {"av": "av", "sigma_e": "sig_e", "age": "age", "ssfr": "ssfr", "re": "re_kpc", "fdust": "fdust", "fh2": "fh2"})
OBS["lfh2"] = _fs(AXES["fh2"], OBS["fh2"])
print(f"observed points: {int((OBS['sample'] == 'ALMA-C11').sum())} ALMA-C11 QGs + {int((OBS['sample'] == 'ALMA-C11 controls').sum())} controls, "
      f"{int((OBS['sample'] == 'Spilker+18').sum())} Spilker+18, {U['name']}")


def run_track(x, y, nbins=NBIN_X, nmin=NMIN_X, nmin_small=5, n=NBOOT, seed=0):
    """Running median of y in equal-count x bins (plot-space columns, one row per galaxy) with a bootstrap 16–84 % band -> (x, y, lo, hi, n);
    a sample smaller than 2 x nmin falls back to 4 bins of >= min(nmin, nmin_small)."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    ok = np.isfinite(x) & np.isfinite(y)
    x, y = x[ok], y[ok]
    if len(x) < 2 * nmin:
        nbins, nmin = 4, min(nmin, nmin_small)
    nb = min(nbins, len(x) // nmin)
    if nb < 2:
        return pd.DataFrame(columns=["x", "y", "lo", "hi", "n"])
    e = np.quantile(x, np.linspace(0, 1, nb + 1))
    ib = np.clip(np.searchsorted(e, x, side="right") - 1, 0, nb - 1)
    rng, out = np.random.default_rng(seed), []
    for b in range(nb):
        yy = y[ib == b]
        bs = np.median(yy[rng.integers(0, len(yy), (n, len(yy)))], axis=1)
        out.append((np.median(x[ib == b]), np.median(yy), np.percentile(bs, 16), np.percentile(bs, 84), len(yy)))
    return pd.DataFrame(out, columns=["x", "y", "lo", "hi", "n"])


def _draw_track(ax, tr, xs, ys, st, band=True, z=4):
    X, Y = _raw(xs, tr["x"]), _raw(ys, tr["y"])
    if band:
        ax.fill_between(X, _raw(ys, tr["lo"]), _raw(ys, tr["hi"]), color=st["color"], alpha=0.14, lw=0, zorder=z - 1)
    ax.plot(X, Y, color=st["color"], lw=st.get("lw", 2.8), ls=st.get("ls", "-"), marker=st.get("marker", None), ms=7, mec="k", mew=0.6, zorder=z,
            path_effects=[withStroke(linewidth=st.get("lw", 2.8) + 1.8, foreground="white")])


def _rows_of(rows, kind, pan, sample, tr, ng, xs, ys):
    for _, r in tr.iterrows():
        rows.append(dict(kind=kind, panel=pan, sample=sample, n_gal_group=ng, x=float(_raw(xs, r["x"])), y=float(_raw(ys, r["y"])),
                         y_lo16=float(_raw(ys, r["lo"])), y_hi84=float(_raw(ys, r["hi"])), n_gal=int(r["n"])))


def draw_obs(ax, xs, ys):
    """The observed points on one plane (colour = A_V, arrows = limits)."""
    for _, r in OBS.iterrows():
        if not (np.isfinite(r[xs["col"]]) and np.isfinite(r[ys["col"]])):
            continue
        x, y = float(_raw(xs, r[xs["col"]])), float(_raw(ys, r[ys["col"]]))
        ax.scatter(x, y, s=r["size"], marker=r["marker"], c=[_cof(r[AXES[OBS_COLOUR]["raw"]])], edgecolors=r["edge"], linewidths=1.4, zorder=6 + (r["marker"] == "*"))
        if ys.get("cens") and r[ys["cens"]] == "ul":
            ax.annotate("", (x, y), xytext=(0, -15), textcoords="offset points", arrowprops=dict(arrowstyle="<|-", color=r["edge"], lw=1.2, mutation_scale=9), zorder=8)
        if xs.get("cens") and r[xs["cens"]] == "ul":
            ax.annotate("", (x, y), xytext=(-15, 0), textcoords="offset points", arrowprops=dict(arrowstyle="<|-", color=r["edge"], lw=1.2, mutation_scale=9), zorder=8)


def ism_panel(ax, xkey, ykey, groups, rows, tag, xlab=True, ylab=True, nbins=NBIN_X, nmin=NMIN_X):
    """One plane: the model tracks of `groups` (name, frame, style; style key `points` also draws the frame's galaxies), the observed points -> {name: n_gal}."""
    xs, ys = AXES[xkey], AXES[ykey]
    pan = f"{xkey}_{ykey}"
    drawn, tracks = {}, {}
    ax.set_xscale("log" if xs["log"] else "linear"); ax.set_yscale("log" if ys["log"] else "linear")
    ax.set_xlim(*xs["lim"]); ax.set_ylim(*ys["lim"])
    for name, frame, st in groups:
        ok = np.isfinite(frame[xs["col"]]) & np.isfinite(frame[ys["col"]])
        if st.get("points"):
            ax.scatter(_raw(xs, frame.loc[ok, xs["col"]]), _raw(ys, frame.loc[ok, ys["col"]]), s=22, marker=st.get("marker", "o") or "o", c=[st["color"]], edgecolors="k", linewidths=0.4,
                       alpha=0.75, zorder=3)
        tr = run_track(frame[xs["col"]], frame[ys["col"]], nbins=nbins, nmin=nmin)
        rho = spearmanr(frame.loc[ok, xs["col"]], frame.loc[ok, ys["col"]]) if ok.sum() >= 5 else (np.nan, np.nan)
        print(f"  {pan:14s} {name:24s} {int(ok.sum()):4d} galaxies, {len(tr)} track points" + (f"; {ys['short']} {_raw(ys, tr['y'].min()):.3g}–{_raw(ys, tr['y'].max()):.3g} over "
              f"{xs['short']} {_raw(xs, tr['x'].min()):.3g}–{_raw(xs, tr['x'].max()):.3g}" if len(tr) else "") + f"; Spearman = {rho[0]:+.2f} (p = {rho[1]:.2g})")
        drawn[name] = int(ok.sum())
        if len(tr):
            tracks[name] = tr
            _draw_track(ax, tr, xs, ys, st, band=st.get("band", True))
            _rows_of(rows, "model_track", pan, name, tr, int(ok.sum()), xs, ys)
    draw_obs(ax, xs, ys)
    if ys.get("cens"):    # the check: detected values against each track interpolated at their x (inside the track's range)
        for name, tr in tracks.items():
            O = OBS[(OBS[ys["cens"]] == "det") & np.isfinite(OBS[xs["col"]]) & np.isfinite(OBS[ys["col"]]) & OBS[xs["col"]].between(tr["x"].min(), tr["x"].max())]
            if len(O) >= 3:
                d = O[ys["col"]].to_numpy(float) - np.interp(O[xs["col"]].to_numpy(float), tr["x"].to_numpy(float), tr["y"].to_numpy(float))
                print(f"    measured − {name} track over the {len(O)} detections inside its {xs['short']} range: median {np.median(d):+.2f} dex, 16–84 % [{np.percentile(d, 16):+.2f}, {np.percentile(d, 84):+.2f}]")
                for _, o in O.iterrows():
                    yt = float(np.interp(o[xs["col"]], tr["x"], tr["y"]))
                    rows.append(dict(kind="obs_vs_track", panel=pan, sample=o["sample"], id=o["id"], track=name, x=float(_raw(xs, o[xs["col"]])), y=float(_raw(ys, o[ys["col"]])),
                                     y_track=float(_raw(ys, yt)), offset_dex=float(o[ys["col"]] - yt)))
    ax.set_xlabel(xs["label"] if xlab else "", fontsize=FONT["label"]); ax.set_ylabel(ys["label"] if ylab else "", fontsize=FONT["label"])
    ax.tick_params(direction="in", top=True, right=True, which="both", labelsize=FONT["tick"], labelbottom=xlab, labelleft=ylab); ax.grid(False)
    for axis, spec in ((ax.yaxis, ys), (ax.xaxis, xs)):
        if spec["log"] and spec["lim"][0] >= 0.1 and spec["lim"][1] < 1e4:
            axis.set_major_formatter(FuncFormatter(lambda v, p: f"{v:g}"))
    if tag:
        ax.set_title(tag, loc="left", fontsize=FONT["tag"], pad=6)
    return drawn


def ism_figure(stem, groups, model_title, extra_notes=(), nbins=NBIN_X, nmin=NMIN_X):
    """The 2 x 3 grid + the legend strip (observed | models | definitions) + the A_V colour bar -> paper_xray_<stem>.{png,pdf,csv,_points.csv};
    returns (what was drawn per panel, the CSV rows: model_track / obs_vs_track)."""
    print(f"\n── {stem}: " + " | ".join(", ".join(f"{AXES[y]['short']} vs {AXES[x]['short']}" for x, y in row) for row in GRID) + " ──")
    rows, nr, nc = [], len(GRID), max(len(r) for r in GRID)
    fig = plt.figure(figsize=(6.0 * nc + 1.6, 4.8 * nr + 2.9))
    gs = GridSpec(nr + 1, nc, height_ratios=[1.0] * nr + [0.46], hspace=0.10, wspace=0.08, figure=fig)
    axes = [[fig.add_subplot(gs[i, j]) for j in range(len(GRID[i]))] for i in range(nr)]
    lax, lax2, lax3 = fig.add_subplot(gs[nr, 0]), fig.add_subplot(gs[nr, 1]), fig.add_subplot(gs[nr, 2:])
    for a in (lax, lax2, lax3):
        a.axis("off")
    drawn, k = [], 0
    for i, row in enumerate(GRID):
        for j, (x, y) in enumerate(row):
            drawn.append(ism_panel(axes[i][j], x, y, groups, rows, f"({'abcdefghij'[k]})", xlab=(i == nr - 1), ylab=(j == 0), nbins=nbins, nmin=nmin)); k += 1
    flat = [a for row in axes for a in row]
    sm = plt.cm.ScalarMappable(cmap=CM, norm=NORM); sm.set_array([])
    cb = fig.colorbar(sm, ax=flat, pad=0.012, extend="max", fraction=0.02, aspect=40)
    cb.set_label(AXES[OBS_COLOUR]["label"] + " of the observed galaxies", fontsize=FONT["label"]); cb.ax.tick_params(labelsize=FONT["tick"])
    nq, ns = int((OBS["sample"] == "ALMA-C11").sum()), int((OBS["sample"] == "Spilker+18").sum())
    h_obs = [(Line2D([], [], marker="o", ls="", ms=10, mfc=_cof(np.mean(OBS_CLIM)), mec=C_OBS_EDGE, mew=1.3), rf"ALMA-C11 QGs, $z\approx0.4$ (N={nq}): colour = $A_V$ (fiducial CIGALE)" + "\narrow = upper limit (dust, CO, sSFR)"),
             (Line2D([], [], marker="D", ls="", ms=7.5, mfc=_cof(0.3), mec="0.2", mew=1.3), rf"Spilker+18 LEGA-C passive, $z\approx0.7$ (N={ns}): CO only"),
             (Line2D([], [], marker="*", ls="", ms=15, mfc=_cof(U["av"]), mec="k", mew=1.2), rf"{U['name']} ({U['ref']}), $z = {U['z']:.2f}$: $A_V$ = {U['av']:g} (FAST++)")]
    if OBS_CTRL:
        h_obs.insert(1, (Line2D([], [], marker="o", ls="", ms=6.5, mfc=_cof(0.15), mec="0.35", mew=1.2), "ALMA-C11 controls (no ALMA detection: limits only)"))
    ng = {name: max(d.get(name, 0) for d in drawn) for name, _, _ in groups}
    h_mod = [(Line2D([], [], color=st["color"], lw=st.get("lw", 2.8), ls=st.get("ls", "-"), marker=st.get("marker", None), ms=7, mec="k", mew=0.6), f"{st['label']} (N={ng[name]})")
             for name, _, st in groups if ng[name]]
    if any(st.get("points") for _, _, st in groups):
        h_mod.append((Line2D([], [], marker="o", ls="", ms=5.5, mfc="0.5", mec="k", mew=0.4), "individual galaxies of the class (small classes: points only)"))
    leg1 = lax.legend([h for h, _ in h_obs], [l for _, l in h_obs], loc="upper left", bbox_to_anchor=(-0.04, 0.80), frameon=False, fontsize=FONT["legend"], handlelength=1.6,
                      title="observed", title_fontsize=FONT["legend"] + 1)
    leg1._legend_box.align = "left"
    leg2 = lax2.legend([h for h, _ in h_mod], [l for _, l in h_mod], loc="upper left", bbox_to_anchor=(-0.02, 0.80), frameon=False, fontsize=FONT["legend"], handlelength=2.2,
                       title=model_title, title_fontsize=FONT["legend"] + 0.5)
    leg2._legend_box.align = "left"
    keys_used = list(dict.fromkeys([x for row in GRID for x, y in row] + [y for row in GRID for x, y in row]))
    notes = [f"{AXES[kk].get('nm', AXES[kk]['short'])}: {AXES[kk]['obs_note']}" for kk in keys_used if AXES[kk].get("obs_note")]
    notes.append(f"catalogues: whole-galaxy caesar quantities, one row per galaxy,\n   <= {nbins} equal-count bins of >= {nmin} galaxies (fewer than {2 * nmin}: 4 bins of >= {min(nmin, 5)})")
    notes += list(extra_notes)
    lax3.text(0.02, 0.80, wrap_caption("\n".join(["definitions"] + notes), width=int(11 * fig.get_figwidth() * 0.42)), transform=lax3.transAxes, fontsize=FONT["note"], color="0.35", ha="left", va="top", linespacing=1.45)
    paper_save(fig, stem)
    plt.show()
    pd.DataFrame(rows).to_csv(os.path.join(PAPERDIR, f"paper_xray_{stem}.csv"), index=False)
    OBS.to_csv(os.path.join(PAPERDIR, f"paper_xray_{stem}_points.csv"), index=False)
    print(f"  tables -> {os.path.join(PAPERDIR, 'paper_xray_' + stem)}{{.csv,_points.csv}}")
    return drawn, pd.DataFrame(rows)


def offset_summary(rows, groups):
    """Median (measured - track) over the dust and the gas panels per track -> the scorecard printed after each figure."""
    o = rows[rows["kind"] == "obs_vs_track"] if len(rows) and "kind" in rows else rows.iloc[:0]
    print(f"  {'track':34s} {'dust offset':>12s} {'n':>3s} {'gas offset':>11s} {'n':>3s}   (median measured − track [dex] over the detections inside the track's x range, the 3 panels of each row pooled)")
    for name, _, st in groups:
        d = o[(o["track"] == name) & o["panel"].str.endswith("_fdust")]["offset_dex"]; g = o[(o["track"] == name) & o["panel"].str.endswith("_fh2")]["offset_dex"]
        if len(d) or len(g):
            print(f"  {name:34s} {d.median() if len(d) else np.nan:+12.2f} {len(d):3d} {g.median() if len(g) else np.nan:+11.2f} {len(g):3d}")


_notes5b = ["; ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES) + f" (the {XTITLE} classes of Part 0)"]
SEL_TXT2 = (f"log $M_\\star$ > {MASS_MIN:g}, passive (sSFR < {PASSIVE_FACTOR:g} / $t_H$),\n>= {NGAS_MIN} gas and >= {NSTAR_MIN} star particles, "
            f"$M_{{\\rm dust}} \\geq 10^{{{np.log10(DUST_TO_H2_MIN):.0f}}} M_{{\\rm H_2}}$")
G5_DRAWN, G5_ROWS = {}, {}
for box in P9_BOXES:
    g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
    groups = [(f"{box} {c}", g[g["xray_class"] == c], dict(color=XCOL[c], ls="-", marker=G5_BOX_MARKER.get(box, "o"), lw=2.8, points=int((g["xray_class"] == c).sum()) <= G5_POINTS_MAX,
                                                           label=f"{XSHORT[c]}: {XNAME[c]}")) for c in XCLASSES]
    groups.append((f"{box} all", g, dict(color="0.15", ls=(0, (1.2, 1.6)), lw=2.0, band=False, label="all classified quenched galaxies")))
    if (QM["box"] == NOX_BOX).sum() >= NMIN_X:                                # the channel-off run as a comparator track in each per-box grid
        groups.append((f"{NOX_BOX} all", QM[QM["box"] == NOX_BOX], dict(color=BOXES[NOX_BOX]["color"], ls="--", marker="^", lw=2.2, band=False,
                                                                        label=f"{BOX_SHORT.get(NOX_BOX, NOX_BOX)} all quenched (X-ray channel off)")))
    G5_DRAWN[box], G5_ROWS[box] = ism_figure(f"ism_grid_{box}", groups, f"SIMBA {BOXES[box]['label']}, catalogue, z = 0.3–2:\n{SEL_TXT2}\nrunning median per X-ray class (band = bootstrap 16–84 %)",
                                             extra_notes=_notes5b)
    offset_summary(G5_ROWS[box], groups)
# every box on one grid: colour = class, marker = box (no points)
groups = [(f"{box} {c}", QM[(QM["box"] == box) & (QM["xray_class"] == c)], dict(color=XCOL[c], ls="-", marker=G5_BOX_MARKER.get(box, "o"), lw=2.4, label=f"{BOX_SHORT[box]} {XSHORT[c]}"))
          for c in XCLASSES for box in P9_BOXES]
if (QM["box"] == NOX_BOX).sum() >= NMIN_X:
    groups.append((f"{NOX_BOX} all", QM[QM["box"] == NOX_BOX], dict(color=BOXES[NOX_BOX]["color"], ls="--", marker="^", lw=2.4,
                                                                    label=f"{BOX_SHORT.get(NOX_BOX, NOX_BOX)} all quenched (X-ray channel off)")))
G5_DRAWN["boxes"], G5_ROWS["boxes"] = ism_figure("ism_grid_boxes", groups, f"SIMBA catalogues ({', '.join(BOX_SHORT[b] for b in P9_BOXES)}), z = 0.3–2:\n{SEL_TXT2}\nrunning median per X-ray class and box (band = bootstrap 16–84 %)",
                                                 extra_notes=_notes5b)
offset_summary(G5_ROWS["boxes"], groups)
```

```python
# ── Section 5 (continued) — the normalised distributions of the dust content per class, and who populates the tail above the threshold: the medians of every class can sit
#    below it while the few galaxies above it belong disproportionately to one class — and that class need not be the one that dominates the box ──
from scipy.stats import hypergeom
from matplotlib.colors import to_rgb

P5_QTY  = [("lfd", r"$\log\,M_{\rm dust}\,/\,M_\star$", -4.0), ("ldg", r"$\log\,M_{\rm dust}\,/\,M_{\rm gas}$", -2.5)]   # (column, label, the tail threshold drawn and tested in the figure)
P5_SCAN = {"lfd": [-4.5, -4.0, -3.5, -3.0], "ldg": [-3.5, -3.0, -2.5, -2.0]}                                           # every threshold of the printed / saved composition table
P5_BIN  = 0.25                      # histogram bin width [dex]; every class normalised to unit area (the zero-dust pile included: the bars at the floor)
P5_HMIN = 2 * P9_MIN                # a class below this many galaxies in a box is drawn as individual marks, not a histogram
P5_PILE = 0.01                      # the zero-dust pile (galaxies at or below the floor) is drawn apart at the floor only when some class has at least this fraction there


def tail_composition(g, col, cut):
    """The classes above `cut` in `col`: per class N, share of the box, N above, share of the tail, fraction of the class above, and the hypergeometric probability of
    the tail count under random draws from the box (P(>= k) when the class is over-represented in the tail, P(<= k) when under-represented) -> (DataFrame, N_tail)."""
    v = g[col].to_numpy(float); t = np.isfinite(v) & (v >= cut); nt = int(t.sum()); rows = []
    for c in XCLASSES:
        m = (g["xray_class"] == c).to_numpy(); n, k = int(m.sum()), int((m & t).sum())
        if n == 0:
            continue
        sa, st = n / len(g), (k / nt if nt else np.nan)
        over = bool(nt) and st >= sa
        p = (float(hypergeom.sf(k - 1, len(g), n, nt)) if over else float(hypergeom.cdf(k, len(g), n, nt))) if nt else np.nan
        rows.append(dict(cls=c, n=n, share_all=sa, n_tail=k, share_tail=st, frac_above=k / n, side="over" if over else "under", p=p))
    return pd.DataFrame(rows), nt


P5_ROWS = []
print(f"who populates the dusty tail (log M* > {MASS_MIN:g}, classified galaxies): per box and threshold, the share of every class among the galaxies above it against its share of "
      f"the box, the fraction of the class above it, and P = the hypergeometric probability of the tail count under random draws from the box (over- / under-represented)")
for box in P9_BOXES:
    g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
    for col, lab, _ in P5_QTY:
        for cut in P5_SCAN[col]:
            T, nt = tail_composition(g, col, cut)
            for r in T.itertuples(index=False):
                P5_ROWS.append(dict(box=box, qty=col, cut=cut, n_box=len(g), n_box_tail=nt, **r._asdict()))
            print(f"  {box:7s} {col} >= {cut:5.1f}: {nt:4d} of {len(g):4d} ({100 * nt / len(g):4.1f} %) | " + " | ".join(
                f"{XABBR[r.cls]} {int(r.n_tail):3d} = {100 * r.share_tail:3.0f} % of the tail vs {100 * r.share_all:3.0f} % of the box "
                f"({100 * r.frac_above:3.0f} % of the class above; P {r.p:.1g} {r.side})" for r in T.itertuples()))
for vref in _dg_ref:                                                        # the runs without the X-ray channel: no classes, but the same thresholds
    gv = QM[QM["box"] == vref]
    for col, lab, _ in P5_QTY:
        vv = gv[col].to_numpy(float); okv = np.isfinite(vv)
        if okv.sum():
            print(f"  {vref:8s} {col}: " + " | ".join(f">= {cut:g}: {100 * float((vv[okv] >= cut).mean()):4.1f} %" for cut in P5_SCAN[col])
                  + f"  of {int(okv.sum())} quenched (X-ray channel off)")
pd.DataFrame(P5_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_dust_tail_composition.csv"), index=False)
print(f"  -> {os.path.join(PAPERDIR, 'paper_xray_dust_tail_composition.csv')}")

# ── the figure: per box a column, per quantity a row; every class as a unit-area histogram (the zero-dust pile as bars at the floor, side by side), the all-classified
#    distribution dotted, the threshold dashed, the no-X-ray reference medians; the inset = the class shares of the whole box against the shares above the threshold ──
if len(P9_BOXES):
    _nb = len(P9_BOXES)
    fig, axes = plt.subplots(len(P5_QTY), _nb, figsize=(6.4 * _nb + 0.8, 5.0 * len(P5_QTY)), squeeze=False, gridspec_kw=dict(hspace=0.36, wspace=0.18))
    for r_, (col, lab, cut) in enumerate(P5_QTY):
        for i, box in enumerate(P9_BOXES):
            ax = axes[r_][i]
            g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES)]
            v = g[col].to_numpy(float); ok = np.isfinite(v)
            present = [c for c in XCLASSES if (g["xray_class"] == c).any()]
            pile = any(float((v[ok & (g["xray_class"] == c).to_numpy()] <= P9_FLOOR).mean()) >= P5_PILE for c in present)   # a class with a zero-dust pile worth drawing apart
            zero = ok & (v <= P9_FLOOR) if pile else np.zeros(len(v), bool)                 # else the few galaxies at / below the floor are binned with the rest
            vmin, vmax = (float(np.min(v[ok & ~zero])), float(np.max(v[ok & ~zero]))) if (ok & ~zero).any() else (cut - 1.0, cut + 1.0)
            edges = np.arange(np.floor(vmin / P5_BIN) * P5_BIN, np.ceil(vmax / P5_BIN) * P5_BIN + P5_BIN + 1e-9, P5_BIN)
            ymax = 0.0
            h_all, _ = np.histogram(v[ok & ~zero], bins=edges); d_all = h_all / (ok.sum() * P5_BIN)
            ax.stairs(d_all, edges, color="0.15", lw=1.8, ls=(0, (1.2, 1.4)), zorder=5, label=f"all classified (N = {int(ok.sum())}, {100 * float((v[ok] >= cut).mean()):.0f} % above {cut:g})")
            ymax = max(ymax, float(d_all.max()) if len(d_all) else 0.0)
            for j, c in enumerate(present):
                m = (g["xray_class"] == c).to_numpy() & ok; n = int(m.sum())
                fa, fz = float((v[m] >= cut).mean()), float(zero[m].sum() / n)
                if n >= P5_HMIN:
                    h, _ = np.histogram(v[m & ~zero], bins=edges); d = h / (n * P5_BIN)
                    ax.stairs(d, edges, color=XCOL[c], lw=2.4, zorder=4, label=f"{XSHORT[c]} (N = {n}, {100 * fa:.0f} % above {cut:g}" + (f", {100 * fz:.0f} % zero dust)" if pile else ")"))
                    ax.stairs(d, edges, color=XCOL[c], fill=True, alpha=0.12, lw=0, zorder=2)
                    ymax = max(ymax, float(d.max()))
                    if pile and fz > 0:                                                     # the zero-dust pile per class, side by side at the floor, on the same density scale
                        xz = P9_FLOOR + (j - (len(present) - 1) / 2) * 0.13
                        ax.bar(xz, fz / P5_BIN, width=0.12, color=XCOL[c], alpha=0.8, lw=0, zorder=3)
                        ymax = max(ymax, fz / P5_BIN)
                else:
                    ax.plot(v[m], np.zeros(n), ls="", marker="|", ms=15, mew=2.2, color=XCOL[c], zorder=6, clip_on=False,
                            label=f"{XSHORT[c]} (N = {n}, individual galaxies, {100 * fa:.0f} % above {cut:g})")
            for vref, (dg, fd, nv) in _dg_ref.items():                                  # the channel-off runs: the full unit-area distribution, not only the median
                gv = QM[QM["box"] == vref]; vv = gv[col].to_numpy(float); okv = np.isfinite(vv)
                if okv.sum() >= P5_HMIN:
                    hv, _ = np.histogram(vv[okv], bins=edges); dv = hv / (okv.sum() * P5_BIN)
                    ax.stairs(dv, edges, color=BOXES[vref]["color"], lw=1.8, ls=(0, (4.0, 1.6)), zorder=3,
                              label=f"{BOX_SHORT.get(vref, vref)}, X-ray channel off (N = {int(okv.sum())}, {100 * float((vv[okv] >= cut).mean()):.0f} % above {cut:g})")
                    ymax = max(ymax, float(dv.max()) if len(dv) else 0.0)
                else:
                    ax.axvline(fd if col == "lfd" else dg, color=BOXES[vref]["color"], lw=1.4, ls=(0, (4.0, 1.6)), zorder=1)      # too few: the median line only
            ax.axvline(cut, color="0.2", lw=1.5, ls="--", zorder=5)
            ax.text(cut + 0.04, 0.02, f"threshold {cut:g}", transform=ax.get_xaxis_transform(), fontsize=FONT["note"] - 0.5, color="0.3", ha="left", va="bottom", rotation=90, zorder=7)
            if pile:
                ax.text(P9_FLOOR + 0.05, 0.02, "zero dust (floor)", transform=ax.get_xaxis_transform(), fontsize=FONT["note"] - 0.8, color="0.45", ha="left", va="bottom", rotation=90, zorder=7)
            ax.set_xlim((P9_FLOOR - 0.5) if pile else (edges[0] - 0.2), edges[-1] + 0.15); ax.set_ylim(0, max(ymax, 1e-3) * 2.6)
            ax.set_xlabel(lab); ax.set_ylabel("probability density  [per dex]" if i == 0 else "")
            ax.tick_params(direction="in", top=True, right=True, labelsize=FONT["tick"]); ax.grid(False)
            ax.set_title(f"({P9_TAGS[r_ * _nb + i]}) {BOXES[box]['label'].split(' (')[0]}: {lab} per class, unit area", loc="left", fontsize=FONT["tag"] - 0.8)
            ax.legend(loc="upper left", frameon=False, fontsize=FONT["legend"] - 1.5, handlelength=1.6, borderaxespad=0.3)
            # the inset: the class shares of the whole box (top bar) against the shares among the galaxies above the threshold (bottom bar)
            T, nt = tail_composition(g, col, cut)
            ins = ax.inset_axes([0.60, 0.57, 0.37, 0.17])
            for yy, key, ylab in ((1, "share_all", f"whole box (N = {len(g)})"), (0, "share_tail", f"$\\geq$ {cut:g} (N = {nt})")):
                left = 0.0
                for r in T.itertuples():
                    w = getattr(r, key)
                    if np.isfinite(w) and w > 0:
                        ins.barh(yy, w, left=left, height=0.74, color=XCOL[r.cls], edgecolor="white", lw=0.6)
                        if w >= 0.09:
                            _lum = np.dot(to_rgb(XCOL[r.cls]), (0.299, 0.587, 0.114))
                            ins.text(left + w / 2, yy, f"{100 * w:.0f}", ha="center", va="center", fontsize=FONT["note"] - 1.2, fontweight="bold", color="white" if _lum < 0.45 else "k")
                        left += w
            ins.set_xlim(0, 1); ins.set_ylim(-0.55, 1.55); ins.set_xticks([]); ins.set_yticks([1, 0])
            ins.set_yticklabels([f"whole box (N = {len(g)})", f"above {cut:g} (N = {nt})"], fontsize=FONT["note"] - 1.0); ins.tick_params(length=0)
            for s in ins.spines.values():
                s.set_visible(False)
            ins.set_title("class shares [%]", fontsize=FONT["note"] - 0.5, pad=2, backgroundcolor="white"); ins.set_facecolor("white"); ins.patch.set_alpha(0.9)
            lines = [f"{XABBR[r.cls]}: {100 * r.share_tail:.0f} % of the tail vs {100 * r.share_all:.0f} % of the box ({r.side}, P = {r.p:.1g})"
                     for r in T.itertuples() if r.cls != XREF] if nt else ["no galaxy above the threshold"]
            ax.text(0.98, 0.545, "\n".join(lines), transform=ax.transAxes, ha="right", va="top", fontsize=FONT["note"] - 0.7, color="0.25", zorder=6, linespacing=1.3)
    fig.suptitle(wrap_caption(f"the dusty tail per class: every class normalised to unit area ({P5_BIN:g} dex bins; the zero-dust galaxies as the bars at the floor, {P9_FLOOR:g}), "
                              f"the tail threshold dashed, the 50 Mpc runs without the X-ray channel as the coloured dashed outline distributions ({', '.join(f'{BOX_SHORT.get(vref, vref)} N = {nv}' for vref, (_, _, nv) in _dg_ref.items())}); the inset compares the class shares of the whole box with the shares among the galaxies above the threshold "
                              f"(P = hypergeometric probability of the tail count under random draws from the box) — the medians can all sit below the threshold while the tail belongs to one class; "
                              f"quenched galaxies, {SEL_TXT}; classes = {XTITLE}: " + ", ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES), width=int(13 * fig.get_figwidth())),
                 fontsize=FONT["note"] + 1, y=0.995)
    fig.subplots_adjust(top=0.90, bottom=0.07)
    paper_save(fig, "dust_distributions")
    plt.show()
```

## Section 5c — the coupling class $\times$ the X-ray class: is the dusty tail a combination of the two divisions?  
The weak / strong division alone leaves both classes on the dusty side (§5): here it is crossed with the onset classes — the cells of the combined division against the tail, the stratified tests (does the coupling add separation at fixed onset timing, and the reverse?), the AUC of each dose / timing quantity, and the tail split by stellar age at 2 Gyr (the young under-dosed tail against the old fully-dosed one).

```python
# ── Section 5c — the AGN-coupling class (weak / strong of B Part 6a) crossed with the X-ray class on the dusty tail: the cells of the combined division, the
#    stratified tests (does the coupling add separation at fixed onset timing, and the reverse?), the dose behind the tail, and the tail split by stellar age ──
from scipy.stats import fisher_exact, chi2_contingency
from scipy.stats import hypergeom as _hg
from matplotlib.patches import Patch

P5C_TAIL = -4.0                     # the dusty tail on log M_dust/M* (the P5 threshold the observed detections sit above)
P5C_COUP = ["weak", "strong"]       # the two-class coupling division (B Part 6a: w_pre < 0.3 -> weak, >= 0.3 -> strong; no_AGN dropped here)
P5C_AGE  = 2.0                      # the tail split into young / old at this mass-weighted age [Gyr]
P5C_DOSE = [("e_x", "$E_x$"), ("f_x", "$f_x$"), ("t_gaspoor", "$T_{\\rm gas-poor}$"), ("dt_onset_sft", "$t_{\\rm onset} - t_{\\rm SFT}$"), ("w_pre", "$w_{\\rm pre}$"), ("age", "age")]
P5C_HATCH = {"weak": "///", "strong": None}


def wilson(k, n, z=1.0):
    """Binomial fraction k/n with its Wilson interval at z sigma -> (p, lo, hi)."""
    if n == 0:
        return np.nan, np.nan, np.nan
    p = k / n; d = 1 + z * z / n; c = (p + z * z / (2 * n)) / d; h = z * np.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / d
    return p, c - h, c + h


def auc_of(x, pos):
    """AUC of x as a score for `pos` against the rest (Mann-Whitney U / n1 n0; > 0.5: a higher x favours `pos`) -> (auc, p)."""
    x, s = np.asarray(x, float), np.asarray(pos, bool); ok = np.isfinite(x)
    a, b = x[ok & s], x[ok & ~s]
    if min(len(a), len(b)) < 2:
        return np.nan, np.nan
    r = mannwhitneyu(a, b)
    return float(r.statistic / (len(a) * len(b))), float(r.pvalue)


def _g5c(box):
    g = QM[(QM["box"] == box) & QM["xray_class"].isin(XCLASSES) & QM["agn_class"].isin(P5C_COUP)].copy()
    g["dusty"] = g["lfd"] >= P5C_TAIL
    return g


P5C_ROWS = []
print(f"the coupling class x the X-ray class ({XSCHEME}) on the dusty tail (log M_dust/M* >= {P5C_TAIL:g}); log M* > {MASS_MIN:g}, classified galaxies with a finite w_pre")
for box in P9_BOXES:
    g = _g5c(box)
    _nall = int(((QM["box"] == box) & QM["xray_class"].isin(XCLASSES)).sum())
    nt = int(g["dusty"].sum())
    tab = pd.crosstab(g["agn_class"], g["xray_class"]).reindex(index=P5C_COUP, columns=XCLASSES, fill_value=0)
    t = tab.to_numpy(); t = t[t.sum(1) > 0][:, t.sum(0) > 0]
    chi2p, v = np.nan, np.nan
    if min(t.shape) >= 2:
        _chi = chi2_contingency(t); chi2p = float(_chi[1]); v = float(np.sqrt(_chi[0] / (t.sum() * (min(t.shape) - 1))))
    print(f"\n{box}: {len(g)} galaxies with both classes ({_nall - len(g)} classified without a coupling class: no BH history), {nt} in the tail ({100 * nt / len(g):.1f} %); "
          f"coupling x onset chi-square p = {chi2p:.2g}, Cramér's V = {v:.2f}")
    print("  " + tab.to_string().replace("\n", "\n  "))
    print(f"  {'cell':>18s} {'N':>5s} {'in tail':>7s} {'% dusty [68 %]':>21s} {'med lfd':>8s} {'med ldg':>8s} {'med age':>8s} {'med E_x':>8s} {'tail vs box share':>19s} {'P':>8s}")
    for oc in XCLASSES:
        for cc in P5C_COUP:
            h = g[(g["xray_class"] == oc) & (g["agn_class"] == cc)]
            if not len(h):
                continue
            k = int(h["dusty"].sum()); p_, lo, hi = wilson(k, len(h))
            sa, st = len(h) / len(g), (k / nt if nt else np.nan)
            over = bool(nt) and st >= sa
            ph = (float(_hg.sf(k - 1, len(g), len(h), nt)) if over else float(_hg.cdf(k, len(g), len(h), nt))) if nt else np.nan
            P5C_ROWS.append(dict(box=box, onset=oc, coupling=cc, n=len(h), n_tail=k, frac=p_, frac_lo=lo, frac_hi=hi, lfd=h["lfd"].median(), ldg=h["ldg"].median(),
                                 age=h["age"].median(), e_x=h["e_x"].median(), share_tail=st, share_all=sa, side="over" if over else "under", p_hyper=ph))
            print(f"  {XABBR[oc] + ' x ' + cc:>18s} {len(h):5d} {k:7d} {100 * p_:7.1f} [{100 * lo:5.1f}, {100 * hi:5.1f}] {h['lfd'].median():8.2f} {h['ldg'].median():8.2f} "
                  f"{h['age'].median():8.2f} {h['e_x'].median():8.2f} {100 * st:7.1f} vs {100 * sa:4.1f} % {'over' if over else 'under':>5s} {ph:8.1g}")
    for oc in XCLASSES:                                   # the coupling at fixed onset timing
        h = g[g["xray_class"] == oc]
        a, b = (h.loc[h["agn_class"] == c, "lfd"].dropna() for c in P5C_COUP)
        if min(len(a), len(b)) < 3:
            continue
        t2 = np.array([[int((h["dusty"] & (h["agn_class"] == c)).sum()), int((~h["dusty"] & (h["agn_class"] == c)).sum())] for c in P5C_COUP])
        odds, pf = fisher_exact(t2)
        print(f"  inside {XABBR[oc]:>7s}: weak vs strong — MW on lfd p = {mannwhitneyu(a, b).pvalue:.2g}; Fisher on the tail p = {pf:.2g}, odds(weak) = {odds:.2f}")
    for cc in P5C_COUP:                                   # the onset timing at fixed coupling
        h = g[g["agn_class"] == cc]
        a, b = h.loc[h["xray_class"] == XREF, "lfd"].dropna(), h.loc[h["xray_class"] != XREF, "lfd"].dropna()
        if min(len(a), len(b)) < 3:
            continue
        print(f"  inside {cc:>7s}: {XABBR[XREF]} ({100 * h.loc[h['xray_class'] == XREF, 'dusty'].mean():.0f} % dusty, N = {len(a)}) vs the rest "
              f"({100 * h.loc[h['xray_class'] != XREF, 'dusty'].mean():.0f} %, N = {len(b)}) — MW on lfd p = {mannwhitneyu(a, b).pvalue:.2g}")
    parts = []
    for c, lab in P5C_DOSE:
        auc, p = auc_of(g[c], g["dusty"])
        parts.append(f"{lab} {auc:.2f} ({p:.1g})")
    print("  AUC of each quantity for the tail (> 0.5: higher favours the tail): " + "; ".join(parts))
    d = g[g["dusty"]]
    for lab, m in ((f"age >= {P5C_AGE:g}", d["age"] >= P5C_AGE), (f"age < {P5C_AGE:g}", d["age"] < P5C_AGE)):
        h = d[m]
        if len(h) < 3:
            continue
        print(f"  the tail, {lab:10s}: N = {len(h):4d} — onset " + str({XABBR[c]: int((h["xray_class"] == c).sum()) for c in XCLASSES})
              + ", coupling " + str({c: int((h["agn_class"] == c).sum()) for c in P5C_COUP})
              + f"; med E_x {h['e_x'].median():.2f}, f_x {h['f_x'].median():.2f}, lfd {h['lfd'].median():+.2f}, z {h['z'].median():.2f}")
pd.DataFrame(P5C_ROWS).to_csv(os.path.join(PAPERDIR, "paper_xray_coupling_onset.csv"), index=False)
print(f"\n-> {os.path.join(PAPERDIR, 'paper_xray_coupling_onset.csv')}")

# ── the figure: (top) the tail fraction of every coupling x onset cell; (bottom) the dose-age plane — the dusty galaxies against the dose typical of their age ──
if len(P9_BOXES):
    _nb = len(P9_BOXES)
    fig, axes = plt.subplots(2, _nb, figsize=(6.4 * _nb + 0.8, 10.4), squeeze=False, gridspec_kw=dict(hspace=0.30, wspace=0.20))
    for i, box in enumerate(P9_BOXES):
        g = _g5c(box)
        ax = axes[0][i]
        for jo, oc in enumerate(XCLASSES):
            for jc, cc in enumerate(P5C_COUP):
                h = g[(g["xray_class"] == oc) & (g["agn_class"] == cc)]
                if not len(h):
                    continue
                p_, lo, hi = wilson(int(h["dusty"].sum()), len(h))
                xx = jo + (jc - 0.5) * 0.38
                ax.bar(xx, 100 * p_, width=0.34, color=XCOL[oc], alpha=0.45 if cc == "weak" else 1.0, hatch=P5C_HATCH[cc], edgecolor="k", lw=0.7,
                       yerr=[[100 * (p_ - lo)], [100 * (hi - p_)]], capsize=2.5, error_kw=dict(ecolor="0.15", lw=1.0))
                ax.text(xx, 1.5, f"N = {len(h)}", ha="center", va="bottom", rotation=90, fontsize=FONT["note"] - 0.5, color="k", zorder=6)
        ax.axhline(100 * g["dusty"].mean(), color="0.2", lw=1.5, ls=(0, (1.2, 1.4)), zorder=1)
        hnd = [Patch(facecolor="0.65", alpha=0.45, hatch=P5C_HATCH["weak"], edgecolor="k", label=r"weak ($w_{\rm pre} < 0.3$)"),
               Patch(facecolor="0.65", edgecolor="k", label=r"strong ($w_{\rm pre} \geq 0.3$)"),
               Line2D([], [], color="0.2", lw=1.5, ls=(0, (1.2, 1.4)), label=f"all classified ({100 * g['dusty'].mean():.0f} %)")]
        ax.legend(handles=hnd, loc="upper left", frameon=False, fontsize=FONT["legend"] - 1.0)
        ax.set_xticks(range(len(XCLASSES))); ax.set_xticklabels([XSHORT[c] for c in XCLASSES], fontsize=FONT["tick"])
        ax.set_ylim(0, 108); ax.set_ylabel(f"% of the cell in the tail (log $M_{{\\rm dust}}/M_\\star \\geq {P5C_TAIL:g}$)" if i == 0 else "")
        ax.tick_params(direction="in", top=True, right=True, labelsize=FONT["tick"]); ax.grid(False)
        ax.set_title(f"({'abcdef'[i]}) {BOXES[box]['label'].split(' (')[0]}: the tail per coupling x onset cell", loc="left", fontsize=FONT["tag"] - 0.8)
        # (bottom) the dose-age plane
        ax = axes[1][i]
        m_d = g["dusty"].to_numpy()
        ax.scatter(g.loc[~m_d, "age"], g.loc[~m_d, "e_x"], s=7, color="0.75", lw=0, zorder=1, label=f"not in the tail (N = {int((~m_d).sum())})")
        for oc in XCLASSES:
            for cc in P5C_COUP:
                m = m_d & (g["xray_class"] == oc).to_numpy() & (g["agn_class"] == cc).to_numpy()
                if not m.any():
                    continue
                ax.scatter(g.loc[m, "age"], g.loc[m, "e_x"], s=26, marker="o", facecolors="none" if cc == "weak" else XCOL[oc], edgecolors=XCOL[oc], lw=1.2,
                           zorder=3, label=f"tail, {XABBR[oc]} x {cc} (N = {int(m.sum())})")
        aa, ee = g["age"].to_numpy(float), g["e_x"].to_numpy(float); ok = np.isfinite(aa) & np.isfinite(ee)
        if ok.sum() >= 4 * P9_MIN:                                     # the dose typical of a classified galaxy of that age
            edges = np.quantile(aa[ok], np.linspace(0, 1, 7)); xm, ym = [], []
            for j in range(6):
                m = ok & (aa >= edges[j]) & (aa <= edges[j + 1])
                if m.sum() >= P9_MIN:
                    xm.append(float(np.median(aa[m]))); ym.append(float(np.median(ee[m])))
            ax.plot(xm, ym, color="0.15", lw=2.0, ls=(0, (1.2, 1.4)), zorder=4, path_effects=[withStroke(linewidth=3.4, foreground="white")],
                    label="median $E_x$ at this age (all classified)")
        ax.axvline(P5C_AGE, color="0.5", lw=1.0, ls=":", zorder=1)
        d = g[g["dusty"]]
        _lines = []
        for lab, m2 in ((f"age < {P5C_AGE:g}", d["age"] < P5C_AGE), (f"age $\\geq$ {P5C_AGE:g}", d["age"] >= P5C_AGE)):
            h = d[m2]
            if len(h) >= 3:
                _lines.append(f"tail, {lab}: N = {len(h)}, med $E_x$ {h['e_x'].median():.2f} Gyr, {XABBR[XREF]} {100 * (h['xray_class'] == XREF).mean():.0f} %, "
                              f"strong {100 * (h['agn_class'] == 'strong').mean():.0f} %")
        if _lines:
            ax.text(0.98, 0.03, "\n".join(_lines), transform=ax.transAxes, ha="right", va="bottom", fontsize=FONT["note"] - 0.5, color="0.15", zorder=6,
                    linespacing=1.35, path_effects=[withStroke(linewidth=2.6, foreground="white")])
        ax.set_xlim(0, float(np.nanmax(aa)) * 1.05 if np.isfinite(aa).any() else 10); ax.set_ylim(-0.15, max(2.0, float(np.nanmax(ee)) * 1.08 if np.isfinite(ee).any() else 2.0))
        ax.set_xlabel("mass-weighted stellar age at the anchor [Gyr]"); ax.set_ylabel(r"$E_x$ [Gyr]" if i == 0 else "")
        ax.tick_params(direction="in", top=True, right=True, labelsize=FONT["tick"]); ax.grid(False)
        ax.set_title(f"({'abcdef'[_nb + i]}) the dose-age plane: who the dusty galaxies are", loc="left", fontsize=FONT["tag"] - 0.8)
        ax.legend(loc="upper left", frameon=False, fontsize=FONT["legend"] - 1.8, handlelength=1.2, borderaxespad=0.3)
    fig.suptitle(wrap_caption(f"the coupling class crossed with the X-ray class on the dusty tail (log M_dust/M* >= {P5C_TAIL:g}; zero dust at the floor {P9_FLOOR:g}): "
                              f"(top) the tail fraction of every coupling x onset cell (68 % Wilson bars; hatched = weak, solid = strong, the colour = the X-ray class; dotted = the box); "
                              f"(bottom) the dusty galaxies on the dose-age plane against the median $E_x$ of the classified galaxies at that age — a tail galaxy below the line "
                              f"is under-dosed for its age, one on it received the full dose and is dusty anyway; quenched galaxies, {SEL_TXT}; classes = {XTITLE}: "
                              + ", ".join(f"{XSHORT[c]} = {XNAME[c]}" for c in XCLASSES), width=int(13 * fig.get_figwidth())), fontsize=FONT["note"] + 1, y=0.995)
    fig.subplots_adjust(top=0.91, bottom=0.06)
    paper_save(fig, "coupling_onset")
    plt.show()
```

## Section 5e — the specificity of the AUC test: every property against the dusty tail, the confounders held fixed, and the no-X-ray box as the control

Section 5c scored the dose quantities for the dusty tail. Before any winner is read causally ("the dusty QGs are rotation-dominated discs with
massive BHs"), three controls in one battery (the 5c sample per box = the classified event galaxies; the channel-off runs = every quenched galaxy):

* **Everything, not just the dose.** The same AUC for every catalogue property, in blocks: dose / kinematics ($\kappa_{\rm rot}$) / structure
  ($R_{\rm e}$, $\Sigma_{\rm e}$) / BH ($M_{\rm BH}$, $f_{\rm Edd}$, jet) / population ($M_\star$, age, $z$, sSFR, central) / gas fractions
  (outcome-adjacent: the tail *is* a dust fraction) + a **drawn random control** (AUC $\approx 0.5$ by construction). If half the catalogue
  scores as well as $E_x$, the tail is simply "a different kind of galaxy" and no single AUC is diagnostic.
* **Age and mass held fixed.** The tail is young, and $M_{\rm BH}$, $\kappa_{\rm rot}$, $R_{\rm e}$, the dose all run with age and $M_\star$ — so
  every correlated property inherits a raw AUC. The stratified column recomputes the AUC inside the (age, $\log M_\star$) quantile cells
  (terciles; medians below `P5E_Q3N` galaxies) with a within-cell label-permutation p, and Benjamini–Hochberg FDR is applied per box across the
  battery (both hand-rolled: the kernel has no statsmodels). A raw-significant property that dies here was riding the confounders.
* **The channel-off control.** s50nox (no X-ray feedback; s50noagn too when present) runs the same battery on its anchor columns: a property that
  separates dusty from dust-poor QGs *where the channel never fired* is generic gas physics, not an X-ray signature. Its dose rows stay empty
  until the s50nox catalogue ladder is on the share and B Parts 6a + 9 build its histories (B Part 9 now lists `cis50nox`) — then $E_x$ there
  becomes the **would-be dose** (the jets fire, the X-ray energy is never injected): dust falling with $E_x$ in the fiducial boxes but *not* with
  the would-be $E_x$ in nox is the causal signature; falling in both is confounding.

Caveat: $\kappa_{\rm rot}$ is resolution-limited outside m25 (~30–90 gas particles per quenched galaxy at 50/100 Mpc; `noxray_analogues` Part 5),
so the kinematics verdict is read from cis25 and the controls' kinematics rows carry that caveat. $\to$ `paper_xray_auc_battery{.png,.pdf,.csv}`.

```python
# ── Section 5e — the specificity of the AUC test: every catalogue property scored for the dusty tail, raw and at fixed (age, log M*), the no-X-ray box as the control ──
from scipy.stats import rankdata

P5E_TAIL   = P5C_TAIL               # the dusty tail of 5c (log M_dust/M* >= -4; zero dust at the floor)
P5E_NPERM  = 1000                   # label permutations inside the (age, log M*) cells (the stratified p)
P5E_Q3N    = 150                    # confounder terciles above this sample size, medians below
P5E_CONF   = ["age", "log_mstar"]   # the confounders held fixed in the stratified column
P5E_Q      = 0.05                   # the BH-FDR threshold marked in the table and the figure
P5E_CONTROLS = [v for v in ("cis50nox", "cis50noagn") if (QM["box"] == v).sum() >= 2 * P9_MIN]   # the channel-off runs: the same battery on the anchor columns
P5E_BOXES  = P9_BOXES + P5E_CONTROLS
P5E_BLOCKS = {"dose": "#d7191c", "kinematics": "#0072B2", "structure": "#009E73", "BH": "#7b3294", "population": "#E69F00", "ISM (adjacent)": "0.45", "control": "0.65"}
P5E_PROPS  = [   # (QM column, label, block); the AUC is rank-based, so a raw and a logged column give the same number
    ("e_x", r"$E_x$", "dose"), ("f_x", r"$f_x = E_x/\Delta t_{\rm SFT}$", "dose"), ("t_gaspoor", r"$T_{\rm gas\text{-}poor}$", "dose"),
    ("dt_onset_sft", r"$t_{\rm onset} - t_{\rm SFT}$", "dose"), ("w_pre", r"$w_{\rm pre}$", "dose"),
    ("dt_sft", r"$\Delta t_{\rm SFT}$ (time since SFT)", "dose"), ("tau_q_over_tqt", r"$\tau_q / t_{\rm QT}$", "dose"),
    ("kappa_gas", r"$\kappa_{\rm rot}$ gas", "kinematics"), ("kappa_star", r"$\kappa_{\rm rot}$ stars", "kinematics"),
    ("re", r"$R_{\rm e}$", "structure"), ("r_half_gas", r"$R_{1/2}$ gas", "structure"), ("sig_e", r"$\Sigma_{\rm e}$", "structure"),
    ("log_mbh", r"$\log M_{\rm BH}$", "BH"), ("fedd", r"$f_{\rm Edd}$", "BH"), ("jet", "jet mode at the anchor (0/1)", "BH"),
    ("log_mstar", r"$\log M_\star$", "population"), ("age", "stellar age", "population"), ("z", r"$z$", "population"),
    ("ssfr", "sSFR", "population"), ("central", "central (0/1)", "population"),
    ("fgas", r"$M_{\rm gas}/M_\star$", "ISM (adjacent)"), ("fh2", r"$M_{\rm H_2}/M_\star$", "ISM (adjacent)"), ("mhi_frac", r"$M_{\rm HI}/M_\star$", "ISM (adjacent)"),
    ("noise", "drawn random control", "control")]


def bh_q(p):
    """Benjamini-Hochberg q-values of the p-values `p` (NaN kept NaN; no statsmodels on the kernel)."""
    p = np.asarray(p, float); q = np.full(p.shape, np.nan)
    ok = np.where(np.isfinite(p))[0]
    if not len(ok):
        return q
    o = ok[np.argsort(p[ok])]
    r = p[o] * len(o) / np.arange(1, len(o) + 1)
    q[o] = np.clip(np.minimum.accumulate(r[::-1])[::-1], 0, 1)
    return q


def strat_auc(x, pos, confs, nperm=P5E_NPERM, seed=0):
    """AUC of x for `pos` with the confounders held fixed: the sample cut into the quantile cells of `confs` (terciles; medians below P5E_Q3N
    galaxies), the Mann-Whitney U summed over the cells, the labels permuted inside each cell only -> (auc, two-sided permutation p, N used)."""
    x, pos = np.asarray(x, float), np.asarray(pos, bool)
    ok = np.isfinite(x)
    for c in confs:
        ok &= np.isfinite(np.asarray(c, float))
    xs, ps = x[ok], pos[ok]
    if len(xs) < 4 * P9_MIN or ps.sum() < 3 or (~ps).sum() < 3:
        return np.nan, np.nan, int(len(xs))
    nq = 3 if len(xs) >= P5E_Q3N else 2
    sid = np.zeros(len(xs), int)
    for c in confs:
        v = np.asarray(c, float)[ok]
        e = np.unique(np.quantile(v, np.linspace(0, 1, nq + 1)))
        sid = sid * (nq + 1) + np.clip(np.searchsorted(e, v, side="right") - 1, 0, max(len(e) - 2, 0))
    rng = np.random.default_rng(seed)
    num = den = 0.0; pnum = np.zeros(nperm); used = 0
    for u in np.unique(sid):
        m = sid == u
        n1, n0 = int(ps[m].sum()), int((~ps[m]).sum())
        if n1 < 1 or n0 < 1:
            continue
        r = rankdata(xs[m])
        num += float(r[ps[m]].sum()) - n1 * (n1 + 1) / 2; den += n1 * n0; used += int(m.sum())
        pick = np.argsort(rng.random((nperm, int(m.sum()))), axis=1)[:, :n1]        # n1 random labels inside the cell, the ranks fixed
        pnum += r[pick].sum(axis=1) - n1 * (n1 + 1) / 2
    if den == 0:
        return np.nan, np.nan, used
    auc = num / den
    return float(auc), float((np.sum(np.abs(pnum / den - 0.5) >= abs(auc - 0.5)) + 1) / (nperm + 1)), used


D5E, N5E = [], {}
for box in P5E_BOXES:
    ctrl = box not in P9_BOXES
    g = QM[QM["box"] == box].copy()
    if not ctrl:
        g = g[g["xray_class"].isin(XCLASSES)].copy()                        # the Section 5 sample: classified event galaxies; controls: every quenched galaxy
    with np.errstate(divide="ignore", invalid="ignore"):
        g["mhi_frac"] = g["mhi"].to_numpy(float) / g["mstar"].to_numpy(float)
    g["noise"] = np.random.default_rng(int.from_bytes(box.encode(), "little") % 2 ** 32).normal(size=len(g))
    dusty = g["lfd"].to_numpy(float) >= P5E_TAIL
    nt = int(dusty.sum()); N5E[box] = (nt, len(g))
    print(f"\n{box}{' (X-ray channel off: the control; the dose rows stay empty until B builds its histories)' if ctrl else ''}: "
          f"{nt} of {len(g)} in the tail (log M_dust/M* >= {P5E_TAIL:g}: {100 * nt / max(len(g), 1):.1f} %)")
    if nt < 3 or len(g) - nt < 3:
        print("  one side of the tail split has under 3 galaxies: no AUC is defined here")
        continue
    rows = []
    for col, lab, blk in P5E_PROPS:
        x = g[col].to_numpy(float)
        auc, p = auc_of(x, dusty)
        auc_s, p_s, n_s = strat_auc(x, dusty, [g[c].to_numpy(float) for c in P5E_CONF], seed=len(rows))
        okx = np.isfinite(x)
        rows.append(dict(box=box, prop=col, label=lab, block=blk, n=int(okx.sum()), n_tail=int((okx & dusty).sum()), auc=auc, p=p, auc_strat=auc_s, p_strat=p_s, n_strat=n_s,
                         med_tail=float(np.nanmedian(x[dusty])) if (okx & dusty).any() else np.nan,
                         med_rest=float(np.nanmedian(x[~dusty])) if (okx & ~dusty).any() else np.nan))
    R = pd.DataFrame(rows)
    R["q"], R["q_strat"] = bh_q(R["p"].to_numpy()), bh_q(R["p_strat"].to_numpy())
    D5E.append(R)
    print(f"  {'property':16s} {'block':14s} {'N':>5s} {'AUC':>5s} {'p':>9s} {'q':>6s} {'AUC|age,M*':>10s} {'p_perm':>8s} {'q':>6s} {'med tail':>9s} {'med rest':>9s}")
    for r in R.itertuples():
        f1 = "*" if (np.isfinite(r.q) and r.q < P5E_Q) else " "
        f2 = "*" if (np.isfinite(r.q_strat) and r.q_strat < P5E_Q) else " "
        print(f"  {r.prop:16s} {r.block:14s} {r.n:5d} {r.auc:5.2f} {r.p:9.1g}{f1}{r.q:6.2f} {r.auc_strat:10.2f} {r.p_strat:8.2g}{f2}{r.q_strat:6.2f} {r.med_tail:9.3g} {r.med_rest:9.3g}")
    surv = R[np.isfinite(R["q_strat"]) & (R["q_strat"] < P5E_Q) & (R["prop"] != "noise")]
    surv = surv.iloc[np.argsort(-np.abs(surv["auc_strat"].to_numpy() - 0.5))]
    dropped = R[np.isfinite(R["q"]) & (R["q"] < P5E_Q) & ~(np.isfinite(R["q_strat"]) & (R["q_strat"] < P5E_Q)) & (R["prop"] != "noise")]
    print(f"  survives at fixed (age, log M*), q < {P5E_Q:g}: " + (", ".join(f"{r.prop} ({r.auc_strat:.2f})" for r in surv.itertuples()) if len(surv) else "nothing"))
    if len(dropped):
        print("  raw-significant but explained by age + M*: " + ", ".join(f"{r.prop} ({r.auc:.2f} -> {r.auc_strat:.2f})" for r in dropped.itertuples()))

B5E = pd.concat(D5E, ignore_index=True) if D5E else pd.DataFrame()
B5E.to_csv(os.path.join(PAPERDIR, "paper_xray_auc_battery.csv"), index=False)
print(f"\n-> {os.path.join(PAPERDIR, 'paper_xray_auc_battery.csv')}")
if len(B5E):
    _claim = ["kappa_gas", "kappa_star", "log_mbh", "fgas", "e_x", "f_x", "age"]      # the "rotation-dominated dusty disc with a massive BH" reading, quantity by quantity
    print(f"\nthe claim check across the boxes and the channel-off controls (AUC raw / at fixed age + M*; kappa_rot is resolved only in m25 — the 50 / 100 Mpc rows are shot-noise limited):")
    print(f"  {'property':12s} " + " ".join(f"{BOX_SHORT.get(b, b):>15s}" for b in P5E_BOXES))
    for c in _claim:
        cells = []
        for b in P5E_BOXES:
            r = B5E[(B5E["box"] == b) & (B5E["prop"] == c)]
            cells.append((f"{r['auc'].iloc[0]:5.2f} /{r['auc_strat'].iloc[0]:5.2f}" if np.isfinite(r["auc"].iloc[0]) else "        —") if len(r) else "        —")
        print(f"  {c:12s} " + " ".join(f"{s:>15s}" for s in cells))

# ── the figure: one panel per box, every property's raw AUC (circle) and (age, log M*)-stratified AUC (open diamond), the blocks coloured, BH-FDR marked ──
if len(B5E):
    _pb = [b for b in P5E_BOXES if (B5E["box"] == b).any()]
    ypos, yticks, ylabels, y, last_blk = {}, [], [], 0.0, None
    for col, lab, blk in P5E_PROPS:
        if last_blk is not None and blk != last_blk:
            y += 0.8
        ypos[col] = y; yticks.append(y); ylabels.append(lab); last_blk = blk; y += 1.0
    fig, axes = plt.subplots(1, len(_pb), figsize=(3.1 * len(_pb) + 2.6, 0.34 * len(P5E_PROPS) + 3.4), sharey=True)
    for ax, b in zip(np.atleast_1d(axes), _pb):
        R = B5E[B5E["box"] == b]
        ax.axvspan(0.45, 0.55, color="0.92", zorder=0)
        ax.axvline(0.5, color="0.4", lw=1.0, zorder=1)
        for r in R.itertuples():
            yy, cc = ypos[r.prop], P5E_BLOCKS[r.block]
            if np.isfinite(r.auc):
                sig = np.isfinite(r.q) and r.q < P5E_Q
                ax.plot([0.5, r.auc], [yy, yy], color=cc, lw=1.0, alpha=0.5, zorder=2)
                ax.plot(r.auc, yy, "o", ms=6.5, color=cc, mec="k" if sig else "none", mew=1.1, alpha=1.0 if sig else 0.45, zorder=3)
            if np.isfinite(r.auc_strat):
                sig_s = np.isfinite(r.q_strat) and r.q_strat < P5E_Q
                ax.plot(r.auc_strat, yy, "D", ms=5.0, mfc="white", mec=cc, mew=1.7 if sig_s else 0.8, alpha=1.0 if sig_s else 0.5, zorder=4)
        nt, n = N5E[b]
        ax.set_title(f"{BOX_SHORT.get(b, b)}" + (" (channel off)" if b not in P9_BOXES else "") + f"\ntail {nt} / {n} ({100 * nt / max(n, 1):.0f} %)", fontsize=FONT["tag"] - 0.8)
        ax.set_xlim(0.0, 1.0); ax.set_xlabel("AUC for the tail")
        ax.tick_params(direction="in", labelsize=FONT["tick"]); ax.grid(False)
    ax0 = np.atleast_1d(axes)[0]
    ax0.set_yticks(yticks); ax0.set_yticklabels(ylabels, fontsize=FONT["tick"] - 0.3); ax0.invert_yaxis()
    hnd = [Line2D([], [], ls="", marker="o", color="0.3", ms=6.5, label=f"raw AUC (black edge: BH q < {P5E_Q:g})"),
           Line2D([], [], ls="", marker="D", mfc="white", mec="0.3", ms=5.0, label=f"AUC at fixed (age, log $M_\\star$) (bold edge: permutation q < {P5E_Q:g})")] + \
          [Patch(facecolor=c, label=b) for b, c in P5E_BLOCKS.items()]
    fig.legend(handles=hnd, loc="lower center", ncol=5, frameon=False, fontsize=FONT["legend"] - 1.0)
    fig.suptitle(wrap_caption(f"the specificity of the dusty-tail AUC test (5c): every catalogue property as a score for tail membership (log M_dust/M* >= {P5E_TAIL:g}), raw (circles) "
                              f"and inside the (age, log M*) quantile cells (open diamonds; the labels permuted only inside the cells, {P5E_NPERM} draws), BH-FDR per box across the battery; "
                              f"grey band |AUC - 0.5| < 0.05, the drawn random control at the bottom; the channel-off runs run the same battery on the anchor columns — a property that "
                              f"separates the tail there too is generic gas physics, not an X-ray signature (their dose rows fill with the would-be dose once B Parts 6a + 9 build their "
                              f"histories); kappa_rot is resolved only in m25 (~30-90 gas particles per quenched galaxy at 50 / 100 Mpc); the ISM block is outcome-adjacent (the tail is "
                              f"defined on the dust fraction); quenched galaxies, {SEL_TXT}; classes = {XTITLE}", width=int(13 * fig.get_figwidth())), fontsize=FONT["note"] + 1, y=0.995)
    fig.subplots_adjust(top=0.87, bottom=0.16, left=0.17)
    paper_save(fig, "auc_battery")
    plt.show()
```

## Conventions, caveats, what the pieces are

* **The class table.** `xray_classes.csv` = every galaxy of every classified anchor of the boxes' histories (B Part 6a/9: cis100, cis50, cis25 — the
  m25 particle sample is the cis25 subset, joined on `(snap, gal_id)`; all 266 Q galaxies of `powderday_quenched_selection.fits` have a row and their
  $\tau_q / t_{\rm QT}$ agrees with the catalogue histories to $< 10^{-3}$ dex). Columns: the exposure quantities of B Part 9 ($E_x$, $T_{\rm gas\text{-}poor}$,
  $r_x$, $f_x$, $t_{\rm anchor} - t_{\rm SFT}$), SFT / QT / $\tau_q$ [Gyr], $t_{\rm onset}$ and $t_{\rm onset} - t_{\rm SFT}$ / $- t_{\rm QT}$ (B Part 6a), $w_{\rm pre}$ and
  the old two- and three-class labels (`agn_class`, `agn_class3`, the m25 pipeline's `agn_class_m25`), `fast_slow`, `xray_class` (= the official scheme's column),
  `onset_class` (the four-way B 6a label: lead / concurrent / late / never) and `rx_class` (always, whatever the scheme), `rx_undefined` (a quench event but never gas-poor since SFT: $T_{\rm gas\text{-}poor} = 0$, $E_x = 0$ → `xlow`
  in the rx scheme), $\log M_{\rm dust}/M_\star$ and $\log M_{\rm dust}/M_{\rm gas}$ with zero dust at the floor $-6.5$.
* **Why the onset timing.** $E_x$ grows with the time since SFT and $T_{\rm gas\text{-}poor}$ is a gas cut, so both correlate with the dust content by
  construction, and $r_x$ is piled up at 1 (most quenched galaxies keep a jet-mode BH whenever gas-poor), so any cut on it is either a tiny tail (0.5) or a
  box-dependent quantile (B Part 9's terciles: 0.91 / 0.85 / 0.97 in cis100 / cis50 / cis25 — no single fixed cut reproduces the three panels). The timing of the
  first gated episode is a discrete, history-level statement — did the channel fire by the end of the quenching, only after it, or never — and B Part 9's
  panels d–f order the dust by it. `early` merges B's `lead` (gas-poor before SFT) and `concurrent` (in [SFT, QT]): the two differ by ~0.5 dex in D/G at fixed
  elapsed time in cis100 / cis50 but not in cis25, and `onset4` keeps them apart; `onset_sft` splits the episode at SFT instead (`before` = `lead`,
  `after` = `concurrent` + `late`): the `before` galaxies are the ones already gas-poor with a jet-mode BH while star-forming (100 % jet lead, older and
  lower-sSFR cores at SFT — close to the m25 `strong` pre-SFT coupling class), 0.5 dex dustier than `after` in cis100 / cis50 (p 4e-35 / 7e-4) at every
  elapsed time, but not in cis25 (−2.78 vs −2.84, p = 1; core A_V 0.083 vs 0.089 mag). `never` is a physical class (no jet-mode BH: the s50nox phenotype inside the
  fiducial box) but tiny in the small boxes; `late` galaxies are recently quenched (short $\Delta t_{\rm SFT}$, all fast quenchers), so their dust excess must be
  read at fixed elapsed time (Section 5, rows 2–3).
* **Two event detectors.** The KS products carry the quench event of the *particle* histories (`has_clock`); the class table that of the *catalogue* histories
  (B Part 6a). They agree for almost every galaxy (Part 0 checks the times); a galaxy classified here but without a KS clock has no track and drops out of
  Sections 2, 4 and 6 (cell 2a prints the crosstab).
* **Floors.** Section 1 stacks need 5 galaxies per clock bin (band: `BAND_MIN` = 8); a KS stage median needs `NMIN_BIN` = 4 uncensored galaxies (open marker +
  arrow below that); Section 3 points need `F_NMIN` = 4 galaxies; Section 4 medians `NMIN_BIN` = 4 with a measurable value ($\kappa$: $\geq 10$ weighted
  particles; the H$_2$ rows are survivors); Section 5 binned medians `P9_MIN` = 6 per $E_x$ bin (a class below that is one median marker, as for the `never` and
  `late` classes of cis50 / cis25); the exposure diagnostic needs $2 \times$ `P9_MIN` galaxies per (box, class).
* **Clocks.** Section 1 runs on $t - t_{\rm SFT}$ (the m25 Part 2d tracks); the KS tracks are drawn at the critical points (SFT, QT, end; Section 4 adds the
  AGN ignition and the jet onset when the KS products carry them). $t_{\rm onset}$ is on the catalogue-history snapshot grid (a discrete time).
* **The mass sample.** Every per-class m25 figure uses $\log M_\star > 10.25$ at the anchor (Section 3 also draws the $< 10.25$ sample); Section 1 draws every
  event galaxy (`P1_MASS_MIN` restricts it). The $A_V$ tercile edges of Section 2 are those of this sample, not of A's $10.5$–$11.2$ window.
* **The boxes' side (Section 5)** uses whole-galaxy catalogue quantities (caesar FOF galaxy, $M_{\rm gas}$ = member gas); the m25 sections use the particle
  products of the m25 pipeline (projected apertures, spherical zones). The two are not the same aperture — the class is the same galaxy-level label.
* **The dusty tail (Section 5's distribution figure).** Under `onset`, the cis100 tail above $M_{\rm dust}/M_\star = 10^{-4}$ (20 % of the box) is tilted to
  `late` (18 % of the tail vs 7 % of the box, hypergeometric P = 4e-27) and `never` (7 vs 1 %, P = 3e-28) while `early` falls from 92 to 75 %, and the tilt
  strengthens with the threshold (above $10^{-3}$: `late` 45 %, `never` 31 %, `early` 25 %). The m25 tail instead matches the box shares — `early` 88 vs 90 %
  (`onset_sft`: `before` 27 vs 29 %, `after` 66 vs 68 %) — except the 4 `never` galaxies (all above $-4$: 7 % of the tail vs 2 % of the box, P = 0.01): in the
  m25 sample no timing class dominates the dusty tail beyond its overall share. cis50 under `onset_sft` does tilt to `before` (54 % of the tail above $-4$ vs
  38 % of the box, P = 0.07; 58 vs 38 % above D/G $= 10^{-2.5}$, P = 0.01).
